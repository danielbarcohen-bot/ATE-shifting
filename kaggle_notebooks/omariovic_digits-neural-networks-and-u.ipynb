```python
import pandas
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import tensorflow as tf
from tensorflow.keras import *
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.datasets import load_digits
```

```python
dataset = pandas.read_csv("/kaggle/input/digit-recognizer/train.csv")
dataset.head()
```

```python
X = dataset.iloc[:, 1:].values/255
y = dataset['label'].values
m, n = X.shape
X.shape, y.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 8, shuffle = True)
```

```python
model = Sequential([
    layers.Flatten(),
    layers.Dense(units = 256, activation = 'relu'),
    layers.Dense(units = 10, activation = 'softmax')
])
```

```python
model.compile(optimizer = 'adam', loss = 'sparse_categorical_crossentropy', metrics = ['accuracy'])
```

```python
model.fit(X_train, y_train, epochs = 60)
```

```python
y_pred = model.predict(X_test)
predictions = []
for it in y_pred:
    predictions.append(np.argmax(it))
```

```python
CM = confusion_matrix(y_test, predictions)
sns.heatmap(CM)
plt.show()
```

```python
p = np.random.randint(1, len(predictions))
print(predictions[p])
sns.heatmap(X_test[p].reshape((28, 28)))
plt.show()
```

```python
print(classification_report(y_test, predictions))
```

```python
test_ds = pandas.read_csv("/kaggle/input/digit-recognizer/test.csv")
test_ds/= 255
y_pred = model.predict(test_ds)
predictions = []
for it in y_pred:
    predictions.append(np.argmax(it))
```
