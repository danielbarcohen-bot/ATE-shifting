## [Optional] Exercise 6 Use Groupby Operation to Find Countries with the Fastest Growth in Ecological Footprint of Consumption

```python
# Step 1: form the condition 1

nfa_df['record'] == 'EFConsPerCap'
```

```python
# Step 2: form the condition 2

(nfa_df['year'] >= 2000) & (nfa_df['year'] <= 2013)

# Or use the one below
# nfa_df['year'].between(2000, 2013)
```

```python
# Step 3: Combine two conditions and put it into loc[]

nfa_df.loc[(nfa_df['record'] == 'EFConsPerCap') & \
           ((nfa_df['year'] >= 2000) & (nfa_df['year'] <= 2013))]
```

```python
# Step 4: Assign the result to the new dataframe

# Use pandas.DataFrame.loc[] to create a dataframe named efcons_period_df 
# for records of Ecological Footprint of Consumption per Capita (EFConsPerCap) between 2000 and 2013

# Replace this line with your code
efcons_period_df = nfa_df.loc[(nfa_df['record'] == 'EFConsPerCap') & \
                              ((nfa_df['year'] >= 2000) & (nfa_df['year'] <= 2013))]
efcons_period_df
```

```python
# Step 1: form the group

efcons_period_df.groupby(by='country', as_index=False)

# list()[0:10] shows the first ten for illustration
list(efcons_period_df.groupby(by='country', as_index=False))[0:10]
```

```python
# Step 2: What is first and last?

# first row in the group
efcons_period_df.groupby(by='country', as_index=False).first()
```

```python
# last row in the group
efcons_period_df.groupby(by='country', as_index=False).last()
```

```python
# Step 3: Use agg()

efcons_period_df.groupby(by='country', as_index=False) \
    .agg({'total': ['first', 'last']})
```

```python
# Step 4: Assign the result to a new dataframe

# Use pandas.DataFrame.groupby() and pandas.DataFrame.agg() to find the Ecological Footprint of Consumption per Capita
# for each country (group by country) at the start and the end of the time period from 2000 to 2013
# Pass the result to a dataframe named efcons_group_df

# Hint: use agg({'total': ['first', 'last']}) to obtain the first and the last rows of the column named total for each group

# Replace this line with your code
efcons_group_df = efcons_period_df.groupby(by='country', as_index=False). \
    agg({'total': ['first', 'last']})

efcons_group_df
```

```python
# Step 1: Calculate the change

efcons_group_df[('total', 'last')] - efcons_group_df[('total', 'first')]
```

```python
# Step 2: Assign the change value to a new column

efcons_group_df[('total', 'change')] = \
    efcons_group_df[('total', 'last')] - efcons_group_df[('total', 'first')]
```

```python
# Step 3: Update the dataframe

# Calculate the change in the total Ecological Footprint of Consumption from 2000 to 2013 for each country
# Create a new column in efcons_group_df to store the result

# Hint: Use a tuple to represent a multi-level column index. 
# E.g., efcons_group_df[('total', 'change')] creates a sub-column named change under the column named total

# Replace this line with your code
efcons_group_df[('total', 'change')] = efcons_group_df[('total', 'last')] - efcons_group_df[('total', 'first')]

efcons_group_df
```

```python
# Step 1: Sorting

efcons_group_df.sort_values([('total', 'change')], ascending=False)
```

```python
# Step 2: Show top 20

# Use pandas.DataFrame.sort_values() to find out the Top 20 countries with the largest growth in 
# the Ecological Footprint of consumption per Capita from 2000 to 2013

# Hint: Use iloc[:20] to select the first 20 rows in a dataframe

# Replace this line with your code
efcons_group_df.sort_values([('total', 'change')], ascending=False).iloc[:20]
```
