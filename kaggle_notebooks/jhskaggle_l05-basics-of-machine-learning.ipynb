### Machine learning
* Machine learning is to let a computer program learn from data without prior knowledge such as `IF THEN` rules
1. we  have a **data** set of iris flowers
1. machine learning is able to build a computational **model**
1. Then the model is able to **predict** the label on new data

Example:  a decision tree model 
* without explicit rules and instructions for classfying iris flowers
* instead, the model generates rules from learning on the data 
* Note: the following code aims to generate a decision tree. However, it is not necessary to understand the code at this stage. We will introduce the decision tree in future lectures.  

```python
# iris flower classification: easy to machine learning, but difficult to conventional programming
from sklearn.datasets import load_iris #iris is a built-in data set in sklearn
iris = load_iris() 
X = iris.data
y = iris.target
from sklearn.model_selection import train_test_split # Split dataset into training set and test set
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)
from sklearn.tree import DecisionTreeClassifier 
clf = DecisionTreeClassifier() # Create a Decision Tree classifer
clf = clf.fit(X_train,y_train)# Train Decision Tree Classifer
from matplotlib import pyplot as plt #plot decision tree
from sklearn import tree 
fig = plt.figure(figsize=(25,20))
_ = tree.plot_tree(clf, feature_names= ['sepal-length', 'sepal-width', 'petal-length', 'petal-width'],
                   class_names=['Iris-setosa','Iris-versicolor','Iris-virginica'], filled=True)
```

### Definition of machine learning

*"A computer program is said to learn from experience E with respect to some class of tasks T and performance measure P, if its performance at tasks in T, as measured by P, improves with experience E."* (Mitchell 1997)

* Machine learning is a subfield of artificial intelligence
* Aim to develop self-learning computer programs
* Ability to gradually improve the performance of computer programs

## Part 3 Supervised learning
### Concept of supervised learning
**Supervised learning** refers to a machine learning process in
which a model learns from input data and generate an output. The  data comprises some examples with labels. 

Goal of supervised learning: 
* learn a model from labelled training data 
* make predictions on unseen or future data without labels

**Supervised** means that a set of samples with desired labels is available for learning

Supervised learning is the most common machine learning type

### Example: Classification
Classification: predict the class label (categorical value) of a dependent variable (target) based on  one or more features (independent variables, predictors)

* Task: predict the categorical class labels of new instances based on past observations.
* Experience: learning on a data set with labels  
* Performance: the prediction error on data

**Example:** 
* given an iris flower data set
* for the sake of visualization, consider **two features (sepal width and sepal length)**
* build a decision tree which can classify the **specie** (class label) of an iris flower.

* The scatter plot shows the distribution of the data

* the boundary between classes is not clear
* it is hard to class iris flowers based on the two features.


```python
#Source: http://scipy-lectures.org/packages/scikit-learn/auto_examples/plot_iris_scatter.html
from sklearn.datasets import load_iris
iris = load_iris() 
from matplotlib import pyplot as plt 
x = iris.data[:, 0] # sepal width
y = iris.data[:, 1] # sepal length
 
plt.figure(figsize=(4,3))
plt.scatter(x,y, c=iris.target) # c for color

plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])  
plt.title('iris data with labels')  
plt.show()
```

### Regression 

**Regression**: predict the numerical value of a target variable (outcome or dependent  variable) based on  one or more features (independent variables or predictors).  

* Task: predict the outcome of a dependent variable on new data. based on past observations 
* Experience: learning dataset with several predictor (features) variables and a continuous dependent variable (outcome)
* Performance: error between prediction and actual value

### Data preprocessing: Feature engineering

The raw data usually are not in a format ready for machine learning. We have to convert them to a required format. This process is called feature engineering.

* Data cleaning such as handling outliers, handling missing values
* Scaling features
* Encoding features
 


### Data preprocessing: Feature selection

Given available features, some features are useful in classification, but other may not. In fact, some of them may degrade the accuracy of classification. 

Feature selection refers to select the most useful features from a data set and improve the performance of machine learning.

Feture selection is to select a subset of original features, but does not create new features

Example: suppose the orginal iris data had two other features, *collection time and collector name*. However, these two features play no role in determing iris flower species.

### Data preprocessing: Feature extraction

Feature extraction refers to creates new features by combining correlated features.

Big data usually have many features. We will create informative new features and removes redundant features.

Feature extraction is often implemented through dimension reduction (e.g. Principal component analysis), which reduces number of features from high to low.

Example: the [URL Reputation Data Set](https://archive.ics.uci.edu/ml/datasets/URL+Reputation) contains 2.4 million examples and **3.2** million features. Imagine how difficult to deal with 3.2 million features.

Model selection and model assessment
* **Model selection** is the process of selecting one final model from a set of candidate models based on their performance on the training data

### Model building: hyperparameter tuning
A model may include types of parameters (configurations)
1. An internal parameter of a model can be automatically estimated from data. This kind of parameters is not required to   manually tune  
2. A hyperparameter cannot be automatically estimated from data.  This parameter is required to manually tune  


### Model evaluation

Performance metrics for Classification includes

## Part 4 Unsupervised Learning  
### Unsupervised Learning
* In unsupervised learning: input data is not labelled  
* however, there exist some *structure* or *distribution* in the data

Goal for unsupervised learning
* discover underlying structure or distribution in the data in order to learn more about the data.

* **unsupervised** refers to no labelled data available (or correct answers).


### Example: Clustering 

Clustering belongs to unspervised learning   

Organize data into meaningful subgroups (clusters) without having any prior knowledge of their group memberships

Members in the same cluster are similar but not similar among different clusters   

Clustering is a great technique for structuring information and deriving meaningful relationships among data 

* Task: cluster data into groups based on similarity
* Experience: clustering data into groups without prior roup memberships (labels)
* Performance:  only subject method but no objective method for evaluating the model performance
 

**Example:**

* given an iris flower data set without species labels (same colour in the figure), build a model for grouping flowers into clusters (species).

```python
#Source: http://scipy-lectures.org/packages/scikit-learn/auto_examples/plot_iris_scatter.html
from sklearn.datasets import load_iris
iris = load_iris() 
from matplotlib import pyplot as plt 
x = iris.data[:, 0] # sepal width
y = iris.data[:, 1] # sepal length
 
plt.figure(figsize=(4,3))
plt.scatter(x,y, c='red') # c for color

plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])  
plt.title('iris data with labels')  
plt.show()
```

* Clustering is to put a group label to each flower (sample).

* however, unlike supervised learning, no test exists in clustering. Learning finishes after adding labels to samples!

### Difference between supervised learning and unsupervised learning
* **Data:** labelled data vs unlabelled data
* **Process:** training + test vs learning (no test)
* **Performance evaluation:** objective metrics such as accuracy vs no objective metrics
