# List

* List is a mutable, ordered sequence of elements.
* Created by placing elements inside square brackets [], separated by commas.

**List indexing & slicing**
* Access an item in a list by referring to its index (index starts from 0).
* Access from the end using negative numbers.
* Extract parts of lists using slice notation [start:end:step]

```python
fruits = ["apple","banana","orange","kiwi","pear","melon","blueberry","coconut","mango","cherry"]
print(fruits[1])  # Output: 'banana'
print(fruits[-1]) # Output: 'cherry'
print(fruits[2:5])  # Output: ['orange', 'kiwi', 'pear']
print(fruits[:4])   # Output: ['apple', 'banana', 'orange', 'kiwi']
print(fruits[::2])  # Output: ['apple', 'orange', 'pear', 'blueberry', 'mango']
```

**Modifying List**
* Adding Items: Use append(), insert(), and extend()
* Removing Items: Use remove(), and pop()

```python
fruits.append("lemon")
fruits.insert(1, "blackberry")
fruits.extend(["strawberry", "durian"])
print(fruits)

fruits.remove("durian")
print(fruits)
fruits.pop()
print(fruits)
fruits.pop(1)
print(fruits)
```

**Common List methods**
* sort(): sorts the list.
* reverse(): reverses the elements of the list.
* count(): returns the number of elements with the specified value.
* index(): finds the first occurrence of the specified value.

```python
fruits.sort()
print(fruits)
fruits.reverse()
print(fruits)
print(fruits.count("apple"))
print(fruits.index("banana"))
```

**List comprehension**

# Dictionary

* Dictionary is an ordered collection, indexed by keys that are associated with values.
* Created with curly braces {} containing key:value pairs separated by commas.
* As of Python 3.7, dictionaries are ordered. In Python 3.6 and earlier, they are unordered.

**Accessing dictionary values**
* Values are accessed by their keys.

```python
person = {"name": "John", "age": 30, "city": "New York"}

print(person["name"])  # Output: 'John'
# If the key doesn't exist, it will raise a KeyError unless you use get()
print(person.get("name"))  # Output: 'John'
```

**Adding and modifying entries**
* Assign a value to a key directly.
* If the key doesn’t exist, it will be added as a new key:value pair.
* If the key already exists, it will update the value associated with the key.

**Removing entries**
* pop() removes the specified key city along with its value.

```python
person["age"] = 31  # Updates if exists, adds if doesn't
person["profession"] = "Engineer"  # Adds a new key-value pair

person.pop("city")
print(person)
```

**Iterating through a Dictionary**
* person.keys() for keys.
* person.values() for values.
* person.items() for key-value pairs (tuples).

```python
print(person.keys())
print(person.values())
for key, value in person.items():
    print(f"{key}: {value}")
```

# Set

* Set is an unordered collection of unique elements.
* Created by placing elements inside curly braces {} or using the set() constructor.
* Unordered: The elements in a set will appear in an arbitrary order.
* Unique elements: Sets automatically remove duplicate elements.
* Immutability of elements: Elements of a set must be immutable.

```python
my_set = {1, 2, 3, 4, 5}
text_set = set("hello")  #{'e', 'h', 'l', 'o'}
print(my_set)
print(text_set)
```

**Adding and removing elements**
* add(): add a single element into a set
* update(): add multiple elements into a set
* remove(): remove an element (raises an error if the element is not found) 
* discard(): remove an element (does not raise an error though the element is not found)

```python
my_set.add(6)
my_set.update([7, 8])
my_set.discard(10)  # No error if it does not exist
my_set.remove(5)
print(my_set)
```

* Union (|): Combines elements from two sets.
* Intersection (&): Elements common to both sets.
* Difference (-): Elements in one set but not the other.
* Symmetric Difference (^): Elements in either of the sets but not in both.
* in keyword: Quickly check if an element exists within a set.

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a | b)  # Union: {1, 2, 3, 4, 5, 6}
print(a & b)  # Intersection: {3, 4}
print(a - b)  # Difference: {1, 2}
print(a ^ b)  # Symmetric Difference: {1, 2, 5, 6}
print(3 in a)  # Output: True
print(7 in a)  # Output: False
```

# Control Structure: if statement

* if statement allows you to execute certain pieces of code based on a condition.

```python
condition = False
another_condition = True

if condition:
    print("This block will be executed if condition = True")
elif another_condition:
    print("This block will be executed if condition = False and another_condition = True")
else:
    print("This block will be executed if condition = False and another_condition = True")
```

```python
score = 85
if score >= 90:
    print("Grade A")
elif score >= 80:
    print("Grade B")
else:
    print("Grade C")
```

**Conditional expression**
* value_if_true if condition else value_if_false

```python
age = 20
status = "teenager" if age < 20 else "adult"
print(status)  # Output: 'adult'
```

# Control Structure: Loop

* Loop allows the execution of a block of code multiple times depending on a condition or an iterator.
* for loop iterates each element over a sequence (like a list, tuple, dictionary, set, or string).

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

# range(N) generates a list of numbers starting from 0 to N (not included).
for i in range(5):
    print(i)
```

* While repeats as long as a certain boolean condition is met.

```python
count = 0
while count < 5:
    print(count)
    count += 1
```

**Controlling loop execution**
* break: Immediately exits the loop.
* continue: Skips the rest of the code inside the loop for the current iteration and moves to the next iteration.

```python
for num in range(10):
    if num == 3:
        continue  # skip print for 3
    if num == 8:
        break  # exit loop if num is 8
    print(num)
```

```python
# Disaplay data from a range of selected rows
# Selecting a range of rows is similar to the list (i.e., position starting from 0)
print(df.iloc[1:3])  # data from the 1st position before the third row (not included)
print(df.iloc[1:])  # data from the 1st position to the last row
print(df.iloc[:1])  # data from the beginning row before the 1st row (not included)
```

* groupby() groups data based on the value of the specified column and performs aggregate functions for each group.

* Using agg() function, you can decide which aggregation to perform on each column
