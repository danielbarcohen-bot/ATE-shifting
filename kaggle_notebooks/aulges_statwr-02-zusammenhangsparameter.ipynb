```python
import matplotlib.pyplot as plt
import numpy as np

def read():
    ''' reads a field of numbers from a textfile.'''
    with open('/kaggle/input/samples/03_samples.txt') as file:
        lines = file.readlines()
    # convert each line from string to sequence of floats. 
    X = list(map(lambda x: [float(t) for t in x.split()], lines))
    X = np.array(X) 
    return X

def plot(X):
    _ = plt.scatter(X[:,0],X[:,1],s=5)
    plt.xlim([3,7]); plt.ylim([-10,30])
    plt.grid(); plt.show()

X = read()
print('X is a dataset with %s rows/points, and with %d columns.' %X.shape)
plot(X)
```
