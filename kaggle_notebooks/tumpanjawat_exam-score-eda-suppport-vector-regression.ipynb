Methods for handling missing values</span></b> </div>

#### **1. Interpolation**

In data analysis, **<span style='color:#E55807'>interpolation</span>** is a method of estimating missing data points in a sequence by taking into consideration the neighboring known data points. 

The **<mark style="background-color:#E55807;color:white;border-radius:5px;opacity:1.0">interpolate</mark>** function in pandas is used to fill missing values in our dataset. It works under the assumption that there's a linear relationship between the data points, and the missing value will lie on the line that connects the nearest known data points.

<ul>
If we have a series of values \(x_1, x_2, ..., x_n\) and \(x_i\) is missing, the interpolated value will be calculated based on its neighboring values \(x_{i-1}\) and \(x_{i+1}\).
For a simple linear interpolation, this can be formally represented as:
</ul>

<p class="formulaDsp">
\[ x_i = \frac{(x_{i-1} + x_{i+1})}{2} \]
</p>

#### **2. Mean/Median Imputation**

**<span style='color:#E55807'>Mean/Median imputation</span>** is another technique for handling missing values in numeric data. This approach involves calculating the mean or median of a feature, and then replacing the missing values with this calculated value. 

<ul>
If we have a series of values \(x_1, x_2, ..., x_n\) and \(x_i\) is missing, then the missing value \(x_i\) is replaced by the mean (\(mean(\vec{x})\)) or the median (\(median(\vec{x})\)) of the feature vector \( \vec{x} \).
</ul>

<p class="formulaDsp">
\[ x_i = mean(\vec{x}) \quad \text{or} \quad x_i = median(\vec{x}) \]
</p>

This approach is useful when the missing data is missing completely at random and the feature has a symmetric or approximately symmetric distribution.

#### **3. Mode Imputation**

**<span style='color:#E55807'>Mode imputation</span>** is often used for handling missing values in categorical data. This method replaces the missing values with the most frequently occurring value, or the mode, of a feature.

<ul>
If we have a series of values \(x_1, x_2, ..., x_n\) and \(x_i\) is missing, then the missing value \(x_i\) is replaced by the mode (\(mode(\vec{x})\)) of the feature vector \( \vec{x} \).
</ul>

<p class="formulaDsp">
\[ x_i = mode(\vec{x}) \]
</p>

This method is most useful when the data are categorical and the mode is a good representation for the missing values.

<br>

<div style="border-radius:10px;border:#E55807 solid;padding: 15px;background-color:#ffffff00;font-size:100%;text-align:left">     
    ??? Note: While these methods are useful, they should be used judiciously. They all make assumptions about the nature of the missing data, and it is critical to understand whether these assumptions are reasonable in the context of a specific dataset and analysis.
</div>

## <div style="padding: 20px;color:white;margin:10;font-size:90%;text-align:left;display:fill;border-radius:10px;overflow:hidden;background-image: url(https://images.pexels.com/photos/50570/sandstone-the-wave-rock-nature-50570.jpeg?auto=compress&cs=tinysrgb&w=1260&h=750&dpr=1)"><b><span style='color:white'> 1. One-Hot Encoding</span></b> </div>

<span style='color:#E55807'>One-Hot Encoding</span> is a crucial preprocessing step for categorical data. Many machine learning models require inputs to be numerical. While some can handle categorical data, most cannot, hence the need for a method to represent these categories numerically.

<mark style="background-color:#E55807;color:white;border-radius:5px;opacity:1.0">OneHotEncoder</mark> transforms categorical data into a format that works better with classification and regression algorithms.

For each category in a feature, the one-hot encoder creates a new binary feature. Each of these features indicates the presence (1) or absence (0) of the category in the original feature.

This way, categorical data can be efficiently used by machine learning algorithms. But be aware, one-hot encoding can significantly increase the data's dimensionality, a problem known as the "curse of dimensionality".

```python
# Define the function for one-hot encoding
def perform_one_hot_encoding(df, column_name):
    # Perform one-hot encoding on the specified column
    dummies = pd.get_dummies(df[column_name], prefix=column_name)

    # Drop the original column and append the new dummy columns to the dataframe
    df = pd.concat([df.drop(column_name, axis=1), dummies], axis=1)

    return df

data = df.copy()
data1 = df.copy()

# Perform one-hot encoding on the categorical variables
for column in ['Gender', 'EthnicGroup', 'ParentEduc', 'LunchType', 'ParentMaritalStatus', 'PracticeSport', 'WklyStudyHours']:
    data = perform_one_hot_encoding(data, column)
    
# Perform one-hot encoding on the categorical variables
for column in ['Gender', 'EthnicGroup', 'LunchType', 'ParentMaritalStatus', 'WklyStudyHours']:
    data1 = perform_one_hot_encoding(data1, column)
```

```python
data = df.copy()
# One-hot encoding for categorical variables
categorical_cols = ['Gender', 'EthnicGroup', 'ParentEduc', 'LunchType', 
                    'ParentMaritalStatus', 'PracticeSport', 'WklyStudyHours']
                   

for col in categorical_cols:
    data = pd.get_dummies(data, columns=[col], prefix = [col], drop_first=True)
```

```python
# Split the dataset into features and targets
features = data.drop(['MathScore', 'ReadingScore', 'WritingScore'], axis=1)
target_cols = ['MathScore', 'ReadingScore', 'WritingScore']
```

```python
# Standardization
scaler = StandardScaler()
features = scaler.fit_transform(features)
```

```python
# Set the parameters for cross-validation
parameters = {'C': np.logspace(-3, 2, 6), 
              'epsilon': np.logspace(-3, 2, 6), 
              'kernel': ['linear', 'poly', 'rbf', 'sigmoid']}

for target in target_cols:
    targets = data[target]

    # Split the dataset into training set and test set
    X_train, X_test, y_train, y_test = train_test_split(features, targets, test_size=0.2, random_state=42)

    # Perform Randomized Search CV
    random_search = RandomizedSearchCV(svm.SVR(), parameters, cv=3, n_jobs=-1, verbose=2)
    random_search.fit(X_train, y_train)

    # Get the best estimator
    best_model = random_search.best_estimator_

    # Predict the scores
    y_pred = best_model.predict(X_test)

    # Evaluation
    r2 = r2_score(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)

    print(f"{target} - R2 Score: {r2}")
    print(f"{target} - Mean Squared Error: {mse}")
    print(f"{target} - Mean Absolute Error: {mae}")

    # Visualization: plot predicted vs actual values
    plt.figure(figsize=(12, 6))
    plt.scatter(y_test, y_pred, alpha=0.5)
    plt.xlabel("Actual Scores")
    plt.ylabel("Predicted Scores")
    plt.title(f"Actual vs Predicted Scores for {target}")
    plt.show()
```
