## Tableaux et matrices avec Numpy

NumPy permet de manipuler efficacement des **vecteurs** (tableaux 1D) et des **matrices** (tableaux 2D), ainsi que d’effectuer des opérations mathématiques vectorisées.


### Vecteurs et matrices

Un **vecteur** est un tableau 1D :

```python
x = np.array([1, 2, 3, 4])
```

Une **matrice** est un tableau 2D :

```python
X = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])
```


### Forme d’un tableau (`shape`)

```python
x.shape
X.shape
```

* un vecteur a une forme `(n,)`
* une matrice a une forme `(n_lignes, n_colonnes)`

La forme est essentielle pour vérifier la compatibilité des opérations.



### Changement de forme avec `reshape`

La méthode `reshape` permet de **modifier la forme d’un tableau** sans changer ses valeurs.

```python
x = np.arange(1, 11)
X = x.reshape(5, 2)
```

* `x` contient 10 éléments
* `X` est une matrice de 5 lignes et 2 colonnes
* le nombre total d’éléments doit rester identique

```python
x.size == X.size
```

`reshape` est souvent utilisé pour passer :

* d’un vecteur à une matrice
* d’une représentation linéaire à une représentation tabulaire



### Sommes et moyennes

```python
np.sum(x)
np.mean(x)
```

Pour une matrice, il est possible de préciser l’axe :

```python
np.sum(X, axis=0)  # somme par colonne
np.sum(X, axis=1)  # somme par ligne
```


### Produit scalaire (dot product)

Le **produit scalaire** entre deux vecteurs est une opération fondamentale en régression linéaire.

```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

np.dot(u, v)
```

Il correspond à :

$$
u \cdot v = \sum_i u_i v_i
$$



### Multiplication matrice-vecteur

```python
X = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

w = np.array([0.5, 1.0])

X @ w
```

Cette opération est centrale dans les modèles linéaires :

$$
\hat{y} = X w
$$



### Différence entre `*` et `@`

```python
X * w    # multiplication élément par élément
X @ w    # produit matrice-vecteur
```

* `*` : opération élément par élément
* `@` : produit matriciel


### Opérations élément par élément

```python
x + 1
x * 2
x ** 2
```

Ces opérations sont vectorisées et remplacent les boucles explicites.

### `pd.read_csv(...)` : lire un fichier CSV dans un DataFrame

`pandas.read_csv` est la porte d’entrée classique pour transformer un fichier texte (CSV, TSV, etc.) en **DataFrame** (une table en mémoire).

**Idée générale**  
- Le fichier contient des lignes (observations) et des colonnes (variables).  
- `read_csv` détecte un séparateur, lit les valeurs, gère les valeurs manquantes, et renvoie un objet `DataFrame`.

**Exemple**  
```python
df = pd.read_csv('/kaggle/input/mini-datasets/co2_mini.csv')
```

- `sep=','` : séparateur (utile si c’est `;` ou `\t`)

**Résultat**  
`df` est un DataFrame.

### Sélectionner une colonne d’un DataFrame

Avec pandas, une colonne se récupère le plus souvent de deux façons :

**1) Avec des crochets**
```python
df['co2']           # une colonne → un objet pandas.Series
df[['co2']]         # une colonne mais avec double crochets → DataFrame (1 colonne)
df[['consumption','co2']]  # plusieurs colonnes → DataFrame
```

**2) Avec la notation point (pratique, mais plus fragile)**
```python
df.co2
```
Elle marche seulement si le nom de colonne :
- n’a pas d’espace,
- ne contient pas de caractères spéciaux,
- et ne rentre pas en conflit avec un attribut de pandas.

**Série vs DataFrame (très important)**
- `df['co2']` → `Series` (1D)  
- `df[['co2']]` → `DataFrame` (2D)

Certaines fonctions (notamment en ML) exigent un tableau 2D, donc on utilise souvent la version `df[[...]]`.

### Supprimer des lignes ou des colonnes 

La méthode `drop` permet de **supprimer des lignes ou des colonnes** d’un DataFrame Pandas.

Elle est fréquemment utilisée pour :

* retirer une colonne inutile
* supprimer une variable cible de `X`
* nettoyer un jeu de données avant un modèle



#### Principe général

La syntaxe générale est :

```python
df.drop(labels, axis=0 ou 1)
```

où :

* `labels` désigne le nom (ou les noms) à supprimer
* `axis` indique s’il s’agit de lignes ou de colonnes



#### Suppression de colonnes

Pour supprimer une ou plusieurs colonnes, on utilise `axis=1`.

```python
df.drop('Rings', axis=1)
```

ou de manière plus explicite :

```python
df.drop(columns=['Rings'])
```

Les deux syntaxes sont équivalentes, la seconde est généralement plus lisible.



#### Suppression de plusieurs colonnes

```python
df.drop(columns=['Length', 'Weight'])
```



#### Suppression de lignes

Pour supprimer des lignes, on utilise `axis=0` (valeur par défaut).

```python
df.drop(0)
```

Supprime la ligne d’indice `0`.



#### Suppression de plusieurs lignes

```python
df.drop([0, 1, 2])
```

Supprime les lignes d’indices `0`, `1` et `2`.



#### Effet sur le DataFrame original

Par défaut, `drop` **ne modifie pas** le DataFrame original.
Il retourne une **nouvelle copie**.

```python
df2 = df.drop(columns=['Rings'])
```

Pour modifier le DataFrame en place :

```python
df.drop(columns=['Rings'], inplace=True)
```



#### Conversion d’un DataFrame Pandas vers un tableau NumPy

### Statistiques descriptives avec `df.describe()`

La méthode `describe()` permet d’obtenir rapidement une **synthèse statistique** des colonnes numériques d’un DataFrame.

```python
df.describe()
```

Elle calcule automatiquement plusieurs indicateurs décrivant la **distribution des données**.


#### Statistiques fournies par `describe()`

Pour chaque colonne numérique, `describe()` retourne les statistiques suivantes.



##### 1. `count`

* Nombre de valeurs **non nulles**
* Peut être inférieur au nombre de lignes si des valeurs sont manquantes



##### 2. `mean` (moyenne)

* Moyenne arithmétique des valeurs
* Donne une valeur centrale globale de la variable


##### 3. `std` (écart-type)

* Mesure la **dispersion** des données autour de la moyenne
* Plus l’écart-type est grand, plus les valeurs sont dispersées



##### 4. `min`

* Valeur minimale observée dans la colonne



##### 5. Quartiles (25 %, 50 %, 75 %)

Les quartiles découpent les données **ordonnées** en quatre parties égales.

###### Premier quartile (25 %)

* 25 % des valeurs sont **inférieures ou égales** à cette valeur
* Noté souvent ( Q_1 )



###### Deuxième quartile (50 %) — médiane

* Valeur qui partage les données en deux parties égales
* 50 % des valeurs sont en dessous, 50 % au-dessus
* Correspond à la **médiane**


###### Troisième quartile (75 %)

* 75 % des valeurs sont **inférieures ou égales** à cette valeur
* Noté souvent ( Q_3 )



###### Interprétation des quartiles

Les quartiles permettent de :

* comprendre la **répartition** des données
* identifier la **dispersion centrale**
* détecter des **valeurs atypiques** (outliers)

L’intervalle interquartile est défini par :

$IQR = Q_3 - Q_1$



##### 6. `max`

* Valeur maximale observée dans la colonne



### Visualisation de la distribution avec les boxplots

Les **boxplots** (ou diagrammes en boîte) permettent de visualiser graphiquement les statistiques descriptives présentées précédemment, en particulier les **quartiles** et les **valeurs aberrantes**.


#### 1. Principe du boxplot

Un boxplot représente la distribution d’une variable numérique à l’aide de cinq valeurs clés :

* la valeur minimale
* le premier quartile ( Q_1 ) (25 %)
* la médiane ( Q_2 ) (50 %)
* le troisième quartile ( Q_3 ) (75 %)
* la valeur maximale

La **boîte** correspond à l’intervalle interquartile :

$IQR = Q_3 - Q_1$



#### 2. Construction du boxplot

* La boîte s’étend de ( Q_1 ) à ( Q_3 )
* Le trait à l’intérieur de la boîte représente la **médiane**
* Les « moustaches » s’étendent vers les valeurs les plus extrêmes **non aberrantes**



#### 3. Définition des valeurs aberrantes (outliers)

Une valeur est considérée comme **aberrante** si elle est située :

$$
x < Q_1 - 1.5 \times IQR
$$

ou
$$
x > Q_3 + 1.5 \times IQR
$$

Ces valeurs sont affichées individuellement sur le boxplot, généralement sous forme de points.



#### 4. Exemple avec Pandas

```python
df["co2"].plot(kind="box")
```

Ce graphique permet de visualiser :

* la concentration des valeurs centrales
* l’asymétrie éventuelle de la distribution
* la présence de valeurs aberrantes



#### 5. Interprétation des outliers

Les valeurs aberrantes peuvent avoir plusieurs origines :

* erreurs de mesure
* erreurs de saisie
* événements exceptionnels
* phénomènes réellement rares mais valides

Il est important de **ne pas supprimer systématiquement** les outliers sans analyse.

### Boxplot interactif avec Plotly

En complément du boxplot réalisé avec Pandas, il est possible d’utiliser **Plotly** afin d’obtenir une **visualisation interactive** de la distribution des données.

Plotly permet notamment :

* d’explorer les quartiles de manière plus fine
* d’identifier précisément les valeurs aberrantes
* d’afficher les valeurs numériques associées à chaque point



#### Exemple avec Plotly

```python
import plotly.express as px

px.box(y=df.co2)
```

Ce boxplot interactif représente les mêmes éléments statistiques que la version Pandas :

* la médiane
* les quartiles
* l’intervalle interquartile
* les valeurs aberrantes



#### Apport de l’interactivité

Grâce à l’interactivité :

* le survol de la souris permet de lire les valeurs exactes
* les outliers peuvent être examinés individuellement
* la structure de la distribution est plus facile à analyser

Cette approche est particulièrement utile lors de l’exploration initiale des données, avant toute modélisation.


#### Lien avec l’analyse statistique

Le boxplot Plotly constitue une **représentation graphique directe** des statistiques fournies par `df.describe()` :

* les quartiles visibles correspondent aux valeurs 25 %, 50 % et 75 %
* les points isolés correspondent aux observations identifiées comme aberrantes selon le critère de l’intervalle interquartile

### Visualisation de la distribution avec Plotly

Plotly permet d’explorer la **distribution des données** à l’aide de graphiques interactifs. Deux visualisations sont particulièrement utiles : l’histogramme et le violin plot.


#### Histogramme interactif

```python
px.histogram(
    x=df.co2,
    nbins=100
)
```

L’histogramme représente la distribution des valeurs en les regroupant dans des **intervalles appelés bins**.

* Chaque bin correspond à une plage de valeurs
* La hauteur d’un bin indique le nombre d’observations dans cette plage
* Le paramètre `nbins` contrôle le nombre d’intervalles utilisés

Un nombre trop faible de bins masque les détails de la distribution, tandis qu’un nombre trop élevé peut introduire du bruit. Le choix du nombre de bins influence donc directement la lisibilité du graphique.



#### Violin plot

```python
px.violin(y=df.co2, box=True)
```

Le violin plot permet de visualiser la **distribution complète** des données.

* La largeur du violon représente la **densité** des valeurs
* Les zones larges indiquent une forte concentration de données
* Les zones étroites indiquent des valeurs plus rares

Lorsque l’option `box=True` est activée, le violin plot inclut :

* la médiane
* les quartiles
* les valeurs aberrantes

Le violin plot combine ainsi une représentation statistique (boxplot) et une représentation continue de la distribution.



#### Complémentarité des graphiques

* L’histogramme montre la distribution par classes discrètes (bins)
* Le violin plot montre la distribution de manière continue
* Les deux graphiques apportent des informations complémentaires

Ces visualisations facilitent l’analyse exploratoire des données avant la modélisation.

### Nuage de points (scatter plot) 

Le **nuage de points** (scatter plot) permet de visualiser la relation entre **deux variables numériques**.
C’est le graphique central pour analyser une relation potentielle et pour introduire la **régression linéaire**.



#### Principe du scatter plot

Chaque point du graphique représente une observation du dataset.

* L’axe horizontal correspond à une variable explicative ( x )
* L’axe vertical correspond à une variable cible ( y )
* Un point correspond à un couple ( (x_i, y_i) )

Le scatter plot permet d’observer :

* l’existence d’une relation entre les variables
* la forme de cette relation (linéaire ou non)
* la dispersion des données
* la présence de valeurs aberrantes



#### Exemple avec Plotly

```python
px.scatter(x=df.consumption, y=df.co2)
```

Ce graphique permet d’examiner visuellement la relation entre la consommation de carburant et la concentration de CO2.



## Interprétation du graphique

À partir du nuage de points, on peut analyser :

* **Tendance globale**
  Une tendance croissante ou décroissante suggère une relation entre les variables.

* **Dispersion**
  La distance des points par rapport à une tendance moyenne indique la variabilité des données.

* **Outliers**
  Les points très éloignés du reste du nuage peuvent correspondre à des observations atypiques.

## Objets en python 

### Construction d’une classe en Python

Une **classe** en Python permet de définir un type d’objet personnalisé en regroupant :

* des **données** appelées *attributs*
* des **fonctions** appelées *méthodes*

L’exemple suivant illustre les éléments fondamentaux d’une classe Python.

```python
# Définition d'une classe très simple pour illustrer attributs + méthodes
class Calculatrice:
    def __init__(self):
        # Attribut d'instance : chaque objet aura sa propre 'memoire'
        self.memoire = 10

    def add(self, x):
        # Méthode qui modifie l'état interne (la mémoire)
        self.memoire = self.memoire + x
```



#### 1. Définition de la classe

```python
class Calculatrice:
```

Le mot-clé `class` permet de définir une nouvelle classe.
`Calculatrice` est le nom du type d’objet qui pourra ensuite être instancié.



#### 2. Le constructeur `__init__`

```python
def __init__(self):
```

* `__init__` est le **constructeur** de la classe
* Il est appelé **automatiquement** lors de la création d’un objet
* Il sert à **initialiser les attributs** de l’objet

Dans cet exemple, le constructeur ne prend aucun paramètre supplémentaire, mais il pourrait en recevoir.



#### 3. Le paramètre `self`

* `self` représente **l’instance courante** de la classe
* Il est toujours le **premier paramètre** des méthodes d’instance
* Il permet d’accéder aux attributs et aux autres méthodes de l’objet

Sans `self`, Python ne saurait pas à quel objet se réfère le code.



#### 4. Les attributs d’instance

```python
self.memoire = 10
```

* `memoire` est un **attribut d’instance**
* Chaque objet `Calculatrice` possède **sa propre valeur** de `memoire`
* L’attribut est créé lors de l’appel du constructeur

Ainsi, deux objets distincts ont des mémoires indépendantes.



#### 5. Les méthodes de la classe

```python
def add(self, x):
```

* Une **méthode** est une fonction définie à l’intérieur d’une classe
* Elle agit sur l’état interne de l’objet via `self`
* Ici, la méthode `add` modifie l’attribut `memoire`

```python
self.memoire = self.memoire + x
```

Cette instruction met à jour l’état interne de l’objet.



#### 6. Création et utilisation d’un objet

```python
calc = Calculatrice()   # appel du constructeur
calc.add(5)             # appel d'une méthode
print(calc.memoire)     # affiche 15
```

* `Calculatrice()` crée une nouvelle instance
* `calc.add(5)` modifie la mémoire interne de cet objet
* Les méthodes sont appelées via la notation pointée


#### 7. Résumé

* Une classe définit la **structure et le comportement** des objets
* `__init__` initialise les attributs
* `self` fait référence à l’instance courante
* Les attributs stockent l’état
* Les méthodes modifient ou exploitent cet état


### Instanciation d’un objet en Python

L’**instanciation** correspond à la création concrète d’un objet à partir d’une classe.
Elle permet de passer d’une **définition abstraite** (la classe) à un **objet utilisable en mémoire**.

```python
# Instanciation : création d'un objet Calculatrice
c = Calculatrice()
```

#### 1. Appel de la classe

```python
Calculatrice()
```

* L’appel de la classe avec des parenthèses déclenche la **création d’un nouvel objet**
* Python réserve un espace mémoire pour cet objet
* Le constructeur `__init__` est alors appelé **automatiquement**



#### 2. Appel automatique du constructeur `__init__`

Lors de l’instruction :

```python
c = Calculatrice()
```

Python exécute implicitement :

```python
Calculatrice.__init__(c)
```

Dans notre exemple, le constructeur initialise l’attribut :

```python
self.memoire = 10
```

L’objet nouvellement créé possède donc immédiatement un état interne.



#### 3. Création de la variable de référence

```python
c = ...
```

* `c` est une **variable de référence**
* Elle pointe vers l’objet `Calculatrice` créé en mémoire
* L’objet existe indépendamment du nom de la variable qui le référence



#### 4. État initial de l’objet

Après l’instanciation :

```python
c.memoire
```

renvoie la valeur :

```text
10
```

Chaque nouvel objet instancié à partir de la classe `Calculatrice` possède sa **propre mémoire indépendante**.



#### 5. Distinction classe / objet

* **Classe** : description du comportement et de la structure (`Calculatrice`)
* **Objet** : instance concrète créée à partir de la classe (`c`)

Une même classe peut donner naissance à **autant d’objets que nécessaire**, chacun avec son propre état.



#### 6. Résumé

* L’instanciation crée un nouvel objet en mémoire
* Le constructeur `__init__` est appelé automatiquement
* Les attributs sont initialisés dès la création
* La variable (`c`) référence l’objet créé



### Accès aux attributs et appel de méthodes

Une fois l’objet instancié, il est possible :

* de **lire ses attributs**
* d’**appeler ses méthodes** pour modifier son état interne

Les instructions suivantes illustrent ces deux mécanismes.

```python
# Lecture de l'attribut memoire
c.memoire

# Appel d'une méthode : ajoute 20 à la mémoire
c.add(20)

# Vérifie la nouvelle valeur de la mémoire
c.memoire
```

#### 1. Lecture d’un attribut

```python
c.memoire
```

* L’attribut `memoire` est accédé via la **notation pointée**
* Python lit la valeur stockée dans l’objet référencé par `c`
* Aucune modification de l’état n’est effectuée

À ce stade, la valeur retournée est :

```text
10
```



#### 2. Appel d’une méthode

```python
c.add(20)
```

* `add` est une **méthode de la classe `Calculatrice`**
* L’appel se fait également via la notation pointée
* L’argument `20` est transmis au paramètre `x`
* Le paramètre `self` est automatiquement lié à l’objet `c`

L’instruction exécutée à l’intérieur de la méthode est :

```python
self.memoire = self.memoire + x
```

Ce qui modifie l’état interne de l’objet.



#### 3. Modification de l’état de l’objet

Après l’appel de la méthode :

* la valeur de `memoire` passe de `10` à `30`
* l’objet conserve cette nouvelle valeur tant qu’il existe en mémoire



#### 4. Nouvelle lecture de l’attribut

```python
c.memoire
```

Cette instruction permet de vérifier que l’état interne a bien été mis à jour.
La valeur retournée est désormais :

```text
30
```



#### 5. Point clé à retenir

* Les **attributs** stockent l’état de l’objet
* Les **méthodes** peuvent lire et modifier cet état
* Toute modification via une méthode est **persistante** pour l’objet concerné

### Rappel — Chaînes de caractères formatées (f-strings)

En Python, les **f-strings** permettent d’insérer directement des variables à l’intérieur d’une chaîne de caractères.

La lettre `f` placée devant la chaîne indique que les expressions entre accolades `{}` doivent être évaluées.



#### Insertion de variables

```python
year = 3
capital = 1234.567

print(f"Année {year} : capital = {capital}")
```

Résultat :

```
Année 3 : capital = 1234.567
```



#### Formatage des nombres

Il est possible de **contrôler l’affichage** des nombres.

```python
print(f"{capital:.2f}")
```

* `.2f` signifie :

  * `f` : nombre réel (float)
  * `.2` : deux chiffres après la virgule

Résultat :

```
1234.57
```

Le nombre est arrondi pour l’affichage.



#### Décomposition de l’instruction

```python
print(f"Année {year} : capital = {self.capital:.2f}")
```

* `f"..."` : chaîne formatée
* `{year}` : insertion de la valeur de la variable `year`
* `{self.capital:.2f}` : insertion du capital avec deux décimales
* `print` : affiche le texte dans la sortie du notebook
