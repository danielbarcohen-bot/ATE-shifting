### <div class="header3">5.2.1 | Outlet_Size</div>
<div class="explain-box">
    Before carrying out the imputation process, <b>an analysis will be carried out first</b> to make sure the imputation values are correct, avoid biased results, and provide a high-quality dataset. Based on the previous section analysis, <b>medium-sized outlets have the highest number</b> in distribution. <mark>However, imputation with a medium value will make the dataset biased since the medium size will dominate the outer outlet size</mark>. Due to this reason, the first step is to <mark>group outlet types and outlet sizes, and then calculate the total available outlets in the dataset</mark>.
</div>

## <div class="header2">5.3 | Handling Outliers ??</div>
<div class="explain-box">
    As mentioned in the previous section, <b>"Item_Visibility" and "Item_Outlet_Sales" have outliers in their distribution</b>. This section will focus on <mark>transforming those columns to eliminate the outliers using log and square root transformation</mark>. After the transformation is done, <mark>the results of the techniques will be compared first</mark> before being applied to the real dataset.
</div>
<!-- Hello world ??. Thank you so much for downloading/forking my codes/works. If you like my works, please support me by giving upvotes and comments on my Kaggle profile (https://www.kaggle.com/caesarmario/). Thank you so much and have a great day ????.
```python
# --- Creating Transformation DataFrame ---
df_transform = df[['Item_Visibility', 'Item_Outlet_Sales']].copy()

# --- Log Transformation ---
df_transform['Item_Visibility_log'] = np.log(df_transform['Item_Visibility'])
df_transform['Item_Outlet_Sales_log'] = np.log(df_transform['Item_Outlet_Sales'])

# --- Using Square Root Transformation
df_transform['Item_Visibility_sqrt'] = np.sqrt(df_transform['Item_Visibility'])
df_transform['Item_Outlet_Sales_sqrt'] = np.sqrt(df_transform['Item_Outlet_Sales'])
```

```python
# --- DataFrame ---
df_transform_skew = pd.DataFrame(df_transform.skew(axis=0, skipna=True)).T
df_transform_kurt = pd.DataFrame(df_transform.kurt(axis=0, skipna=True)).T

# --- Variables ---
ax_num_iv, ax_num_os = 0, 0
tick_params = dict(length=3, width=1, color=color_line)
suptitle = dict(x=0.5, y=0.94, fontsize=20, weight='heavy', ha='center', va='bottom', fontname=font_main)
title = dict(x=0.5, y=0.93, fontsize=15, weight='normal', ha='center', va='bottom', fontname=font_alt)
highlight_textprops = [{'weight':'bold', 'color': colors[1]}, {'weight':'bold'}, {'weight':'bold'}]

# --- Functions ---
def distribution_plot(ax, col_name):
    skew_value = df_transform_skew.iloc[0][col_name]
    kurt_value = df_transform_kurt.iloc[0][col_name]

    ax.xaxis.set_label_position('top')
    ax.set_title(f"\n{col_name}", fontsize=13, fontweight='bold')
    ax.set_xlabel('Skewness Value: {:.4f}\nKurtosis Value: {:.4f}'.format(skew_value, kurt_value), fontsize=10, fontname=font_alt)
    ax.set_ylabel('')
    ax.grid(axis='y', alpha=0, zorder=2)
    ax.grid(axis='x', which='major', alpha=0.3, color=color_grid, linestyle='dotted', zorder=1)
    for spine in ax.spines.values(): spine.set_color('None')
    ax.spines['bottom'].set_visible(True)
    ax.spines['bottom'].set_color(color_line)
    ax.tick_params(bottom='on', **tick_params)
    
# --- Plot Size ---
fig, ax = plt.subplots(3, 2, figsize=(19, 21))
for col, axes in zip(list(df_transform.columns), ax.flatten()):
    if col.find("Item_Visibility") == 0:
        eda_transform = sns.histplot(data=df_transform, x=col, kde=True, ax=axes, color=colors[3])
        distribution_plot(axes, col)
        ax_num_iv += 1
    else:
        eda_transform = sns.histplot(data=df_transform, x=col, kde=True, ax=axes, color=colors[-2])
        distribution_plot(axes, col)
        ax_num_os += 1
fig_text(s='Transformation Comparison', **suptitle)
fig_text(s='<Distribution Comparison after Transformation> on <"Item_Visibility"> and <"Item_Outlet_Sales">', highlight_textprops=highlight_textprops, **title)
plt.gcf().text(0.78, 0.07, 'kaggle.com/caesarmario', style='italic', fontsize=12)
fig.tight_layout()
plt.show();
```

<div class="explain-box">
    Based on histograms, skewness, and kurtosis values shown above, <mark>square root transformation is better than log transformation</mark> since <b>the histogram distributions of square root transformation are closer to normal distribution</b> even though outliers are still detected in both columns. In addition, the <b>skewness value showed that square root transformation can reduce skewness value effectively</b>. The next step will <b>apply square root transformation</b> for both columns in a real dataset.
