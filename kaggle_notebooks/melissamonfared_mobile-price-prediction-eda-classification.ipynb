**Introducing DataSet:**
- Bob has started his own mobile company. He wants to give tough fight to big companies like Apple,Samsung etc.
He does not know how to estimate price of mobiles his company creates. In this competitive mobile phone market you cannot simply assume things. To solve this problem he collects sales data of mobile phones of various companies.
Bob wants to find out some relation between features of a mobile phone(eg:- RAM,Internal Memory etc) and its selling price. But he is not so good at Machine Learning. So he needs your help to solve this problem.
**In this problem you do not have to predict actual price but a price range indicating how high the price is.**
- **This dataset contains 2000 rows and has information about the 21 features.**
- **There are no empty or NaN values in this dataset and all categorical data are represented with numbers.**

#### pixel resolution height and screen width of mobile in cm contains many values of zero or close to zero and they look like noise, 
- by searching on the internet, we can see the earliest commercially available mobile phones, like the Motorola DynaTAC 8000X, had a screen resolution of only 5 pixels in height and some of the newest mobile phones had a screen width of around 2-3 centimeters. therefore, we consider the value of 5 for px_height and the value of 2 for sc_w as the minimum acceptable values and detect noises but do not clean them until modeling with noise-sensitive models like decision tree.

```python
sc_w_Noise = df[df['sc_w'] < 2].index
px_height_Noise = df[df['px_height'] < 5].index
print('detected noise values of sc_w: ', len(sc_w_Noise))
print('detected noise values of px_height: ', len(px_height_Noise))
```

```python
corr = df.corr()
f, ax = plt.subplots(figsize=(12, 5))
mask = np.triu(np.ones_like(corr, dtype=bool))
cut_off = 0.25
extreme_1 = 0.5
extreme_2 = 0.75
extreme_3 = 0.9
mask |= np.abs(corr) < cut_off
corr = corr[~mask]
remove_empty_rows_and_cols = True
if remove_empty_rows_and_cols:
    wanted_cols = np.flatnonzero(np.count_nonzero(~mask, axis=1))
    wanted_rows = np.flatnonzero(np.count_nonzero(~mask, axis=0))
    corr = corr.iloc[wanted_cols, wanted_rows]

annot = [[f"{val:.4f}"
          + ('' if abs(val) < extreme_1 else '\n*')
          + ('' if abs(val) < extreme_2 else '*')
          + ('' if abs(val) < extreme_3 else '*')
          for val in row] for row in corr.to_numpy()]
heatmap = sns.heatmap(corr, vmin=-1, vmax=1, annot=annot, fmt='', cmap='GnBu')
heatmap.set_title('Triangle Correlation Heatmap', fontdict={'fontsize': 13}, pad=16)
plt.show()
```

### Normal distribution, also known as the Gaussian distribution, is a probability distribution that is symmetric about the mean, showing that data near the mean are more frequent in occurrence than data far from the mean.
### In graphical form, the normal distribution appears as a "bell curve".
### The normal distribution describes a symmetrical plot of data around its mean value, where the width of the curve is defined by the standard deviation. It is visually depicted as the "bell curve."

### Skewness measures the degree of symmetry of a distribution. The normal distribution is symmetric and has a skewness of zero. If the distribution of a data set instead has a skewness less than zero, or negative skewness (left-skewness), then the left tail of the distribution is longer than the right tail; positive skewness (right-skewness) implies that the right tail of the distribution is longer than the left.

SVM 🤖:
Support Vector Machine (SVM) is a widely-used machine learning algorithm utilized for classification and regression tasks. Its operation involves identifying the optimal hyperplane that separates data points into distinct classes. This hyperplane is positioned by maximizing the margin, which denotes the distance between the hyperplane and the nearest data points of each class.

Advantages of SVM classifier 🌟:
1. Robust to noise & outliers 🛡️ - SVM exhibits lower sensitivity to noise and outliers compared to other algorithms, rendering it suitable for tasks where the presence of noise and outliers is anticipated.
2. Good generalization 🧠 - SVM is recognized for its proficient generalization capabilities, allowing it to perform effectively on new and unseen data.
3. Handles non-linear data 🔄 - SVM can manage non-linear decision boundaries, proving valuable in scenarios where the data is not linearly separable.
4. Flexibility in kernel selection 🌐 - SVM offers versatility in choosing diverse kernel functions like linear, polynomial, and radial basis function (RBF), enabling it to function across a broad array of datasets.
5. Effective in high-dimensional spaces 🚀 - SVM can operate effectively with high-dimensional datasets, a feature beneficial in instances where the number of features surpasses the number of samples.

Disadvantages of SVM classifier 📉:
1. Computationally intensive 💻 - SVM demands significant computational resources, potentially resulting in slower performance and heightened memory requirements when handling voluminous datasets.
2. Sensitivity to hyperparameters ⚖️ - SVM's performance is notably influenced by the selection of parameters such as the kernel function and regularization parameters, impacting its overall efficacy.
3. Limited interpretability ❓ - SVM lacks explicit models for interpreting the relationship between features and the target variable, posing challenges in result interpretation.
4. Bias-Variance Tradeoff ⚖️ - Similar to all machine learning algorithms, SVM is subject to the bias-variance tradeoff, hence it can either overfit or underfit the data, leading to subpar performance.

🌳 Decision Tree: 
A Decision Tree is a supervised machine learning technique that employs a tree-like structure to represent decisions and their potential outcomes. It operates by repetitively dividing the data into subsets based on feature values to minimize a chosen criterion like Gini impurity or entropy.

Advantages of Decision Tree:
1. 👍 Easy to comprehend and articulate - Decision Trees are simple to understand and explain, even for intricate models.
2. 📊 Handles both numerical and categorical data types - Decision trees can manage both numerical and categorical data categories.
3. 🔄 Insensitive to nonlinear relationships - Decision trees can manage nonlinear relationships between variables without impacting performance.

Disadvantages of Decision Tree:
1. 🧠 Prone to overfitting - Decision trees are prone to overfitting, notably when trained on noisy or incomplete datasets.
2. 🔀 Unstable - Minor fluctuations in the data can lead to the creation of radically different decision trees.
3. 📉 Limited approximation of complex functions - Decision Trees might not be ideal for approximating complex functions due to their constraint by axis-parallel splits at each node.

Decision Tree Hyperparameter Tuning:
Optimizing hyperparameters is critical for decision trees as they are susceptible to overfitting. Overfitting happens when the tree is overly intricate and closely matches the training data, capturing even the noise present. Fine-tuning the hyperparameters can aid in preventing overfitting and enhancing the model's performance on new, unseen data.

Methods to Avoid Overfitting in Decision Trees:
1. Pruning - Eliminating branches from the tree that have minimal impact on classification.
2. Utilizing ensemble methods - Merging multiple decision trees to create a more resilient model.
3. Adjusting Hyperparameters - Tuning hyperparameters can prevent decision tree classifiers from overfitting by managing the tree's complexity.

Most Common Hyperparameters in Decision Trees:
1. Criterion: Determines the method used to assess split quality, with Gini Impurity and Information Gain being popular criteria.
2. Maximum depth: Controls the tree's deepest level. A deeper tree increases complexity, potentially leading to overfitting. Setting a maximum depth can mitigate overfitting by constraining tree size.
3. Minimum samples per split: Specifies the minimum samples required to split an internal node. If samples at a node are below this threshold, the node cannot split further, helping limit tree size and prevent overfitting.
4. Minimum samples per leaf: Establishes the minimum samples needed for a leaf node. If a leaf has fewer samples than this threshold, it can be removed.
5. Maximum features: Sets the maximum features to consider when splitting a node, reducing model complexity to prevent overfitting.

Pruning is a potent method employed in machine learning and search algorithms to decrease the complexity of decision trees by eliminating non-essential and duplicative parts. The primary objective of Pruning is to condense the information and streamline the ultimate classifier, leading to enhanced predictive precision and decreased overfitting. 

```python
clf = DecisionTreeClassifier() # decision tree classifier
path = clf.cost_complexity_pruning_path(X_train, y_train) # cost complexity pruning path for decision tree using training data
ccp_alphas = path.ccp_alphas
impurities = path.impurities
```

```python
fig, ax = plt.subplots(figsize=(10,5))
ax.plot(ccp_alphas[:-1], impurities[:-1], marker='o', color='#52796f', drawstyle="steps-post")
ax.set_xlabel("effective alpha")
ax.set_ylabel("total impurity of leaves")
ax.set_title("total Impurity vs. effective alpha for training set")
plt.show()
```

To prepare a pruned decision tree, we can adjust the ccp_alpha parameter in the DecisionTreeClassifier to specific alpha values. Subsequently, we can iterate through these alpha values, training a decision tree for each and assessing the accuracy of the pruned decision tree on both training and test datasets. By analyzing the accuracy results across various alpha values, we can choose the alpha value that offers the optimal trade-off between complexity and accuracy for our decision tree.

```python
clfs = []
accuracy_train = []
accuracy_test = []
recall_train = []
recall_test = []
# loop
for ccp_alpha in ccp_alphas:
# decision tree classifier object with given value of ccp_alpha
    clf = DecisionTreeClassifier(ccp_alpha=ccp_alpha)
    clf.fit(X_train, y_train) # fit classifier to training set
    y_train_pred = clf.predict(X_train) # predict
    y_test_pred = clf.predict(X_test)
    accuracy_train.append(clf.score(X_train, y_train))
    accuracy_test.append(clf.score(X_test, y_test))
    recall_train.append(metrics.recall_score(y_train, y_train_pred, average='micro'))
    recall_test.append(metrics.recall_score(y_test, y_test_pred, average='micro'))
    clfs.append(clf)
```

# <p style="background-color:#354f52; font-family:calibri; font-size:120%; color:white; text-align:center; border-radius:15px 50px; padding:10px">Random Forest</p>

🌲 Random Forest is a classification model that combines multiple decision trees to boost prediction accuracy. It selects random features and data points for each tree, then merges predictions through voting.

Advantages of Random Forest:
1. ✨ Less prone to overfitting than decision trees, especially with highly correlated features.
2. 🎯 High accuracy and adept at handling high-dimensional data.
3. 🔍 Offers insights into feature importance, aiding in feature selection and data comprehension.

Disadvantages of  Random Forest:
1. 💻 Can be computationally intensive and resource-demanding.
2. 🔎 Reduced model interpretability due to multiple decision trees.
3. 📉 May not perform well on small or imbalanced datasets.

Random Forest Hyperparameter Tuning:
Fine-tuning the hyperparameters of a Random Forest can enhance the model's accuracy and mitigate issues like overfitting or underfitting on the dataset. A Random Forest model comprises multiple decision trees, which can be categorized as either shallow (limited branches or levels) or deep (many branches/levels and incompletely developed). Deep trees typically exhibit low bias but high variance, making them suitable for bagging techniques that target variance reduction. When selecting hyperparameters for a Random Forest model, it's crucial to consider incorporating deep trees into the range of values for the hyperparameters.

Methods to Prevent Overfitting in Decision Trees:
Pruning - Eliminating less significant branches from the tree that contribute minimally to the classification.
Utilizing ensemble methods - Merging multiple decision trees to create a more resilient model.
Optimizing Hyperparameters - Tuning hyperparameters can help avert overfitting in decision tree classifiers by managing the tree's complexity.

Most Common Hyperparameters in Random Forest:
n_estimators: The quantity of trees in the forest.
criterion: The metric used to assess split quality, with common criteria being Gini impurity and information gain.
max_depth: The highest depth of a tree, utilized to regulate model complexity and prevent overfitting.
min_samples_split: The minimum number of samples required to split an internal node.
min_samples_leaf: The minimum number of samples needed to form a leaf node.
bootstrap: Deciding whether or not to sample with replacement during forest tree construction.
max_features: The maximum number of features to evaluate when splitting a node, with options including a numerical value, a percentage, 'sqrt,' or 'log2.'
