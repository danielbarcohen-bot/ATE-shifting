## 1. Vorbereitung

Zuerst laden wir wie üblich einen kleinen Datensatz. Außerdem definieren wir eine Funktion mit der wir später unsere Daten plotten werden. Führen Sie einfach die folgende Zelle aus, es sind keine Änderungen nötig.

Es sollte ein Plot einer seltsame geschwungenen Punktewolke erscheinen.

```python
import pandas
import matplotlib.pyplot as plt
import numpy as np

def plot(x, y, a=None, b=None, c=None, d=None, e=None):
    """ 
          ! DO NOT TOUCH !

          This method plots a 2D point cloud. Optionally, 
          if you specify a/b/c/d/e, the method also
          plots the following fitted models:
          
          o OPTION 1:
            When passing a and b, the line
            y = a * x + b is plotted.
            
          o OPTION 2:
            When passing c, the model
            y = sin(x) + c * x is plotted.
            
          o OPTION 3:
            When passing d and e, the model
            y = sin(x) + d * x + e is plotted.
    """
    plt.plot(x, y, 'o', color='red', markersize=3)
    plt.axis([np.min(x),np.max(x),-3,13])
    xticks = np.arange(np.min(x), np.max(x), 0.01)
    if a is not None and b is not None:
        plt.plot(xticks, a*xticks+b, '-', color='blue')
    if c is not None:
        plt.plot(xticks, np.sin(xticks) + c*xticks, '-', color='green')
    if d is not None and e is not None:
        plt.plot(xticks, np.sin(xticks) + d*xticks + e, '-', color='orange')
    plt.grid(True)
    plt.show()
    
def read_data(path):
    dataframe = pandas.read_csv(path, sep=';', header=None)
    data = dataframe.values.astype('float')
    return data[:,0], data[:,1]
    
x,y = read_data('/kaggle/input/data-leastsquares/data_leastsquares.csv') 
plot(x,y)
```
