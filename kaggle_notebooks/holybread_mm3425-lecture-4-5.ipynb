## **Demonstration 6: Advanced data table formatting**

Use the function pandas.DataFrame.applymap() to apply style changes element-wise  
https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.applymap.html

```python
# Define the function for selecting negative values
def style_negative(v, props=''):
    return props if v < 0 else None

# Use slicer to select the column
slice_ = ['Calculated Change']

# Format numeric values and then highlight negative values in red
df_style = car_df.style.format({'Calculated Change':'{:.2%}'})\
                       .applymap(func=style_negative, props='color:red;', subset=slice_)

# Display the formatted data table
display(df_style)
```

### **Demonstration 10.5: Customize the calculation of descriptive statistics**

Explain Kurtosis and Skewness
https://medium.com/@atanudan/kurtosis-skew-function-in-pandas-aa63d72e20de
* If the skewness is between -0.5 and 0.5, the data are fairly symmetrical
* If the skewness is between -1 and — 0.5 or between 0.5 and 1, the data are moderately skewed
* If the skewness is less than -1 or greater than 1, the data are highly skewed

* If the distribution is tall and thin it is called a leptokurtic distribution (Kurtosis > 3)
* A flat distribution where the values are moderately spread out (i.e., unlike leptokurtic) is called platykurtic (Kurtosis <3) distribution
* A distribution which has a shape between a leptokurtic distribution and a platykurtic distribution is called a mesokurtic (Kurtosis=3) distribution, or a normal distribution

* High kurtosis in a data set is an indicator that data has heavy outliers
* Low kurtosis in a data set is an indicator that data has lack of outliers
* A positive kurtosis value means pointy and a negative one means flat

```python
car_df.agg(
    {
        'Sales (March 2010)': ['sum', 'mad', 'median', 'var', 'skew', 'kurt'],
        'Sales (March 2011)': ['sum', 'mad', 'median', 'var', 'skew', 'kurt']
    }
)
```

## **Demonstration 15: Handle Missing Values**

```python
# Generate two data columns, one is an incremental sequence from 0 to 9, the other is a random sequence
data = {'Increment':[i for i in range(10)], 'Random':np.random.randn(10)}
missing_value_df = pd.DataFrame(data)

# Make use of np.nan to add missing values in the dataframe
missing_value_df.loc[2:4, :] = np.nan
missing_value_df
```

### **Demonstration 15.1: Check missing values in a dataframe**

Use function pandas.DataFrame.isna() to check misisng values, often represented by NaN, in a dataframe https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.isna.html

```python
# Check missing values in a datafram
missing_value_df.isna()
```

```python
# Use isna() to single out missing values
missing_value_df.loc[(missing_value_df['Increment'].isna()),]
```

### **Demonstration 15.2: Drop missing data in a dataframe**

Use function pandas.DataFrame.drop() drop rows/columns with missing values in a dataframe https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.dropna.html

```python
# Set axis to 0 to drop rows that contain missing values
missing_value_df.dropna(axis = 0)
```

```python
# Set axis to 1 to drop columns that contain missing values
missing_value_df.dropna(axis = 1)
```

### **Demonstration 15.3: Fill missing data in a dataframe**

Use function pandas.DataFrame.fillna() to fill missing values in a dataframe https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.fillna.html

```python
# Fill missing values with a constant number 0
missing_value_df.fillna(0)
```

```python
# Fill missing values with the mean of the column
missing_value_df.fillna(missing_value_df.mean())
```
