```python
# applymap() example
def add_little(v):
    return v+0.0001

sale_df = car_df.loc[:,'Sales (March 2011)':'Sales (March 2010)']

sale_df.applymap(func=add_little)
```

```python
# another applymap() example

def change_red(v, props=''):
    return props

sale_df = car_df.loc[:,'Sales (March 2011)':'Sales (March 2010)']

sale_df.style.format().applymap(func=change_red, props='color:red')
```

```python
# another applymap() example

def change_red(v, props=''):
    return props if v > 30000 else None

sale_df = car_df.loc[:,'Sales (March 2011)':'Sales (March 2010)']

sale_df.style.format().applymap(func=change_red, props='color:red', subset=['Sales (March 2010)'])
```

```python
# Define the function for selecting negative values
def style_negative(v, props=''):
    return props if v < 0 else None

# Use slicer to select the column
slice_ = ['Calculated Change']

# Format numeric values and then highlight negative values in red
df_style = car_df.style.format()\
            .applymap(func=style_negative, props='color:red;', subset=slice_)
   
# Display the formatted data table
df_style
```

```python
# Version 2 of style_negative

def style_negative_2(v):
    if v < 0:
        return 'color:red'
    
    else:
        return 'color:blue'
    
    
df_style = car_df.style.format({'Calculated Change':'{:.2%}'})\
            .applymap(func=style_negative_2, subset=['Calculated Change'])

df_style
```

### **Demonstration 10.5: Customize the calculation of descriptive statistics**

Check the full list of descriptive statistics available via the following link

Explain Kurtosis and Skewness
https://medium.com/@atanudan/kurtosis-skew-function-in-pandas-aa63d72e20de
* If the skewness is between -0.5 and 0.5, the data are fairly symmetrical
* If the skewness is between -1 and — 0.5 or between 0.5 and 1, the data are moderately skewed
* If the skewness is less than -1 or greater than 1, the data are highly skewed

* If the distribution is tall and thin it is called a leptokurtic distribution (Kurtosis > 3)
* A flat distribution where the values are moderately spread out (i.e., unlike leptokurtic) is called platykurtic (Kurtosis <3) distribution
* A distribution which has a shape between a leptokurtic distribution and a platykurtic distribution is called a mesokurtic (Kurtosis=3) distribution, or a normal distribution

* High kurtosis in a data set is an indicator that data has heavy outliers
* Low kurtosis in a data set is an indicator that data has lack of outliers
* A positive kurtosis value means pointy and a negative one means flat

```python
car_df.agg(
    {
        'Sales (March 2010)': ['sum', 'mad', 'median', 'var', 'skew', 'kurt'],
        'Sales (March 2011)': ['sum', 'mad', 'median', 'var', 'skew', 'kurt']
    }
)
```
