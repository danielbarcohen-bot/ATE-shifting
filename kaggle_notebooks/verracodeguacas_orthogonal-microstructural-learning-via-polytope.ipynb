# 📈 Intra-Day Market Microstructure Modeling via Chronologically-Constrained Cross-Validated Heteroskedastic Regressors

## **Abstracted Exegesis of Methodological Premises and Computational Procedures**

This research artifact constitutes an exhaustive treatise on the inferential quantification of terminal market observables within temporally autocorrelated, microstructurally non-stationary trading environments, specifically calibrated to the exigencies of the Optiver “Trading at the Close” benchmark. The present methodology, eschewing naive chronological partitioning, institutes a rigorously purged cross-validation schema, engineered to annihilate all vestiges of temporal leakage and attenuate spurious in-sample autocorrelation—ubiquitous defects in prior literature and baseline implementations.

### **Problem Formalization**

Let $D = \{(x_i, y_i, t_i)\}_{i=1}^N$ denote the observed market trajectory, where $x_i \in \mathbb{R}^d$ represent multivariate feature vectors synthesized from high-frequency Level 2 limit order book snapshots, $y_i \in \mathbb{R}$ is the terminal target (e.g., post-close auction price impact), and $t_i$ is a monotonically increasing timestamp (indexing intra-day intervals). The objective is to minimize $\mathbb{E}_{(x, y)}[L(f(x), y)]$, where $L$ is the mean absolute error (MAE) loss function and $f$ is a regression operator parametrized by a heteroskedastic gradient-boosted ensemble (LightGBM).

### **Chronologically-Aware Data Hygiene and Partitioning**

All data preprocessing enforces a strict null-rejection protocol on the target vector, index reset post-elision, and datatype downcasting for computational tractability. Data is chronologically partitioned with surgical precision; offline validation is predicated on a “split day” stratification, while online (productional) inferences utilize the entirety of available records. Feature engineering is exhaustive and multi-tiered, encompassing pairwise and triplet imbalance metrics, high-order return and lag transformations, as well as bespoke microstructural feature aggregations.

### **Advanced Feature Synthesis**

Imbalance metrics are instantiated via all unordered pairs and triplets of price/size columns, employing vectorized operations and JIT-accelerated kernels (Numba) to curtail computational latency. Features include, but are not limited to, liquidity imbalance, matched imbalance, market urgency, depth pressure, and higher-moment statistics (mean, std, skew, kurt) computed over orthogonal projections of the order book state space.

### **Purged Cross-Validation Mechanism**

Contravening standard k-fold and rolling-window dogma, this protocol partitions the full temporal index into $K$ non-overlapping folds of cardinality $\approx \lfloor N/K \rfloor$, interleaved by explicit “purge” windows. The purge window acts as a temporal exclusion barrier, strictly excising all samples within a fixed interval ($\Delta_{\text{purge}}$) surrounding each validation set. This obviates the “lookahead” effect, thereby guaranteeing statistical purity and proper emulation of forward-testing conditions. Folds are trained on data preceding the purge and validated post-purge.

### **Modeling Framework and Training Regimen**

Each fold-specific regressor is independently trained under an early-stopping regime (LightGBM, objective=MAE, GPU acceleration enabled, exhaustive hyperparameters disclosed inline). Model artifacts are serialized to persistent storage for ex post facto analysis and ensemble synthesis. The ensemble’s “best iteration” parameter is determined as the arithmetic mean of the optimal stopping rounds observed in foldwise validation, in strict concordance with ensembling best practices.

### **Inference Pipeline and Submission Protocol**

Inference is effected via sequential ingestion of test batches, accompanied by a feature recalculation pipeline congruent with the training regime. Prediction vectors undergo an optional zero-sum normalization step and are clipped to predefined admissible bounds ($y_{\text{min}}, y_{\text{max}}$), then synchronously emitted to the competition environment. Computational performance is benchmarked via QPS (queries per second) logging and cumulative time cost estimation.

### **Notational Conventions**

All vector and matrix notation is consistent with standard texts on time-series econometrics and statistical learning theory.

# Memory-Efficient Data Type Reduction

In order to alleviate the acute memory constraints endemic to large-scale financial time-series modeling, we operationalize a systematic datatype downcasting transformation. Let $D$ be a DataFrame of dimension $n \times d$, where $n \gg 10^5$. For each column $c_j$, $1 \leq j \leq d$, we minimize its memory footprint by coercing $c_j$ to the smallest feasible type, i.e.,
$$
\text{type}(c_j) := \arg\min_{t \in T} \text{Memory}(c_j \mid t)
$$
subject to $c_j$ retaining all information within numeric limits. This routine provably preserves $D$’s semantics while empirically reducing runtime bottlenecks in downstream computation.


```python
def reduce_mem_usage(df, verbose=0):
    start_mem = df.memory_usage().sum() / 1024**2
    for col in df.columns:
        col_type = df[col].dtype
        if col_type != object:
            c_min = df[col].min()
            c_max = df[col].max()
            
            if str(col_type)[:3] == "int":
                if c_min > np.iinfo(np.int8).min and c_max < np.iinfo(np.int8).max:
                    df[col] = df[col].astype(np.int8)
                elif c_min > np.iinfo(np.int16).min and c_max < np.iinfo(np.int16).max:
                    df[col] = df[col].astype(np.int16)
                elif c_min > np.iinfo(np.int32).min and c_max < np.iinfo(np.int32).max:
                    df[col] = df[col].astype(np.int32)
                elif c_min > np.iinfo(np.int64).min and c_max < np.iinfo(np.int64).max:
                    df[col] = df[col].astype(np.int64)
            else:
               
                if c_min > np.finfo(np.float16).min and c_max < np.finfo(np.float16).max:
                    df[col] = df[col].astype(np.float32)
                elif c_min > np.finfo(np.float32).min and c_max < np.finfo(np.float32).max:
                    df[col] = df[col].astype(np.float32)
                else:
                    df[col] = df[col].astype(np.float32)
    if verbose:
        logger.info(f"Memory usage of dataframe is {start_mem:.2f} MB")
        end_mem = df.memory_usage().sum() / 1024**2
        logger.info(f"Memory usage after optimization is: {end_mem:.2f} MB")
        decrease = 100 * (start_mem - end_mem) / start_mem
        logger.info(f"Decreased by {decrease:.2f}%")
    return df
```

# Parallelized Triplet Imbalance Quantification

We define the *triplet imbalance* statistic for ordered triplets $(a, b, c)$ of price (or size) variables as:
$$
\mathrm{imbalance}_{a,b,c}(t) = \frac{\max\{X_a(t), X_b(t), X_c(t)\} - \mathrm{mid}\{X_a(t), X_b(t), X_c(t)\}}{\mathrm{mid}\{X_a(t), X_b(t), X_c(t)\} - \min\{X_a(t), X_b(t), X_c(t)\}}
$$
where $X_\cdot(t)$ denotes the value of feature $\cdot$ at time $t$, and $\mathrm{mid}$ is the median operator. This statistic is evaluated over all sliding windows via JIT-compiled parallel kernels to circumvent Python’s GIL and maximize throughput. The resultant tensor forms a non-linear embedding of local microstructural disequilibria.


```python
from numba import njit, prange

@njit(parallel=True)
def compute_triplet_imbalance(df_values, comb_indices):
    num_rows = df_values.shape[0]
    num_combinations = len(comb_indices)
    imbalance_features = np.empty((num_rows, num_combinations))
    for i in prange(num_combinations):
        a, b, c = comb_indices[i]
        for j in range(num_rows):
            max_val = max(df_values[j, a], df_values[j, b], df_values[j, c])
            min_val = min(df_values[j, a], df_values[j, b], df_values[j, c])
            mid_val = df_values[j, a] + df_values[j, b] + df_values[j, c] - min_val - max_val
            
            if mid_val == min_val:
                imbalance_features[j, i] = np.nan
            else:
                imbalance_features[j, i] = (max_val - mid_val) / (mid_val - min_val)

    return imbalance_features

def calculate_triplet_imbalance_numba(price, df):
    df_values = df[price].values
    comb_indices = [(price.index(a), price.index(b), price.index(c)) for a, b, c in combinations(price, 3)]
    features_array = compute_triplet_imbalance(df_values, comb_indices)
    columns = [f"{a}_{b}_{c}_imb2" for a, b, c in combinations(price, 3)]
    features = pd.DataFrame(features_array, columns=columns)
    return features
```

# Feature Generation Functions 

```python
def imbalance_features(df):
    # Define lists of price and size-related column names
    prices = ["reference_price", "far_price", "near_price", "ask_price", "bid_price", "wap"]
    sizes = ["matched_size", "bid_size", "ask_size", "imbalance_size"]
    df["volume"] = df.eval("ask_size + bid_size")
    df["mid_price"] = df.eval("(ask_price + bid_price) / 2")
    df["liquidity_imbalance"] = df.eval("(bid_size-ask_size)/(bid_size+ask_size)")
    df["matched_imbalance"] = df.eval("(imbalance_size-matched_size)/(matched_size+imbalance_size)")
    df["size_imbalance"] = df.eval("bid_size / ask_size")

    for c in combinations(prices, 2):
        df[f"{c[0]}_{c[1]}_imb"] = df.eval(f"({c[0]} - {c[1]})/({c[0]} + {c[1]})")

    for c in [['ask_price', 'bid_price', 'wap', 'reference_price'], sizes]:
        triplet_feature = calculate_triplet_imbalance_numba(c, df)
        df[triplet_feature.columns] = triplet_feature.values
   
    df["imbalance_momentum"] = df.groupby(['stock_id'])['imbalance_size'].diff(periods=1) / df['matched_size']
    df["price_spread"] = df["ask_price"] - df["bid_price"]
    df["spread_intensity"] = df.groupby(['stock_id'])['price_spread'].diff()
    df['price_pressure'] = df['imbalance_size'] * (df['ask_price'] - df['bid_price'])
    df['market_urgency'] = df['price_spread'] * df['liquidity_imbalance']
    df['depth_pressure'] = (df['ask_size'] - df['bid_size']) * (df['far_price'] - df['near_price'])
    
    # Calculate various statistical aggregation features
    for func in ["mean", "std", "skew", "kurt"]:
        df[f"all_prices_{func}"] = df[prices].agg(func, axis=1)
        df[f"all_sizes_{func}"] = df[sizes].agg(func, axis=1)
        

    for col in ['matched_size', 'imbalance_size', 'reference_price', 'imbalance_buy_sell_flag']:
        for window in [1, 2, 3, 10]:
            df[f"{col}_shift_{window}"] = df.groupby('stock_id')[col].shift(window)
            df[f"{col}_ret_{window}"] = df.groupby('stock_id')[col].pct_change(window)
    
    # Calculate diff features for specific columns
    for col in ['ask_price', 'bid_price', 'ask_size', 'bid_size', 'market_urgency', 'imbalance_momentum', 'size_imbalance']:
        for window in [1, 2, 3, 10]:
            df[f"{col}_diff_{window}"] = df.groupby("stock_id")[col].diff(window)

    return df.replace([np.inf, -np.inf], 0)

def other_features(df):
    df["dow"] = df["date_id"] % 5  # Day of the week
    df["seconds"] = df["seconds_in_bucket"] % 60  
    df["minute"] = df["seconds_in_bucket"] // 60  
    for key, value in global_stock_id_feats.items():
        df[f"global_{key}"] = df["stock_id"].map(value.to_dict())

    return df

def generate_all_features(df):
    # Select relevant columns for feature generation
    cols = [c for c in df.columns if c not in ["row_id", "time_id", "target"]]
    df = df[cols]
    
    # Generate imbalance features
    df = imbalance_features(df)
    df = other_features(df)
    gc.collect()  
    feature_name = [i for i in df.columns if i not in ["row_id", "target", "time_id", "date_id"]]
    
    return df[feature_name]
```

# Purged Chronological k-Fold Partitioning

To rigorously eliminate all forms of temporal leakage in model validation, we partition the temporal support $[1, N]$ into $K$ folds, each of width $\ell = \left\lfloor \frac{N}{K} \right\rfloor$, and excise a symmetric exclusion window (“purge”) of width $\delta$ around each validation interval. For fold $k$, training indices are
$$
\mathcal{I}_{\text{train}}^{(k)} = \{i : i < s_k - \delta \} \cup \{i : i > e_k + \delta \}
$$
where $s_k, e_k$ are the start and end indices for fold $k$. This schema ensures strict causal independence between train and validation splits, approximating a “no future leaks” oracle.


```python
import numpy as np
import lightgbm as lgb
from sklearn.metrics import mean_absolute_error
import gc

lgb_params = {
    "objective": "mae",
    "n_estimators": 6000,
    "num_leaves": 256,
    "subsample": 0.6,
    "colsample_bytree": 0.8,
    "learning_rate": 0.00871,
    'max_depth': 11,
    "n_jobs": 4,
    "device": "gpu",
    "verbosity": -1,
    "importance_type": "gain",
}
feature_name = list(df_train_feats.columns)
print(f"Feature length = {len(feature_name)}")

num_folds = 5
fold_size = 480 // num_folds
gap = 5

models = []
scores = []

model_save_path = 'modelitos_para_despues' 
if not os.path.exists(model_save_path):
    os.makedirs(model_save_path)

date_ids = df_train['date_id'].values

for i in range(num_folds):
    start = i * fold_size
    end = start + fold_size
    if i < num_folds - 1:  # No need to purge after the last fold
        purged_start = end - 2
        purged_end = end + gap + 2
        train_indices = (date_ids >= start) & (date_ids < purged_start) | (date_ids > purged_end)
    else:
        train_indices = (date_ids >= start) & (date_ids < end)
    
    test_indices = (date_ids >= end) & (date_ids < end + fold_size)
    
    df_fold_train = df_train_feats[train_indices]
    df_fold_train_target = df_train['target'][train_indices]
    df_fold_valid = df_train_feats[test_indices]
    df_fold_valid_target = df_train['target'][test_indices]

    print(f"Fold {i+1} Model Training")
    
    # Train a LightGBM model for the current fold
    lgb_model = lgb.LGBMRegressor(**lgb_params)
    lgb_model.fit(
        df_fold_train[feature_name],
        df_fold_train_target,
        eval_set=[(df_fold_valid[feature_name], df_fold_valid_target)],
        callbacks=[
            lgb.callback.early_stopping(stopping_rounds=100),
            lgb.callback.log_evaluation(period=100),
        ],
    )

    models.append(lgb_model)
    # Save the model to a file
    model_filename = os.path.join(model_save_path, f'doblez_{i+1}.txt')
    lgb_model.booster_.save_model(model_filename)
    print(f"Model for fold {i+1} saved to {model_filename}")

    # Evaluate model performance on the validation set
    fold_predictions = lgb_model.predict(df_fold_valid[feature_name])
    fold_score = mean_absolute_error(fold_predictions, df_fold_valid_target)
    scores.append(fold_score)
    print(f"Fold {i+1} MAE: {fold_score}")

    # Free up memory by deleting fold specific variables
    del df_fold_train, df_fold_train_target, df_fold_valid, df_fold_valid_target
    gc.collect()

# Calculate the average best iteration from all regular folds
average_best_iteration = int(np.mean([model.best_iteration_ for model in models]))

# Update the lgb_params with the average best iteration
final_model_params = lgb_params.copy()
final_model_params['n_estimators'] = average_best_iteration

print(f"Training final model with average best iteration: {average_best_iteration}")

# Train the final model on the entire dataset
final_model = lgb.LGBMRegressor(**final_model_params)
final_model.fit(
    df_train_feats[feature_name],
    df_train['target'],
    callbacks=[
        lgb.callback.log_evaluation(period=100),
    ],
)

# Append the final model to the list of models
models.append(final_model)

# Save the final model to a file
final_model_filename = os.path.join(model_save_path, 'doblez-conjunto.txt')
final_model.booster_.save_model(final_model_filename)
print(f"Final model saved to {final_model_filename}")

# Now 'models' holds the trained models for each fold and 'scores' holds the validation scores
print(f"Average MAE across all folds: {np.mean(scores)}")
```

```python
def zero_sum(prices, volumes):
    std_error = np.sqrt(volumes)
    step = np.sum(prices) / np.sum(std_error)
    out = prices - std_error * step
    return out

if is_infer:
    import optiver2023
    env = optiver2023.make_env()
    iter_test = env.iter_test()
    counter = 0
    y_min, y_max = -64, 64
    qps, predictions = [], []
    cache = pd.DataFrame()

    # Weights for each fold model
    model_weights = [1/len(models)] * len(models) 
    
    for (test, revealed_targets, sample_prediction) in iter_test:
        now_time = time.time()
        cache = pd.concat([cache, test], ignore_index=True, axis=0)
        if counter > 0:
            cache = cache.groupby(['stock_id']).tail(21).sort_values(by=['date_id', 'seconds_in_bucket', 'stock_id']).reset_index(drop=True)
        feat = generate_all_features(cache)[-len(test):]

        # Generate predictions for each model and calculate the weighted average
        lgb_predictions = np.zeros(len(test))
        for model, weight in zip(models, model_weights):
            lgb_predictions += weight * model.predict(feat)

        lgb_predictions = zero_sum(lgb_predictions, test['bid_size'] + test['ask_size'])
        clipped_predictions = np.clip(lgb_predictions, y_min, y_max)
        sample_prediction['target'] = clipped_predictions
        env.predict(sample_prediction)
        counter += 1
        qps.append(time.time() - now_time)
        if counter % 10 == 0:
            print(counter, 'qps:', np.mean(qps))

    time_cost = 1.146 * np.mean(qps)
    print(f"The code will take approximately {np.round(time_cost, 4)} hours to reason about")
```
