<blockquote style="font-size: 12px; color: #000000;">
    ?? <mark><b>Heteroscedasticity</b></mark> is a concept that describes to <b>circumstances in which the residual variance is <mark>uneven</mark> throughout a spectrum of measured data</b>. A fan or cone form on a plot of the residuals suggests the existence of heteroscedasticity (as the variable values rise, the variance of the residuals increases proportionately). Heteroscedasticity is considered a concern in statistics because regressions using ordinary least squares (OLS) presume that the residuals are generated from a population with constant variance.
    </blockquote>
</div>
