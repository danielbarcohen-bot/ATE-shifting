```python
import pandas
import numpy
```

```python
# Đọc dữ liệu
data = pandas.read_csv('/kaggle/input/nesarc-huit/nesarc.csv', low_memory=False)
```

```python
# Xem xét kích thước dữ liệu
data.shape
```

```python
selected_columns = data[['TAB12MDX', 'CHECK321', 'S2AQ8A', 'S3AQ3C1']]
# Xem thông tin của các cột được chọn
selected_columns.info()
```

```python
# Chuyển sang dạng số
data['TAB12MDX'] = pandas.to_numeric(data['TAB12MDX'], errors='coerce')
data['CHECK321'] = pandas.to_numeric(data['CHECK321'], errors='coerce')
data['S2AQ8A'] = pandas.to_numeric(data['S2AQ8A'], errors='coerce')
data['S3AQ3C1'] = pandas.to_numeric(data['S3AQ3C1'], errors='coerce')
```

```python
selected_columns = data[['TAB12MDX', 'CHECK321', 'S2AQ8A', 'S3AQ3C1']]
selected_columns.info()
```

```python
# Tinh chỉnh câu hỏi nghiên cứu
sub1 = data[(data['AGE']>=18) & (data['AGE']<=25) & (data['CHECK321']==1)]
sub2 = sub1.copy()
```

```python
selected_columns = sub2[['AGE', 'TAB12MDX', 'S2AQ8A', 'S3AQ3C1']]
selected_columns.shape
```

```python
# Đếm cho thuộc tính S3AQ3B1 ban đầu
c1 = sub2['S3AQ3B1'].value_counts(dropna=False).sort_index()
print(c1)
```

```python
# Loại bỏ các dữ liệu thiếu
sub2['S3AQ3B1']=sub2['S3AQ3B1'].replace(9, numpy.nan)
sub2['S3AQ3C1']=sub2['S3AQ3C1'].replace(99, numpy.nan)
```

```python
# Đếm cho S3AQ3B1 với 9 được thay bằng NAN
c2 = sub2['S3AQ3B1'].value_counts(dropna=False).sort_index()
print(c2)
```

```python
# Mã hóa lại để trở thành định lượng, số ngày hút thuốc trong 1 tháng
recode = {1: 30, 2: 22, 3: 14, 4: 5, 5: 2.5, 6: 1}
sub2['USFREQMO']= sub2['S3AQ3B1'].map(recode)
c5 = sub2['USFREQMO'].value_counts(dropna=True).sort_index()
p5 = sub2['USFREQMO'].value_counts(dropna=True, normalize=True).sort_index()
print(c5)
print(p5)
```

```python
# Biến phụ: số điếu thuốc hút trong 1 tháng
sub2['NUMCIGMO_EST']=sub2['USFREQMO'] * sub2['S3AQ3C1']
sub3 = sub2[['IDNUM', 'S3AQ3C1', 'USFREQMO', 'NUMCIGMO_EST', 'TAB12MDX']]
sub3.head(25)
```

```python
# Khảo sát phân bố tuổi
print('đếm cho AGE')
c6 = sub2['AGE'].value_counts().sort_index()
print(c6)

print('phần trăm cho AGE')
p6 = sub2['AGE'].value_counts(normalize=True).sort_index()
print(p6)
```

```python
sub2['AGEGROUP3'] = pandas.cut(sub2.AGE, bins=[17, 20, 22, 25], 
                               labels=['Nhóm 18 - 20', 'Nhóm 21 - 22', 'Nhóm 23 - 25'])
c8 = sub2['AGEGROUP3'].value_counts(dropna=True).sort_index()
print(c8)
```

```python
pandas.crosstab(sub2['AGEGROUP3'], sub2['AGE'])
```

```python
# Biểu đồ tần số
import matplotlib.pyplot as plt

value_counts = sub2['AGEGROUP3'].value_counts()

plt.figure(figsize=(10, 6))
value_counts.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Biểu đồ tần số theo nhóm tuổi')
plt.xlabel('Nhóm tuổi')
plt.ylabel('Tần số')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```

```python
# Giả sử bạn có một DataFrame df với biến phân loại 'biến_cần_vẽ'
value_counts = sub2['AGEGROUP3'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(value_counts, labels=value_counts.index, autopct='%1.1f%%', colors=plt.cm.Paired(range(len(value_counts))))
plt.title('Biểu đồ tròn theo nhóm tuổi')
plt.show()
```
