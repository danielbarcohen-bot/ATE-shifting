# 4.Cleansing colom Hari

mengubah nama hari dari singkatan menjadi full nama

```python
# Mapping singkatan hari ke nama lengkap
day_mapping = {
    "Sun": "Sunday",
    "Sat": "Saturday",
    "Thur": "Thursday",
    "Fri": "Friday"
}


```

```python
# 4. Ubah kolom 'day'
df['day'] = df['day'].replace(day_mapping)
```

Hasil Data colom Hari setelah Di cleansing

```python
print(df.head(10))
```

# 5. Cleansing Data Colom Total Bill

```python
# Urutkan berdasarkan total_bill dari yang termurah
df_sorted = df.sort_values(by='total_bill', ascending=True)
```

```python
# Tampilkan hasil urutan
print(df_sorted.head(20))
```

# Kesimpulan



1.   Cleansing Data Hari


*   Kolom day yang sebelumnya masih menggunakan singkatan (Sun, Sat, Fri, Thur) berhasil diubah menjadi nama lengkap (Sunday, Saturday, Friday, Thursday).
*   Hal ini membuat data lebih jelas, mudah dipahami, dan konsisten untuk analisis lebih lanjut.


2.   Pengurutan Data

*   Dataset berhasil diurutkan berdasarkan kolom total_bill dari yang termurah ke termahal.
