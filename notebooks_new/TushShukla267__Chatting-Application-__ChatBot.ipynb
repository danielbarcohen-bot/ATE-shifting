```python
# !Install the Google AI Python SDK

%pip install google-generativeai

import os
import google.generativeai as genai
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

# Get the API key from the environment
api_key = os.getenv('GEMINI_API_KEY')

# Check if the API key is available
if not api_key:
    raise ValueError("API key for Gemini is missing! Make sure it's defined in the .env file.")

# Configure the API key for Google Generative AI
genai.configure(api_key=api_key)

# Create the model
generation_config = {
    "temperature": 1,
    "top_p": 0.95,
    "top_k": 64,
    "max_output_tokens": 8192,
    "response_mime_type": "text/plain",
}

# Instantiate the generative model
model = genai.GenerativeModel(
    model_name="gemini-1.5-flash",
    generation_config=generation_config,
  # safety_settings = Adjust safety settings
  # See https://ai.google.dev/gemini-api/docs/safety-settings
)

# Start a chat session
chat_session = model.start_chat(
    history=[],
)

# Send a message and get the response
response = chat_session.send_message("Tell me about the Sun")

# Print the response text
print(response.text)
```

```python
import nltk
from newspaper import Article

# Download the punkt tokenizer
nltk.download('punkt')

# Get the article
article = Article('https://en.wikipedia.org/wiki/Solar_System')

# Download the article
article.download()

# Parse the article
article.parse()

# Get the text from the article
corpus = article.text

```

```python
#printing the artical text
print(corpus)
```

```python
import re

# Alternate sentence tokenization using regular expressions
def simple_sent_tokenize(text):
    # Regex pattern for sentence boundaries (basic approach)
    sentence_endings = re.compile(r'(?<!\w\.\w.)(?<![A-Z][a-z]\.)(?<=\.|\?)\s')
    sentences = sentence_endings.split(text.strip())
    return [sentence.strip() for sentence in sentences if sentence]

# Tokenization example using the custom function
text = corpus
sentence_list = simple_sent_tokenize(text)
print(sentence_list)
```

```python
#function to return a random greeting to  user greeting
def greeting_response(text):
  text = text.lower()
  #Bots greeting response
  bot_greetings = ['greetings','hi','hey','hello']
  #User greeting
  user_greetings = ['hi','hey','hello','greetings']
  #If user input is a greeting return a random greeting
  for word in text.split():
    if word in user_greetings:
      return random.choice(bot_greetings)
```

Purpose of the Functions Together:
Similarity-Based Response: The bot_response function is intended to generate a chatbot response by comparing the latest user input with a list of sentences. It uses cosine similarity to determine how closely the user input matches previous inputs or predefined responses.
Sorting Similarity Scores: The index_sort function is used to sort the indices of the similarity scores so that the bot can identify the most similar sentence in the sentence_list and use it to generate an appropriate response.

```python
#This function is intended to sort the indices of the elements in list_var based on their values, in descending order
def index_sort(list_var):
  #The length of the input list list_var is determined.
  length = len(list_var) #Fixed typo here
  #list_index is a list of indices ranging from 0 to length - 1. This list will be rearranged based on the values in list_var.
  list_index = list(range(0, length))
  x = list_var
# This is a nested loop that sorts the indices in list_index based on the values in list_var.
# The condition inside the if statement is intended to compare the values in list_var at positions list_index[i] and list_index[j].
  for i in range(length): #Fixed typo here
    for j in range(length): #Fixed typo here
        if x[list_index[i]] > x[list_index[j]]:
            #it swaps the indices if the condition is met
            temp = list_index[i]
            list_index[i] = list_index[j]
            list_index[j] = temp
# The function returns the sorted indices, which represent the order of elements in list_var from highest to lowest value.
  return list_index
```

```python
print(sentence_list)
def bot_response(user_input):
    # The user input is converted to lowercase to ensure that comparisons are case-insensitive, which makes the matching process more accurate and consistent.
    user_input = user_input.lower()
    # The user's input is added to sentence_list, which presumably contains previous user inputs or predefined sentences that the bot can use to generate responses
    sentence_list.append(user_input)
    # Initializes an empty string for bot_response, which will eventually hold the bot's response.
    bot_response = ''
    # CountVectorizer is a tool from the sklearn library that converts a collection of text documents (in this case, sentence_list)
    # into a matrix of token counts. Essentially, it creates a matrix where each row represents a sentence,
    # and each column represents a word, with the values being the count of how many times that word appears in the sentence.
    # fit_transform does two things: it fits the CountVectorizer model to the data (i.e., it learns the vocabulary from the sentences) and then transforms the data into the matrix.
    cm = CountVectorizer().fit_transform(sentence_list)
    # cosine_similarity is a measure of similarity between two non-zero vectors. In this context, it calculates the similarity between the vector
    # representation of the latest user input (i.e., cm[-1]) and all other vectors in cm.
    # The result, similarity_scores, is an array containing the cosine similarity scores between the latest user input and all the sentences in sentence_list.
    similarity_scores = cosine_similarity(cm[-1], cm)
    # Converts the similarity_scores array into a one-dimensional list. This step is necessary because the
    # cosine_similarity function returns a 2D array, even if there?s only one row.
    similarity_scores_list = similarity_scores.flatten()
    # Calls the index_sort function to get the indices of the sentences in sentence_list sorted by their similarity to the latest user input, in descending order.
    index = index_sort(similarity_scores_list)
    index = index[1:]  # Skip the first index since it's the user_input itself
    response_flag = 0

    j = 0
    for i in range(len(index)):
        if similarity_scores_list[index[i]] > 0.0:
            if sentence_list[index[i]] != user_input:  # Ensure the response is not identical to the user input
                bot_response += ' ' + sentence_list[index[i]]
                response_flag = 1
                j += 1
                if j > 3:  # Limit to 3 sentences to avoid overly long responses
                    break

    if response_flag == 0:  # If no similar sentences were found, use a default response
        bot_response = "I apologize, I don't understand."

    sentence_list.remove(user_input)  # Remove the user input from the sentence list to avoid confusion in future responses
    return bot_response.strip()  # Return the response and ensure no extra spaces
```
