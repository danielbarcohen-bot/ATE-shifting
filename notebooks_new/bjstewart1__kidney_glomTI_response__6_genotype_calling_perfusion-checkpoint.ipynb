```r
setwd("~/lustre/scratch126/cellgen/clatworthy/bs16/kidney_glomTI_response/seq_data")
#10X3p
seq_inventory = read.csv("~/nfs/team297/bs16/sample_donor_data/sequencing_inventory.csv")
seq_inventory = seq_inventory[seq_inventory$experiment %in% "10X_3'_V3", ]
tissue_idx = seq_inventory$tissue %in% c('spleen_lymphoid_single_cell_suspension', 'spleen_myeloid_single_cell_suspension',
                                 'TI_compartment_single_cell_suspension', 'TI_compartment_single_cell_suspension_AND_Glomerulus_single_cell_suspension',
                                 'kidney_single_cell_suspension', 'Glomerulus_single_cell_suspension')
seq_inventory = seq_inventory[tissue_idx, ]
#remove perfusion
exclude_idx = seq_inventory$sequencing_sample_name %in% grep('10X_LPS_perfusion_sc', seq_inventory$sequencing_sample_name, value = TRUE)
seq_inventory = seq_inventory[!exclude_idx, ]
seq_inventory$irods_ID <- gsub(" ", "", seq_inventory$irods_ID)
#keep the depth samples - we will do these later because they are tricky and require supervised rerun of souporcell
#remove poorQC samples
seq_inventory = seq_inventory[!seq_inventory$poor_QC, ]
#copy
multiplex_design = seq_inventory
```

```r
perfusion_kidneys = c("K288", "K287", "K281", "K290", "K289", "K282", "K283")
multiplex_design = multiplex_design[grep(paste(perfusion_kidneys, collapse = "|"), multiplex_design$donor), ]
```

```r
#get channels
channels <- multiplex_design$sanger_ID
#get the correct directory 
seq_dir = multiplex_design$`10xDIR`
#reset some of the directory names names
seq_dir[seq_dir == '10x3p'] = '10X_3p'
seq_dir[seq_dir == '10xATAC'] = '10X_snATACseq'
seq_dir[seq_dir == '10x_multiome'] = '10X_multiome'
#then build a multiplexing matrix
channel_donors <- strsplit(multiplex_design$donor, "_")
names(channel_donors) <- multiplex_design$sanger_ID
donors <- unique(unlist(channel_donors))
multiplexing_mat <- do.call(rbind, lapply(names(channel_donors), function(i){
  donors %in% channel_donors[[i]]*1
}))
colnames(multiplexing_mat) <- donors
rownames(multiplexing_mat) <- names(channel_donors)
```

```r
options(repr.plot.width = 8, repr.plot.height = 8)
plot_experimental_design(multiplexing_mat)
```

```r
#get the locations of the files ####
dirs <- multiplex_design[match(multiplex_design$sanger_ID, rownames(multiplexing_mat)), "10xDIR"]
locations <- data.frame("channel" = rownames(multiplexing_mat), 
                        "SOC_location" = file.path("/lustre/scratch126/cellgen/clatworthy/bs16/kidney_glomTI_response/seq_data/10X_3p/SOC", multiplex_design$irods_ID, "SOC_common_variants"))
```

```r
#make multiplexing mat into a donor mat
multiplexing_mat = data.frame(multiplexing_mat)
multiplexing_mat$donor1 = multiplexing_mat[, 'K281']
multiplexing_mat$donor2 = rowSums(multiplexing_mat[, c('K282', 'K283')])
multiplexing_mat$donor3 = rowSums(multiplexing_mat[, c('K287', 'K288')])
multiplexing_mat$donor4 = rowSums(multiplexing_mat[, c('K289', 'K290')])
multiplexing_mat = multiplexing_mat[, -c(1:7)]
```
