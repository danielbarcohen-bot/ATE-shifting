## 2. ?tica em IA ? An?lise de Vi?s

```python
import numpy as np

# Simula??o de um dataset com vi?s hist?rico
np.random.seed(42)

# Dataset sint?tico: contrata??o de funcion?rios
# Features: nota_tecnica (0-10), anos_exp (0-10)
# Label: contratado (1) ou n?o (0)
# Vi?s: historicamente, homens foram mais contratados

n = 200
genero = np.random.choice([0,1], n)  # 0=mulher, 1=homem
nota   = np.random.uniform(5, 10, n)
exp    = np.random.uniform(0, 10, n)

# Decis?o com vi?s: homens t?m 20% mais chance
prob_base = (nota/10)*0.5 + (exp/10)*0.3
prob_viesada = np.clip(prob_base + genero*0.2, 0, 1)
contratado = (np.random.random(n) < prob_viesada).astype(int)

# An?lise do vi?s
taxa_homem  = contratado[genero==1].mean()
taxa_mulher = contratado[genero==0].mean()

print("=== An?lise de Vi?s no Dataset ===")
print(f"Taxa de contrata??o ? Homens:  {taxa_homem:.1%}")
print(f"Taxa de contrata??o ? Mulheres: {taxa_mulher:.1%}")
print(f"Diferen?a:                      {(taxa_homem-taxa_mulher)*100:.1f} p.p.")
print()

# Distribui??es de notas por g?nero
print("Nota t?cnica m?dia ? Homens:  ", nota[genero==1].mean().round(2))
print("Nota t?cnica m?dia ? Mulheres:", nota[genero==0].mean().round(2))
print("Anos de exp. m?dio ? Homens:  ", exp[genero==1].mean().round(2))
print("Anos de exp. m?dio ? Mulheres:", exp[genero==0].mean().round(2))
print()
print("??  Apesar de qualifica??es similares, homens s?o contratados mais.")
print("   Um modelo ML treinado nesse dataset vai perpetuar o vi?s!")

# Visualiza??o
fig, axes = plt.subplots(1,2,figsize=(12,5))
for g,rotulo,cor in [(0,'Mulheres','red'),(1,'Homens','blue')]:
    mask = genero==g
    axes[0].scatter(nota[mask],exp[mask],c=cor,label=rotulo,
                   marker='o' if contratado[mask].mean()>0.5 else 'x',alpha=0.5)
axes[0].set_xlabel('Nota T?cnica'); axes[0].set_ylabel('Anos de Exp.')
axes[0].set_title('Dataset de Contrata??o (com vi?s)'); axes[0].legend()
axes[1].bar(['Mulheres','Homens'],[taxa_mulher,taxa_homem],color=['red','blue'],alpha=0.7)
axes[1].set_ylabel('Taxa de Contrata??o'); axes[1].set_title('Disparidade na Contrata??o'); axes[1].set_ylim(0,1)
for i,(t,lab) in enumerate([(taxa_mulher,'Mulheres'),(taxa_homem,'Homens')]):
    axes[1].text(i, t+0.02, f'{t:.0%}', ha='center', fontsize=12)
plt.tight_layout(); plt.show()
```
