```python
import requests
from bs4 import BeautifulSoup
import json

```

```python
# URL of the job listings page
url = "https://en.wikipedia.org/wiki/Data_science"

```

```python
# Send a GET request to the URL
response = requests.get(url)

# Parse the HTML content of the page using BeautifulSoup
soup = BeautifulSoup(response.content, "html.parser")

# Find all job listings based on their specific HTML structure
job_listings = soup.find_all()

```

```python
#status_check function:
def status_check(r):
    if r.status_code==200:
        print("Success!")
        return 1
    else:
        print("Failed!")
        return -1
```

```python
#Check the response using the status_check command:
status_check(response)
```

```python
#Checking the Encoding of the Web Page
```

```python
def encoding_check(r):
    return (r.encoding)

#Check the response:

encoding_check(response)
```

```python
#utility function to decode the contents of the response:
def decode_content(r,encoding):

    return (r.content.decode(encoding))

contents = decode_content(response,encoding_check(response))
```

```python
#Check the type of the decoded object:
type(contents)
```

```python
#Import the package and then pass on the whole string (HTML content) to a method for parsing:
from bs4 import BeautifulSoup

soup = BeautifulSoup(contents, 'html.parser')

#Execute the following code in your notebook:
txt_dump=soup.text

#Find the type of the txt_dmp:
type(txt_dump)

#Find the length of the txt_dmp:
len(txt_dump)

#Now, the length of the text dump is much smaller than the raw HTML's string length. This is because bs4 has parsed through the HTML and extracted only human-readable text for further processing.
#Print the initial portion of this text.
print(txt_dump[0:11000])
```

```python
# Extract relevant information and store it in a list of dictionaries
jobs_data = []
for job in job_listings:
    title = job.find("h2").text.strip()
    company = job.find("span", class_="company").text.strip()
    location = job.find("span", class_="location").text.strip()
    link = job.find("a")["href"].strip()
    
    job_info = {
        "title": title,
        "company": company,
        "location": location,
        "link": link
    }
    jobs_data.append(job_info)

# Save the data in JSON format
with open("jobs_data.json", "w") as json_file:
    json.dump(jobs_data, json_file, indent=4)

print("Job data scraped and saved in jobs_data.json.")
```
