# **Hand Written Digit Prediction**

**OBJECTIVE -**

The objective of a handwritten digit prediction machine learning project using 8x8 pixel images of digits is to develop and evaluate a model that can accurately classify these low-resolution images into their respective digit classes (0-9).

**Data Source -**

The digits dataset consists of 8x8 pixel images of digits. The images attribute of the dataset stores 8x8 arrays of grayscale values for each image. We will use these arrays to visualise the first 4 images. The target attribute of the dataset stores the digit each image represents.

**Import Library**

```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

**Import Data**

```python
from sklearn.datasets import load_digits
```

```python
df = load_digits()
```

```python
_, axes = plt.subplots(nrows=1, ncols=4, figsize=(10,3))
for ax, image, label in zip(axes, df.images, df.target):
    ax.set_axis_off()
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation="nearest")
    ax.set_title("Training: %i" % label)
```

**Data Preprocessing**

```python
df.images.shape
```

```python
df.images[0]
```

```python
df.images[0].shape
```

```python
len(df.images)
```

```python
n_samples = len(df.images)
data = df.images.reshape((n_samples,-1))
```

```python
data[0]
```

```python
data[0].shape
```

```python
data.shape
```

**Scaling Image Data**

```python
data.min()
```

```python
data.max()
```

```python
data = data/16
```

```python
data.min()
```

```python
data.max()
```

```python
data[0]
```

**Train Test Split Data**

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(data, df.target, test_size = 0.3)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

**Random Forest Model**

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier()
```

```python
rf.fit(X_train, y_train)
```

**Predict Test Data**

```python
y_pred = rf.predict(X_test)
```

```python
y_pred
```

**Model Accuracy**

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
confusion_matrix(y_test, y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

## **Explanation -**

**Data -**
The digits dataset consists of 1797 samples of handwritten digits, each represented as an 8x8 pixel grayscale image. Each pixel value ranges from 0 to 16, indicating the intensity.

**Preprocessing -**
Flattening the 8x8 images into 1D arrays is crucial for feeding the data into the random forest classifier model. Standardizing the data ensures that each feature contributes equally to the result.

**Model -**
Random Forest Classifier is chosen for its simplicity and effectiveness in multiclass classification problems. The model is trained on the standardized training data.

**Evaluation -**
The accuracy score provides a quick measure of the model's performance. The confusion matrix and classification report offer deeper insights into the model's performance across different classes.

**Visualization -**
Visualizing the results helps in understanding the model's strengths and weaknesses. By comparing predicted and actual labels, we can identify specific instances where the model might struggle.

**Conclusion -**
This handwritten digit prediction project demonstrates a typical workflow in machine learning, from data preprocessing to model evaluation and visualization. It provides a practical introduction to image classification and the use of machine learning techniques for solving real-world problems.
