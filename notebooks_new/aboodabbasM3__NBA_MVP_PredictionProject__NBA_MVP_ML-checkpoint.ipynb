#### Determining features

```python
#We removed player, team, and position
#I will also remove the label we are trying to predict which is share
#along with Points won and points max which is used to directly derive share

features = ['Age', 'G', 'GS', 'MP', 'FG', 'FGA', 'FG%',
       '3P', '3PA', '3P%', '2P', '2PA', '2P%', 'eFG%', 'FT', 'FTA', 'FT%',
       'ORB', 'DRB', 'TRB', 'AST', 'STL', 'BLK', 'TOV', 'PF', 'PTS', 'Year',
        'W', 'L', 'W/L%', 'GB', 'PS/G', 'PA/G','SRS']
```

```python
X = stats[features]

y = stats["Share"]
```

#### Splitting for training and testing
I won't randomly split the data using train test split because it would mix different NBA seasons together and cause data leakage, where the model learns from future information. This leads to unrealistic performance, so the data should instead be split in chronological order to properly simulate predicting future MVP shares

```python
#training data is all seasons before 2025
train = stats[stats["Year"] < 2025]
```

```python
#testing data is the 2025 season
test = stats[stats["Year"] == 2025]
```

```python
train_X = train[features]
train_y = train["Share"]

test_X = test[features]
test_y = test["Share"]
```

```python
from sklearn.linear_model import Ridge

#Creating a ridge regression model
model = Ridge(alpha=.1)
```

```python
#Training the model
model.fit(train_X,train_y)
```

```python
#Getting predictions
predictions = model.predict(test_X)
```

```python
#Turning our predictions from an array to a df
predictions = pd.DataFrame(predictions, columns = ["predictions"], index = test.index)
predictions
```

```python
#Combining our predictions to player names
combination = pd.concat([test[["Player", "Share"]], predictions], axis=1)
combination
```

```python
#Sorting to get an idea of which players got the most predictions
combination.sort_values("predictions", ascending= False)
```

```python
#I will use a custom metric called average precision tha evaluates how well the model ranks the true top 5 MVP players
#First I take the actual top 5 players based on MVP share then I iterate through all players sorted by predicted score
#Each time I encounter a player who is actually in the top 5 I compute the precision at that point which is found / seen so far
#The final score is the average of these precision values which will reward models that rank true top players earlier in the list

def find_precision(combination):
    actual = combination.sort_values("Share", ascending = False).head(5)
    predicted = combination.sort_values("predictions", ascending = False)
    ps = []
    found = 0
    seen = 1

    for index, row in predicted.iterrows():
        if row["Player"] in actual["Player"].values:
            found +=1
            ps.append(found/seen)
        seen +=1

    return sum(ps)/len(ps)
        
```

### Implementing Backtesting

```python
#Here I will test the model on every year on the years before it
years = list(range(1991,2026))
```

```python
#Backtesting the MVP prediction model by simulating real world yearly predictions
#for each season i train only on previous years data to predict that seasons MVP vote share
#then I evaluate performance and store results over time

precisions = []
allpredictions = []

for year in years[5:]:
    train = stats[stats["Year"]< year]
    test = stats[stats["Year"] == year]

    model.fit(train[features], train["Share"])
    predictions = model.predict(test[features])

    predictions = pd.DataFrame(predictions, columns = ["predictions"], index = test.index)
    combination = pd.concat([test[["Player", "Share"]], predictions], axis=1)

    allpredictions.append(combination)
    precisions.append(find_precision(combination))
```

```python
#Mean precision over all the years

sum(precisions) / len(precisions)
```

```python
#This function adds ranking columns to compare actual MVP voting performance to model predictions
#and it measures how far off the models ranking is from reality

def add_ranks(combination):
    combination = combination.sort_values("Share", ascending = False)
    combination["Rk"] = list(range(1,combination.shape[0]+1))
    combination = combination.sort_values("predictions", ascending = False)
    combination["PredictedRk"] = list(range(1,combination.shape[0]+1))
    combination["Difference"] = combination["Rk"] - combination["PredictedRk"]

    return combination
```

```python
ranking = add_ranks(allpredictions[1])
ranking[ranking["Rk"] <= 5].sort_values("Difference", ascending = False)
```

```python
#This function backtests the MVP prediction model by simulating real seasons one by one
#for each year it trains only on past data and predicts that seasons MVP vote share and
#then ranks players by real vs predicted performance and checks how good the ranking is using precision
#Lastly it saves everything so we can see how the model performs over time and across all seasons

def backtest(stats, model, year, features):
    precisions = []
    allpredictions = []

    for year in years[5:]:
        train = stats[stats["Year"]< year]
        test = stats[stats["Year"] == year]

        model.fit(train[features], train["Share"])
        predictions = model.predict(test[features])

        predictions = pd.DataFrame(predictions, columns = ["predictions"], index = test.index)
        combination = pd.concat([test[["Player", "Share"]], predictions], axis=1)

        combination = add_ranks(combination)

        allpredictions.append(combination)
        precisions.append(find_precision(combination))

    return sum(precisions) / len(precisions), precisions, pd.concat(allpredictions)
        
```

```python
average_precision, precisions, allpredictions = backtest(stats, model, years[5:], features)
```

```python
#These are the main misdiagnosed players from our model, these are players who ranked much higher than predicted
allpredictions[allpredictions["Rk"] <= 5].sort_values("Difference").head(10) 
```

```python
#These are the most important features that the model takes into account
pd.concat([pd.Series(model.coef_), pd.Series(features)], axis=1).sort_values(0, ascending = False)
```

### Feature Engineering

```python
#This will give us the ratio of a players stats to the league average for that year
stat_ratios = stats[["PTS", "AST", "STL", "BLK", "3P","Year"]].groupby("Year").apply(lambda x: x/x.mean())
```

```python
stat_ratios.index = stat_ratios.index.droplevel()
stat_ratios
```

```python
#Adding these new features to the dataframe
stats[["PTS_Ratio", "AST_Ratio", "STL_Ratio", "BLK_Ratio", "3P_Ratio"]] = stat_ratios[["PTS", "AST", "STL", "BLK", "3P"]]
```

```python
stats.head()
```

```python
#Adding these new features to our features list
features += ["PTS_Ratio", "AST_Ratio", "STL_Ratio", "BLK_Ratio", "3P_Ratio"]
```

```python
average_precision, precisions, allpredictions = backtest(stats, model, years[5:], features)
```

```python
#There is a slight improvement in precision with the new features
average_precision
```
