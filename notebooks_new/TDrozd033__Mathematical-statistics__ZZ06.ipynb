### Zadanie 1: Estymacja warto?ci oczekiwanej [8 pkt]

Przygotuj funkcj? `mean_estimation()`, kt?ra b?dzie dokonywa?a estymacji przedzia?owej warto?ci oczekiwanej dla danych wej?ciowych w postaci szeregu szczeg??owego zgodnie ze schematem z za??czonego do notatnika zestawu wzor?w.

Opr?cz zwracanych warto?ci granic przedzia?u funkcja powinna wy?wietla? nast?puj?cy komunikat:

`?? nale?y do przedzia?u [X, Y] przy za?o?eniu poziomu ufno?ci 1-?? = Z`

gdzie X, Y i Z s? automatycznie uzupe?niane przez funkcj?, a X i Y dodatkowo sformatowane w taki spos?b, ?eby wy?wietla?y si? z dok?adno?ci? do 4 miejsc po przecinku.

```python


def mean_estimation(data, alpha, population_std="unknown"):
    """
    Parameters
    -------
    data: DataFrame
        Tablica zawieraj?ca domy?lny indeks i pojedyncz? kolumn? "X" z warto?ciami pr?by losowej,
    alpha: float
        Warto?? okre?laj?ca poziom istotno?ci.
    population_std: str or float
        Je?eli odchylenie standardowe populacji nie jest znane to parametr przyjmuje warto?? "unknown",
        w przeciwnym wypadku jest to warto?? okre?laj?ca odchylenie standardowe populacji.
    
    Returns
    -------
    mean_low: float
        Dolna granica wyliczonego przedzia?u ufno?ci.
    mean_high: float
        G?rna granica wyliczonego przedzia?u ufno?ci.
    """
    # YOUR CODE HERE
    n = data.shape[0] # liczba wierszy w ramce 
    mean = data.iloc[:,0].sum() / n # obliczamy srednia elementow 1 kolumny 
    S = 0
    for i in range(n):
        S += (data.iloc[i,0] - mean) ** 2 # suma kwadratow roznic 
    S = S/n
    S = np.sqrt(S)
   
    if population_std != "unknown":
        temp = sp.stats.norm.ppf(1 - alpha/2)
        temp = temp * population_std / np.sqrt(n)
    else:
        rozklad_t = sp.stats.t.ppf(1 - alpha/2, n-1)
        temp = rozklad_t * S/np.sqrt(n-1)
  


    mean_low = mean - temp
    mean_high = mean + temp

    print(f"?? nale?y do przedzia?u [{mean_low: .4f}, {mean_high:.4f}] przy za?o?eniu poziomu ufno?ci 1-{alpha} = {1-alpha}")
    return mean_low, mean_high
            
        
            
    # ppf do obliczania 


```

### Zadanie 2: Estymacja wariancji [8 pkt]

Przygotuj funkcj? `variance_estimation()`, kt?ra b?dzie dokonywa?a estymacji przedzia?owej wariancji dla danych wej?ciowych w postaci szeregu szczeg??owego zgodnie ze schematem z za??czonego do notatnika zestawu wzor?w.

Opr?cz zwracanych warto?ci granic przedzia?u funkcja powinna wy?wietla? nast?puj?cy komunikat:

`??^2  nale?y do przedzia?u [X, Y] przy za?o?eniu poziomu ufno?ci 1-?? = Z`

gdzie X, Y i Z s? automatycznie uzupe?niane przez funkcj?, a X i Y dodatkowo sformatowane w taki spos?b, ?eby wy?wietla?y si? z dok?adno?ci? do 4 miejsc po przecinku.

```python
def variance_estimation(data, alpha, population_mean="unknown"):
    """
    Parameters
    -------
    data: DataFrame
        Tablica zawieraj?ca domy?lny indeks i pojedyncz? kolumn? "X" z warto?ciami pr?by losowej,
    alpha: float
        Warto?? okre?laj?ca poziom istotno?ci.
    population_mean: str or float
        Je?eli warto?? oczekiwana populacji nie jest znane to parametr przyjmuje warto?? "unknown",
        w przeciwnym wypadku jest to warto?? okre?laj?ca warto?? oczekiwan? populacji.
    
    Returns
    -------
    var_low: float
        Dolna granica wyliczonego przedzia?u ufno?ci.
    var_high: float
        G?rna granica wyliczonego przedzia?u ufno?ci.
    """
    # YOUR CODE HERE
    n = data.shape[0] # liczba wierszy w ramce danych zeby wiedziec ile n ( probek) 
    # wyliczamy wariancje
    mean = data.iloc[:,0].sum() / n # wyliczamy srednia - suma poszczegolnych elementow w 1 kolumnie podzielona przez liczbe probek ( czyli liczbe wierszy )
    var = 0
    var1 =0
    for i in range(n):
        var += (data.iloc[i,0] - mean ) ** 2
        
    var = var / n
    
    if population_mean == "unknown": # jesli nieznane 
        temp1 = sp.stats.chi2.ppf(1 - (alpha/2), n-1)
        var_low = (n * var)/temp1

        temp2 = sp.stats.chi2.ppf( alpha/2, n-1)
        var_high = (n * var)/temp2

    else: #  jesli population_mean != unknown  
        for k in range(n): # obliczamy osobno wykorzystujac population_mean wartosc wariancji 
            var1 += (data.iloc[k,0] - population_mean) **2
             
        var1 = var1 / n
        temp1 = sp.stats.chi2.ppf(1 - (alpha/2), n-1)
        var_low = (var1*n)/ temp1
        temp2 = sp.stats.chi2.ppf(alpha/2, n-1)
        var_high = (var1 * n)/ temp2



    print(f" ??^2  nale?y do przedzia?u [{var_low:.4f}, {var_high:.4f}] przy za?o?eniu poziomu ufno?ci 1 - {alpha} = {1 - alpha}")
    return var_low, var_high

    
```

```python
### Kom?rka testowa

# Test na test_data_1 przy za?o?eniu, ?e znane jest odchylenie standardowe populacji, z kt?rej zosta?a pobrana pr?bka losowa
assert(np.all(np.isclose(variance_estimation(test_data_1, 0.1, population_mean=5), (0.01855573624556403, 0.05528660848892888))));
print()
assert(np.all(np.isclose(variance_estimation(test_data_1, 0.05, population_mean=5), (0.017025745016269588, 0.06280068548350207))));
print()
assert(np.all(np.isclose(variance_estimation(test_data_1, 0.02, population_mean=5), (0.015455150810030724, 0.07328116757245808))));
print()

# Test na test_data_1 przy za?o?eniu, ?e nie jest znane odchylenie standardowe populacji, z kt?rej zosta?a pobrana pr?bka losowa
assert(np.all(np.isclose(variance_estimation(test_data_1, 0.1), (0.018268125369359626, 0.05442967510189219))));
print()
assert(np.all(np.isclose(variance_estimation(test_data_1, 0.05), (0.0167618487538223, 0.06182728513230527))));
print()
assert(np.all(np.isclose(variance_estimation(test_data_1, 0.02), (0.015215598500840796, 0.07214532146342346))));
```

### Zadanie 3: Estymacja warto?ci wsp??czynnika korelacji liniowej [6 pkt]

Przygotuj funkcj? `correlation_estimation()`, kt?ra b?dzie dokonywa?a estymacji przedzia?owej warto?ci wsp??czynnika korelacji liniowej Pearsona dla danych wej?ciowych w postaci szeregu szczeg??owego zgodnie ze schematem z za??czonego do notatnika zestawu wzor?w.

Opr?cz zwracanych warto?ci granic przedzia?u funkcja powinna wy?wietla? nast?puj?cy komunikat:

`r nale?y do przedzia?u [X, Y] przy za?o?eniu poziomu ufno?ci 1-?? = Z`

gdzie X, Y i Z s? automatycznie uzupe?niane przez funkcj?, a X i Y dodatkowo sformatowane w taki spos?b, ?eby wy?wietla?y si? z dok?adno?ci? do 4 miejsc po przecinku.

```python
def correlation_estimation(data, alpha):
    """
    Parameters
    -------
    data: DataFrame
        Tablica zawieraj?ca domy?lny indeks i dwie kolumny "X" i "Y" z wynikami pr?by losowej.
    alpha: float
        Warto?? okre?laj?ca poziom istotno?ci.
    Returns
    -------
    r_corr_low: float
        Dolna granica wyliczonego przedzia?u ufno?ci.
    r_corr_high: float
        G?rna granica wyliczonego przedzia?u ufno?ci.
    """   
    # YOUR CODE HERE
    # wyliczamy r 
    # do wzorow sp.stats.norm.ppf(1 - (alpha/2))
    # bez przypadkow jak wczesniej z unknown czyli nie bedzie if i else
    # dane z drugiego data framu 
    # x - kolumna zerowa, y - kolumna pierwsza 
    n = data.shape[0] # obliczamy liczbe wierszy w ramce danych 
    # obliczamy srednia z X i srednia z Y
    mean_x = data.iloc[:,0].sum() / n
    mean_y = data.iloc[:,1].sum() / n

    # obliczamy po kolei licznik i mianownik zeby podstawic pozniej do r 
    licznik =0;
    for i in range(n):
        licznik += (data.iloc[i,0] - mean_x) * (data.iloc[i,1] - mean_y)


    mianownik_1 = 0
    mianownik_2 =0
    mianownik = 0
    for j in range(n):
        mianownik_1 += (data.iloc[j,0] - mean_x)**2
        mianownik_2 += (data.iloc[j,1] - mean_y) **2

    mianownik_1 = np.sqrt(mianownik_1)
    mianownik_2 = np.sqrt(mianownik_2)
    mianownik = mianownik_1 * mianownik_2
        

    # obliczamy wartosc r
    r = 0
    r = licznik /mianownik


    # obliczamy estymacje 
    temp = 0
    norm = 0
    norm = sp.stats.norm.ppf(1 - (alpha/2))
    temp = norm * ((1 - r ** 2) / np.sqrt(n))
    r_corr_low = r - temp
    r_corr_high = r + temp

    
    print(f" r nale?y do przedzia?u [{r_corr_low:.4f}, {r_corr_high:.4f}] przy za?o?eniu poziomu ufno?ci 1-{alpha} = {1 - alpha}")
    return r_corr_low, r_corr_high
    

```

### Zadanie 4: Estymacja wsp??czynnik?w r?wnania regresji liniowej [8 pkt]

Przygotuj funkcj? `interval_linear_regression_coefficients_estimation()`, kt?ra b?dzie dokonywa?a estymacji przedzia?owej warto?ci wsp??czynnik?w r?wnania regresji liniowej dla danych wej?ciowych w postaci szeregu szczeg??owego zgodnie ze schematem z za??czonego do notatnika zestawu wzor?w.

Opr?cz zwracanych warto?ci granic przedzia?u funkcja powinna wy?wietla? nast?puj?cy komunikat:

`a nale?y do przedzia?u [X1, Y1] przy za?o?eniu poziomu ufno?ci 1-?? = Z1` </br>
`b nale?y do przedzia?u [X2, Y2] przy za?o?eniu poziomu ufno?ci 1-?? = Z2`

gdzie X1, X2, Y1, Y2, Z1 i Z2 s? automatycznie uzupe?niane przez funkcj?, a X1, X2, Y1 i Y2 dodatkowo sformatowane w taki spos?b, ?eby wy?wietla?y si? z dok?adno?ci? do 4 miejsc po przecinku.

```python
def linear_regression_coefficients_estimation(data, alpha):
    """
    Parameters
    -------
    data: DataFrame
        Tablica zawieraj?ca domy?lny indeks i dwie kolumny "X" i "Y" z wynikami pr?by losowej.
    alpha: float
        Warto?? okre?laj?ca poziom istotno?ci.
    Returns
    -------
    a_ci: tuple
        Zmienna typu tuple zawierajaca granice przedzia?u ufno?ci parametru a (a_low, a_high).
    b_ci: tuple
        Zmienna typu tuple zawierajaca granice przedzia?u ufno?ci parametru b (b_low, b_high).
    """   
    # YOUR CODE HERE
    # uzywac bedzie rozklad tStudenta np.stats.t.ppf(1 - (alpha/2), n-2)
    # najpierw trzeba policzyc estymatory 
    # pozniej y_daszek i su, pozniej D(a) i D(b)
    # na koncu estymacja 
    n = data.shape[0]
    mean_x = data.iloc[:,0].sum() / n
    mean_y = data.iloc[:,1].sum() / n
    # liczymy a
    licznik_a = 0
    mianownik_a = 0
    a = 0
    b = 0
    for i in range(n):
        licznik_a += (data.iloc[i,0] - mean_x) * (data.iloc[i,1] - mean_y)
        mianownik_a += (data.iloc[i,0] - mean_x) ** 2
    a = licznik_a / mianownik_a
    # obliczamy b
    b = mean_y - ( a * mean_x)

    y_daszek = np.zeros(n) # deklarujemy pusta tablice o dlugosci n
    for j in range(n):
        y_daszek[j] = a * data.iloc[j,0] + b

    # obliczamy su
    zmienna = 0
    for i in range(n):
        zmienna += (data.iloc[i,1] - y_daszek[i]) **2

    su = zmienna/(n - 2)
    su = np.sqrt(su)

    # obliczamy D(a) i D(b)
    mianownik_Da = 0
    #mian_Da = 0
    D_a = 0
    for i in range(n):
        mianownik_Da += (data.iloc[i,0] - mean_x) ** 2

    mian_Da = np.sqrt(mianownik_Da)
    D_a = su/mian_Da


    #D(b)
    #mianownik_Db = 0
    mianownik_Db = n * mianownik_Da
    licznik_Db = 0
    for k in range(n):
        licznik_Db += data.iloc[k,0] ** 2

    res = 0
    D_b = 0
    res = licznik_Db / mianownik_Db
    res = np.sqrt(res)

    D_b = su * res

    # liczymy estymacje
    rozklad_t = sp.stats.t.ppf(1 - (alpha/2), n-2)
    est_a = rozklad_t * D_a
    a_low = a - est_a
    a_high = a + est_a

    est_b = rozklad_t * D_b
    b_low = b - est_b
    b_high = b + est_b



    print(f" {a} nale?y do przedzia?u [{a_low:.4f}, {a_high:.4f}] przy za?o?eniu poziomu ufno?ci 1-{alpha} = {1 - alpha}")
    print(f" {b} nale?y do przedzia?u [{b_low:.4f}, {b_high:.4f}] przy za?o?eniu poziomu ufno?ci 1-{alpha}= {1 - alpha}")
    a_ci = (a_low,a_high)
    b_ci = (b_low, b_high)
    return (a_ci, b_ci)

    

    
        
        
        


    
        
```
