## Combine years of dataset

```python
import pandas as pd

# List of filenames
filenames = [f"{super_name}_2023.csv", f"{super_name}_2024.csv", f"{super_name}_2025.csv"]

# Read and combine
combined_df = pd.concat([pd.read_csv(file) for file in filenames], ignore_index=True)

# Save to new file
combined_df.to_csv(f"{super_name}_combined.csv", index=False)

print('Files combined and saved to', f"{super_name}_combined.csv")
```

## Merge stocks and news by date

```python
import pandas as pd
from datetime import datetime, timedelta

# Load the Qualcomm historical data
qcom_data = pd.read_csv(f"HistoricalData_{super_code}.csv")

# Load the NVIDIA combined data
nvidia_data = pd.read_csv(f"{super_name}_combined.csv")  # Corrected file name

# Convert the 'Date' column in Qualcomm data to datetime format (MM/DD/YYYY)
qcom_data['Date'] = pd.to_datetime(qcom_data['Date'], format='%m/%d/%Y')

# Function to clean and standardize the 'Date' column in NVIDIA data
def clean_date(date_str):
    if pd.isna(date_str):  # Handle NaN or missing values
        return pd.NaT
    try:
        # Try to parse the date in the format '%b %d, %Y' (e.g., "May 25, 2023")
        return datetime.strptime(str(date_str), '%b %d, %Y')
    except ValueError:
        # Handle relative dates like "7 days ago"
        if isinstance(date_str, str) and "days ago" in date_str:
            days_ago = int(date_str.split()[0])  # Extract the number of days
            return datetime.now() - timedelta(days=days_ago)
        elif isinstance(date_str, str) and "yesterday" in date_str.lower():
            return datetime.now() - timedelta(days=1)
        else:
            # If the date format is unrecognized, return NaT (Not a Time)
            return pd.NaT

# Apply the cleaning function to the 'Date' column in NVIDIA data
nvidia_data['Date'] = nvidia_data['Date'].apply(clean_date)

# Drop rows with invalid or missing dates (NaT)
nvidia_data = nvidia_data.dropna(subset=['Date'])

# Select only the required columns from NVIDIA data
nvidia_data = nvidia_data[['Title', 'Link', 'Description', 'Date']]

# Merge the two datasets on the 'Date' column using a LEFT JOIN
merged_data = pd.merge(qcom_data, nvidia_data, on='Date', how='left')

# Save the merged data to a new CSV file
merged_data.to_csv(f"{super_name}_merged.csv", index=False)

print("Merged data has been saved to", f"{super_name}_merged.csv")
```

## Fill missing data by prev data

```python
import pandas as pd

# Load the CSV file
file_path = f"{super_name}_merged.csv"  # Replace with the path to your CSV file
df = pd.read_csv(file_path)

# Reverse the DataFrame to process in chronological order (2020 to 2025)
df = df.iloc[::-1]

# Fill missing values in 'Title', 'Link', and 'Description' using forward fill
columns_to_fill = ['Title', 'Link', 'Description']
df[columns_to_fill] = df[columns_to_fill].fillna(method='ffill')

# Reverse the DataFrame back to its original order (2025 to 2020)
df = df.iloc[::-1]

# Save the updated DataFrame back to a CSV file
output_file_path = f"{super_name}_merged_filled.csv"  # Replace with the desired output file path
df.to_csv(output_file_path, index=False)

print(f"Missing values filled and saved to {output_file_path}")
```
