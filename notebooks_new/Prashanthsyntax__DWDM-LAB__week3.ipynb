Loading the Dataset

a) Manual Loading

```python
# Manual function to read CSV
def load_manual(filename):
    with open(filename, 'r') as file:
        data = file.readlines()
    # Split lines and handle header
    header = data[0].strip().split(',')
    rows = [line.strip().split(',') for line in data[1:]]
    return header, rows

header, rows = load_manual('data.csv')
print("Header:", header)
print("Rows:", rows)
```

b) Using numpy.loadtxt

```python
import numpy as np

# Assuming numeric data only (Age, Salary)
data = np.loadtxt('data1.csv', delimiter=',', skiprows=1, usecols=(1,2))
print(data)
```

c) Using numpy.genfromtxt

```python
import numpy as np

data = np.genfromtxt('data.csv', delimiter=',', skip_header=1, usecols=(1,2), filling_values=np.nan)
print(data)
```

d) Using pandas.read_csv

```python
import pandas as pd

df = pd.read_csv('data.csv')
print(df)
```

Dealing with Missing Data

a) Identify missing data

```python
print(df.isnull())          # True for missing values
print(df.isnull().sum())    # Count missing per column
```

b) Remove missing data

```python
# Drop rows with any missing values
df_drop = df.dropna()
print(df_drop)
```

c) Fill missing data

```python
# Fill numeric columns with mean
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].mean(), inplace=True)
print(df)
```

d) Fill with a constant value

```python
df['Salary'].fillna(50000, inplace=True)
```

```python
df
```

e) Forward/Backward Fill

```python
df.fillna(method='ffill', inplace=True)  # forward fill
df.fillna(method='bfill', inplace=True)  # backward fill
```
