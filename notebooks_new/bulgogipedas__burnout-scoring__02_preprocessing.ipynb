# 02. Preprocessing & Feature Engineering
This notebook tests and validates the modular preprocessing class and feature engineering module.

```python
import os
import sys
sys.path.append('../')
import pandas as pd
from src.preprocessing import BurnoutPreprocessor
from src.features import engineer_features
```

## 1. Load Data

```python
train = pd.read_csv('../data/raw/train.csv')
print(f'Original shape: {train.shape}')
```

## 2. Apply Feature Engineering

```python
train_fe = engineer_features(train)
print(f'After feature engineering: {train_fe.shape}')
train_fe[['Date of Joining', 'tenure_days', 'workload_mental_ratio', 'burnout_risk_index']].head()
```

## 3. Apply Modular Preprocessing (Impute, Encode, Scale)

```python
preprocessor = BurnoutPreprocessor()
# For class-based setup, we add engineered numeric features to numerical columns list
preprocessor.num_cols.extend(['tenure_days', 'workload_mental_ratio', 'burnout_risk_index'])

X, y = preprocessor.fit_transform(train_fe, is_train=True)
print(f'Processed feature matrix X shape: {X.shape}')
print(f'Processed target vector y shape: {y.shape}')
X.head()
```

## 4. Convert Target to Binary Label

```python
# Threshold the Burn Rate at 0.5 to get binary Burnout Risk classes
y_bin = (y >= 0.5).astype(int)
print('Class distribution of binary target (Burnout Risk >= 0.5):')
print(y_bin.value_counts(normalize=True))
```
