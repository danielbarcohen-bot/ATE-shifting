```python
import pandas as pd 
df=pd.read_csv('datalab.csv')
print(df.shape)

df.head(100)
# df.info()
# df.describe()
```

```python
import matplotlib.pyplot as plt
type_count=df['type'].value_counts()
plt.figure(figsize=(6,4))
plt.bar(type_count.index,type_count.values,color=['skyblue','orange'])
plt.title("Movies Vs Tv Show")
plt.xlabel('Type')
plt.ylabel('Count')
plt.tight_layout
plt.show()




rating_counts=df['rating'].value_counts()
plt.figure(figsize=(6,4))
plt.pie(rating_count ,labels=rating_counts.index,autopct='%1.1f%%',startangle=90)
plt.title("rating Count")
plt.tight_layout
plt.show()
```
