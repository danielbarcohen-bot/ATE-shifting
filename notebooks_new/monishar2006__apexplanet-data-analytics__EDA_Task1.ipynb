```python
df.drop_duplicates(inplace=True)
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")

df.columns
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df.info()
```

```python
Q1 = df["total"].quantile(0.25)
Q3 = df["total"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["total"] < lower) | (df["total"] > upper)]

print(outliers.shape)
```

```python
df = df[(df["total"] >= lower) & (df["total"] <= upper)]
```

```python
df["date"] = pd.to_datetime(df["date"])
df.info()
```

## Date Conversion

The `date` column was converted from object type to datetime format.

```python
Q1 = df["total"].quantile(0.25)
Q3 = df["total"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["total"] < lower) | (df["total"] > upper)]

print("Number of outliers:", outliers.shape[0])
```

```python
df = df[(df["total"] >= lower) & (df["total"] <= upper)]
```

## Outlier Detection

The IQR method was used to detect outliers in the `total` column.

### Observation

Outliers were identified and handled appropriately.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df["total"], bins=20)
plt.title("Distribution of Total Sales")
plt.xlabel("Total Sales")
plt.ylabel("Frequency")
plt.show()
```

## Observation

Most sales are concentrated between lower and medium values, with fewer high-value transactions.

```python
import seaborn as sns

plt.figure(figsize=(8,5))
sns.boxplot(x=df["total"])
plt.title("Boxplot of Total Sales")
plt.show()
```

## Observation

The boxplot highlights the spread of total sales and indicates the presence of a few outliers.

```python
df["gender"].value_counts().plot(kind="bar")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

## Observation

The number of male and female customers is nearly balanced.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x="gross_income", y="total")

plt.title("Gross Income vs Total Sales")
plt.show()
```

## Observation

Gross income increases as the total sales amount increases, showing a positive relationship.

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    df.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

## Observation

The heatmap shows strong positive correlations between Total, COGS, Tax, and Gross Income.
