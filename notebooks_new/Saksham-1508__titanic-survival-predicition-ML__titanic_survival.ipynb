## Titanic Survival Prediction (LOGISTIC REGRESSION)

```python
import pandas as pd
import numpy as np 
import seaborn as sn
import matplotlib.pyplot as plt
```

### Reading the data from csv files

```python
# reading the data from csv file
df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
df
```

### Handling NULL values

```python
# checking all columns with null values
df.isna().sum()
```

```python
# dropping 'Cabin' because majority values(687) are NULL hence replacing them with default values is not a good idea
df.drop(['Cabin'],axis=1,inplace=True)
```

#### Filling NULL values with mean median or mode

```python
df['Age'].fillna(df['Age'].mean(),inplace=True)
```

```python
df['Embarked'].fillna(df['Embarked'].mode()[0],inplace=True)
```

### Finding Corelation Between Attributes

Finding corelation between attributes is important for feature selection.
Features with strong positive or negative corelation with the labels are selected for training the model.

```python
df.corr()['Survived']
```

```python
sn.countplot('Survived',hue='Pclass',data=df)
```

```python
sn.countplot('Survived',hue='Sex',data=df)
```

```python
df.replace({'Sex':{'male':1,'female':0}},inplace=True)
```

```python
df.replace({'Embarked':{'S':1,'C':2,'Q':3}},inplace=True)
```

```python
labels = df['Survived']
df = df[['Pclass','Sex','Age','SibSp','Parch','Fare','Embarked']]
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
x_train,x_test,y_train,y_test = train_test_split(df,labels,test_size=0.25,random_state=42)
```

```python
x_train
```

```python
x_test.describe()
```

```python
# model = LogisticRegression()
model = RandomForestClassifier()
model.fit(x_train,y_train)
```

```python
prediction = model.predict(x_test)
```

```python
accuracy_score(prediction,y_test)
```
