```python
def contar_palavra(arquivo, palavra):
  try:
      # Abre o arquivo para leitura
      with open(arquivo, 'r', encoding='utf-8') as f:
          # L? o conte?do do arquivo
          conteudo = f.read()

      # Verifica se a palavra n?o ? uma string vazia
      if not palavra:
          print("A palavra para contar n?o pode ser uma string vazia.")
          return 0

      # Divide o conte?do em palavras e conta as ocorr?ncias da palavra
      palavras = conteudo.lower().split()

      # Conta as ocorr?ncias da palavra
      return palavras.count(palavra.lower())

  except FileNotFoundError:
      print(f"O arquivo {arquivo} n?o foi encontrado.")
      return 0
  except Exception as e:
      print(f"Ocorreu um erro: {e}")
      return 0
```
