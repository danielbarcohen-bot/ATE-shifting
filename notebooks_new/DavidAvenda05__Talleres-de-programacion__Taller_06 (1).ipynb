## Ejercicio 04 - Leer una lista de n?meros

Escribir un **programa** en Python que lea la edad de $N$ estudiantes del curso, los almacene en una lista de n?meros y luego imprima la lista

````python
# Programa en Python para leer la edad de varios estudiantes. 
# Vamos a preguntar cuantos estudiantes hay = N.
c=0
estudiantes=int(input("cuantos estudiantes hay: "))
edades1=[]
while True:
    if(c==estudiantes):
        break
    edad=int(input("ingrese edad: "))
    edades1.append(edad)
    c=c+1
print("la lista: ", edades1)
````

````python
# Funci?n que cuenta en una lista
def contar_temperaturas(temps: [float]) -> int:
    c=0
    for i in temps:
        if(i>=15 and i<=20):
            c=c+1
    return c


# Pruebas de la funci?n anterior
assert contar_temperaturas([11, 18, 23.5, 19.5, 6, 16.3, 22.1, 4.6, 18.9, 18.6, 15.1, 16.1, 19.9, 1.5, 12, 21.7]) == 8
print("Prueba superada ????")
````

## Ejercicio 12 - Contar en listas de tuplas

Escriba una funci?n que reciba esta lista de tuplas con los resultados de los partidos, y que retorne cu?ntos partidos gan? en total, cu?ntos empat? y cu?ntos perdi?.

````python
# Funci?n para contar partidos
def resultados_partidos(partidos:[(int, int)]) -> (int, int, int):
    g=0
    p=0
    e=0
    for a,b in partidos:
        if(a==b):
            e=e+1
        elif (a>b):
            g=g+1
        else:
            p=p+1
    return g,e,p
# prueba de la funci?n anterior
assert (resultados_partidos([(1, 3), (0, 0), (4, 0), (5, 3), (2, 2), (4, 3), (1, 0), (1, 2), (0, 0), (3, 2), (3, 1), (7, 0), (0, 2), (3, 3), (4, 2), (3, 4)])) == (8, 4, 4)
print("Prueba superada ????")
````
