## 정제

불필요한 단어 제거

- 사용 빈도가 매우 적은 단어
- 불용어 (자주 사용되지만, 큰 의미를 갖지 않은 단어)
- 특수문자

```python
# 단어 빈도 기반 토큰 필터링 
from nltk.tokenize import word_tokenize
from collections import Counter

text = "The quick brown fox jumps over the lazy dog. The fox is quick and agile."

tokens = word_tokenize(text)        # 단어 토큰화
word_counts = Counter(tokens)       # 토큰별 등장 횟수 계산

filtered_tokens = [token for token in tokens if word_counts[token] >=2 ]       # 2회 이상 나온 토큰만 남김
filtered_tokens
```

```python
# 길이가 3이상인 토큰만 남기기
filtered_tokens = [token for token in tokens if len(token) > 2 ]
filtered_tokens
```

```python
# 불용어 제거
from nltk.corpus import stopwords

en_stopwords = stopwords.words('english')   # 영어 불용어 리스트
en_stopwords
```

```python
text = "The quick brown fox jumps over the lazy dog. The fox is quick and agile."

tokens = word_tokenize(text)

filtered_tokens = [token for token in tokens if token.lower() not in en_stopwords]
filtered_tokens
```

```python
# 불용어 파일 로드 후 토큰 필터링

def load_stopwords(filepath):
    with open(filepath, 'r', encoding='UTF-8') as f:
        stopwords = [line.strip() for line in f]        # 줄바꿈 제거 후 한 줄 = 불용어 로 리스트 생성
    return stopwords

ko_stopwords = load_stopwords('ko_stopwords.txt')
print(len(ko_stopwords))
cleand_tokens = [token for token in tokens if token not in ko_stopwords]
cleand_tokens
```
