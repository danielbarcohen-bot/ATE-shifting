```python
# 10/29/20 fm
# This notebook demonstrates the "Baby Research Visualization" scenario described in 
#       https://secureailabs.atlassian.net/wiki/spaces/SAILConfluenceHome/pages/882016268/Computation+Roadmap

# prep, load data
import seaborn as sns

# note that the dataset is loaded as a pandas dataframe (because this is Seaborn)
df = sns.load_dataset('iris')

# plot it!
# see https://seaborn.pydata.org/generated/seaborn.boxplot.html for info on parameters
# note that Seaborn uses matplotlib, so parameters can get passed through to matplotlib function: matplotlib.axes.Axes.boxplot()
sns.boxplot( x=df["species"], y=df["sepal_length"] )
```

```python
# Discussion:
#
# This example, like with other "Baby" scenarios, is a very basic illustration, in this case, for visualization.
# Note that I chose Seaborn as the library - we could also have used matplotlib directly, since we do not need to 
# expose the visualization API. That is, most of this should be GUI in the future - at first, we'll have the researcher
# construct a function call with parameters.
#
# For this example, the function specification will be more like:
#   boxplot(x = 'species', y = 'sepal_length')
# i.e. there should be no references to the dataframe

# In real use cases, these visualizations can be used in roughly two different ways:
#  1. Producing visualizations to show a result of some research work product (i.e. visualization of some computation result)
#    1a. This might be driven programmatically (i.e. the "arbitrary code execution" scenario)
#    1b. ... or to a computation result, as a result of a previously executed "computation job". 
#    1c. In the "ML Platform" scenario, some visualization may be generated automatically (e.g. result of evaluating a series of models)
#  2. Visualization as an initial step for researchers to understand the data set; also can be used to figure out the 
#      data pre-processing steps. For this use case, we would like to automate as much as we can, so that the researcher
#      does not have to interactively generate different visualizations on the data set itself. See https://github.com/pandas-profiling/pandas-profiling
#      for an example. 

# The initial POC for visualization viewing from the Researcher Tool is to simply show some visualization output - 
# to have the infrastructure in place for passing image files from the Research engine to the UI. 
```
