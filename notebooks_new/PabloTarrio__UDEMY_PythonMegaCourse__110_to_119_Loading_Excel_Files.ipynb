# Establecemos una columna como "INDEX"
# El objeto que creamos así no modifica el original, por lo que el cambio no es permanente

```python
df8 = pandas.read_csv ("./supermarkets/data.txt", header= None)
df8.columns = ["ID","Address","City","ZIP","Country","Name", "Employees"]
df8.set_index("ID")
```

```python
df8
```

# Si queremos que el data frame tenga ese cambio de INDEX permanente, tenemos dos opciones:
#     1. Crear un nuevo dataframe con el index en una nueva variable

```python
df9 = df8.set_index("ID")
df9
```

#     2. Utilizar el parámetro "inplace=True" al usar el método set_index()
# Esta segunda opción tiene el inconveniente de que al establecer una columna como index no devuelve a columna el index que está puesto,
# simplemente lo elimina. Por ejemplo, si usasemos "Address" como index, la columna "ID" desaparecería.

```python
df8.set_index("ID", inplace=True)
df8
```
