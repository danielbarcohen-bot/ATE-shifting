## ?? Step 2: Load & Clean the Data

```python
import pandas as pd

# Load CSV file
df = pd.read_csv('Sample - Superstore.csv', encoding='ISO-8859-1')
df.head()
```

```python
# prompt: profit margin by category

# Calculate profit margin per order
df['Profit Margin'] = df['Profit'] / df['Sales']

# Group by category and calculate the average profit margin
profit_margin_by_category = df.groupby('Category')['Profit Margin'].mean().reset_index()

# Display the result
print("Average Profit Margin by Category:")
print(profit_margin_by_category)

# Optional: Sort by profit margin
profit_margin_by_category_sorted = profit_margin_by_category.sort_values(by='Profit Margin', ascending=False)
print("\nAverage Profit Margin by Category (Sorted):")
profit_margin_by_category_sorted
```

## ?? Step 3: Business Analytics Tasks
### ?? A. Total Sales by Region

Total Sales were highest in the West and Lowest in the South.

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
region_sales.plot(kind='bar', title='Total Sales by Region')
```

### ?? B. Average Profit Margin by Category

Technology has the highest profit margin.

```python
category_margin = df.groupby('Category')['Profit Margin'].mean()
category_margin.plot(kind='bar', title='Avg. Profit Margin by Category')
```

### ?? C. Top 5 Products by Sales

The top product by sales is the Canon Image Class 2200 Advanced copier.

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(5)
top_products.plot(kind='bar', title='Top 5 Products by Sales')
```

## ?? Step 4: Define and Calculate 2 KPIs
### ? KPI 1: Total Sales and Profit by Region

```python
df.groupby('Region')[['Sales', 'Profit']].sum()
```

### ? KPI 2: Average Delivery Time

```python
df['Delivery Time (Days)'] = (df['Ship Date'] - df['Order Date']).dt.days
df.groupby('Region')['Delivery Time (Days)'].mean()
```

```python
# Convert date columns to datetime
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

# Create Profit Margin
df['Profit Margin'] = df['Profit'] / df['Sales']

# Check missing values
df.isnull().sum()
```
