```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn import linear_model, tree, ensemble, svm, metrics
from sklearn.preprocessing import LabelEncoder, scale
from sklearn.model_selection import train_test_split
from collections import Counter
from xgboost import XGBRegressor
import pickle

# Load dataset
data = pd.read_csv('traffic volume.csv')

# Basic info
data.head()
data.describe()
data.info()

# Check for missing values
data.isnull().sum()

# Fill missing values
#data['temp'].fillna(data['temp'].mean(), inplace=True)
#data['rain'].fillna(data['rain'].mean(), inplace=True)
#data['snow'].fillna(data['snow'].mean(), inplace=True)
#data['weather'].fillna('Clouds', inplace=True)
data.fillna({
    'temp': data['temp'].mean(),
    'rain': data['rain'].mean(),
    'snow': data['snow'].mean()
}, inplace=True)

data['weather'] = data['weather'].fillna('Clouds')


# Check weather category counts
print(Counter(data['weather']))

#data.select_dtypes(include=['float64', 'int64']).corr()
sns.heatmap(cor)
sns.pairplot(data)

# Date-time features
data['datetime'] = pd.to_datetime(data['date'] + ' ' + data['Time'], dayfirst=True)
data['day'] = data['datetime'].dt.day
data['month'] = data['datetime'].dt.month
data['year'] = data['datetime'].dt.year
data['hours'] = data['datetime'].dt.hour
data['minutes'] = data['datetime'].dt.minute
data['seconds'] = data['datetime'].dt.second

# Label Encoding
for col in ['holiday', 'weather']:
    if data[col].dtype == 'object':
        le = LabelEncoder()
        data[col] =  Data Visualization (keep as-is)
cor = le.fit_transform(data[col])

# Select features
columns_to_scale = ['holiday', 'temp', 'rain', 'snow', 'weather',
                    'day', 'month', 'year', 'hours', 'minutes', 'seconds']
x = data[columns_to_scale]
y = data['traffic_volume']

# Scaling
x_scaled = scale(x)
x = pd.DataFrame(x_scaled, columns=columns_to_scale)

# Split the data
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Check the shape of the splits
print("Training features shape:", x_train.shape)
print("Testing features shape:", x_test.shape)
print("Training labels shape:", y_train.shape)
print("Testing labels shape:", y_test.shape)

# Train models
LR = linear_model.LinearRegression()
Dtree = tree.DecisionTreeRegressor()
Rand = ensemble.RandomForestRegressor()
svr = svm.SVR()
XGB = XGBRegressor()

# Fit models
LR.fit(x_train, y_train)
Dtree.fit(x_train, y_train)
Rand.fit(x_train, y_train)
svr.fit(x_train, y_train)
XGB.fit(x_train, y_train)

# Predict on train set
p1_train = LR.predict(x_train)
p2_train = Dtree.predict(x_train)
p3_train = Rand.predict(x_train)
p4_train = svr.predict(x_train)
p5_train = XGB.predict(x_train)

# Print R2 scores on training data
print("\nTraining R2 Scores:")
print("Linear Regression:", metrics.r2_score(y_train, p1_train))
print("Decision Tree:", metrics.r2_score(y_train, p2_train))
print("Random Forest:", metrics.r2_score(y_train, p3_train))
print("SVR:", metrics.r2_score(y_train, p4_train))
print("XGBoost:", metrics.r2_score(y_train, p5_train))

# Predict on test set
p1 = LR.predict(x_test)
p2 = Dtree.predict(x_test)
p3 = Rand.predict(x_test)
p4 = svr.predict(x_test)
p5 = XGB.predict(x_test)

# Print R2 scores on test data
print("\nTesting R2 Scores:")
print("Linear Regression:", metrics.r2_score(y_test, p1))
print("Decision Tree:", metrics.r2_score(y_test, p2))
print("Random Forest:", metrics.r2_score(y_test, p3))
print("SVR:", metrics.r2_score(y_test, p4))
print("XGBoost:", metrics.r2_score(y_test, p5))

# Calculate RMSE for Random Forest
MSE = metrics.mean_squared_error(y_test, p3)
RMSE = np.sqrt(MSE)
print("\nRandom Forest RMSE:", RMSE)

# Save best model (Random Forest) and encoder
pickle.dump(Rand, open("model.pkl", 'wb'))
pickle.dump(le, open("encoder.pkl", 'wb'))
```
