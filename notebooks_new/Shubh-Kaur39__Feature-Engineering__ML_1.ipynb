Q.6 How do we handle categorical variables in Machine Learning? What are the common techniques?

-> Machine Learning models doesn't understand categorical values So, we handle this by converting the values into numerical form. This process of converting categorical variables into numerical variables is called as encoding. The most common techniques in encoding are:

1.) One-hot encoding (Nomial Encoding):

It creates binary columns for each unique value of the categorical variable. It is useful for nominal data when their is no order. However, if we have a lot of categories then many new columns will be made which cause the curse of dimentionality.

2.) Label Encoding or Ordinal Encoding:

In Label Encoding we convert each category into a unique integer so their is no curse of dimentionality. However, it is good for ordinal data since for nominal data it can learn patterns from the integers.
In Ordinal Encoding we explicitly give order like low < med < high. Since, we provide order explicitely, it ensures model sees true order rather then making itself.

3.) Target Guided Ordinal Encoding:

It is based on the relationship with the target variable. We first make groups for each category and then impute the value with the mean or median of the corresponding target variable. It is useful when we have large number of unique categories in categorical value.

Q.16 What is an Optimizer? What are different types of optimizers? Explain each with an example.

-> Optimizer are the functions or algorithms which are used to update the weights or internal parameters by comparing the loss value. It not only ensures minimum loss value and high accuracy scores but also helps to decide the direction and speed to learn a model. The different types of optimizer are listed below:

i) Gradient-descent : It calculates the gradient of entire dataset and then update the weights at last.

Example: when we find the lowest point in the graph, we check the value at every point in range and then gave the minimum one.

ii) Stochastic Gradient-descent: It calculates the gradient after every data sample.

Example: In this we check only one point ahead and move towards the lower value.

iii) Mini-Batch Gradient descent: It calculated the gradient on a given particular batch and then update the weights.

Example: In this approach, we look small parts of the graph before taking each step forward.

iv) AdaGrad: It lowers the learning rate for each parameter and is inversely proportional to the square root of sum of squared past gradient values. It is only good for sparse data.

Example:If the slope is too steap it starts taking small step and vice-versa.

Q.21 What is feature scaling? How does it help in Machine Learning?

-> Feature scaling is a process of rescaling the features on the same scale or range. It ensures that all features contribute equally to the model.

 It helps in Machine Learning since, interpretation became easier when all features are on same scale. Moreover, in M.L many of the algorithms are distance based, so lowering the magnitude can help in optimization. It also prevents dominance of one feature as the feature with high values can overpower the feature with low value.

Q.22 How do we perform scaling in Python?

-> In python, scaling is done using functions from sklearn.preprocessing module. But let's first see the type of scaling:

i) Standardization: It normalizes each feature value to have mean = 0 and variance = 1. In python, StandardScaler() is used to initialize the scaler and then we fit and transform this scaler.

ii) Normalization: (also known as Min-Max scaler) It rescales the value to a fixed range like 0 and 1. In python, MinMaxScaler() is used to initialize the scaler and then we fit and transform this scaler.

iii) Unit Vector: It is useful when only direction matters and not magnitude. In python, we do this using normalize() function from sklearn.preprocessing. In this we don't need to fit or transform, we just call the function.
