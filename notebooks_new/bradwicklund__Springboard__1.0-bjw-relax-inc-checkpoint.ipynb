### 3.1 Create Target Feature: "Adopted"

```python
# Normalize to dates
engagement['date'] = engagement['time_stamp'].dt.normalize()

# Drop duplicates (if a user somehow has multiple rows for the same day)
engagement = engagement.drop_duplicates(subset=['user_id', 'date'])

# Container for results
adopted_list = []

# Loop through each user
for user_id, group in engagement.groupby('user_id'):
    # Sort that user's active dates
    dates = sorted(group['date'].tolist())
    
    adopted = 0  # default
    
    # Check 7-day windows
    for i in range(len(dates) - 2):   # need at least 3 dates
        # Look at a slice of 3 dates
        window = dates[i:i+3]
        
        # If the first and third date are within 7 days → adopted
        if (window[-1] - window[0]).days <= 7:
            adopted = 1
            break
    
    adopted_list.append({'user_id': user_id, 'adopted': adopted})

# Turn into a DataFrame
adopted_df = pd.DataFrame(adopted_list)
```

### 3.2 Merge Tables

```python
users_with_adopted = users.merge(adopted_df, left_on='object_id', right_on='user_id', how='left')
```

### 4.0 Modeling

### 4.1 Preprocessing Cleanup

```python
# Create copy prior to modeling
df = users_with_adopted.copy()
```

```python
# Check dtypes
df.info()
```

```python
# Flag invited users
if 'invited' not in df.columns:
    df['invited'] = 0
```

```python
# Treat users with no engagement as not adopted
df['adopted'] = df['adopted'].fillna(0).astype(int)
```

```python
# Create helper flag
df['has_session'] = df['last_session_creation_time'].notna().astype(int)
```

```python
# Create reference date
ref_date = pd.to_datetime(
    pd.Series([df['creation_time'].max(), df['last_session_creation_time'].max()]).max()
)
```

```python
# Time based features
df['account_age_days'] = (ref_date - df['creation_time']).dt.days
df['days_since_last_session'] = (ref_date - df['last_session_creation_time']).dt.days
```

```python
# Keep creation source as category
df['creation_source'] = df['creation_source'].astype('category')
```

```python
df.info()
```

```python
# Drop features with high cardinality or not needed for modeling
drop_cols = ['object_id','user_id','name','email',
             'creation_time','last_session_creation_time','org_id']

df_model = df.drop(columns=[c for c in drop_cols if c in df.columns])

df_model.head()
```

### 4.2 Modeling

```python
# Define target
X = df_model.drop(columns=['adopted'])
y = df_model['adopted']
```

```python
# Train-test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42
)
```

```python
# Preprocess and Logistic Regression
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

num_features = ['account_age_days','days_since_last_session']
bin_features = ['invited','opted_in_to_mailing_list','enabled_for_marketing_drip']
cat_features = ['creation_source']

preprocess = ColumnTransformer(
    transformers=[
        ('num', SimpleImputer(strategy='median'), num_features),
        ('bin', 'passthrough', bin_features),
        ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), cat_features),
    ]
)

model = Pipeline(steps=[
    ('prep', preprocess),
    ('clf', LogisticRegression(max_iter=1000, class_weight='balanced'))
])

model.fit(X_train, y_train)
```

```python
# Evalute
from sklearn.metrics import classification_report, roc_auc_score

y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:,1]

print("ROC-AUC:", roc_auc_score(y_test, y_proba).round(3))
print(classification_report(y_test, y_pred, digits=3))
```

```python
# Get feature names
ohe = model.named_steps['prep'].named_transformers_['cat']
cat_names = list(ohe.get_feature_names_out(cat_features))
feature_names = num_features + bin_features + cat_names

coefs = model.named_steps['clf'].coef_.ravel()
odds = np.exp(coefs)

coef_df = pd.DataFrame({
    'feature': feature_names,
    'coef': coefs,
    'odds_ratio': odds
}).sort_values('odds_ratio', ascending=False)

coef_df
```
