Angenommen wir erstellen eine Sicht auf die Relation ```ManagerIn(Name, Adresse, ManagerinID, Gehalt)```, welche alle Manager\*Innen mit einem ```Gehalt``` gr??er als 2 Millionen beinhaltet. 

````python
%sql
CREATE VIEW Reiche AS
    SELECT Name, Gehalt, ManagerinID
    FROM ManagerIn
    WHERE Gehalt > 2000000;
````

````python
%sql
UPDATE Reiche SET Gehalt = 1500000
WHERE ManagerinID = 25;
````

Um Tupelmigration zu verhindern, k?nnen problematische Update durch ```WITH CHECK OPTION``` abgelehnt werden.
