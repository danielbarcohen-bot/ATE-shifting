```python
import os
import yaml

folder_path = 'C:/Users/gthir/Desktop/Data_Driven_stock_project/data/2023-11'  # change this to your folder

comb_yamil = []

for filename in os.listdir(folder_path):
    if filename.endswith('.yaml') or filename.endswith('.yml'):
        file_path = os.path.join(folder_path, filename)
        with open(file_path, 'r') as file:
            data = yaml.safe_load(file)
            if data:
                comb_yamil.extend(data)  # for list-type YAML files
```

```python
folder_yamil = Path(r'D:/Data_Driven_stock_project/data')  # Main folder path
# Use rglob to recursively search for all YAML files in subfolders
yaml_files = [f for f in folder_yamil.rglob('*.yaml')] # uses list comprehension to collect all ...
#.yamil files from folder directory and alll its subfolders
comb_yamil=[] # create a list to store all the dictionary
# Loop through each YAML file
for file_links in yaml_files: 
    with open(file_links, 'r') as f:
        data = yaml.safe_load(f) 
        if data: #data is a list of dicts, so you want to flatten them all into one big list.
            comb_yamil.extend(data)  # extend() adds each dict inside the list, 
                                     # while append() would add the whole list as a single 
                                     # item (which you don?t want).
```

```python
df=pd.DataFrame(comb_yamil)
df.to_csv("comb_stock.csv",index=False)
print("Merged YAML data saved to comb_stock.csv")
```

```python
#save the Ticker data in separate csv files under one folder

import os

# Create a folder to store output files
output_folder = 'nifty50'
os.makedirs(output_folder, exist_ok=True)

# Loop through each unique ticker
for ticker in df['Ticker'].unique(): # here picks each unique ticker name is SBIN', 'BAJFINANCE', 'TITAN', 'ITC ...
    ticker_df = df[df['Ticker'] == ticker]  # filter dat
    file_name = f"{ticker}.csv"             # create filename like SBIN.csv...so on...
    file_path = os.path.join(output_folder, file_name)  # full path
    ticker_df.to_csv(file_path, index=False)  # save CSV
    print(f"Saved: {file_name}")
```
