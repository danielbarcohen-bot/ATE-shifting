# Step 4: Baseline Evaluation

```python
baseline_data = df.copy(deep=True)
X_baseline = baseline_data.drop('y',axis=1)
y_baseline = baseline_data.y
```

```python
X_train_baseline,X_test_baseline, y_train_baseline, y_test_baseline = train_test_split(X_baseline, y_baseline, test_size=0.25)
```

```python
dummyClassifier = DummyClassifier(strategy="most_frequent", random_state=42)
dummyClassifier.fit(X_train_baseline, y_train_baseline)
```

```python
y_pred_baseline = dummyClassifier.predict(X_test_baseline)
baseline_balanced_score = balanced_accuracy_score(y_test_baseline, y_pred_baseline)
baseline_accuracy_score = accuracy_score(y_test_baseline, y_pred_baseline)
# baseline_roc = roc_auc_score(y_test_baseline, y_pred_baseline)
baseline_cm = confusion_matrix(y_test_baseline, y_pred_baseline)
baseline_plot = ConfusionMatrixDisplay(baseline_cm, display_labels=dummyClassifier.classes_)
baseline_plot.plot()
baseline_report = classification_report(y_test_baseline, y_pred_baseline)
print(baseline_report)
print(baseline_balanced_score, baseline_accuracy_score)
```

# Step 5: Preprocessing and Feature Engineering

```python
numerical_columns = df.select_dtypes(include="int64").columns
categorical_columns = df.select_dtypes(exclude="int64").columns
```

```python
data = df.copy(deep=True)
```

```python
labelEncoder = LabelEncoder()
data['y'] = labelEncoder.fit_transform(data['y'])
```

```python
data
```

```python
data.info()
```

```python
X = data.drop('y', axis=1)
y = data['y']
```

```python
numerical_columns = X.select_dtypes(include="int64").columns
categorical_columns = X.select_dtypes(exclude="int64").columns
```

```python
print(numerical_columns)
```

```python
print(categorical_columns)
```

```python
week_days = ['mon', 'tue', 'wed', 'thu', 'fri', 'sat', 'sun']
months = list(map(str.lower, ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']))
```

```python
def numerize(df, col,tuple):
    if type(df[col].iloc[0]) == str:
        df[col] = df[col].apply(lambda x: tuple.index(x))
```

```python
numerize(X, 'day_of_week', week_days)
numerize(X, 'month', months)
```

```python
X
```

```python
poutcome = ('failure','nonexistent','success')
contact = ('cellular','telephone')
loan = ('no','yes','unknown')
numerize(X, 'poutcome', poutcome)
numerize(X, 'contact', contact)
numerize(X, 'loan', loan)
```

```python
X
```

```python
job=('admin.','blue-collar','entrepreneur','housemaid','management','retired','self-employed','services','student','technician','unemployed','unknown')
marital=('divorced','married','single','unknown')   # note: 'divorced' means divorced or widowed)
education=('basic.4y','basic.6y','basic.9y','high.school','illiterate','professional.course','university.degree','unknown')
default=('no','yes','unknown')
housing=('no','yes','unknown')
numerize(X, 'job', job)
numerize(X, 'marital', marital)
numerize(X, 'education', education)
numerize(X, 'default', default)
numerize(X, 'housing', housing)
```

```python
X
```

```python
X.info()
```

```python
standardScaler = StandardScaler()
decisionTree = DecisionTreeClassifier(random_state=42)
```

```python
pipeline = Pipeline(steps=[
    ('standardScaler', standardScaler),
    ('decisionTree', decisionTree)
])
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.25, random_state=42)
```

```python
pipeline.fit(X_train,y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

```python
balanced_score = balanced_accuracy_score(y_test, y_pred)
accuracy_score = accuracy_score(y_test, y_pred)
# baseline_roc = roc_auc_score(y_test_baseline, y_pred_baseline)
cm = confusion_matrix(y_test, y_pred)
cm_plot = ConfusionMatrixDisplay(baseline_cm, display_labels=dummyClassifier.classes_)
cm_plot.plot()
report = classification_report(y_test, y_pred)
print(report)
print(balanced_score, accuracy_score)
```

```python
def dump_file(obj, file_name):
    with open(file_name,"wb") as file:
        pickle.dump(obj, file)
```

```python
dump_file(obj=pipeline,file_name='pipeline.pkl')
```
