# Globale Variablen

In Python k?nnen Variablen mit zwei Geltungsbereichen deklariert werden: global und lokal.

* Lokale Variablen: Diese werden innerhalb einer Funktion oder Methode deklariert. Auf sie kann nur innerhalb dieser Funktion oder Methode zugegriffen und sie ver?ndert werden. Sobald die Ausf?hrung der Funktion abgeschlossen ist, wird die lokale Variable zerst?rt und kann au?erhalb der Funktion nicht mehr aufgerufen werden.

* Globale Variablen: Diese werden au?erhalb aller Funktionsdefinitionen deklariert. Sie sind im gesamten Programm zug?nglich, auch innerhalb von Funktionen. Jede Funktion oder Methode kann auf eine globale Variable zugreifen und sie ver?ndern.

Der Hauptunterschied zwischen ihnen liegt in ihrem Geltungsbereich. Globale Variablen haben einen programmweiten Geltungsbereich, w?hrend lokale Variablen einen auf die Funktion, in der sie deklariert sind, begrenzten Geltungsbereich haben.

```python
# Beispiel f?r lokale Variablen (unerw?nschtes Verhalten)
value = 10

def double_variable(x):
    """
    Double the value.

    Parameters:
    x (int or float): The number to be doubled.

    Returns:
    None
    """
    x = x*2
    
double_variable(value)
print(value)
```

```python
# Beispiel f?r lokale Variablen (erw?nschtes Verhalten)
value = 10

def double_variable(x):
    """
    Return double the value.

    Parameters:
    x (int or float): The number to be doubled.

    Returns:
    int or float: The doubled value of the input number.
    """
    return x*2
    
value = double_variable(value)
print(value)
```

```python
# Beispiel f?r globale Variable (erw?nschtes Verhalten, aber siehe Warnung unten)
value = 10

def double_variable():
    """
    Double the value of the global variable 'value'.

    It doesn't take any parameters. The function directly modifies the global variable 'value'.

    Returns:
    None
    """
    
    # Please note that this function relies on the existence of a global variable 
    # named ?value?, and it will raise a NameError if ?value? is not defined in 
    # the global scope.
    global value 
    value *= 2
    
double_variable()
print(value)
```

Von der Verwendung globaler Variablen wird jedoch im Allgemeinen aus mehreren Gr?nden abgeraten:

* Schwer zu debuggen: Globale Variablen k?nnen von jeder Stelle im Programm aus ge?ndert werden, so dass es schwierig ist, nachzuvollziehen, woher ein Wert kommt oder ge?ndert wird.
* Gefahr von Namenskonflikten: Wenn Sie denselben Namen f?r eine globale und eine lokale Variable verwenden, kann es zu unerwartetem Verhalten kommen.
* Macht den Code weniger flexibel: Funktionen, die von globalen Variablen abh?ngen, sind weniger flexibel und schwieriger wiederzuverwenden, da sie von dem spezifischen Kontext abh?ngen, der durch diese globalen Variablen bereitgestellt wird.

Anstatt globale Variablen zu verwenden, ist es oft besser, Funktionsargumente und R?ckgabewerte zu verwenden, um Daten in Ihrem Programm weiterzugeben. Dadurch wird Ihr Code modularer und ist leichter zu verstehen, zu debuggen und zu testen. Wenn Sie feststellen, dass Sie viele globale Variablen verwenden, k?nnte dies ein Zeichen daf?r sein, dass Sie Ihren Code umstrukturieren oder Klassen zur Kapselung des Zustands verwenden sollten.
