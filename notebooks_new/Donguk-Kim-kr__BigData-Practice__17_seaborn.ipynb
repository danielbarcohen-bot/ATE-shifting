# 붓꽃 데이터 시각화

```python
import matplotlib.pyplot as plt
import koreanize_matplotlib

iris.head()
```

## 꽃받침의 길이 시각화 - 막대그래프

```python
iris.sepal_length[:10].plot(kind='bar', rot=0) # 10개의 데이터만 막대그래프, 회전 0도
plt.title('꽃 받침의 시각화') # 차트 제목
plt.xlabel('Data') # x축 제목
plt.ylabel('꽃 받침의 길이') # y축 제목
plt.show()
```

## 붓꽃의 품종별 그룹화 - 평균 - 새로운 데이터프레임으로 저장 -> 차트 생성

```python
df2 = iris.groupby(iris['species']).mean()
df2
```

### 품종별 꽃받침 길이 그래프

```python
df2.sepal_length[:].plot(kind='bar', rot=0)
plt.title('품종별 꽃받침 길이 그래프')
plt.xlabel('품종')
plt.ylabel('꽃받침 길이의 평균값')
plt.show()
```

# 히스토그램(빈도수 그래프)

```python
iris.plot(kind='hist')
plt.title('각 특징 값들의 빈도수 히스토그램')
plt.show()
```

## 상자 그래프 --> 이상치 발견

```python
iris.plot(kind='box')
plt.title('각 특징 값들의 빈도수에 대한 박스 플롯')
plt.xlabel('특징')
plt.ylabel('데이터 값')
plt.show()
```

## 산점도 그래프 (산포그래프)

```python
iris.plot(x='sepal_length', y='petal_length', kind='scatter')
plt.show()
```

# 4. 타이타닉

### 4-1) 성별에 따른 생존율

```python
titanic = pd.read_csv('train.csv')
```

```python
titanic
```

```python
titanic_2 = titanic[['Sex', 'Survived']] # 성별, 생존율
titanic_2
```

```python
titanic_2.groupby('Sex').mean()
```

### 4-2) 함께 승선한 부모와 자식 수 (Parch)에 따른 생존율

```python
titanic_3 = titanic[['Parch', 'Survived']] # 부모와 자식 수, 생존율
titanic_3
```

```python
titanic_3.groupby('Parch').mean()
```

```python
titanic_3.groupby('Parch', as_index=False).mean()
```

```python
titanic_3.groupby('Parch', as_index=False).mean().sort_values('Survived', ascending=False)
```

### 4-3) 객실 등급(Pclass), 성별(Sex) 컬럼별로 생존율, 나이의 평균값
- 그룹 --> 2개
- 평균값 --> 2개

```python
titanic_4 = titanic[['Pclass', 'Sex', 'Survived', 'Age']]
titanic_4
```

```python
titanic_4.groupby(['Pclass', 'Sex']).mean()
```

#### 데이터프레임명.groupby(['그룹할 컬럼1', '그룹할 컬럼2])[['계산하고 싶은 컬럼1, 게산하고 싶은 컬럼2]].mean()

```python
titanic.groupby(['Pclass', 'Sex'])[['Survived', 'Age']].mean()
```

### 4-4) 항구별 생존자
