```python
# ds0 is the dataset of the actual study population (first-order features)
ds0 = Dataset.from_cached_db(
    db, config.dataset_options, config.database_options)
ds0

# itemsets are the ones actually used for analysis
itemsets = pd.read_csv('stage_fhp_ffm/selected_itemsets.csv', index_col=[0], header=[0, 1])

# mapping features for ds0
fm = FeatureMapper(cmax=4, fixed_itemsets=itemsets)
ds = fm.fit_transform(ds0)
```

```python
# From the Dataset containing features from T1 and targets from T2
# we build a new dataset, containing (mapped) features (from T1)
# AND original (single, unmapped) diagnoses from T1,
# all together as the dataset's features (ds.X).
# ds.Y contains the unmodified (non-monotonic) targets from T2.
joined_ds = Dataset(X=ss.hstack([ds.X, ds0.X]), Y=ds.Y)
joined_ds.feature_names = \
    np.concatenate([ds.feature_names,
                    list(map(lambda f: 'o' + f, ds0.feature_names))])
joined_ds.target_names = ds.target_names
joined_ds.obs_names = ds.obs_names
ds = joined_ds
```

```python
def stats_for(var):
    mi = ds.binary_mutual_information(var)
    ci = ds.co_information(var)
    tc = ds.total_correlation(var)
    phi_min, phi_max = ds.phi_coefficient(var, agg=[np.min, np.max])
    igp_min, igp_max = ds.igp(var, agg=[np.min, np.max])
    print("{var0}; {others}:\n   "
          "mi={mi:+0.2e} igp=[{igp_min:+0.2e}, {igp_max:+0.2e}]\n   "
          "ci={ci:+0.2e} phi=[{phi_min:+0.2e}, {phi_max:+0.2e}] tc={tc:+0.2e}"
          .format(var0=var[0], others=', '.join(var[1:]),
                  mi=mi, ci=ci, phi_min=phi_min, phi_max=phi_max,
                  tc=tc, igp_min=igp_min, igp_max=igp_max))
```

```python
# our standard feature names
feature_names = [f for f in ds.feature_names if 'o' not in f]
```

```python
# we transform the names to groups of o-Names
ho_orig_feat = [['f_o' + s for s in feat.split('_')] for feat in feature_names if len(feat.split('_')) > 2]
display(ho_orig_feat)
```

```python
fn = 'live-analysis-stats.csv'
if os.path.isfile(fn):
    stats_df = pd.read_csv(fn, index_col=[0])
else:
    stats = [[ds.ci(x), *ds.igp(x, agg=[np.min, np.max])]
             for x in tqdm(ho_orig_feat)]
    stats_df = pd.DataFrame(stats, columns=['ci', 'igp_min', 'igp_max'],
                            index=pd.Index([f for f in feature_names if len(f.split('_')) > 2], name='name'))
    stats_df = stats_df.join(itemsets, how='left')
    stats_df.to_csv('live-analysis-stats.csv')
```

```python
stats_df.columns = ['ci', 'igp_min', 'igp_max', 'support_mean', 'metadata_mean', 'support_std', 'metadata_std', 'support_count', 'metadata_count', 'card']
```

### Re-compute Apriori metadata for the full study population

```python
### APRIORI WITH -AN OPTION
ap = Apriori(cmax=4, smin=30, apriori_options="-ei -v' (%a) [%e]' -an -d0")
app = ap.process(ds0.X)
itemsets, supports, metadata_an = zip(*app)

# put Apriori output into a Dataframe
app_df_an = pd.DataFrame({
    'itemsets': itemsets,
    'support': supports,
    'metadata_an': metadata_an,
})

### APRIORI WITH -AM OPTION
ap = Apriori(cmax=4, smin=30, apriori_options="-ei -v' (%a) [%e]' -am -d0")
app = ap.process(ds0.X)
itemsets, supports, metadata_am = zip(*app)

# put Apriori output into a Dataframe
app_df_am = pd.DataFrame({
    'itemsets': itemsets,
    'support': supports,
    'metadata_am': metadata_am,
})
```

```python
app_df = pd.concat([app_df_an.set_index('itemsets'),
                    app_df_am.set_index('itemsets')], axis=1)
app_df
```

```python
app_df['card'] = app_df.index.map(len)
app_df = app_df.query('card > 1')
app_df['phenotype'] = app_df.index.map(lambda i: Feature(frozenset(ds0.feature_names[list(i)])))
app_df.set_index('phenotype', inplace=True)
```
