```python
##Visualization modules
import pandas
from pandas import DataFrame
import matplotlib.pyplot as plt ##Creates an alias
##Machine learning modules
from sklearn.linear_model import LinearRegression
```

```python
data = pandas.read_csv('cost_revenue_dirty.csv')
```

```python
#Shows summary statistics
data.describe()
```

## Visualization

```python
#Creates pandas dataframes
X = DataFrame(data, columns=['production_budget_usd'])
y = DataFrame(data, columns=['worldwide_gross_usd'])
```

```python
#Creates a scatterplot using matplotlib
plt.figure(figsize=(10,6))
plt.scatter(X,y, alpha=0.3)
plt.title('Film Cost vs Global Revenue')
plt.xlabel('Production Budget $')
plt.ylabel('Worldwide Gross $')
plt.ylim(0, 3000000000)
plt.xlim(0, 450000000)
plt.show() 
```

## Evaluation and Analysis

```python
#Linear regression - Residual sum of squares
```

```python
#Creates a linear regression object
regression = LinearRegression()
regression.fit(X, y) #Inputs features = X , labels = y
```

Slope Coefficient:

```python
#Slope or m
regression.coef_ # theta_1
# for every $1 invested, $3.11 is made
```

Intercept Coefficient:

```python
#Intercept or b
regression.intercept_
# a budget of $0 will lose this much money 
# This is a dramatic simplification. Take the prediction with a grain of salt. 
```

Fits line into the scatter plot

```python
plt.figure(figsize=(10,6))
plt.scatter(X,y, alpha=0.3)
plt.plot(X, regression.predict(X), color='red', linewidth=4) #plots regression line

plt.title('Film Cost vs Global Revenue')
plt.xlabel('Production Budget $')
plt.ylabel('Worldwide Gross $')
plt.ylim(0, 3000000000)
plt.xlim(0, 450000000)
plt.show() 
```

## R<sup>2</sup> goodness of fit

```python
regression.score(X, y) #Inputs features = X , labels = y
# 55% of the data can be explained by the model.
```
