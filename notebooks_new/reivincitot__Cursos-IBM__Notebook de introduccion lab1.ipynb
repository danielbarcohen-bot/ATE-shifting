## Leer la información

Usaremos la función pandas.read_csv() para leer el archivo csv.en los brackets, ponemos la dirección del archivo entre comillas, asi pandas leerá el archivo en un dataframe desde esa dirección. La dirección puede ser una Url o una dirección local en tu computador.

Ya que la información no incluye encabezados, podemos agregar el argumento `headers = None` dentro del método `read_csv()` asi pandas no asignará de manera automática la primera fila como encabezados.

También puedes asignar al dataset cualquier variable que crees.

```python
import urllib.request

path= "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-SkillsNetwork/labs/Data%20files/auto.csv"
archivo_destino = "auto.csv"

urllib.request.urlretrieve(path,archivo_destino)
print(f"Se ha descargado el archivo {archivo_destino} correctamente")
```

```python
# Importamos la librería de pandas
import pandas as pd 

# Leemos el archivo en linea a traves de la url proporcionada arriba y asignamos la variable "df"

df = pd.read_csv(path, header=None)
```

Luego de leer el dataset, podemos usar el método dataframe.head(n) para verificar las filas n superiores del dataframe, donde n es un integer. Contrario a dataframe.head(n), dataframe.tail(n) mostrara las filas del final del dataframe.

```python
print("The first 5 rows of the dataframe")
df.head(5)
```

## Agregar Encabezados

Hecha un vistazo a nuestro dataset. Pandas automáticamente ajusta el encabezado con un integer que comienza en 0.

Para describir de mejor manera nuestra información, podemos introducir encabezados. Esta información esta disponible en: https://archive.ics.uci.edu/ml/datasets/Automobile.
Por lo tanto, debemos agregarla de forma manual.

Primero, creamos una lista "headers" que incluye el nombre de todas las columnas en orden. Luego, usamos `dataframe.columns = headers` para reemplazar los por la lista que creamos.

```python
# En el lab original ya estaban separadas y ordenadas como correspondían, pero yo no quería copiar y pegar y me dio pereza abrir y cerrar comillas asi que lo hice de esta manera.

headers = "symboling, normalized-losses, make, fuel-type, aspiration, num-of-doors, body-style, drive-wheels, engine-location, wheel-base, length, width, height, curb-weight, engine-type, num-of-cylinders, engine-size, fuel-system, bore, stroke, compression-ratio, horsepower, peak-rpm, city-mpg, highway-mpg, price"

headers=[header.strip() for header in headers.split(",")]

print(headers)
```

Reemplazamos los headers y revisamos nuestro dataframe.

```python
df.columns = headers
df.head(10)
```

Necesitamos reemplazar el símbolo *`"?"`* por NaN así que dropna() removerá los valores faltantes:

```python
df1 = df.replace('?',np.nan)
```

Podemos remover los valores faltantes de la columna "price" de la siguiente manera:

```python
df = df1.dropna(subset=['price'], axis=0)
df.head(20)
```

Ahora, hemos leído exitosamente nuestro dataset en bruto y agregamos correctamente los encabezados en el dataframe.

## Descripción

Si queremos obtener un sumario estadístico de cada columna e.j. el valor promedio de la columna `count`, la desviación estándar de la columna, etc usamos el método describe.

**`dataframe.describe()`**

Este método nos proporcionara varios sumarios estadísticos, excluyendo los valores `NaN` (Not a Number).

```python
df.describe()
```

Esto muestra el sumario estadístico de todas las columnas de tipo numéricos(int, float).

Por ejemplo, el atributo "symboling" tiene un conteo de 205, el valor promedio de esta columna es de 0.83, la desviación estándar es 1.25, el valor mínimo es -2, el percentil 25 es 0 el percentil 50 es 1 el percentil 75 es 2 y el valor máximo es 3

Sin embargo, ¿y si también quisiéramos revisar todas las columnas, incluidas aquellas que son de tipo objeto?

Puedes agregar un argumento `include= all` dentro del bracket, intentemos lo de nuevo.

```python
# Describe todas las columnas en "df"

df.describe(include="all")
```

Ahora nos entrega un sumario de todas las columnas, incluyendo los atributos de tipo object.

Ahora también podemos ver cuantos valores únicos hay. Cual es el valor mas alto y cual es la frecuencia de ese valor en las columnas de tipo object.

Algunos valores en la tabla de arriba son mostrados como "NaN", esto es por que no se encuentran disponibles con respecto a un tipo de columna en particular.
