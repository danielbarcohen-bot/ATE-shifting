```python
# Importaci?n de librer?as necesarias
import pandas as pd

# Cargar el archivo amazonbaby5000.csv (aseg?rate de subirlo a tu entorno de Colab)
df = pd.read_csv('../data/amazonbaby5000.csv')

# Verificar los valores ?nicos en la columna rating
print("Valores ?nicos en la columna 'rating':", df['rating'].unique())

# Mostrar las primeras filas del dataframe para verificar la carga correcta
df.head()
```

```python
import torch

print("Versi?n de PyTorch:", torch.__version__)
print("CUDA disponible:", torch.cuda.is_available())
print("Nombre de la GPU:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "No hay GPU disponible")
```

```python
# Importaci?n de la funci?n de partici?n de scikit-learn
from sklearn.model_selection import train_test_split
from datasets import Dataset

# Partici?n de los datos (80% entrenamiento, 20% prueba)
train_texts, test_texts, train_labels, test_labels = train_test_split(df['review'], df['rating'], test_size=0.2, random_state=42)

# Convertir los datos a formato de Dataset de Hugging Face
train_dataset = Dataset.from_dict({'text': train_texts, 'label': train_labels})
test_dataset = Dataset.from_dict({'text': test_texts, 'label': test_labels})

# Mostrar el tama?o de los conjuntos de datos
print(f"Tama?o del conjunto de entrenamiento: {len(train_dataset)}")
print(f"Tama?o del conjunto de prueba: {len(test_dataset)}")
print("")

# Calcular el porcentaje de clases positivas y negativas en el conjunto de entrenamiento
train_positive_pct = (train_labels.sum() / len(train_labels)) * 100
train_negative_pct = 100 - train_positive_pct

print(f"Porcentaje de clases positivas en el conjunto de entrenamiento: {train_positive_pct:.2f}%")
print(f"Porcentaje de clases negativas en el conjunto de entrenamiento: {train_negative_pct:.2f}%")
print("")

# Calcular el porcentaje de clases positivas y negativas en el conjunto de prueba
test_positive_pct = (test_labels.sum() / len(test_labels)) * 100
test_negative_pct = 100 - test_positive_pct

print(f"Porcentaje de clases positivas en el conjunto de prueba: {test_positive_pct:.2f}%")
print(f"Porcentaje de clases negativas en el conjunto de prueba: {test_negative_pct:.2f}%")
print("")
```

Interpretaci?n y consideraciones:

Equilibrio de clases: Los conjuntos de entrenamiento y prueba muestran una distribuci?n relativamente equilibrada entre clases positivas y negativas. Esto es beneficioso para el modelo, ya que le permite aprender de manera efectiva las caracter?sticas distintivas de ambas clases durante el entrenamiento.

Generalizaci?n del modelo: Con un conjunto de prueba de 1000 muestras, el modelo tiene suficientes datos para ser evaluado de manera significativa en condiciones similares a las del mundo real. La distribuci?n similar de clases en el conjunto de prueba asegura que las m?tricas de evaluaci?n reflejen el desempe?o del modelo de manera justa para ambas clases.

Monitoreo de desempe?o: Durante el entrenamiento y la evaluaci?n, es esencial monitorear m?tricas como precisi?n, recall, F1-score y matrices de confusi?n para comprender c?mo el modelo est? clasificando correctamente las instancias positivas y negativas en ambos conjuntos de datos. Esto proporciona insights sobre la capacidad del modelo para manejar diferentes tipos de comentarios y sentimientos.  Este evaluaci?n la veremos m?s adelante con los resultados obtenidos.

Avanzaremos, ya que la distribuci?n equilibrada de clases y el tama?o adecuado de los conjuntos de datos de entrenamiento y prueba son elementos clave para asegurar un entrenamiento efectivo y una evaluaci?n robusta del modelo de an?lisis de sentimientos en este contexto espec?fico.

```python
# Importaci?n de librer?as necesarias para el modelo
from transformers import DistilBertTokenizer, DistilBertForSequenceClassification, Trainer, TrainingArguments
import torch

# Tokenizaci?n usando el tokenizer de DistilBERT
tokenizer = DistilBertTokenizer.from_pretrained('distilbert-base-uncased')

# Funci?n para tokenizar los datos
def tokenize_function(examples):
    return tokenizer(examples['text'], truncation=True, padding='max_length')

# Tokenizaci?n de los conjuntos de entrenamiento y prueba
train_dataset = train_dataset.map(tokenize_function, batched=True)
test_dataset = test_dataset.map(tokenize_function, batched=True)

# Formateo de los datos para PyTorch
train_dataset.set_format(type='torch', columns=['input_ids', 'attention_mask', 'label'])
test_dataset.set_format(type='torch', columns=['input_ids', 'attention_mask', 'label'])

# Asegur?monos de tener GPU disponible para acelerar el proceso (si Colab ofrece GPU)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Using device:', device)
```

```python
# Definici?n de los argumentos de entrenamiento base
training_args_base = TrainingArguments(
    output_dir='./results_base',                 # Directorio donde se guardar?n los resultados del entrenamiento base
    num_train_epochs=3,                          # N?mero de ?pocas de entrenamiento base
    per_device_train_batch_size=8,               # Tama?o del batch de entrenamiento por dispositivo (GPU) base
    per_device_eval_batch_size=8,                # Tama?o del batch de evaluaci?n por dispositivo (GPU) base
    warmup_steps=500,                            # N?mero de pasos de calentamiento para el optimizador base
    weight_decay=0.01,                           # Decaimiento de peso para la regularizaci?n base
    logging_dir='./logs_base',                   # Directorio donde se guardar?n los logs del entrenamiento base
    logging_steps=10,                            # Frecuencia de pasos para el logging base
    evaluation_strategy="epoch",                 # Estrategia de evaluaci?n por ?poca
    save_strategy="epoch",                       # Estrategia de guardado por ?poca
    report_to="tensorboard",                     # Reportar a TensorBoard para visualizaci?n
    no_cuda=not torch.cuda.is_available()        # Forzar uso de GPU si est? disponible
)

# Creaci?n del modelo DistilBERT para clasificaci?n de secuencias
model_base = DistilBertForSequenceClassification.from_pretrained('distilbert-base-uncased', num_labels=2).to(device)

# Entrenamiento del modelo base
trainer_base = Trainer(
    model=model_base,                           # Modelo base a ser entrenado
    args=training_args_base,                    # Argumentos de entrenamiento base definidos anteriormente
    train_dataset=train_dataset,                # Dataset de entrenamiento
    eval_dataset=test_dataset,                  # Dataset de evaluaci?n
)
```

```python
# Entrenamiento, puede tomar un par de minutos.
trainer_base.train()
```

Interpretaci?n de -Evaluation Results-:

P?rdida (eval_loss): Una p?rdida de 0.4265 es relativamente baja, lo que sugiere que el modelo est? haciendo predicciones precisas en el conjunto de prueba. Sin embargo, la interpretaci?n exacta depende del contexto del problema y de las m?tricas espec?ficas de evaluaci?n (precisi?n, recall, F1-score).

Tiempo de ejecuci?n (eval_runtime): El tiempo de ejecuci?n de 2.1622 segundos indica cu?nto tard? el modelo en evaluar todas las muestras del conjunto de prueba. Es ?til para evaluar la eficiencia computacional del modelo.

Velocidad de procesamiento (eval_samples_per_second y eval_steps_per_second): Estas m?tricas indican la velocidad a la que el modelo puede procesar las muestras o los pasos durante la evaluaci?n. Una alta velocidad de procesamiento es deseable, ya que indica una capacidad de escalar a conjuntos de datos m?s grandes o a aplicaciones en tiempo real de manera efectiva.

De forma preliminar, estos resultados son indicativos del desempe?o del modelo en t?rminos de precisi?n, eficiencia y capacidad de procesamiento durante la evaluaci?n en el conjunto de prueba. Son ?tiles para entender c?mo se comporta el modelo en condiciones similares a las de producci?n y para identificar posibles ?reas de mejora o ajuste en futuras iteraciones del modelo.

```python
# Importaci?n de m?tricas de scikit-learn
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, f1_score
import numpy as np

# Evaluaci?n del modelo base
predictions_base = trainer_base.predict(test_dataset)
preds_logits_base = predictions_base.predictions
preds_labels_base = predictions_base.label_ids
preds_class_base = torch.argmax(torch.tensor(preds_logits_base), dim=1).numpy()

# Calcular m?tricas del modelo base
accuracy_base = accuracy_score(preds_labels_base, preds_class_base)
precision_base = precision_score(preds_labels_base, preds_class_base)
recall_base = recall_score(preds_labels_base, preds_class_base)
f1_base = f1_score(preds_labels_base, preds_class_base)

# Calcular la matriz de confusi?n
conf_matrix_base = confusion_matrix(preds_labels_base, preds_class_base)

print(f"Exactitud del modelo base: {accuracy_base:.4f}")
print(f"Precisi?n del modelo base: {precision_base:.4f}")
print(f"Recall del modelo base: {recall_base:.4f}")
print(f"F1-score del modelo base: {f1_base:.4f}")

print("Matriz de confusi?n del modelo base:")
print(conf_matrix_base)

# Convertir la matriz de confusi?n a porcentajes
conf_matrix_pct = conf_matrix_base.astype('float') / conf_matrix_base.sum(axis=1)[:, np.newaxis] * 100

print("Matriz de confusi?n del modelo (porcentual):")
print(conf_matrix_pct)
```

```python
training_args_finetune = TrainingArguments(
    output_dir='./results_finetune',             # Directorio donde se guardar?n los resultados del fine-tuning
    num_train_epochs=3,                          # N?mero de ?pocas de fine-tuning
    per_device_train_batch_size=16,              # Tama?o del batch de entrenamiento por dispositivo (GPU) fine-tuning
    per_device_eval_batch_size=16,               # Tama?o del batch de evaluaci?n por dispositivo (GPU) fine-tuning
    warmup_steps=200,                            # N?mero de pasos de calentamiento para el optimizador fine-tuning
    weight_decay=0.01,                           # Decaimiento de peso para la regularizaci?n fine-tuning
    logging_dir='./logs_finetune',               # Directorio donde se guardar?n los logs del fine-tuning
    logging_steps=50,                            # Frecuencia de pasos para el logging fine-tuning
    evaluation_strategy="epoch",                 # Estrategia de evaluaci?n por ?poca
    save_strategy="epoch",                       # Estrategia de guardado por ?poca
    report_to="tensorboard",                     # Reportar a TensorBoard para visualizaci?n
    no_cuda=not torch.cuda.is_available()        # Forzar uso de GPU si est? disponible
)

# Partici?n del conjunto de entrenamiento para fine-tuning en entrenamiento y validaci?n
train_texts_finetune, val_texts_finetune, train_labels_finetune, val_labels_finetune = train_test_split(train_texts, train_labels, test_size=0.2, random_state=42)

# Construir Dataset de entrenamiento y validaci?n para fine-tuning
train_dataset_finetune = Dataset.from_dict({'text': train_texts_finetune, 'label': train_labels_finetune})
val_dataset_finetune = Dataset.from_dict({'text': val_texts_finetune, 'label': val_labels_finetune})

# Tokenizaci?n de los conjuntos de entrenamiento y validaci?n para fine-tuning
train_dataset_finetune = train_dataset_finetune.map(tokenize_function, batched=True)
val_dataset_finetune = val_dataset_finetune.map(tokenize_function, batched=True)

# Formateo de los datos para PyTorch
train_dataset_finetune.set_format(type='torch', columns=['input_ids', 'attention_mask', 'label'])
val_dataset_finetune.set_format(type='torch', columns=['input_ids', 'attention_mask', 'label'])

# Crear el modelo DistilBERT para fine-tuning
model_finetune = DistilBertForSequenceClassification.from_pretrained('distilbert-base-uncased', num_labels=2)

# Definir Trainer para fine-tuning
trainer_finetune = Trainer(
    model=model_finetune,
    args=training_args_finetune,
    train_dataset=train_dataset_finetune,
    eval_dataset=val_dataset_finetune
)
```

```python
# Entrenamiento del modelo con fine-tuning
trainer_finetune.train()
```

```python
# Evaluaci?n del modelo con fine-tuning en el conjunto de validaci?n
predictions_finetune = trainer_finetune.predict(val_dataset_finetune)
preds_logits_finetune = predictions_finetune.predictions
preds_labels_finetune = predictions_finetune.label_ids
preds_class_finetune = torch.argmax(torch.tensor(preds_logits_finetune), dim=1).numpy()

# Calcular m?tricas del modelo con fine-tuning
accuracy_finetune = accuracy_score(preds_labels_finetune, preds_class_finetune)
precision_finetune = precision_score(preds_labels_finetune, preds_class_finetune)
recall_finetune = recall_score(preds_labels_finetune, preds_class_finetune)
f1_finetune = f1_score(preds_labels_finetune, preds_class_finetune)

conf_matrix_finetune = confusion_matrix(preds_labels_finetune, preds_class_finetune)
```

```python
print(f"Exactitud del modelo con fine-tuning en validaci?n: {accuracy_finetune:.4f}")
print(f"Precisi?n del modelo con fine-tuning en validaci?n: {precision_finetune:.4f}")
print(f"Recall del modelo con fine-tuning en validaci?n: {recall_finetune:.4f}")
print(f"F1-score del modelo base: {f1_finetune:.4f}")

print("Matriz de confusi?n del modelo con fine-tuning:")
print(conf_matrix_finetune)

# Convertir la matriz de confusi?n a porcentajes
conf_matrix_pct = conf_matrix_finetune.astype('float') / conf_matrix_finetune.sum(axis=1)[:, np.newaxis] * 100

print("Matriz de confusi?n del modelo (porcentual):")
print(conf_matrix_pct)
```
