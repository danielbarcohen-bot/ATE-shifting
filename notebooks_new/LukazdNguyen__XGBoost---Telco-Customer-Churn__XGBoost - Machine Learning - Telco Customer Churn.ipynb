# Load Libraries

```python
import pandas as pd
import xgboost as xgb 
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
%matplotlib inline
```

```python
#T?i d? li?u v?o
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
df.head()
```

# Cleansing Data

```python
#Chuy?n ??i d? li?u c?t TotalCharges v? d?ng integer
df['TotalCharges'] = pd.to_numeric(df['MonthlyCharges'], errors='coerce')
```

```python
#Xem c?c tr??ng b? Null
df.isnull().sum()
```

```python
#Xem t?ng c?t v? d?ng
df.shape
```

```python
#X?a c?t CustomerID trong b?ng df
df = df.drop(['customerID'], axis=1)
df.head()
```

```python
#Ti?n h?nh ??a d? li?u ??u v?o b?ng hai m?c X, y
X = df.drop(['Churn'], axis=1)
y = df['Churn'].copy()
```

In ra hai t?p KQ hi?n th? nh? sau:

```python
X.head()
```

```python
y.head()
```

```python
#Chuy?n c?c c?t c? tr??ng Object v? d?ng s? trong ph??ng th?c  one-hot encoding
X_ec = pd.get_dummies(X, drop_first=True)
```

```python
X_ec.head()
```

```python
#Chuy?n c?c c?t c?a y v? d?ng s? trong ph??ng th?c  one-hot encoding
y_ec = pd.get_dummies(y, drop_first=True)
```

```python
y_ec.head()
```

```python
#T?nh t?ng KH r?i b?/s? KH r?i b? d?ch v?
Ratio_CustomerChurn = y_ec.sum()/len(y_ec)
Ratio_CustomerChurn
```

```python
#??a d? li?u v?o c?c t?p hu?n luy?n train & test
X_train, X_test, y_train, y_test = train_test_split(X_ec, y_ec, random_state=42, stratify = y_ec)
```

```python
#T?nh t?ng KH r?i b?/s? KH r?i b? d?ch v?
Ratio_CustomerChurn = y_train.sum()/len(y_train)

# Convert the result to a float if it's a single-element Series
if isinstance(Ratio_CustomerChurn, pd.Series):
    Ratio_CustomerChurn = Ratio_CustomerChurn.item()
    
# Print the ratio with two decimal places
print(f'Ratio of Customer Churn: {Ratio_CustomerChurn:.2f}')
```

```python
# ??nh ngh?a m? h?nh
cdf_xgb = xgb.XGBClassifier(
    objective='binary:logistic',
    missing=None,
    seed=42
)

# C?i ??t c?c tham s? b?ng ph??ng th?c set_params
cdf_xgb.set_params(
    eval_metric='aucpr',
    early_stopping_rounds=10
)

# Hu?n luy?n m? h?nh
cdf_xgb.fit(X_train, y_train, 
            eval_set=[(X_test, y_test)], 
            verbose=2)
```

# Ti?n h?nh Grid Search ?? t?i ?u m? h?nh XGBoost

```python
from sklearn.model_selection import GridSearchCV
import xgboost as xgb

# ??nh ngh?a m? h?nh
model = xgb.XGBClassifier(objective='binary:logistic', seed=42)

# ??nh ngh?a c?c tham s? ?? t?m ki?m
param_grid = {
    'learning_rate': [0.01, 0.1, 0.2],
    'max_depth': [3, 6, 9],
    'n_estimators': [50, 100, 150],
    'subsample': [0.8, 0.9, 1.0],
    'colsample_bytree': [0.8, 0.9, 1.0]
}

# Kh?i t?o GridSearchCV v?i scoring='roc_auc'
grid_search = GridSearchCV(estimator=model, param_grid=param_grid, scoring='roc_auc', cv=3, verbose=2)

# Hu?n luy?n v?i GridSearchCV
grid_search.fit(X_train, y_train)

# L?y tham s? t?t nh?t
best_params = grid_search.best_params_
print("Best parameters found: ", best_params)

# Hu?n luy?n m? h?nh v?i c?c tham s? t?t nh?t
best_model = grid_search.best_estimator_
```
