# Logistic Regression

Logistic regression is a supervised machine learning algorithm that is used for classification tasks. In classification tasks, the dependent variable (target) is categorical (i.e. has discrete classes). Logistic regression builds a model that estimates the probability of a class occurring based on some independent variables.

## The Equation

The hypothesis function of logistic regression, which predicts the probability that an observation in the data belongs to a certain class, needs to be a function that produces values between 0 and 1. Thus, the sigmoid function can be used:
$$h_\theta(x) = g(\theta^T x) = \frac{1}{1 + e^{-\theta^T x}}$$

$$\theta^T x$$ is just a shorthand for the linear regression hypothesis
- ? = array of all your thetas ? [?_0, ?_1, ?_2, ...]
- x = array of all your features ? [x_0, x_1, x_2, ...]
- ??x = multiply each theta by its feature and sum them all up

So in Logistic Regression:
$$g(\theta^T x) = \frac{1}{1 + e^{-\theta^T x}}$$
We're just feeding the same linear equation from Linear Regression into the sigmoid function to squish the output between 0 and 1.

In binary logistic regression, the target has only two possible outcomes (e.g. 0 or 1). In this case, to perform classification, an estimated probability that is less than 0.5 will predict class 0 while an estimated probability greater than 0.5 will predict class 1.

The cost function used in binary logistic regression is the binary cross-entropy (log loss) which is calculated as follows:

$$J(\theta) = \frac{1}{m} \sum_{i=1}^{m} \left[ -y^{(i)} \log(h_\theta(x^{(i)})) - (1 - y^{(i)}) \log(1 - h_\theta(x^{(i)})) \right]$$

If y = 1, look at the formula ? the second term disappears:
$$= \left[ -y^{(i)} \log(h_\theta(x^{(i)})) - (1 - y^{(i)}) \log(1 - h_\theta(x^{(i)})) \right]$$

If y = 1:

$$- \log(h_\theta(x^{(i)}))$$

if h = 0.01
-log(0.01) is a big positive number ? big penalty!
And if h = 0.99 (predicted almost 1, which is correct when y=1):
-log(0.99) ? very small number ? small penalty

- It measures how wrong the model's predicted probabilities are
- Only one term activates at a time depending on the true label:
    - y = 1 ? first term ? penalizes low predicted probability
    - y = 0 ? second term ? penalizes high predicted probability
- The penalty is large when the model is confidently wrong, small when it's correct
- 1/m averages the cost over all training examples
- Gradient descent then minimizes this cost to find the best ? values

# Example:

Suppose you are the administrator of a university department and you want to determine each applicant's chance of acceptance based on their results on two exams. You have data from previous applicants which consists of the applicants' scores on two exams and the admissions decision. You can use logistic regression to build a classification model that estimates an applicant's probability of acceptance based on the scores from those two exams.

```python
import pandas
import numpy as np
import matplotlib.pyplot as plot
from sklearn import linear_model
from sklearn.model_selection import train_test_split
```

```python
data = pandas.read_csv('acceptance_data.txt', names = ['exam1','exam2','decision'])

# Step 1
print(data.describe())
data.plot(kind='box', subplots=True)
plot.show()
```

```python
# Steps 2 and 3
X = data.drop(columns=['decision']).round(2) #rounds all the values in the DataFrame to 2 decimal places.
y = data['decision']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=100)

print(type(X_train))
X_train.plot(kind='box', subplots=True, sharey=True)
plot.show()

# Task 1: Perform feature scaling

X_train = X_train.to_numpy().reshape((-1,2)) #2 means 2 columns and -1 means "hey numpy count how many rows there"
X_test = X_test.to_numpy().reshape((-1,2))
y_train = y_train.to_numpy() # What if the decision was "Yes" and "No"? 
#- machine learning models dont work with categorical data it needs numbers so we will encode
y_test = y_test.to_numpy()
xmin = X_train.min(axis=0)
xmax = X_train.max(axis=0)

# 2. Apply the Min-Max formula to scale the data between 0 and 1
X_train_scaled = (X_train - xmin) / (xmax - xmin)
X_test_scaled = (X_test - xmin) / (xmax - xmin)

#What if we have skewed data?
#Skewed data degrades the model?s ability (especially regression-based models) to describe typical cases as
#it has to deal with rare cases on extreme values. Skewed data also does not work well with many statistical methods.
#One way to deal with skewed data is to perform log-scale transformation,
#i.e. transform the skewed distribution to be a bit similar to the normaldistribution. 
#Note that there is no log for zero or negative values (so, add a constant to all values to ensure values are greater than one).
```

```python
# Step 4
model = linear_model.LogisticRegression(solver='sag') #Activity:What is the purpose of solver parameter, what does 'sag' stand for, and briefly describe two other solver options available
#The solver parameter specifies the optimization algorithm used to minimize the cost function (instead of the basic gradient descent we implemented manually).
#'sag' stands for Stochastic Average Gradient ? a faster variant of gradient descent that works well on large datasets.
model.fit(X_train_scaled, y_train)
```

```python
print('Coefficients: \n', model.coef_, " ", model.intercept_)

# Step 5
y_pred = model.predict(X_test_scaled)
print('Correct predicitions ratio: %.2f'% model.score(X_test_scaled, y_test))
#for a classification model like Logistic Regression, model.score(X, y) returns the accuracy ? 
#the proportion of correctly classified examples.
#model.score(X_test, y_test)
# equivalent to:
#accuracy = np.sum(y_pred == y_test) / len(y_test)
```

```python
# Task 2:
def predict_class(x_data):
    #Extract weights and calculate the linear equation instantly using dot product
    theta = model.coef_[0]
    bias = model.intercept_[0]
    z = np.dot(x_data, theta) + bias
    
    #Pass through sigmoid and apply the 0.5 threshold
    h = 1 / (1 + np.exp(-z))
    return (h >= 0.5).astype(int) #if h>= 0.5 return 1 otherwise 0

# Task 3:
def calc_accuracy(y_true, y_pred):
    # Calculate the mean of the exact matches
    return np.mean(y_true == y_pred)

# Task 4: Test these 2 functions before and after feature Scaling
my_preds = predict_class(X_test)
my_acc = calc_accuracy(y_test, my_preds)
print(f"Accuracy Before Scaling: {my_acc}")

my_preds = predict_class(X_test_scaled)
my_acc = calc_accuracy(y_test, my_preds)
print(f"Accuracy After Scaling: {my_acc}")
```
