First plot data points on a scatter plot chart to see if linear regression model can be applied. If yes, then build a model that can answer following questions,

1) Predict price of a mercedez benz that is 4 yr old with mileage 45000

2) Predict price of a BMW X5 that is 7 yr old with mileage 86000

3) Tell me the score (accuracy) of your model. (Hint: use LinearRegression().score())

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
```

```python
df = pd.read_csv(r"E:\machine_learning_practice\supervise_learning\linear_reg\datasets\carprices.csv")
df
```

```python
# using dummy variables
dummies = pd.get_dummies(df['Car Model'],dtype=int)
dummies
```

```python
# Avoid Dummy Variable Trap
dummies = dummies.drop('Mercedez Benz C class',axis=1)
dummies
```

```python
# concat dataframes
df2 = pd.concat([df,dummies],axis=1)
df2 = df2.drop('Car Model',axis=1)
df2
```

```python
X = df2.drop('Sell Price($)',axis=1)
y = df2['Sell Price($)']
```

```python
reg = LinearRegression()
reg.fit(X,y)
```

```python
predict = reg.predict([[45000,4,0,0]])
print(predict)
```

```python
reg.score(X,y)
```

```python
predict = reg.predict([[86000,7,0,1]])
print(predict)
```
