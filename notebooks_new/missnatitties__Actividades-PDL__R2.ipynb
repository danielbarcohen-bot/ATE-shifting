#### .loc e .iloc

Hay algunos otros objetos de indexaci?n en pandas, los cuales toman un valor para elegir las filas y un valor para elegir las columnas.

  - `df.iloc` est? **basado en la posici?n**.  Este indexador acepta enteros y trozos de enteros, y esencialmente trata el marco de datos como si fuera una simple matriz.
  - `df.loc` est? **basado en etiquetas**.  Este indexador trabaja con ?ndices/etiquetas de filas y columnas.

````python
df = pd.DataFrame({
    'some_integers': [0, 0, 1, 1, 2, 2],
    'some_strings': ['x', 'y', 'z', 'x', 'y', 'z'],
    'some_booleans': [0, 0, 1, 0, 1, 1]},
    index=['a', 'b', 'c', 'd', 'e', 'f']
)
df
````

````python
df.iloc[2:4, 0:2]
````

````python
df.loc['b':'e', ['some_integers', 'some_booleans']]
````

## ?ndice mixto

Hacer una indexaci?n mixta en los pandas modernos es m?s expl?cito, menos m?gico.  Es necesario utilizar los atributos `df.index` y `df.columns` para convertir expl?citamente las posiciones en etiquetas.

````python
df.loc[df.index[2:4], ['some_integers', 'some_booleans']]
````

`sort` controla el orden de las columnas

````python
pd.concat([df1, df2], axis=0, join='outer', sort=True)
````

````python
pd.concat([df1, df2], axis=0, join='inner', sort=True)
````

#### Horizontal

````python
pd.concat([df1, df2], axis=1)
````
