```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.neighbors import  KNeighborsClassifier
from sklearn.metrics import confusion_matrix,classification_report
from sklearn.model_selection import train_test_split
```

```python
x=np.array(
    [
        [3,3],[3,2],[2,1],[2,2],[3,3], #cat
        [5,5],[6,7],[5,6],[7,5],[6,6]  #dog
    ]
)
y=np.array(
    [
        "cat","cat","cat","cat","cat","dog","dog","dog","dog","dog"
    ]
)
```

```python
xxTest=[7,7]
knnModel=KNeighborsClassifier(n_neighbors=3)
xTrain,xTest,yTrain,yTest=train_test_split(x,y,test_size=.4,random_state=42)

knnModel.fit(xTrain,yTrain)
yPredict=knnModel.predict(xTest)
conM=confusion_matrix(yTest,yPredict)
print(conM)
ans=knnModel.predict([xxTest])
print(ans)
```

```python
report=classification_report(yTest,yPredict)
print(report)
```
