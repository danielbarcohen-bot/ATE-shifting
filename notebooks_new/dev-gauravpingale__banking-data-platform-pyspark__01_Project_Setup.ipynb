# Business Problem

ABC Bank receives data from multiple operational systems including Customer Management, Core Banking, Card Management, Merchant Management, and Branch Management systems.

Business users currently generate reports directly from operational databases, resulting in:
- Slow report performance
- Inconsistent business metrics
- Duplicate and poor-quality data
- High load on transactional systems
- Lack of a centralized analytics platform

To address these challenges, the Data Engineering team will build a centralized Banking Data Platform using the Medallion Architecture.

The platform will ingest raw data, apply business rules and data quality checks, and produce analytics-ready datasets for reporting and business intelligence.

# Medallion Architecture

```text
Raw Files
    │
    ▼
Bronze
(Raw Data)
    │
    ▼
Silver
(Cleansed & Validated Data)
    │
    ▼
Gold
(Business Ready Data)
```

### Bronze
Stores raw data exactly as received.

### Silver
Applies data cleansing, validation, deduplication, and business rules.

### Gold
Provides analytics-ready datasets such as fact and dimension tables for reporting and dashboards.\
