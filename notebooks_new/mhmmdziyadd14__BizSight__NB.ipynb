```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix, classification_report

# 1. Load dataset
df = pd.read_csv('balanced_cinnamon_quality_dataset.csv')

# 2. Ambil 15 data acak sebagai data uji untuk demo
test_data = df.sample(n=15, random_state=42)
train_data = df.drop(test_data.index)

# 3. Pisahkan Fitur dan Label
X_train = train_data.drop(columns=['Sample_ID', 'Quality_Label'])
y_train = train_data['Quality_Label']
X_test = test_data.drop(columns=['Sample_ID', 'Quality_Label'])
y_test = test_data['Quality_Label']

# 4. Implementasi Naive Bayes
model = GaussianNB()
model.fit(X_train, y_train)

# 5. Evaluasi (Pastikan bagian print di bawah ini bersih dari tanda [cite])
y_pred = model.predict(X_test)

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
import pandas as pd
import numpy as np
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# 1. Load dataset
df = pd.read_csv('balanced_cinnamon_quality_dataset.csv')

# 2. DATA CLEANING: Menangani Missing Values
# Mengisi data yang kosong (jika ada) dengan nilai rata-rata tiap kolom
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

# 3. PISAHKAN FITUR DAN LABEL
X = df.drop(columns=['Sample_ID', 'Quality_Label'])
y = df['Quality_Label']

# 4. NORMALISASI (Min-Max Scaling)
# Mengubah rentang nilai fitur menjadi 0 sampai 1
scaler_minmax = MinMaxScaler()
X_norm = scaler_minmax.fit_transform(X)

# 5. STANDARISASI (Standard Scaling / Z-Score)
# Mengubah data agar memiliki rata-rata 0 dan standar deviasi 1 (Distribusi Normal)
scaler_std = StandardScaler()
X_final = scaler_std.fit_transform(X_norm)

# Konversi kembali ke DataFrame agar mudah diambil sampelnya
X_final_df = pd.DataFrame(X_final, columns=X.columns)

# 6. SPLIT DATA (15 data acak sebagai data uji)
test_data = X_final_df.sample(n=15, random_state=42)
y_test = y.loc[test_data.index]

train_data = X_final_df.drop(test_data.index)
y_train = y.drop(test_data.index)

# 7. TRAINING MODEL NAIVE BAYES
model = GaussianNB()
model.fit(train_data, y_train)

# 8. EVALUASI HASIL
y_pred = model.predict(test_data)

print("--- Hasil Evaluasi dengan Preprocessing ---")
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
