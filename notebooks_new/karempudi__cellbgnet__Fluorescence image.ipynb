```python
import numpy as np
from skimage import io
import matplotlib.pyplot as plt
import pathlib
from mpl_toolkits.axes_grid1.axes_divider import make_axes_locatable
from pathlib import Path
%matplotlib qt5
```

```python
image = io.imread('/home/pk/Pictures/thesis_figures/test_fluor.tiff')
```

```python
SAVE_DIR = Path('/home/pk/Pictures/thesis_figures/')
```

```python
save_fluor_image = SAVE_DIR / Path('test.pdf')
fig, ax = plt.subplots(nrows=1, ncols=1)
im = ax.imshow(image, cmap='gray')
ax.set_axis_off()

divider = make_axes_locatable(ax)
cax = divider.append_axes("right", size="5%", pad=0.05)
cbar = plt.colorbar(im, cax=cax)
cbar.ax.tick_params(labelsize=16)
fig.savefig(save_fluor_image, bbox_inches='tight', pad_inches=0, transparent=True, dpi=300)
plt.show()
```
