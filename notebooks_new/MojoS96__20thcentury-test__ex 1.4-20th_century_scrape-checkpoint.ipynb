### Data Sourcing Criteria
- Scrape a list of countries from a web page [BONUS]
- TEXT Scrape a twentieth-century history and politics page from Wikipedia. [MAIN]

### As such, the following web pages will be scrape: 
- List of Countries > https://simple.wikipedia.org/wiki/List_of_countries
- General text scrape for key 20th Century events > https://en.wikipedia.org/wiki/Key_events_of_the_20th_century#

```python
# Step 1: Importing Libraries - using Selenium instead of BeautifulSoup as per CF training.

# Import libraries

import pandas as pd
import time
from selenium import webdriver
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import matplotlib.pyplot as plt
import os
import logging
import requests
import bs4
from bs4 import BeautifulSoup
```

### Section 1: Generating a list of countries

```python
## Setup chrome options

chrome_options = Options()
chrome_options.add_argument("--no-sandbox")
```

```python
# Install Chrome driver manager

service = Service(executable_path=ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)
```

```python
# Get the page's contents & convert it to HTML to be able to parse through

page_url = "https://simple.wikipedia.org/wiki/List_of_countries"
driver.get(page_url)
html_code = driver.page_source
soup = BeautifulSoup(html_code, 'html.parser')
```

```python
# Create a collection of countries in table:

a_elements = soup.find_all('a', title=True)
```

```python
# The <a> pull brought over too many non country lines of text - by using .iloc we can remove the preceding and following additional rows
countries = [a.get('title') for a in a_elements]
df = pd.DataFrame(countries, columns=['Country'])
df_filtered = df.iloc[252:561]
df_filtered
```

```python
df_filtered.to_csv('countries.csv')
```

### Section 2: performing a general text scrape

```python
# Get URL

page = requests.get("https://en.wikipedia.org/wiki/Key_events_of_the_20th_century")
soup = BeautifulSoup(page.text, 'html.parser')
print(soup.title)
```

```python
print(soup.text)
```

```python
#This line of code saves the .txt file right this directory
text = soup.get_text()
text = text.encode('utf-8')
with open('key_events_of_the_20th_century.txt', 'wb') as f: 
    f.write(text)
```
