```python
import numpy as np
import numpy_financial as npf

# Parametry
cena = 120000                # Cena pocz?tkowa mieszkania
roczny_wzrost = 0.05         # Roczny wzrost cen mieszka?
lata = 5
okres_miesiecy = lata * 12

# Obliczenie przysz?ej ceny mieszkania
przyszla_cena = cena * (1 + roczny_wzrost) ** lata

# Parametry lokaty
stopa_roczna = 0.12
stopa_miesieczna = stopa_roczna / 12

# Wyliczenie miesi?cznej wp?aty potrzebnej do zgromadzenia tej kwoty
rata = npf.pmt(rate=stopa_miesieczna, nper=okres_miesiecy, pv=0, fv=przyszla_cena)

print(f"Cena mieszkania za 5 lat: {przyszla_cena:.2f} z?")
print(f"Potrzebna miesi?czna wp?ata: {-rata:.2f} z?")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# O? czasu: 60 miesi?cy (5 lat)
miesiace = np.arange(1, okres_miesiecy + 1)

# Cena mieszkania rosn?ca liniowo
cena_mieszkania = cena * (1 + roczny_wzrost) ** (miesiace / 12)

# Warto?? lokaty w ka?dym miesi?cu
wartosc_lokaty = npf.fv(rate=stopa_miesieczna, nper=miesiace, pmt=-rata, pv=0)

# Tworzenie wykresu
plt.figure(figsize=(12, 6))
plt.plot(miesiace, cena_mieszkania, label='Cena mieszkania', color='red')
plt.plot(miesiace, wartosc_lokaty, label='Warto?? lokaty', color='green')
plt.title('Por?wnanie ceny mieszkania i warto?ci lokaty (miesi?c po miesi?cu)')
plt.xlabel('Miesi?ce')
plt.ylabel('Kwota [z?]')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
