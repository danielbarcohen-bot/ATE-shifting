```python
import csv

with open("weather_data.csv") as f:
    datas=csv.reader(f)
    temp=[]
    #print(type(datas))
    for row in datas:
        if row[1]!="temp":
            #print(row)
            temp.append(int(row[1]))

    print(temp)
```
