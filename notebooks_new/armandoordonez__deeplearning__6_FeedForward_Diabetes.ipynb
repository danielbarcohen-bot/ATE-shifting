```python
# separa los datos en X y y
X = data[:,:8]
y = data[:,8].astype(int)

# imprime el tama?o de X y y
print("[INFO]\nX: %s\ny: %s"%(X.shape,y.shape))
```

**Sequential**  es un modelo que permite crear redes neuronales de forma secuencial, capa por capa, sin bifurcaciones ni conexiones complicadas entre capas. Cada capa en un modelo Sequential toma la salida de la capa anterior como entrada y produce una salida para la siguiente capa. Es una forma simple y directa de construir modelos de redes neuronales en Keras.

**Dense**  es una capa completamente conectada en una red neuronal, donde cada neurona en una capa est? conectada a todas las neuronas de la capa anterior y siguiente. La capa Dense es fundamental para la mayor?a de las arquitecturas de redes neuronales y se utiliza com?nmente en tareas de clasificaci?n y regresi?n.

**Dropout** es una t?cnica de regularizaci?n utilizada para prevenir el sobreajuste en modelos de redes neuronales al azarmente desactivar un porcentaje de unidades de una capa durante el entrenamiento. Esto ayuda a mejorar la generalizaci?n del modelo al reducir la dependencia de unidades espec?ficas.

**Adam**es un algoritmo de optimizaci?n popular utilizado para ajustar los pesos de una red neuronal durante el entrenamiento. Combina t?cnicas de descenso de gradiente estoc?stico con adaptaci?n de la tasa de aprendizaje por cada par?metro de la red. Adam ajusta autom?ticamente la tasa de aprendizaje durante el entrenamiento, lo que lo hace eficiente y efectivo en una variedad de problemas de optimizaci?n.

```python

from keras.models import Sequential
from keras.layers import Dense, Dropout
from keras.optimizers import Adam

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

Para construir un modelo en Keras, se utiliza la clase Sequential para definir la arquitectura de la red neuronal, donde se pueden agregar capas utilizando la clase Dense. Cada capa Dense representa una capa completamente conectada en la red, especificando el n?mero de unidades (neuronas) y la funci?n de activaci?n a utilizar. La funci?n de activaci?n determina la salida de cada neurona y puede ser relu, sigmoid, tanh, entre otras. Estas capas se encadenan secuencialmente, y al final se compila el modelo especificando la funci?n de p?rdida, el optimizador (como Adam) y las m?tricas a seguir durante el entrenamiento.

En este caso se utiliza una sola neurona en la capa de salida porque esta neurona representa la probabilidad de pertenencia a una de las dos clases mediante una salida sigmoidal, donde valores cercanos a 0 indican la clase negativa y valores cercanos a 1 indican la clase positiva. Esta configuraci?n simplifica el modelo y es adecuada para problemas de clasificaci?n binaria.

```python
num_classes = 2
hl1 = 12
hl2 = 8
hl3 = 1
#multilayer feedforward

model = Sequential()
model.add(Dense(hl1, input_dim=8, activation='relu'))
model.add(Dense(hl2, activation='relu'))
model.add(Dense(hl3, activation='sigmoid'))  
model.summary()
```

```python
#separar datos train test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
# Escalar caracter?sticas
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

La funci?n **compile** se utiliza para configurar el proceso de entrenamiento de un modelo neuronal, donde se especifican el tipo de p?rdida (loss) a minimizar durante el entrenamiento, el optimizador (optimizer) para ajustar los pesos de la red y las m?tricas para evaluar el rendimiento del modelo. La p?rdida representa la discrepancia entre las predicciones del modelo y los valores reales, mientras que el optimizador determina c?mo se actualizan los pesos de la red para minimizar esta p?rdida durante el entrenamiento. Estos par?metros son cruciales para definir c?mo el modelo aprende a partir de los datos y c?mo se eval?a su rendimiento.

En un modelo de Keras para clasificaci?n binaria, se utiliza la funci?n de p?rdida `binary_crossentropy`, que calcula la p?rdida entre las etiquetas y las predicciones para un problema de dos clases. 

En Keras, el m?todo **fit** se utiliza para entrenar un modelo de red neuronal con un conjunto de datos. Los par?metros **epochs** especifican el n?mero de veces que se recorrer?n todos los datos durante el entrenamiento, mientras que **batch_size** determina el tama?o del lote de datos que se utilizar? en cada iteraci?n del entrenamiento. Juntos, estos par?metros controlan la duraci?n y la granularidad del proceso de entrenamiento del modelo.

```python
#'binary_crossentropy'

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=100, batch_size=10)
```

Cuando se hace una predicci?n con un modelo de Keras que utiliza una neurona con salida sigmoidal, la salida de las predicciones estar? en el rango de 0 a 1, representando la probabilidad de que cada entrada pertenezca a la clase positiva. Valores cercanos a 0 indican baja probabilidad de pertenencia a la clase positiva, mientras que valores cercanos a 1 indican alta probabilidad. Las predicciones pueden interpretarse como la confianza del modelo en la clasificaci?n de cada entrada.

# Crear una funci?n que automatice el proceso
La funci?n agrega capas de dropout entre las capas densas. Los par?metros de configuraci?n se envian a la funci?n. 

```python
def create_model(neurons_per_layer, input_dim, activation_func, dropout_rate):
    model = Sequential()
    model.add(Dense(neurons_per_layer[0], input_dim=input_dim, activation=activation_func))
    model.add(Dropout(dropout_rate))
    neurons_per_layer.pop(0)
    for neurons in neurons_per_layer:
        model.add(Dense(neurons, activation=activation_func))
        model.add(Dropout(dropout_rate))
    model.add(Dense(1, activation='sigmoid'))
    return model
```

```python
neurons_per_layer = [64, 32]        # N?mero de neuronas en cada capa oculta
activation_func   = 'relu'          # Funci?n de activaci?n
dropout_rate      = 0.2             # Tasa de dropout
input_dim         =  X_train.shape[1] #dimension de datos

model = create_model(neurons_per_layer, input_dim, activation_func, dropout_rate)
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
model.fit(X_train, y_train, epochs=50, batch_size=32, validation_data=(X_test, y_test))
loss, accuracy = model.evaluate(X_test, y_test)
print(f'Loss: {loss}, Accuracy: {accuracy}')
```
