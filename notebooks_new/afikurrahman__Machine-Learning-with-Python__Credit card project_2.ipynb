# Separet x and y

```python
x=df.drop(['Class'],axis=1)   # dataset theke class column drop 
```

```python
x
```

```python
y=df['Class']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split   # train_test_split import 
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=1)
```

```python
xtrain
```

```python
from sklearn.svm import SVC   # support vector machine(SVM)algoritm import
```

```python
model=SVC()  # object
```

```python
model_=model.fit(xtrain,ytrain)
```

```python
model_.score(xtest,ytest)  # support vector machine(SVM) algorithm Accuracy
```

```python
from sklearn.linear_model import LogisticRegression  # LogisticRegression algoritm import 
```

```python
model=LogisticRegression()  # object
```

```python
model_=model.fit(xtrain,ytrain)
```

```python
model_.score(xtest,ytest)  # LogisticRegression algorithm Accuracy
```

```python
from sklearn.ensemble import RandomForestClassifier  #RandomForestClassifier algoritm import 
```

```python
model=RandomForestClassifier() # object
```

```python
model_=model.fit(xtrain,ytrain)
```

```python
model_.score(xtest,ytest)  # RandomForestClassifier algorithm Accuracy
```

```python
from sklearn.tree import DecisionTreeClassifier  # DecisionTreeClassifier algoritm import 
```

```python
model=DecisionTreeClassifier()  # object
```

```python
model_=model.fit(xtrain,ytrain)
```

```python
model_.score(xtest,ytest)  # DecisionTreeClassifier algorithm Accuracy
```
