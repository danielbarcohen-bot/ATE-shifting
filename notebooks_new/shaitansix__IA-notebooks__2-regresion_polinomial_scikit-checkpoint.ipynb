## Gradiente descendente

<h1>
    $$\hat{y}_i = w_2 x_2 + w_1 x_1 + w_0$$
</h1>

### Funcion de perdida (EMC)

<h1>
    $$loss(\hat{y}) = \frac{1}{N} \sum_{i=1}^n{(y_i - \hat{y}_i) ^ 2}$$
</h1>

### Derivada parcial de la funcion de perdida con respecto a $w_0$

<h3>
    $\frac{\partial}{\partial w_0} = (y - (w_2 x_2 + w_1 x_1 + w_0)) ^ 2$ <br /><br />
    $\frac{\partial}{\partial w_0} = 2 * (y_i - w_2 x_2 - w_1 x_1 - w_0) * -1$ <br /><br />
    $\frac{\partial}{\partial w_0} = -2(y - \hat{y})$
</h3>

<h1>
    $$\frac{\partial loss(\hat{y})}{\partial w_0} = -\frac{2}{N} \sum_{i=1}^n {(y_i - \hat{y}_i)}$$
</h1>

### Derivada parcial de la funcion de perdida con respecto a $w_1$

<h3>
    $\frac{\partial}{\partial w_1} = (y - (w_2 x_2 + w_1 x_1 + w_0)) ^ 2$ <br /><br />
    $\frac{\partial}{\partial w_1} = 2 * (y_i - w_2 x_2 - w_1 x_1 - w_0) * -x_1$ <br /><br />
    $\frac{\partial}{\partial w_1} = -2x_1(y - \hat{y})$
</h3>

<h1>
    $$\frac{\partial loss(\hat{y})}{\partial w_1} = -\frac{2}{N} \sum_{i=1}^n {x_1(y_i - \hat{y}_i)}$$
</h1>

### Derivada parcial de la funcion de perdida con respecto a $w_2$

<h3>
    $\frac{\partial}{\partial w_2} = (y - (w_2 x_2 + w_1 x_1 + w_0)) ^ 2$ <br /><br />
    $\frac{\partial}{\partial w_2} = 2 * (y_i - w_2 x_2 - w_1 x_1 - w_0) * -x_2$ <br /><br />
    $\frac{\partial}{\partial w_2} = -2x_2(y - \hat{y})$
</h3>

<h1>
    $$\frac{\partial loss(\hat{y})}{\partial w_2} = -\frac{2}{N} \sum_{i=1}^n {x_2(y_i - \hat{y}_i)}$$
</h1>

<strong>$N:$ </strong> Numero de muestras <br />
<strong>$y:$ </strong> Valor objetivo <br />
<strong>$\hat{y}:$ </strong> Valor predicho

## Algotritmo de regresion polinomial(sin sklearn)

```python
from matplotlib import pyplot as plt
import numpy as np

# each degree of x becomes a new feature
X_1 = np.arange(0, 10).reshape(10, 1)
X_2 = X_1 ** 2
X = np.hstack((X_1, X_2))

# weights for each degree of x, including x^0
W = np.random.uniform(0.001, 1, 3)
X, W
```

```python
# y = ax^2 + bx + c => y = w_2 * x_2 + w_1 * x_1 + w_0
y = W[2] * X[:, 1] + W[1] * X[:, 0] + W[0]

# graph
fig, ax = plt.subplots(figsize = (5, 5))
ax.scatter(X[:, 0], y, label = '$f(x) = w_2x^2 + w_1x + w_0$')
ax.legend()
```

```python
X_1 = (np.arange(0, 40) + np.random.uniform(0.001, 5, 40)).reshape(40, 1)
X_2 = X_1 ** 2 + np.random.uniform(0.001, 5, (40, 1))
print(np.unique(X_1).shape, np.unique(X_2).shape)
X = np.hstack((X_1, X_2))

W = np.random.uniform(0.001, 1, 3)
X[:5, :], W
```

```python
desv = np.random.uniform(-20, 20, 40)
y = (W[2] * X[:, 1] + W[1] * X[:, 0] + W[0]) + desv
fig, ax = plt.subplots(figsize = (5, 5))
ax.scatter(X[:, 0], y, label = '$f(x) = w_2x^2 + w_1x + w_0$')
ax.legend()
```

```python
class RegressionModel: 
    def __init__(self, alpha = 0.01): 
        # params
        self.W = np.random.uniform(0, 1, 3)
        self.loss = np.array([])

        # hyperparams
        self.alpha = alpha

    # methods
    def predict(self, X): 
        return self.W[2] * X[:, 1] + self.W[1] * X[:, 0] + self.W[0]
    
    def gradient(self, X, y): 
        n = X.shape[0]
        y_ = self.predict(X)
        
        dw_0 = -2 * np.sum((y - y_))
        dw_1 = -2 * np.sum(X[:, 0] * (y - y_))
        dw_2 = -2 * np.sum(X[:, 1] * (y - y_))
        dw = np.array([dw_0, dw_1, dw_2])
        new_W = self.W - self.alpha * dw / n
        return new_W
    
    def EMC(self, X, y): 
        n = X.shape[0]
        y_ = self.predict(X)
        return np.sum((y - y_) ** 2) / n
    
    def fit(self, X, y, epochs = 1000): 
        for _ in range(epochs): 
            self.W = self.gradient(X, y)
            emc = self.EMC(X, y)
            self.loss = np.append(self.loss, emc)
    
    def get_params(self): 
        return { 'W': self.W, 'loss': self.loss }
```

```python
def plotRegression(X, y, epochs, model): 
    y_ = model.predict(X) # predict
    params = model.get_params()
    W = params['W'] # w value
    loss = params['loss'] # loss

    # graph
    fig, axs = plt.subplots(1, 2, figsize = (20, 10))
    for i, ax in enumerate(axs): 
        if i == 0: 
            ax.scatter(X[:, 0], y)
            ax.plot(X[:, 0], y_, c = 'orange', label = f'$y = {W[2]:.2f}x_2 + {W[1]:.2f}x_1 + {W[0]:.2f} = {W[2]:.2f}x^2 + {W[1]:.2f}x + {W[0]:.2f}$')
            ax.set_title('Regression', fontsize = 16)
            ax.set_xlabel('Feature', fontsize = 16)
            ax.set_ylabel('Target', fontsize = 16)
        elif i == 1: 
            ax.plot(range(epochs), loss, c = 'grey', label = f'Loss: {loss[-1]:.2f}')
            ax.set_title('Epochs vs Loss', fontsize = 16)
            ax.set_xlabel('Epoch', fontsize = 16)
            ax.set_ylabel('Loss', fontsize = 16)
        ax.legend(prop = {'size': 12})
        ax.grid(True)
        ax.tick_params(labelsize = 12)
```

```python
# hyperparams
alpha = 0.000001
epochs = 10

# model
model = RegressionModel(alpha)
model.fit(X, y, epochs)
model.get_params()
```

```python
plotRegression(X, y, epochs, model)
```
