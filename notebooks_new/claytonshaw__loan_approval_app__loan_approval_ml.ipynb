```python
import pandas as pd
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
```

```python
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')

train.head()
```

```python
# convert all object columns to category
for col in train.select_dtypes(include=['object']).columns:
    train[col] = train[col].astype('category')
    test[col] = test[col].astype('category')

# drop id from train and test
train.drop('id', axis=1, inplace=True)
test.drop('id', axis=1, inplace=True)
```

```python
X = train.drop('loan_status', axis=1)
y = train['loan_status']

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=123)
```

```python
bst = XGBClassifier(n_estimators=1000, learning_rate=0.05, objective='binary:logistic', enable_categorical=True)

bst.fit(X_train, y_train)
```

```python
from sklearn.model_selection import cross_val_score
import numpy as np

scores = cross_val_score(bst, X_train, y_train, cv=5)
print("Cross-validated accuracy: {:.2f} ? {:.2f}".format(scores.mean(), scores.std() * 2))
```

```python
import joblib

joblib.dump(bst, 'xgboost_model.joblib')
```

```python
# i want to see in the test set who gets approved
test_pred = bst.predict(test)

# add the predictions to the test set
test['loan_status'] = test_pred
test.head()
```
