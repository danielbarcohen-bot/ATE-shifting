# Project 01: Ecommerce Sales & Profit Trends (2022–2024)

Sales and profit were explored to understand overall business performance and identify patterns across the available records.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("ecommerce_data.csv")
df.head()
```

## Data Inspection

```
df.info()
df.describe(include="all")
df.isnull().sum()
```

## Data Cleaning

```
df = df.drop_duplicates()

for c in df.select_dtypes(include="object").columns:
    df[c] = df[c].fillna(df[c].mode()[0])

for c in df.select_dtypes(exclude="object").columns:
    df[c] = df[c].fillna(df[c].median())

df.isnull().sum()
```

## Descriptive Statistics

```
df.describe()
```

## Correlation

```
numeric = df.select_dtypes(include=np.number)
corr = numeric.corr()
corr
```

## Basic Visualizations

```
numeric.hist(figsize=(10,7))
plt.tight_layout()
plt.show()

plt.figure(figsize=(8,6))
plt.imshow(corr, aspect="auto")
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.tight_layout()
plt.show()
```

### Observations

Write a few short observations after running the notebook. Focus on visible patterns rather than assumptions.

### Summary

The dataset was cleaned, explored, and summarized using descriptive statistics and simple visualizations.
