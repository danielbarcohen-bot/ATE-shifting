Performing initial data inspection and cleaning of the UiS4ADL and PAAL ADL datasets.

```python
import os
import warnings
import pandas as pd

# Suppress warnings from Python and external libraries
os.environ["PYTHONWARNINGS"] = "ignore"  # Suppress Python warnings
warnings.filterwarnings("ignore")
```

# UiS4ADL

```python
fs_UiS4ADL = 100 # sampling frequency in Hz
path_UiS4ADL = "../Data/UiS4ADL/Raw/UiS4ADL_100hz_raw.csv"

df_UiS4ADL = pd.read_csv(path_UiS4ADL)
```

```python
df_UiS4ADL
```

## Basic data cleaning and overview

```python
df_UiS4ADL = df_UiS4ADL[df_UiS4ADL['adl'] != 0] # removing the 0 class

print(f'The dataset contains {df_UiS4ADL["adl"].nunique()} unique ADL classes after removing the 0 class.')
```

```python
df_UiS4ADL.columns
```

```python
num_unique_file_ids = df_UiS4ADL['fileID'].nunique()
print(f"Number of unique file IDs: {num_unique_file_ids}")
```

```python
df_UiS4ADL.info()
```

```python
# Identify how many NaN values the df dataframe has
nan_count = df_UiS4ADL.isnull().sum().sum()
print(f"Total NaN values in the dataframe: {nan_count}")

# Identify which files contain the empty rows
files_with_nan = df_UiS4ADL.groupby('fileID').apply(lambda group: group.isnull().sum().sum() > 0)
files_with_nan = files_with_nan[files_with_nan].index.tolist()
print(f"Files with NaN values: {files_with_nan}")

# Remove these fileIds from the df
df_UiS4ADL = df_UiS4ADL[~df_UiS4ADL['fileID'].isin(files_with_nan)]

# Verify the rows are removed
print(f"Files with NaN values removed. Remaining rows: {len(df_UiS4ADL)}.")
print(f"New number of unique file IDs: {df_UiS4ADL['fileID'].nunique()}.")
```

```python
# Count the number of unique subjects in the DataFrame
num_subjects = df_UiS4ADL['subject'].nunique()

print(f"The dataset contains {num_subjects} unique subjects.")
```

```python
# Check for duplicate rows
num_dups = df_UiS4ADL.duplicated().sum()

if num_dups > 0:
    print(f"Found {num_dups} duplicates. Removing them...")
    df_UiS4ADL = df_UiS4ADL.drop_duplicates()
else:
    print("No duplicates found.")

# Verify the rows are removed
print(f"Remaining rows: {len(df_UiS4ADL)}")
```

```python
# Group the DataFrame by 'adl', 'subject', and 'session' and count unique 'fileID' values
grouped_file_counts = df_UiS4ADL.groupby(['adl', 'subject', 'session'])['fileID'].nunique()

# Filter and print only the groups that have more than one unique 'fileID' value
multiple_fileID_groups = grouped_file_counts[grouped_file_counts > 1]
if not multiple_fileID_groups.empty:
    print(multiple_fileID_groups)
else:
    print("No groups have more than one unique 'fileID' value.") # every session performed by a subject for a specific ADL is unique
```

```python
# Describe the DataFrame to show the summary statistics for numerical columns
print("Basic Statistics:")
display(df_UiS4ADL.describe().iloc[1:,0:-4])
print()
```

## Save the initially cleaned dataset

The 'adl'==0 was removed. Furthermore, there were two files (with IDs 4092 and 4093) that contained NaN values, which lead to 7969 rows being removed as well. No duplicates were found in the dataset.

```python
# Save the DataFrame after initial cleaning
output_file = f"../Data/Processing/combined_UiS4ADL_{fs_UiS4ADL}hz_initially_cleaned.csv"

# Remove the file if it already exists
if os.path.exists(output_file):
    os.remove(output_file)
    print(f"{output_file} has been removed successfully.")
else:
    print(f"{output_file} does not exist.")
    
# Save the cleaned DataFrame to a CSV file
df_UiS4ADL.to_csv(output_file, index=False)
print(f"The dataset was saved to {output_file}")
```
