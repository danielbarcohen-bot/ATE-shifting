```python
import pandas as pd

data = pd.DataFrame({
    'edges' : ['soomth, toothed, lobed', 'lobed', None],
    'arrangement' : ['opposite', 'alternate, opposite', 'alternate']
})

print("Original DataFrame: ", data, sep="\n")
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(sparse_output=False, dtype=int)
enoded = ohe.fit_transform(data[['edges', 'arrangement']])

encoded_df = pd.DataFrame(enoded, columns=ohe.get_feature_names_out())

encoded_df
```

```python
from itertools import product

edges_cols = [col for col in encoded_df.columns if col.startswith('edges')]
arrangement_cols = [col for col in encoded_df.columns if col.startswith('arrangement')]

cross_features ={ f"{e}__{a}": (encoded_df[e] & encoded_df[a]) for e, a in product(edges_cols, arrangement_cols) }

cross_df = pd.DataFrame(cross_features)
cross_df
```

```python
final = pd.concat([data, encoded_df, cross_df], axis=1)
final
```
