# BTVN 02:

# T?o 1 tensor c? gi? tr? ban ??u b?ng 2

# ??nh ngh?a h?m s? v? t?nh  gradient

# y  = x^3 + 2x^2 + 5x + 1

# D?ng ph??ng ph?p Gradient Descent v?i learning rate alpha = 0.1 ?? c?p nh?t gi? tr? x trong 10 v?ng l?p

```python
# BTVN 02:
import torch

x = torch.tensor(2.0, requires_grad=True)
alpha = 0.1

for i in range(10):
    y = x**3 + 2*x**2 + 5*x + 1
    y.backward()
    x = x - alpha * x.grad
    x = x.detach().requires_grad_()
    
    print(x.item())
```

# BTVN 03:

## T?o 1 t?p d? gi? l?p v?i x l? s? gi? t? h?c (ng?u nhi?n  t? 1 ??n 10) v? y l? ?i?m s? ???c t?nh theo c?ng th?c y = 3x + 5 + noise v?i noise l? gi? tr? ng?u nhi?n nh?

1. Kh?i t?o tham s? w, b ng?u nhi?n v?i requires_grad=True
2. T?nh MSE
3. T?nh Gradient
4. C?p nh?t tham s? w v? b b?ng Gradient Descent v?i learning rate alpha = 0.01
5. L?p l?i qu? tr?nh tr?n trong 100 v?ng ?? xem s? h?i t? m? h?nh

```python
# BTVN 03:
import torch
# T?o 1 t?p d? gi? l?p v?i x l? s? gi? t? h?c (ng?u nhi?n  t? 1 ??n 10) v? y l? ?i?m s? ???c t?nh theo c?ng th?c y = 3x + 5 + noise v?i noise l? gi? tr? ng?u nhi?n nh?
x = torch.rand(20) * 9 + 1
noise = torch.randn(20) * 0.5
y = 3*x + 5 + noise
# 1. Kh?i t?o tham s? w, b ng?u nhi?n v?i requires_grad=True
w = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)
# 2. T?nh MSE
y_pred = w * x + b
mse = ((y_pred - y)**2).mean()
# 3. T?nh Gradient
mse.backward()
# 4. C?p nh?t tham s? w v? b b?ng Gradient Descent v?i learning rate alpha = 0.01
alpha = 0.01
with torch.no_grad():
    w -= alpha * w.grad
    b -= alpha * b.grad
# 5. L?p l?i qu? tr?nh tr?n trong 100 v?ng ?? xem s? h?i t? m? h?nh
for i in range(100):
    y_pred = w * x + b
    mse = ((y_pred - y)**2).mean()
    mse.backward()
    with torch.no_grad():
        w -= alpha * w.grad
        b -= alpha * b.grad
    w.grad.zero_()
    b.grad.zero_()
    if (i+1) % 10 == 0:
        print(f"V?ng {i+1}: loss={mse.item():.3f}, w={w.item():.3f}, b={b.item():.3f}")


```
