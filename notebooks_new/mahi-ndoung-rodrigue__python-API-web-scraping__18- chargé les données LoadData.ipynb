After the import command, we now have access to a large number of pre-built classes and functions. Let's go through the process to go from a comma separated values (<b>.csv</b>) file to a dataframe. This variable <code>csv_path</code> stores the path of the <b>.csv</b>, that is  used as an argument to the <code>read_csv</code> function. The result is stored in the object <code>df</code>, this is a common short form used for a variable referring to a Pandas dataframe.


Après la commande import, nous avons maintenant accès à un grand nombre de classes et de fonctions pré-construites. Cela suppose que la bibliothèque est installée ; Dans notre environnement de laboratoire, toutes les bibliothèques nécessaires sont installées. L’une des façons dont pandas vous permet de travailler avec des données est un dataframe. Passons en revue le processus pour passer d’un fichier de valeurs séparées par des virgules (<b>.csv</b>) à une trame de données. Cette variable <code>stocke csv_path</code> le chemin de la <b>.csv</b>, qui est utilisée comme argument de la fonction <code>read_csv</code>. Le résultat est stocké dans l’objet <code>df</code>, il s’agit d’une forme abrégée courante utilisée pour une variable faisant référence à une trame de données Pandas.

````python
# Read data from CSV file

# csv_path = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%204/data/TopSellingAlbums.csv'
# df = pd.read_csv(csv_path)

from pyodide.http import pyfetch
import pandas as pd

filename = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%204/data/TopSellingAlbums.csv"

async def download(url, filename):
    response = await pyfetch(url)
    if response.status == 200:
        with open(filename, "wb") as f:
            f.write(await response.bytes())


await download(filename, "TopSellingAlbums.csv")
df = pd.read_csv("TopSellingAlbums.csv")
````

## Viewing Data and Accessing Data

## Consulter les données et accéder aux données

You can also get a column as a series. You can think of a Pandas series as a 1-D dataframe. Just use one bracket:

Vous pouvez également obtenir une chronique en série. Vous pouvez considérer une série Pandas comme une trame de données 1D. Utilisez simplement un support :


````python
# Get the column as a series

x = df['Length']
x
````

You can also get a column as a dataframe. For example, we can assign the column <b>Artist</b>:


Vous pouvez également obtenir une colonne en tant que dataframe. Par exemple, nous pouvons attribuer à la colonne <b>Artiste</b> :


````python
# Get the column as a dataframe

x = df[['Artist']]
type(x)
````

````python
x.head()
````

You can do the same thing for multiple columns; we just put the dataframe name, in this case, <code>df</code>, and the name of the multiple column headers enclosed in double brackets. The result is a new dataframe comprised of the specified columns:


Vous pouvez faire la même chose pour plusieurs colonnes ; Nous avons simplement mis le nom de la trame de données, dans ce cas, <code>DF</code>, et le nom des en-têtes de plusieurs colonnes entre doubles crochets. Le résultat est une nouvelle trame de données composée des colonnes spécifiées :


````python
# Access to multiple columns

y = df[['Artist','Length','Genre']]
y
````
