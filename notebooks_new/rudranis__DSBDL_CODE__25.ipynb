```python
# =========================================
# Hepatitis Dataset (FIXED)
# =========================================

import pandas as pd

# Load dataset WITHOUT header
df = pd.read_csv('hepatitis.csv', header=None)

# Assign simple column names
df.columns = ['Class','Age','Sex','Steroid','Antivirals','Fatigue',
              'Malaise','Anorexia','LiverBig','LiverFirm','SpleenPalpable',
              'Spiders','Ascites','Varices','Bilirubin','AlkPhosphate',
              'SGOT','Albumin','Protime','Histology']

print("Columns:")
print(df.columns)


# =========================================
# Data Cleaning
# =========================================

df.replace('?', pd.NA, inplace=True)
df.dropna(inplace=True)


# =========================================
# a. Create subsets for Sex
# =========================================

male = df[df['Sex'] == 1]
female = df[df['Sex'] == 2]

print("\nMale Data:")
print(male.head())

print("\nFemale Data:")
print(female.head())


# =========================================
# b. Merge
# =========================================

merged_data = pd.concat([male, female])

print("\nMerged Data:")
print(merged_data.head())


# =========================================
# c. Sort
# =========================================

# Convert to numeric
df['Age'] = pd.to_numeric(df['Age'])
df['SGOT'] = pd.to_numeric(df['SGOT'])
df['Protime'] = pd.to_numeric(df['Protime'])

sorted_data = df.sort_values(by=['Age','SGOT','Protime'], ascending=False)

print("\nSorted Data:")
print(sorted_data[['Age','SGOT','Protime']].head())


# =========================================
# d. Transpose
# =========================================

print("\nTranspose:")
print(df.head().T)


# =========================================
# e. Melt
# =========================================

melted_data = pd.melt(df,
                     id_vars=['Sex'],
                     value_vars=['Age','SGOT','Protime'],
                     var_name='Feature',
                     value_name='Value')

print("\nMelted Data:")
print(melted_data.head())


# =========================================
# f. Pivot
# =========================================

pivot_data = melted_data.pivot_table(index='Sex',
                                     columns='Feature',
                                     values='Value',
                                     aggfunc='mean')

print("\nPivot Table:")
print(pivot_data)
```
