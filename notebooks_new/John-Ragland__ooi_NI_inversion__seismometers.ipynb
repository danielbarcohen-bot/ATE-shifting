# Seismometer exploration

```python
import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
import ooipy
from ni_tools import calculate
```

```python
starttime = datetime(2017,7,1)
endtime = datetime(2017,7,8)
node1 = 'AXEC2'
node2 = 'AXCC1'
channel = 'HNZ'

data1 = ooipy.get_acoustic_data_LF(starttime, endtime, node1, channel=channel, verbose=True)
data2 = ooipy.get_acoustic_data_LF(starttime, endtime, node2, channel=channel, verbose=True)
```

```python
ds = xr.Dataset({
    'AXEC2':xr.DataArray(data1.data[:35000], dims=['time'], attrs=data1.stats),
    'AXCC1':xr.DataArray(data2.data[:35000], dims=['time'], attrs=data2.stats)
})
```

```python
ds.chunk({'time':3600*200})
```

```python
NCCFs = calculate.compute_NCCF_stack(ds, dim='time')
```
