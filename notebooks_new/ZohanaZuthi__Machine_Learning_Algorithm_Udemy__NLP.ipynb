## **Natural Language Processing**

at first we will clean the sentences:

1.lowercase all words.

2.remove punctuation

3.remove stopwords like those words that
doesn't have that much value

4.tokenize the words. meaning split the sentence into words

Next we create the bag of words:

1.Making a list of all unique words from your dataset.

2.Counting how many times each word appears in each review.



```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## **Importing the libraries**

```python
dataset=pd.read_csv('Restaurant_Reviews.tsv',delimiter='\t',quoting=3)
# as tsv file is a tab separated value so setting delimiter to tab means that after a tab there is a separated word
# In pandas panda might treat the quoting as a full word which might lead to an error
# by setting the quoting to 3 means csv.QUOTE_NONE
```

## **Cleaning the texts**

```python
import re
# for making lowercase letters
import nltk
# to remove all the stopwords
nltk.download('stopwords')
from nltk.corpus import stopwords
# for stemming meaning taking all the base words, after making them in their base form
# for example love=loveing, loved
from nltk.stem.porter import PorterStemmer
corpus=[]
# we have 1000 reviews in out dataset
for i in range(0,1000):
  # following sub function will transform every special character with space
  # in computer science ^ means not
  # we want changes taking place in dataset review column
  review=re.sub('[^a-zA-Z]',' ',dataset['Review'][i])
  # lowercase
  review=review.lower()
  # splitting sentence in words
  review=review.split()
  # stemming
  ps=PorterStemmer()
  all_stopwords=stopwords.words('english')
  all_stopwords.remove('not')
  # we are not taking stopwords
  # then we will apply stemming in the set
  review=[ps.stem(word) for word in review if not word in set(all_stopwords)]
  review=' '.join(review)
  corpus.append(review)
```

```python
print(corpus)
# here this programe also removing not from the sentence
# so we need to update a little bit
```

## **Creating the Bag of Words Model**

Tokenaization meaning to count the number of the words and vectorize it

```python
from sklearn.feature_extraction.text import CountVectorizer
# our target is to take the most frequent words
cv=CountVectorizer(max_features=1500)
# bag of words model
# now we will make a sparse matrix
# this fit transform method will make all the words in different column
x=cv.fit_transform(corpus).toarray()
# making dependent variable
y=dataset.iloc[:,-1].values
```

```python
len(x[0])
```

## **Splitting the dataset into the Trainig set and Test set**

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=0)
```

## **Training the Naive Bayes Model on the Training set**

we can also train the data using other classification methods like decision tree or logistic regression

```python
# from sklearn.naive_bayes import GaussianNB
# classifier=GaussianNB()
# classifier.fit(x_train,y_train)
# 0.73

# from sklearn.linear_model import LogisticRegression
# classifier = LogisticRegression(random_state = 0)
# classifier.fit(x_train, y_train)
# 0.775

from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion = 'entropy', random_state = 0)
classifier.fit(x_train, y_train)
# 0.75
# so i think logistic regression is much better for this dataset
```

## **Predicting the Test set results**

```python
y_pred=classifier.predict(x_test)
```

### **Making The Confusion Matrix**

```python
from sklearn.metrics import confusion_matrix,accuracy_score
cm=confusion_matrix(y_test,y_pred)
print(cm)
k=accuracy_score(y_test,y_pred)
print(k)
```
