# Preprocessing

```python
# rows with at least 1 NaN value
df[pd.isnull(df).any(axis=1)].head()
```

```python
# drop rows with NaN vlues
df = df.dropna() # removes rows with any NaN values
df = df.reset_index() # reset's row indexes in case any row was dropped
df.head()
```

```python
# dropping multiple columns
df = df.drop(['name', 'cabin', 'ticket'], axis=1) # we won't use text features for the initial basic models
df.head()
```

```python
# map feature values
df['sex'] = df['sex'].map({'female': 0, 'male': 1}).astype(int)
df['embarked'] = df['embarked'].dropna().map({'S': 0, 'C': 1, 'Q': 2}).astype(int)
df.head()
```

# Feature Engineering

```python
# lambda expression to create new features
def get_family_size(sibsp, parch):
  family_size = sibsp + parch
  return family_size
```

```python
# determine family size
df['family_size'] = df[['sibsp', 'parch']].apply(lambda x: get_family_size(x['sibsp'], x['parch']), axis=1)
df.head()
```

```python
# reorganize headers
df =df[['pclass', 'sex', 'age', 'sibsp', 'parch', 'family_size', 'fare', 'embarked', 'survived']]
df.head()
```

# Save Data

```python
# saving dataframe to CSV
df.to_csv('processed_titanic.csv', index=False)
```
