```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import pandas as pd
```

```python
Dataset_Path = "../dataset/csiro-biomass/train_eda.csv"
df = pd.read_csv(Dataset_Path)
df.head()
```

```python
target_columns = [
    'target',
    'Std_Scaled_Target',
    'MinMax_Scaled_Target',
    'Log_Scaled_Target',
    'Log2_Scaled_Target'
]
exclude_columns = [
    'Sampling_Date',
    'State',
    'Species',
    'target_name',
    'Pre_GSHH_NDVI',
    'Height_Ave_cm'
]
X_train, X_test, y_train, y_test = train_test_split(
    df.drop(columns=target_columns + exclude_columns), 
    df[target_columns[4]], 
    test_size=0.2, 
    random_state=42,
    shuffle=True
)
```

```python
X_train.shape, X_test.shape
```

```python
X_train_sampleId = X_train['sample_id']
X_test_sampleId = X_test['sample_id']

X_train_imagePath = X_train['image_path']
X_test_imagePath = X_test['image_path']

X_train.drop(columns=['sample_id', 'image_path'], inplace=True)
X_test.drop(columns=['sample_id', 'image_path'], inplace=True)


# just image embedding columns
X_train = X_train.filter(like='emb_', axis=1)
X_test = X_test.filter(like='emb_', axis=1)
```

```python
X_train.shape, X_test.shape
```

```python
rf = RandomForestRegressor(
    n_estimators=200, 
    max_depth=50,
    random_state=42, 
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

```python
def predict_and_evaluate(model, X, y, dataset_type="Validation"):
    # if dataset_type == "Testing" or dataset_type == "Training":
    #     X = X.copy()
    #     X = X.filter(like='emb_', axis=1)
    #     print(f"Number of features used for {dataset_type}: {X.shape[1]}")

    y_pred = model.predict(X)
    mse = mean_squared_error(y, y_pred)
    r2 = r2_score(y, y_pred)
    print(f"{dataset_type} MSE: {mse:.4f}")
    print(f"{dataset_type} R^2: {r2:.4f}")
```

```python
predict_and_evaluate(rf, X_train, y_train, dataset_type="Training")
predict_and_evaluate(rf, X_test, y_test, dataset_type="Testing")
```

```python
# run-1
# params - n_estimators=150, random_state=42
predict_and_evaluate(rf, X_train, y_train, dataset_type="Training")
predict_and_evaluate(rf, X_test, y_test, dataset_type="Testing")
```

```python
# run-2
# params - n_estimators=100, random_state=42, n_jobs=-1
predict_and_evaluate(rf, X_train, y_train, dataset_type="Training")
predict_and_evaluate(rf, X_test, y_test, dataset_type="Testing")
```
