```python
# VIF -- Variance Inflation Factor

#It is going to check and measure the multicollinearity scores
```

```python
X = df.drop(['charges'], axis = 1) #axis = 1 =columns
```

```python
X
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
vif_data =pd.DataFrame()
```

```python
vif_data['column_name'] = X.columns
vif_data['VIF Values'] = [variance_inflation_factor(X.values,i) for i in range(len(X.columns))]
```

# Model Building

```python
x = df.drop(['charges'], axis = 1)
y = df['charges']
```

```python
x
```

```python
y
```

```python
df.sample(5, random_state = 6)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.2, random_state =7 )
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
# Training the model

model.fit(x_train, y_train)
```

```python
# Testing the model , predictions

y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
# accuracy metrics are used to check if the answers are correct or not

# r2_score, rmse, mae, mse
```

```python
from sklearn.metrics import r2_score
```

```python
print(r2_score(y_test,y_pred))
```
