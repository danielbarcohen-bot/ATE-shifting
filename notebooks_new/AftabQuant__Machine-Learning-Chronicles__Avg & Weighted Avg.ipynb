```python
x = df[['households', 'median_income']]
y = df['median_house_value']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
svm = SVR()
knn = KNeighborsRegressor()
```

```python
model = {
    'lr' : lr,
    'dt' : dt,
    'svm' : svm,
    'knn' : knn,
}
```

```python
def train_regression(reg, x_train, x_test, y_train, y_test):
    reg.fit(x_train, y_train)
    y_pred = reg.predict(x_test)
    test = reg.score(x_test, y_test)
    train = reg.score(x_train, y_train)
    return test, train
```

```python
for name, reg in model.items():
    test, train = train_regression(reg, x_train, x_test, y_train, y_test)
    print(f"For {name}:, test - {test*100:.2f}, train - {train*100:.2f}")
```

```python
from sklearn.ensemble import VotingRegressor
```

```python
li = [('dt1', DecisionTreeRegressor()), ('lr1', LinearRegression()), ('svm1', SVR()), ('knn1', KNeighborsRegressor())]
```

```python
vc = VotingRegressor(li)
vc.fit(x_train, y_train)
```

```python
test, train  = train_regression(vc, x_train, x_test, y_train, y_test)
```
