```python
# =========================================
# Adult Dataset (ML - Income Prediction)
# =========================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder


# =========================================
# Load dataset WITHOUT header
# =========================================

df = pd.read_csv('adult.csv', header=None)

# Assign simple column names
df.columns = ['age', 'workclass', 'fnlwgt', 'education', 'edu_num',
              'marital_status', 'occupation', 'relationship',
              'race', 'sex', 'capital_gain', 'capital_loss',
              'hours', 'country', 'income']

print("Original Data:")
print(df.head())


# =========================================
# m. Data Cleaning
# =========================================

# Replace '?' with NaN
df.replace('?', np.nan, inplace=True)

# Remove missing values
df.dropna(inplace=True)

print("\nAfter Cleaning:")
print(df.head())


# =========================================
# n. Outlier Removal
# =========================================

# Use only numeric columns
num_cols = df.select_dtypes(include=np.number)

Q1 = num_cols.quantile(0.25)
Q3 = num_cols.quantile(0.75)
IQR = Q3 - Q1

# Remove outliers
df = df[~((num_cols < (Q1 - 1.5 * IQR)) | 
          (num_cols > (Q3 + 1.5 * IQR))).any(axis=1)]

print("\nAfter Outlier Removal:")
print(df.head())


# =========================================
# o. Data Transformation
# =========================================

# Convert categorical columns to numeric using Label Encoding
le = LabelEncoder()

for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = le.fit_transform(df[col])

print("\nAfter Transformation:")
print(df.head())


# =========================================
# p. Model Building
# =========================================

X = df.drop('income', axis=1)
y = df['income']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)


# Logistic Regression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_test)
acc_lr = accuracy_score(y_test, y_pred_lr)

print("\nLogistic Regression Accuracy:", acc_lr)


# Naive Bayes
nb = GaussianNB()
nb.fit(X_train, y_train)

y_pred_nb = nb.predict(X_test)
acc_nb = accuracy_score(y_test, y_pred_nb)

print("Naive Bayes Accuracy:", acc_nb)


# Compare
if acc_lr > acc_nb:
    print("\nLogistic Regression is better")
else:
    print("\nNaive Bayes is better")
```
