### Descri??o das colunas

* **`date`** - Data do registro
* **`currency_code`** - C?digo ISO 4217 internacional da moeda do pa?s
* **`name`** - Nome do pa?s
* **`local_price`** - Pre?o do Big Mac na moeda local
* **`dollar_ex`** - Taxa de Convers?o do d?lar americano para a moeda local
* **`local_price`** - Pre?o do Big Mac em d?lares americanos


#### Vamos criar uma coluna cujo resultado ? 1 se o bigmac custa mais do que 4 d?lares e 0 se custa igual ou menos do que 4 d?lares

```python
def e_caro(preco):
    if preco > 4:
        return 1
    return 0
```

```python
df['bigmac_caro'] = df['dollar_price'].apply(e_caro)
```

```python
df.head()
```

```python
df.describe()
```

#### Vamos fazer um Intervalo com 95% Confian?a para a propor??o de Big Mac caros

```python
big_mac_caro_prop = df["bigmac_caro"].mean()
big_mac_caro_prop_desvio_padrao = (big_mac_caro_prop * (1 - big_mac_caro_prop)) ** 0.5

print('\n')
print("Propor??o de Big Mac's caros: {}".format(big_mac_caro_prop))
print("Desvio Padr?o da propor??o de Big Mac's caros: {}"\
      .format(big_mac_caro_prop_desvio_padrao))
```

```python
from scipy.stats import t

# df = graus de liberdade , que vai ser o tamanho da amostra menos 1.
# loc = m?dia da amostra
# scale = desvio padr?o da amostra

graus_de_liberdade = df.shape[0] - 1

intervalo_95_big_mac_prop = t.interval(0.95 , 
                                  df = graus_de_liberdade,
                                  loc = big_mac_caro_prop,
                                  scale = big_mac_caro_prop_desvio_padrao)


print('Intervalo de 95% de Confian?a para Propor??o: {}'.format(intervalo_95_big_mac_prop))

```

```python
margem_de_erro_big_mac_prop = intervalo_95_big_mac_prop[1] - big_mac_caro_prop
print('Margem de erro do intervalo: {}'.format(margem_de_erro_big_mac_prop))
```
