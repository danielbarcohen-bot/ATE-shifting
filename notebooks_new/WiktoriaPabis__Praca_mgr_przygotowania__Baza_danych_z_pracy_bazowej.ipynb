```python
# Zamiast wyrzucać cały pomiar w przypadku stwierdzenia nieprawidłowej pojedynczej wartości zamieniam ją na nan, 
# dając szansę wykorzystania danych w przypadku, gdy w danym wierszu ilość danych w ustalonych przeze mnie dopuszczlanych 
# zakresach mieści się w kryteriach ustalonych później do odrzucenia danego wiersza (więcej niż 2 nan)

# Zastąpienie wartości D.O. (mg/l), które są mniejsze niż 0 lub większe niż 100, wartością np.nan
df.loc[(df['D.O. (mg/l)'] < 0.6) | (df['D.O. (mg/l)'] > 100), 'D.O. (mg/l)'] = np.nan

# Zastąpienie wartości PH, które są mniejsze niż 0 lub większe niż 14, wartością np.nan
df.loc[(df['PH'] < 0) | (df['PH'] > 14), 'PH'] = np.nan

# Zastąpienie wartości CONDUCTIVITY (µmhos/cm), które są mniejsze niż 0 lub większe niż 100000, wartością np.nan
df.loc[(df['CONDUCTIVITY (µmhos/cm)'] < 11) | (df['CONDUCTIVITY (µmhos/cm)'] > 9753), 'CONDUCTIVITY (µmhos/cm)'] = np.nan

# Zastąpienie wartości B.O.D. (mg/l), które są mniejsze niż 0 lub większe niż 10000, wartością np.nan
df.loc[(df['B.O.D. (mg/l)'] < 0) | (df['B.O.D. (mg/l)'] > 88), 'B.O.D. (mg/l)'] = np.nan

# Zastąpienie wartości NITRATENAN N+ NITRITENANN (mg/l), które są mniejsze niż 0 lub większe niż 150, wartością np.nan
df.loc[(df['NITRATENAN N+ NITRITENANN (mg/l)'] < 0) | (df['NITRATENAN N+ NITRITENANN (mg/l)'] > 21), 'NITRATENAN N+ NITRITENANN (mg/l)'] = np.nan

# Zastąpienie wartości FECAL COLIFORM (MPN/100ml), które są mniejsze niż 0 lub większe niż 200000000, wartością np.nan
df.loc[(df['FECAL COLIFORM (MPN/100ml)'] < 0) | (df['FECAL COLIFORM (MPN/100ml)'] > 2350), 'FECAL COLIFORM (MPN/100ml)'] = np.nan

# Zastąpienie wartości TOTAL COLIFORM (MPN/100ml)Mean, które są mniejsze niż 0 lub większe niż 200000000, wartością np.nan
df.loc[(df['TOTAL COLIFORM (MPN/100ml)Mean'] < 0) | (df['TOTAL COLIFORM (MPN/100ml)Mean'] > 3000), 'TOTAL COLIFORM (MPN/100ml)Mean'] = np.nan
```

```python
# do konsultacji, ja przykładowo wyrzucam przypadki, gdy brakujących danych jest powyżej 2 wartości w każdym recordzie, 
# czyli wyrzucone zostanie (przed filtracją min max pam 3,3%) obecnie 5,2% danych 

# usunięcie wierszy, w których są conajmniej 3 puste wartości, a conajmniej 5 występuje z uwzględnieniem tylko kolumn 4-10
# df_filtered = df.dropna(subset=df.columns[4:11], thresh=5).copy()
# df_filtered.reset_index(drop=True, inplace=True)
```

```python
df_filtered = df.copy()
# Policzenie mediany dla kolumn 4-10
median_values = df_filtered.iloc[:, 4:11].median()
# Zastąpienie brakujących wartości medianą dla odpowiednich kolumn
df_filtered.iloc[:, 4:11] = df_filtered.iloc[:, 4:11].fillna(median_values)
```

```python
# do wyliczenia WQI jako etykiet mam 7 parametrów w każdym recordzie, przy czym przynajmniej 5 posiada wartość liczbową w każdym recordzie (NA TEN MOMENT NIEAKTUALNE)
```

```python
# Vi to wartość parametru- próbki wody

# wartości referencyjne 
Videal = {
  'D.O. (mg/l)': 14.6,                   
  'PH': 7.0,                             
  'CONDUCTIVITY (µmhos/cm)': 0,      
  'B.O.D. (mg/l)': 0,                     
  'NITRATENAN N+ NITRITENANN (mg/l)': 0, 
  'FECAL COLIFORM (MPN/100ml)': 0,        
  'TOTAL COLIFORM (MPN/100ml)Mean': 0  
}

# wartość oczekiwana dla każdego z parametrów czystej wody 
Si = {
    'D.O. (mg/l)': 10,                   
    'PH': 8.5,                             
    'CONDUCTIVITY (µmhos/cm)': 1000,        
    'B.O.D. (mg/l)': 45,                     
    'NITRATENAN N+ NITRITENANN (mg/l)': 5,  
    'FECAL COLIFORM (MPN/100ml)': 100,        
    'TOTAL COLIFORM (MPN/100ml)Mean': 1000 
}



# waga każdego z parametrów
def calculate_wi(S_i,valid_Si):
    # stała proporcjonalności
    K = 1/sum(valid_Si.values())
    return (K/S_i) 

# skala oszacowania jakości
def calculate_qi(Vi, param):
    return 100 * (Vi - Videal.get(param)) / (Si.get(param) - Videal.get(param))

# indeks jakości wody
def calculate_wqi(row):
    sum_qi_wi = 0
    sum_wi = 0



    valid_Si = Si.copy()
    for param in ['D.O. (mg/l)', 'PH', 'CONDUCTIVITY (µmhos/cm)', 'B.O.D. (mg/l)', 'NITRATENAN N+ NITRITENANN (mg/l)', 
                  'FECAL COLIFORM (MPN/100ml)', 'TOTAL COLIFORM (MPN/100ml)Mean']:   
        if pd.isna(row[param]):
            valid_Si[param] = 0
   
    for param in ['D.O. (mg/l)', 'PH', 'CONDUCTIVITY (µmhos/cm)', 'B.O.D. (mg/l)', 'NITRATENAN N+ NITRITENANN (mg/l)', 
                  'FECAL COLIFORM (MPN/100ml)', 'TOTAL COLIFORM (MPN/100ml)Mean']:                                  
        if not pd.isna(row[param]):
            qi = calculate_qi(row[param], param)
            wi = calculate_wi(Si[param], valid_Si)
            sum_qi_wi += (qi * wi)
            sum_wi += wi
    return round(sum_qi_wi / sum_wi,2)
    
print(len(df_filtered))
df_filtered.shape

for idx, row in df_filtered.iterrows():
    if idx in df_filtered.index:
        df_filtered.loc[idx, 'WQI'] = calculate_wqi(row)

```

```python
import matplotlib.pyplot as plt
from statsmodels.distributions.empirical_distribution import ECDF
from scipy.stats import norm
%matplotlib inline

#wycignięcie tylko kolumny temperatura
temperature = df_filtered.iloc[:, 3]

# usuniecie wszystkich NaN w celu uzyskania dystrybuanty do 100%

all_values = temperature.shape[0]
print("Liczba wszystkich wartości:", all_values)

count_nan = temperature.isna().sum()
print("Liczba wartości będcych NaN:", count_nan)

count_not_nan = temperature.notna().sum()
print("Liczba wartości niebędcych NaN:", count_not_nan)


df_filteredtemp = temperature.dropna()


all_values = df_filteredtemp.shape[0]
print("Liczba wszystkich wartości po przefiltrowaniu:", all_values)

count_nan = df_filteredtemp.isna().sum()
print("Liczba wartości będacych NaN po przefiltrowaniu:", count_nan)

count_not_nan = df_filteredtemp.notna().sum()
print("Liczba wartości niebędcych NaN po przefiltrowaniu:", count_not_nan)


# obliczenie dystrybuanty empirycznej
ecdf = ECDF(df_filteredtemp)

# dopasowanie rozkładu normalnego do danych
mean_temp = np.mean(df_filteredtemp)
std_temp = np.std(df_filteredtemp)

# generowanie teoretycznej dystrybuanty normalnej
x = np.linspace(min(df_filteredtemp), max(df_filteredtemp), 1000)
normal_cdf = norm.cdf(x, loc=mean_temp, scale=std_temp)

plt.figure(figsize=(10, 6))
plt.plot(x, normal_cdf, label="Teoretyczna dystrybuanta normalna", color="orange")
plt.step(ecdf.x, ecdf.y, label="Dystrybuanta empiryczna", color="blue", where="post")
plt.xlabel("Temperatura")
plt.ylabel("Dystrybuanta")
plt.title("Dystrybuanta empiryczna i teoretyczna rozkładu normalnego")
plt.legend()
plt.grid()
plt.show()
```
