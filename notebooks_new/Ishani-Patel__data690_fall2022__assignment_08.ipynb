###6. Cleanse the data to make the year, population, cost, cost per person numerical type

```python
def cleanse_year(a_string):
  if '*' in a_string:
    return int(a_string.replace("*",""))
  return int(a_string)
```

```python
def cleanse_total_pop(a_string):
  return int(a_string.replace(",",""))
```

```python
cost="$123,456,789"
type(int(cost.replace("$","").replace(",","")))
```

```python
def cleanse_census_cost(a_string):
  if '$' in a_string and 'Billion' in  a_string:
    a_string = float(a_string.replace("$","").replace("Billion",""))
    return int(a_string * 1000000000) 
  return int(a_string.replace("$","").replace(",",""))
```

```python
test_list=["$123","123 cents"]
test_list=[int(num.replace("$","").replace(" cents","")) for num in test_list]
```

```python
test_list
```

```python
def cleanse_avg_cost(a_string):
  if '$' in a_string:
    return float(a_string.replace("$",""))
  if 'cents' in a_string:
    a_string = a_string.replace("cents","").strip()
    return float(a_string)/100.00
```

```python
df["Census Year"] = df["Census Year"].apply(cleanse_year)
df["Total Population"] = df["Total Population"].apply(cleanse_total_pop)
df["Census Cost"]=df["Census Cost"].apply(cleanse_census_cost)
df["Average Cost Per Person"]=df["Average Cost Per Person"].apply(cleanse_avg_cost)
```

###8. Use Plotly to plot the average cost per person over time.

###9. Use Plotly to plot the scatter plot of Total Population vs Total Cost.

```python
fig = px.scatter(
    df,
    x="Census Cost",
    y="Total Population", 
    title="Total Population vs Total Cost", 
    color="Census Year",
    template="plotly_dark"
)

fig.update_layout(showlegend=False)

fig.show()
```
