```python
def build_losses(confounder_dic):
    losses = {}
    lossWeights = {}
    for k in confounder_dic:
        losses[k] = correlation_coefficient_loss
        lossWeights[k] = 1.0/len(confounder_dic)
    return losses,lossWeights
```

```python
def train_gan(gan, training_data, labels, confounders, epochs=50, batch_size = 10):
    ce_confounder_loss = []
    pbs_confounder_loss = []
    for epoch in range(epochs):
        print("Epoch:",epoch)
        ce_confounder_loss_batch = []
        pbs_confounder_loss_batch = []
        #for b in range(len(training_data)//batch_size):
        for b in range(50):
            #idx_batch = numpy.array(list(range(b*batch_size,(b+1)*batch_size)))
            idx_batch = numpy.random.choice(range(len(training_data)),batch_size,replace=False)
            training_batch = training_data[idx_batch]
            confounder_batch = [ c[idx_batch] for c in confounders ]
            labels_batch = labels[idx_batch]


            # ---------------------
            #  Train confounder predictor
            # ---------------------
            encoded_features = gan.encoder.model.predict(training_batch)
            r_loss = gan.regressor.model.train_on_batch(encoded_features, confounder_batch)
            ce_confounder_loss_batch.append(r_loss)
            #print("regressor loss (CE)",r_loss)

            # ---------------------
            #  Train Distiller
            # ---------------------
            g_loss = gan.distiller.train_on_batch(training_batch, confounder_batch)
            pbs_confounder_loss_batch.append(g_loss)
            #print("distiller loss (PBS)",g_loss)
            

            # ---------------------
            #  Train Encoder & Classifier
            # ---------------------
            c_loss = gan.workflow.train_on_batch(training_batch, labels_batch)
        print("predictor loss",c_loss)
            
        ce_confounder_loss.append(numpy.array(ce_confounder_loss_batch))
        pbs_confounder_loss.append(numpy.array(pbs_confounder_loss_batch))
    return numpy.mean(ce_confounder_loss,axis=1),numpy.mean(pbs_confounder_loss,axis=1)
            
def predict_gan(gan, input_data):
    return gan.workflow.predict(input_data)            

def my_mse(model,feature,true):
    return numpy.sqrt(numpy.mean((predict_gan(model,feature)[:,0]-true)**2))

def my_mae(model,feature,true):
    return numpy.mean(numpy.abs(predict_gan(model,feature)[:,0]-true))
```
