## Ejercicio 3: API REST con FastAPI + Langgraph 

**Objetivo:** Crear una API para la conversacion con un agente react. El agente debe contar con dos herramientas 'math_calculator' para realizar calculos matematicos simples y 'datetime_info' la cual es una funcion sin parametros que retorna informacion de fecha y hora en formato ISO. La API debe contener un endpoint para cambiar la zona horaria de la tool 'datetime_info', un endpoint para reiniciar la conversacion y un enpoint para conversar con el agente. El estado del agente puede ser almacenado de la forma que considere mas conveniente.

**Requisitos:**
1. Endpoint POST que reciba una pregunta
2. Endpoint DELETE que reinicie el estado del contexto
3. Endpoint POST que reciba una zona horaria
2. Validación de entrada con Pydantic
3. Integración con LangGraph para el agente
4. Manejo de errores y rate limiting básico
5. Respuesta estructurada con metadata (tokens, tiempo de respuesta)
6. La implementacion debe ser capaz de responder consultar del estilo  'Calcula el logaritmo del dia de hoy, luego eleva el resultado por el numero de este mes' a travez de un solo llamado
