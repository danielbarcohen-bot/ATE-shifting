```python
import pandas as pd

# Load dataset from local file
data = pd.read_csv("diabetes.csv")

print("Dataset Loaded Successfully!")
print(data.head())

```

```python
# Remove the class/label column
X = data.drop("Outcome", axis=1)

print("Feature Data (X):")
print(X.head())
```

```python

from sklearn.preprocessing import MinMaxScaler

# Min-Max scaling
scaler = MinMaxScaler(feature_range=(0, 1))
X_minmax = scaler.fit_transform(X)

# Convert back to DataFrame
X_minmax_df = pd.DataFrame(X_minmax, columns=X.columns)

print("\nMin-Max Scaled Data (Head):")
print(X_minmax_df.head())
```

```python
from sklearn.preprocessing import Binarizer

# Binarization with threshold 0.0
binarizer = Binarizer(threshold=0.0)

X_binary = binarizer.fit_transform(X)

# Convert to DataFrame
X_binary_df = pd.DataFrame(X_binary, columns=X.columns)

print("\nBinarized Data (Head):")
print(X_binary_df.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Standardization (Z-score normalization)
std_scaler = StandardScaler()
X_standard = std_scaler.fit_transform(X)

# Convert back to DataFrame
X_standard_df = pd.DataFrame(X_standard, columns=X.columns)

print("\nStandardized Data (Head):")
print(X_standard_df.head())
```
