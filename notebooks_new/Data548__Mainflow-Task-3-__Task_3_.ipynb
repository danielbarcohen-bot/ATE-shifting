```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('householdtask3.csv')
```

```python
data.head(10)
```

```python
#Scatter plot between year against own column
plt.scatter(data['year'], data['own'])

#Adding title to the plot
plt.title("Scatter Plot")

#Setting the x and y lables
plt.xlabel('year')
plt.ylabel('own')

#Showing the results
plt.show()
```

```python
#Line chart between year and own
plt.plot(data['year'])
plt.plot(data['own'])

#Adding title to the plot
plt.title("Line Chart")

#Setting the x and y lables
plt.xlabel('year')
plt.ylabel('own')

#Showing the results
plt.show()
```

```python
#Line chart between year and own
plt.bar(data['year'],data['own'])

#Adding title to the plot
plt.title("Bar Chart")

#Setting the x and y lables
plt.xlabel('year')
plt.ylabel('own')

#Showing the results
plt.show()
```

```python
#Histogram
plt.hist(data['income'])

plt.title("Histogram")

plt.show()
```
