### Registros duplicados

Los valores o registros duplicados en los datos es un problema prevalente, ya que consumen mucho espacio de almacenamiento y en memoria, lo que resulta en una inexactitud de conocimiento u objetivos del negocio. Lo mejor, es remover estos registros, pero para esto, es importante identificarlos y analizar los diferentes caminos para hacerlo. En pandas para el tratamioento del problema de registros duplicados cuenta con las funciones **DataFrame.duplicated()** y **DataFrame.drop_duplicates()**. 

Se consideran datos duplicados cuando en un registro tienen el mismo valor en todas las columnas, para este ejercicio nos apoyaremos del siguiente archivo:

```python
# carga de archivo csv con duplicados
df_dup = pd.read_csv('../data/organizations_duplicados.csv')
df_dup
```

Observamos que las filas con los **Id_org = [101, 104, 108]** son registros duplicados. Es muy importante analizar si un registro en verdad es duplicado y, s? es necesario eliminarlo, veamos los registros con **Id_org = [106, 107]**, se observa que hay coincidencias de valores en algunas columnas, pero esto no significa que necesariamente sea un registro duplicado, al final, ser? criterio del analista eliminar o no registros.

Para conocer el n?mero de registros duplicados podemos escribir el siguiente c?digo haciendo uso de las funciones **duplicated()** y **sum()**. 

```python
num_reg_dup = df_dup.duplicated().sum()
print("N?mero total de registros duplicados:", num_reg_dup)
```

Para esta tarea tambi?n podemos ayudarnos de la columna *Id_org*, ya que por su misma naturaleza se entiende que no debe duplicarse ning?n valor. 

```python
# se asignan los valores duplicados al objeto declarado
num_reg_dup_id_org = df_dup[ df_dup.duplicated('Id_org')]
num_reg_dup_id_org
```

Para eliminar los registros duplicados, pandas proporciona la funci?n **drop_duplicates()**, si pasamos una ?nica columna como argumento elimina solamente los valores duplicados en ella, pero si pasamos una lista de columnas eliminar? los valores duplicados en todas.

El c?digo siguiente muestra la eliminaci?n de registros duplicados para una columna en especifico (Id_org) y el sigueinte para varias columnas pasadas como una lista ([Name, Country, Founded]), as? como la suma de registros duplicados despu?s de ejecutar el c?digo correspondiente.

```python
df_dup_drop_id = df_dup.drop_duplicates('Id_org')
df_dup_drop_id
```

```python
num_reg_df_dup_drop_id = df_dup_drop_id.duplicated().sum()
print("N?mero de registros duplicados:", num_reg_df_dup_drop_id)
```

```python
df_dup_drop_colum = df_dup.drop_duplicates(['Name', 'Country', 'Founded'])
df_dup_drop_colum
```

```python
num_reg_df_dup_drop_column = df_dup_drop_colum.duplicated().sum()
print("N?mero de registros duplicados:", num_reg_df_dup_drop_column)
```

### Valores perdidos

Manipular valores perdidos es una tarea crucial, en ocasiones el enfoque m?s directo es eliminar los registros con los valores faltantes. Sin embargo, esto no funciona en todos los casos, pues podemos perder hechos importantes de los datos. Otra manera de manejar esta situaci?n es imputar los valores perdidos, sin embargo, no existe un m?todo ?nico para encontrar ?stos valores. Realizar la imputaci?n depender? de varios factores, como son el tipo de problema, la necesidad del negocio, afectaciones a los resultados esperados, entre otros m?s.

En pandas se tiene una lista importante de funciones para manipular los datos faltantes o perdidos, algunos de ellas se ven a continuaci?n.

* **Eliminar filas con valores perdidos**

Se tiene la funci?n **pandas.dropna()** para eliminar los registros con valores perdidos, esta funci?n recibe como par?metros *how=all* y *how=any*, para el primer caso, se requiere que todas las columnas de la fila presenten valores perdidos (NaN-Not a Number), para el segundo basta que alguna de las columnas exista un valor faltante para que sea eliminada la fila. A parte de los par?metros anteriores, tambi?n puede recibir el par?metro *subset=[col1, col2,...]*, el cual recibe una lista espec?fica de columnas donde se desea aplicar la eliminaci?n.  

Para mostrar los ejemplos de esta funci?n se usar? el archivo "organizations_mv.csv".

```python
df_na = pd.read_csv('../data/organizations_mv.csv')
df_na
```

```python
# borrado de registros donde todas sus columnas presenten valores perdidos (?ndices 5 y 7)
df_na.dropna(how='all')
```

```python
# borrado de registros donde alguna de sus columnas presenten valores perdidos (?nices 1, 2, 3, 5, 7 y 9)
df_na.dropna(how='any')
```

```python
# borrado de registros donde las columnas especif?cadas presentan valores perdidos (?ndices 2, 5, 7 y 9)
df_na.dropna(subset=['Name', 'Industry'])
```

* **Llenado o imputaci?n de valores perdidos**

Como se coment? no siempre lo mejor es eliminar los registros con valores perdidos, ya que en algunas ocasiones surge la necesidad de inferir los valores para imputarlos. Para realizar esta acci?n pandas posee las funciones **fillna()** y **replace()**, las cuales pueden ser usadas con varios par?metros dependiendo las necesidades; a continuaci?n, se craer? un DataFrame a partir de un diccionario para ejemplificar los m?todos mencionados.   

```python
import numpy as np # libreria especializada en el c?lculo num?rico
# np.nan significa insertar valores NaN (Not a Number) o valores perdidos
org_dict = {
    'Name': ['Abraham', 'James', 'NA', 'Victoria', 'Hinako'],
    'Department': ['Adversiting', np.nan, 'Accounting', '', 'Accounting'], 
    'Salary_USD': [7920, 8500, np.nan, 7450, 0], 
    'City': ['Madrid', 'Chicago', np.nan, 'Cali', 'Tokio']
    }

df_imp = pd.DataFrame(org_dict)
df_imp
```

**Nota:** En un DataFrame los valores NaN (Not a Number) significan que son valores perdidos, sin embargo, en pandas los valores de cero, espacios en blanco y NA no son considerados valores faltantes.  

* **DataFrame.fillna(<valor_de_llenado_o_imputaci?n>)**

En seguida se har? uso del m?todo **fillna()**, el cual recibe como par?metro el valor con el que se imputa el o los valores perdidos, en este caso un string o cadena de texto *'Valor Faltante'*. 

```python
df_imp_fillna = df_imp.fillna('Valor Faltante')
df_imp_fillna
```

El m?todo **ffill()** (forward fill) hace un llenado hacia adelante, es decir, el valor perdido se sustituye con el valor previo de la misma columna. 

```python
df_imp_fillna_ffill = df_imp.ffill()
df_imp_fillna_ffill
```

El m?todo **bfill()** (backward fill) hace una imputaci?n hacia atr?s, es decir, el valor perdido se sustituye con el valor de adelante de la misma columna. 

```python
df_imp_fillna_bfill = df_imp.bfill()
df_imp_fillna_bfill
```

Regresando con la funci?n **fillna()**, ahora se sustituir?n los valores perdidos con alg?n valor estad?stico de la propia columna n?merica, como son la media (*mean()*), mediana (*median()*) y moda (*mode()*). Para ejemplificar estas funciones imputaremos el valor perdido de la columna *Salary_USD*.  

```python
df_imp.describe().T # .T transpone los datos
```

```python
salary_mean = df_imp['Salary_USD'].fillna(df_imp['Salary_USD'].mean())
df_imp_fillna_salary_mean = pd.DataFrame(salary_mean)
df_imp_fillna_salary_mean
```

```python
salary_median = df_imp['Salary_USD'].fillna(df_imp['Salary_USD'].median())
df_imp_fillna_salary_median = pd.DataFrame(salary_median)
df_imp_fillna_salary_median
```

En el siguiente ejemplo se imputar?n valores perdidos con la moda, pero ser? para una columna categ?rica, es importante pasar el par?metro que indica que la imputaci?n ser? por filas [0].  

```python
department_mode = df_imp['Department'].fillna(df_imp.Department.mode()[0]) # cero hace referencia a los valores por fila
df_imp_fillna_department = pd.DataFrame(department_mode) 
df_imp_fillna_department
```

* **DataFrame.replace()**

Como se observ? con las funciones **fillna()**, **ffill()** y **bfill()** s?lo los valores NaN son imputados, pero no llena valores NA y espacios en blanco, para esta situaci?n se puede usar la funci?n **replace()**, a la cual se pasa como par?metros el valor actual y el valor nuevo, obteniendo un remplazo de estos valores. 

A continuaci?n, veremos algunos aplicaciones de la funci?n **replace()**.

```python
# seguiremos haciendo uso del DataFrame df_imp 
df_imp
```

```python
# se define un diccionario con los valores a reemplazar, la clave o key es el valor actual y value el valor de reemplazo
df_replace = df_imp.replace({'NA':'Not applicable', 0:5500, "":'Blank', np.nan:'Not a Number'})
df_replace
```

```python
# para una columna especif?ca 
salary_replace = pd.Series(df_imp['Salary_USD'].replace(
                                                        {0:5000, # valor cero es reemplazado por 5,000 
                                                         np.nan:df_imp['Salary_USD'].mean() # valor NaN es reemplazado por la media
                                                        })
                          )
salary_replace
```

## Agrupar y agregar

Cuando se necesita obtener un resumen de los datos se puede hacer uso de las funciones de agrupaci?n y agregaci?n. Para agrupar pandas posee la funci?n **groupby()** y para la agregaci?n tiene varias funciones, como son: **min()**, **max()**, **count()**, **sum()**, entre otras.

### Agrupar

Para conocer el uso de la funci?n **groupby()** se har? uso del archivo correspondiente al Censo Poblacional 2020 del Instituto Nacional de Estad?stica y Geograf?a (INEGI) para la Ciudad de M?xico, cabe mencionar que este solo contiene una parte del total de todas las columnas. 

A la funci?n **groupby()** se le pasar? el argumento 'NOM_MUN', el cual corresponde al nombre de las alcald?as, para fines pr?cticos se har? referencia a ?stas como municipios que es como se les conoce a estas delimitaciones territoriales en la mayor parte del pa?s. El objeto o variable que contiene la agrupaci?n define a los nombres de los municipios como keys o llaves.

```python
# crear el agrupamiento 
df_censo_cdmx_agrup_nom_mun = df_censo_cdmx.groupby('NOM_MUN')
print("Objeto agrupado: ", df_censo_cdmx_agrup_nom_mun)
```

```python
# obtener los valores agrupados o keys del objeto agrupado
grupos_mun = df_censo_cdmx_agrup_nom_mun.groups.keys() 
print("Valores agrupados:\n", grupos_mun)
```

Si se requiere agrupar por m?s de una columna se puede usar la funci?n *groupby()* con una lista de los nombres de las columnas a agrupar como argumento: **df.groupby(['col1, col2, ..., coln'])**.

### Agregar

Despu?s de mostrar el uso de la funci?n de agrupar, lo siguiente es hacer uso de las funciones de agregaci?n, la s?ntaxis puede verse as?: **df.groupby([col1, col2...]).funci?n_de_agregaci?n()**. Lo primero es agrupar por la columna de municipio (NUM_MUN) y aplicar la funci?n de agregaci?n de suma [*sum()*] para la columna poblaci?n total (POBTOT).

```python
df_censo_cdmx.groupby(['NOM_MUN'])[['POBTOT']].sum()
```

Existe otra forma de usar las funciones de agregaci?n, para esto es necesario es hacer uso de otra funci?n, siendo la siguiente: **agg([funci?n_de_agregaci?n1(), funci?n_de_agregaci?n2()...])**, las funciones de agregaci?n se pueden pasar a la funci?n **agg()** como una lista o un diccionario. En el siguiente ejemplo retomamos la agrupaci?n de los municipios, las funciones de agregaci?n aplicaran a la columna de poblaci?n total, siendo: valor m?nimo y m?ximo, suma, conteo, media y desviaci?n est?ndar, para esta ?ltima se obtienen valores NaN por ser un solo ser un registro desagregado.

```python
df_censo_cdmx_agg_pob = df_censo_cdmx.groupby(['NOM_MUN'])['POBTOT'].agg(['min', 'sum', 'mean', 'std', 'max', 'count'])
df_censo_cdmx_agg_pob
```

### Fusionar

B?sicamente la funci?n **merge()** toma dos DataFrames como par?metros de entrada y retorna un DataFrame fusionado como salida. La sintaxis de la funci?n **merge()** y algunos de los par?metros esenciales son: 

**pandas.merge(left, right, how='inner', on=None, left_on=None, right_on=None)**

* **left:** DataFrame izquierdo.
* **right:** DataFrame derecho. 
* **how:** Aqu? se indica c?mo se desea fusionar los DataFrames, similar a SQL: ['inner', 'left', 'right', 'outer', 'cross'], por defecto se aplica 'inner'.
* **on:** La columna de identificador o ?ndices que deben estar presentes en ambos DataFrames.
* **left_on:** Columna de identificador o ?ndices con el cual se debe fusionar el DataFrame de la izquierda.
* **right_on:** Columna de identificador o ?ndices con el cual se debe fusionar el DataFrame de la derecha.

A continuaci?n se presentan algunos ejemplos de la funci?n **merge()** usando los par?metros antes descritos para su mejor comprensi?n. 

* Cuando ambos DataFrames poseen el mismo nombre de la columna de identificador o ?ndice

Se observa que los DataFrames de mun_df y loc_df poseen una columna en com?n (id_mun), esta posee los ?ndices con los cuales se van a fusionar los registros que tienen en com?n el valor de dicha columna, el nombre de esta columna se pasa en el par?metro 'on' (**on='id_mun'**), asimismo, para este primer ejemplo en el par?metro 'how' se definir? con 'inner'(**how='inner'**), esto significa que solo se fusionar?n los registros con coincidencias o intersecci?n en ambos DataFrames.

```python
df_merge_inner = pd.merge(mun_df, loc_df, on='id_mun', how='inner')
df_merge_inner
```

Para el siguiente ejemplo en el parametro 'how' se asigna el string 'left'(**how='left'**), es decir, fusionar? aquellos registros donde solo exista coincidencia con los del DataFrame izquierdo. Considere que en caso de no encontrar valores coincidentes en las columnas anexadas se asignara valores NaN. 

```python
df_merge_left = pd.merge(mun_df, loc_df, on='id_mun', how='left')
df_merge_left
```

En el siguiente se asigna 'right' en el parametro 'how' (**how='right'**), lo que significa que fusionar? aquellos registros donde solo exista coincidencia con los del DataFrame derecho. De la misma manera, de no encontar valores existentes en las columnas que se devuelven en el DataFrame fusionado se asignara valores NaN.

```python
df_merge_right = pd.merge(mun_df, loc_df, on='id_mun', how='right')
df_merge_right
```

Ahora en el parametro 'how' se asigna 'outer'(**how='outer'**), con lo cual se obtienen absolutamente todos registros exista o no coincidencia de los valores de la columna del identificador.

```python
df_merge_outer = pd.merge(mun_df, loc_df, on='id_mun', how='outer')
df_merge_outer
```

Con la asignaci?n de 'cross' en el par?metro 'how' (**how='cross'**) se obtiene un producto cartesiano, en este caso no es necesario el uso del par?metro *on=*. Dicho de otra manera, el total de filas del DataFrame resultante es igual al n?mero de registros del primer DataFrame multiplicado por el total registros del segundo DataFrame.  

```python
df_merge_cross = pd.merge(mun_df, loc_df, how='cross')
df_merge_cross
```

Cuando los DataFrames a fusionar no poseen el mismo nombre en la columna con los valores coincidentes pandas posee los par?metros **left_on=None** y **right_on=None**, con estos pr?metros se se?alan los nombres de las columnas a intersectar.
```python
df_merge_right_on = pd.merge(mun_df, loc_df_id, left_on='id_mun', right_on='id', how='inner')
df_merge_right_on
```

### Unir

La funci?n **join()** es usada para unir columnas de otro DataFrame, es similar a la funci?n **merge()**, sin embargo, pero la funci?n **join()** realiza la uni?n con base en los ?ndices del DataFrame. La sintaxis as? como par?metros m?s significativos de esta funci?n son: 

**DataFrame.join(other, on=None, how='left', lsuffix='', rsuffix='', validate='str')**

* **other:** Un DataFrame, serie o lista que contenga la combinaci?n de estos. 
* **on:** Columna de identificador que debe estar presentes en ambos DataFrames.
* **how:** Forma de uni?n de los DataFrames ['inner', 'left', 'right', 'outer', 'cross'], por defecto se aplica 'inner'.
* **lsuffix:** Sufijo para los nombres de las columnas superpuestas en el DataFrame izquierdo.
* **rsuffix:** Sufijo para los nombres de las columnas superpuestas en el DataFrame derecho.
* **validate:** Si se espec?fica, valida que la uni?n sea del tipo espec?ficado, es decir, comprueba la cardinalidad (1:1, 1:m, m:1, m:m).

Para conocer el uso de la funci?n **join()** se har? uso de los DataFrames que se han utilizado, se asignaran los datos a nuevas variables con ?ndices de una lista mediante el par?metro **index=**. 

```python
mun_df_idx = pd.DataFrame({'id_mun':[8, 9, 4, 13, 12], 
                    'nom_mun':['La Magdalena Contreras', 'Milpa Alta', 'Cuajimalpa de Morelos', 'Xochimilco', 'Tlalpan']},
                          index=[1, 2, 3, 4, 5]
                    )
mun_df_idx
```

```python
loc_df_idx = pd.DataFrame({'id_mun':[8, 9, 4, 13, 11], 
                    'nom_loc':['Rancho Viejo', 'San Antonio Tec?mitl', 'Cruz Blanca', 'Paraje Zacapa', 'San Andr?s Mixquic']},
                          index=[0, 2, 3, 5, 6]
                   )
loc_df_idx
```

En el siguiente c?digo se realiza la uni?n de los DataFrames anteriores con base en los ?ndices asignados, el DataFrame de municipios (mun_df_idx) tomar? lo posici?n de derecha y el de localidades (loc_df_idx) la izquierda, para identificar las columnas de coincidencia procedentes de cada DataFrame se hace uso de los par?metros **lsuffix=''** y **rsuffix=''**, a los cuales se pasa una cadena de texto, asimismo, se pasa al par?metro **how=''** con el instring 'inner', para obtener una superposici?n solo de intersecci?n de valores coincidentes de los ?ndices.
```python
joined_df_idx = mun_df_idx.join(loc_df_idx, lsuffix='_suf_izq', rsuffix='_suf_der', how='inner')
joined_df_idx
```
