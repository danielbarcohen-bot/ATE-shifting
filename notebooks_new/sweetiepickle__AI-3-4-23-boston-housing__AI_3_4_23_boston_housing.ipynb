```python
from keras.datasets import boston_housing
(x_train, y_train), (x_test, y_test)= boston_housing.load_data()
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_train.dtype)
print(y_train.dtype)
print(x_test.shape)
print(y_test.shape)
print(x_test.dtype)
print(y_test.dtype)
```

```python
# display data
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from keras import models
from keras.models import Sequential 
from keras.layers import Dense    # full-connected
model= Sequential()
```

```python
model.add(Dense(800,activation='relu',input_shape=(13,),name='layers1'))
model.add(Dense(200,activation='relu',name='layers2'))
model.add(Dense(1,activation='linear',name='output'))
# input_shape : t?n hi?u v?o 
# activation : h?m truy?n
# 50 : ??u ra
model.summary()
```

```python
# compile
model.compile(optimizer='Adam',loss='mean_absolute_error',metrics=['accuracy'])
model.fit(x_train, y_train, epochs= 10)
test_loss, test_acc= model.evaluate(x_test, y_test)
print('?? ch?nh x?c tr?n t?p ki?m tra: ', test_acc)
print('H?m m?t m?t tr?n t?p ki?m tra: ', test_loss)
```
