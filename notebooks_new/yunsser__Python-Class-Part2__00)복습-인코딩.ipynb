```python
# ?? ?? -> ?? :: LabelEncoder? ??
le = LabelEncoder()
labels = le.fit_transform(items)
print(labels)

## ?? labels? -> 2????... 
labels_2d = labels.reshape(-1,1)

## ?? ??? ??
oh_e = OneHotEncoder()
labels_oh = oh_e.fit_transform(labels_2d)

print('\n oh_data:\n',labels_oh.toarray()) 
print('\n oh_shape:\n',labels_oh.toarray().shape) 
```
