### ??3. Improve The Simple Time Series Plot By Calculating A Running 10-Year Average For The Columns "Anomaly" And "Uncertainty".


We now want to calculate a 10-year mean value for the columns "anomaly" and "uncertainty" and plot them.
The moving average is calculated by combining the methods `rolling()` and `mean()`.


- The `rolling()` method takes several arguments. We need to find a suitable value for the `window` argument. We also set `center=True`.
- After the `rolling()` method has been called, it is concatenated with the `mean()` method to calculate the 10-year average. Just like in the task above, will use the `plot()` method to check the result of the calculation.
- We will then repeat the procedure for the columns "anomaly" and "uncertainty" and add the two columns to the data frame, labeled "10-year-anomaly" and "10-year-uncertainty".
- Finally we will plot the "anomaly", the "10-year-anomaly" and the "10-year-uncertainty" in a single plot. Note that the `uncertainty` corresponds to the absolute value of the deviation. We are calculating `10-year-anomaly` $+$ `10-year-uncertainty` and `10-year-anomaly` $-$ `10-year-uncertainty` to obtain the upper and lower bounds of uncertainty.


```python
df_10_y_anomaly = df.anomaly.rolling(window = 12*10, center = True).mean()
df_10_y_uncertainty = df.uncertainty.rolling(window = 12*10, center = True).mean()

df_upper_uncertainty = df_10_y_anomaly + df_10_y_uncertainty
df_bottom_uncertainty = df_10_y_anomaly - df_10_y_uncertainty
df.anomaly.plot(alpha=0.4, label="anomaly")
df_10_y_anomaly.plot(label="10-year-anomaly")

plt.legend()
```

```python
# Solution
def read_temp_data(FILEPATH):
    '''
    Read the temperature data from Berkeley Earth's open database
    '''
    COLUMN_NAMES = ["year", "month", "anomaly", "uncertainty"]
    df = pd.read_csv(FILEPATH,
                     comment="%", 
                     delim_whitespace=True, 
                     header=None, 
                     usecols=[0, 1, 2, 3], 
                     names=COLUMN_NAMES)
    return(df)
```

To ensure reproducibility, you should revise your code and write a function called `monthly_index` that has only a single input argument, a `DataFrame` object.


```python
# Solution:
def monthly_index(df):
    '''
    Provision of a time-based index for a data frame
    '''    
    ## Concatenate columns with strings
    date_string = df.year.astype(str) + "-" + df.month.astype(str)
    
    ## Creating a PeriodIndex object
    idx = pd.to_datetime(date_string).dt.to_period('M')

    ## Assign PeriodIndex to the DataFrame object
    df.set_index(idx, inplace=True) # use inplace=True to make the change permanent
    return(df)
```

### ??3. Improve The Simple Time Series Plot By Calculating A Running 10-Year Average For The Columns "Anomaly" And "Uncertainty".


To ensure consistency, revise the code and write a function called `plot_anomalies` with two input arguments, a `DataFrame` object and a `string` corresponding to the title of the plot.


```python
# perfect solution:

def plot_anomalies(df, title):
    '''
    Function for creating a time series plot
    '''
    
    # Calculation of time span (10 years --> 12 * 10 months)
    w = 12*10
    
    # Calculate the 10-year running average for the anomaly and add the result to the data frame
    df["10-year-anomaly"] = df.anomaly.rolling(window=w, center=True).mean()
    # Calculate the 10-year running mean for the uncertainty and add the result to the data frame
    df["10-year-uncertainty"] = df.uncertainty.rolling(window=w, center=True).mean()

    # Calculation of the upper and lower limits of the temperature anomaly
    upper_bound = df["10-year-anomaly"] + df["10-year-uncertainty"]
    lower_bound = df["10-year-anomaly"] - df["10-year-uncertainty"]
    
    # Plotting
    # Note that we use the matplotlib library, as this gives us even more control over the representation of the figure
    
    import matplotlib.pyplot as plt
    
    # Create figures and axes
    fig, ax = plt.subplots(figsize=(12,6))
    # Plot monthly anomaly
    df["anomaly"].plot(ax=ax, alpha=0.3, linewidth=0.5, color="blue")
    # Plot 10-year-anomaly
    df["10-year-anomaly"].plot(ax=ax, linewidth=2, color="red")
    # Fill the range between the upper and lower uncertainty limits
    ax.fill_between(df.index, lower_bound, upper_bound,color="gray", alpha=0.8)
    
    # Add line at y=0
    ax.axhline(y=0, linewidth=0.85, linestyle="--", color="black")

    # Set the figure title
    ax.set_title(title, size=18)
    # Set the label for the y-axis
    ax.set_ylabel("$^\circ$C")
    
    # Add a legend
    plt.legend()
    
    # plt.savefig(fname="global_temp_anomaly")   # Save as ...
    
    return(fig, ax)
```
