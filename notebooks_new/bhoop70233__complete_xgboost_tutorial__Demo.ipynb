```python
from xgboost import XGBClassifier #It is an implementation of gradient boosting that is specifically designed to be efficient and scalable, making it a popular choice for working with large datasets. Mathematically, XGBoost is an ensemble learning method that combines the predictions of multiple weak models to produce a strong prediction.
#read data
from sklearn.metrics import accuracy_score

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
data=load_iris()
```

```python
X_train,X_test,y_train,y_test=train_test_split(data['data'],data['target'],test_size=0.2)
```

```python
#create model instance
bst=XGBClassifier(n_estimators=2,max_depth=2,learning_rate=1,objective='binary:logistic')
```

```python
#fit model
bst.fit(X_train,y_train)
```

```python
#make predictions
preds=bst.predict(X_test)
```

```python
# Calculate accuracy
accuracy = accuracy_score(y_test, preds)
print(f'Accuracy: {accuracy:.2f}')
```
