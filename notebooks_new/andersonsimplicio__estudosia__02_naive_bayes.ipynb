```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
from sklearn.preprocessing import StandardScaler,LabelEncoder,OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
import pickle
# GausianNB ? a distribui??o Estat?stica, neste caso a Distribui??o Normal
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix,classification_report
from yellowbrick.classifier import ConfusionMatrix
```

```python
with open('database/credit.pkl','rb') as f:
    Xtreinamento, Xteste,ytreinamento,yteste = pickle.load(f)
```

```python
Xtreinamento.shape, ytreinamento.shape
```

```python
Xteste.shape,yteste.shape
```

```python
naive_credit = GaussianNB()
naive_credit.fit(Xtreinamento,ytreinamento)
previsoes = naive_credit.predict(Xteste)
previsoes
```

```python
yteste
```

```python
accuracy_score(yteste,previsoes);
confusion_matrix(yteste,previsoes)
```

```python
cm = ConfusionMatrix(naive_credit);
cm.fit(Xtreinamento,ytreinamento);
cm.score(Xteste,yteste);
```

```python
print(classification_report(yteste,previsoes))
```
