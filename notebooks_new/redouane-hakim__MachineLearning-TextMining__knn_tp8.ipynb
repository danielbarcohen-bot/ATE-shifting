### programme 2

fonction utiliser : `knn` : le model , `mean` : moyenne , `mode` : le label qui a plus d'occurance,<br>
`euclidean_distance` : $  = \sqrt{x^2 + y^2}$<br>

La fonction mean : Elle calcule la moyenne arithmétique des étiquettes des voisins.  
La fonction mode : Elle identifie la valeur la plus fréquente (le mode) parmi les étiquettes des voisins.

```python

from collections import Counter
import math
def mean(labels):
    """Calcul de la moyenne des k étiquettes"""
    return sum(labels) / len(labels)

def mode(labels):
    "Calcul du mode des k étiquettes"
    return Counter(labels).most_common(1)[0][0]

def euclidean_distance(point1, point2):
    """Calcul de la distance euclidienne"""
    sum_squared_distance = 0
    for i in range(len(point1)):
        sum_squared_distance += math.pow(point1[i]-point2[i], 2)
    return math.sqrt(sum_squared_distance)

def knn(data, query, k, distance_utilisee, choix_knn_regress_ou_classif):
    """Algorithme des k plus proches voisins kNN"""
    distances_et_positions = []
    for index, example in enumerate(data):
        distance = distance_utilisee(example[ :- 1], query)
        distances_et_positions.append((distance, index))
        #ici une liste de tuples sera triée selon les distances
        sorted_distances_et_positions =sorted(distances_et_positions)
        k_proches_distances_and_indices =sorted_distances_et_positions[:k]
        k_proches_classes =[]
        for dist_indice in k_proches_distances_and_indices:
            num=dist_indice[1]; k_proches_classes+=[data[num][1]]
    # 7. Si régression (choix_knn_regress_ou_classif = mean),
    #retourner la moyenne des k étiquettes
    # 8. Si classification, retourner le mode des k étiquettes
    return k_proches_distances_and_indices ,choix_knn_regress_ou_classif(k_proches_classes),k_proches_classes
```

### programme 4

La distance Euclidienne est plus performante que Manhattan/Minkowski. Augmenter k de 1 à 5 permet d'améliorer le score en réduisant l'overfitting, mais aller trop loin k=9 fait rechuter la précision (86.0%).

```python

from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV, train_test_split

for k in [1,3,5,7,9]:
    print("k = ", k)
    for distances in ['euclidean', 'manhattan','minkowski']:
        data=pandas.read_csv("images_chiffres_codees_niveau_de_gris.csv")
        data=np.array(data); x=data[0:1000,1:]; y=data[0:1000,0];
        x_train, x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
        model=KNeighborsClassifier(n_neighbors=k,metric=distances)
        model.fit(x_train,y_train); score=model.score(x_test, y_test)
        print("score sur test-set par k="+str(k)+" et distance="+str(distances)+"  -- > "+str(score*100)+'%')
```

l'utilisation de gridSearch pour la comparaision entre les models , on obtient le même conclusion au-dessus : que la distance euclidean avec k = 5 est la meilleur au niveau de accuracy (peut etre differente par rapport a un autre mesure de performance ) et que l'utilsation de gridSearch sert a rapidement trouver les paramétres optimals sans essayer de manuallement distinguer entre les differentes scores obtenu pour chaque paramteres changer (K , distance ....)

```python

distances=['euclidean', 
           'manhattan'] ; 
valeurs_de_k=np.arange(1,8)
parametres_grid={"n_neighbors": valeurs_de_k, 
                 'metric' : distances}

model=KNeighborsClassifier()
grid=GridSearchCV(model,parametres_grid,cv=3,scoring='accuracy')
grid.fit(x_train,y_train)
print("best parameters:",grid.best_params_)
model=grid.best_estimator_ 
model.fit(x_train,y_train)
score=model.score(x_test, y_test)
print("score sur test-set-paramètres automatiques="+str(score*100)+'%')
```
