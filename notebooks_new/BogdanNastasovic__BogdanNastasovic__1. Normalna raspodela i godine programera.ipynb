```python
#Vr?imo ?i??enje i izdvajamo kolone koje su nam od zna?aja#
#Za na?e potrebe zadatka, to su 'MainBranch', 'Ag'e i 'RemoteWork'#
#Novodobijenu tabelu nazva?emo sa df1#

df1 = df.loc[:, ['MainBranch', 'Age', 'RemoteWork']]
```

```python
#Menjamo nazive kolona na srpskom jeziku#
#Konstrui?emo 'dictionary' i onda njega primenjujemo na df1#

nazivi_kolona = {'MainBranch':'Profesija',
                 'Age':'Godine',
                 'RemoteWork':'Na?in rada'}
df1 = df1.rename(columns= nazivi_kolona)
```

```python
#Zapazimo da su odgovori zabele?eni na engleskom jeziku#
#U svakom redu promenimo ih da budu na srpskom, kako bi nam krajnji rezultat bio na srpskom#
#Prvo to radimo za kolonu "Profesija", nakon za "Godine" i na kraju za "Na?in rada"#
```

```python
#Najpre gledamo koje sve vrednosti sadr?i kolona "Profesija"#

df1['Profesija'].value_counts()
```

```python
nazivi_profesija = {'I am a developer by profession':'Programira zbog posla',
                    'I am not primarily a developer, but I write code sometimes as part of my work/studies':'Programira zbog potrebe fakulteta',
                    'I am learning to code':'U?i da programira',
                    'I code primarily as a hobby':'Programira iz hobija', 
                    'I used to be a developer by profession, but no longer am':'Nekada je programiralo ali vi?e ne',
                    'None of these':'Neizja?njeno'
                   }
df1['Profesija'] = df1['Profesija'].map(nazivi_profesija)

#Primetimo da je ovde dobro da prilikom same ankete i prikupljanja podataka, ispitanicima ponudite odgovore u "scroll list"#
#Razlog tome je da ne do?e do gre?aka u kucanju i nagomilavanju 'razli?itih', a su?tinski istih vrednosti#
#Time se pri prikupljanju podataka javlja manji broj gre?aka i stoga kasnije ne dolazi do njihovog nagomilavanja#
#Ova anketa je profesionalno sprovedena, tako da ima malo sre?ivanja#
```

```python
#Gorenavedeni postupak robavljamo i sa kolonama 'Godine' i 'Na?in rada' (linije 15,16,17,18)#
```

```python
df1['Godine'].value_counts()
```

```python
nazivi_godina = {'Under 18 years old':'18-',
                 '18-24 years old':'18-24',
                 '25-34 years old':'25-34',
                 '35-44 years old':'35-44',
                 '45-54 years old':'45-54',
                 '55-64 years old':'55-64',
                 '65 years or older': '65+',
                 'Prefer not to say':'Neizja?njeno'
                }
df1['Godine'] = df1['Godine'].map(nazivi_godina)
```

```python
df1['Na?in rada'].value_counts()
```

```python
nazivi_na?ina_rada = {'Hybrid (some remote, some in-person)':'Kombinovano',
                     'Remote':'Od ku?e',
                     'In-person':'U?ivo, u prostorijama firme'
                     }
df1['Na?in rada'] = df1['Na?in rada'].map(nazivi_na?ina_rada)
```

```python
#Grafi?ki prikazujemo odnos programera koji rade od ku?e, na poslu ili kombinovano#
#Ubacujemo legendu da bi znali ?ta svaka boja predstavlja, a zbog toga izbacujemo labels#
#Radi bolje vizualizacije, koristimo %#

plt.style.use('fivethirtyeight')
df1['Na?in rada'].value_counts().plot(kind = 'pie', 
                                      legend = True, 
                                      autopct = '%1.1f%%', 
                                      labeldistance=None,
                                      )


plt.title('Na?in rada')
plt.xlabel('')
plt.ylabel('')
plt.legend(bbox_to_anchor=(1, 1))
plt.show()
```

```python
#Izbacujemo one koji se nisu izjasnili povodom godina#
#Ovo radimo jer odnos godina i broja programera ?elimo da prika?emo broj?ano preko trakastog dijagrama#
#Osobe koje se nisu izjasnile nam remete taj koncept#

df1 = df1.loc[df1["Godine"] != 'Neizja?njeno']
```

```python
#Sortiramo godine od najmanje ka najve?em#
#Ovo je bitno da bismo kasnije uspostavili veyu sa Gausovom raspodelom#

df1["Godine"].value_counts().sort_index()
```

```python
#PREDSTAVLJANJE ODNOSA GODINA I BROJA PROGRAMERA PREKO TRAKASTOG DIJAGRAMA I VEZA S GAUSOVOM RASPODELOM#
fig, (ax1, ax2) = plt.subplots(1, 2, figsize = (12,6))
fig.subplots_adjust(wspace=0.3)

#TRAKASTI DIJAGRAM DOBIJEN NA OSNOVU NA?IH PODATAKA#
df1["Godine"].value_counts().sort_index().plot(ax=ax1, 
                                               kind='bar', 
                                               ylabel="Broj Programera", 
                                               title = "Na osnovu na?ih podataka", 
                                               width=0.7,
                                              )

#GAUSOVA RASPODELA SA MATEMATI?KIM O?EKIVANJEM 35 I STANDARDNOM DEVIJACIJOM 12#
x_data = np.arange(5, 65, 0.01)
y_data = (stats.norm.pdf(x_data, x_data.mean() , 12))

ax2.plot(x_data, pow(10,6)*y_data)
plt.title("Gausova raspodela")
plt.xlabel("Godine")
plt.ylabel("Broj Programera")

plt.show()
```
