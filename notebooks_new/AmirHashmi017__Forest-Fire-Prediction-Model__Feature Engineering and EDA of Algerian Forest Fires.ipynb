## Algerian Forest Fires Dataset 
Data Set Information:

The dataset includes 244 instances that regroup a data of two regions of Algeria,namely the Bejaia region located in the northeast of Algeria and the Sidi Bel-abbes region located in the northwest of Algeria.

122 instances for each region.

The period from June 2012 to September 2012.
The dataset includes 11 attribues and 1 output attribue (class)
The 244 instances have been classified into fire(138 classes) and not fire (106 classes) classes.

Attribute Information:

1. Date : (DD/MM/YYYY) Day, month ('june' to 'september'), year (2012)
Weather data observations
2. Temp : temperature noon (temperature max) in Celsius degrees: 22 to 42
3. RH : Relative Humidity in %: 21 to 90
4. Ws :Wind speed in km/h: 6 to 29
5. Rain: total day in mm: 0 to 16.8
FWI Components
6. Fine Fuel Moisture Code (FFMC) index from the FWI system: 28.6 to 92.5
7. Duff Moisture Code (DMC) index from the FWI system: 1.1 to 65.9
8. Drought Code (DC) index from the FWI system: 7 to 220.4
9. Initial Spread Index (ISI) index from the FWI system: 0 to 18.5
10. Buildup Index (BUI) index from the FWI system: 1.1 to 68
11. Fire Weather Index (FWI) Index: 0 to 31.1
12. Classes: two classes, namely Fire and not Fire

## Data Cleaning

```python
# Handelling Missing Values
dataset.isnull().sum()
```

```python
# Getting the rows having missing values
dataset[dataset.isnull().any(axis=1)]
```

The dataset is converted into two sets based on Region from 122th index, we can make a new column based on the Region

1 : "Bejaia Region Dataset"

2 : "Sidi-Bel Abbes Region Dataset"

Add new column with region

```python
dataset.loc[:122,'Region']=0
dataset.loc[122:,'Region']=1
df=dataset
```

```python
# Convert Region into int
df['Region']=df['Region'].astype(int)
df.info()
```

```python
df.head()
```

```python
# Taking rows with null values and remove them
df[df.isnull().any(axis=1)]
df=df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df.iloc[[122]]
```

```python
# Dropped the additional column row
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
# You can see that we have spaces in column names let's fix it
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

```python
# We can see that there are 13 columns whihc are objects let's conver them into int or float
# We noticed that day,month,year,Temperature,RH,Ws all can be int so let;s do it
df[['day','month','year','Temperature','RH','Ws']]=df[['day','month','year','Temperature','RH','Ws']].astype(int)
df.info()
```

```python
df.head()
```

```python
# Now you can see Rain,FFMC,DMC,DC,ISI,,BUI,FWI can be float so let's change it
# Let's convert all columns that are object into int instead of classes
object_features=[features for features in df.columns if df[features].dtype=='O']
object_features
```

```python
for column in object_features:
    if column!='Classes':
        df[column]=df[column].astype(float)
df.info()
```

```python
# Let's save the cleaned Dataset
df.to_csv("Algerian_forest_fires_cleaned_dataset.csv",index=False)
```

```python
# Now we are doing the monthly Fire Analysis
df_original=pd.read_csv("Algerian_forest_fires_cleaned_dataset.csv")
df_original.head()
df_original['Classes']=np.where(df_original["Classes"].str.contains("not fire"),"not fire","fire")
```

```python
# Visualization monthly Fire Analysis of Bejai Region
dfbejai=df_original[df_original['Region']==0]
plt.subplots(figsize=(13,6))
sns.countplot(x="month",hue="Classes",data=dfbejai)
plt.title("Fire Analysis For Bejai Region")
plt.show()
```

```python
# Visualization monthly Fire Analysis of Sidibell Region
df_SidiBel=df_original[df_original['Region']==1]
plt.subplots(figsize=(13,6))
sns.countplot(x="month",hue="Classes",data=df_SidiBel)
plt.title("Fire Analysis of Sidibell Region")
plt.show()
```
