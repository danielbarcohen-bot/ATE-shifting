# Raw Flight Data Merging and Uploading to S3

```python
# import libraries
import pandas as pd
import boto3
import os
```

```python
# import the csv files
Jan23 = pd.read_csv('raw_flight_data/Jan23_OT_report.csv')
Feb23 = pd.read_csv('raw_flight_data/Feb23_OT_report.csv')
Mar23 = pd.read_csv('raw_flight_data/Mar23_OT_report.csv')
Apr23 = pd.read_csv('raw_flight_data/Apr23_OT_report.csv')
May23 = pd.read_csv('raw_flight_data/May23_OT_report.csv')
Jun23 = pd.read_csv('raw_flight_data/Jun23_OT_report.csv')
Jul23 = pd.read_csv('raw_flight_data/Jul23_OT_report.csv')
Aug23 = pd.read_csv('raw_flight_data/Aug23_OT_report.csv')
Sep23 = pd.read_csv('raw_flight_data/Sep23_OT_report.csv')
Oct23 = pd.read_csv('raw_flight_data/Oct23_OT_report.csv')
Nov23 = pd.read_csv('raw_flight_data/Nov23_OT_report.csv')
Dec23 = pd.read_csv('raw_flight_data/Dec23_OT_report.csv')
Jan24 = pd.read_csv('raw_flight_data/Jan24_OT_report.csv')
Feb24 = pd.read_csv('raw_flight_data/Feb24_OT_report.csv')
Mar24 = pd.read_csv('raw_flight_data/Mar24_OT_report.csv')
Apr24 = pd.read_csv('raw_flight_data/Apr24_OT_report.csv')
May24 = pd.read_csv('raw_flight_data/May24_OT_report.csv')
Jun24 = pd.read_csv('raw_flight_data/Jun24_OT_report.csv')
Jul24 = pd.read_csv('raw_flight_data/Jul24_OT_report.csv')
Aug24 = pd.read_csv('raw_flight_data/Aug24_OT_report.csv')
Sep24 = pd.read_csv('raw_flight_data/Sep24_OT_report.csv')
Oct24 = pd.read_csv('raw_flight_data/Oct24_OT_report.csv')
Nov24 = pd.read_csv('raw_flight_data/Nov24_OT_report.csv')
Dec24 = pd.read_csv('raw_flight_data/Dec24_OT_report.csv')
```

```python
# list the df's in chronological order
dfs = [Jan23, Feb23, Mar23, Apr23, May23, Jun23, Jul23, Aug23, Sep23, Oct23, Nov23, Dec23,
       Jan24, Feb24, Mar24, Apr24, May24, Jun24, Jul24, Aug24, Sep24, Oct24, Nov24, Dec24]
```

```python
# merge df's
merged_df = pd.concat(dfs, ignore_index=True)
```

```python
# save the merged df to csv
merged_df.to_csv('raw_flight_data/merged_flights.csv', index=False)
```

```python
# examine list of buckets available
s3 = boto3.client('s3')
response = s3.list_buckets()
print("Existing buckets:")
for bucket in response['Buckets']:
    print(bucket['Name'])
```

```python
# set local path of merged CSV file
local_file_path = 'raw_flight_data/merged_flights.csv'
```

```python
# set the raw data bucket variable
raw_bucket = "group9-ml-project-raw-data-bucket-grw"
```

```python
# set the S3 key path, we are using flat structure because this will only hold two files
s3_key = "data/merged_flights.csv"
```

```python
# upload file to the S3 bucket
try:
    s3.upload_file(local_file_path, raw_bucket, s3_key)
    print(f"Uploaded {local_file_path} to s3://{raw_bucket}/{s3_key}")
except Exception as e:
    print("Error uploading file to S3:", e)
```
