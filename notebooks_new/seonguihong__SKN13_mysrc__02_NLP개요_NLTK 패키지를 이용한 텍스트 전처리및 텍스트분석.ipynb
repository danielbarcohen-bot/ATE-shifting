````python
# Pos-tag ?? ??? ????(? ???? ????)? WordNetLemmatizer? ????? ??
def get_wordnet_pos(pos_tag):
    """
    ? ???? ????? WordNetLemmatizer?? ???? ????? ??
    ???/??/??/?? ?? ??
    """
    if pos_tag.startswith("J"):
        return wordnet.ADJ
    elif pos_tag.startswith("V"):
        return wordnet.VERB
    elif pos_tag.startswith("N"):
        return wordnet.NOUN
    elif pos_tag.startswith("R"):
        return wordnet.ADV
    else:
        return None
````
