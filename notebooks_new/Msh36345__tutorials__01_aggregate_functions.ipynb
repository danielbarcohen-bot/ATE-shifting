# 📊 פונקציות אגרגציה ב-PySpark - מדריך מקיף
[PySpark Aggregate Functions with Examples](https://sparkbyexamples.com/pyspark/pyspark-aggregate-functions/)
## מבוא

פונקציות אגרגציה ב-PySpark הן חיוניות לסיכום נתונים על פני מערכי נתונים מבוזרים. הן מאפשרות ביצוע חישובים כמו סכום, ממוצע, ספירה, מקסימום ומינימום בצורה יעילה במקביל על פני מספר צמתים באשכול.

פונקציות אלו משמשות בעיקר עם מערכי נתונים גדולים כדי לקבל ערכים בודדים או קבוצות תוצאות קטנות יותר, ומספקות תובנות על מאפייני מערך הנתונים.

### 🔑 עקרונות חשובים:

- פונקציות אגרגציה פועלות על קבוצה של שורות ומחשבות ערך החזרה בודד לכל קבוצה
- כל פונקציות האגרגציה מקבלות קלט מסוג Column או שם עמודה כמחרוזת
- פונקציות אגרגציה ב-PySpark SQL מקובצות תחת "agg_funcs"

---

## 🔧 הגדרת הסביבה וייבוא ספריות

ראשית, נייבא את כל הספריות והפונקציות הדרושות:

```python
# ייבוא ספריות
import pyspark
from pyspark.sql import SparkSession
from pyspark.sql.functions import approx_count_distinct, collect_list
from pyspark.sql.functions import collect_set, sum, avg, max, countDistinct, count
from pyspark.sql.functions import first, last, kurtosis, min, mean, skewness
from pyspark.sql.functions import stddev, stddev_samp, stddev_pop, sumDistinct
from pyspark.sql.functions import variance, var_samp, var_pop
```

## 🚀 יצירת SparkSession

ניצור SparkSession - נקודת הכניסה לעבודה עם PySpark:

```python
spark = SparkSession.builder.appName('PySpark-Aggregate-Functions').getOrCreate()
```

## 📝 יצירת DataFrame לדוגמה

ניצור DataFrame פשוט עם נתוני עובדים שישמש אותנו להדגמת פונקציות האגרגציה:

```python
# הגדרת הנתונים
simpleData = [
    ("James", "Sales", 3000),
    ("Michael", "Sales", 4600),
    ("Robert", "Sales", 4100),
    ("Maria", "Finance", 3000),
    ("James", "Sales", 3000),
    ("Scott", "Finance", 3300),
    ("Jen", "Finance", 3900),
    ("Jeff", "Marketing", 3000),
    ("Kumar", "Marketing", 2000),
    ("Saif", "Sales", 4100)
]

# הגדרת הסכימה
schema = ["employee_name", "department", "salary"]

# יצירת DataFrame
df = spark.createDataFrame(data=simpleData, schema=schema)
```

### 👀 הצגת המבנה והנתונים

נציג את סכימת ה-DataFrame ואת הנתונים שלו:

```python
# הצגת הסכימה
df.printSchema()
```

```python
# הצגת הנתונים
df.show(truncate=False)
```

---

# 📊 פונקציות אגרגציה - דוגמאות

כעת נעבור על כל פונקציות האגרגציה השונות עם דוגמאות מעשיות.

## 1️⃣ approx_count_distinct() - ספירה משוערת של ערכים ייחודיים

הפונקציה מחזירה את המספר המשוער של פריטים ייחודיים בקבוצה. זוהי פונקציה יעילה לספירת ערכים ייחודיים במערכי נתונים גדולים.

```python
# approx_count_distinct()
print(df.select(approx_count_distinct("salary")).collect()[0][0])
```

## 2️⃣ avg() - חישוב ממוצע

הפונקציה מחזירה את הממוצע של הערכים בעמודת הקלט.

```python
# avg
print(df.select(avg("salary")).collect()[0][0])
```

## 3️⃣ collect_list() - איסוף כל הערכים כרשימה

הפונקציה מחזירה את כל הערכים מעמודת הקלט **כולל ערכים כפולים**.

```python
# collect_list
df.select(collect_list("salary")).show(truncate=False)
```

## 4️⃣ collect_set() - איסוף ערכים ייחודיים כסט

הפונקציה מחזירה את כל הערכים מעמודת הקלט **ללא ערכים כפולים**.

```python
# collect_set
df.select(collect_set("salary")).show(truncate=False)
```

## 5️⃣ countDistinct() - ספירת ערכים ייחודיים

הפונקציה מחזירה את מספר האלמנטים הייחודיים בעמודות. ניתן להעביר מספר עמודות.

```python
# countDistinct
df2 = df.select(countDistinct("department", "salary"))
df2.show(truncate=False)
print(df2.collect()[0][0])
```

## 6️⃣ count() - ספירת אלמנטים

הפונקציה מחזירה את מספר האלמנטים בעמודה.

```python
# count
print(df.select(count("salary")).collect()[0][0])
```

## 7️⃣ first() - האלמנט הראשון

הפונקציה מחזירה את האלמנט הראשון בעמודה. כאשר `ignoreNulls` מוגדר ל-true, היא מחזירה את האלמנט הלא-null הראשון.

```python
# first
df.select(first("salary")).show(truncate=False)
```

## 8️⃣ last() - האלמנט האחרון

הפונקציה מחזירה את האלמנט האחרון בעמודה. כאשר `ignoreNulls` מוגדר ל-true, היא מחזירה את האלמנט הלא-null האחרון.

```python
# last
df.select(last("salary")).show(truncate=False)
```

## 9️⃣ kurtosis() - חישוב קורטוזיס

הפונקציה מחזירה את הקורטוזיס (kurtosis) של הערכים בקבוצה. קורטוזיס מודד את "הזנב" של ההתפלגות.

```python
# kurtosis
df.select(kurtosis("salary")).show(truncate=False)
```

## 🔟 max() - ערך מקסימלי

הפונקציה מחזירה את הערך המקסימלי בעמודה.

```python
# max
df.select(max("salary")).show(truncate=False)
```

## 1️⃣1️⃣ min() - ערך מינימלי

הפונקציה מחזירה את הערך המינימלי בעמודה.

```python
# min
df.select(min("salary")).show(truncate=False)
```

## 1️⃣2️⃣ mean() - חישוב ממוצע (שם נרדף ל-avg)

הפונקציה מחזירה את הממוצע של הערכים בעמודה. זהו שם נרדף (alias) לפונקציה `avg()`.

```python
# mean
df.select(mean("salary")).show(truncate=False)
```

## 1️⃣3️⃣ skewness() - חישוב סטיית אסימטריה

הפונקציה מחזירה את ה-skewness (סטיית אסימטריה) של הערכים בקבוצה. Skewness מודד את חוסר הסימטריה של ההתפלגות.

```python
# skewness
df.select(skewness("salary")).show(truncate=False)
```

## 1️⃣4️⃣ סטיית תקן - stddev(), stddev_samp() ו-stddev_pop()

- **stddev()** - שם נרדף ל-stddev_samp
- **stddev_samp()** - מחזירה את סטיית התקן של המדגם
- **stddev_pop()** - מחזירה את סטיית התקן של האוכלוסייה

```python
# stddev, stddev_samp, stddev_pop
df.select(stddev("salary"), stddev_samp("salary"), \
          stddev_pop("salary")).show(truncate=False)
```

## 1️⃣5️⃣ sum() - חישוב סכום

הפונקציה מחזירה את הסכום של כל הערכים בעמודה.

```python
# sum
df.select(sum("salary")).show(truncate=False)
```

## 1️⃣6️⃣ sumDistinct() - סכום ערכים ייחודיים

הפונקציה מחזירה את הסכום של כל הערכים **הייחודיים** בעמודה.

```python
# sumDistinct
df.select(sumDistinct("salary")).show(truncate=False)
```

## 1️⃣7️⃣ שונות - variance(), var_samp() ו-var_pop()

- **variance()** - שם נרדף ל-var_samp
- **var_samp()** - מחזירה את השונות הלא-מוטה של הערכים בעמודה
- **var_pop()** - מחזירה את שונות האוכלוסייה של הערכים בעמודה

```python
# variance, var_samp, var_pop
df.select(variance("salary"), var_samp("salary"), var_pop("salary")) \
    .show(truncate=False)
```

## 📋 סיכום

במדריך זה למדנו על פונקציות האגרגציה השונות ב-PySpark:

### 🎯 פונקציות ספירה:
- `approx_count_distinct()` - ספירה משוערת של ערכים ייחודיים
- `count()` - ספירת אלמנטים
- `countDistinct()` - ספירת ערכים ייחודיים

### 📊 פונקציות סטטיסטיות בסיסיות:
- `sum()` / `sumDistinct()` - סכום ערכים
- `avg()` / `mean()` - ממוצע
- `min()` / `max()` - ערכים קיצוניים

### 📈 פונקציות סטטיסטיות מתקדמות:
- `stddev()`, `stddev_samp()`, `stddev_pop()` - סטיית תקן
- `variance()`, `var_samp()`, `var_pop()` - שונות
- `skewness()` - סטיית אסימטריה
- `kurtosis()` - קורטוזיס

### 🔍 פונקציות איסוף נתונים:
- `collect_list()` - איסוף כל הערכים (כולל כפולים)
- `collect_set()` - איסוף ערכים ייחודיים
- `first()` / `last()` - ערך ראשון/אחרון

### 💡 טיפים חשובים:

1. **שימוש עם GroupBy**: פונקציות אגרגציה משמשות לעתים קרובות יחד עם פעולות `groupBy()` לביצוע חישובים על נתונים מקובצים
2. **ביצועים**: פונקציות אלו מותאמות לעבודה על נתונים מבוזרים ויעילות מאוד במערכי נתונים גדולים
3. **גמישות**: כל הפונקציות מקבלות שם עמודה כמחרוזת או אובייקט Column
