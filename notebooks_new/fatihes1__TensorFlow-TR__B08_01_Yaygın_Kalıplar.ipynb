```python
def plot_series(time, series, format="-", start=0, end=None, label=None):
    plt.plot(time[start:end], series[start:end], format, label=label)
    plt.xlabel("Zaman")
    plt.ylabel("De?erler")
    if label:
        plt.legend(fontsize=14)
    plt.grid(True)
```

## Trend ve Mevsimsellik

Sadece yukar? do?ru giden bir zaman serisi olu?turarak ba?layal?m.

```python
def trend(time, slope=0):
    return slope * time
```

Olu?turaca??m?z zaman serisi 4 x 365 + 1 = 1461 birimlik olacakt?r.

```python
time = np.arange(4 * 365 + 1)
baseline = 10
series = baseline + trend(time, 0.1)

plt.figure(figsize=(10, 6))
plot_series(time, series)
plt.show()
```

?imdi de mevsimsel bir desene sahip bir zaman serisi olu?tural?m.

```python
def seasonal_pattern(season_time):
    return np.where(season_time < 0.4,
                    np.cos(season_time * 2 * np.pi),
                    1 / np.exp(3 * season_time))

def seasonality(time, period, amplitude=1, phase=0):
    """Her periotta ayn? kal?b? tekrarlamas?n? sa?layal?m."""
    season_time = ((time + phase) % period) / period
    return amplitude * seasonal_pattern(season_time)
```

Genlik (`amplitude`) de?erimizi 40 olarak ve period de?erimizi 365 olarak ayarlad?ktan sonta mevsimsel bir desene sahip zaman serimizi olu?turup, g?rselle?tirebiliriz.

```python
amplitude = 40
series = seasonality(time, period=365, amplitude=amplitude)

plt.figure(figsize=(10, 6))
plot_series(time, series)
plt.show()
```

?imdi de hem tren gem de mevsimsellik i?eren bir zaman serisi olu?tural?m ve nas?l g?r?nd???ne g?z atal?m.

```python
slope = 0.05
series = baseline + trend(time, slope) + seasonality(time, period=365, amplitude=amplitude)

plt.figure(figsize=(10, 6))
plot_series(time, series)
plt.show()
```

## G?r?lt?

Pratikte ?ok az ger?ek zamanl? zaman serisi yukar?daki gibi d?zg?n bir sinyala sahiptir. Genellikle biraz g?r?lt?ye sahiptirler ve sinyal-g?r?lt? oran? bazen ?ok d???k olabilir. ?imdi beyaz g?r?lt? (*white noise*) ?retelim.

```python
def white_noise(time, noise_level=1, seed=None):
    rnd = np.random.RandomState(seed)
    return rnd.randn(len(time)) * noise_level
```

```python
noise_level = 5
noise = white_noise(time, noise_level, seed=42)

plt.figure(figsize=(10, 6))
plot_series(time, noise)
plt.show()
```

Son olarak yukar?da olu?turdu?umuz ve trend + mevsimsellik i?eren zaman serimizi g?r?lt? ile birle?tirelim ve nas?l bir g?r?nt? ile kar??la?aca??m?z? g?relim.

```python
series += noise

plt.figure(figsize=(10, 6))
plot_series(time, series)
plt.show()
```

Evet tam da beklenildi?i gibi. Y?kselen bir trend, belli periyotlarda ayn? deseni tekrarlayan bir mevsimsellik ve elbette biraz beyaz g?r?lt? ile zaman serimizi olu?turduk.
