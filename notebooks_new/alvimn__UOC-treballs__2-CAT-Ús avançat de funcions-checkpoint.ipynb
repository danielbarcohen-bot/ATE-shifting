Per qu? podem voler encapsular el codi? Suposem, per exemple, que estem analitzant un conjunt de dades amb les localitzacions actuals d'un conjunt de persones, aix? com la localitzaci? del seu domicili i del seu lloc de treball, i que volem saber si aquestes persones es troben a prop tant del lloc de treball com del domicili. Per fer-ho, implementem una funci? `is_close` que ens retorna un boole? que indica si estan a prop o no d'ambdues localitzacions. Aquesta funci? necessitar? calcular la dist?ncia entre dos punts dues vegades (una per calcular la dist?ncia entre la localitzaci? actual i el domicili, i una segona vegada per saber la dist?ncia entre la localitzaci? actual i el lloc de treball). Per tant, per no repetir codi, definirem una altra funci? `dist` que calculi la dist?ncia entre dos punts. Ara b?, per a l'an?lisi que volem fer, no farem servir la dist?ncia euclidiana, sin? que farem servir la [dist?ncia de Manhattan](https://en.wikipedia.org/wiki/Taxicab_geometry). A la resta del nostre codi, per?, mai farem servir aquesta definici? de dist?ncia, i volem evitar que algun altre programador de l'equip, per error, cridi la nostra funci? dist?ncia `dist` (pensant, potser, que calcula la dist?ncia m?s habitual, l'euclidiana). Una possible manera de fer-ho ?s definir la funci? `dist` com a una funci? local a `is_close`. D'aquesta manera, podem fer servir una funci? i no haurem de repetir codi en calcular les dist?ncies i, alhora, evitarem que es cridi aquesta funci? des de fora de la funci? `is_close`:

````python
def is_close(x, y, l1_x, l1_y, l2_x, l2_y):
    """Computes whether a location is nearby two other locations.

    Computes if location (x,y) is close to two other locations,
    (l1_x, l1_y) and (l2_x, l2_y)."""

    def dist(x1, y1, x2, y2):
        """Computes the Manhattan distance between (x1, y1) and (x2, y2)."""
        return abs(x1 - x2) + abs(y1 - y2)

    # Considerem que un punt ?s a prop d'un altre si la dist?ncia que els separa
    # ?s inferior a lim
    lim = 10
    # Retornem si (x, y) ?s a prop tant de (l1_x, l1_y) com de (l2_x, l2_y)
    return (dist(x, y, l1_x, l1_y) < lim) and (dist(x, y, l2_x, l2_y) < lim)
````

````python
# Calculem si la localitzaci? actual (0, 0) ?s a prop de (1, 4) i (-7, 1)
r1 = is_close(0, 0, 1, 4, -7, 1)
print("(0, 0) is close to (1, 4) and (-7, 1)?: {}".format(r1))

# Calculem si la localitzaci? actual (0, 0) ?s a prop de (1, 4) i (15, 1)
r2 = is_close(0, 0, 1, 4, 15, 1)
print("(0, 0) is close to (1, 4) and (15, 1)?: {}".format(r2))
````

Abans d'acabar aquesta secci?, un apunt sobre la implementaci? de la funci? `is_close`. Fixeu-vos que la funci? ha de retornar `True` si es compleix una determinada condici?     return True

Aquest codi fa servir una expressi? (que inclou el c?lcul de dist?ncies) per diferenciar si la funci? ha de retornar `True` o `False`. ?s important tenir en compte que la mateixa expressi? ja retorna un boole? i que, per tant, pot fer-se servir directament com a valor de retorn:

````python


# Definim una funci? sense return
def print_sum_v0(x, y):
    print("Result is: {}".format(x+y))


# Definim una funci? amb retorn buit
def print_sum_v1(x, y):
    print("Result is: {}".format(x+y))
    return


# El valor de retorn de les dues funcions ?s None
r = print_sum_v0(3, 5)
print(r)

r = print_sum_v1(3, 5)
print(r)
````

Una funci? retorna un ?nic **objecte**. Aix?, si cal que una funci? retorni m?s d'un valor, podem fer que la funci? retorni una tupla amb els diversos valors.

````python


# Definim la funci? de nom 'sum_two_values' amb dos arguments: 'x' i 'y':
def sum_two_values(x, y):
    """Return the value of the sum."""
    return x + y


# Definim la funci? de nom 'sum_and_mult_two_values' amb dos arguments:
# 'x' i 'y':
def sum_and_mult_two_values(x, y):
    """Return a tuple with the value of the sum and the product."""
    return x + y, x * y


r1 = sum_two_values(5, 11)
r2 = sum_and_mult_two_values(5, 11)
print("The result of sum_two_values(5, 11) is {} ({})".
      format(r1, type(r1)))
print("The result of sum_and_mult_two_values(5, 11) is {} ({})".
      format(r2, type(r2)))
````

### 1.3.3. Nombre indeterminat d'arguments


Python tamb? permet definir funcions que acceptin un nombre arbitrari (indefinit en el moment de la definici? de la funci?) d'arguments. ````python


````python
# Definim una funci? amb dos par?metres obligatoris i
# un nombre indeterminat de par?metres posicionals
def sum_two_or_three_values_p(x, y, *extra_arguments):
    print("Compulsory arguments are: x={}, y={}".format(x, y))
    print("Additional arguments are: {}".format(extra_arguments))
    print("extra_arguments type is: {}".format(type(extra_arguments)))
    return x + y + sum(extra_arguments)
````

### 1.4.1. Map

La funci? [`map`](https://docs.python.org/3/library/functions.html#map) rep com a par?metres una funci? i un iterable, i retorna un iterador que aplica la funci? proporcionada a cadascun dels elements de l'iterable.

````python
def sum_2(x):
    """Add 2 to the value received by parameter."""
    return x + 2


# Apliquem la funci? sum_2 a cadascun dels elements de la llista a_list
a_list = [1, 2, 3, 4]
r = map(sum_2, a_list)
print(list(r))
````

````python


# Definim una funci? que converteix cadenes de car?cters que expressen
# valors num?rics (possiblement decimals) en enters
def convert_to_int(x):
    """Convert strings to int."""
    return int(float(x))


# Apliquem la funci? convert_to_int a cadascun dels elements de la
# llista a_list
a_list = ["42.45", "13.4", "12000"]
list(map(convert_to_int, a_list))
````

Als exemples anteriors, la funci? que es passa com a argument a [`map`](https://docs.python.org/3/library/functions.html#map) (`sum_2` i `convert_to_int`) t? un ?nic par?metre, de manera que [`map`](https://docs.python.org/3/library/functions.html#map) la pot aplicar directament sobre cadascun dels elements de la llista (llista que rep com a segon argument). Ara b?, la funci? que es passa com a par?metre a [`map`](https://docs.python.org/3/library/functions.html#map) pot tenir m?s d'un par?metre. En aquest cas, la funci? [`map`](https://docs.python.org/3/library/functions.html#map) rebr? tantes llistes com par?metres necessiti la funci? que rep. 

En l'exemple seg?ent, definim una funci? que calcula el preu d'un pis a partir dels metres quadrats que t? (`sqm`), l'estat de conservaci? (`status`) i el ve?nat en qu? es troba (`neigh`):

````python
def compute_price(sqm, status, neigh):
    """Compute the price of a flat."""

    price_per_sqm = 1000
    nice_neigh = ["A", "B"]
    nice_neigh_factor = 1.25
    new_factor = 2

    # Preu base ?s metres quadrats per preu per metre quadrat
    price = sqm * price_per_sqm

    # Si el pis es troba en un barri considerat bo s'aplica un factor
    # multiplicatiu al preu del pis
    if neigh in nice_neigh:
        price *= nice_neigh_factor

    # Si el pis ?s nou s'aplica un factor multiplicatiu al
    # preu del pis
    if status == "New":
        price *= new_factor

    return price
````

````python
# Calculem el preu d'un pis nou de 100m^2 que es troba en un barri
# catalogat com a "A"
compute_price(100, "New", "A")
````

Una immobili?ria t? 5 pisos a la seva disposici? i vol fer servir la funci? `compute_price` per calcular el preu que hauria de tenir cadascun dels pisos. Podem fer servir la funci?
