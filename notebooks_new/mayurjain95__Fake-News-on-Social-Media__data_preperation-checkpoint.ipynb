```python
#importing Headers

import pandas 
import csv
import numpy 
import nltk
import seaborn 
from nltk.tokenize import word_tokenize
```

```python
# loading the files 

file_train = 'train.csv'
file_test = 'test.csv'
file_valid = 'valid.csv'

```

```python
training_mat = pandas.read_csv(file_train)
```

```python

testing_mat = pandas.read_csv(file_test)
```

```python
valid_mat = pandas.read_csv(file_valid,encoding='iso-8859-1')
```

```python
#printing data to check it is loaded
def print_Data():
    print("Training Data set\n")
    print(training_mat.shape)
    print(training_mat.head(15))
    
    print("Testing Data set\n")
    print(testing_mat.shape)
    print(testing_mat.head(15))
    
    print("Valid Data set\n")
    print(valid_mat.shape)
    print(valid_mat.head(15))
    
print_Data()
```
