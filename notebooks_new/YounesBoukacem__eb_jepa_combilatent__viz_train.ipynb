```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# plot the loss curves
batch_losses = np.lib.format.open_memmap("batch_losses.npy", dtype=np.float32, mode="r")
print(batch_losses.shape)
last_batch_loss_idx = np.lib.format.open_memmap("last_batch_loss_idx.npy", dtype=np.float32, mode="r")
plt.figure(figsize=(10, 6))
plt.plot(batch_losses[:last_batch_loss_idx[...]+1], label="training Loss", color="blue")
plt.xlabel("batch iteration")
plt.ylabel("loss")
plt.title("training loss curve")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# plot the val loss curves
val_losses = np.lib.format.open_memmap("val_losses.npy", dtype=np.float32, mode="r")
last_val_loss_idx = np.lib.format.open_memmap("last_val_loss_idx.npy", dtype=np.float32, mode="r")
plt.figure(figsize=(10, 6))
plt.plot(val_losses[:last_val_loss_idx[...]+1], label="validation loss", color="red")
plt.xlabel("checkpoint")
plt.ylabel("loss")
plt.title("validation loss curve")
plt.legend()
plt.grid(True)
plt.show()
```
