```python
#T03
# Select all columns for the first clinical trial in the dataset
first_clinical_trial = df.iloc[0, :]
print("First clinical trial details:\n", first_clinical_trial)

# Retrieve the Title and Status of the clinical trial with the NCT Number 'NCT04595136'
trial_details = df.loc[df['NCT Number'] == 'NCT04595136', ['Title', 'Status']]
print("Title and Status of NCT04595136:\n", trial_details)

# Get the Sponsor/Collaborators and Start Date for clinical trials that are Recruiting
recruiting_trials = df.loc[df['Status'] == 'Recruiting', ['Sponsor/Collaborators', 'Start Date']]
print("Sponsor/Collaborators and Start Date for Recruiting trials:\n", recruiting_trials)

# Select the first 5 rows and columns Title, Conditions, and Outcome Measures
selected_columns = df.loc[:4, ['Title', 'Conditions', 'Outcome Measures']]
print("First 5 rows with selected columns:\n", selected_columns)

# Find the Completion Date and URL for the last 3 clinical trials in the dataset
last_3_trials = df.iloc[-3:, [df.columns.get_loc('Completion Date'), df.columns.get_loc('URL')]]
print("Completion Date and URL for the last 3 clinical trials:\n", last_3_trials)

```

```python
# How many studies have an enrollment greater than a certain threshold?
threshold = 1000  # Replace with your threshold value
enrollment_threshold_count = df[df['Enrollment'] > threshold].shape[0]
print(f"Number of studies with enrollment greater than {threshold}:", enrollment_threshold_count)

# How many clinical trials have 'No Results Available'?
no_results_count = df[df['Study Results'] == 'No Results Available'].shape[0]
print("Number of clinical trials with 'No Results Available':", no_results_count)

# How many clinical trials are in "Completed" and "Recruiting" status?
completed_trials_count = df[df['Status'] == 'Completed'].shape[0]
recruiting_trials_count = df[df['Status'] == 'Recruiting'].shape[0]
print("Number of 'Completed' trials:", completed_trials_count)
print("Number of 'Recruiting' trials:", recruiting_trials_count)

# How many clinical trials are related to 'COVID-19'?
covid_trials_count = df[df['Conditions'].str.contains('COVID-19', case=False, na=False)].shape[0]
print("Number of clinical trials related to 'COVID-19':", covid_trials_count)

# How many clinical trials started after January 1, 2020?
start_date_threshold = '2020/01/01'
trials_after_date_count = df[pd.to_datetime(df['Start Date']) > start_date_threshold].shape[0]
print("Number of clinical trials started after January 1, 2020:", trials_after_date_count)
```
