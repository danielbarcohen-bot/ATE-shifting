```python
import pandas as pd #it is library it gives function and methods
```

```python
dataset = pd.read_csv("Salary_Data.csv") #here invoke the salary data files
```

```python
dataset #it invoked readed outputs
```

```python
independent = dataset[["YearsExperience"]] #split the input datas from invoked file
independent # display the input values 
```

```python
dependent= dataset[["Salary"]]# split the output from invoked file
dependent  #display the output values
```

```python
from sklearn.model_selection import train_test_split # it is import for traning and testing purpose 
X_train,X_test,Y_train,Y_test=train_test_split(independent,dependent, test_size=0.30,random_state=0)#here assign variables and call for model creation 
```

```python
from sklearn.linear_model import LinearRegression #here use for taking particular model function 
regressor=LinearRegression() # here assigned the function to variable
regressor.fit(X_train,Y_train)#here call for train model creation 
```

```python
weight = regressor.coef_  #here calculate weight 
weight #it is containe weight value
```

```python
bais = regressor.intercept_ #it is calculate bais 
bais #it is containe bais value
```

```python
y_pred=regressor.predict(X_test) #give test input value for model performance 
```

```python
from sklearn.metrics import r2_score# import the metrics library from sklearn for evaluating 
r_score=r2_score(Y_test,y_pred) # call fuction for evaluate
```

```python
r_score # here evaluate value  for model good or bad reslut
```

```python
import pickle #it is used for deployment file create 
filename="finalized_model_linear_practice.sav" #here we can create your file like
```

```python
pickle.dump(regressor,open(filename,'wb')) #here trained modal put on inside file 
```
