```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import tkinter as tk
from tkinter import filedialog
from google.colab import files # Import google.colab.files
import io # Import io for BytesIO

plt.close('all')

# Modified function to load file in Colab
def cargar_archivo():
    uploaded = files.upload()
    if uploaded:
        for filename in uploaded.keys():
            return io.BytesIO(uploaded[filename]) # Return a file-like object
    return None # Return None if no file was uploaded

def procesar_grupo(ruta, nombre):
    # Now, ruta will be a file-like object, so read_csv directly from it
    df = pd.read_csv(ruta)

    if "HorasEstudio" not in df.columns or "Calificacion" not in df.columns:
        print(f"ERROR: El archivo del {nombre} no contiene columnas v?lidas.\n")
        return

    print("\n==============================")
    print(f"RESULTADOS {nombre}")
    print("==============================")
    print("Descripci?n estad?stica del conjunto de datos:")
    print(df.describe())

    # Histograma
    plt.figure()
    plt.hist(df["Calificacion"], bins=10, edgecolor='black')
    plt.xlabel("Calificaci?n")
    plt.ylabel("Frecuencia")
    plt.title(f"Histograma de Calificaciones - {nombre}")
    plt.show()

    X = df["HorasEstudio"].values.reshape(-1, 1)
    y = df["Calificacion"].values

    modelo = LinearRegression()
    modelo.fit(X, y)

    pendiente = modelo.coef_[0]
    intercepto = modelo.intercept_
    r2 = modelo.score(X, y)

    print(f"\nCoeficiente de determinaci?n (R?): {r2:.4f}")
    print(f"Ecuaci?n de la recta de regresi?n: y = {pendiente:.4f}x + {intercepto:.4f}")

    if pendiente > 0:
        interpretacion = "Existe una relaci?n lineal positiva: a mayor n?mero de horas de estudio, mayor calificaci?n."
    elif pendiente < 0:
        interpretacion = "Existe una relaci?n lineal negativa: a mayor n?mero de horas de estudio, menor calificaci?n."
    else:
        interpretacion = "No existe relaci?n lineal entre las horas de estudio y las calificaciones."

    print("Interpretaci?n:", interpretacion)
    print("======================================================\n")

    # Gr?fica de dispersi?n y l?nea de regresi?n
    plt.figure()
    plt.scatter(df["HorasEstudio"], df["Calificacion"])
    plt.plot(df["HorasEstudio"], modelo.predict(X))
    plt.xlabel("Horas de Estudio")
    plt.ylabel("Calificaci?n")
    plt.title(f"Regresi?n Lineal - {nombre}")
    plt.show()


# =======================
# PROCESAMIENTO DE LOS 3 GRUPOS
# =======================

print("Por favor, selecciona el archivo CSV para el GRUPO 1")
ruta1 = cargar_archivo()
if ruta1 is not None:
    procesar_grupo(ruta1, "GRUPO 1")

print("\nPor favor, selecciona el archivo CSV para el GRUPO 2")
ruta2 = cargar_archivo()
if ruta2 is not None:
    procesar_grupo(ruta2, "GRUPO 2")

print("\nPor favor, selecciona el archivo CSV para el GRUPO 3")
ruta3 = cargar_archivo()
if ruta3 is not None:
    procesar_grupo(ruta3, "GRUPO 3")
```
