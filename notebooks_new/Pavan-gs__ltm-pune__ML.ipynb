```python
y = df1['Exited']
```

```python
df1.shape
```

```python
x=df1.iloc[:,2:12]
```

```python
x.drop('Gender', axis=1, inplace=True)
```

```python
x[5:10]['Geography']
```

```python
pd.get_dummies(x['Geography'], drop_first=True)
```

```python
x1 = pd.concat([x,pd.get_dummies(x['Geography'], drop_first=True)], axis=1)
```

```python
x1.drop('Geography', axis=1, inplace=True)
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x1,y, test_size=0.2, random_state=2)
```

```python
ytrain.value_counts()
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
```

```python
model2 = lr.fit(xtrain,ytrain)
```

```python
pred = model2.predict(xtest)
```

```python
pred[:5]
```

```python
ytest[:5]
```

```python
pd.crosstab(ytest, pred)
```

```python
# Decision tree
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtree = DecisionTreeClassifier(max_depth=5, min_samples_leaf=3, min_samples_split=5)
```

```python
model_tree = dtree.fit(xtrain,ytrain)
```

```python
pred_tree = model_tree.predict(xtest)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
confusion_matrix(ytest, pred_tree)
```

```python
print(classification_report(ytest,pred_tree))
```

```python
import matplotlib.pyplot as plt
```

```python
from sklearn import tree
```

```python
plt.figure(figsize=(30,12))
tree.plot_tree(decision_tree=model_tree, feature_names=x1.columns, class_names=['pos','neg'], 
               filled=True, fontsize=12)
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier()
```

```python
model_rf = rf.fit(xtrain,ytrain)
```

```python
pred_rf = model_rf.predict(xtest)
```

```python
print(classification_report(ytest,pred_rf))
```
