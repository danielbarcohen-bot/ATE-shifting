```python
# 🎬 Netflix Data Analysis with Python
# Author: Hardik Jodhani

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Load the dataset
df = pd.read_csv(r"D:\Netflix\netflix_titles.csv\netflix_titles.csv")


# Step 2: Basic info
print("Shape:", df.shape)
print(df.info())
print(df.head())

# Step 3: Handle missing values
df['country'] = df['country'].fillna("Unknown")

# Strip spaces and handle errors safely
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')

df.dropna(subset=['release_year'], inplace=True)

# Step 4: Movies vs TV Shows
type_count = df['type'].value_counts()
sns.barplot(x=type_count.index, y=type_count.values, hue=type_count.index, palette='viridis', legend=False)

plt.title("Movies vs TV Shows on Netflix")
plt.ylabel("Count")
plt.show()

# Step 5: Top 10 Countries with Most Content
top_countries = df['country'].value_counts().head(10)
top_countries.plot(kind='bar', color='skyblue')
plt.title("Top 10 Countries with Most Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.show()

# Step 6: Content Added Over the Years
df['year_added'] = df['date_added'].dt.year
yearly_trend = df['year_added'].value_counts().sort_index()
yearly_trend.plot(kind='line', marker='o', color='crimson')
plt.title("Netflix Content Added Over the Years")
plt.xlabel("Year")
plt.ylabel("Titles Added")
plt.show()

# Step 7: Top 10 Genres
genres = df['listed_in'].str.split(',').explode().str.strip()
top_genres = genres.value_counts().head(10)
sns.barplot(y=top_genres.index, x=top_genres.values, hue=top_genres.index, palette='mako', legend=False)

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.show()
```

## 📊 Project Summary: Netflix Data Analysis

This beginner data science project explores the Netflix dataset to understand content distribution and trends.

### 🔍 Key Insights:
- **Movies vs TV Shows:** Movies make up the majority of the content on Netflix.  
- **Popular Genres:** Drama, Comedy, and Action & Adventure appear most frequently.  
- **Top Countries:** The United States and India produce a large share of Netflix titles.  
- **Recent Additions:** Most new content has been added after 2015, showing Netflix’s rapid expansion.
