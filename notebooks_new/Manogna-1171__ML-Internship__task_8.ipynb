```python
df_filtered['y']=df_filtered['y'].fillna(df_filtered['y'].mode()[0])
```

```python
df_filtered['cons.conf.idx']=df['cons.conf.idx']
df_filtered['cons.conf.idx']=df_filtered['cons.conf.idx'].fillna(df_filtered['cons.conf.idx'].mean())
df_filtered['euribor3m']=df_filtered['euribor3m'].fillna(df_filtered['euribor3m'].mean())
df_filtered['nr.employed']=df_filtered['nr.employed'].fillna(df_filtered['nr.employed'].mean())
# df_filtered=df_filtered.drop(columns=['cons.conf.idx\t'])
```

```python
df_filtered.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

```python
le=LabelEncoder()
df_filtered['y']=le.fit_transform(df_filtered['y'])
# df_filtered
```

```python
ohe=OneHotEncoder(handle_unknown='ignore',sparse_output=False).set_output(transform="pandas")
encoded=ohe.fit_transform(df_filtered[['job','marital','education','default','housing','loan','contact','month','day_of_week','poutcome']])
```

```python
encoded
```

```python
df_filtered=df_filtered.drop(columns=['job','marital','education','default','housing','loan','contact','month','day_of_week','poutcome'])
```

```python
df_filtered
```

```python
df_filtered=pd.concat([df_filtered,encoded],axis=1)
df_filtered
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=df_filtered.drop(columns=['y'])
y=df_filtered['y']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
clf=DecisionTreeClassifier(max_depth=4, random_state=42)
clf.fit(x_train,y_train)
```

## **Visualize the tree using plot_tree():**

```python
from sklearn.tree import plot_tree
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(20,10))
plot_tree(clf,feature_names=x.columns,class_names=['no','yes'],filled=True)
plt.savefig("decisiontree.png")
plt.show()
```

```python
from sklearn.metrics import classification_report, accuracy_score

y_pred = clf.predict(x_test)

"Test Accuracy:", accuracy_score(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```

## **Train Accuracy Vs. Test Accuracy:**

```python
print("Train Accuracy:", clf.score(x_train, y_train))
print("Test Accuracy:", clf.score(x_test, y_test))
```
