## Types of Errors

There are two main types of errors – **systematic** and **random**:

**1. Systematic Errors or "Biased" Errors**

Systematic errors are consistent and predictable. If you repeat the measurement, the error stays the same or follows a specific pattern. These errors affect the accuracy of your results—how close you are to the "true" value.

- Cause: Usually a feature of the experimental setup, e.g. experimental design, accuracy of our measuring equipment or inherent offsets in values recorded, etc..

- Fix: These can’t be removed by averaging. You have to calibrate your tools or change your method.

Examples:

- Zero Error: A digital scale that reads 0.5g even when nothing is on it. Every single measurement will be exactly 0.5g too high.

- Calibration Issues: Using a cloth measuring tape that has stretched out over time. Every meter you measure will actually be shorter than a real meter.

- Environmental Interference: Measuring the length of a metal rod in a very hot room where the metal has expanded.

**2.  Random Errors or "Noisy" Errors**

Random errors are unpredictable fluctuations that happen in every experiment. They don't have a specific direction, i.e. sometimes the reading is too high, sometimes it's too low. These affect the precision of your results.

- Cause: Uncontrollable variables like vibrations, small temperature changes, the limit of human reaction time, natural variation in a sample, etc..

- Fix: You can minimize the impact of random errors by taking multiple measurements and calculating an average.

Examples:

- Human Reaction Time: Using a manual stopwatch to time a 100m sprint. You might click a millisecond too early one time and a millisecond too late the next.

- Parallax Error: Looking at a liquid level in a graduated cylinder from slightly different angles each time.

- Electronic Noise: Small, "jumpy" fluctuations in the last digit of a digital voltmeter caused by nearby electromagnetic interference.

## Summary Comparison

|Feature | Systematic Error |Random Error|
| --- | --- | --- |
|Consistency |Constant/Predictable |Unpredictable/Varies|
|Direction | Always in one direction (too high or too low) |In both directions (+ or -)|
Effect | Affects **Accuracy** |Affects **Precision**|
|Solution |Calibrate equipment or improve method |Take multiple measurements and average them|

## Precision vs. Accuracy

While people often use these words interchangeably in casual conversation, in science and engineering, they mean two very different things.

### Example: Temperature measurements with a Thermometer

Imagine a room that is exactly 25°C:

- Case A: You measure the temperature three times and get 24.9°C, 25.1°C, and 25.0°C. You are both **accurate** and **precise**.

- Case B: You get 30.1°C, 30.2°C, and 30.1°C. You are **precise** (the numbers are close together) but **inaccurate** (the thermometer is calibrated incorrectly).

- Case C: You get 20°C, 30°C, and 25°C. The average is 25°C, so it is **accurate** on average, but the individual measurements have low **precision**.

## Mean, standard deviation and standard error

### Mean:

The mean, often referred to as the average, is calculated by adding all values $x_i$ in a dataset and dividing by the number of values (sample size $n$). It represents the central point of the data:

$\bar{x} = \frac{1}{n}\sum_{i=1}^{n}x_i$.


### Standard Deviation:

Standard deviation ($\sigma$) measures the amount of variation or dispersion in a set of values. A low $\sigma$ indicates that values are close to the mean, while a high $\sigma$ indicates that values are spread out over a wider range:

$\sigma_x = \sqrt{\frac{\sum_{i=1}^{n}(x_i-\bar{x})^2}{n}}$.

Note: $\sigma^2$ is called variance.


### Standard Error:

The standard error (SE) is the standard deviation of the sampling distribution. It estimates how far the sample mean is likely to be from the population mean. The formula for SE is:

$SE = \frac{\sigma_x}{\sqrt{n}}$.
