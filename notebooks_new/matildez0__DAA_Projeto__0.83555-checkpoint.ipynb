## 2 - Preprocess Function

```python
def preprocess(df, le_luminosity=None, le_cloudiness=None, ohe_rain=None, scaler=None, is_train=True):

    # 1. Record Date Handling
    df['record_date'] = pd.to_datetime(df['record_date'])
    df['year'] = df['record_date'].dt.year
    df['month'] = df['record_date'].dt.month
    df['day'] = df['record_date'].dt.day
    df['hour'] = df['record_date'].dt.hour
    df['day_of_week'] = df['record_date'].dt.dayofweek 
    df = df.drop(columns=['record_date'])

    # 2. Add isWeekend Feature (This feature is wrong, but fixing it, or dropping it, makes the model worse... even though importance is 0.0000)
    df['isWeekend'] = df['day'].apply(lambda x: 1 if x in [5, 6] else 0)

    # 3. Create Cyclical Features (example: Makes it so the model know 23h is close to 1h)
    df['hour_sin'] = np.sin(2 * np.pi * df['hour'] / 24.0)
    df['hour_cos'] = np.cos(2 * np.pi * df['hour'] / 24.0)
    df['month_sin'] = np.sin(2 * np.pi * df['month'] / 12.0)
    df['month_cos'] = np.cos(2 * np.pi * df['month'] / 12.0)
    df['day_of_week_sin'] = np.sin(2 * np.pi * df['day_of_week'] / 7.0)
    df['day_of_week_cos'] = np.cos(2 * np.pi * df['day_of_week'] / 7.0)

    # 4. Consolidate Categories and Translate
    df['AVERAGE_CLOUDINESS'] = df['AVERAGE_CLOUDINESS'].replace({
        'nuvens quebrados': 'broken clouds', 'nuvens quebradas': 'broken clouds',
        'céu claro': 'clear sky', 'céu limpo': 'clear sky',
        'tempo nublado': 'overcast', 'nublado': 'overcast',
        'nuvens dispersas': 'scattered clouds',
        'céu pouco nublado': 'few clouds'
    })
    
    df['AVERAGE_RAIN'] = df['AVERAGE_RAIN'].replace({
        # Light Rain
        'chuviscos': 'light rain', 'chuva fraca': 'light rain',
        'chuvisco fraco': 'light rain', 'chuva leve': 'light rain',
        'aguaceiros fracos': 'light rain', 'chuvisco e chuva fraca': 'light rain',
        # Rain
        'chuva': 'rain', 'chuva moderada': 'rain', 'aguaceiros': 'rain',
        # Heavy Rain
        'chuva forte': 'heavy rain', 'chuva de intensidade pesada': 'heavy rain',
        'chuva de intensidade pesado': 'heavy rain',
        # Thunderstorm
        'trovoada com chuva': 'thunderstorm', 'trovoada com chuva leve': 'thunderstorm'
    })
    
    df['LUMINOSITY'] = df['LUMINOSITY'].replace('LOW_LIGHT', 'DARK')

    # 5. Fill Missing Values
    if is_train:
        cloudiness_mode = df['AVERAGE_CLOUDINESS'].mode()[0]
        df['AVERAGE_CLOUDINESS'] = df['AVERAGE_CLOUDINESS'].fillna(cloudiness_mode)
    else:
        cloudiness_mode = 'clear sky'
        df['AVERAGE_CLOUDINESS'] = df['AVERAGE_CLOUDINESS'].fillna(cloudiness_mode)
    df['AVERAGE_RAIN'] = df['AVERAGE_RAIN'].fillna('no rain')

    # 6. Encoding
    # Label Encoding
    if is_train:
        le_luminosity = LabelEncoder()
        le_cloudiness = LabelEncoder()
        df['luminosity_encoded'] = le_luminosity.fit_transform(df['LUMINOSITY'])
        df['cloudiness_encoded'] = le_cloudiness.fit_transform(df['AVERAGE_CLOUDINESS'])
    else:
        df['luminosity_encoded'] = le_luminosity.transform(df['LUMINOSITY'])
        df['cloudiness_encoded'] = le_cloudiness.transform(df['AVERAGE_CLOUDINESS'])

    # One-Hot Encoding
    rain_categories = sorted(df['AVERAGE_RAIN'].unique())
    if is_train:
        ohe_rain = OneHotEncoder(categories=[rain_categories], handle_unknown='ignore', sparse_output=False)
        rain_encoded = ohe_rain.fit_transform(df[['AVERAGE_RAIN']])
    else:
        rain_encoded = ohe_rain.transform(df[['AVERAGE_RAIN']])
    rain_df = pd.DataFrame(rain_encoded, columns=[f"rain_{cat}" for cat in ohe_rain.categories_[0]], index=df.index)
    df = pd.concat([df, rain_df], axis=1)
    
    # 7. Scale Numerical Features (Places all numerical features on the same scale)
    numerical_cols = ['AVERAGE_FREE_FLOW_SPEED', 'AVERAGE_TIME_DIFF', 'AVERAGE_FREE_FLOW_TIME',
                      'AVERAGE_TEMPERATURE', 'AVERAGE_ATMOSP_PRESSURE', 'AVERAGE_HUMIDITY', 'AVERAGE_WIND_SPEED']
    
    numerical_cols = [col for col in numerical_cols if col in df.columns]

    if is_train:
        scaler = StandardScaler()
        df[numerical_cols] = scaler.fit_transform(df[numerical_cols])
    else:
        df[numerical_cols] = scaler.transform(df[numerical_cols])

    # 8. Drop original and unused columns
    cols_to_drop = [
        'LUMINOSITY',
        'AVERAGE_CLOUDINESS',
        'AVERAGE_RAIN',
        'isHoliday',
        'city_name',
        'hour',
        'month',
        'day_of_week'#,
        #'isWeekend',
        #'rain_heavy rain',
        #'rain_thunderstorm'
    ]
    if 'AVERAGE_PRECIPITATION' in df.columns:
        cols_to_drop.append('AVERAGE_PRECIPITATION')
    df = df.drop(columns=[col for col in cols_to_drop if col in df.columns])

    if is_train:
        return df, le_luminosity, le_cloudiness, ohe_rain, scaler
    else:
        return df
```

## 3 - Load Data and Apply Preprocess

```python
# Load data
nalist = ["NULL", ""]
df = pd.read_csv('datasets/training_data.csv', encoding='latin1', keep_default_na=False, na_values=nalist)

# Encode Target Variable
le_target = LabelEncoder()
df['speed_diff_encoded'] = le_target.fit_transform(df['AVERAGE_SPEED_DIFF'])
y = df['speed_diff_encoded']

# Preprocess training data, fitted encoders and scaler
df_train, le_lum, le_cld, ohe_r, scaler = preprocess(df.drop(columns=['speed_diff_encoded']), is_train=True)
```

## 4 - Define and Split Data for Testing

```python
# Define features
features = [col for col in df_train.columns if col not in ['AVERAGE_SPEED_DIFF']]
X = df_train[features]

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=2025, stratify=y
)

print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}, y_test shape: {y_test.shape}")
print(f"Features being used: {features}")
```

## 4.1 - KMeans, Medoids and DBSCAN

```python
from sklearn.cluster import KMeans, DBSCAN
from sklearn.neighbors import KNeighborsClassifier
try:
    from sklearn_extra.cluster import KMedoids
    HAS_KMEDOIDS = True
except ImportError:
    HAS_KMEDOIDS = False
    print("sklearn_extra not found. Skipping K-Medoids (using K-Means proxy instead).")

print("--- Starting Clustering Feature Engineering ---")

# Create copies so we dont change the originals
X_train_clus = X_train.copy()
X_test_clus = X_test.copy()

# 1. K-Means Clustering
# Try 5 clusters
print("1. Running K-Means (k=5)...")
kmeans = KMeans(n_clusters=5, random_state=2025, n_init=10)
kmeans.fit(X_train)

# Assign Clusters
X_train_clus['kmeans_cluster'] = kmeans.predict(X_train)
X_test_clus['kmeans_cluster'] = kmeans.predict(X_test)

# 2. DBSCAN Clustering
# DBSCAN finds outliers
print("2. Running DBSCAN (eps=3.0)...")
dbscan = DBSCAN(eps=3.0, min_samples=10)
train_clusters = dbscan.fit_predict(X_train)

# DBSCAN has no .predict(), so we use a KNN to project clusters to Test Data
knn_proxy = KNeighborsClassifier(n_neighbors=1)
knn_proxy.fit(X_train, train_clusters)

X_train_clus['dbscan_cluster'] = train_clusters
X_test_clus['dbscan_cluster'] = knn_proxy.predict(X_test)

# 3. K-Medoids (Robust K-Means)
if HAS_KMEDOIDS:
    print("3. Running K-Medoids (k=5)...")
    kmedoids = KMedoids(n_clusters=5, random_state=2025, method='pam')
    kmedoids.fit(X_train)
    
    X_train_clus['kmedoids_cluster'] = kmedoids.predict(X_train)
    X_test_clus['kmedoids_cluster'] = kmedoids.predict(X_test)
else:
    print("   (Skipping K-Medoids)")

# 4. Evaluate: Did it help?
print("\n--- Evaluating XGBoost with Clustering Features ---")

# Train Baseline XGBoost (Optimized) on NEW Data
xgb_clus = XGBClassifier(
    n_estimators=100, 
    learning_rate=0.2, 
    max_depth=3, 
    min_child_weight=2, 
    gamma=0,
    random_state=2025, 
    use_label_encoder=False, 
    eval_metric='mlogloss'
)

xgb_clus.fit(X_train_clus, y_train)
y_pred_clus = xgb_clus.predict(X_test_clus)
acc_clus = accuracy_score(y_test, y_pred_clus)

#The baseline accuracy is not dynamic, this code could be improved in the future
print(f"Original Baseline Accuracy: {0.80039:.5f}")
print(f"Clustered Model Accuracy:   {acc_clus:.5f}")

if acc_clus > 0.80039:
    print("SUCCESS! Clustering improved the score.")
    # Update the global variables to use this new data for the Tournament
    X_train = X_train_clus
    X_test = X_test_clus
else:
    print("FAILURE! Clustering added noise. Discarding changes.")
```

## 5 - Model Training

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier, StackingClassifier
from xgboost import XGBClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import accuracy_score
import pandas as pd
import numpy as np

# 1. PyTorch Wrapper
class PyTorchTrafficClassifier(BaseEstimator, ClassifierMixin):
    def __init__(self, input_dim, output_dim, class_weights=None, epochs=300, lr=0.001, batch_size=64):
        self.input_dim = input_dim
        self.output_dim = output_dim
        self.epochs = epochs
        self.lr = lr
        self.batch_size = batch_size
        self.class_weights = class_weights
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        self.model = None
        
    def fit(self, X, y):
        self.model = nn.Sequential(
            nn.Linear(self.input_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, self.output_dim)
        ).to(self.device)
        
        if self.class_weights is not None:
            weights_t = torch.tensor(self.class_weights, dtype=torch.float32).to(self.device)
            criterion = nn.CrossEntropyLoss(weight=weights_t)
        else:
            criterion = nn.CrossEntropyLoss()
            
        optimizer = optim.Adam(self.model.parameters(), lr=self.lr)
        
        X_array = X.values if hasattr(X, 'values') else X
        y_array = y.values if hasattr(y, 'values') else y
        
        X_t = torch.tensor(X_array, dtype=torch.float32)
        y_t = torch.tensor(y_array, dtype=torch.long)
        
        dataset = TensorDataset(X_t, y_t)
        loader = DataLoader(dataset, batch_size=self.batch_size, shuffle=True)
        
        self.model.train()
        for epoch in range(self.epochs):
            for batch_X, batch_y in loader:
                batch_X, batch_y = batch_X.to(self.device), batch_y.to(self.device)
                optimizer.zero_grad()
                outputs = self.model(batch_X)
                loss = criterion(outputs, batch_y)
                loss.backward()
                optimizer.step()
        return self

    def predict(self, X):
        self.model.eval()
        X_array = X.values if hasattr(X, 'values') else X
        with torch.no_grad():
            X_t = torch.tensor(X_array, dtype=torch.float32).to(self.device)
            outputs = self.model(X_t)
            _, predicted = torch.max(outputs, 1)
        return predicted.cpu().numpy()

# 2. Define Models

# XGBoost (Optimized)
xgb_model = XGBClassifier(
    n_estimators=100, 
    learning_rate=0.2, 
    max_depth=3, 
    min_child_weight=2, 
    gamma=0,
    random_state=2025, 
    #use_label_encoder=False, deprecated in py3.13
    eval_metric='mlogloss'
)

# Random Forest
rf_model = RandomForestClassifier(n_estimators=300, random_state=2025, n_jobs=-1)

# Gradient Boosting
gbc_model = GradientBoostingClassifier(n_estimators=200, learning_rate=0.1, max_depth=3, random_state=2025)

# PyTorch Neural Network
classes = np.unique(y_train)
weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_train)
input_dim = X_train.shape[1]
output_dim = len(classes)

pytorch_model = PyTorchTrafficClassifier(
    input_dim, output_dim, class_weights=weights, epochs=300, lr=0.001, batch_size=64
)

# Stacking
estimators = [
    ('xgb', xgb_model),
    ('rf', rf_model),
    ('gbc', gbc_model)
]
stacking_model = StackingClassifier(
    estimators=estimators,
    final_estimator=LogisticRegression(),
    cv=5,
    n_jobs=-1
)

# Dictionary
models = {
    "Random Forest": rf_model,
    "Gradient Boosting": gbc_model,
    "XGBoost (Optimized)": xgb_model,
    "Neural Network (PyTorch)": pytorch_model,
    "Stacking (XGB+RF+GBC)": stacking_model
}

# 3. The Tournament Loop - shows the accuracy of all models
best_acc = 0
best_model_name = ""

print("-" * 40)
print(f"{'Model Name':<25} | {'Accuracy':<10}")
print("-" * 40)

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred_loop = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred_loop)
    
    print(f"{name:<25} | {acc:.5f}")
    
    if acc > best_acc:
        best_acc = acc
        best_model = model
        best_model_name = name

print("-" * 40)
print(f"🏆 Winner: {best_model_name} with {best_acc:.5f}")
```
