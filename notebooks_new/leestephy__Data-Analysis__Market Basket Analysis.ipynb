### B3. MARKET BASKET ASSUMPTION
- Market basket analysis assumes that items frequently purchased together are associated, indicating a meaningful relationship, even if they are not directly related.

### C1. TRANSFORMING THE DATA SET

```python
# convert to list of lists
rows = []
for i in range(0, 7501):
	rows.append([str(df.values[i,j])
for j in range(0, 20)])
```

```python
# encode list
encoder = TransactionEncoder()
array = encoder.fit(rows).transform(rows)
```

```python
# Convert array back to a DataFrame
pd.set_option('display.max_columns',None)
transaction = pd.DataFrame(array, columns = encoder.columns_)

transaction
```

```python
# Display items as column
for col in transaction.columns:
	print(col)
```

```python
# Remove nan column from transformed dataset
df_clean = transaction.drop(['nan'], axis = 1)
df_clean.head()
df_clean.shape
```

```python
# Write transformed dataset to CSV
df_clean.to_csv(r'C:\Users\steph\Documents\GitHub\Data Analysis\11. Market Basket Analysis in Python\Outputs\df_clean.csv')
df_clean.columns
```

### C2. CODE EXECUTION

```python
# Create Apriori object called rules
rules = apriori(df_clean, min_support = 0.02, use_colnames = True)
rules.head(5)
```

### C3. ASSOCIATION RULES TABLE

```python
# Create a rules table
# metric set to lift b/c the higher the lift, the stronger the likelihood of the items getting purchased together
# any rule below 1 will not be considered
rules_table = association_rules(rules, metric = 'lift', min_threshold = 1)
rules_table.head(10)
```

### C4. TOP THREE RULES

```python
top_three_rules = rules_table.sort_values('support', ascending = False).head(3)
top_three_rules
```

```python
top_three_rules = rules_table.sort_values('lift', ascending = False).head(3)
top_three_rules
```

```python
# Sorting rules by metric in descending order
top_three_rules = rules_table.sort_values('confidence', ascending = False).head(3)
top_three_rules
```

### D1. SIGNIFICANCE OF FINDINGS
- **Support**: The leading support-based rules show that customers purchasing a Dust-Off Compressed Gas 2 pack are likely to also buy a VIVO Dual LCD Monitor Desk mount and HP 61 ink, with support values of 0.059725 and 0.052660, indicating that about 6% of transactions involve these items together.

- **Lift**: The primary rules derived from the lift metric demonstrate strong associations, such as when purchasing a VIVO Dual LCD Monitor Desk mount, which has a lift value of 2.291162 with a SanDisk Ultra 64GB card, suggesting that these items are significantly more likely to be bought together than by chance.

- **Confidence**: The confidence metric indicates a 45.6% probability that customers buying a 10ft iPhone Charger Cable 2 Pack will also purchase a Dust-Off Compressed Gas 2 pack. While this suggests a notable relationship, the confidence levels of 41.9% for FEIYOLD Blue Light Blocking Glasses and 41.6% for the SanDisk Ultra 64GB card indicate that these associations may not be particularly strong, as none exceed the 50% threshold

### D2. PRACTICAL SIGNIFICANCE OF FINDINGS
- **Substantial Co-Purchases**: With 7,501 transactions, a support value of 5-6% indicates that approximately 375 to 450 transactions involve associated items being purchased together, reflecting a significant pattern of co-purchases.

- **Strong Lift Associations**: The lift metric shows all three rules have lift values around 2, meaning the associated items are about twice as likely to be purchased together compared to independently.

- **Weak Confidence Correlation**: Although there is a reasonable likelihood of purchasing a Dust-Off Compressed Gas 2 pack alongside items like the 10ft iPhone Charger Cable 2 Pack and FEIYOLD Blue Light Blocking Glasses, the confidence values are below 50%, indicating a weaker correlation.
