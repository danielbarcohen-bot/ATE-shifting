Do you, like everyone else, get annoyed when your credit card clients default on their payments and cost you money? Well, this time on Intelligent Capitalism we are building out a machine learning model to predict which people are likely to default on their next payment! 

Our dataset was collected in Taiwan from April 2005 to September 2005, and it focuses on financial and basic demographic information of this particular bank's clients. First, we'll do some simple exploratory data analysis on our dataset to see if there are any features that clearly show whether someone will default or not. Figure 1 shows some histogram plots of various features in the dataset; we are looking for features that show some separability in the x-axis. Unfortunately we don't really see much separation, so we'll just go ahead with the model and see later on how it looks. 

![BILLAMT histogram](img/billamt.png) ![LIMITBAL histogram](img/limitbal.png) ![PAYAMT histogram](img/payamt.png) ![EDUCATION histogram](img/educationhist.png)

*Figure 1: From left to right and top to bottom, histograms of recent bill statement, amount of given credit, recent payment amount, and education level.*

Since a lot of our features don't seem to separate our two classes very well, we will add some of our own features to help out our model and try to get better performance. There are three features that I thought to add in here. First, credit utilization is probably useful in predicting defaults - the amount of money people borrow is likely to be correlated with failing to pay it off in time, especially if it's a large amount. Second, a debt accumulation trend might be interesting to look at, to see if someone is actively paying off debt or racking up more. Finally, since the amount of credit limit is related to a person's age, we should specify that for our model so it doesn't think that a 16 year old can have a $500K limit. 

Now that we have all our features sorted, it's time to build out our model! We first need to scale all of our columns with numeric data and encode our categorical data so that the model can read it all properly. We also drop irrelevant features here, like sex, since it probably isn't a good idea to try and segregate people based on that. For the actual model, I tried out a couple different ones with default hyperparameters, and landed on LightGBM as my model of choice. I then used RFECV to try and get rid of any useless features that the model wasn't learning from to save time, but it ended up keeping almost all of the features in place so I just ignored the results from that. Finally, I did hyperparameter optimization for the model using RandomizedSearchCV (which includes cross validation), and used that finalized model to evaluate our performance on the test data. The final LGBM model got an ROC AUC score on our test data of 0.784, which tells us that our model will assign the correct class (default) to the defaulting client with a 78.4% probability compared to a non-defaulting client. Overall, this isn't a terrible result given the size of the dataset and the amount of time spent training the model. The training and cross validation scores are also shown in Figure 2. 

![ROC scores for the LGBM models](<img/LGBM Scores.png>)

*Figure 2: ROC scores for the various LGBM models that were tested.* 

I also generated a SHAP plot on our model in Figure 3 to see what features it was prioritizing the most; we can see from the graph that recent repayment status and credit utilization were the most important factors in predicting if a client would default or not. 

![SHAP feature importance plot](img/shap_graph.png)

*Figure 3: SHAP plot of feature importances.*

Of course, this model isn't perfect, and there are plenty of issues with both the dataset in general and my approach to the problem. Regarding the dataset, it is missing really useful info like credit score and employment status, which would probably be a lot more indicative of default. It also only has data over 6 months, which is a really short time, and it was recorded almost 20 years ago so it may not be relevant today. It also just comes from one bank in Taiwan, so the results again may not generalize well to other banks or countries. 

One strange thing I noticed was that RFECV selected all of the features as being important enough to include in the model, but our SHAP graph tells us that there are quite a few features that don't really matter a ton for predictions. There could have been some issues with implementation here, or maybe my model is using unnecessary features that could've been cut out for faster training. Another issue with my approach is that this dataset is actually a time series, but I split the data randomly rather than using only future data for the test set. This means that we introduced some temporal data leakage into our model and we weren't actually predicting on "future" data, which a real-world model would have had to do. In the future, I should redo the data splitting with this in mind, and maybe add some temporal features to capture the passage of time in the model. 

Everything being said, this was an interesting dataset to work on, and I think there is plenty of potential with this to improve our banking systems to more efficiently reject people's loans.
