Data cleaning and manipulation

```python
import pandas as pd
df = pd.read_csv('data_jobs.csv')
df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
```

```python
mean_year_avg=df['salary_year_avg'].median()
mean_hour_avg=df['salary_hour_avg'].median()
```

```python
df_filled = df.copy()
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(mean_hour_avg)
df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(mean_year_avg)
```
