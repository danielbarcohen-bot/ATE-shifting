````python
# DO NOT MODIFY NOR ADD CODE TO THIS CELL

df = pd.read_csv('https://raw.githubusercontent.com/jnin/information-systems/refs/heads/main/data/superheroes_complete.csv')
df
````

Feature Selection and Engineering: 
- The goal of feature selection is to retain the most relevant features for the machine learning model while removing irrelevant or redundant ones. In this case, the Name and Publisher columns were excluded because they are either identifiers (e.g., Name) or potentially non-predictive for alignment classification (e.g., Publisher).
- Alignment is used as the target variable (y), as this is what we aim to predict. It represents whether a character is good, bad, or neutral.
- Any column that does not provide a meaningful contribution to predicting the alignment (such as names or publisher information) is discarded to reduce noise and improve model efficiency.
- Upon reviewing the number of missing values across different columns, we found that 588 out of 659 values in the "Skin color" column were missing, representing approximately 90% of the data. We concluded that this column does not provide a meaningful contribution to predicting alignment and, therefore, removed it.

````python
X = df.drop(columns=['Name','Publisher','Alignment','Skin color'])
y = df[['Alignment']]

missing_columns = X.isna().sum()  # Get the sum of NaNs per column
missing_columns = missing_columns[missing_columns > 0]  # Filter columns with NaNs
print(missing_columns)
````

#Encoding Categorical and Ordinal Features
- We use OneHotEncoder for categorical features like Gender, Eye color, Race, etc., because these variables do not have an inherent order. One-hot encoding converts these features into binary columns to represent their categories.
- For features like Power Level and Intelligence Level, we use OrdinalEncoder because these variables have a natural order. Encoding them as ordinal ensures that the model captures the relative ranking of these values.
- ColumnTransformer allows the application of different transformations to different subsets of features. This is necessary when dealing with mixed types of data (categorical vs. ordinal).

Alternative consideration
- We could use Target Encoding or Binary Encoding as an alternative to One-Hot Encoding, especially for high-cardinality categorical features (like Race). This would reduce the dimensionality but might introduce bias if not carefully handled.

````python

from sklearn.preprocessing import OneHotEncoder
one_hot_encoder = OneHotEncoder(sparse_output=False)

from sklearn.preprocessing import OrdinalEncoder
categories=[['Weak','Below Average','Average','Above Average','Extremely Powerful'],
            ['Low Intelligence','Average Intelligence','Smart','Genius','Super-Genius']]

ordinal_encoder = OrdinalEncoder(categories=categories)

from sklearn.compose import ColumnTransformer
categorical_columns=['Gender','Eye color','Race','Hair color']
ordinal_columns=['Power Level','Intelligence Level']


transformer = ColumnTransformer(
    transformers=
        [('categorical', one_hot_encoder, categorical_columns),
        ('ordinal', ordinal_encoder , ordinal_columns)],
    remainder='passthrough')

X_encoded= transformer.fit_transform(X)

X_encoded

````

#Handling missing Data TO UPDATE
- Although our code scales the data, handling missing data would typically involve using techniques such as mean, median, or mode imputation before scaling the data. The scaling step here uses StandardScaler, which assumes data is normally distributed.

Alternative consideration
- For non-normal data, alternatives like RobustScaler or MinMaxScaler could be considered.

````python
missing_columns = X_encoded.isna().sum()  # Get the sum of NaNs per column
missing_columns = missing_columns[missing_columns > 0]  # Filter columns with NaNs
print(missing_columns)

from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='mean')
X_imputed = imputer.fit_transform(X_encoded)
X_imputed
````

Standardizing Features 
- Many machine learning algorithms (e.g., linear regression or neural networks) implicitly assume that all features are equally important. Standardization ensures that each feature has a comparable range (typically mean 0 and standard deviation 1), so no single feature dominates the model just because of its magnitude.
- Features like weight and height have much larger ranges compared to binary features (which are 0 and 1). Without standardization, these larger values can dominate the learning process because many machine learning algorithms (e.g., distance-based ones like k-Nearest Neighbors (k-NN), SVMs, or gradient descent-based models like logistic regression) rely on measuring distances or gradients (source:'https://neptune.ai/blog/performance-metrics-in-machine-learning-complete-guide')
- Since we have many categorical binary values alongside continuous variables like for example weight and height, we decided to use a StandardScaler to standardize the our numerical features for more balanced feature scaling

````python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)
X_scaled
````

````python
from sklearn.pipeline import Pipeline
pipe = Pipeline(steps=[
    ('Transformer', transformer),
    ('Imputer', imputer),
    ('Scaler',scaler)
])
X_pipe= pipe.fit_transform(X)
X_pipe
````
