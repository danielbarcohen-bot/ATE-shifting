#### Tamb?m ? poss?vel optar por uma sele??o por linha e coluna. Criamos um arranjo de $(3 \times 4)$.

#### Como exemplo de aplica??o de `Fancy Indexing` vamos selecionar pontos aleat?rios, usando par?metros da [distribui??o normal multivariada](https://support.minitab.com/pt-br/minitab/18/help-and-how-to/probability-distributions-and-random-data/supporting-topics/distributions/multivariate-normal-distribution/).

```python
rand = np.random.RandomState(42)

mean = [0, 0]

cov = [[1, 2],
       [2, 5]]
```

#### Pegamos 100 pontos (s?o amostras de duas dimens?es) [`.multivariate_normal()`](https://docs.scipy.org/doc/numpy-1.10.1/reference/generated/numpy.random.multivariate_normal.html#numpy-random-multivariate-normal).

```python
X = rand.multivariate_normal(mean, cov, 100)
#X
X.shape
```

#### Fazendo um `slice` de todas as linhas da coluna $0$ ou $1$.

```python
X[:, 0]
#X[:, 1]
```

#### Podemos plotar os valores considerados acima com a ajuda da fun??o [`plt.scatter()`](https://matplotlib.org/3.2.1/api/_as_gen/matplotlib.pyplot.scatter.html#matplotlib-pyplot-scatter)), do pacote [`MatPlotLib`](https://matplotlib.org/). Importamos a biblioteca [`seaborn`](https://seaborn.pydata.org/), para a visualiza??o dos dados.

#### Observe a seguir o comando m?gico [%matplotlib inline](https://jakevdp.github.io/PythonDataScienceHandbook/04.00-introduction-to-matplotlib.html), que gera as imagens est?ticas de seu plot, embutidas no caderno (Jupyter Notebook):

```python
#%matplotlib inline
import matplotlib.pyplot as plt
import seaborn; seaborn.set()  

plt.scatter(X[:, 0], X[:, 1]);
```

#### Selecionando-se 20 dos 100 (X.shape[0] = 100) sem reposi??o [`.random.choice`](https://docs.scipy.org/doc/numpy-1.16.0/reference/generated/numpy.random.choice.html#numpy-random-choice), que gera uma amostra aleat?ria de uma determinada matriz `1D`.

```python
indices = np.random.choice(X.shape[0], 
                           size = 20, 
                           replace = False
                          )
indices
```

#### Temos agora um arranjos de ?ndices, que podemos usar para uma sele??o de pontos.

```python
selection = X[indices] # fancy indexing!!
selection.shape
```

#### E essa sele??o de pontos pode ser replotada.

```python
plt.scatter(X[:, 0], 
            X[:, 1]
           )

plt.scatter(selection[:, 0], 
            selection[:, 1], 
            facecolor = 'red'
           );
```
