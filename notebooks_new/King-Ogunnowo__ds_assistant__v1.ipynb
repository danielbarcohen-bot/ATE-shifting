```python
import os
import json
import ollama
import numpy as np
import pandas as pd
import logging

logging.basicConfig(level=logging.INFO)

# state

_conversation_path = None

def set_conversation_path(path):
    global _conversation_path
    _conversation_path = path

def get_conversation_path():
    if _conversation_path is None:
        raise RuntimeError("Conversation path is not set. Call `load_conversations()` first.")
    return _conversation_path





def get_project_root(project_name="eda_assistant"):
    current = os.getcwd()
    while True:
        if os.path.basename(current) == project_name:
            return current
        parent = os.path.dirname(current)
        if parent == current:
            raise FileNotFoundError(f"Could not find project root named '{project_name}'")
        current = parent

def load_conversations(project_name = None):
    if project_name is None:
        raise ValueError("Project name must be provided. Must be same as root folder name.")
    abs_project_path = get_project_root(project_name)
    files_in_abs_project_path = os.listdir(abs_project_path)
    conversations = [f for f in files_in_abs_project_path if f.startswith(".conversations")]
    if len(conversations) == 0:
        logging.info(f"No existing conversations found in {abs_project_path}. Creating new directory.")
        conversations_path = os.path.join(abs_project_path, ".conversations")
        os.makedirs(conversations_path, exist_ok=True)
        set_conversation_path(conversations_path)
        logging.info(f"Created new conversations directory at: {conversations_path}")

    else:
        conversations_path = os.path.join(abs_project_path, conversations[0])
        set_conversation_path(conversations_path)
        logging.info(f"Found existing conversations directory at: {conversations_path}")
 


load_conversations(project_name = "eda_assistant")


```
