Assim como ? poss?vel interromper uma repeti??o usando o comando ```break```, ? poss?vel interromper a execu??o de um procedimento usando o comando ```return```.

````python
def raiz_quadrada(n):
    if n < 0:
        return
    print("Raiz:", n ** 0.5)

raiz_quadrada(4)
raiz_quadrada(16)
raiz_quadrada(-1)
````

## Produzindo um valor a partir de um procedimento

Tanto ? poss?vel retornar valores diretamente, como usando as var?aveis associadas a esses valores.

````python
def soma_n_primeiros(n):
    soma = 0
    for i in range(n):
        soma += i
    return soma

print(soma_n_primeiros(3))
print(soma_n_primeiros(5))
print(soma_n_primeiros(10))
````
