```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from warnings import filterwarnings
filterwarnings("ignore")


df=pd.DataFrame({"status": ["single", "married", "separated", "single","married"]})
```

```python
df
```

```python
encoder=OneHotEncoder()
encoder
```

```python
encoded=encoder.fit_transform(df[["status"]]).toarray()
encoded
```

```python
encoder.get_feature_names_out()
encoded_df=pd.DataFrame(encoded, columns=encoder.get_feature_names_out())
encoded_df
```

```python
encoder.transform([["single"]]).toarray()
```

```python
pd.get_dummies(encoded_df, drop_first=True) 
```

```python
final_df=pd.concat([df,encoded_df], axis=1)
```

```python
#target guided ordinal encoding >> based on the target variable we can assign the order
#relationship with tgarget variable 
#A lot of categories
#replace the category with mean and median with respectative group
```

```python
df3 = pd.DataFrame({
    'time': ["lunch", "breakfast", "dinner", "lunch", "breakfast", "dinner", "lunch", "breakfast", "dinner"],
    'total_bill': [120, 130, 90, 125, 150, 190, 160, 180, 189]
})
df3
```

```python
mean_price=df3.groupby("time")["total_bill"].mean().to_dict()
```

```python
df3["time_encoded"]=df3['time'].map(mean_price)
```
