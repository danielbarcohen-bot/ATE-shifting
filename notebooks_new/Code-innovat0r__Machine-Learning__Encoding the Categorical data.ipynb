```python
#Label Encoding
#One-Hot Encoding
```

```python
from sklearn import preprocessing
import pandas as pd
```

```python
# defining the object for the labelEncoder
le = preprocessing.LabelEncoder()
```

```python
#creating the dataframe
data = pd.DataFrame({'animals':['cat','dog','horse']})
```

```python
data
```

```python
#pass the data to the LabelEncoder
le.fit(data)
```

```python
encoded_data = le.transform(data)
```

```python
encoded_data
```

```python
#One Hot Encoding
```

```python
ohe = preprocessing.OneHotEncoder()
ohe.fit(data)
encoded_ohe_data = ohe.transform(data)
```

```python
encoded_ohe_data.toarray()
```
