```python
import json
import joblib
import numpy as np
from tqdm import tqdm
from xgboost import XGBClassifier

# 1. 加载模型
# 正确加载方式
model_path = 'xgboost_classify_model_v02_mean.pkl'
model = joblib.load(model_path)
# 2. 读取原始JSON数据
input_path = 'C:\job\python\TTTTTTT\Cool\guomao_probability_mean_2.json'  # 请替换为实际输入路径
with open(input_path, 'r', encoding='utf-8') as f:
    buildings = json.load(f)


# 预计算总信号量
total_signals = sum(len(b['signals']) for b in buildings)

# 带进度条的处理流程
with tqdm(total=total_signals, desc="🚀 处理信号", unit="signal") as pbar:
    for building in buildings:
        for signal in building['signals']:
            # 构建特征向量
            # 特征工程（保持与修改后一致）
            feature = [
            signal['cn0'],
            signal['elevation'],
            signal['azimuth'],
            #signal['tan'],
            signal['location_signal_length'],
            #signal['location_length'],
            #signal['location_azimuth'],
            #len(signal.get('passed', {})),
            #abs(signal['location_azimuth']-signal['azimuth'])%180,
            #float(building['didi_height'])

            ]

            # 模型预测
            features = np.array([feature], dtype=np.float32)
            signal['is_blocked'] = bool(model.predict(features))

            pbar.update(1)

# 4. 保存结果到新JSON文件
output_path = r'C:\job\python\TTTTTTT\Cool\guomao_probability_mean_output_2.json'  # 请替换为期望的输出路径
with open(output_path, 'w', encoding='utf-8') as f:
    json.dump(buildings, f, ensure_ascii=False, indent=4)

print("处理完成，结果已保存至:", output_path)
```
