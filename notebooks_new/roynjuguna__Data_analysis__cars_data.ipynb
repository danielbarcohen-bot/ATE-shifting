# Cars dataset

Here the data of different cars is given with their specifications.
This data is available in a csv file. We are going to analyze using pandas dataframe.

```python
import pandas as pd
```

```python
cars = pd.read_csv(r"cars_Data.csv")
```

```python
cars.head()
```

```python
cars.shape
```

1. Instruction (For data cleansing)
   + Find all null values in the data set. If there is any null value, fill it with the mean of that column.


```python
cars.isnull().sum()
```

```python
cars["EngineSize"].fillna(cars["EngineSize"].mean(), inplace= True)
cars["Cylinders"].fillna(cars["Cylinders"].mean(), inplace= True)
cars["Horsepower"].fillna(cars["Horsepower"].mean(), inplace= True)
cars["MPG_City"].fillna(cars["MPG_City"].mean(), inplace= True)
cars["MPG_Highway"].fillna(cars["MPG_Highway"].mean(), inplace= True)
cars["Weight"].fillna(cars["Weight"].mean(), inplace= True)
cars["Wheelbase"].fillna(cars["Wheelbase"].mean(), inplace= True)
cars["Length"].fillna(cars["Length"].mean(), inplace= True)
```

```python
cars.dropna(subset=['Make'], inplace=True)
```

```python
cars.isnull().sum()
```

   + Check different types of makes in our data set and the count of each make in the data set

```python
cars["Make"].value_counts()
```

   + Remove all the records( rows) where weight is greater than 4000

5. Instruction (Applying a function on a column).
   + Increase all the values of MPG_City by 3.

```python
cars["MPG_City"] = cars["MPG_City"].apply(lambda x:x+3)
```
