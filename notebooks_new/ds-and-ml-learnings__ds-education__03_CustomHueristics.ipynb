```python
# Creating a title column

def create_title(df):
    title = []
    for passenger_index, passenger in df.iterrows():
        name = passenger['Name']

        if (name.find('Miss.') > 0):
            title.append('Miss.')
        elif (name.find('Master.') > 0):
            title.append('Master.')
        elif (name.find('Mrs.') > 0):
            title.append('Mrs.')
        elif (name.find('Mr.') > 0):
            title.append('Mr.')
        else:
            title.append('Other')
    
    return title
```
