# Evitar n?meros m?gicos
Ao ler c?digo, ?s vezes encontramos n?meros no c?digo onde n?o ? ?bvio o que eles fazem. No entanto, quando os alteramos, de repente nosso programa deixa de funcionar. Chamamos esses n?meros de [n?meros m?gicos](https://en.wikipedia.org/wiki/Magic_number_(programming)). Por exemplo, voc? sabe o que o `3` e o `7` no c?digo abaixo fazem?

```python
from skimage.io import imread
from skimage.filters import gaussian, threshold_otsu
from skimage.measure import label
```

```python
image = imread("../../data/BBBC007_batch/17P1_POS0013_D_1UL.tif")

# noise removal
blurred = gaussian(image, 3)

# instance segmentation
binary = blurred > threshold_otsu(blurred)
labels = label(binary)

# quantitative measurement
labels.max()
```

```python
image = imread("../../data/BBBC007_batch/17P1_POS0013_D_1UL.tif")

# noise removal
blurred = gaussian(image, 7)

# instance segmentation
binary = blurred > threshold_otsu(blurred)
labels = label(binary)

# quantitative measurement
labels.max()
```

## Uma se??o de configura??o no in?cio
Para evitar n?meros m?gicos e tornar o c?digo mais leg?vel, ? recomend?vel ter uma se??o de c?digo de configura??o no in?cio de cada script / notebook. L? voc? tamb?m pode usar coment?rios para explicar o que as vari?veis significam. Ali?s, dar bons nomes a essas vari?veis ? fundamental.

```python
# entre aqui o nome do arquivo de imagem a ser processado
file_to_process = "../../data/BBBC007_batch/17P1_POS0013_D_1UL.tif"

# entre aqui o raio esperado dos n?cleos, em unidades de pixel
approximate_nuclei_radius = 3
```

```python
image = imread(file_to_process)

# noise removal
blurred = gaussian(image, approximate_nuclei_radius)

# instance segmentation
binary = blurred > threshold_otsu(blurred)
labels = label(binary)

# quantitative measurement
labels.max()
```
