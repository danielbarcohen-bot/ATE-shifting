### Sliding Window

**Sliding Window Technique:**

* The sliding window technique involves using past data points as inputs to predict future data points. By shifting the data by a fixed number of days (10 days in this case), you create a window where the closing price of today (Close) is used to predict the closing price 10 days later (Close in 10 changes)

* below tail, 68.787498 used to predict 72.879997

* 69.964996 for testing part

```python
df["Close in 10 changes"]  = df["Close"].shift(-10) # Adding new 'Colse in 10 changes' column to same data frame 
```

```python
df.tail(20)
```

```python
dfc = df.iloc[:-10, :]   # Create new data frame that contain all data without last 'NaN' 10 empty data rows of 
dfc.head()               # 'Close in 10 changes' column
```

```python
dfc.tail() # 'dfc' haven't 'Nan' rows, data have column that relevnt to 'Nan' take for testing
```

### Machine Learning

```python
print(dfc.shape)
print(dfc.shape[0])
```

```python
dfc.iloc[:, 1].values
```

```python
x = dfc.iloc[:, 0].values.reshape(dfc.shape[0], 1)   # 'Close' column
y = dfc.iloc[:, 1].values   # 'Close in 10 changes' column
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
rf = RandomForestRegressor(n_estimators=500)
rf.fit(x_train, y_train)
y_pred = rf.predict(x_test)
np.sqrt(mean_squared_error(y_test, y_pred))
```

### Forcasting

```python
y_actual_10 = dfc.iloc[-10: , :]["Close in 10 changes"].values
y_actual_10
```

```python
y_pred_10 = rf.predict(dfc.iloc[-10 : , : ]["Close"].values.reshape(10, 1))
```

```python
y_pred_10
```

```python
plt.plot(range(1, 11), y_pred_10)
plt.plot(range(1, 11), y_actual_10)
plt.legend(["predicted", "Actual"])
plt.title("Last 10 days predictions")
plt.show()
```
