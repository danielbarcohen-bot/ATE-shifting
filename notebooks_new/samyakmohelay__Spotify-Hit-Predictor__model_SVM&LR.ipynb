```python
ds00 = pd.read_csv('Original_data/dataset-of-00s.csv')
ds10 = pd.read_csv('Original_data/dataset-of-10s.csv')
ds60 = pd.read_csv('Original_data/dataset-of-60s.csv')
ds70 = pd.read_csv('Original_data/dataset-of-70s.csv')
ds80 = pd.read_csv('Original_data/dataset-of-80s.csv')
ds90 = pd.read_csv('Original_data/dataset-of-90s.csv')
```

```python
# Combining all data into one dataframe
merge = [ds00, ds10, ds60, ds70, ds80, ds90]
data = pd.concat(merge)
data.reset_index(drop=True, inplace=True)
```

```python
data
```

```python
data2 = data.copy()
data2.loudness = (data2.loudness +50)/max(data2.loudness +50)
data2.duration_ms = (data2.duration_ms / 1000)/max(data2.duration_ms / 1000)
data2.key = (data2.key)/max(data2.key)
data2.tempo = (data2.tempo)/max(data2.tempo)
data2.time_signature = (data2.time_signature)/max(data2.time_signature)
data2.chorus_hit = (data2.chorus_hit)/max(data2.chorus_hit)
data2.sections = (data2.sections)/max(data2.sections)

data2
```

```python
data2.columns
```

```python
y = data2.target
x = data2[['danceability', 'energy', 'key', 'loudness','mode', 'speechiness', 'acousticness', 'instrumentalness', 'liveness','valence', 'tempo', 'duration_ms', 'time_signature', 'chorus_hit','sections']]
```

```python
print(x.shape, y.shape)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=12)
```

## With SVM

```python
from sklearn.svm import SVC 

model = SVC(kernel='poly',verbose=True, random_state=20)
model.fit(x_train, y_train)
# predictions = model.predict(x_test)
```

```python
print(f"Training Data Score: {model.score(x_train, y_train)}")
print(f"Testing Data Score: {model.score(x_test, y_test)}")
```
