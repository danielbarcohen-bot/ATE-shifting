# Importing the libraries for data preprocessing

```python
import numpy as np
import pandas as pd
```

```python
#Importing Dataset
df=pd.read_csv('data/Thyroid_Diff.csv') 
```

```python
df.shape
# dataset has 383 rows and 17 columns
```

```python
df.info()
# Info about Data
```

```python
#Renaming the columns into understandable context
df=df.rename(columns={'T':'Tumor Classification', 'N': 'Nodal classification', 'M':'Metastasis classification', 'Hx Smoking': 'Smoking History'
, 'Hx Radiothreapy': 'Radiotheraphy History'})
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
## duplicated shows that there are 19 duplicates in the data
```

```python
df=df.drop_duplicates()
#dropping duplicate rows in the dataset
```

```python
df.shape
#Now we have 364 rows and 17 columns in the dataset
```

```python
df = df.reset_index(drop=True)
#Resetting the indexes
```

```python
df.nunique()
#A view into the unique features of each column for a stronger exploration of the features
```

**Encoding Categorical Data**

```python
df.head()
```

```python
# One Hot Encoding
columns_onehot = df[['Thyroid Function','Physical Examination','Adenopathy', 'Pathology','Response']]
# Label Encoding
columns_labelencoders = df[['Gender','Smoking','Smoking History','Radiotheraphy History','Focality','Risk','Tumor Classification','Nodal classification','Metastasis classification','Stage','Recurred']]
```

```python
columns_onehot
```

```python
columns_labelencoders 
```

**Importing the libraries for encoding categorical data**


```python
from sklearn.preprocessing import OneHotEncoder, LabelEncoder
#Create the encoders - OneHotencoding for Nominal data and LabelEncoder for Ordinal and Binay Class Data
encoder = OneHotEncoder(sparse_output = False, drop='first')
le = LabelEncoder()
```

```python
#Appling One Hot Encoding
one_hot_encoded = pd.DataFrame(encoder.fit_transform(columns_onehot),
                               columns=encoder.get_feature_names_out(columns_onehot.columns))
```

```python
one_hot_encoded #A new onehotencoded features
```

```python
#Applying Label Encoding
label_encoded = columns_labelencoders.apply(le.fit_transform)
```

```python
label_encoded
```

```python
#Concatenating it
df = pd.concat([df['Age'],label_encoded,one_hot_encoded],axis=1)
```

```python
df.head()
```

```python
df.info()
```

**Creating Dependent and Independent Variable**

```python
#Create the independent variable into X and dependent variable into Y
X = df.drop(columns='Recurred')
y = df['Recurred']
```

**Splitting the dataset into training and test sets**

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.25, random_state=1)
```

```python
#importing the machine learning model
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter = 1000)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
## Importing Metrics
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='binary') 
recall = recall_score(y_test, y_pred, average='binary')
f1 = f1_score(y_test, y_pred, average='binary')
```

```python
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
```
