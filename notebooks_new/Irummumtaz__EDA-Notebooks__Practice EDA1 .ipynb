
?	How many clinical trials are in an "Completed" and "Recruiting" status? 
?	How many clinical trials are related to 'COVID-19'? 
?	How many clinical trials started after January 1, 2020?


```python
#?	How many studies have an enrollment greater than a certain threshold?
Enrol_Threshold=cd['Enrollment'].median()
TotalEnrolment=cd['Enrollment']>Enrol_Threshold
TStudies=TotalEnrolment.sum()
print('Studies have an enrollment greater than  Median ',TStudies)
```

```python
#?	How many clinical trials have 'No Results Available'? 
Result=cd['Study Results']
count=0
for i in Result:
    cd['Study Results']=='No Results Available'
    count+=1
print(count)
```

```python
len(cd['Study Results']=='No Results Available')
```

```python
#?	How many clinical trials are in an "Completed" and "Recruiting" status? 
clinicaltrailscount=cd.loc[cd['Status'].isin(['Completed','Recruiting'])]

len(clinicaltrailscount)
```

```python
#?	How many clinical trials are related to 'COVID-19'? 
covidtrail=cd.loc[cd['Conditions']=='COVID-19']
len(covidtrail)
```

```python
#?	How many clinical trials started after January 1, 2020?
cd['Start Date']=pd.to_datetime(cd['Start Date'],format='mixed')
tr=cd[cd['Start Date']>='2020-1-1']
len(tr)
```
