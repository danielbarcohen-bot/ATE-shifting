### Machine learning

De agora em diante, vamos explorar recursos de machine learning para tentar entender as rela??es entre as vari?veis a classe das esp?cies.

16) Salvando os valores a serem processados nas analises. A biblioteca scikit-learn usa numpy arrays.

```python
array = iris.values
print(type(array))
```

```python
print(array.shape)
print(array)
```

16) Isolando Iris-setosa

```python
raw_target = array[:,4]
print(raw_target)
y = np.where(raw_target == 'Iris-setosa', 1, 0)
print(y)
```

```python
X = array[:,0:4]
X.shape
```

17) Algoritmos de machine learning normalmente exigem que se tenha um conjunto de dados para "treinar" o modelo e outro para testar. Isso ser? feito abaixo. O par?metro "Seed" indica o ponto de partida da "aleatoridade" da escolha. Ao fix?-lo garantimos que a escolha ser? sempre a mesma.

```python
test_size = 0.20
seed = 7
X_train, X_test, y_train, y_test = \
    model_selection.train_test_split(X, y, test_size=test_size, random_state=seed)
    
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
print(X_train[:10])
print(X_test[:10])
print(y_train[:10])
print(y_test[:30])
```

18) Criando o modelo, validando o resultado

```python
model = LogisticRegression()
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
cm = confusion_matrix(y_pred,y_test)
print(cm)
```

19) Iris-virinica e Iris-versicolor se sobrep?em parcialmente. Vamos tentar modelar a Iris-virginica.

```python
raw_target = array[:,4]
y = np.where(raw_target == 'Iris-virginica', 1, 0)
print(raw_target)
print(y)
```

```python
X = array[:,0:4]
test_size = 0.20
seed = 7
X_train, X_test, y_train, y_test = \
    model_selection.train_test_split(X, y, test_size=test_size, random_state=seed)
model = LogisticRegression()
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
cm = confusion_matrix(y_pred,y_test)
print(cm)
```

20) E agora a Iris-versicolor

```python
raw_target = array[:,4]
y = np.where(raw_target == 'Iris-versicolor', 1, 0)
print (y)
X = array[:,0:4]
test_size = 0.20
seed = 7
X_train, X_test, y_train, y_test = \
    model_selection.train_test_split(X, y, test_size=test_size, random_state=seed)
model = LogisticRegression()
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
cm = confusion_matrix(y_pred,y_test)
print(cm)
```

Talvez a previs?o ruim das ?ltimas tentativas deva-se ao fato do modelo tentar separar os grupos atrav?s de uma linha reta, o que funcionaria para Iris-setosa mas n?o para as outras esp?cies.

21) Vamos tentar um outro modelo:

```python
X = array[:,0:4]
raw_target = array[:,4]
y = np.where(raw_target == 'Iris-versicolor', 1, 0)
test_size = 0.20
seed = 7
X_train, X_test, y_train, y_test = \
    model_selection.train_test_split(X, y, test_size=test_size, random_state=seed)
model = KNeighborsClassifier(n_neighbors=15, weights='distance')
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
cm = confusion_matrix(y_pred,y_test)
print(cm)
```

Muito melhor, n?? 

22) Talvez o modelo log?stico tenha falhado porque a classe Iris-setosa tenha "empurrado" as outras classes uma em dire??o ? outra. Vamos remover a Iris-setosa e tentar criar um novo modelo log?stico.

```python
small_array = array[array[:,4]!='Iris-setosa']
print('Shape of small_array: ', small_array.shape)
print(small_array)
```

23) Aplicando novamente o modelo linear, agora com uma amostra de treinamento maior, porque s?o menos dados.

```python
X = small_array[:,0:4]
raw_target = small_array[:,4]
y = np.where(raw_target == 'Iris-versicolor', 1, 0)

test_size = 0.30

seed = 7
X_train, X_test, y_train, y_test = \
    model_selection.train_test_split(X, y, test_size=test_size, random_state=seed)
model = LogisticRegression()
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
cm = confusion_matrix(y_pred,y_test)
print(cm)
```

24) Tentando um outro modelo n?o linear

```python
from sklearn.svm import SVC

raw_target = array[:,4]
y = np.where(raw_target == 'Iris-versicolor', 1, 0)
print(y)
X = array[:,0:4]
test_size = 0.20
seed = 7
X_train, X_test, y_train, y_test = \
    model_selection.train_test_split(X, y, test_size=test_size, random_state=seed)
model = SVC()
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
cm = confusion_matrix(y_pred,y_test)
print(cm)
```
