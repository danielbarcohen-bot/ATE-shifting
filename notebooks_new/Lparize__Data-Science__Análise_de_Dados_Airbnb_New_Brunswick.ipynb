### **Q3. Qual o tipo de distribui??o das vari?veis?** 
Filtros para as loca??es de at? 600 d?lares e at? 10 dias de disponibilidade para loca??o.

```
# plotar o histograma das vari?veis num?ricas
# Foi excluido o gr?fico License, pois estava sem valor.
df_clean = df.copy()
df_clean.drop(df_clean[df_clean.price > 600].index, axis=0, inplace=True)
df_clean.drop(df_clean[df_clean.minimum_nights > 10].index, axis=0, inplace=True)

df_clean.drop('license', axis=1, inplace=True)
df_clean.hist(bins=15, figsize=(15,10));
```

### **Q4. Qual a m?dia dos pre?os de aluguel?**

```
# ver a m?dia da coluna `price``
media = df["price"].mean()

print(media)
```

### **Q4. Qual a correla??o existente entre as vari?veis**

```
# criar uma matriz de correla??o
corr = df_clean[['price', 'minimum_nights', 'number_of_reviews', 'reviews_per_month', 
                 'calculated_host_listings_count', 'availability_365']].corr()
# mostrar a matriz de correla??o
display(corr)
```

```
# plotar um heatmap a partir das correla??es
sns.heatmap(corr, cmap='RdBu', fmt='.2f', square=True, linecolor='white', annot=True);
```
