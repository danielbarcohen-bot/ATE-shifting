```python
import pandas as pd

# 1. Load dataset
df = pd.read_csv("student_performance_data.csv")


# ---------- Data Cleaning Steps----------------
# 2. View basic info
print("First 3  rows:\n", df.head(3))
print("\nData Info:\n")
print(df.info())

# 3. Check missing values
print("\nMissing values:\n", df.isnull().sum())

# Handle missing values (example: fill or drop)
df = df.dropna()   # or use df.fillna(method='ffill')

# 4. Remove duplicates
df = df.drop_duplicates()
```

```python
# 5 detect outliers and remove it
from scipy import stats
import numpy as np

num_cols = ['Age', 'StudyHoursPerWeek', 'AttendanceRate', 'GPA']

# Apply Z-score only on numeric columns
z_scores = np.abs(stats.zscore(df[num_cols]))

outliers = df[(z_scores > 3).all(axis=1)]

print("This all are outliers : ",outliers)

# Keep rows where all z-scores < 3
df_no_outliers = df[(z_scores < 3).all(axis=1)]

print("After Z-score cleaning:", df_no_outliers.shape)


# 5: Handle Duplicates
# Count duplicate rows
print(df.duplicated().sum())
# Drop duplicates
df = df.drop_duplicates()
# 6: check data Inconsistencies 
print("Major : ",df['Major'].unique())
print("",df['Gender'].unique())
print("",df['PartTimeJob'].unique())
print("",df['ExtraCurricularActivities'].unique())

# 7: Correct Data Inconsistencies


# 1. Select all categorical columns
cat_cols = df.select_dtypes(include=['object']).columns

# 2. Standardize text (lowercase + strip spaces)
for col in cat_cols:
    df[col] = df[col].str.lower().str.strip()

# 3. Check inconsistencies: print unique values for each categorical column
for col in cat_cols:
    print(f"\nColumn: {col}")
    print(df[col].value_counts())

# 4. Replace inconsistent values manually according to what you see
# Example: adjust based on your dataset output
# df['Department'].replace({'hr': 'human resources', 'it dept': 'it'}, inplace=True)
# df['Gender'].replace({'m': 'male', 'f': 'female'}, inplace=True)
    
```
