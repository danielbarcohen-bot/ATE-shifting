Matching the column names in both datasets to combine the data

```python
df1_renamed = df1.rename(columns={
    "Number": "Question Number",
    "Questions": "Question",
    "Answers": "Answer",
    "language": "Category",
    " level ": "Difficulty"  
})
```

```python
df1_reordered = df1_renamed[["Question Number", "Question", "Answer", "Category", "Difficulty"]]
```

```python
df1_reordered.info()
```

Creating a combined dataset

```python
combined_df = pd.concat([df2, df1_reordered], ignore_index=True)
```
