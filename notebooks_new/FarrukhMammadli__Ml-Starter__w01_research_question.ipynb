## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import numpy as np
import pandas as pd
data_url = "https://raw.githubusercontent.com/FarrukhMammadli/Ml-Starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(data_url)

corr_matrix = df[
    [
        "avg_position",
        "days_since_last_update",
        "content_age_days",
        "ctr",
        "word_count",
    ]
].corr()
pos_corr = corr_matrix["avg_position"].drop("avg_position")

top10_stale_avg = df[df["avg_position"] <= 10][
    "days_since_last_update"
].mean()
others_stale_avg = df[df["avg_position"] > 10]["days_since_last_update"].mean()

high_ctr_pos = df[df["ctr"] > df["ctr"].median()]["avg_position"].mean()
low_ctr_pos = df[df["ctr"] <= df["ctr"].median()]["avg_position"].mean()

print("=== Correlation with avg_position ===")
print(pos_corr.round(3))
print(
    f"\nTop 10 rank pages average staleness: {top10_stale_avg:.1f} days"
)
print(f"Other pages average staleness: {others_stale_avg:.1f} days")
print(f"\nAverage position for High-CTR pages: {high_ctr_pos:.2f}")
print(f"Average position for Low-CTR pages: {low_ctr_pos:.2f}")

```
