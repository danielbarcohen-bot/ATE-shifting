```python
import sqlite3, sqlalchemy as sa, pandas
df = pandas.read_csv('books2.csv')
engine = sa.create_engine('sqlite://')
with engine.connect() as conn:
    conn.execute(sa.text('''CREATE TABLE books (title VARCHAR(20) PRIMARY KEY, author VARCHAR(20), year INT)'''))
    df.to_sql('books', con=engine, if_exists='append', index=False)
    
    rows = conn.execute(sa.text('SELECT title FROM books'))
    mylist = []
    for row in rows:
        mylist.append(row)
    mylist.sort()
    print(mylist)
```
