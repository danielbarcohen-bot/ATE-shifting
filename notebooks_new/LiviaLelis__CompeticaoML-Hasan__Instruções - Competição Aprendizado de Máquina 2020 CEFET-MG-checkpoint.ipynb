## Representa??o Bag of Words

Muitas vezes, temos textos que podem ser relevantes para uma determinada tarefa de aprendizado d m?quina. Por isso, temos que representar tais elementos para nosso m?todo de aprendizado de m?quina. 

A forma mais usual para isso, ? a `Bag of Words` em que cada palavra ? um atributo e, o valor dela, ? a frequencia dele no texto (ou algum outro valor que indique a importancia dessa palavra no texto).

Por exemplo, caso temos as frases `A casa ? grande`, `A casa ? verde verde` em que cada frase ? uma instancia diferente. A representa??o seria da seguinte forma: 

````python
dic_bow = {"a":[1,1],
         "casa":[1,1],
         "?":[1,1],
         "verde":[0,2]
        }
df_bow = pd.DataFrame.from_dict(dic_bow)
df_bow
````

Da forma que fizemos acima, usamos a frequencia de um termo para definir sua importancia no texto, por?m, existem termos que possuem uma frequencia muito alta e importancia baixa: s?o os casos dos artigos e preposi??es por exemplo, pois, eles n?o discriminam o texto. 

Uma forma de mensurar o porder discriminativo das palavras ? usando a m?trica `TF-IDF`. Para calcularmos essa m?trica, primeiramente calculamos a frequencia de um termo no documento (TF) e, logo ap?s multiplamos pelo IDF. 
A f?rmula para calcular o TF-IDF do termo $i$ no documento (ou instancia) $j$ ? a seguinte:

\begin{equation}
    TFIDF_{ij} = TF_{ij} \times IDF_i
\end{equation}
\begin{equation}
    TF_{ij} = log(f_{ij})
\end{equation}

em que $f_{ij}$ ? a frequencia de um termo $i$ no documento $j$. Usa-se o `log` para suavizar valores muito altos e o $IDF$ (do ingl?s, _Inverse Document Frequency_) do termo $i$ ? calculado da seguinte forma:

\begin{equation}
    IDF_i = log(\frac{N}{n_i})
\end{equation}

em que $N$ ? o n?mero de documentos da cole??o e $n_i$ ? o n?mero de documentos em que esse termo $i$ ocorre. Espera-se que, quanto mais discriminativo o termo, em menos documentos esse termo ir? ocorrer e, consequentemente, o $IDF$ deste termo ser? mais alto. 

Por exemplo, considere as palavras `de`, `bebida` e `cerveja`. `cerveja` ? uma palavra mais discriminativa do que `bebida`; e `bebibda` ? mais discriminativo do que a preposi??o `de`. Muito provavelmente teremos mais frequentemente termos menos discriminativos. Por exemplo, se tivermos uma cole??o de 1000 documentos,   `de` poderia ocorrer em 900 documentos,  `bebida` em 500 e `cerveja` em 100 documentos. Se fizermos o calculo, veremos que quanto mais discriminativo um termo, mais alto ? seu IDF:

````python
import math
N = 1000
n_de = 900
n_bebida = 500
n_cerveja = 100

IDF_de = math.log(N/n_de)
IDF_bebida = math.log(N/n_bebida)
IDF_cerveja = math.log(N/n_cerveja)

print(f"IDF_de: {IDF_de}\tIDF_bebida:{IDF_bebida}\tIDF_cerveja:{IDF_cerveja}")
````

## Como funciona o SVM?

Considere que temos acesso a um conjunto de treino $\{(\textbf{x}_1, \textbf{y}_1), (\textbf{x}_1, \textbf{y}_1), ...,  (\textbf{x}_n, \textbf{y}_1n)\}$ em que $\textbf{x}_i$ ? um vetor de valores num?ricos $x_{i,1}, x_{i,2}, ...., x_{i,m}$ representando os $m$ atributos da instancia $i$ e $y_i$ ? a classe (valor alvo) a ser predito desta mesma instancia. 

Dado o vetor de pesos $w_1, w_2, w_m$ e a fun??o linear na forma $f(x_{i,1}, x_{i,2}, ...., x_{i,m}) = w_1 x_{i,}$. Essa fun??o pode tamb?m ser representada como $f(x_i)$. Assim, a classifica??o do SVM ? dada por: 
\begin{equation}    
    g(x_i) = \begin{cases} 1 &\mbox{se } f(x_i) >= 0 \\ 
                         -1 & caso contr?rio 
  \end{cases} 
\end{equation}
em que 1 e -1 s?o os valores (r?tulos) das classes a serem preditas. Instancias da classe $-1$ s?o denominadas exemplos negativos e, da classe $+1$ exemplos possitivos. Podemos representar um exemplo de modelo constru?do pelo SVM usando o seguinte plano cartesiano:

Neste plano, o objetivo ? encontrar $\textbf{w}$ em que minimizamos o erro da fun??o $f(x_i)$ e, ao mesmo tempo maximizamos a margem. A margem ? dada pela a distancia do exemplo positivo mais pr?ximo a fun??o $f(x_i)$, representado por $d_+$, e pela distancia do exemplo negativo mais pr?ximo, representada por $d_-$.
