### Regular (M,g)

Dado a m?trica 
\begin{align}
ds^2 = -(1 - 2\mu(r)/r)dt^2 + (1 - 2\mu(r)/r)^{-1} dr^2 + r^2 d \Omega^2,
\end{align}
vamos supor
\begin{align}
\mu(r) = \mu e^{- \alpha r^2} r^3,
\end{align}
de modo que
\begin{align}
\rho(r) = \frac{\mu e^{- \alpha r^2}}{4 \pi}(3 - 2\alpha r^2) \implies \rho(r = 0) = \frac{\mu}{4\pi/3}.
\end{align}
Al?m disso,
\begin{align}
\rho(r) \ge 0 \iff r \le \sqrt{\frac{3}{2\alpha}} \ \ \ \text{ou} \ \ \ \alpha \le \frac{3}{2r^2}.
\end{align}
Por fim, notemos que
\begin{align}
ds^2 = -(1 - 2\mu r^2e^{- \alpha r^2})dt^2 + (1 - 2\mu r^2e^{- \alpha r^2})^{-1} dr^2 + r^2 d \Omega^2
\end{align}
? assintoticamente plano.

Al?m disso, ? poss?vel estender a coordenada radial no intervalo $r \in (- \infty, + \infty)$.

#### Equa??es de Campo

Dado as equa??es de campo na forma 
\begin{align}
G^{\mu}_{\ \nu} = 8 \pi T^{\mu}_{\ \nu}& = 8\pi\Big(\rho e_0^{\ \mu}e^0_{\ \nu} + p e_1^{\ \mu}e^1_{\ \nu} + p_{\Omega}(e_2^{\ \mu}e^2_{\ \nu} + e_3^{\ \mu}e^3_{\ \nu} ) \Big),
\end{align}
com $v^{\mu} = e_{0}^{\ \mu} = g^{-1/2}\delta^{\mu}_{\ t}$. Componente a componente, temos
\begin{align}
& G^{t}_{\ t} = - 8 \pi \rho \implies \rho(r) = \frac{\mu e^{- \alpha r^2}}{4 \pi}(3 - 2\alpha r^2)\\
& G^{r}_{\ r} =  8 \pi p \implies p(r) = - \rho(r) = - \frac{\mu e^{- \alpha r^2}}{4 \pi}(3 - 2\alpha r^2)\\
& G^{\theta}_{\ \theta} = 8 \pi p_{\Omega} \implies p_{\Omega} = \frac{\mu e^{- \alpha r^2}}{4\pi}(3 - \alpha r^2)(2 \alpha r^2 - 1).
\end{align}
A fim de que $\rho \ge 0$ ? necess?rio que $r \le \sqrt{3/2\alpha}$, e, portanto, $p = - \rho \le 0, \ \ \forall \ r \in [0, \sqrt{3/2\alpha}]$. Al?m disso, $p_{\Omega} \ge 0, \forall \ r \in [1/\sqrt{2\alpha},\sqrt{3/2\alpha}]$, enquanto que $p_{\Omega} \le 0, \forall \ r \in [0, 1/\sqrt{2\alpha}].$


#### Condi??es de Energia
\begin{align}
& \rho \ge 0, \forall r \le \sqrt{3/2\alpha}, \\
& \rho + p = 0, \forall r \le \sqrt{3/2\alpha}, \\
& \rho + p_{\Omega} = \frac{\mu}{4\pi} (5 - 2\alpha r^2)\alpha r^2 e^{\alpha r^2} \ge 0, \forall r \le \sqrt{3/2\alpha},\\
& \rho + p + 2p_{\Omega} = 2 p_{\Omega} = \frac{\mu e^{- \alpha r^2}}{2\pi}(3 - \alpha r^2)(2 \alpha r^2 - 1) \le 0, \forall r \le 1/\sqrt{2\alpha}. 
\end{align}

#### Condi??es de Jun??o

Se impormos que temos mat?ria apenas at? $r = \sqrt{3/2\alpha}$, definindo assim a superf?cie do objeto atrav?s de $p = \rho = 0$ (embora seja poss?vel estender tal solu??o at? infinito), ent?o temos que colar com $(M,g)_{sch}$. Tamb?mm como mostraremos abaixo, ? poss?vel supor que existe uma fronteira $r = a$ que define a superf?cie do objeto e aplicando as condi??es de jun??o surge naturalmente que tal superf?cie est? localizada em $r = a =  \sqrt{3/2\alpha}$ onde $p(a) = \rho(a) = 0$. Para isso, seja a m?trica interior denotada por $g^-_{\mu \nu}$ e a m?trica de Schwarzschild por $g^+_{\mu \nu}$. Assim, na fronteira, devemos ter que
\begin{align}
& g^-_{tt}|_{r = a} = g^{+}_{tt}|_{r = a}, \ \ \ g^-_{rr}|_{r = a} = g^{+}_{rr}|_{r = a} \implies \mu = \frac{m}{a^3} e^{\alpha a^2},\\
& K^-_{ab} = K^+_{ab} \implies a =  \sqrt{3/2\alpha}
\end{align}
onde $K_{ab} = \nabla_{\nu} n_{\mu} \frac{\partial x^{\mu}}{\partial \xi^a}\frac{\partial x^{\nu}}{\partial \xi^b}$ ? a curvatura extr?nseca da superf?cie $\Sigma : r = a$.
como esperado uma vez que a fronteira ? justamente suposta de inicial em $r = a =  \sqrt{3/2\alpha}$. Na superf?cie $\Sigma$ temos 
\begin{align}
ds^2_{\Sigma} = - d \tau^2 + a^2(d \theta^2 + \sin^2 \theta d \phi^2), \ \ \ \ \xi^a = (\tau, \theta, \phi).
\end{align}
Definindo o vetor normal ? $\Sigma$ como $n^{\mu} = g_{rr}^{-1/2} \delta^{\mu}_{\ r}$ tal que $g(n,n) = 1$, ent?o $n$ ? tipo espa?o e a superf?cie $\Sigma$ ? escolhida tipo tempo. Assim
\begin{align}
K_{a b} = g_{rr}^{1/2} \nabla_{\nu} \delta_{\mu}^{\ r} \frac{\partial x^{\mu}}{\partial \xi^a}\frac{\partial x^{\nu}}{\partial \xi^b} = -g_{rr}^{1/2} \Gamma_{\mu \nu}^{r} \frac{\partial x^{\mu}}{\partial \xi^a}\frac{\partial x^{\nu}}{\partial \xi^b},
\end{align}
de modo que
\begin{align}
K_{\tau \tau} =- g_{rr}^{1/2}  \Gamma_{\mu \nu}^{r} \frac{d x^{\mu}}{d \tau}\frac{d x^{\nu}}{d \tau} =  -g_{rr}^{1/2} \Gamma_{t t}^{r}  \frac{d t}{d \tau}\frac{d t}{d \tau} = \frac{1}{2} g_{rr}^{1/2} g^{r r}\partial_r g_{tt} \frac{d t}{d \tau}\frac{d t}{d \tau}.
\end{align}
Como $v^{\mu} :=\frac{d t}{d \tau} \delta^{\mu}_{\ t}$ representa a velocidade do flu?do com?vel (ou observadores) com a superf?cie, ent?o $$g^{rr}\frac{d t}{d \tau} \frac{d t}{d \tau}  = g_{tt}\frac{d t}{d \tau} \frac{d t}{d \tau}  = -1, $$ e portanto, temos que
\begin{align}
K_{\tau \tau} = -\frac{1}{2}g_{rr}^{1/2}\partial_r g_{tt}.
\end{align}

Ainda, lembrando que
\begin{align}
& ds_-^2 = -(1 - 2\mu r^2e^{- \alpha r^2})dt^2 + (1 - 2\mu r^2e^{- \alpha r^2})^{-1} dr^2 + r^2 d \Omega^2, \\
& ds_+^2 = -(1 - \frac{2m}{r})dt^2 + (1 - \frac{2m}{r})^{-1} dr^2 + r^2 d \Omega^2,
\end{align}
ent?o temos que
\begin{align}
 K^-_{ab}|_{r = a} = K^+_{ab}|_{r = a} \implies (g^-_{rr})^{1/2}\partial_r g^-_{tt} = (g^+_{rr})^{1/2}\partial_r g^+_{tt}   \implies \partial_r g^-_{tt}|_{r = a} = \partial_r g^+_{tt}|_{r = a} \implies a =  \sqrt{3/2\alpha}, 
\end{align}
de modo que $p(a) = \rho(a) = 0$. Ainda, vale ressaltar que temos 4 par?metros: $\mu, \alpha, m, a$. No entanto, apenas dois deles s?o independentes. Por exemplo, podemos escrever
\begin{align}

\end{align}
ou
\begin{align}
\mu = m\Big(\frac{3}{2\alpha}\Big)^{-3/2}e^{3/2}, \ \ \ \alpha = \frac{3}{2 a^2}.
\end{align}
Por exemplo, se escrevermos $m, a$ em termos de $\mu, \alpha$, ent?o
\begin{align}
g^{+}_{tt}(r) = 1 - \frac{2\mu}{r} \Big(\frac{3}{2\alpha}\Big)^{3/2}e^{-3/2}
\end{align}
sendo interessante notar que, para $\alpha = 1$ e $\mu = 3/2$, $g^{-}_{tt}(r)  = g^{+}_{tt}(r) $ coincidem em $r = a$
#### Horizontes

#### Condi??es para haver Horizontes

Como para uma m?trica esfericamente sim?trica e est?tica, o horizonte pode ser localizado atrav?s de $g_{tt}(r)|_{r = r_h} = 0$, para haver horizontes podemos olhar para o m?nimo de $g_{tt}$. Se o m?nimo de $g_{tt} < 0$ ent?o haver? horizontes, enquanto que se o m?nimo $g_{tt} > 0$ n?o haver? horizontes. Logo, como
\begin{align}

\end{align}
ent?o
\begin{align}
\frac{d}{dr} g_{tt} |_{r = r_*} = 0 \iff r_* = 1/\sqrt{\alpha}.
\end{align}
Neste caso, temos que
\begin{align}
g_{tt}(r_*) = 1 - 2\frac{\mu}{\alpha}e^{-1}.
\end{align}
Portanto,

#### Quasi BH?

From Lemos & Weinberg, Phys.Rev. D69 (2004) 104004: 
"To see how such a solution might come about, consider a static spacetime with a spherically symmetric concentration of matter near the origin, r = 0. If the spacetime is non-singular, then A(0) = 1. As one moves out from the origin, $A^{-1}$ initially decreases until it reaches a minimum value, typically at a radius near the edge of the mass distribution, and then increases towards its asymptotic value. The minimum of  $A^{-1}$ becomes deeper as the
density of the mass distribution is increased. This suggests that one could approach the black hole limit simply by making the density large enough."

Al?m disso, dado que 
\begin{align}
& ds_-^2 = -(1 - 2\mu r^2e^{- \alpha r^2})dt^2 + (1 - 2\mu r^2e^{- \alpha r^2})^{-1} dr^2 + r^2 d \Omega^2, \\
& ds_+^2 = -(1 - \frac{2m}{r})dt^2 + (1 - \frac{2m}{r})^{-1} dr^2 + r^2 d \Omega^2,
\end{align}
em $\Sigma: r = a$, devemos ter
\begin{align}
-(1 - 2\mu r^2e^{- \alpha r^2})dT^2 = -(1 - \frac{2m}{r})dt^2
\end{align}
de modo que
\begin{align}
dT^2 = \frac{1 - \frac{2m}{r}}{1 - 2\mu r^2e^{- \alpha r^2}} dt^2 \implies dT^2 \to 0, \ \ \ \text{para} \ \ \ r \to 2m.
\end{align}

A partir daqui podemos 'deduzir' a m?trica suposta acima. Dado que
\begin{align}
G^{t}_{\ t} = 8 \pi T^{t}_{\ t} \implies  -\frac{2}{r^2} \frac{d}{dr}\mu(r) = - 8 \pi \rho(r)
\end{align}
temos que
\begin{align}
\mu(r) = 4\pi \int_0^r \rho(r) r^2 dr.
\end{align}
Supondo que
\begin{align}
\rho(r) = \frac{\mu e^{- \alpha r^2}}{4 \pi}(3 - 2\alpha r^2), \ \ \ \alpha,\mu > 0
\end{align}
obtemos
\begin{align}
\mu(r) = \mu e^{- \alpha r^2} r^3,
\end{align}
e, portanto,
\begin{align}
ds^2 = -(1 - 2\mu r^2e^{- \alpha r^2})dt^2 + (1 - 2\mu r^2e^{- \alpha r^2})^{-1} dr^2 + r^2 d \Omega^2.
\end{align}

#### Botando o tro?o pra girar


Seguindo o procedimento de Mustapha, Phys. Rev. D 90, 064041 (2014), ou seja, usando o algoritmo de Newman-Janis sem complexifica??o ? poss?vel encontrar uma m?trica estacion?ria a partir da m?trica original
\begin{align}
ds^2 = -(1 - 2\mu r^2e^{- \alpha r^2})dt^2 + (1 - 2\mu r^2e^{- \alpha r^2})^{-1} dr^2 + r^2 d \Omega^2,
\end{align}
de tal modo que a transforma??o de coordenadas EF para coordenadas BL s?o integr?veis. Neste caso, temos
\begin{align}
du = dt + \frac{r^2 + a^2}{r^2 + a^2 - 2\mu r^4e^{\alpha r^2}} dr, \ \ \ d \varphi = d \phi - \frac{a}{r^2 + a^2 - 2\mu r^4e^{\alpha r^2}}dr,
\end{align}
de modo que os elementos da nova m?trica s?o dados por
\begin{align}
& g_{tt} = -\Big(1 - \frac{2\mu r^4e^{-\alpha r^2}}{\rho^2}\Big), \ \ \ g_{rr} = \frac{r^2 + a^2 \cos^2 \theta}{r^2 + a^2 - 2\mu r^4e^{-\alpha r^2}}, \ \ \ g_{\theta \theta} = \rho^2 := r^2 + a^2 \cos^2 \theta, \\
& g_{t \phi} = -\frac{2\mu ar^4e^{-\alpha r^2}\sin^2 \theta}{\rho^2}, \ \ \, g_{\phi \phi} = \Big(r^2 + a^2 + \frac{2\mu r^4e^{-\alpha r^2}a^2 \sin^2 \theta}{\rho^2} \Big)\sin^2 \theta. 
\end{align}
Como $\lim_{r \to \infty} r^4 e^{-\alpha r^2} = 0$, ent?o ? poss?vel ver que a m?trica acima ? assintoticamente plana, se reduzindo a Minkowski em coordenada esferoidais e tendo uma forma bem parecida com a m?trica de Kerr. Para obter Kerr, basta fazer fazer a substitui??o $r^4e^{-\alpha r^2} \to r$.  Ou ainda, podemos ver que tal m?trica possui a mesma forma que m?trica de Kerr se escrevermos
\begin{align}
& g_{tt} = -\Big(1 - \frac{2\mu(r)r}{\rho^2}\Big), \ \ \ g_{rr} = \frac{r^2 + a^2 \cos^2 \theta}{r^2 + a^2 - 2\mu(r)r}, \ \ \ g_{\theta \theta} = \rho^2 := r^2 + a^2 \cos^2 \theta, \\

#### Condi??es de Jun??o

Supondo que existe uma fronteira $r = a$ que define a superf?cie do objeto, seja a m?trica interior denotada por $g^-_{\mu \nu}$ e a m?trica de Kerr por $g^+_{\mu \nu}$. Ou seja, temos que
\begin{align}

\end{align}
