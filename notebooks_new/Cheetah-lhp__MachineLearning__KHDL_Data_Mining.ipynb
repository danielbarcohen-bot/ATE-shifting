````python
# B??c 3.3.1: Ph?n b? c?c bi?n ph?n lo?i (Categorical Variables)
# Step 3.3.1: Distribution of Categorical Variables
print("PH?N T?CH PH?N B? C?C BI?N PH?N LO?I")  # Print header / In ti?u ??
print("="*50)                                  # Decorative line / D?ng ph?n c?ch

# T?o subplot cho 4 bi?n categorical / Create subplots for 4 categorical variables
fig, axes = plt.subplots(2,2,figsize=(15,10))   # TODO: t?o figure v? axes (g?i ?: subplots(2,2,figsize=(15,10)))
# Suggestions:
# - plt.subplots(2,2) ? 4 subplots
# - figsize=(w,h) ? set figure size

fig.suptitle('Some Title', fontsize=16, fontweight='bold')
# TODO: nh?p ti?u ?? ch?nh cho figure / main title for figure

# Danh s?ch c?c bi?n c?n v? / List of columns to plot
categorical_cols = ["Sex","BP","Cholesterol","Drug"]   # TODO: li?t k? t?n c?c c?t ph?n lo?i (v? d?: ["Sex","BP","Cholesterol","Drug"])
colors = ['skyblue', 'red', 'blue', 'lightcoral']       # TODO: ch?n danh s?ch m?u (g?i ?: ['skyblue','lightcoral',...])

# V? t?ng bi?u ?? / Plot each chart
for i, col in enumerate(categorical_cols):
    row, col_idx = i // 2, i % 2
    ax = axes[row, col_idx]

    # ??m t?n su?t / Count frequency
    value_counts = my_data[col].value_counts()    # TODO: ??m s? l??ng c?c gi? tr?
    # Suggestions:
    # - value_counts() ? count each category
    # - nunique() ? number of unique categories (less used here)

    # V? bi?u ?? c?t / Draw bar chart
    sns.countplot(data=my_data, x=col, ax=ax, color=colors[i], order=value_counts.index)
    # TODO: ch?n h?m v? (g?i ?: countplot)

    # Thi?t l?p ti?u ?? v? labels / Set title and labels
    ax.set_title(f'Ph?n b? {col}', fontweight='bold', fontsize=12)
    ax.set_xlabel("X", fontweight='bold')   # TODO: g?n nh?n tr?c X / X-axis label
    ax.set_ylabel("Y", fontweight='bold')   # TODO: g?n nh?n tr?c Y / Y-axis label

    # Th?m s? li?u tr?n m?i c?t / Annotate bars with values
    for p in ax.patches:
        ax.annotate(f'{int(p.get_height())}',
                   (p.get_x() + p.get_width()/2., p.get_height()),
                   ha='center', va='bottom', fontweight='bold')

    # In th?ng k? summary / Print frequency summary
    print(f"\n{col.upper()}:")
    for category, count in value_counts.items():
        percentage = (count / len(my_data)) * 100
        print(f"  ? {category}: {count} b?nh nh?n ({percentage:.1f}%)")  # Patients count and percentage

plt.tight_layout()    # TODO: c?n ch?nh layout (g?i ?: tight_layout()) / adjust layout
plt.show()    # TODO: hi?n th? bi?u ?? (g?i ?: show()) / display figure

# T?ng k?t insights / Summarize key insights
print(f"\nKEY INSIGHTS:")

print(f"? Gi?i t?nh: TODO")
# G?i ? / Suggestion:
# - D?ng value_counts() ?? so s?nh nam vs n? / use value_counts() to compare M vs F
# - ?i?u ki?n if ? else ?? in ?C?n b?ng? ho?c ?Kh?ng c?n b?ng? / if-else for "Balanced" or "Unbalanced"

print(f"? Huy?t ?p: TODO")
# G?i ?: l?y mode ho?c gi? tr? t?n su?t cao nh?t / take mode or value_counts().index[0]

print(f"? Cholesterol: TODO")
# G?i ?: t??ng t? nh? tr?n / same as above

print(f"? Drug distribution: TODO")
# G?i ?: value_counts().index[0] ? lo?i thu?c ph? bi?n nh?t / most prescribed drug
````
