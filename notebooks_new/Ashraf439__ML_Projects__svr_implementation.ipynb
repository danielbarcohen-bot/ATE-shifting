```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
```

```python
df.info()
```

We're predicting the total_bill

```python
X = df[['day','sex','tip','time','size','smoker']]
y = df['total_bill']
```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
l1 = LabelEncoder()
l2 = LabelEncoder()
l3 = LabelEncoder()
```

```python
X_train['sex'] = l1.fit_transform(X_train['sex'])
X_train['time'] = l2.fit_transform(X_train['time'])
X_train['smoker'] = l3.fit_transform(X_train['smoker'])
```

```python
X_train.head()
```

```python
X_train['day'].value_counts()
```

```python
X_test['sex'] = l1.fit_transform(X_test['sex'])
X_test['time'] = l2.fit_transform(X_test['time'])
X_test['smoker'] = l3.fit_transform(X_test['smoker'])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct = ColumnTransformer(transformers=[("onehot",OneHotEncoder(drop='first'),[0])],remainder="passthrough")
```

```python
import sys
np.set_printoptions(threshold=sys.maxsize)
```

```python
X_train = ct.fit_transform(X_train)
X_test = ct.transform(X_test)
```

```python
X_test
```

```python
from sklearn.svm import SVR
```

```python
svr = SVR()
```

```python
svr.fit(X_train,y_train)
```

```python
y_pred = svr.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
r2 = r2_score(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
```

```python
print(f"{r2}\n{mae}")
```

```python
from sklearn.model_selection import GridSearchCV

paramsgrid = {
    'C':[0.1,10,100,1000],
    'gamma':[1,0.1,0.01,0.001,0.0001],
    'kernel':['rbf']
}
```

```python
grid = GridSearchCV(svr,param_grid=paramsgrid,refit=True,verbose=3)
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred1 = grid.predict(X_test)
```

```python
r2_score(y_test,y_pred1)
```

```python
mean_absolute_error(y_test,y_pred1)
```
