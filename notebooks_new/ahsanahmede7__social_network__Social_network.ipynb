```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer,make_column_selector
from sklearn.metrics import r2_score
from sklearn.preprocessing  import StandardScaler,OneHotEncoder,PolynomialFeatures
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.impute import SimpleImputer
import numpy as np


```

```python
data = pd.read_csv("Social_Network_Ads.csv")
data.head()
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
x = data.drop("Purchased",axis=1)
y = data.Purchased
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.20)
```

```python
model = LogisticRegression()
```

```python
num_pipeline = Pipeline([
    ('handlemissing',SimpleImputer(strategy="median")),
    ("Scaler",StandardScaler())
])
cat_pipeline = Pipeline([
    ("hanldeMissing",SimpleImputer(strategy="most_frequent")),
    ("Encode",OneHotEncoder(drop="first",sparse_output=False,handle_unknown="ignore"))
])
```

```python
Column_transfer = ColumnTransformer([
    ("nums",num_pipeline,make_column_selector(dtype_include=np.number)),
    ("Columns",cat_pipeline,make_column_selector(dtype_include=object))
])
```

```python
model_fit = Pipeline([
    ('Preprosser',Column_transfer),
    ("poly",PolynomialFeatures(degree=2,include_bias=False)),
    ("model",model)
])
```

```python
model_fit.fit(X_train,y_train)
```

```python
y_pred = model_fit.predict(X_train)
```

```python
from sklearn.metrics import accuracy_score,classification_report,recall_score,precision_score,f1_score,confusion_matrix
```

```python
accuracy_score(y_train,y_pred)
```

```python
print(classification_report(y_train,y_pred))
```
