```python
from utils import  calculate_ate_dml, calculate_ate_dr
from experiments import largest_data_transformations
from utils import apply_data_preparations_seq
from data_loader import TwinsDataLoader, LalondeDataLoader, ACSDataLoader, IHDPDataLoader, WalmartDataLoader
import pandas as pd
```

```python
df_lalonde = LalondeDataLoader().load_data()
df_acs = ACSDataLoader().load_data()
df_walmart = WalmartDataLoader().load_data()
df_ihdp = IHDPDataLoader().load_data()
df_twins = TwinsDataLoader().load_data().dropna()
```

```python
# =========================
# Prepare datasets
# =========================

df_acs_copy = df_acs.copy()
df_acs_copy['treatment'] = df_acs_copy['treatment'] - 1


# =========================
# Data preparation sequences
# =========================

sequence_lalonde = (('IQR', 'education'), ('zscore_filter_3', 'age'), ('bin_equal_frequency_5', 'age'), ('drop_duplicates', 'TABLE'), ('bin_equal_frequency_2', 'education'), ('isolationForest', 'TABLE'))

sequence_acs = (('bin_equal_frequency_5', 'education'), ('norm_log', 'education'), ('isolationForest', 'TABLE'))

sequence_walmart = (('zscore_clip_3', 'sqft_basement'), ('bin_equal_frequency_10', 'sqft_living'), ('IQR', 'sqft_basement'), ('IQR', 'bedrooms'), ('bin_equal_width_5', 'sqft_living15'), ('bin_equal_frequency_5', 'yr_built'))

sequence_ihdp = (('IQR', 'x4'), ('zscore_filter_3', 'x5'), ('isolationForest', 'TABLE'))

sequence_twins = (('fill_drop_na', 'TABLE'), ('zscore_filter_3', 'cigar6'), ('isolationForest', 'TABLE'))
```

```python
# =========================
# Apply transformations
# =========================

transformed_df_lalonde = apply_data_preparations_seq(
    df_lalonde,
    sequence_lalonde,
    largest_data_transformations
)

transformed_df_acs = apply_data_preparations_seq(
    df_acs_copy,
    sequence_acs,
    largest_data_transformations
)

transformed_df_walmart = apply_data_preparations_seq(
    df_walmart,
    sequence_walmart,
    largest_data_transformations
)

transformed_df_ihdp = apply_data_preparations_seq(
    df_ihdp,
    sequence_ihdp,
    largest_data_transformations
)

transformed_df_twins = apply_data_preparations_seq(
    df_twins,
    sequence_twins,
    largest_data_transformations
)
```

```python
# =========================
# Print results
# =========================
results_data = [
    {
        "Dataset": "acs",
        "DR_Before": calculate_ate_dr(df_acs_copy),
        "DR_After": calculate_ate_dr(transformed_df_acs),
        "DML_Before": calculate_ate_dml(df_acs_copy),
        "DML_After": calculate_ate_dml(transformed_df_acs)
    },
    {
        "Dataset": "walmart",
        "DR_Before": calculate_ate_dr(df_walmart),
        "DR_After": calculate_ate_dr(transformed_df_walmart),
        "DML_Before": calculate_ate_dml(df_walmart),
        "DML_After": calculate_ate_dml(transformed_df_walmart)
    }
]

# Create DataFrame and save to CSV
results_df = pd.DataFrame(results_data)
results_df.to_csv("DR_DML_ate_results.csv", index=False)

# =========================================================
# Print results
# =========================================================

for row in results_data:
    print(
        f"{row['Dataset']}\n"
        f"start ATE:\n"
        f"DR:\n"
        f"before: {row['DR_Before']} | after: {row['DR_After']}\n"
        f"DML:\n"
        f"before: {row['DML_Before']} | after: {row['DML_After']}"
    )
    print("~" * 120)
```
