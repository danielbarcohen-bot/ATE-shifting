## Tokenization

Tokenization refers to converting a text string into individual tokens. Tokens may be words or punctations

```python
import nltk
#import os
```

```python
#Read the base file into a raw text variable
base_file = open("/kaggle/input/text-handling-ipba-kn/1_TextRepresentation/data_science.txt", 'rt')
raw_text = base_file.read()
base_file.close()
```

```python
#Extract tokens
token_list = nltk.word_tokenize(raw_text)
print("Token List : ",token_list[:100])
print("\n Total Tokens : ",len(token_list))
```

## Cleansing Text

We will see examples of removing punctuation and converting to lower case

#### Remove Punctuation

```python
#Use the Punkt library to extract tokens
token_list2 = list(filter(lambda token: nltk.tokenize.punkt.PunktToken(token).is_non_punct, token_list))
print("Token List after removing punctuation : ",token_list2[:100])
print("\nTotal tokens after removing punctuation : ", len(token_list2))
```

#### Convert to Lower Case

```python
token_list3=[word.lower() for word in token_list2 ]
print("Token list after converting to lower case : ", token_list3[:20])
print("\nTotal tokens after converting to lower case : ", len(token_list3))
```

## Stop word Removal

Removing stop words by using a standard stop word list available in NLTK for English

```python
#Download the standard stopword list
#nltk.download('stopwords')
from nltk.corpus import stopwords

#Remove stopwords
token_list4 = list(filter(lambda token: token not in stopwords.words('english'), token_list3))
print("Token list after removing stop words : ", token_list4[:20])
print("\nTotal tokens after removing stop words : ", len(token_list4))
```

## Stemming

```python
#Use the PorterStemmer library for stemming.
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

#Stem data
token_list5 = [stemmer.stem(word) for word in token_list4 ]
print("Token list after stemming : ", token_list5[:20])
print("\nTotal tokens after Stemming : ", len(token_list5))
```

# Negation

```python
import nltk
from nltk.sentiment.util import mark_negation

# Tokenize a sentence into words
sentence = "The movie was not good, but the acting was great."
words = nltk.word_tokenize(sentence)

# Apply negation handling using mark_negation() function
words_neg_handled = mark_negation(words)

# Print the result
print(words_neg_handled)
```

# Expanding Abbreviations

```python
# Define a dictionary of abbreviations and their full forms
abbreviations = {
    'mr.': 'mister',
    'mrs.': 'missus',
    'dr.': 'doctor',
    'st.': 'saint',
    'ave.': 'avenue',
    'rd.': 'road',
    'blvd.': 'boulevard'
}

# Define a function to expand abbreviations in a given sentence
def expand_abbreviations(text):
    words = text.split()
    expanded_words = []
    for word in words:
        if word.lower() in abbreviations:
            expanded_words.append(abbreviations[word.lower()])
        else:
            expanded_words.append(word)
    return ' '.join(expanded_words)

# Example usage
sentence = "Please take me to Dr. Smith's office on Park Ave."
expanded_sentence = expand_abbreviations(sentence)
print(expanded_sentence)
```

# Repeated Letters/Words

```python
import re

# Define a function to remove repeated characters from a given string
def remove_repeated_chars(text):
    # Replace any sequence of 3 or more repeated characters with the same character
    pattern = re.compile(r"(.)\1{2,}")
    return pattern.sub(r"\1", text)

# Example usage
text = "Hellooooo, how areeeee youuuuu? "
clean_text = remove_repeated_chars(text)
print(clean_text)
```
