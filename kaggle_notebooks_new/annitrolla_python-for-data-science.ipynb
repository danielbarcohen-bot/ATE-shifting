# Modelling

We want to predict who to approve based on the history of earlier approvals. We will be using 'Approved' as our **target variable**, it is usually indicated as **y** in data science. We can use any other features (variables) to predict it. Features are usually indicated as **X**. 

```python
# Choose the columns you want to use in your model
dt_modelling = cc_approval[['Gender','Age', 'Married', 'BankCustomer',
       'Employed', 'CreditScore', 'Approved', 'Citizen']]
```

Often, algorithms are not capable of dealing with non-numerical features. Therefore, we need to represent them as numeric ones. Various methods of changing feature type from non-numeric to numeric exists. We will look here at **one-hot encoding** that is suitable for **categorical features** - features that do not have any ranking. 

```python
df = pd.get_dummies(dt_modelling, drop_first=True)
df.head()
```

<div class="alert alert-block alert-info"> 🐱‍💻 What happened with Citizen_ByBirth? Why it is not in the columns? </div>

Now, when features are pre-processed, we need to separate our target variable from the rest. Also, we need to split our dataset into two distinct parts: training and test. We need to **learn** our model from the training set and validate how well we performed on the test. Data that is used for learning can't be used for testing. Think about it as a school assignment, we want to test our model without it *peeking* at the right answers.

![](https://i.kym-cdn.com/photos/images/newsfeed/000/221/868/cannot_be_unseen.jpg)

```python
X = df.drop('Approved', axis=1)
y = df['Approved']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
```

```python
print(X_train.shape)
print(X_test.shape)
```

We train our model using logistic regression

```python
logit = LogisticRegression(max_iter=400)
logit.fit(X_train, y_train)
```

```python
y_pred = logit.predict(X_test)
y_pred
```

```python
y_pred_proba = logit.predict_proba(X_test)
y_pred_proba[:,1]
```

```python
print("Accuracy of logistic regression classifier: ", logit.score(X_test, y_test))
```

```python
X_test['prediction'] = y_pred
X_test['Approved'] = y_test
```

```python
pd.crosstab(X_test['Approved'], X_test['prediction'])
```

# Additional Tips & Tricks of Pandas

```python
cc_approval.groupby('Industry').size().reset_index()
```

```python
cc_approval.groupby('Industry')['YearsEmployed'].mean().reset_index()
```

```python
cc_approval.groupby(['Industry', 'Married']).aggregate({'YearsEmployed': 'mean', 'Income': 'median', 'PriorDefault': 'sum'})
```

```python
%%timeit # measures time of the execution
cc_approval['SpecialOffer'] = cc_approval.apply(lambda x: 'qualified' if (x.BankCustomer == 1) & (x.CreditScore > 0) else 'not_qualified', axis=1)
```

Compare with the classical `for cycle`

```python
%%timeit
for i in range(cc_approval.shape[0]):
    if (cc_approval['BankCustomer'].iloc[i] == 1) & (cc_approval['CreditScore'].iloc[i] > 0):
        cc_approval['SpecialOffer'].iloc[i] = 'qualified'
    else:
        cc_approval['SpecialOffer'].iloc[i] = 'not_qualified'
```

```python
cc_approval.SpecialOffer.value_counts()
```
