```python
# ????? ????
import pandas as pd

# ??? ????
data = pd.read_csv('../input/mtcars/mtcars.csv')
print('minmax ?? ?: \n',data['qsec'].head())

# min-max scale
def minmax(data):
    data = (data - min(data)) / (max(data) - min(data))
    return data

data['qsec'] = minmax(data['qsec'])
print('minmax ?? ?: \n',data['qsec'].head())

# 0.5?? ? ? (row ? ??)
print(len(data[data['qsec']>0.5]))
```
