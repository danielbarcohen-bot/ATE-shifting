```python
import pandas as pd
import numpy as np
import pandas 
import numpy 
```

```python
veri = pandas.read_csv("/kaggle/input/deneyap-hayv/hayvanatbahcesi.csv",encoding="unicode_escape")
```

```python
veri
```

```python
girdi = numpy.array(veri.drop(["hayvan adi","sinifi"],axis = 1))
```

```python
girdi
```

```python
etiketler = numpy.array(pandas.DataFrame(veri["sinifi"]))
```

```python
from sklearn.model_selection import train_test_split
```

```python
ozellik_egitim,ozellik_test,etiket_egitim,etiket_test = train_test_split(girdi,etiketler,test_size = 0.10,random_state = 70)
```

```python
ozellik_test
```

```python
from sklearn.naive_bayes import CategoricalNB
```

```python
bayes = CategoricalNB()
```

```python
bayes.fit(ozellik_egitim,etiket_egitim)
```

```python
test_cikti = bayes.predict(ozellik_test)
```

```python
test_cikti
```

```python
from sklearn.metrics import confusion_matrix 
import seaborn as sns 
import matplotlib.pyplot as plt 
cm= confusion_matrix(etiket_test, test_cikti) 
index = ['1','2','3','4','5','6','7'] 
columns = ['1','2','3','4','5','6','7'] 
cm_df = pd.DataFrame(cm,columns,index) 
plt.figure(figsize=(10,6)) 
sns.heatmap(cm_df, annot=True,fmt="d") 
from sklearn import metrics 
print("Accuracy:",metrics.accuracy_score(etiket_test, test_cikti))
```
