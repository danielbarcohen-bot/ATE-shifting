### 📁 Load and Split Image Dataset (Training & Validation)

This code uses `tf.keras.preprocessing.image_dataset_from_directory` to load image data from a directory and automatically split it into training and validation sets.

- `validation_split=0.2` means 20% of the data will be used for validation.
- `subset="training"` and `"validation"` specify the two splits.
- `seed=100` ensures the same shuffling order is applied to both subsets.
- `target_size=(224, 224)` resizes all images to a uniform shape.
- `batch_size=64` groups the data into batches for efficient training.

The final loop prints the shape of one batch of training images and labels to confirm the data format. The expected output shapes are:
- `images.shape`: (64, 224, 224, 3) → 64 RGB images of size 224x224
- `labels.shape`: (64,) → 64 corresponding class labels


```python
train_data_dir =image_data
batch_size = 64
target_size = (224,224)
validation_split = 0.2
train= tf.keras.preprocessing.image_dataset_from_directory(
    train_data_dir,
    validation_split=validation_split,
    subset="training",
    seed=100,
    image_size=target_size,
    batch_size=batch_size,
)
validation= tf.keras.preprocessing.image_dataset_from_directory(
    train_data_dir,
    validation_split=validation_split,
    subset="validation",
    seed=100,
    image_size=target_size,
    batch_size=batch_size,
)

for images, labels in train.take(1):
    print(images.shape)  
    print(labels.shape)  
```

```python
sns.set_theme(style="whitegrid")

def get_dataset_size(dataset):
    return sum([len(batch[0]) for batch in dataset])

train_size = get_dataset_size(train)
validation_size = get_dataset_size(validation)

sizes = [train_size, validation_size]
labels = ['Training', 'Validation']
sizes_labels = [f'{label}: {size}' for label, size in zip(labels, sizes)]


colors = sns.color_palette("coolwarm", len(sizes))

plt.figure(figsize=(8, 8))
plt.pie(
    sizes, 
    labels=sizes_labels, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=colors, 
    wedgeprops={'edgecolor': 'white', 'linewidth': 2}
)


plt.title('Training vs. Validation Data Size', fontsize=16, pad=20)
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
plt.gca().add_artist(centre_circle)


plt.axis('equal')
plt.show()
```

<h3 style="font-weight:700">Keras Application</h3>

ConvNeXt-Tiny is a modern convolutional neural network architecture that rethinks ConvNet design using principles from transformer-based models. Despite being fully convolutional, ConvNeXt achieves competitive or superior performance compared to vision transformers such as Swin Transformer, while retaining the efficiency advantages of ConvNets.

To develop ConvNeXt, the authors start with a standard ResNet and gradually modify its components to incorporate design patterns found in transformers, resulting in a ConvNet that scales well and trains efficiently.

Architecturally the main differences are:

- ConvNeXt replaces traditional stem convolutions with a patchify-like operation similar to Vision Transformers (i.e., using a large kernel with stride-4 to divide the image into non-overlapping patches).
- It uses depthwise separable convolutions (like in MobileNet) and large kernel sizes (7x7) to increase receptive field while keeping parameters manageable.
- The architecture employs Layer Normalization instead of Batch Normalization, aligning with transformer-style normalization.
- Activation functions are replaced with GELU instead of ReLU for smoother gradients.
- ConvNeXt uses inverted bottlenecks and residual connections throughout, inspired by ResNet and MobileNet.
- It uses fewer downsampling stages, allowing it to preserve more spatial resolution in the early layers.

ConvNeXt-Tiny is especially suitable for deployment in environments where both performance and inference speed matter, such as mobile and edge devices.


### ⚙️ Model Configuration Settings

#### 🔹 `input_shape`
- **Type**: Tuple (height, width, channels)
- **Description**: Optional. Define the input image size if it's different from the default `(224, 224, 3)`.
- **Important**: 
  - Must have **3 channels** (RGB), e.g., `(160, 160, 3)` is valid.
  - If both `input_tensor` and `input_shape` are provided:
    - If shapes match → `input_shape` is used.
    - If shapes don't match → error will be raised.
  - Can be omitted if using `input_tensor`.

---

#### 🔹 `include_top`
- **Type**: Boolean
- **Default**: `True`
- **Description**: Whether to include the fully-connected classification layer (usually softmax) at the top of the network.
- **Tip**: Set to `False` if you want to use the model as a **feature extractor**.

---

#### 🔹 `weights`
- **Type**: String
- **Options**:
  - `'imagenet'`: Use pre-trained weights from ImageNet.
  - `None`: Initialize with random weights.
  - `path`: Provide a custom weights file path to load.

### 📉 **Training Accuracy is Lower than Validation Accuracy**
This may indicate **underfitting** or **insufficient learning** from the training data.  
Here are strategies already applied or recommended to improve training performance:

---

### 🔧 1. **Unfreeze Additional Layers for Fine-Tuning**
Allow more layers of a pre-trained model (e.g., ResNet, EfficientNet) to be trainable.

---

### 📉 2. **Reduce Regularization**
Lower dropout rate, L2 weight decay, or other forms of regularization to allow the model to learn more.

---

### 🔄 3. **Add Batch Normalization**
Helps stabilize and accelerate training, especially in deeper models.

---

### 📈 4. **Increase Model Capacity**
Use a more complex architecture or add more layers/nodes to allow the model to capture more patterns.

---

### ⚙️ 5. **Adjust Optimizer or Learning Rate**
Try learning rate scheduling, or experiment with optimizers like `AdamW`, `SGD`, `RMSprop`.

---

### 🌐 6. **Use GlobalAveragePooling Instead of Flatten**
Reduces overfitting and the number of parameters while preserving spatial information better than `Flatten`.

```python
base_model = tf.keras.applications.ConvNeXtTiny(input_shape=(224, 224, 3), include_top=False, weights='imagenet')
base_model.trainable = True

fine_tune_at = int(len(base_model.layers) * 0.7)
for layer in base_model.layers[:fine_tune_at]:
    layer.trainable = False

keras_model = keras.models.Sequential([
    base_model,
    keras.layers.GlobalAveragePooling2D(),
    keras.layers.BatchNormalization(),
    keras.layers.Dense(64, activation='relu', kernel_regularizer=regularizers.l2(0.01)),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(6, activation='softmax')
])

keras_model.compile(
    optimizer=Adam(learning_rate=1e-4),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

keras_model.summary()
```

### 🎯 Model Training Callbacks: Checkpoint, Early Stopping & Learning Rate Scheduler

- **ModelCheckpoint**  
  Saves the model weights to `'best_model.h5'` whenever the validation loss (`val_loss`) improves.  
  - `save_best_only=True`: Only keep the best-performing model  
  - `mode='min'`: Monitor for minimum validation loss

- **EarlyStopping**  
  Stops training early if the validation loss does not improve for **6** consecutive epochs (`patience=6`).  
  - `restore_best_weights=True`: After stopping, restore the model weights from the best epoch

- **ReduceLROnPlateau**  
  Automatically reduces the learning rate when the validation loss plateaus.  
  - `monitor='val_loss'`: Watch the validation loss  
  - `factor=0.5`: Halves the learning rate  
  - `patience=3`: Waits 3 epochs with no improvement before reducing  
  - `min_lr=1e-6`: Ensures learning rate doesn’t fall below this value

These callbacks help optimize the training process by adapting the learning rate and avoiding overfitting.


```python
hist = keras_model.fit(
    train,
    epochs=40,
    validation_data=validation,
    callbacks=[checkpoint, early_stopping, reduce_lr]
)
```

### 🏋️‍♂️ Train the Keras Model with Validation and Callbacks

- `keras_model.fit()` starts training the model using the `train` dataset.
- `epochs=40`: Set maximum number of training epochs to 40.
- `validation_data=validation`: Use the `validation` dataset for evaluating model performance after each epoch.
- `callbacks=[checkpoint, early_stopping]`:  
  - `checkpoint` saves the best model based on validation loss.  
  - `early_stopping` halts training early if validation loss stops improving, preventing overfitting.

The training history is stored in `hist` for further analysis or plotting.

# **<h1 id="6" style="background-color:#F0EFEB;font-family:newtimeroman;font-size:150%;color:#283618;text-align:center;border-radius:15px 15px;padding:7px;border:solid 3px #B7B7A4;">Model Performance</h1>**

```python
def get_true_and_predicted_labels(model, dataset):
    true_labels = []  
    predicted_labels = []  
    
    for img_batch, label_batch in dataset:
        predictions = model.predict(img_batch, verbose=0) 
        predicted_labels.extend(np.argmax(predictions, axis=1))  
        
 
        true_labels.extend(label_batch.numpy())  
    
    return np.array(true_labels), np.array(predicted_labels)  

true_labels, predicted_labels = get_true_and_predicted_labels(keras_model, validation)

conf_matrix = confusion_matrix(true_labels, predicted_labels)

plt.figure(figsize=(10, 10))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', 
            xticklabels=class_labels, yticklabels=class_labels)
plt.xlabel('Predicted Label')  
plt.ylabel('True Label')  
plt.title('Confusion Matrix') 
plt.show() 
```

### 🔍 Generate and Visualize Confusion Matrix for Validation Set

1. **Function `get_true_and_predicted_labels`**:  
   - Iterates over the dataset batches.
   - For each image batch, uses the trained model to predict class probabilities.
   - Converts predictions to class indices (`argmax`).
   - Collects true labels and predicted labels into arrays.

2. **Calculate confusion matrix**:  
   - Uses `sklearn.metrics.confusion_matrix` to compare true and predicted labels.
   - The confusion matrix shows counts of correct and incorrect predictions per class.

3. **Plot confusion matrix**:
