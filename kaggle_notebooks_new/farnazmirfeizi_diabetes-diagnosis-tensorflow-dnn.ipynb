## <p style="color: #2554C7;font-size:130%;text-shadow: 1px 1px 2px gray">4.1 | Zero Values</p>

<div style="border-radius: 10px; padding: 5px 15px; background-color: #d1e6ff; font-size: 110%; text-align: justify; display: flex; box-shadow: 2px 2px 5px rgba(0, 0, 0, 0.2); ">
<p style="text-align: justify;color: #223a76;">
Among the features, there are several features with a minimum value of 0. Let's explore more of these features.<br><br>
<b>Pregnancies :</b><br> Indicates an individual's pregnancy history, and a value of zero means the person has never been pregnant.<br><br>
<b>Glucose :</b><br> This variable represents the plasma glucose concentration measured 2 hours after an oral glucose tolerance test. If it is close to zero, it may indicate issues related to glucose utilization in the body, such as type 2 diabetes or other metabolic problems.<br><br>
<b style="color:red">BloodPressure :</b><br> Indicates individuals' blood pressure, and <u>a value of zero doesn't seem logical</u> for this variable.<br><br>
<b style="color:red">SkinThickness :</b><br> Represents the thickness of subcutaneous fat, and people naturally have subcutaneous fat. This fat is essential for the functioning of the skin, muscles, and maintaining body temperature. <u>Therefore, a value of zero may either be treated as a missing value or might have been entered mistakenly.</u><br><br>
<b style= "color:orange">Insulin :</b><br>Indicates the serum insulin level after 2 hours of a test, and a value of zero or close to zero in this case may indicate pancreatic insulin problems. Therefore, a zero value in this variable may signify diabetes. However, approximately 236 individuals diagnosed as healthy have a test value of zero, which may not be accurate. <u>Hence, we consider these values as missing data (only for healthy individuals).</u><br><br>
<b style="color:red">BMI :</b><br> Body mass index, calculated based on an individual's weight and height. <u>A value of zero in this context appears meaningless.</u><br><br>
Given the information provided, we consider the presence of zero values in the variables "SkinThickness", "BMI", and "BloodPressure", as well as in the "Insulin" values of healthy individuals, as missing data. To address these missing values, we will use the method of replacing them with the mean of the other samples.<p>
</div>

```python
min_values = df.describe().loc[['min']]
style(min_values)
```

```python
columns_zero = ['BloodPressure','SkinThickness','BMI']

# Replacing Zero values with mean :
for i in columns_zero:
    mean = df[df[i] != 0][i].mean()
    df.replace({i: 0}, mean, inplace=True)
```

```python
# Helthy individuals with zero test value :
non_diabetic_test_zero = df[(df['Outcome'] == 0) & (df['Insulin'] == 0)]

# Calculating mean test value of nondiabetic individuals :
non_diabetic_test_mean = df[df['Outcome'] == 0]['Insulin'].mean()

# Replacing the Zero test values of healthy individuals with the mean test value of nondiabetic individuals :
df.loc[(df['Outcome'] == 0) & (df['Insulin'] == 0), 'Insulin'] = non_diabetic_test_mean

print(f'Number of Healthy samples with zero test value: {non_diabetic_test_zero.shape[0]}') 
print(f'Average test value of the remaining healthy samples: {non_diabetic_test_mean}')
```

## <p style="color: #2554C7;font-size:130%;text-shadow: 1px 1px 2px gray">5.1 | Define Features and Target</p>

```python
# Features:
X = df.iloc[:,:-1]

# Target :
y = df.iloc[:,-1:]
```

📄 [Back to Tabel of Contents](#contents_tabel)
<a id="scale"></a>
## <p style="color: #2554C7;font-size:130%;text-shadow: 1px 1px 2px gray">5.2 | Scaling</p>

```python
ss = StandardScaler()
X = ss.fit_transform(X)
```

📄 [Back to Tabel of Contents](#contents_tabel)
<a id="tts"></a>
## <p style="color: #2554C7;font-size:130%;text-shadow: 1px 1px 2px gray">5.3 | Splitting Train and Test data</p>

```python
X_train, X_test, y_train ,y_test = train_test_split(X, y, random_state = 42)
```

📄 [Back to Tabel of Contents](#contents_tabel)
<a id="ohe"></a>
## <p style="color: #2554C7;font-size:130%;text-shadow: 1px 1px 2px gray">5.4 | One-Hot Encoding</p>

```python
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)
```

📄 [Back to Tabel of Contents](#contents_tabel)
<a id="arch"></a>
## <p style="color: #2554C7;font-size:130%;text-shadow: 1px 1px 2px gray">5.5 | Architecture of a Neural Network Model</h1>

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Dense(8, activation='relu', input_shape=(8,)),
    Dropout(0.2),
    tf.keras.layers.Dense(30, activation='relu'), 
    Dropout(0.2),
    tf.keras.layers.Dense(10, activation='relu'), 
    Dropout(0.2),
    tf.keras.layers.Dense(2, activation='softmax'), 
])
```

```python
model.summary()
```

```python
plot_model(model, show_layer_activations=True, show_shapes=True)
```

📄 [Back to Tabel of Contents](#contents_tabel)
<a id="comp"></a>
## <p style="color: #2554C7;font-size:130%;text-shadow: 1px 1px 2px gray">5.6 | Compilation and Fitting model</p>

```python
model.compile(optimizer=tf.keras.optimizers.Adam(0.002),
              loss=tf.keras.losses.CategoricalCrossentropy(),
              metrics=[tf.keras.metrics.CategoricalAccuracy()]
             )
```

```python
history = model.fit(X_train, y_train, epochs=50, validation_data=(X_test, y_test), batch_size=10)
```

```python
best_score = max(history.history['val_categorical_accuracy'] )
print(f"Best Validation score is: {best_score}")
```
