**import Data**

```python
import pandas
from sklearn import linear_model
import seaborn as sns
import matplotlib.pyplot as plt

df = pandas.read_csv("../input/stress-level-detection/Stress-Lysis.csv")

df.head
```

**Stress Level : Temperature**

```python
sns.violinplot(x=df['Stress Level'], y=df['Temperature'])
```

**Stress Level : Step Count**

```python
sns.violinplot(x=df['Stress Level'], y=df['Step count'])
```

**Stress Level : Humidity**

```python
sns.violinplot(x=df['Stress Level'], y=df['Humidity'])
```

```python
X = df[['Humidity', 'Temperature','Step count']]
y = df['Stress Level']

regr = linear_model.LinearRegression()
regr.fit(X.values, y)

print('intercept :' , regr.intercept_)
print('coefficient :' , regr.coef_)
```

```python
print("Prediction : ['Humidity', 'Temperature','Step count']")
prediction = regr.predict([[25, 89,52]])
print('[13, 82,20] : ',prediction)
prediction = regr.predict([[13, 82,20]])
print('[13, 82,20] : ',prediction)
prediction = regr.predict([[20, 89,80]])
print('[20, 89,80] : ',prediction)
prediction = regr.predict([[28, 96,155]])
print('[28, 96,155] : ',prediction)
print("-- Maximum --")
prediction = regr.predict([[30, 0,0]])
print('[30, 0,0] : ',prediction)
prediction = regr.predict([[0,99,0]])
print('[0, 99,0] : ',prediction)
prediction = regr.predict([[0, 0,200]])
print('[0, 0,200] : ',prediction)
```
