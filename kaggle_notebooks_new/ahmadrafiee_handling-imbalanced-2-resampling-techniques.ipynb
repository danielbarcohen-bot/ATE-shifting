<font style="font-family:Georgia;color:black; font-size:110%;">   &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;  ◼&nbsp; Class imbalance is a problem that occurs in machine learning classification problems. It merely tells that the target class’s frequency is highly imbalanced, i.e., the occurrence of one of the classes is very high compared to the other classes present. In other words, there is a bias or skewness towards the majority class present in the target. Suppose we consider a binary classification where the majority target class has 10000 rows, and the minority target class has only 100 rows. In that case, the ratio is 100:1, i.e., for every 100 majority class, there is only one minority class present. This problem is what we refer to as class imbalance. Some of the general areas where we can find such data are fraud detection, churn prediction, medical diagnosis, e-mail classification, etc.

<font style="font-family:Georgia;color:black; font-size:160%;"> 🔻Tackling Imbalanced Datasets  </font>

<font style="font-family:Georgia;color:navy;font-weight: bold;  font-size:110%;">   &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;  ◼&nbsp; Be aware to use fine metric: (What I explained in PART ONE (link) </font>

     <br>   
    <font style="font-family:Georgia;color:navy;font-weight: bold;  font-size:110%;">   &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;  ◼&nbsp; Model-based approaches: Using class weights and tree-based algorithms that naturally handle imbalanced data(What I do in part two) </font> <br>   
<br>    
<font style="font-family:Georgia;color:navy;font-weight: bold;  font-size:110%;">   &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;  ◼&nbsp; Data-based approaches: Sampling Techniques(What I do in this notebook): </font> <font style="font-family:Georgia;color:black;  font-size:110%;">Under-sampling data related to majority class: Under-sampling is a common technique used to address the issue of class imbalance in machine learning models. Class imbalance occurs when the training data is not evenly distributed between classes, which can lead to biased models. Under-sampling involves randomly removing samples from the majority class with or without replacement until the class distribution is more balanced. This can be done either before or after splitting the data into train and test sets. This is also called random undersampling. Although under-sampling can improve model performance, it may also reduce the overall accuracy of the model if the minority class is very small. Therefore, it is important to carefully consider whether under-sampling is the right approach for your data set. <br> 
Over-sampling data related to minority classes: Oversampling is a technique used to solve the class imbalance problem in machine learning models. It involves randomly selecting samples from the minority class and replicating them until the classes are balanced. This technique can improve the performance of machine learning models because it ensures that the model is trained on data that is representative of the test data. Moreover, oversampling can also help to reduce the variance of the model, which can further improve performance. This technique is especially useful when the dataset is small and there is a danger of overfitting. By oversampling, we can ensure that the model is trained on a balanced dataset and that all classes are represented equally. <br> 
Synthetic sampling: Synthetic sampling is a technique used to address class imbalance in machine learning models by generating new synthetic samples from the minority class. The idea behind synthetic sampling is to create new samples that are similar to the minority class, thus increasing the representation of the minority class in the dataset. One of the most widely used algorithms for synthetic training examples is Synthetic Minority Over-sampling Technique (SMOTE). SMOTE works by creating new synthetic data samples that are similar to existing data samples in the minority class based on finding its k nearest neighbors. These synthetic data samples can then be used to train the machine learning model, providing a more balanced training set. In addition, SMOTE can also help to improve the generalizability of the model by increasing the number of training examples. <br>

````python
from sklearn.model_selection import  cross_val_score
# now you can import normally from model_selection
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score, roc_auc_score
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
from sklearn.preprocessing import LabelEncoder


le = LabelEncoder()
df1 = df.copy(deep = True)


col = list(df.columns)
categorical_features = []
numerical_features = []
for i in df1.columns:
    if len(df[i].unique()) > 6:
        numerical_features.append(i)
    else:
        df1[i]=le.fit_transform(df1[i])
        categorical_features.append(i)
````

````python
X=df1.drop(['HeartDisease',"Race"], axis=1)
y=df1['HeartDisease']
````

````python

X_train , X_test, y_train,  y_test = train_test_split(X, y, test_size=0.35, random_state=0)

scaler= StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

logreg_ss=LogisticRegression(solver="liblinear", penalty="l2", C=0.00001, max_iter=10000)
logreg_ss.fit(X_train_scaled,y_train)
y_pred_log=logreg_ss.predict(X_test_scaled)

y_pred_proba_log = logreg_ss.predict_proba(X_test_scaled)
fpr_log, tpr_log, _ = metrics.roc_curve(y_test, y_pred_proba_log[:,1])
auc_log = round(metrics.auc(fpr_log, tpr_log),5)


simple_log = pd.DataFrame(data=[accuracy_score(y_test, y_pred_log),
                                                                    precision_score(y_test, y_pred_log, average='binary'),
                                                                    recall_score(y_test, y_pred_log, average='binary'),
                                                                    f1_score(y_test, y_pred_log, average='binary'),
                                                                    roc_auc_score(y_test, y_pred_proba_log[:,1])],
                                                                    index=['Accuracy','Precision','Recall','F1-score','AUC'],
                                                                     columns = ["Logestic_regression_simple"])

simple_log
````

Random Oversampling involves supplementing the training data with multiple copies of some of the minority classes. Oversampling can be done more than once (2x, 3x, 5x, 10x, etc.) This is one of the earliest proposed methods, that is also proven to be robust. Instead of duplicating every sample in the minority class, some of them may be randomly chosen with replacement.
<br>
<br><blockquote>
<font style="color:navy;font-weight: bold ">sampling_strategy='minority':</font> : <br>
This means that if the majority class had 1,000 examples and the minority class had 100, this strategy would oversampling the minority class so that it has 1,000 examples.
<br><br>
<font style="color:navy;font-weight: bold ">sampling_strategy=0.5: </font> : <br>
 This means that if the majority class had 1,000 examples and the minority class had 100, the transformed dataset would have 500 examples of the minority class.
    </blockquote>

 where n is the number of samples. This means that if it takes 0.8 seconds for 7.5K data points, it should take [3, 48] minutes for 115K,
[(115/7.5)^2×0.8,(115/7.5)^3×0.8]s=[3,48]m,

<font style="color:Crimson;font-weight: bold ">So I used feature importance and selecting a part of dataset to solve this issue: </font>

````python
set_background("#F0FFF0")
#Use "make_scorer" to convert a metric to scorer
def f1_metric(model, X_train, y_train):
        return f1_score(y_train, model.predict(X_train), average='binary')
````

````python
set_background("#F0FFF0")
from sklearn.base import clone 


def Featureimportance(model, X_train, y_train, random_state=0):
    # list containing feature importances
    importances = []
    # Clone the model
    model_clone = clone(model)
    # Set random_state for comparability
    model_clone.random_state = random_state
    # Train the model
    model_clone.fit(X_train, y_train)
    # Create the cross-validation object using StratifiedKFold 
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
    # Score the benchmark model using cross-validation
    benchmark_score = cross_val_score(model_clone, X_train, y_train, cv=cv, scoring=f1_metric).mean()

    
    # Iterate over all features and store feature importance
    for col in X_train.columns:
        # Clone the model
        model_clone = clone(model)
        # Set random_state for comparability
        model_clone.random_state = random_state
        # Traint the model on the the dataset with a single feature removed
        model_clone.fit(X_train.drop(col, axis = 1), y_train)
        
        # Score the droped-column model
        drop_column_score = cross_val_score(model_clone, X_train.drop(col,axis=1), y_train, cv=cv, scoring=f1_metric).mean()
        # Store feature importance which is defined as the difference between the benchmark and the new model score
        importances.append(benchmark_score - drop_column_score)
        
    # Return the features along with their importances in the form of a dataframe
    importances_df = pd.DataFrame({'feature': X_train.columns, 'feature importance': importances}) \
                     .sort_values('feature importance', ascending = False).reset_index(drop = True)
           
    return importances_df       
````

````python
set_background("#F0FFF0")

def plotFeatureimportance(model, X_train, y_train):
    # Call drop-column feature importance function
    df_drop_column = Featureimportance(model, X_train, y_train, random_state=0)
    # Rename columns
    df_drop_column.columns = ['Feature', 'Feature Importance']
    
    # Plot barchart
    plt.figure(figsize=(12,10))
    sns.barplot(data=df_drop_column, x='Feature Importance', y='Feature', orient='h',
                              color="white",
                              hatch='//',
                              edgecolor="darkred",
                              ecolor="black",
                              linewidth=0.6,)
    
    
    plt.xlabel("Feature Importance", size=16)
    plt.ylabel("Feature", size=16)
    plt.title(' Feature Importance', fontsize=20)
    plt.show()
````

One way of handling imbalanced datasets is to reduce the number of observations from all classes but the minority class. The minority class is that with the least number of observations. The most well known algorithm in this group is random undersampling, where samples from the targeted classes are removed at random.<br>
<br>
But there are many other algorithms to help us reduce the number of observations in the dataset. These algorithms can be grouped based on their undersampling strategy into:<br>
<blockquote>
<font style="color:navy;font-weight: bold ">1_Prototype generation methods:</font> prototype generation techniques will reduce the number of samples in the targeted classes but the remaining samples are generated — and not selected — from the original set. <font style="color:crimson;font-weight: bold ">Like Cluster Centroid</font><br>
<br>
<font style="color:navy;font-weight: bold ">2_Prototype selection methods: </font>
Prototype selection algorithms can be divided into two groups: <br>
<blockquote>
<font style="color:DarkOrange;font-weight: bold ">(i) controlled under-sampling techniques and Controlled under-sampling methods:</font> reduce the number of observations in the majority class or classes to an arbitrary number of samples specified by the user. Typically, they reduce the number of observations to the number of samples observed in the minority class.<font style="color:crimson;font-weight: bold ">Random Under_sampling, NearMiss, CNN</font><br><br>
<font style="color:DarkOrange;font-weight: bold ">(ii) cleaning under-sampling techniques:</font>In contrast, cleaning under-sampling techniques “clean” the feature space by removing either “noisy” or “too easy to classify” observations, depending on the method. The final number of observations in each class varies with the cleaning method and can’t be specified by the user. <font style="color:crimson;font-weight: bold ">ENN, RNN, ALLKNN, Tomek links, OSS, NCR</font><br>
</blockquote>

Near Miss refers to a collection of undersampling methods that select examples based on the distance of majority class examples to minority class examples.<br>
<br>
<font style="color:Crimson;font-weight: bold ">NearMiss has 3 versions:</font>
1- **NearMiss-1** selects the positive samples for which the average distance to the k closest samples of the negative class is the smallest.
2- **NearMiss-2** selects the positive samples for which the average distance to the k farthest samples of the negative class is the smallest.( is an inverse to NearMiss-1)
3- **NearMiss-3** is a 2-steps algorithm. First, for each negative sample, their m nearest-neighbors will be kept. Then, the positive samples selected are the ones for which the average distance to the k nearest-neighbors is the largest.
<br>
<br>
<blockquote>
<font style="color:navy;font-weight: bold ">sampling_strategy:</font>
<br>
'majority': resample only the majority class;<br>
'not minority': resample all classes but the minority class;<br>
'not majority': resample all classes but the majority class;<br>
'all': resample all classes;<br>
'auto': equivalent to 'not minority'.<br>
<br><br>
<font style="color:navy;font-weight: bold ">version:</font>Version of the NearMiss to use. Possible values are 1, 2 or 3. default=1
<br><br>
<font style="color:navy;font-weight: bold ">n_neighbors:</font>Size of the neighbourhood to consider to compute the average distance to the minority point samples.
<br> </blockquote>

Tomek Links is one of a modification from Condensed Nearest Neighbors. Unlike the CNN method that are only randomly select the samples with its k nearest neighbors from the majority class that wants to be removed, the Tomek Links method uses the rule to selects the pair of observation (say, a and b) that are fulfilled these properties:
<br>
The observation a’s nearest neighbor is b.<br>
The observation b’s nearest neighbor is a.<br>
Observation a and b belong to a different class. That is, a and b belong to the minority and majority class (or vice versa), respectively.
<br><br>
<blockquote>
<font style="color:navy;font-weight: bold ">sampling_strategy:</font><br>
'majority': resample only the majority class <br>
'not minority': resample all classes but the minority class<br>
'not majority': resample all classes but the majority class<br>
'all': resample all classes<br>

 'majority': resample only the majority class<br>

<br><br>
<font style="color:navy;font-weight: bold ">kind_sel:</font>Strategy to use in order to exclude samples in the ENN sampling.

If 'mode', the majority vote of the neighbours will be used in order to exclude a sample.
The strategy "all" will be less conservative than 'mode'. Thus, more samples will be removed when kind_sel="all" generally.
<br><br>
<font style="color:navy;font-weight: bold ">threshold_cleaning:</font>Threshold used to whether consider a class or not during the cleaning after applying ENN.

Undersample by generating centroids based on clustering methods.

Method that under samples the majority class by replacing a cluster of majority samples by the cluster centroid of a KMeans algorithm. This algorithm keeps N majority samples by fitting the KMeans algorithm with N cluster to the majority class and using the coordinates of the N cluster centroids as the new majority samples.
<br>
<blockquote>
<font style="color:navy;font-weight: bold ">sampling_strategy:</font><br>
'majority': resample only the majority class<br>
'not minority': resample all classes but the minority class<br>

<br><br>
<font style="color:navy;font-weight: bold ">estimator:</font>A scikit-learn compatible clustering method that exposes a n_clusters parameter and a cluster_centers_ fitted attribute. 
<br><br>
<font style="color:navy;font-weight: bold ">voting:</font>:
Voting strategy to generate the new samples:
If 'hard', the nearest-neighbors of the centroids found using the clustering algorithm will be used.
If 'soft', the centroids found by the clustering algorithm will be used.
If 'auto', if the input is sparse, it will default on 'hard' otherwise, 'soft' will be used.
<br>
    </blockquote>

<a href="https://www.sciencedirect.com/science/article/abs/pii/S0031320311000823">Learn More About This Paper</a>
In this paper, a dynamic over-sampling procedure is proposed for improving the classification of imbalanced datasets with more than two classes. The base over-sampling procedure is the synthetic minority over-sampling technique (SMOTE) . SMOTE is an over-sampling method, where the minority class is over-sampled by taking each minority class sample and introducing synthetic examples throughout the line segments joining any/all of the k minority class nearest neighbours. Depending upon the amount of over-sampling required, neighbours from the k nearest neighbours are randomly chosen.<br>
<br>
This procedure is incorporated into a memetic algorithm (MA)  that optimizes radial basis functions neural networks (RBFNNs). The MA combines an evolutionary algorithm (EA), a clustering process, and a local search (LS) procedure. For this reason, this approach is called memetic radial basis function (MRBF).<br>
<br>
We propose two different methodologies which add an over-sampling procedure in the learning algorithm for dealing with imbalanced multi-class datasets: the static smote radial basis function (SSRBF) method and the dynamic smote radial basis function (DSRBF) method. The SSRBF method performs the over-sampling procedure in the preprocessing stage and the DSRBF runs the over-sampling procedure both in the preprocessing stage and in different generations of the MA.

This algorithm is known as EasyEnsemble. The classifier is an ensemble of AdaBoost learners trained on different balanced bootstrap samples. The balancing is achieved by random under-sampling.<br>
<blockquote>
<font style="color:navy;font-weight: bold ">n_estimators:</font><br>
Number of AdaBoost learners in the ensemble.<br><br>
<font style="color:navy;font-weight: bold ">estimator:</font><br>The base AdaBoost classifier used in the inner ensemble.<br><br>
<font style="color:navy;font-weight: bold ">warm_start:</font><br> When set to True, reuse the solution of the previous call to fit and add more estimators to the ensemble, otherwise, just fit a whole new ensemble.<br><br>
<font style="color:navy;font-weight: bold ">sampling_strategy:</font><br>
    'majority': resample only the majority class<br>
'not minority': resample all classes but the minority class<br>
'not majority': resample all classes but the majority class<br>
'all': resample all classes<br>
'auto': equivalent to 'not minority'.<br><br>
<font style="color:navy;font-weight: bold ">verbose:</font><br>Controls the verbosity of the building process.
    </blockquote>

````python
set_background("#FAEBD7")

from imblearn.ensemble import EasyEnsembleClassifier
from sklearn.ensemble import AdaBoostClassifier

X_train , X_test, y_train,  y_test = train_test_split(X, y , test_size=0.35, random_state=0)


scaler= StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)


estimator = AdaBoostClassifier(n_estimators=10)
eec = EasyEnsembleClassifier(n_estimators=10, estimator=estimator)
eec.fit(X_train_scaled,y_train)
y_pred_eec = eec.predict(X_test_scaled)


y_pred_proba_eec = eec.predict_proba(X_test_scaled)

Easyec = pd.DataFrame(data=[accuracy_score(y_test, y_pred_eec),
                               precision_score(y_test, y_pred_eec, average='binary'),
                               recall_score(y_test, y_pred_eec, average='binary'),
                               f1_score(y_test, y_pred_eec, average='binary'),
                               roc_auc_score(y_test, y_pred_proba_eec[:,1])],
                               index=['Accuracy','Precision','Recall','F1-score','AUC'],
                               columns = ["EasyEnsembleClassifier"])
Easyec
````

Random under-sampling integrated in the learning of AdaBoost.

During learning, the problem of class balancing is alleviated by random under-sampling the sample at each iteration of the boosting algorithm.<br>
<blockquote>
<font style="color:navy;font-weight: bold ">estimator: </font><br>
 The base estimator from which the boosted ensemble is built. <br><br>
<font style="color:navy;font-weight: bold ">n_estimators:</font><br>The maximum number of estimators at which boosting is terminated. In case of perfect fit, the learning procedure is stopped early.<br>
<font style="color:navy;font-weight: bold ">learning_rate:</font><br>Learning rate shrinks the contribution of each classifier by learning_rate.<br><br>
<font style="color:navy;font-weight: bold ">algorithm:</font><br>If ‘SAMME.R’ then use the SAMME.R real boosting algorithm. base_estimator must support calculation of class probabilities. If ‘SAMME’ then use the SAMME discrete boosting algorithm. The SAMME.R algorithm typically converges faster than SAMME, achieving a lower test error with fewer boosting iterations.<br><br>
<font style="color:navy;font-weight: bold ">sampling_strategy:</font><br>'majority': resample only the majority class<br>
'not minority': resample all classes but the minority class<br>
'not majority': resample all classes but the majority class<br>
'all': resample all classes<br>
'auto': equivalent to 'not minority'.
</blockquote>

````python
set_background("#FAEBD7")
from imblearn.ensemble import  RUSBoostClassifier

X_train , X_test, y_train,  y_test = train_test_split(X, y , test_size=0.35, random_state=0)


scaler= StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

estimator = AdaBoostClassifier(n_estimators=10)
rusboost = RUSBoostClassifier(n_estimators=10, estimator=estimator)
rusboost.fit(X_train_scaled,y_train)
y_pred_rusboost = rusboost.predict(X_test_scaled)



y_pred_proba_rusboost = rusboost.predict_proba(X_test_scaled)

RUSBoost = pd.DataFrame(data=[accuracy_score(y_test, y_pred_rusboost),
                               precision_score(y_test, y_pred_rusboost, average='binary'),
                               recall_score(y_test, y_pred_rusboost, average='binary'),
                               f1_score(y_test, y_pred_rusboost, average='binary'),
                               roc_auc_score(y_test, y_pred_proba_rusboost[:,1])],
                               index=['Accuracy','Precision','Recall','F1-score','AUC'],
                               columns = ["RUSBoostClassifier"])
RUSBoost
````

A balanced random forest classifier.
<br>
A balanced random forest differs from a classical random forest by the fact that it will draw a bootstrap sample from the minority class and sample with replacement the same number of samples from the majority class.<br>
<blockquote>
<font style="color:navy;font-weight: bold ">n_estimators:</font><br>
The number of trees in the forest.<br><br>
<font style="color:navy;font-weight: bold ">criterion:</font><br>The function to measure the quality of a split. Supported criteria are “gini” for the Gini impurity and “entropy” for the information gain. Note: this parameter is tree-specific.<br><br>
<font style="color:navy;font-weight: bold ">max_depth:</font><br>The maximum depth of the tree<br><br>
<font style="color:navy;font-weight: bold ">min_samples_split:</font><br>The minimum number of samples required to split an internal node<br><br>
<font style="color:navy;font-weight: bold ">min_samples_leaf:</font><br>The minimum number of samples required to be at a leaf node<br><br>
<font style="color:navy;font-weight: bold ">min_weight_fraction_leaf:</font><br>The minimum weighted fraction of the sum total of weights (of all the input samples) required to be at a leaf node.<br><br>
<font style="color:navy;font-weight: bold ">max_features:</font><br>The number of features to consider when looking for the best split. “auto”, “sqrt”, “log2”,None<br><br>
<font style="color:navy;font-weight: bold ">max_leaf_nodes:</font><br>Best nodes are defined as relative reduction in impurity<br><br>
<font style="color:navy;font-weight: bold ">bootstrap:</font><br> Whether bootstrap samples are used when building trees<br><br>
<font style="color:navy;font-weight: bold ">oob_score:</font><br> Whether to use out-of-bag samples to estimate the generalization accuracy<br><br>
<font style="color:navy;font-weight: bold ">sampling_strategy:</font><br>'majority': resample only the majority class;<br>
'not minority': resample all classes but the minority class<br>
'not majority': resample all classes but the majority class<br>
'all': resample all classes<br>
'auto': equivalent to 'not minority'.<br><br>
<font style="color:navy;font-weight: bold ">verbose:</font><br>Controls the verbosity of the tree building process.<br><br>
<font style="color:navy;font-weight: bold ">warm_start:</font><br>When set to True, reuse the solution of the previous call to fit and add more estimators to the ensemble, otherwise, just fit a whole new forest.<br><br>
<font style="color:navy;font-weight: bold ">class_weight:</font><br>
ccp_alpha:Complexity parameter used for Minimal Cost-Complexity Pruning. 
</blockquote>


````python
set_background("#FAEBD7")

from imblearn.ensemble import BalancedRandomForestClassifier


X_train , X_test, y_train,  y_test = train_test_split(X, y , test_size=0.35, random_state=0)


scaler= StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)




brf = BalancedRandomForestClassifier( n_estimators=50, sampling_strategy="all", replacement=True,
                                                                        bootstrap=False,random_state=0,)
brf.fit(X_train_scaled,y_train)
y_pred_brf = brf.predict(X_test_scaled)

y_pred_proba_brf = brf.predict_proba(X_test_scaled)



balancedRF = pd.DataFrame(data=[accuracy_score(y_test, y_pred_brf),
                               precision_score(y_test, y_pred_brf, average='binary'),
                               recall_score(y_test, y_pred_brf, average='binary'),
                               f1_score(y_test, y_pred_brf, average='binary'),
                               roc_auc_score(y_test, y_pred_proba_brf[:,1])],
                               index=['Accuracy','Precision','Recall','F1-score','AUC'],
                               columns = ["BalancedRandomForestClassifier"])

balancedRF



````
