```python
import os
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))

import pandas 
train_data = pandas.read_csv("../input/titanic/train.csv")
test_data = pandas.read_csv('../input/titanic/test.csv')
```

```python
train_data.isnull().sum()
```

```python
# fill NA 
train_data["Age"] = train_data["Age"].fillna(train_data["Age"].median())
train_data["Embarked"] = train_data["Embarked"].fillna('S')

# S→0；C→1；Q→2
train_data.loc[train_data["Embarked"] == "S", "Embarked"] = 0
train_data.loc[train_data["Embarked"] == "C", "Embarked"] = 1
train_data.loc[train_data["Embarked"] == "Q", "Embarked"] = 2

# male→0，female→1
train_data.loc[train_data["Sex"] == "male", "Sex"] = 0
train_data.loc[train_data["Sex"] == "female", "Sex"] = 1

# ignore cabin 
```

```python
train_data.isnull().sum()
```

```python
test_data.isnull().sum()
```

```python
# fill NA
test_data["Age"] = test_data["Age"].fillna(test_data["Age"].median())
test_data["Fare"] = test_data["Fare"].fillna(test_data["Fare"].median())

# S→0；C→1；Q→2
test_data.loc[test_data["Embarked"] == "S", "Embarked"] = 0
test_data.loc[test_data["Embarked"] == "C", "Embarked"] = 1
test_data.loc[test_data["Embarked"] == "Q", "Embarked"] = 2

# male→0，female→1
test_data.loc[test_data["Sex"] == "male", "Sex"] = 0
test_data.loc[test_data["Sex"] == "female", "Sex"] = 1
```

```python
test_data.isnull().sum()
```

```python
from sklearn.preprocessing import StandardScaler

# predict the features of train_data
predictors = ["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked"]
x_data = train_data[predictors]
y_data = train_data["Survived"]

x2_data = test_data[predictors]


# Data standardization
scaler = StandardScaler()
x_data = scaler.fit_transform(x_data)
x2_data = scaler.fit_transform(x2_data)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#LogisticRegression
from sklearn import model_selection
from sklearn.linear_model import LogisticRegression

params = [1,10,15,20,25,30,40,50,60,70]
for param in params:
    LR = LogisticRegression()
    scores = model_selection.cross_val_score(LR, x_data, y_data, cv=10)

plt.plot(params,scores,c='r')
print(scores.mean()) 
```

```python
#nural_network
from sklearn.neural_network import MLPClassifier

params = [1,10,15,20,25,30,40,50,60,70]
for param in params:
    mlp = MLPClassifier(hidden_layer_sizes=(50,20,10),max_iter=1000)
    scores = model_selection.cross_val_score(mlp, x_data, y_data, cv=10)
    
plt.plot(params,scores,c='r')
plt.show()
print(scores.mean())
```

```python
#KNN
from sklearn import neighbors

params = [1,10,15,20,25,30,40,50,60,70]
for param in params:
    # k=20
    knn = neighbors.KNeighborsClassifier(20)
    scores = model_selection.cross_val_score(knn,x_data,y_data,cv=10)
    
plt.plot(params,scores,c='r')
print(scores.mean())   
```

```python
# decision tree
from sklearn import tree

params = [1,10,15,20,25,30,40,50,60,70]

for param in params:
    dtree = tree.DecisionTreeClassifier(max_depth=3,min_samples_split=4)
    scores = model_selection.cross_val_score(dtree,x_data,y_data,cv=10)

plt.plot(params,scores,c='r')
print(scores.mean())   
```

```python
# Random Foreset
from sklearn.ensemble import RandomForestClassifier

params = [1,10,15,20,25,30,40,50,60,70]

for param in params:
    RF1 = RandomForestClassifier(random_state=1,n_estimators=100,min_samples_split=2)
    scores = model_selection.cross_val_score(RF1,x_data,y_data,cv=10)
    
plt.plot(params,scores,c='r')
print(scores.mean())   
```

```python
# Bagging
from sklearn.ensemble import BaggingClassifier

params = [1,10,15,20,25,30,40,50,60,70]
for param in params:
    bagging_clf = BaggingClassifier(RF1,n_estimators=params)
    scores = model_selection.cross_val_score(RF1, x_data, y_data, cv=10)
    
plt.plot(params,scores,c='r')
print(scores.mean()) 
```

```python
# Adaboost
from sklearn.ensemble import AdaBoostClassifier

params = [1,10,15,20,25,30,40,50,60,70]
for param in params:
    adaboost = AdaBoostClassifier(bagging_clf,n_estimators=10)
    scores = model_selection.cross_val_score(RF1, x_data, y_data, cv=10)
    
plt.plot(params,scores,c='r')
print(scores.mean())  
```
