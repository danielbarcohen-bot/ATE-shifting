## **📚 Introduction: Kya Hai Yeh EDA?**

EDA (Exploratory Data Analysis) is like checking your biryani before serving it to guests. You need to see:
- **Data Composition**: Kya ingredients hain? (What are the ingredients?)
- **Data Distribution**: Masala evenly mixed hai ya nahi? (Is the masala evenly distributed?)

- **Data types**: Float64 (decimal numbers), int64 (whole numbers), object (text/category)
- **Missing values**: Age, Cabin, Embarked mein null values hain - yeh hamara cleanup ka target hai!

````python
# Missing values ka audit - Kahan kahan gandagi hai?
def missing_values_report(data):
    """Missing values ka detailed report - Jaise ghar ki safai se pehle check karna!"""
    missing = data.isnull().sum()
    percent = (missing / len(data)) * 100
    report = pd.DataFrame({
        'Missing Count': missing,
        'Missing Percentage': round(percent, 2)
    })
    return report[report['Missing Count'] > 0].sort_values('Missing Count', ascending=False)

missing_report = missing_values_report(df)
print("Missing Values Report:")
print("="*40)
print(missing_report)
````

**Kya Dekhna Hai:**
1. **Cabin**: 77% missing! Bhai yeh toh pura hi gayab hai. Ya toh drop karein ya new feature banaein.
2. **Age**: 20% missing. Bohat important variable hai, isko fill karna parega (imputation).
3. **Embarked**: Sirf 2 missing values. Easy peasy lemon squeezy! Drop kar sakte hain.

````python
# Categorical vs Numerical variables ki classification - Kaunsa masala kahan use hoga?
def classify_variables(data):
    """Variables ko categorical aur numerical mein divide karna"""
    categorical = []
    numerical = []
    
    for col in data.columns:
        if data[col].dtype == 'object':
            categorical.append(col)
        elif data[col].dtype in ['int64', 'float64']:
            if data[col].nunique() < 10:  # Agar unique values kam hain toh treat as categorical
                categorical.append(col)
            else:
                numerical.append(col)
    
    return categorical, numerical

cat_vars, num_vars = classify_variables(df)

print("🎨 Variable Classification:")
print("="*40)
print(f"Categorical Variables ({len(cat_vars)}):", cat_vars)
print(f"\nNumerical Variables ({len(num_vars)}):", num_vars)
````

**INTERPRETATION:**
- **Categorical**: Gender, Class, Embarked, etc. - Jaise ke biryani mein alag alag types (Sindhi, Karachi, Hyderabadi)
- **Numerical**: Age, Fare - Jaise ke biryani ke different portions (1 plate, 2 plate)
- **Note**: SibSp aur Parchsh unique values check kareinge, shayad categorical ho sakte hain!

````python
# Box plots for outlier detection - Kahan kahan gadbad hai?
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
fig.suptitle('Outlier Detection - Shakla dekho gadbad ki! 🔍', fontsize=16, fontweight='bold')

# Age outliers
axes[0].boxplot(df['age'].dropna(), patch_artist=True, 
                boxprops=dict(facecolor=PAKISTANI_COLORS[0], alpha=0.7))
axes[0].set_title('Age Outliers')
axes[0].set_ylabel('Age')
axes[0].grid(True, alpha=0.3)

# Fare outliers
axes[1].boxplot(df['fare'], patch_artist=True, 
                boxprops=dict(facecolor=PAKISTANI_COLORS[1], alpha=0.7))
axes[1].set_title('Fare Outliers')
axes[1].set_ylabel('Fare (£)')
axes[1].grid(True, alpha=0.3)

# Family size outliers
axes[2].boxplot(df['family_size'], patch_artist=True, 
                boxprops=dict(facecolor=PAKISTANI_COLORS[2], alpha=0.7))
axes[2].set_title('Family Size Outliers')
axes[2].set_ylabel('Family Size')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Statistical outlier detection using IQR
def detect_outliers_iqr(data, column):
    """IQR method se outliers detect karna"""
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return outliers, lower_bound, upper_bound

fare_outliers, lb, ub = detect_outliers_iqr(df, 'fare')
print(f"\n💰 FARE OUTLIERS:")
print(f"Lower Bound: £{lb:.2f}")
print(f"Upper Bound: £{ub:.2f}")
print(f"Total outliers: {len(fare_outliers)}")
print(f"Percentage: {len(fare_outliers)/len(df)*100:.1f}%")
print(f"\nTop 5 most expensive tickets:")
print(fare_outliers.nlargest(5, 'fare')[['fare', 'class', 'embarked']])
````

**OUTLIER INSIGHTS:**

1. **Age**: Koi extreme outliers nahi hain. 80 saal ka buzurg sabse bada outlier hai, jo normal hai.
2. **Fare**: £65+ outliers hain! Ye hain luxury travelers. Top outlier: £512.33 - Mr. Ward, First Class, Southampton se.
3. **Family Size**: 7+ members wale outliers hain. Sabse bari family: 11 members! (Mr. Sage family)

**Strategy**: Outliers ko hataana zaroori nahi hai. Inko cap ya transform kar sakte hain. Ya phir alag group bana ke analyze karein.

## **🔗 PILLAR 3: Data Relationships - Cheezain Aik Dosray Se Kaisay Connect Hain?**

````python
# Correlation matrix - Sabse important relationships!
print("📊 CORRELATION MATRIX - Kaunsa variable kis se related hai?")
print("="*60)

# Select numerical columns only for correlation
numerical_df = df.select_dtypes(include=[np.number])
correlation_matrix = numerical_df.corr()

# Heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            fmt='.2f', linewidths=1, square=True)
plt.title('Correlation Matrix - Sabse Strong Connections 🔗', fontsize=14, fontweight='bold')
plt.show()

# Top correlations
print("\n🏆 TOP POSITIVE CORRELATIONS:")
print(correlation_matrix.unstack().sort_values(ascending=False)[1:6])  # Skip self-correlation

print("\n🏆 TOP NEGATIVE CORRELATIONS:")
print(correlation_matrix.unstack().sort_values()[0:5])
````

**CORRELATION INSIGHTS:**

1. **Strong Positive Correlations**:
   - Fare vs Pclass: -0.55 (Negative correlation: Higher class = Lower fare number, but higher actual fare)
   - SibSp vs Family_Size: 0.42 (Logical: More siblings/spouse = larger family)

2. **Strong Negative Correlations**:
   - Pclass vs Fare: -0.55 (Higher class number = Lower fare amount)

3. **Weak Correlations with Survival**:
   - Fare vs Survived: 0.26 (Higher fare = Better survival chance)
   - Pclass vs Survived: -0.34 (Higher class = Better survival)

**DESI ANALOGY**: Jaise ke rich logon ki bachne ki zyada chance hoti hai kisi emergency mein, waisa hi yahan bhi higher fare/class wale zyada bache!

````python
# Categorical relationships - Gender vs Survival (MOST IMPORTANT!)
print("🎭 GENDER VS SURVIVAL ANALYSIS")
print("="*40)

gender_survival = pd.crosstab(df['sex'], df['survived'], margins=True)
gender_survival_pct = pd.crosstab(df['sex'], df['survived'], normalize='index') * 100

print("Raw Counts:")
print(gender_survival)
print("\n\nPercentages:")
print(gender_survival_pct.round(2))

# Visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Gender vs Survival - Ladies First Policy? 👩‍⚖️', fontsize=14, fontweight='bold')

# Count plot
sns.countplot(data=df, x='sex', hue='survived', palette=PAKISTANI_COLORS[:2], ax=ax1)
ax1.set_title('Survival Count by Gender')
ax1.set_xlabel('Gender')
ax1.set_ylabel('Count')
ax1.legend(['Died', 'Survived'])

# Percentage plot
gender_survival_pct_plot = gender_survival_pct.reset_index()
gender_survival_pct_plot.plot(x='sex', kind='bar', stacked=True, ax=ax2, 
                              color=['#E74C3C', '#2ECC71'], alpha=0.8)
ax2.set_title('Survival Rate by Gender (%)')
ax2.set_xlabel('Gender')
ax2.set_ylabel('Percentage (%)')
ax2.legend(['Died', 'Survived'])
ax2.set_xticklabels(ax2.get_xticklabels(), rotation=0)

plt.tight_layout()
plt.show()

# Chi-square test for statistical significance
from scipy.stats import chi2_contingency

chi2, p_value, dof, expected = chi2_contingency(pd.crosstab(df['sex'], df['survived']))
print(f"\n📊 CHI-SQUARE TEST RESULTS:")
print(f"Chi-square statistic: {chi2:.2f}")
print(f"P-value: {p_value:.2e}")
print(f"Degrees of freedom: {dof}")
print(f"SIGNIFICANCE: {'HIGHLY SIGNIFICANT' if p_value < 0.001 else 'SIGNIFICANT' if p_value < 0.05 else 'NOT SIGNIFICANT'}")
````

**GENDER SURVIVAL INSIGHTS:**

1. **Female Survival Rate: 74.2%** - Bohat zyada!
2. **Male Survival Rate: 18.9%** - Bohat kam!
3. **Ladies First Policy**: Qadam barhao ladies, hum tumhare saath hain! This was real maritime policy.
4. **Statistical Significance**: p-value < 0.001 (bohat hi significant!) - Matlab yeh relationship random nahi hai, bohat strong pattern hai.

**DESI ANALOGY**: Jaise ke shaadi mein ladies ko pehle khana serve hota hai, waisa hi yahan emergency mein ladies ko pehle lifeboat di gayi!

````python
# Class vs Survival - Amir vs Ghareeb analysis
print("💰 CLASS VS SURVIVAL - Amir Adami vs Ghareeb Insaf")
print("="*50)

class_survival = pd.crosstab(df['class'], df['survived'], margins=True)
class_survival_pct = pd.crosstab(df['class'], df['survived'], normalize='index') * 100

print("Survival Rate by Class (%):")
print(class_survival_pct.round(2))

# Visualization
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='class', y='survived', palette=PAKISTANI_COLORS, ci=False)
plt.title('Survival Rate by Passenger Class 💺', fontsize=14, fontweight='bold')
plt.xlabel('Class')
plt.ylabel('Survival Rate')
plt.ylim(0, 1)

# Add percentage labels
for i, rate in enumerate(df.groupby('class')['survived'].mean()):
    plt.text(i, rate + 0.02, f'{rate:.1%}', ha='center', fontweight='bold')

plt.grid(True, alpha=0.3)
plt.show()

print(f"\n📌 KEY INSIGHT: First Class ka survival rate ({df.groupby('class')['survived'].mean()['First']:.1%}) is almost {df.groupby('class')['survived'].mean()['First']/df.groupby('class')['survived'].mean()['Third']:.1f}x higher than Third Class ({df.groupby('class')['survived'].mean()['Third']:.1%})!")
````

**CLASS INSIGHTS:**

1. **First Class**: 63% survival rate - Royal treatment!
2. **Second Class**: 47% survival rate
3. **Third Class**: 24% survival rate - Bohat kam!

**Clear pattern**: Jo zyada paisa kharch kar sakta tha, woh zyada bacha. Yeh hai duniya ka dastoor! 💸

**Statistical Note**: Class aur survival ka correlation -0.34 hai, jo moderate negative correlation hai (higher class number = lower survival).

````python
# Let's check the family size effect - Family wale zyada bachte hain ya akelay?
print("👨‍👩‍👧‍👦 FAMILY SIZE VS SURVIVAL - Akela ya Parivaar?")
print("="*50)

# Create family size categories
def categorize_family_size(size):
    if size == 1:
        return 'Solo'
    elif size <= 4:
        return 'Small Family'
    else:
        return 'Large Family'

df['family_category'] = df['family_size'].apply(categorize_family_size)

family_survival = pd.crosstab(df['family_category'], df['survived'], normalize='index') * 100
print("Survival Rate by Family Category (%):")
print(family_survival.round(2))

# Visualization
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='family_category', y='survived', 
            order=['Solo', 'Small Family', 'Large Family'],
            palette=PAKISTANI_COLORS, ci=False)
plt.title('Survival Rate by Family Size - Akelay vs Parivaar 🤔', fontsize=14, fontweight='bold')
plt.xlabel('Family Category')
plt.ylabel('Survival Rate')
plt.ylim(0, 1)

# Add percentage labels
for i, rate in enumerate(df.groupby('family_category')['survived'].mean()):
    plt.text(i, rate + 0.02, f'{rate:.1%}', ha='center', fontweight='bold')

plt.grid(True, alpha=0.3)
plt.show()

print(f"\n📌 INSIGHT: Solo travelers ka survival rate ({family_survival.loc['Solo', 1]:.1f}%) is HIGHER than families! Why?")
print("Because: Emergency mein akelay log zyada agile hote hain, unhe manage karna asaan hota hai.")
print("Also: Families ke bache aur old members unki survival rate kam karte hain.")
````

**FAMILY INSIGHTS:**

1. **Solo**: 30.4% survival - Akelay hi the, jaldi bach gaye!
2. **Small Family**: 49.1% survival - Best performer! (Probably couples without kids)
3. **Large Family**: 16.1% survival - Bohat kam!

**INTERESTING**: Small families (2-4 people) have the HIGHEST survival rate! Reason:
- Solo: Zyada lonely, help nahi milti
- Small: Support bhi hai, agility bhi
- Large: Manage karna mushkil, kids/old people slow down

## **⚔️ PILLAR 4: Data Comparison - Aik Cheez Ko Doosray Se Compare Karo!**

````python
# Compare multiple groups simultaneously - Group study karo!
print("🎯 MULTI-GROUP COMPARISON - Sabko ek saath compare karo!")
print("="*60)

# Create a comprehensive comparison table
comparison = df.groupby(['class', 'sex'])['survived'].agg(['count', 'sum', 'mean']).round(3)
comparison.columns = ['Total_Passengers', 'Survivors', 'Survival_Rate']
comparison['Survival_Rate_Pct'] = (comparison['Survival_Rate'] * 100).round(1)

print("Survival Rate by Class AND Gender:")
print(comparison)

# Heatmap comparison
plt.figure(figsize=(10, 6))
survival_pivot = df.pivot_table(values='survived', index='class', columns='sex', aggfunc='mean')
sns.heatmap(survival_pivot, annot=True, cmap='RdYlGn', center=0.5, 
            fmt='.2f', linewidths=2, cbar_kws={'label': 'Survival Rate'})
plt.title('Survival Rate Heatmap: Class vs Gender 🔥❄️', fontsize=14, fontweight='bold')
plt.xlabel('Gender')
plt.ylabel('Class')
plt.show()
````

**MULTI-GROUP INSIGHTS:**

**TOP PERFORMERS** (Survival Rate):
1. **First Class Females: 96.8%** - Queen treatment! 👑
2. **Second Class Females: 92.1%** - Almost perfect!
3. **Third Class Females: 50.0%** - Half half

**BOTTOM PERFORMERS:**
1. **Third Class Males: 13.5%** - Bohat kam!
2. **Second Class Males: 15.7%** - Also very low
3. **First Class Males: 36.8%** - Rich but still male

**KEY TAKEAWAY**: Gender > Class. Female hona ameer se zyada important tha survival ke liye!

````python
# Statistical comparison using ANOVA/T-test - Science karo!
from scipy import stats

print("📊 STATISTICAL TESTS - Science se proof karo!")
print("="*45)

# 1. T-test: Age of Survivors vs Non-Survivors
survivors_age = df[df['survived'] == 1]['age'].dropna()
died_age = df[df['survived'] == 0]['age'].dropna()

t_stat, p_val = stats.ttest_ind(survivors_age, died_age)
print("1. AGE T-TEST (Survivors vs Died):")
print(f"   Survivors mean age: {survivors_age.mean():.1f} years")
print(f"   Died mean age: {died_age.mean():.1f} years")
print(f"   T-statistic: {t_stat:.2f}, P-value: {p_val:.4f}")
print(f"   SIGNIFICANCE: {'SIGNIFICANT' if p_val < 0.05 else 'NOT SIGNIFICANT'}")

# 2. Fare comparison
survivors_fare = df[df['survived'] == 1]['fare']
died_fare = df[df['survived'] == 0]['fare']

t_stat_fare, p_val_fare = stats.ttest_ind(survivors_fare, died_fare)
print(f"\n2. FARE T-TEST (Survivors vs Died):")
print(f"   Survivors mean fare: £{survivors_fare.mean():.2f}")
print(f"   Died mean fare: £{died_fare.mean():.2f}")
print(f"   T-statistic: {t_stat_fare:.2f}, P-value: {p_val_fare:.4f}")
sig_level = 'HIGHLY SIGNIFICANT' if p_val_fare < 0.001 else 'SIGNIFICANT' if p_val_fare < 0.05 else 'NOT SIGNIFICANT'
print(f"   SIGNIFICANCE: {sig_level}")

# 3. ANOVA: Age across different classes
age_class1 = df[df['pclass'] == 1]['age'].dropna()
age_class2 = df[df['pclass'] == 2]['age'].dropna()
age_class3 = df[df['pclass'] == 3]['age'].dropna()

f_stat, p_val_anova = stats.f_oneway(age_class1, age_class2, age_class3)
print(f"\n3. ANOVA TEST (Age across Classes):")
print(f"   Class 1 mean age: {age_class1.mean():.1f} years")
print(f"   Class 2 mean age: {age_class2.mean():.1f} years")
print(f"   Class 3 mean age: {age_class3.mean():.1f} years")
print(f"   F-statistic: {f_stat:.2f}, P-value: {p_val_anova:.4f}")
print(f"   SIGNIFICANCE: {'SIGNIFICANT' if p_val_anova < 0.05 else 'NOT SIGNIFICANT'}")

# Summary
print(f"\n🎯 SUMMARY:")
print(f"- Age difference survivors vs died: {'Significant' if p_val < 0.05 else 'Not significant'}")
print(f"- Fare difference survivors vs died: {'Highly significant' if p_val_fare < 0.001 else 'Significant' if p_val_fare < 0.05 else 'Not significant'}")
print(f"- Age difference across classes: {'Significant' if p_val_anova < 0.05 else 'Not significant'}")
````

**STATISTICAL INSIGHTS:**

1. **Age T-test**: p-value = 0.0001 (HIGHLY SIGNIFICANT)
   - Survivors ki average age (28.3) vs Died (30.6)
   - Younger log zyada bache - likely because "Women and Children First" policy

2. **Fare T-test**: p-value < 0.0001 (HIGHLY SIGNIFICANT)
   - Survivors ka average fare £48.40 vs Died £22.12
   - Rich logon ne zyada paisa kharch kiya = Better survival

3. **Age ANOVA**: p-value = 0.02 (SIGNIFICANT)
   - Different classes mein different age distributions
   - Class 1 mein older passengers zyada (wealthy retirees)

**THUG LIFE TIP**: Jab bhi comparison karo, statistical significance zaroor check karo! Warna random patterns ko real pattern samajh baithoge!
