```python
import pandas
import requests
from datetime import date
import io
```

### Request na API do Bacen:

A documenta??o de acesso a API est? dispon?vel nos dados abertos do bacen! 

```python
today=date.today()
today=today.strftime("%m-%d-%Y")
req=requests.get(f"https://olinda.bcb.gov.br/olinda/servico/PTAX/versao/v1/odata/CotacaoDolarPeriodo(dataInicial=@dataInicial,dataFinalCotacao=@dataFinalCotacao)?@dataInicial='01-01-1980'&@dataFinalCotacao='{today}'&$format=text/csv&$select=cotacaoCompra,cotacaoVenda,dataHoraCotacao")
```

### Tratamento das informa??es obtidas. Observe que eu utilizo o seguinte c?digo para converter as informa??es da requisi??o:
> df=pandas.read_csv(io.StringIO(req.content.decode('utf-8')))
### Visto que a request nos retorna um texto!

```python

print("O request tem o retorno: ",req.status_code)
if req.status_code==200:
   df=pandas.read_csv(io.StringIO(req.content.decode('utf-8')))
   print("Salvando informa??o!")
   df.rename(columns = {'cotacaoCompra' : 'Cotacao Compra', 'cotacaoVenda' : 'Cotacao Venda','dataHoraCotacao' : 'Data e hora da Cotacao'}, inplace = True)
   df.to_csv("Cota??o Atualizada Dolar_Real.csv", sep=";",index=False, encoding='utf-8', mode="w", decimal=',')

else:
  print("N?o ? poss?vel atualizar dolar Di?rio")

```
