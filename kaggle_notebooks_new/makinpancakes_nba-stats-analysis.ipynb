```python
import scipy.cluster

# This is an attempt to heirarchicaly cluster the above dataset

# Why is there no documentation for this >>>>?????
pairwise_dist_mat = scipy.spatial.distance.pdist(corr)
linkage_matrix = scipy.cluster.hierarchy.linkage(pairwise_dist_mat, 'complete')
idx = scipy.cluster.hierarchy.fcluster(linkage_matrix, 0.8 * pairwise_dist_mat.max(), 'distance')

columns = [df.columns[4:].tolist()[i] for i in list((numpy.argsort(idx)))]
corr = df[columns].corr()
corr.style.background_gradient(cmap='coolwarm')
```

# Working on a baseline Linear Regressor model

Lets work with a baseline model to predict a players points from just their 3P,2P,FT. This should be just 

$$ \text{PTS} = 3*\text{3P} + 2*\text{2P} + 1*\text{FT} $$

- Dependent stats = Points per game
- Independent stats = 3P, 2P, FT

```python
# baseline correlation
columns_considered = ['3P', '2P', 'FT', 'PTS']
```

```python
corr = df[columns_considered].corr()
corr.style.background_gradient(cmap='coolwarm')
```

```python
import sklearn.linear_model
import sklearn.model_selection

data: pandas.DataFrame = df[columns_considered]
x : pandas.DataFrame = data[['3P', '2P', 'FT']]
y : pandas.Series = data['PTS']
train_x, test_x, train_y, test_y = sklearn.model_selection.train_test_split(x, y, test_size = 0.33, random_state=42)

regressor : sklearn.linear_model.LinearRegression = sklearn.linear_model.LinearRegression().fit(train_x, train_y)
regressor.coef_
```

## Notes

With this basic regressor we can see how it learnt the correct linear formula to calculate out for points in the game.

```python
regressor.score(test_x, test_y) # makes sense
```
