**1. Loading the Data**

This section loads the dataset using pandas.read_csv() and provides a brief overview of the data.
> Цей блок завантажує набір даних за допомогою функції pandas.read_csv() і надає короткий огляд даних.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import pairwise_distances
from sklearn.preprocessing import LabelEncoder

df = pd.read_csv('/kaggle/input/heart-attack-dataset-tarik-a-rashid/Medicaldataset.csv')
```

```python
df.info()
```

**2. Encoding Categorical Data**

The Result column is encoded to numerical values using LabelEncoder to convert 'positive' to 1 and 'negative' to 0.
> Стовпець Result кодується в числові значення за допомогою LabelEncoder, щоб перетворити 'positive' на 1, а 'negative' на 0.

```python
to_encode = ['Result']
df[to_encode] = df[to_encode].apply(LabelEncoder().fit_transform)
```

**3. Checking the First Few Rows**

Display the first few rows of the dataset to inspect the structure.
> Виводимо перші кілька рядків набору даних, щоб перевірити його структуру.

```python
df.head()
```

**4. Correlation Calculation Before Dropping Columns**

Compute the correlation matrix before dropping columns and then sort the correlation with respect to the target column 'Result'.

> Обчислюємо кореляційну матрицю до видалення стовпців і сортуємо кореляцію щодо цільового стовпця 'Result'.

```python
correlation_matrix = df.corr()
correlation_with_result = correlation_matrix['Result'].sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=correlation_with_result.values, y=correlation_with_result.index, color='blue')
plt.title("Кореляція ознак з Result (До видалення стовпців)")
plt.xlabel("Кореляція з Result")
plt.ylabel("Ознаки")
plt.show()
```

**5. Dropping Columns and Recalculating Correlation**

Drop unnecessary columns like 'Heart rate', 'Diastolic blood pressure', etc., and recalculate the correlation matrix for the remaining features.

> Видаляємо непотрібні стовпці, такі як 'Heart rate', 'Diastolic blood pressure' тощо, і перераховуємо кореляційну матрицю для залишкових ознак.

```python
df.drop(['Heart rate', 'Diastolic blood pressure', 'Systolic blood pressure', 'Blood sugar'], axis=1, inplace=True)
```

```python
correlation_matrix_dropped = df.corr()
correlation_with_result_dropped = correlation_matrix_dropped['Result'].sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=correlation_with_result_dropped.values, y=correlation_with_result_dropped.index, color='blue')
plt.title("Кореляція ознак з Result (Після видалення стовпців)")
plt.xlabel("Кореляція з Result")
plt.ylabel("Ознаки")
plt.show()
```

**9. Distance Matrices Calculation**

We compute distance matrices using various distance metrics (Euclidean, Manhattan, and Cosine).

> Обчислюємо матриці відстаней за допомогою різних метрик відстаней (Евклідова, Манхеттенська та Косинусна).

```python
distance_matrix_euclidean = pairwise_distances(df.drop('Result', axis=1), metric='euclidean')
distance_matrix_manhattan = pairwise_distances(df.drop('Result', axis=1), metric='manhattan')
distance_matrix_cosine = pairwise_distances(df.drop('Result', axis=1), metric='cosine')
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

sns.heatmap(distance_matrix_euclidean, ax=axes[0], cmap="coolwarm", cbar=True)
axes[0].set_title("Distance Matrix (euclidean)")

sns.heatmap(distance_matrix_manhattan, ax=axes[1], cmap="coolwarm", cbar=True)
axes[1].set_title("Distance Matrix (manhattan)")

sns.heatmap(distance_matrix_cosine, ax=axes[2], cmap="coolwarm", cbar=True)
axes[2].set_title("Distance Matrix (cosine)")

plt.suptitle("Візуалізація Матриць Відстаней Для Різних Метрик", fontsize=16)

plt.tight_layout()
plt.show()
```

1. Distance Matrix (Euclidean metric): The visualization shows distinct blocks with low distances between values, indicating high similarity between certain data points.

2. Distance Matrix (Manhattan metric): Similar to the Euclidean one, but with a more evenly distributed set of distances, suggesting less clear grouping or relationships between the points.

3. Distance Matrix (Cosine metric): More evenly distributed, showing weaker correlations between points, indicating less obvious dependencies.

> 1. Матриця відстаней (Евклідова метрика): Візуалізація показує чітко виражені блоки з низькими відстанями між значеннями, що вказує на високий рівень схожості між певними точками даних.

> 2. Матриця відстаней (Манхеттенська метрика): Схожа на евклідову, але з більш рівномірним розподілом відстаней, що може свідчити про менш чітке групування або зв’язки між точками.

> 3. Матриця відстаней (Косинусна метрика): Більш рівномірно розподілена, показуючи слабші кореляції між точками, що вказує на менш очевидні залежності.


**10. Correlation Matrices for Different Metrics**

Compute and plot correlation matrices for different methods (Pearson, Spearman, Kendall).

> Обчислюємо та будуємо кореляційні матриці для різних методів (Пірсона, Спірмана, Кендалла).

```python
correlation_pearson = df.corr(method='pearson')
correlation_spearman = df.corr(method='spearman')
correlation_kendall = df.corr(method='kendall')
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

sns.heatmap(correlation_pearson, annot=True, cmap="coolwarm", fmt=".2f", cbar=True, square=True, ax=axes[0])
axes[0].set_title("Pearson Correlation")

sns.heatmap(correlation_spearman, annot=True, cmap="coolwarm", fmt=".2f", cbar=True, square=True, ax=axes[1])
axes[1].set_title("Spearman Correlation")

sns.heatmap(correlation_kendall, annot=True, cmap="coolwarm", fmt=".2f", cbar=True, square=True, ax=axes[2])
axes[2].set_title("Kendall Correlation")

plt.suptitle("Correlation Matrices for Different Metrics", fontsize=16)

plt.tight_layout()
plt.show()
```

1. Pearson Correlation: Shows weak positive correlations between age, cardiac biomarkers (CK-MB, Troponin), and the result, but all correlation values are low, indicating a lack of strong linear relationships between these variables.

2. Spearman Correlation: A high correlation is observed between Troponin and Result (0.66), indicating a strong monotonic relationship between troponin levels and the result. Other correlations are weak.

3. Kendall Correlation: Here, a moderate correlation is seen between Troponin and Result (0.55), confirming the previous metric observations. Overall, the correlations are still relatively low.

> 1. Пірсонова кореляція: Відображає слабкі позитивні кореляції між віком, серцевими біомаркерів (CK-MB, Troponin) і результатом, але всі значення кореляції є низькими, що вказує на відсутність сильної лінійної залежності між цими змінними.
>
> 2. Спірманова кореляція: Висока кореляція спостерігається між Troponin та Result (0.66), що показує сильний монотонний зв'язок між рівнем тропоніну та результатом. Інші кореляції слабкі.
>
> 3. Кендаллова кореляція: Тут спостерігається помірна кореляція між Troponin та Result (0.55), що підтверджує спостереження з попередніх метрик. Загалом, кореляції в цьому методі також незначні.
