```python
import numpy
import pandas
```

```python
data = pandas.read_csv('/kaggle/input/nesarc-huit/nesarc.csv', low_memory=False)
```

```python
data['S4AQ4A18'] = pandas.to_numeric(data['S4AQ4A18'], errors='coerce')
```

```python
sub1=data[(data['AGE']>=18) & (data['AGE']<=25)]
```

```python
sub1['S4AQ4A18'].value_counts()
```

```python
# Tạo một dataframe mới sub2 bằng cách loại bỏ các dòng có giá trị S4AQ4A18 = 9
sub2 = sub1[sub1['S4AQ4A18'] != 9.0].copy()
```

```python
sub2['S4AQ4A18'].value_counts()
```

```python
sub2['MAJORDEPLIFE'].value_counts()
```

```python
# Thay thế giá trị 2 bằng 0 trong cột S4AQ4A18
sub2['S4AQ4A18'].replace(2.0, 0.0, inplace=True)
sub2['S4AQ4A18'].value_counts()
```

# Kiểm định Chi Square

```python
# Tạo bảng tần số chéo (contingency table)
contingency_table = pd.crosstab(sub2['S4AQ4A18'], sub2['MAJORDEPLIFE'])
contingency_table
```

```python
# Thực hiện kiểm định Chi-square
from scipy.stats import chi2_contingency
chi2, p, dof, expected = chi2_contingency(contingency_table)
```

```python
# In kết quả
print("Chi-square value:", chi2)
print("p-value:", p)
print("Degrees of freedom:", dof)
print("Expected frequencies:\n", expected)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Vẽ biểu đồ cột sử dụng catplot
sns.catplot(
    x='MAJORDEPLIFE',  # Trục X là MAJORDEPLIFE
    y='S4AQ4A18',    # Chia cột theo giá trị của S4AQ4A18
    data=sub2,         # Sử dụng dữ liệu từ dataframe sub2
    kind='bar',        # Loại biểu đồ là bar chart
    palette='Set2',    # Bảng màu Set2
    height=5,          # Chiều cao biểu đồ
    aspect=1,           # Tỉ lệ kích thước biểu đồ
    errorbar=None
)

# Thiết lập tiêu đề và nhãn
plt.xlabel('Trầm cảm')
plt.ylabel('Tỷ lệ nghĩ đến cái chết')

# Hiển thị đồ thị
plt.show()
```
