## Coordinates And GeoDataFrame

Creating geodataframe, a data structure used for working with geospatial data in Python.

```python
asian_data_w_coordinates = geopandas.GeoDataFrame(
    asian_data, geometry=geopandas.points_from_xy(asian_data.Longitude, asian_data.Latitude), crs="EPSG:4326"
)
asian_data_w_coordinates
```

## Create scatterplot on the map based on the coordinates

```python
fig = px.scatter_geo(asian_data_w_coordinates,
                    lat=asian_data_w_coordinates.geometry.y,
                    lon=asian_data_w_coordinates.geometry.x,
                    color="Rate",
                    hover_name="Location",
                    template="plotly_dark",
                    color_continuous_scale=px.colors.sequential.Reds,
                    title="Myanmar Has The Highest Intentional Homicide Rate In Asia")
lat_foc = 21.916221
lon_foc = 95.955974
fig.update_layout(
        geo = dict(
            projection_scale=2.5, 
            center=dict(lat=lat_foc, lon=lon_foc), 
        ))
fig.update_traces(marker={'size': 9})
fig.show()
```
