## Iris dataset

```python
import os
import numpy as np
import tensorflow as tf
import random
from keras.models import Sequential
from keras.layers import Dense
from sklearn.preprocessing import OneHotEncoder
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from keras.optimizers import Adam

# Set seed for reproducibility
def set_seed(seed_value=42):
    os.environ['PYTHONHASHSEED'] = str(seed_value)
    random.seed(seed_value)
    np.random.seed(seed_value)
    tf.random.set_seed(seed_value)

# Set the seed
set_seed(42)

# Load the dataset
dataset = load_iris()
features = dataset.data
y = dataset.target.reshape(-1, 1)

# One-hot encode the labels
encoder = OneHotEncoder()
targets = encoder.fit_transform(y).toarray()

# Split the dataset
train_features, test_features, train_targets, test_targets = train_test_split(
    features, targets, test_size=0.2, random_state=5
)

# Define the model
model = Sequential()
model.add(Dense(10, input_dim=4, activation='relu'))
model.add(Dense(3, activation='softmax'))

# Compile the model
model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])

# Show model summary
model.summary()

# Train the model
model.fit(train_features, train_targets, epochs=10, batch_size=20, verbose=1)

# Evaluate the model
loss, accuracy = model.evaluate(test_features, test_targets)

print("Accuracy on the test dataset: %.2f" % accuracy)
```

## Cifar10 dataset

```python
from keras.models import Sequential
from keras.layers import Dense
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from keras.optimizers import Adam
from keras.utils import to_categorical
from keras.optimizers import SGD
from matplotlib import pyplot as plt
from keras.datasets import cifar10

# Set seed for reproducibility
def set_seed(seed_value=42):
    os.environ['PYTHONHASHSEED'] = str(seed_value)
    random.seed(seed_value)
    np.random.seed(seed_value)
    tf.random.set_seed(seed_value)

# Set the seed
set_seed(42)
```

```python
(x_train,y_train),(x_test,y_test) = cifar10.load_data()

class_names = ['airplane','automobile','bird','cat','deer','dog','frog','horse','ship','truck']
```

```python
fig = plt.figure()
for i in range(0,25):
  ax = plt.subplot(5,5,i+1)
  plt.imshow(x_train[i])
  code = y_train[i]
  print("{}-->{}".format(class_names[int(code)],code))
```

```python
print("X_train: {},{}".format(x_train.shape,x_train.dtype))
print("Y_train: {}".format(y_train.shape))
print("X_test: {}".format(x_test.shape))
print("Y_test: {}".format(x_test.shape))
```

```python
nTrain = x_train.shape[0]
nDimTrain = x_train.shape[1]*x_train.shape[2]*x_train.shape[3]
nTest = x_test.shape[0]
print("#Train: {},#Test:{}, nDim:{}".format(nTrain,nTest,nDimTrain))
```

```python
x_train = x_train.reshape(nTrain,nDimTrain)
x_test = x_test.reshape(nTest,nDimTrain)
print("# reshape")
print("X_train: {}".format(x_train.shape))
print("Y_train: {}".format(y_train.shape))
print("X_test: {}".format(x_test.shape))
print("Y_test: {}".format(y_test.shape))
```

```python
y_train = to_categorical(y_train,10)
y_test = to_categorical(y_test,10)

print(y_train[1])

print("X_train: {}".format(x_train.shape))
print("Y_train: {}".format(y_train.shape))
print("X_test: {}".format(x_test.shape))
print("Y_test: {}".format(y_test.shape))
```

```python
# Perceptron training
model =Sequential()
model.add(Dense(units=10,activation="softmax",input_shape=(nDimTrain,)))
# Compile the model
epochs = 25
lrate = 0.01
decay = lrate/epochs
sgd = SGD(learning_rate=lrate, momentum=0.9, decay=decay, nesterov=False)
model.compile(loss='categorical_crossentropy',
                  optimizer=sgd,
                  metrics=['accuracy'])
print(model.summary())

# Train the model
model.fit(x_train / 255.0, y_train,
              batch_size=128,
              shuffle=True,
              epochs=epochs)

# Evaluate the model
scores = model.evaluate(x_test / 255.0, y_test)

print('Loss: %.3f' % scores[0])
print('Accuracy: %.3f' % scores[1])
```
