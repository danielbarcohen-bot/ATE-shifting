```python
import numpy as np # ??Numpy
import pandas as pd # ??Pandas
import os # ??OS
import cv2 # ??Open CV???

print(os.listdir('../input/flowers-recognition/flowers')) #??????
daisy_dir='../input/flowers-recognition/flowers/daisy' #????
rose_dir='../input/flowers-recognition/flowers/rose' #????
sunflower_dir='../input/flowers-recognition/flowers/sunflower' #?????
tulip_dir='../input/flowers-recognition/flowers/tulip' #?????

X = [] #???
y_label = [] #???
imgsize = 150 #????
# ????????????
def training_data(label,data_dir):
    print ("????:", data_dir) 
    for img in os.listdir(data_dir): #??
        path = os.path.join(data_dir,img) #??+???
        img = cv2.imread(path,cv2.IMREAD_COLOR) #????
        img = cv2.resize(img,(imgsize,imgsize)) #????????
        X.append(np.array(img)) #X???
        y_label.append(str(label)) #y??,?????
# ????????
training_data('daisy',daisy_dir) #????
training_data('rose',rose_dir) #????
training_data('sunflower',sunflower_dir) #?????
training_data('tulip',tulip_dir) #?????
```

```python
import matplotlib.pyplot as plt # ??matplotlib
import random as rdm # ???????
# ??????????????
fig,ax=plt.subplots(5,2) #??
fig.set_size_inches(15,15) #??
for i in range(5):
    for j in range (2):
        r=rdm.randint(0,len(X)) #??????
        ax[i,j].imshow(X[r]) #????
        ax[i,j].set_title('Flower: '+y_label[r]) #????
plt.tight_layout() #??
```

```python
from sklearn.preprocessing import LabelEncoder # ????????
from tensorflow.keras.utils import to_categorical # ??One-hot????
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y_label) # ????
y = to_categorical(y,4) # ??????One-hot??
X = np.array(X) # ?X??????????
```

```python
y #???y?????,??????
```

```python
X = X/255 # ?X?????
```

```python
from sklearn.model_selection import train_test_split # ??????
X_train, X_test, y_train, y_test = train_test_split(X, y, #?????
                             test_size=0.2,random_state=1)
```

```python
from tensorflow.keras import layers # ????? ?1
from tensorflow.keras import models # ?????? ?2
cnn = models.Sequential() # ???? ?3
cnn.add(layers.Conv2D(32, (3, 3), activation='relu', # ????? ?4
                        input_shape=(150, 150, 3))) 
cnn.add(layers.MaxPooling2D((2, 2))) # ????? ?5
cnn.add(layers.Conv2D(64, (3, 3), activation='relu')) # ??? ?6
cnn.add(layers.MaxPooling2D((2, 2))) # ????? ?7
cnn.add(layers.Conv2D(128, (3, 3), activation='relu')) # ??? ?8
cnn.add(layers.MaxPooling2D((2, 2))) # ????? ?9
cnn.add(layers.Conv2D(128, (3, 3), activation='relu')) # ??? ?10
cnn.add(layers.MaxPooling2D((2, 2))) # ????? ?11
cnn.add(layers.Flatten()) # ??? ?12
cnn.add(layers.Dense(512, activation='relu')) # ???? ?13
cnn.add(layers.Dense(4, activation='softmax')) # ????? ?14
cnn.compile(loss='categorical_crossentropy', # ???? ?15
            optimizer='RMSprop', # ???
            metrics=['acc']) # ????
```

```python
from IPython.display import SVG # ??????????????
from tensorflow.keras.utils import model_to_dot # ??model_to_dot??
SVG(model_to_dot(cnn).create(prog='dot', format='svg')) # ??
```

```python
# ??????????????history??
history = cnn.fit(X_train,y_train, #????
                  epochs=30, #????(????)
                  validation_split=0.2) #?????????
```

```python
def show_history(history): # ????????????
    loss = history.history['loss'] #????
    val_loss = history.history['val_loss'] #????
    epochs = range(1, len(loss) + 1) #????
    plt.figure(figsize=(12,4)) # ????
    plt.subplot(1, 2, 1) #??1
    plt.plot(epochs, loss, 'bo', label='Training loss') #????
    plt.plot(epochs, val_loss, 'b', label='Validation loss') #????
    plt.title('Training and validation loss') #??
    plt.xlabel('Epochs') #X???
    plt.ylabel('Loss') #Y???
    plt.legend() #??
    acc = history.history['acc'] #?????
    val_acc = history.history['val_acc'] #?????
    plt.subplot(1, 2, 2) #??2
    plt.plot(epochs, acc, 'bo', label='Training acc') #?????
    plt.plot(epochs, val_acc, 'b', label='Validation acc') #?????
    plt.title('Training and validation accuracy') #??
    plt.xlabel('Epochs') #X???
    plt.ylabel('Accuracy') #Y???
    plt.legend() #??
    plt.show() #??
show_history(history) # ??????
```

```python
result = cnn.evaluate(X_test, y_test) #??????????
print('CNN???????',"{0:.2f}%".format(result[1]))
```
