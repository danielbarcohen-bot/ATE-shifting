```python
import numpy as np
import matplotlib.pyplot as plt
import pandas
import seaborn as sns
import warnings
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
warnings.filterwarnings('ignore')
```

```python
dataset = pandas.read_csv("/kaggle/input/diamonds-prices/Diamonds Prices2022.csv")
dataset
```

```python
dataset = dataset.drop(['Unnamed: 0'], axis = 1)
```

```python
target = dataset['price']
data = dataset.drop(['price'], axis = 1)
m, n = data.shape
print(m, n)
data
```

# Visualizing the data vesrus the target

```python
cols = data.select_dtypes(np.number).columns
plt.figure(figsize = (7, 7))
for i in range(len(cols)):
    plt.subplot(2, 3, i+1)
    plt.plot(data.loc[:, cols[i]], target, '.', color = 'crimson')
    plt.title(cols[i])
plt.show()
```

# Showing the distribution

```python
cols = data.select_dtypes(np.number).columns
for i in range(len(cols)):
    sns.displot(data[cols[i]])
    plt.title(cols[i])
plt.show()
```

# Handling skewed distributions

```python
'''
depth: >=55, <= 70
table: >= 50, <= 70
X: < 3
y: < 3, > 10
z: < 2,> 6
'''
idx = ((data['x'] >= 3) & (data['y'] >= 3) & (data['z'] >= 2) 
       & (data['y'] <= 10) & (data['z'] <= 6) & (data['depth'] >= 55)
      & (data['depth'] <= 70) & (data['table'] <= 70) &(data['table'] >= 50))
target = target.loc[idx]
data = data.loc[idx, :]
```

```python
cols = data.select_dtypes(np.number).columns
for i in range(len(cols)):
    sns.displot(data[cols[i]])
    plt.title(cols[i])
plt.show()
```

```python
cols = data.select_dtypes(np.number).columns
plt.figure(figsize = (30, 20))
for i in range(len(cols)):
    plt.subplot(2, 3, i+1)
    plt.plot(data.loc[:, cols[i]], target, '.', color = 'crimson')
    plt.title(cols[i])
plt.show()
```

# Numerating Categorical Values

```python
mp1 = {'Ideal': 4, 'Good': 1, 'Very Good': 2, 'Fair': 0, 'Premium': 3}
mp2 = {'J': 0, 'I': 1, 'H': 2, 'G': 3, 'F': 4, 'E': 5, 'D': 6}
mp3 = {'I1':0, 'SI2':1, 'SI1':2, 'VS2':3, 'VS1':4, 'VVS2':5, 'VVS1':6, 'IF':7}
data['cut'] = data['cut'].map(mp1)
data['color'] = data['color'].map(mp2)
data['clarity'] = data['clarity'].map(mp3)
```

```python
data
```

```python
data = data.drop(['depth', 'table'], axis = 1)
```

# Scaling the Dataset

```python
data = StandardScaler().fit_transform(data)
```

```python
data
```

```python
X_train, X_test, y_train, y_test = train_test_split(data, target, test_size = 0.25, random_state = 44)
```

# Training and evaluating the Random Forest Regressor model

```python
rf_reg = RandomForestRegressor(n_estimators = 100)
rf_reg.fit(X_train, y_train)
rf_reg.score(X_test, y_test)
```
