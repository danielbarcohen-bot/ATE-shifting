## Categorising

Let's make a new column **Age Category** to store the age ranges. 

?? We're going to compare each value to a lower and upper boundary, and *if* the 'age' falls inbetween them, we return a name for the range - that's an **if** statement!

?? **elif** is the code for an 'else if' statement, where you can ask the computer to check for multiple things, <u> IF the first if statement is not fulfilled.</u> For example here we're checking the age category by category, by asking elif after elif statement, and waiting for one to be true.

```python
# Function to categorise age
def categorise_age(age):
    if age <= 12:
        ageRange = 'Kids'
    elif 13 <= age <= 17:
        ageRange = 'Teens'
    elif 18 <= age <= 25:
        ageRange = '18-25'
    elif 26 <= age <= 35:
        ageRange = '26-35'
    elif 36 <= age <= 50:
        ageRange = '36-50'
    else:
        ageRange = '51+'
    return ageRange
```

?? To see summary statistics, we can perform an aggregate function on columns that we need.
