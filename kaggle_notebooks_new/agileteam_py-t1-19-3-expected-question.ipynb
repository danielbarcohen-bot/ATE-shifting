```python
# ????? ????
import pandas as pd
```

```python
# ??? ???? (datatime?? ??)
df = pd.read_csv("../input/bigdatacertificationkr/basic2.csv", 
                 parse_dates=['Date'])
```

```python
df.head()
```

```python
# ?? ?? ??
df['year'] = df['Date'].dt.year
df['month'] = df['Date'].dt.month
df['day'] = df['Date'].dt.day
df['dayofweek'] = df['Date'].dt.dayofweek
```

```python
# ???? 1? sales?? 80%? ??

def event_sales(x):
    if x['Events'] == 1:
        x['Sales2'] = x['Sales']*0.8
    else:
        x['Sales2'] = x['Sales']
    return x


df = df.apply(lambda x: event_sales(x), axis=1) #1? ?? row, 0? ?? ??
df.head()
```

```python
# 2022? ?? ?? ? ?? ? ?
cond = df['year'] == 2022
df1 = df[cond]
sale1 = df1.groupby('month')['Sales2'].sum().max()
sale1
```

```python
# 2023? ?? ?? ? ?? ? ?
cond = df['year'] == 2023
df2 = df[cond]
sale2 = df2.groupby('month')['Sales2'].sum().max()
sale2
```
