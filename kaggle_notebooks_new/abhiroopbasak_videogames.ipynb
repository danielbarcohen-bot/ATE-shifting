### Reading Data

This should be kept in mind that the file is diificult to read only with pandas.read_csv(). Some of the faulty rows need to be skipped to avoid the db from collapsing. 

```python
filepath='/kaggle/input/metacritics-best-video-games-of-all-time-2021/Best Video Games of All Time.csv'
```

```python
df = pd.read_csv(filepath, delimiter=';', skiprows=0, low_memory=False)
```

```python
df.info()
```

### Data Analysis

Now that data has been read into columns, we can analze the top contenders on various parameters such as Ranking and Metascore

```python
df.groupby('Platform')['Ranking'].nlargest(10)
```

```python
df.groupby('Platform')['Metascore'].nlargest(10)
```
