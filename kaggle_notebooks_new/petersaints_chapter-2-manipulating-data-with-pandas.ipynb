## 17. Define a Function for Safe Float Conversion
#
Looking at the `.info()` output, the 'budget' column is of type `object`, meaning it might contain non-numeric values.
To perform calculations, we need it to be numeric (specifically, a float, which can handle decimals).
#
This function, `to_float`, tries to convert its input `x` into a float.
- `try:`: Attempts the conversion `float(x)`.
- `except:`: If the conversion fails (e.g., if `x` is text like 'Unknown'), the code inside the `except` block runs.
- `x = np.nan`: If conversion fails, assign `np.nan` (NumPy's Not a Number) to `x`. `NaN` is the standard way to represent missing numerical data in pandas and NumPy.
- `return x`: Returns the converted float value or `np.nan`.

```python
# Define a function named 'to_float'.
def to_float(x):
    # Try to convert the input 'x' to a floating-point number.
    try:
        x = float(x)
    # If the conversion fails (raises an error)...
    except:
        # ...set 'x' to NumPy's Not a Number (NaN) value.
        x = np.nan
    # Return the result (either the float or NaN).
    return x
```

## 18. Apply the Conversion Function to 'budget' Column
#
The `.apply()` method is used to apply a function (like our `to_float`) to each element
in a pandas Series (a column).
We apply `to_float` to the 'budget' column and overwrite the original column with the results.
Any values in 'budget' that couldn't be converted to numbers will now be `NaN`.

```python
# Apply the 'to_float' function to every value in the 'budget' column of 'small_df'.
# The result replaces the original 'budget' column content.
small_df['budget'] = small_df['budget'].apply(to_float)
```

## 19. Explicitly Set Data Type using `.astype()`
#
Although applying the function likely resulted in floats and NaNs, we can explicitly
ensure the column's data type is `float` using the `.astype()` method.
This is sometimes done for clarity or to handle cases where the `apply` function might
not perfectly set the overall column dtype.

```python
# Force the data type of the 'budget' column to be 'float'.
small_df['budget'] = small_df['budget'].astype('float')
```

## 21. Convert 'release_date' to Datetime Objects
#
The 'release_date' column is currently `object` (likely text). To perform date-based operations
(like extracting the year), we need to convert it to a proper datetime format.
#
- `pd.to_datetime()`: This pandas function is specifically designed to parse strings into datetime objects.
- `errors='coerce'`: This is crucial. If the function encounters a date string it cannot understand (e.g., badly formatted or non-date text), instead of stopping with an error, it will put `NaT` (Not a Time) in that position. `NaT` is the datetime equivalent of `NaN`.

```python
# Convert the 'release_date' column from text strings to pandas datetime objects.
# If a date cannot be parsed, it will be replaced with NaT (Not a Time).
small_df['release_date'] = pd.to_datetime(small_df['release_date'], errors='coerce')
```

## 22. Extract Year from Release Date
#
Now that 'release_date' contains datetime objects, we can easily extract parts like the year.
We create a new column called 'year'.
#
- `.apply(lambda x: ...)`: We use `.apply()` again with a `lambda` function (a small, anonymous function). This function takes a single value `x` (a datetime object or `NaT`) from the 'release_date' column.
- `if x is not pd.NaT`: Check if the value is not `NaT`.
- `str(x).split('-')[0]`: If it's a valid date, convert it to a string (e.g., '1995-10-30 ...'), split it by the hyphen '-', and take the first part (`[0]`), which is the year ('1995').
- `else np.nan`: If the value is `NaT` (meaning the original date was invalid), assign `np.nan` to the 'year' column for that row.

```python
# Create a new column 'year'. For each row:
# If 'release_date' is a valid date (not NaT), convert it to string, split by '-', take the first part (year).
# If 'release_date' is NaT, put NaN in the 'year' column.
small_df['year'] = small_df['release_date'].apply(lambda x: str(x).split('-')[0] if x is not pd.NaT else np.nan)
```

## 25. Handle Missing Revenue Values (`.fillna()`)
#
Numerical columns like 'revenue' might contain missing values (`NaN`).
For certain operations like filtering or calculations, it might be desirable to replace these NaNs.
The `.fillna(0)` method replaces all `NaN` values in the 'revenue' column with `0`.
This assumes that a missing revenue likely means the movie made $0 or the data is unavailable,
and we choose to treat it as $0 for this analysis.

```python
# Replace any missing values (NaN) in the 'revenue' column with the value 0.
small_df['revenue'] = small_df['revenue'].fillna(0)
```

## 26. Filtering Data (Boolean Indexing)
#
We can select rows based on conditions. This is often called boolean indexing.
- `small_df['revenue'] > 1e9`: This creates a boolean Series (True/False values). It's `True` for rows where revenue is greater than 1 billion (1e9 is scientific notation for 1 * 10^9) and `False` otherwise.
- `small_df[...]`: When we put this boolean Series inside the square brackets `[]` of the DataFrame, it selects only the rows where the condition is `True`.
The result is a new DataFrame `new` containing only movies with revenue over $1 billion.

```python
# Create a condition: 'revenue' must be greater than 1 billion (1,000,000,000).
# Select only the rows from 'small_df' where this condition is True.
# Store the result in a new DataFrame called 'new'.
new = small_df[small_df['revenue'] > 1e9]
# Display the 'new' DataFrame containing the billion-dollar movies.
new
```

## 27. Handle Missing Budget Values
#
Similar to revenue, we replace missing budget values (`NaN`) with 0 before applying further filters that involve the budget.

```python
# Replace any missing values (NaN) in the 'budget' column with the value 0.
small_df['budget'] = small_df['budget'].fillna(0)
```

## 28. Filtering with Multiple Conditions
#
We can combine multiple conditions using logical operators:
- `&` means AND (both conditions must be True)
- `|` means OR (at least one condition must be True)
- `~` means NOT (reverses the condition)
#
Here, we want movies where:
1. `small_df['revenue'] > 1e9` (Revenue is greater than 1 billion)
2. `small_df['budget'] < 1.5e8` (Budget is less than 150 million)
#
Each condition must be enclosed in parentheses `()` when combined with `&` or `|`.
The result `new2` contains movies that were highly profitable relative to their budget (at least based on these thresholds).

```python
# Define the first condition: revenue greater than 1 billion.
condition1 = (small_df['revenue'] > 1e9)
# Define the second condition: budget less than 150 million (1.5 * 10^8).
condition2 = (small_df['budget'] < 1.5e8)

# Select rows from 'small_df' where BOTH condition1 AND condition2 are True.
# Store the result in a new DataFrame 'new2'.
new2 = small_df[condition1 & condition2]
# Display the 'new2' DataFrame.
new2
```
