## Removing outliers by IQR method

```python
sorts = data['tripduration'].sort_values()

q1 = sorts.quantile(0.25)
q3 = sorts.quantile(0.75)

iqr = q3-q1

upper = q1-1.5*iqr
lower = q3+1.5*iqr

outliers = sorts[~((sorts<(upper))  |  (sorts>(lower)))]
```

## Boxplot after removing outliers

```python
sns.boxplot(outliers)
plt.show()
```
