## 💡 Research Gaps & Opportunities

1. Gap 1: **Generalization of Utilization and Performance Regression to Unseen GPU Architectures and Workloads**

- **Problem**: Current regression approaches for predicting GPU utilization and performance (such as linear regression on FLOPs or bandwidth, or MLPs on kernel properties) struggle to generalize to **unseen GPU architectures** and **novel/heterogeneous workloads**[1][3]. These models often require access to the hardware or target workload for calibration, limiting utility when deploying to new clouds or hardware generations.
  
- **Current Limitations**:  
  - **Feature engineering** is ad hoc and generally hand-crafted, making the models brittle to architectural or workload changes.
  - Models often **do not encode domain constraints** (e.g., hardware utilization limits, tile-level dependencies, microarchitecture behaviors), leading to over/under-prediction and high error, especially on new workloads or hardware[3].
  - Existing regression models do not naturally encode *uncertainty* or *confidence* intervals, which hinders robust decision making for cost and scheduling.

- **Proposed Solution**:  
  - Develop **domain-adaptive regression architectures** that incorporate both *hardware-level invariants* (e.g., laws of utilization scaling, memory bottlenecks) and *meta-learning components* to enable transfer to new GPU architectures with minimal calibration.
  - Integrate **uncertainty estimation** (e.g., Bayesian regression, quantile regression, or ensemble techniques) to quantify prediction confidence for dynamic scheduling and pricing.
  - Apply **automated feature engineering** leveraging hardware counters, program ASTs, and representation learning to move beyond hand-crafted features and capture underlying structure.

- **Expected Impact**:  
  - Reduce error rates by at least **30-50%** for unseen GPU/workload pairs compared to prior regression baselines[3].
  - Enable zero-shot or low-shot transfer to new data center GPUs, leading to measurable improvements in cost prediction and bin-packing reliability.

- **Difficulty**: Advanced

---

2. Gap 2: **Integrated Modeling of Utilization, Cost, and Bin-Packing in a Unified Predictive Optimization Pipeline**

- **Problem**: Research for GPU utilization prediction (regression), cost optimization (pricing models), and scheduling (bin packing) commonly treats these as *separate* tasks. There is no principled approach to **jointly regress utilization, forecast costs, and optimize scheduling performance** with all constraints and feedback in one pipeline, especially in the presence of missing, partial, or noisy data.

- **Current Limitations**:  
  - Existing works often model **utilization and cost independently**, leading to misaligned optimizations—e.g., a schedule optimal for predicted utilization might not minimize costs due to mispredictions under price tiers or burst billing[1].
  - **Missing value handling** and *imputation* for resource traces is simplistic (e.g., mean imputation), which can bias downstream estimates in bin packing.
  - **Feature dependencies** across utilization, pricing, and scheduling are under-explored (e.g., how memory pressure and utilization co-vary with spot vs. reserved pricing).

- **Proposed Solution**:  
  - Design a **multitask regression framework** where utilization, hourly cost, and scheduling efficiency (bin-packing residuals) are predicted in a coupled fashion, enabling the model to learn interactions and tradeoffs directly.
  - Incorporate **joint uncertainty propagation**, so downstream cost and scheduling models account for uncertainty in resource regression.
  - Use advanced imputation techniques (e.g., model-based or conditional GANs) for missing data in both temporal and feature dimensions to prevent bias in cost and scheduling.

- **Expected Impact**:  
  - Increase overall scheduling efficiency by **10-25%** due to better aligned resource and cost forecasts.
  - Reduce cost wastage from mis-schedules and overprovisioning in the face of regression error or missing data.
  
- **Difficulty**: Intermediate to Advanced

---

3. Gap 3: **Performance and Power Regression under Diverse Micro-Architectural Behaviors and Missingness**

- **Problem**: Utilization alone does not capture **microarchitecture-level variations** in different workloads, and **missing features** (such as unobserved instruction mix, cache behaviors) can significantly degrade regression performance[2]. Thus, regression for power/cost prediction or utilization from trace-based data is highly inaccurate between domains, limiting reliable infrastructure automation.
  
- **Current Limitations**:  
  - Existing regression models oversimplify by **mapping utilization directly to power or cost**, failing for workloads with distinct instruction mixes or cache friendliness (e.g., rendering vs. neural nets can have 30%+ difference in power at the same utilization)[2].
  - Missing bars in utilization/cost traces (data holes) are not well handled, leading to incomplete or biased models, especially in production cloud environments[2].
  - Regression models are not robust to **range gaps**—if some utilization/power/cost levels are underrepresented in training, extrapolation suffers.

- **Proposed Solution**:  
  - Develop **hybrid models** that combine utilization-based regression with workload/microarchitecture fingerprints (e.g., opcode histograms, cache miss rates) for power and cost regression.
  - Use **robust regression and missing data strategies** (e.g., multiple imputation, regression with missingness indicators) to gracefully handle holes in the feature or trace data.
  - Leverage **active learning** to prioritize data collection in underrepresented operational regimes.

- **Expected Impact**:  
  - Boost prediction reliability for cost/power regression by **20-40%** in workflows with missing or incomplete features.
  - Substantially reduce misallocations and outlier power/cost events due to regression errors in production settings.

- **Difficulty**: Intermediate

## 4. Data Preprocessing

**Competition:** manual-tabular-regression

**Note:** Following research-based implementation strategy

```python
# Data Preprocessing
try:
    print('🔧 === DATA PREPROCESSING ===\n')
    
    # Identify all loaded DataFrames from previous cells
    df_vars = {k: v for k, v in globals().items() if isinstance(v, pd.DataFrame)}
    if not df_vars:
        print('❌ No DataFrames found for preprocessing. Please check previous steps.')
    else:
        for name, df in df_vars.items():
            print(f'\n📂 Preprocessing DataFrame: {name}')
            print('-' * 60)
            
            # 1. Drop duplicate rows
            initial_shape = df.shape
            df.drop_duplicates(inplace=True)
            print(f'Dropped {initial_shape[0] - df.shape[0]} duplicate rows.')
            
            # 2. Handle missing values
            missing = df.isnull().sum()
            missing_cols = missing[missing > 0].index.tolist()
            if missing_cols:
                print(f'Missing value columns: {missing_cols}')
                # Numeric: fill with median, Categorical: fill with mode
                for col in missing_cols:
                    if df[col].dtype in ['float64', 'int64']:
                        median_val = df[col].median()
                        df[col].fillna(median_val, inplace=True)
                        print(f'Filled missing numeric "{col}" with median: {median_val}')
                    else:
                        mode_val = df[col].mode(dropna=True)
                        if not mode_val.empty:
                            mode_val = mode_val[0]
                            df[col].fillna(mode_val, inplace=True)
                            print(f'Filled missing categorical "{col}" with mode: {mode_val}')
                        else:
                            df[col].fillna('Unknown', inplace=True)
                            print(f'Filled missing categorical "{col}" with "Unknown"')
            else:
                print('No missing values to fill.')
            
            # 3. Encode categorical features
            cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
            if cat_cols:
                print(f'Categorical columns to encode: {cat_cols}')
                df = pd.get_dummies(df, columns=cat_cols, drop_first=True)
                print(f'Applied one-hot encoding. New shape: {df.shape}')
            else:
                print('No categorical columns to encode.')
            
            # 4. Scale numeric features
            from sklearn.preprocessing import StandardScaler
            num_cols = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
            if num_cols:
                print(f'Numeric columns to scale: {num_cols}')
                scaler = StandardScaler()
                df[num_cols] = scaler.fit_transform(df[num_cols])
                print('Applied standard scaling to numeric features.')
            else:
                print('No numeric columns to scale.')
            
            # 5. Visualize distributions after preprocessing
            print('\n🔹 Numeric Feature Distributions After Scaling:')
            n = min(6, len(num_cols))
            if n > 0:
                fig, axes = plt.subplots(1, n, figsize=(5*n, 4))
                if n == 1:
                    axes = [axes]
                for i, col in enumerate(num_cols[:n]):
                    sns.histplot(df[col], kde=True, ax=axes[i], color='lightgreen')
                    axes[i].set_title(col)
                plt.tight_layout()
                plt.show()
            else:
                print('(No numeric columns to plot)')
            
            print('\n✅ Preprocessing complete for DataFrame:', name)
            # Optionally, update the variable in globals for downstream use
            globals()[name] = df
    
    print('✅ Data Preprocessing complete!')
    
except Exception as e:
    print(f'✗ Error in Data Preprocessing: {e}')
    import traceback
    traceback.print_exc()
```

```python
# Model Architecture
try:
    print('🔧 === MODEL ARCHITECTURE ===\n')

    # Identify all preprocessed DataFrames
    df_vars = {k: v for k, v in globals().items() if isinstance(v, pd.DataFrame)}
    if not df_vars:
        raise ValueError('No preprocessed DataFrames found. Please check preprocessing steps.')

    # Use the largest DataFrame as main (assume it's the main train set)
    main_df_name, main_df = max(df_vars.items(), key=lambda x: x[1].shape[0])
    print(f'Using DataFrame "{main_df_name}" for model architecture. Shape: {main_df.shape}')

    # Identify target column (assume last column if not specified)
    # You may customize this if you know the target column name
    target_candidates = [col for col in main_df.columns if 'target' in col.lower() or 'utilization' in col.lower() or 'price' in col.lower()]
    if target_candidates:
        target_col = target_candidates[0]
    else:
        target_col = main_df.columns[-1]
    print(f'Assuming target column: "{target_col}"')

    # Features and target
    feature_cols = [col for col in main_df.columns if col != target_col]
    X = main_df[feature_cols].values.astype(np.float32)
    y = main_df[target_col].values.astype(np.float32).reshape(-1, 1)

    # Split for validation
    from sklearn.model_selection import train_test_split
    X_train, X_val, y_train, y_val = train_test_split(
        X, y, test_size=0.15, random_state=42
    )
    print(f'Train shape: {X_train.shape}, Validation shape: {X_val.shape}')

    # PyTorch Dataset
    class TabularDataset(Dataset):
        def __init__(self, X, y):
            self.X = torch.from_numpy(X)
            self.y = torch.from_numpy(y)
        def __len__(self):
            return len(self.X)
        def __getitem__(self, idx):
            return self.X[idx], self.y[idx]

    train_ds = TabularDataset(X_train, y_train)
    val_ds = TabularDataset(X_val, y_val)

    batch_size = 256
    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True)
    val_loader = DataLoader(val_ds, batch_size=batch_size, shuffle=False)

    # Model Architecture
    class TabularRegressor(nn.Module):
        def __init__(self, input_dim, hidden_dims=[256, 128, 64], dropout=0.2):
            super().__init__()
            layers = []
            dims = [input_dim] + hidden_dims
            for i in range(len(dims)-1):
                layers.append(nn.Linear(dims[i], dims[i+1]))
                layers.append(nn.BatchNorm1d(dims[i+1]))
                layers.append(nn.ReLU())
                layers.append(nn.Dropout(dropout))
            layers.append(nn.Linear(dims[-1], 1))
            self.model = nn.Sequential(*layers)
        def forward(self, x):
            return self.model(x)

    input_dim = X_train.shape[1]
    model = TabularRegressor(input_dim=input_dim).to(device)
    print(model)

    # Loss and optimizer
    criterion = nn.MSELoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', patience=3, factor=0.5, verbose=True)

    # Training loop
    n_epochs = 20
    best_val_loss = float('inf')
    train_losses, val_losses = [], []

    for epoch in range(1, n_epochs+1):
        model.train()
        running_loss = 0.0
        for xb, yb in train_loader:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad()
            preds = model(xb)
            loss = criterion(preds, yb)
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * xb.size(0)
        train_loss = running_loss / len(train_loader.dataset)
        train_losses.append(train_loss)

        # Validation
        model.eval()
        val_running_loss = 0.0
        with torch.no_grad():
            for xb, yb in val_loader:
                xb, yb = xb.to(device), yb.to(device)
                preds = model(xb)
                loss = criterion(preds, yb)
                val_running_loss += loss.item() * xb.size(0)
        val_loss = val_running_loss / len(val_loader.dataset)
        val_losses.append(val_loss)
        scheduler.step(val_loss)

        print(f'Epoch {epoch:02d}/{n_epochs} | Train Loss: {train_loss:.5f} | Val Loss: {val_loss:.5f}')
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_model_state = model.state_dict().copy()

    # Load best model
    model.load_state_dict(best_model_state)

    # Plot loss curves
    plt.figure(figsize=(8,5))
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('MSE Loss')
    plt.title('Training & Validation Loss')
    plt.legend()
    plt.show()

    # Evaluate on validation set
    model.eval()
    val_preds = []
    val_targets = []
    with torch.no_grad():
        for xb, yb in val_loader:
            xb = xb.to(device)
            preds = model(xb).cpu().numpy()
            val_preds.append(preds)
            val_targets.append(yb.numpy())
    val_preds = np.vstack(val_preds)
    val_targets = np.vstack(val_targets)

    from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
    mse = mean_squared_error(val_targets, val_preds)
    mae = mean_absolute_error(val_targets, val_preds)
    r2 = r2_score(val_targets, val_preds)
    print(f'\nValidation Metrics:')
    print(f'MSE: {mse:.5f}')
    print(f'MAE: {mae:.5f}')
    print(f'R2: {r2:.5f}')

    # Visualize predictions vs. targets
    plt.figure(figsize=(6,6))
    plt.scatter(val_targets, val_preds, alpha=0.2)
    plt.plot([val_targets.min(), val_targets.max()], [val_targets.min(), val_targets.max()], 'r--')
    plt.xlabel('True Values')
    plt.ylabel('Predictions')
    plt.title('Validation: Predictions vs. True Values')
    plt.show()

    print('✅ Model Architecture complete!')

except Exception as e:
    print(f'✗ Error in Model Architecture: {e}')
    import traceback
    traceback.print_exc()
```
