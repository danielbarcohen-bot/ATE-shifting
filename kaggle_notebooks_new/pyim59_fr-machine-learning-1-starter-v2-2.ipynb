
### Visualisation de la distribution avec Seaborn : les boxplots

**Seaborn** est une biblioth?que de visualisation statistique construite au-dessus de Matplotlib.

#### 1. Rappel : principe du boxplot

Un **boxplot** synth?tise une distribution num?rique ? partir de cinq statistiques descriptives :

* minimum
* premier quartile ( Q_1 ) (25 %)
* m?diane ( Q_2 ) (50 %)
* troisi?me quartile ( Q_3 ) (75 %)
* maximum

La bo?te repr?sente l’intervalle interquartile :

[
IQR = Q_3 - Q_1
]

Cet intervalle contient les 50 % centraux des donn?es.


#### 2. Structure graphique

* La bo?te s’?tend de ( Q_1 ) ? ( Q_3 )
* Le trait central correspond ? la **m?diane**
* Les moustaches s’?tendent jusqu’aux valeurs extr?mes non aberrantes

Ce type de graphique permet d’?valuer rapidement :

* la dispersion
* l’asym?trie
* la concentration des donn?es



#### 3. D?finition des valeurs aberrantes

Une observation est consid?r?e comme aberrante si :

[
x < Q_1 - 1.5 \times IQR
]

ou

[

#### D?composition de l’instruction

```python
print(f"Ann?e {year} : capital = {self.capital:.2f}")
```

* `f"..."` : cha?ne format?e
* `{year}` : insertion de la valeur de la variable `year`
* `{self.capital:.2f}` : insertion du capital avec deux d?cimales
* `print` : affiche le texte dans la sortie du notebook

```python
class CompoundInterestCalculator:
    def __init__(self, capital_initial, taux):
        self.capital = capital_initial
        self.taux = taux

    def next_year(self):
        self.capital = self.capital * (1 + self.taux)

    def simulate(self, n):
        for year in range(1, n + 1):
            self.next_year()
            print(f"Ann?e {year} : capital = {self.capital:.2f}")
```

```python
calc = CompoundInterestCalculator(1000, 0.05)
calc.simulate(5)
```
