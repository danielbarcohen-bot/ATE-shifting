# <span style="color:#E888BB; font-size: 1%|;"> Imputation for Missing Values</span>




<div style="border-radius:10px;border:#D2222D solid;padding: 15px;background-color:#ffffff00;font-size:100%;text-align:left">

<strong>1. Numerical Features:</strong>

When dealing with missing values in numerical columns, the KNNImputer can be a valuable tool. The KNNImputer replaces missing values by considering the similarity between samples. It identifies the k-nearest neighbors for each sample with missing values and imputes the missing values based on the average of the neighbors' values.

<strong>2.Categorical Columns:</strong>

While KNNImputer is designed for numerical data, it doesn't directly handle categorical columns. For categorical columns, we can use a separate imputation technique such as SimpleImputer with the 'most_frequent' strategy. This strategy replaces missing categorical values with the most frequently occurring value in each column.

<strong>3. Combining imputation:</strong>

By combining KNNImputer for numerical columns and SimpleImputer for categorical columns, we create a comprehensive strategy for handling missing values in both types of columns in dataset.

    
 </div>  

```python
# Display information about missing values in each column
missing_values = df_train.isnull().sum()

# Identify numerical and categorical columns
numerical_columns = df_train.select_dtypes(include=['number']).columns
categorical_columns = df_train.select_dtypes(include=['object', 'category']).columns

# Filter missing values based on numerical and categorical columns
missing_numerical = missing_values[numerical_columns]
missing_categorical = missing_values[categorical_columns]

# Print the results
print("Numerical columns with missing values:")
print(missing_numerical[missing_numerical > 0])

print("\nCategorical columns with missing values:")
print(missing_categorical[missing_categorical > 0])
```

```python
from sklearn.preprocessing import LabelEncoder

# Identify object-type columns
object_columns = df_train.select_dtypes(include=['object']).columns

# Initialize LabelEncoder
label_encoder = LabelEncoder()

# Convert object-type columns to numerical using Label Encoding
for col in object_columns:
    df_train[col] = label_encoder.fit_transform(df_train[col])
    
# Convert object-type columns to numerical using Label Encoding
# Identify object-type columns
object_columns1 = df_test.select_dtypes(include=['object']).columns
for col1 in object_columns1:
    df_test[col1] = label_encoder.fit_transform(df_test[col1]) 
```

# <span style="color:#E888BB; font-size: 1%|;">  Boosting Technique</span>



```python
import xgboost as xgb

k_fold = RepeatedStratifiedKFold(n_splits=n_folds, n_repeats=repeats, random_state=42) 

modelsXGB = []
predsXGB = []

PATIENCE = 50

xgbr_params = {'n_estimators'     : 2048,
          'max_depth'             : 10,
          'learning_rate'         : 0.05,
          'booster'               : 'gbtree',
          'subsample'             : 0.75,
          'colsample_bytree'      : 0.30,
          'reg_lambda'            : 1.00,
          'reg_alpha'             : 1.00,
          'gamma'                 : 1.00,
          'random_state'          : 42,
          'objective'             : 'binary:logistic',
         # 'tree_method'           : 'gpu_hist',
          'eval_metric'           : 'auc',
          #'early_stopping_rounds' : 256,
          'n_jobs'                : -1,
         }


for train_index, test_index in k_fold.split(X, y):
    X_train, X_valid = X.iloc[train_index], X.iloc[test_index]
    y_train, y_valid = y.iloc[train_index], y.iloc[test_index]
    
    modelxgb = xgb.XGBClassifier(**xgbr_params)
    
    modelxgb.fit(
        X_train, y_train,
        eval_set=[(X_valid, y_valid)],
        early_stopping_rounds=PATIENCE,
        verbose=100
    )
    
    modelsXGB.append(modelxgb)
    predsXGB.append(modelxgb.predict_proba(df_test))  
```
