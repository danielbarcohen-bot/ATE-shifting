```python
import pandas
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pandas.read_csv("/kaggle/input/diabetes-dataset/diabetes.csv")
```

```python
data
```

**Slicing Dataset**

```python
x=data.iloc[:,:-1].values
y=data.iloc[:,-1].values
```

```python
x
```

```python
y
```

# Oversampler

```python
from imblearn.over_sampling import RandomOverSampler
Z=RandomOverSampler()
R_data,P_data=Z.fit_resample(x,y)
```

```python
R_data
```

```python
from collections import Counter
print(Counter(P_data))
```

# Undersampler

```python
from imblearn.under_sampling import RandomUnderSampler
Z=RandomUnderSampler()
R_data,P_data=Z.fit_resample(x,y)
```

```python
from collections import Counter
print(Counter(P_data))
```

# SMOTE

```python
from imblearn.over_sampling import SMOTE
Z=SMOTE()
R_data,P_data=Z.fit_resample(x,y)
```

```python
from collections import Counter
print(Counter(P_data))
```

# Accuracy

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(R_data,P_data,test_size=0.3,random_state=15)
```

**Logistic Regression**

```python
from sklearn.linear_model import LogisticRegression
l1=LogisticRegression()
l1.fit(x_train,y_train)
```

**Y prediction**

```python
y_prashant=l1.predict(x_train)
```

```python
y_prashant
```

**Accuracy** 

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_train,y_prashant)*100
```
