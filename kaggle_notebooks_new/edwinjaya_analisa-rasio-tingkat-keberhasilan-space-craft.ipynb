## Memeriksa Awak Space Craft

Pertama-tama kita akan mencari semua value unik untuk mencari tipe misi yang memiliki crew dengan menggunakan fungsi <code>data.unique()</code>. Seperti yang tampak pada cells dibawah, terdapat beberapa tipe misi yang memiliki awak space craft. Tipe misi tersebut ialah : 
1. Crewed Orbiter
2. Crewed Orbiter/Lander
3. Crewed Orbiter/Lander/Rover

```python
print(mission_data_subset_nan_payload['Mission Type'].unique())
```

```python
total_crewed_success_data=mission_data_subset_nan_payload[
    (mission_data_subset_nan_payload['Outcome']=="Successful") & 
    (mission_data_subset_nan_payload['Mission Type']=="Crewed orbiter") |
    (mission_data_subset_nan_payload['Mission Type']=="Crewed orbiter/lander") | 
    (mission_data_subset_nan_payload['Mission Type']=="Crewed orbiter/lander/rover")]
total_crewed_success_data
```

Dengan membuat sebuah subset yang berisikan kolom dan data yang relevan, kita dapat mencari informasi yang dibutuhakan dengan lebih mudah. Subset yang dibuat hanya berisikan kolom Outcome yang memiliki values "Successful" dan kolom Mission Type yang memiliki values antara "Crewed orbiter", "Crewed orbiter/lander", atau "Crewed orbiter/lander/rover".

Tetapi jika kita cermati, pada urutan ke empat terdapat Spacecraft bernama Apollo 13 yang memiliki"Outcome" bernilai "Spacecraft failure" tetapi termasuk ke dalam filter "Successful". Mengapa? Ini karena dalam misinya, Apollo 13 disebutkan gagal untuk mendarat di bulan tetapi berhasil membawa kembali seluruh krunya dengan selamat setelah terjadi kebocoran gas oksigen.
```python
total_crewed_success_outcome=numpy.product(mission_data_subset_nan_payload[
    (mission_data_subset_nan_payload['Outcome']=="Successful") & 
    (mission_data_subset_nan_payload['Mission Type']=="Crewed orbiter") |
    (mission_data_subset_nan_payload['Mission Type']=="Crewed orbiter/lander") | 
    (mission_data_subset_nan_payload['Mission Type']=="Crewed orbiter/lander/rover")].shape)
total_creworbiter_success_percentage=(total_crewed_success_outcome/total_nan_payload)*100
print("Total Spacecraft Success without payload : {:.2f}%".format(total_creworbiter_success_percentage))
```

## Groupby Dan Convert Dtypes
Langkah berikutnya ialah melakukan convert terlebih dahulu, jika menjalankan fungsi <code>.info()</code> terhadap dataset maka akan terlihat tipe dari variables tersebut. Dikarenakan saya ingin untuk mengambil hari, bulan, dan tahun dari tabel tersebut untuk dijadikan bahan plot maka harus diubah terlebih dahulu ke tipe datetime dari yang awalnya object. Untuk mengubah tipe ke datetime dapat menggunakan salah satu fungsi yang terdapat di dalam pandas yaitu <code>pandas.to_datetime()</code>, lalu menjelaskan format yang diinginkan.

```python
mission_data_success_without_payload.loc[:,"Launch Date"]=pandas.to_datetime(
    mission_data_success_without_payload["Launch Date"],format=("%d %B %Y"))

mission_data_success.loc[:,"Launch Date"]=pandas.to_datetime(
    mission_data_success["Launch Date"],format=("%d %B %Y"))
```

Setelah melakukan convert, saya melakukan groupby berdasarkan sebuah ketentuan yaitu persamaan tahun. Filter tahun ini dilakukan menggunakan <code>.dt.strftime()</code>, dimana didalam <code>.strftime()</code> dijelaskan hal yang kita inginkan yaitu tahun. Setiap tahun yang sama maka akan dimasukan ke tahun yang sesuai, dan dihitung jumlahnya menggunakan fungsi <code>.count()</code>. Lalu data hasil hitungan tersebut diubah kedalam dictionary menggunakan fungsi <code>.to_dict()</code>.

```python
ketentuan_tanpa_payload=mission_data_success_without_payload["Launch Date"].dt.strftime("%Y")
total_tahun_tanpa_payload=mission_data_success_without_payload.groupby(ketentuan_tanpa_payload)["Launch Date"].count().to_dict()
total_tahun_tanpa_payload
```

```python
ketentuan_tahun_pl=mission_data_success[(mission_data_success["Launch Date"]>="1965")
                                    & (mission_data_success["Launch Date"]<"2014")]
ketentuan_dengan_payload=ketentuan_tahun_pl["Launch Date"].dt.strftime("%Y")
total_tahun_dengan_payload=mission_data_success.groupby(ketentuan_dengan_payload)["Launch Date"].count().to_dict()
total_tahun_dengan_payload
```

```python
ketentuan_tahun=mission_data_success_without_payload[(mission_data_success_without_payload["Launch Date"]>="1965")
                                                    & (mission_data_success_without_payload["Launch Date"]<"2014")]
ketentuan_tanpa_payload=ketentuan_tahun["Launch Date"].dt.strftime("%Y")
total_tahun_tanpa_payload=ketentuan_tahun.groupby(ketentuan_tanpa_payload)["Launch Date"].count().to_dict()
total_tahun_tanpa_payload
```

Dikarenakan untuk membuat plot dari dua data diharuskan kedua data tersebut memiliki panjang yang sama, saya melakukan filter terhadap subset data sebelum melakukan groupby.

```python
print(len(total_tahun_tanpa_payload))
print(len(total_tahun_dengan_payload))
```

Lalu menggunakan fungsi <code>.keys()</code> dan <code>.values()</code> untuk mengambil masing-masing kunci dan nilai yang terdapat di dalam dictionary tersebut.

```python
tahun_payload_keys=total_tahun_dengan_payload.keys()
tahun_payload_values=total_tahun_dengan_payload.values()
tahun_tanpa_payload_keys=total_tahun_tanpa_payload.keys()
tahun_tanpa_payload_values=total_tahun_tanpa_payload.values()
```

## Bar chart

Untuk membuat sebuah komparasi menggunakan bar chart dari dua data yang berbeda, kita dapat menggunakan fungsi <code>.subplots()</code> dari matplotlib.

```python
panjang_interval=numpy.arange(0,len(total_tahun_tanpa_payload))
panjang_bar = 0.35

figure , combine_plt = plt.subplots(figsize=(12,10))
figure.tight_layout()
fig, (ax1, ax2) = plt.subplots(1, 2,figsize=(15,5))

tnp_tnp_pl = ax1.bar(panjang_interval,
                    tahun_tanpa_payload_values,
                    panjang_bar,
                    label="Space Craft Tanpa Payload")

tnp_dgn_pl = ax2.bar(panjang_interval,
                    tahun_payload_values,
                    panjang_bar,
                    label="Space Craft Dengan Payload",
                    color=["#ff7f0e"])

tanpa_payload_bar = combine_plt.bar(panjang_interval,
                                 tahun_tanpa_payload_values,
                                 panjang_bar,
                                 label="Space Craft Tanpa Payload"
)

dengan_payload_bar = combine_plt.bar(panjang_interval+panjang_bar,
                                 tahun_payload_values,
                                  panjang_bar,
                                 label="Space Craft Dengan Payload")

ax1.set_ylim([0, 7])
ax2.set_ylim([0, 7])
combine_plt.set_title("Total Keberhasilan Space Craft selama 1965-2013")
combine_plt.set_ylabel("Total Keberhasilan") 
combine_plt.set_xticks([])
ax1.set_title("Total Keberhasilan Space Craft Tanpa Payload Selama 1965-2013")
ax2.set_title("Total Keberhasilan Space Craft Dengan Payload Selama 1965-2013")
ax1.set_ylabel("Total Keberhasilan") 
ax2.set_ylabel("Total Keberhasilan") 
ax1.set_xticks(range(0,len(tahun_tanpa_payload_values)))
ax1.set_xticklabels(tahun_tanpa_payload_keys,rotation="90")
ax2.set_xticks(range(0,len(tahun_payload_values)))
ax2.set_xticklabels(tahun_payload_keys,rotation="90")
combine_plt.legend()
ax1.legend()
ax2.legend()
plt.show()
```

## Pie Chart
Untuk melihat rasio dari keberhasilan space craft dengan dan tanpa payload dapat menggunakan pie chart. Langkah pertama yang dilakukan untuk membuat pie chart tersebut ialah membuat sebuah kolom baru yang berisi hasil dari klasifikasi apakah space craft tersebut termasuk kedalam kategori "Successful With Payload, "Successful Without Payload", atau "None Of Both"/Tidak kedua-duanya. Untuk melakukan hal tersebut saya akan menggunakan fungsi <code>numpy.where()</code>.

```python
mission_data_subset.loc[:,"Successful With Payload"]=numpy.where((mission_data_subset["Payload"].notnull()) 
                                                                 & (mission_data_subset["Outcome"]=="Successful"),True, False)
```

```python
mission_data_subset.loc[:,"Successful Without Payload"]=numpy.where((mission_data_subset["Payload"].isnull()) 
                                                                 & (mission_data_subset["Outcome"]=="Successful"),True, False)
```

Setelah diklasifikasikan, langkah berikutnya adalah menghitung jumlah dari setiap kategori dan memasukannya ke dalam dictionary.

```python
hasil_satu=mission_data_subset[mission_data_subset["Successful With Payload"]==True]["Successful With Payload"].count()
hasil_dua=mission_data_subset[mission_data_subset["Successful Without Payload"]==True]["Successful Without Payload"].count()
hasil_tiga=mission_data_subset[(mission_data_subset["Successful With Payload"]==False) & 
                               (mission_data_subset["Successful Without Payload"]==False)]["Successful Without Payload"].count()

total_hasil={"Successful With Payload":hasil_satu,
             "Successful Without Payload":hasil_dua,
             "None Of Both":hasil_tiga}

print(total_hasil)
```

Lalu menggunakan fungsi <code>plt.pie()</code> untuk membuat plot pie chart berdasarkan data dari dictionary <B>total_hasil</B> di atas.

```python
hasil=total_hasil.values()
label_hasil=total_hasil.keys()
meledak=[0.2,0.2,0]
plt.pie(hasil,
       labels=label_hasil,
       autopct= '%1.1f%%',
        explode=meledak,
       shadow = True)
plt.title('Rasio Keberhasilan Space Craft Dengan Dan Tanpa Payload')
plt.show()
```

Dari pie chart di atas dapat terlihat rasio keberhasilan antara space craft dengan dan tanpa payload, dimana rasio tersebut sesuai dengan perhitungan kita sebelumnya.
