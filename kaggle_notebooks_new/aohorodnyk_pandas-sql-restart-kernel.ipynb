# How to fix an issue with PandaSQL not runnin

To solve the issue we need to install older dependencies for Padnas, but it have to be done in two steps:

1. [Installing dependencies](#Installing-dependencies)
1. [Restarting Kernel](#Restarting-Kernel)
1. [Run the code](#Run-the-code)

> !!! Do not forget to follow the first two steps **Installing dependencies** and **Restarting Kernel** EVERY time you start new server !!!

## Installing dependencies

To install dependencies, create a cell with the content:

```
pip install SQLAlchemy==1.4.49 ipython-sql==0.4.1
```

Run the cell and wait untill it will finish

## Restarting Kernel

When dependencies are successfuly reinstalled, we need to **Restart Kernel**.
To do it, follow these steps:

1. To make it happen, click by mouse's right button on any **Cell** and scroll menu to bottom.
1. Choose **Restart Kernel...** menu item 

## Run the code

Now we are ready to test our fixes.

Use this code to test it:
```
import pandasql

meatdf = pandasql.load_meat()
pandasql.sqldf('SELECT * FROM meatdf', globals())
```

Instead of `load_meat`, feel free to use your DataSet:
```
import pandas
import pandasql


licenseddf = pandas.read_csv('/kaggle/input/licensed-drivers-by-state-gender-and-age-group/Licensed_drivers_By_State.csv')

pandasql.sqldf('SELECT * FROM licenseddf', globals())
```
