```python
def create_improved_model(input_shape):
    model = Sequential()
    model.add(Dense(256, activation='relu', input_shape=input_shape))
    model.add(Dropout(rate=0.2))  
    model.add(Dense(128, activation='relu'))
    #model.add(Dropout(rate=0.1)) 
    model.add(Dense(64, activation='relu'))  
    model.add(Dense(1))    
    
    model.compile(optimizer='adam', loss='mse')
    return model


input_shape = (X_train.shape[1],)
improved_model = create_improved_model(input_shape)


model_checkpoint = ModelCheckpoint(
    filepath="/kaggle/working/best_model.h5",
    monitor="val_loss",
    verbose=1,
    save_best_only=True
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=10,
    verbose=1,
    mode="auto"
)


history = improved_model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=1000,
    batch_size=32,
    verbose=1,
    callbacks=[model_checkpoint, early_stopping]
)
```

```python
def plot_weights_and_biases(model):
    layer_names = []
    weights = []
    biases = []

    for layer in model.layers:
        if isinstance(layer, Dense):
            layer_weights, layer_biases = layer.get_weights()
            layer_names.append(layer.name)
            weights.append(layer_weights)
            biases.append(layer_biases)
    
    
    for i, weight_matrix in enumerate(weights):
        plt.figure(figsize=(10, 4))
        plt.title(f'Weights of Layer {layer_names[i]}')
        plt.hist(weight_matrix.flatten(), bins=50)
        plt.xlabel('Weight Value')
        plt.ylabel('Frequency')
        plt.show()


    for i, bias_vector in enumerate(biases):
        plt.figure(figsize=(10, 4))
        plt.title(f'Biases of Layer {layer_names[i]}')
        plt.hist(bias_vector, bins=50)
        plt.xlabel('Bias Value')
        plt.ylabel('Frequency')
        plt.show()
plot_weights_and_biases(improved_model)
```
