### Categorising

Question 2

Hint 1:

The format of the dictionary should be *\"category\": (lower_limit, upper_limit)*. The function should return the same categories.

Hint 2: 

You might want to iterate over the dictionary items using a [for-loop](https://en.wikipedia.org/wiki/For_loop#:~:text=In%20computer%20science%2C%20a%20for,certain%20condition%20has%20been%20satisfied.).

Answer:

```python
age_categories = {
    'kids': (0, 12),
    'teens': (13, 17),
    '18-25': (18, 25),
    '26-35': (26, 35),
    '36-50': (36, 50),
    '51+': (51, None)
}


def categorise_age(age):
    for category in age_categories: # iterating over age_categories
        lower, upper = age_categories[category] #accessing lower and upper limit 
        if upper is None: #checking if upper limit is none e.g. if we run out of categories and this is the last one
            if age >= lower: 
                return category
        elif lower <= age <= upper:
            return category
    return 'Unknown'

# use the same 'apply' function in Pandas to get an identical result
```

### Calculations

Question 3

Hint:

Find target women of needed age and target men of needed age

Hint 2:

Since the same entry's gender cannot be Male and Female at once, we can't use *and* operator and need to use *or*

Answer:

```python
target_women_age = DF['Age'].between(18, 35)
target_men_age = DF['Age'].between(25, 40)

target_women = DF["Gender"] == "Female"
target_men = DF["Gender"] == "Male"


target_audience = DF[(target_men_age & target_men) | (target_women_age & target_women)]

target_audience.head()
```
