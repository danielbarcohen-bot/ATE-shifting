```python
def convert_emojis(text):
    for emoji, meaning in UNICODE_EMO.items():
        text = text.replace(emoji, meaning)
    return text

def convert_emoticons(text):
    for emoticon, meaning in EMOTICONS.items():
        text = re.sub(emoticon, meaning, text)
        text = re.sub(emoticon.lower(), meaning, text)
    return text

def process_and_tokenize(text):
    """
    1. Replace URLs in text
    2. Replace emojis and emoticons with their meanings
    3. Replace contractions with expanded forms
    4. Tokenize text
    """
    text = re.sub('http[s]?://(?:[a-zA-Z]|[0-9]|[$-_@.&+]|(?:%[0-9a-fA-F][0-9a-fA-F]))+', 'url', text).strip()
    text = convert_emojis(text)
    text = convert_emoticons(text)
    text = contractions.fix(text)
    return list(gensim_tokenize(text, lower=True))

def get_mbti_vector(mbti_type):
    I, N, F, P = mbti_type
    mbti_vector = [I == 'I', N == 'N', F == 'F', P == 'P']
    return numpy.array(mbti_vector).astype(int)

def plot_num_posts_by_indicator(df):
    fig = make_subplots(rows=2, cols=2)
    for i, (one, opp) in enumerate(zip(one_slice, opposite)):
        counts = df.groupby(one).type.count().sort_index()
        row = 1 if i <= 1 else 2
        col = 1 if i in [0, 2] else 2
        fig.append_trace(go.Bar(x=[opp, one], y=counts.tolist()), row=row, col=col)
    fig.update_layout(showlegend=False, title='Number of posts split by each indicator')
    fig.show()

def plot_num_posts_by_length(title, df):
    token_len = df.tokens.apply(len)
    token_len_dist = df.groupby(token_len).type.count()
    fig = px.bar(token_len_dist.reset_index(), x='tokens', y='type', title=title)
    fig.update_layout(xaxis_title="Sentence Length (Num. of tokens)", yaxis_title="Count")
    fig.show()
    return token_len, token_len_dist

def plot_num_posts_by_type(df):
    type_posts = df.groupby('type').posts.count().sort_values(ascending=False).reset_index()
    fig = px.bar(type_posts, x='type', y='posts', title='Number of posts by MBTI type')
    fig.update_layout(xaxis_title='MBTI Type', yaxis_title='Count')
    fig.show()

def get_length_filtered_df(df, token_len, min_tokens):
    return df[token_len >= min_tokens]

def get_token_freq(df):
    token_freq = defaultdict(int)
    for token_list in df.tokens:
        for token in token_list:
            token_freq[token] += 1
    token_freq = pandas.Series(token_freq)
    token_freq_ser = token_freq.groupby(token_freq).count()
    token_freq_ser.name = 'NumTokens'
    token_freq_ser.index.name = 'Freq'
    return token_freq, token_freq_ser

def get_token_analysis(min_freq):
    total_words = (token_freq_ser.index * token_freq_ser).sum()
    total_num_tokens = token_freq_ser.sum()
    invalid_token_freq = token_freq_ser[token_freq_ser.index <= min_freq]
    invalid_total_words = (invalid_token_freq.index * invalid_token_freq).sum()
    invalid_tokens = invalid_token_freq.sum()
    return round(invalid_total_words/total_words*100, 2), round(invalid_tokens/total_num_tokens*100, 2), total_num_tokens - invalid_tokens

def plot_changes():
    min_freqs = list(range(0, 11))
    total_tokens_list, unique_tokens_list, valid_tokens_list = [], [], []
    for min_freq in min_freqs:
        total_tokens, unique_tokens, valid_tokens = get_token_analysis(min_freq)
        total_tokens_list.append(100 - total_tokens)
        unique_tokens_list.append(100 - unique_tokens)
        valid_tokens_list.append(valid_tokens)

    fig = go.Figure()
    fig.add_trace(go.Scatter(x=min_freqs, y=total_tokens_list , name="Percentage of total number of words remaining"))
    fig.add_trace(go.Scatter(x=min_freqs, y=unique_tokens_list, name="Percentage of number of unique tokens remaining"))
    fig.update_layout(showlegend=True, legend=dict(yanchor="top", y=0.75, xanchor="right", x=0.99), xaxis_title="Min. Freq. Count", 
                      yaxis_title="Percentage")
    fig.show()

    fig = go.Figure()
    fig.add_trace(go.Scatter(x=min_freqs, y=valid_tokens_list , name="Valid unique tokens"))
    fig.update_layout(showlegend=True, legend=dict(yanchor="top", y=0.99, xanchor="right", x=0.99), xaxis_title="Min. Freq. Count", 
                      yaxis_title="Count of Valid Unique Tokens")
    fig.show()
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.layers import (
    Dense, LSTM, Dropout, SpatialDropout1D, 
    Embedding, Flatten, Input
)
from tensorflow.keras.models import Sequential, Model
from gensim.models import KeyedVectors
from tqdm import tqdm
import transformers

def load_glove_embeddings(fname, vocab, embedding_dim):
    """Load GloVe embeddings from text file"""
    print(f"Loading GloVe embeddings from {fname}")
    embedding_matrix = np.zeros((len(vocab) + 1, embedding_dim))
    found_words = 0
    
    try:
        with open(f'/kaggle/input/{fname}', 'r', encoding='utf-8') as f:
            for line in tqdm(f, desc='Loading GloVe'):
                values = line.split()
                word = values[0]
                if word in vocab:
                    vector = np.asarray(values[1:], dtype='float32')
                    embedding_matrix[vocab[word]] = vector
                    found_words += 1
                    
        print(f'Found GloVe embeddings for {found_words}/{len(vocab)} words')
        return Embedding(
            input_dim=len(vocab) + 1,
            output_dim=embedding_dim,
            weights=[embedding_matrix],
            trainable=False,
            mask_zero=True,
            name='GloVeEmbedding'
        )
    except Exception as e:
        print(f"Error loading GloVe embeddings: {str(e)}")
        return None

def load_word2vec_embeddings(fname, vocab, embedding_dim):
    """Load Word2Vec embeddings from binary file"""
    print(f"Loading Word2Vec embeddings from {fname}")
    try:
        w2v = KeyedVectors.load_word2vec_format(
            f'/kaggle/input/{fname}', 
            binary=True
        )
        
        embedding_matrix = np.zeros((len(vocab) + 1, embedding_dim))
        found_words = 0
        
        for word, i in tqdm(vocab.items(), desc='Creating W2V matrix'):
            if word in w2v.key_to_index:
                embedding_matrix[i] = w2v.get_vector(word)
                found_words += 1
                
        print(f'Found Word2Vec embeddings for {found_words}/{len(vocab)} words')
        return Embedding(
            input_dim=len(vocab) + 1,
            output_dim=embedding_dim,
            weights=[embedding_matrix],
            trainable=False,
            mask_zero=True,
            name='Word2VecEmbedding'
        )
    except Exception as e:
        print(f"Error loading Word2Vec embeddings: {str(e)}")
        return None

def get_embeddings(vocab, max_tokens):
    """Get all available embeddings"""
    embeddings = []
    
    # Try loading Word2Vec
    w2v = load_word2vec_embeddings(
        'googlenewsvectorsnegative300/GoogleNews-vectors-negative300.bin',
        vocab,
        300
    )
    if w2v is not None:
        embeddings.append(w2v)
    
    # Try loading GloVe
    glove = load_glove_embeddings(
        'glovetwitter27b100dtxt/glove.twitter.27B.100d.txt',
        vocab,
        100
    )
    if glove is not None:
        embeddings.append(glove)
    
    
    return embeddings


def get_model(embedding, sequence, max_tokens, Y, strategy):
    """Create the main model with fixed masking"""
    lr_schedule = tf.keras.optimizers.schedules.ExponentialDecay(
        initial_learning_rate=0.001,
        decay_steps=10000, # 10000
        decay_rate=0.9
    )
    
    with strategy.scope():
        # Create input layer
        inputs = Input(shape=(max_tokens,), dtype=tf.int32, name="WordIDs")
        
        # Add embedding layer
        x = embedding(inputs)
        x = SpatialDropout1D(0.2, name='DroppingEmbeddings')(x)
        
        # Add sequence layer with explicit cudnn settings
        if isinstance(sequence, (LSTM, Bidirectional)) and 'LSTM' in sequence.name:
            modified_sequence = LSTM(
                units=Y,
                return_sequences=True,
                use_bias=True,
                use_cudnn=False,  # Disable CUDNN to avoid masking issues
                name=sequence.name
            )
        elif isinstance(sequence, (GRU, Bidirectional)) and 'GRU' in sequence.name:
            modified_sequence = GRU(
                units=Y,
                return_sequences=True,
                use_bias=True,
                use_cudnn=False,  # Disable CUDNN to avoid masking issues
                name=sequence.name
            )
        else:
            modified_sequence = sequence
            
        if isinstance(sequence, Bidirectional):
            x = Bidirectional(modified_sequence)(x)
        else:
            x = modified_sequence(x)
        
        # Add remaining layers
        x = Flatten(name='FlattenSequenceOutputs')(x)
        x = Dense(
            units=int(np.sqrt(max_tokens * Y)),
            activation='relu',
            name='HiddenLayer'
        )(x)
        x = Dropout(0.2, name='DroppingHiddenNodes')(x)
        outputs = Dense(units=4, activation='sigmoid', name='OutputLayer')(x)
        
        # Create and compile model
        model = Model(inputs=inputs, outputs=outputs, name="Model")
        model.compile(
            loss=tf.keras.losses.BinaryCrossentropy(),
            optimizer=tf.keras.optimizers.Adam(learning_rate=lr_schedule),
            metrics=['binary_accuracy'],
            steps_per_execution=32
        )
    
    return model

```

```python
# Evaluate all models and create detailed results DataFrame
def create_detailed_results(models, x_test, y_test, bx_test):
    """Create detailed evaluation results for all models"""
    results = []
    
    for embed_name, embed_models in models.items():
        for seq_name, model in embed_models.items():
            # Use appropriate test data based on embedding type
            test_data = bx_test if embed_name == 'BERT' else x_test
            
            # Evaluate model
            loss, accuracy = model.evaluate(test_data, y_test, verbose=0)
            
            # Make predictions
            y_pred = model.predict(test_data)
            y_pred_binary = (y_pred > 0.5).astype(int)
            
            # Calculate metrics for each personality dimension
            for i, dim in enumerate(['I/E', 'N/S', 'F/T', 'P/J']):
                dimension_metrics = {
                    'Model': f"{embed_name}+{seq_name}",
                    'Dimension': dim,
                    'Accuracy': (y_pred_binary[:, i] == y_test[:, i]).mean(),
                    'Precision': precision_score(y_test[:, i], y_pred_binary[:, i]),
                    'Recall': recall_score(y_test[:, i], y_pred_binary[:, i]),
                    'F1': f1_score(y_test[:, i], y_pred_binary[:, i])
                }
                results.append(dimension_metrics)
    
    return pd.DataFrame(results)

# Create and display detailed results
detailed_results = create_detailed_results(models, x_test, y_test, bx_test)
print("\nDetailed Model Evaluation Results:")
print(detailed_results.to_string(index=False))

# Create visualization of results
def plot_model_comparison(detailed_results):
    """Create visualizations comparing model performance"""
    # Create figure with subplots
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Accuracy by Model and Dimension',
                       'F1 Score by Model and Dimension',
                       'Precision by Model and Dimension',
                       'Recall by Model and Dimension')
    )
    
    metrics = ['Accuracy', 'F1', 'Precision', 'Recall']
    colors = px.colors.qualitative.Set3
    
    for idx, metric in enumerate(metrics):
        row = idx // 2 + 1
        col = idx % 2 + 1
        
        # Create grouped bar chart for each metric
        for i, model in enumerate(detailed_results['Model'].unique()):
            model_data = detailed_results[detailed_results['Model'] == model]
            
            fig.add_trace(
                go.Bar(
                    name=model,
                    x=model_data['Dimension'],
                    y=model_data[metric],
                    marker_color=colors[i],
                    showlegend=(idx == 0)
                ),
                row=row, col=col
            )
    
    # Update layout
    fig.update_layout(
        height=1000,
        width=1200,
        title_text="Model Performance Comparison",
        barmode='group'
    )
    
    # Update axes labels
    for i in range(4):
        fig.update_yaxes(title_text=metrics[i], row=i//2+1, col=i%2+1)
        fig.update_xaxes(title_text="MBTI Dimension", row=i//2+1, col=i%2+1)
    
    fig.show()

# Plot model comparison
plot_model_comparison(detailed_results)

# Save the best model based on average accuracy
model_avg_accuracy = detailed_results.groupby('Model')['Accuracy'].mean()
best_model_name = model_avg_accuracy.idxmax()
embed_name, seq_name = best_model_name.split('+')
best_model = models[embed_name][seq_name]

print(f"\nSaving best model ({best_model_name}) with average accuracy: {model_avg_accuracy[best_model_name]:.4f}")
best_model.save('best_mbti_model.keras')

# Create summary visualization of training histories
def plot_training_summary(histories):
    """Create summary visualization of training histories"""
    fig = make_subplots(
        rows=1, cols=2,
        subplot_titles=('Training Loss', 'Training Accuracy')
    )
    
    colors = px.colors.qualitative.Set3
    
    for i, (embed_name, embed_hist) in enumerate(histories.items()):
        for seq_name, history in embed_hist.items():
            model_name = f"{embed_name}+{seq_name}"
            
            # Plot final loss and accuracy values
            final_loss = history['loss'][-1]
            final_val_loss = history['val_loss'][-1]
            final_acc = history['binary_accuracy'][-1]
            final_val_acc = history['val_binary_accuracy'][-1]
            
            # Add loss points
            fig.add_trace(
                go.Scatter(
                    x=['Train', 'Validation'],
                    y=[final_loss, final_val_loss],
                    name=model_name,
                    mode='lines+markers',
                    marker=dict(color=colors[i]),
                    showlegend=True
                ),
                row=1, col=1
            )
            
            # Add accuracy points
            fig.add_trace(
                go.Scatter(
                    x=['Train', 'Validation'],
                    y=[final_acc, final_val_acc],
                    name=model_name,
                    mode='lines+markers',
                    marker=dict(color=colors[i]),
                    showlegend=False
                ),
                row=1, col=2
            )
    
    # Update layout
    fig.update_layout(
        height=500,
        title_text="Final Training Metrics Comparison",
        showlegend=True
    )
    
    # Update axes labels
    fig.update_yaxes(title_text="Loss", row=1, col=1)
    fig.update_yaxes(title_text="Accuracy", row=1, col=2)
    
    fig.show()

# Plot training summary
plot_training_summary(histories)

# Print final summary
print("\nTraining Complete!")
print(f"Best performing model: {best_model_name}")
print(f"Model saved to: best_mbti_model")
print("\nRecommended next steps:")
print("1. Load the saved model using tf.keras.models.load_model('best_mbti_model')")
print("2. Use model.predict() for making new predictions")
print("3. Consider further fine-tuning on specific MBTI dimensions if needed")
```

```python
def preprocess_input(text, tokenizer, max_tokens=40):  # Changed from 512 to match model's expected input
    """Tokenize and preprocess the input text."""
    # Add proper BERT sentence structure with [CLS] and [SEP]
    token_ids = tokenizer.encode(
        text,
        padding='max_length',
        truncation=True,
        max_length=max_tokens,
        return_tensors='np'
    )
    return token_ids


def predict_personality(text, model, tokenizer):
    """Predict MBTI personality type from input text."""
    # Process input
    processed_input = preprocess_input(text, tokenizer)
    
    # Make prediction
    prediction = model.predict(processed_input, verbose=0)
    
    # Define personality dimensions and threshold
    personality_dimensions = ['I/E', 'N/S', 'F/T', 'P/J']
    threshold = 0.5
    
    # Convert predictions to MBTI type
    predicted_personality = [
        f"{dim[0] if p < threshold else dim[2]}" 
        for dim, p in zip(personality_dimensions, prediction[0])
    ]
    
    return "".join(predicted_personality)
```
