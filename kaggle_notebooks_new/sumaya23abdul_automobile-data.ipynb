### Data Normalization

##### why normalization
Normalization is the process of transforming values of several variables into a similar range.
Typical normalizations include scaling the variable so the variable average is 0, 
scaling the variable so the variance is 1, 
or scaling the variable so the variable values range from 0 to 1.

##### Example
To demonstrate normalization, let's we want to scale the colmns "length", "width", " height"
##### Target: would like to normalize those variables so their values ranges from 0 to 1.
##### Approach : replace original value by (original value)/(maxium value)



```python
df[["length", "width", "height"]].head()
```

```python
# replace (original value) by (original value)/(maxium value)
df["length"]= df["length"]/df["length"].max()
df["width"]=df["width"]/df["width"].max()
df["height"]= df["height"]/df["height"].max()

# show the scaled columns
df[["length","width", "height"]].head()
```

#### Here we can see we've normlaized "length, "width" and"height" in the avergae of [0,1]

We would like 3 bins of equal size bandwidth so we use numpy's linspace(start_value, end_value, numbers_generated function
Since we want to include the minimum value of horsepower , we want to set start_value= min(df["horsepower"]).
Since we want to include the maximum value of horsepower, we want to set end_value = max(df["horsepower"]).

Since we are building 3 bins of equal length, there should be 4 dividers, so numbers_generated=4


We build a bin array with a minium value to a maxium value by using bandwidth calculated above. The values will determine when one bin ends and another begins

```python
bins=np.linspace(min(df["horsepower"]), max(df["horsepower"]),4)
bins
```

Next we create a list of "group_names" that contains the different bin names

### Indicator Variable (or Dummy Variable)
#### what is an indicator variable?
An idicator variable (or dummy variable) is a numerical variable used to label categories. They are called "dummies" because the numbers themeselves don't have meaning

#### Why we use indicator variable?
We use indicator variable so we can use categorical variables for regression analysis

#### Example
We see the column "fuel-type" has two unique values: "gas or "diesel". Regression doesnot understand words, only numbers.
To use this attribute in regression analysis, we convert "fuel-type" to indicator variables.
We use pandas method "get_dummies" to assign numerical values to different categories of fuel type.



```python
df["fuel-type"]
```

```python
dummy_variable_1=pd.get_dummies(df["fuel-type"])
dummy_variable_1.head()
```

```python
# Rename the column names for clarity
dummy_variable_1.rename(columns={"gas":"fuel-type-gas", "diesel": "fuel-type-diesel"}, inplace=True)
dummy_variable_1.head()
```

In the dataframe, column "fuel-type has values for "gas" and "diesel" as 0 and 1 now

```python
# merge data frame "df" and "dummy_varaible_1"
df=pd.concat([df, dummy_variable_1], axis=1)

# drop original column "fuel-type" from "df"
df.drop("fuel-type", axis=1, inplace=True)

df.head()
```
