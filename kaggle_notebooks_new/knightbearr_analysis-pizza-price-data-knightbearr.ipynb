# **Data Cleaning**

Data Cleaning is the process of fixing or removing incorrect, corupted, incorrectly formatted, duplicate, or incomplete data within a dataset. When combining multiple data sources, there are many opportunities for data to be duplicated or mislabeled.

But why we clean the data? Data cleansing ensures you only have the most recent files and important documents, so when you need to, you can find them with ease. It also helps ensure that you do not have significant amounts of personal information on your computer, which can be a security risk.

```python
# let's see the duplicates in the columns
for col in categoric:
    if pizza_data[col].duplicated().any() == True:
        print(f"column {col} contains duplicates.")
    else:
        print(f"column {col} does not contain duplicates.")
```

> Check if the columns contains words in different case and extra spaces.

```python
# check if spelling errors have been hit
for col in categoric:
    print(f"before {col}: {len(set(pizza_data[col]))} after {col}: {len(set(pizza_data[col].str.title().str.strip()))}")
```

> Great! look like does not have any spelling error! let's move on to the next steps!

### **Removing Diameter**

In my case, since I don't really need the diameter of the pizza, let's remove that from our pizza data, and just get the size of the pizza, if you want to use that diameter, you can do that! but now, I'm not gonna use that.

```python
# create the function
def get_size(size):
        return size.split(" ")[0]
# apply lambda    
pizza_data["Size"] = pizza_data["Size"].apply(lambda x: get_size(x))
# let's see
pizza_data.head(2)
```

> Great! looks better now, and now let's remove the dollar sign and change that feature to floating point.

```python
# let's replace that dollar sign
pizza_data["Price"] = pizza_data["Price"].str.replace("$", "").astype("float64")
# let's have a look
pizza_data.head()
# and let's get the information
pizza_data.info(verbose=True)
```

## **Univariate Analysis**

Univariate analysis is perhaps the simplest form of statistical analysis. Like other forms of statistics, it can be inferential or descriptive. The key fact is that only one variable is involved. Univariate analysis can yield misleading results in cases in which multivariate analysis is more appropriate, in here, I'll function from my notebook called [Analysis: Sales Data](https://kaggle.com/knightbearr/analysis-sales-data-knightbearr/), you should check that notebook if you want to gain more insight!

```python
# set the subplots
fig, (ax1, ax2) = plt.subplots(
    nrows=1, ncols=2, 
    figsize=(24, 9.5)
)

# Distplot axes 1
sns.distplot(
    pizza_data["Price"], 
    ax=ax1
)
ax1.set_title(
    "Price Distribution", 
    weight="bold", fontsize=25, 
    pad=30
)
# QQplot axes 2
qqplot(
    pizza_data["Price"], 
    ax=ax2, 
    line="s"
)
ax2.set_title(
    "Quantile Quantile Plot", 
    weight="bold", fontsize=25, 
    pad=30
);
```

<p style="font-name: monospace; line-height: 1; font-size: 16.5px; font-weight: bold; letter-spacing: 1px; text-align: left;">Q5: For example, the average success probability (success makes customers happy) in the service of each existing company is $87\%$</p>
<br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: left;">1. What is the probability for any company to succeed in making customers happy?</p>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: left;">2. What is the probability for each company to fail to make customers happy?</p>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: left;">3. What is the probability of at least one of the four companies succeeding in making customers happy?</p>
<br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: center;">$P(S) = 0.87$</p>
<br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: left;">1. What is the probability for any company to succeed in making customers happy?</p>
<br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: center;">Answer:</p>
$$
\begin{aligned}
P(4S) &= P(S) \times P(S) \times P(S) \times P(S) \\
      &= 0.87 \times 0.87 \times 0.87 \times 0.87 \\
      &= 0.5729 \ (even \ close)
\end{aligned}
$$
<br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: center;">Interpretation:</p>
The probability for each pizza company to successfully make customers happy is $57\%$ (if there are $100$ customers, there are only $57$ customers who are satisfied with the service of each pizza company). 
<br><br><br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: left;">2. What is the probability for each company to fail to make customers happy?</p>
<br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: center;">Answer:</p>
$$
\begin{aligned}
P(S') &= 1 - P(S) = 1 - 0.87 =  0.13\\
P(4S')&= P(S') \times P(S') \times P(S') \times P(S') \\
      &= 0.13 \times 0.13 \times 0.13 \times 0.13 \\
      &= 0.000286 \ (unusual \ events)
\end{aligned}
$$
<br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: center;">Interpretation:</p>
The probability of every company failing to make customers happy is $0.0002\%$ and this includes unusual events, or arguably very rarely every company fails to make customers happy in its service.
<br><br><br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: left;">3. What is the probability of at least one of the four companies succeeding in making customers happy?</p>
<br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: center;">Answer:</p>
$$
\begin{aligned}
P(at \ least) 1 S &= 1 - P(4S') \\
                  &= 1 - 0.000285 \\
                  &= 0.999714 \ (likely \ events)
\end{aligned}
$$
<br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: center;">Interpretation:</p>
$99\%$ Every pizza company can successfully make customers satisfied, and this is included in the event that appears very often, you could say customers will definitely be satisfied with the service of each company.
<br><br><br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: center;">Code:</p>

```python
# 1. What is the probability for any company to succeed in making customers happy?
p_s = 0.87
p_s_4 = p_s * p_s * p_s * p_s
print(f"probability: {p_s_4:.4f}")
```

```python
# 2. What is the probability for each company to fail to make customers happy?
p_s_prime = 1 - p_s
p_s_4_prime = p_s_prime * p_s_prime * p_s_prime * p_s_prime
print(f"probability: {p_s_4_prime:.6f}")
```

```python
# 3. What is the probability of at least one of the four companies succeeding in making customers happy?
p_at_least_1 = 1 - p_s_4_prime
print(f"probability: {p_at_least_1:.6f}")
```

<p style="font-name: monospace; line-height: 1; font-size: 16.5px; font-weight: bold; letter-spacing: 1px; text-align: left;">Q6: In a city, the city has $4$ pizza selling companies, namely Domino's Pizza, Pizza Hut, Godfather's Pizza, and IMO's Pizza, Employees want to buy pizza to accompany their lunch break, unfortunately these employees can only choose to buy from $2$ companies only, How many choices can the five employees choose?</p>
<br>
<p style="font-name: monospace; line-height: 1; font-size: 16.5px; font-weight: bold; letter-spacing: 1px; text-align: center;">$n = 4, r = 2$</p>
<br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: center;">Combination Formula:</p>
$$
\begin{aligned}
_nC_r = \frac{n!}{(n - r)! \times r!} \ \  condition= r \le n
\end{aligned}
$$
<br><br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: center;">Answer:</p>
$$
\begin{aligned}
_nC_r &= \frac{n!}{(n - r)! \times r!} \\
_4C_2 &= \frac{4!}{(4 - 2)! \times 2!} \\
      &= \frac{4!}{2! \times 2!} \\
      &= \frac{24}{4} \\
      &= 6
\end{aligned}
$$
<br><br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: center;">Interpretation:</p>
There are $6$ combination of pizza companies that can be formed.
<br><br><br>
<p style="font-name: monospace; line-height: 1; font-size: 16px; font-weight: bold; letter-spacing: 1px; text-align: center;">Code:</p>

```python
companies = ["Domino's Pizza", "Pizza Hut", "Godfather's Pizza", "IMO's Pizza"]
combination, i = itertools.combinations(companies, 2), 1
for comb in combination:
    print(f"combination {i}: {comb}")
    i += 1
```
