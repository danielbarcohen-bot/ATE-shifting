## Data Preparation
### Encoding Fitur Kategori
Encoding Fitur Kategori adalah proses mengubah variabel kategorikal menjadi variabel numerik agar dapat diproses oleh algoritma machine learning. Pada dataset ini digunakan teknik encoding fitur kategori yaitu One-Hot Encoding

```python
from sklearn.preprocessing import OneHotEncoder

def one_hot_encode(data, columns):
    encoder = OneHotEncoder(handle_unknown='ignore')
    encoded = encoder.fit_transform(data[columns])
    labels = encoder.get_feature_names_out(columns)
    encoded_df = pd.DataFrame(encoded.toarray(), columns=labels)
    data = pd.concat([data.drop(columns, axis=1), encoded_df], axis=1)
    return data
```

One-Hot Encoding dengan Mengubah nilai kategori menjadi variabel dummy. Contohnya, untuk variabel "region" dengan kategori "northeast", "southeast", "southwest", dan "northwest", maka one-hot encoding akan membuat empat kolom baru yang masing-masing merepresentasikan satu kategori dan bernilai 1 jika sampel memiliki kategori tersebut, dan 0 jika tidak.

```python
data = one_hot_encode(data, categorical_features)

data.head()
```

```python
null_counts = data.isnull().sum()
print(null_counts)
```

```python
data = data.dropna()
data.shape
```

## Train-Test-Split

Train-Test-Split adalah melakukan pemisahan dataset menjadi data train dan data test. Pemisahan ini dilakukan untuk menguji performa model pada data yang belum pernah dilihat sebelumnya. Dalam hal ini, dataset dibagi menjadi data train dan data test dengan perbandingan 80:20.

```python
from sklearn.model_selection import train_test_split
 
X = data.drop(["charges"],axis =1)
y = data["charges"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 123)
```

## Modeling
Setelah melakukan data preparation dan exploratory data analysis, tahap selanjutnya adalah pemodelan. Dalam kasus ini, kita ingin memprediksi biaya asuransi kesehatan berdasarkan faktor-faktor yang ada pada dataset. Ada beberapa model machine learning yang dapat digunakan untuk menyelesaikan permasalahan ini, di antaranya adalah:
1. K-Nearest Neighbor (KNN)
adalah algoritma machine learning yang digunakan untuk klasifikasi dan regresi. Algoritma ini akan mencari k-nearest neighbor (tetangga terdekat) dari suatu data point yang ingin diprediksi, kemudian mengambil rata-rata atau mode dari label (untuk klasifikasi) atau nilai target (untuk regresi) dari k tetangga terdekat tersebut sebagai prediksi. Parameter yang perlu diatur pada KNN antara lain nilai k, metrik jarak, dan teknik weighting.
2. Random Forest
adalah algoritma machine learning yang termasuk dalam kategori ensemble learning. Ensemble learning adalah teknik yang menggabungkan beberapa model machine learning menjadi satu model yang lebih kuat. Pada algoritma Random Forest, terdapat beberapa pohon keputusan (decision tree) yang dibuat dengan sampel data yang diambil secara acak (random). Setiap pohon akan memberikan prediksi, kemudian hasil akhir prediksi akan diambil berdasarkan mayoritas prediksi dari semua pohon. Parameter yang perlu diatur pada Random Forest antara lain jumlah pohon, kedalaman pohon, dan jumlah fitur yang diambil secara acak.
3. Boosting Algorithm
adalah algoritma machine learning yang juga termasuk dalam kategori ensemble learning. Pada algoritma ini, terdapat beberapa model machine learning yang dibuat secara berurutan. Setiap model akan belajar dari kesalahan model sebelumnya, sehingga secara bertahap performa model akan semakin meningkat. Ada beberapa jenis algoritma Boosting salah satunya adalah AdaBoost. Parameter yang perlu diatur pada Boosting Algorithm antara lain jumlah model (n_estimators), learning rate, dan kedalaman pohon (pada Gradient Boosting dan XGBoost). 
Setelah memilih beberapa model machine learning yang sesuai untuk menyelesaikan permasalahan ini, kita perlu melakukan tuning pada parameter-parameter yang ada pada masing-masing algoritma. Hal ini dilakukan untuk mencari kombinasi parameter terbaik yang dapat memberikan performa model yang optimal. Setelah melakukan tuning, kita dapat memilih model terbaik berdasarkan metrik evaluasi yang dipilih yaitu Root Mean Squared Error.

```python
# Siapkan dataframe untuk analisis model
models = pd.DataFrame(index=['train_mse', 'test_mse'], 
                      columns=['KNN', 'RandomForest', 'Boosting'])
```

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error
 
knn = KNeighborsRegressor(n_neighbors=5)
knn.fit(X_train, y_train)
 
models.loc['train_mse','knn'] = mean_squared_error(y_pred = knn.predict(X_train), y_true=y_train)
```

```python
# Impor library yang dibutuhkan
from sklearn.ensemble import RandomForestRegressor
 
# buat model prediksi
RF = RandomForestRegressor(n_estimators=100, max_depth=35, random_state=123, n_jobs=-1)
RF.fit(X_train, y_train)
 
models.loc['train_mse','RandomForest'] = mean_squared_error(y_pred=RF.predict(X_train), y_true=y_train)   
```

```python
from sklearn.ensemble import AdaBoostRegressor
 
boosting = AdaBoostRegressor(learning_rate=0.05, random_state=55)                             
boosting.fit(X_train, y_train)
models.loc['train_mse','Boosting'] = mean_squared_error(y_pred=boosting.predict(X_train), y_true=y_train)
```

```python
# Lakukan scaling terhadap fitur numerik pada X_test sehingga memiliki rata-rata=0 dan varians=1
X_test.loc[:, numerical_features] = scaler.transform(X_test[numerical_features])
```

## Evaluation
Untuk kasus regresi, salah satu metrik evaluasi yang umum digunakan adalah Root Mean Squared Error (RMSE). Metrik ini mengukur seberapa jauh hasil prediksi dari nilai sebenarnya dalam satuan yang sama dengan variabel target.

RMSE dihitung dengan menghitung selisih antara nilai prediksi dan nilai sebenarnya, kemudian mengkuadratkan hasilnya, menjumlahkan seluruh nilai tersebut, membaginya dengan jumlah data, dan akar dari hasil tersebut. Formula untuk menghitung RMSE adalah sebagai berikut:

RMSE = sqrt((1/n) * sum((y_pred - y_true)^2))

Dalam hal ini, y_pred adalah nilai yang diprediksi oleh model, y_true adalah nilai sebenarnya, dan n adalah jumlah data.

Berdasarkan metrik RMSE, semakin kecil nilai RMSE, semakin baik performa model dalam memprediksi nilai target.

Dalam kasus dataset Medical Cost Personal, model terbaik dapat dipilih berdasarkan nilai RMSE terkecil. Semakin kecil nilai RMSE, semakin akurat model dalam memprediksi biaya asuransi kesehatan.

```python
# Buat variabel mse yang isinya adalah dataframe nilai mse data train dan test pada masing-masing algoritma
mse = pd.DataFrame(columns=['train', 'test'], index=['KNN','RF','Boosting'])
 
# Buat dictionary untuk setiap algoritma yang digunakan
model_dict = {'KNN': knn, 'RF': RF, 'Boosting': boosting}
 
# Hitung Mean Squared Error masing-masing algoritma pada data train dan test
for name, model in model_dict.items():
    mse.loc[name, 'train'] = mean_squared_error(y_true=y_train, y_pred=model.predict(X_train))/1e3 
    mse.loc[name, 'test'] = mean_squared_error(y_true=y_test, y_pred=model.predict(X_test))/1e3
 
# Panggil mse
mse
```

```python
fig, ax = plt.subplots()
mse.sort_values(by='test', ascending=False).plot(kind='barh', ax=ax, zorder=3)
ax.grid(zorder=0)
```

Dari hasil tersebut, dapat dilihat bahwa Random Forest memiliki performa terbaik pada data train dengan MSE yang lebih rendah dibandingkan KNN dan Boosting Algorithm, namun pada data test, performa terbaik dimiliki oleh Boosting Algorithm dengan MSE yang lebih rendah dibandingkan KNN dan Random Forest.

```python
prediksi = X_test.iloc[:1].copy()
pred_dict = {'y_true':y_test[:1]}
for name, model in model_dict.items():
    pred_dict['prediksi_'+name] = model.predict(prediksi).round(3)
 
pd.DataFrame(pred_dict)
```
