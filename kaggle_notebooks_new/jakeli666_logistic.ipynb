```python
import pandas
Data = pandas.read_csv("/kaggle/input/titanic/train.csv")
```

```python
Data.head()
```

```python
train_data = Data[['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Survived']]
```

```python
train_data.head()
train_data.dropna(inplace = True)
```

```python
train_data['Sex']
```

```python
train_data['Sex'].replace('male', 1, inplace = True)
train_data['Sex'].replace('female', 0, inplace = True)
```

```python
train_data['Sex']
```

```python
train_x = train_data[['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']]
train_y = train_data[['Survived']]
```

```python
train_y.head()
```

```python
from sklearn import preprocessing
import numpy as np
train_X = np.array(train_x)
train_Y = np.array(train_y)
scaler = preprocessing.StandardScaler().fit(train_X)
train_scaled = scaler.transform(train_X)
```

```python
train_X
```

```python
from sklearn.linear_model import LogisticRegression
clf = LogisticRegression(random_state=0).fit(train_X, train_Y)
```

```python
test = pandas.read_csv('/kaggle/input/titanic/test.csv')
```

```python
test
```

```python
Test = test.fillna(method="ffill")
Test
```

```python
Test_data = Test[['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']]
```

```python
Test_data['Sex'].replace('male', 1, inplace = True)
Test_data['Sex'].replace('female', 0, inplace = True)
```

```python
testy = clf.predict(Test_data)
```

```python
testy
```

```python
output = pandas.DataFrame({'PassengerId': Test.PassengerId, 'Survived': testy})
```
