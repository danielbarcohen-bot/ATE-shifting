# Uso de **.kde** y **.displot** para gr?ficar distribuciones.
Quiz?s el enfoque m?s com?n para visualizar una distribuci?n es el histograma . Este es el enfoque predeterminado en [displot()](https://seaborn.pydata.org/generated/seaborn.displot.html#seaborn.displot), que usa el mismo c?digo subyacente que [histplot()](https://seaborn.pydata.org/generated/seaborn.histplot.html#seaborn.histplot) . 
> ***Un histograma es un gr?fico de barras en el que el eje que representa la variable de datos se divide en un conjunto de contenedores discretos y el recuento de observaciones que se encuentran dentro de cada contenedor se muestra usando la altura de la barra correspondiente:***

````python
sns.histplot(data = tips, x= 'tip', bins = 15, cumulative=True)
# El atributo "bins=n" Cuantas barras debe usar para la frecuencia
# Aqu? estamos agregando el par?metro "cumulative=bool" para poder acumular nuestros valores
plt.show()
````

````python
sns.histplot(data = tips, x= 'tip', bins = 15, cumulative=False,hue='sex')
# Al cambiar el valor de cumulative a False, nos quita la gr?fica acumulativa
# Agregamos la variabl hue para ver la distribuci?n del "sex"
plt.show()
````

````python
sns.histplot(data = tips, x= 'tip', bins = 15, cumulative=True, hue='sex',stat='probability')
# Agregando "stat='str" nos manda las estadisticas.
plt.show()
````

````python
sns.histplot(data = tips, x= 'tip', bins = 15, cumulative=True, hue='sex',stat='count',multiple='stack')
# Con el valor de multiple= es como queremos presentar nuestra gr?fica
plt.show()
````

## Para el p?ramametro "multiple='str'" que nos sirve para poder elegir como representar nuestra gr?fica.
Algunos valores:
* Stack = termina uno y empieza otro (uno encima del otro)
* Layer = Valor predeterminado
* Dodge = Uno a lado de otro
* fill = Dando toda la probabilidad.

## Elegir tama?o del contenedor:
**El tama?o de los contenedores es un par?metro _importante_**, y el uso de un tama?o de contenedor incorrecto puede **inducir a error** al oscurecer caracter?sticas importantes de los datos o al crear caracter?sticas aparentes a partir de la variabilidad _aleatoria_. De forma predeterminada **displot()/ histplot()** elije un tama?o de intervalo predeterminado en funci?n de la varianza de los datos. **(Nos ayuda ajustar la informaci?n)** y el n?mero de observaciones. Pero no debe **confiar demasiado en estos enfoques autom?ticos**, ya que dependen de suposiciones particulares sobre la estructura de sus datos. Siempre es recomendable verificar que sus impresiones de la distribuci?n sean consistentes en diferentes tama?os de contenedores. Para elegir el tama?o directamente, configure el par?metro 
```python
binwidth=n
```

````python
sns.histplot(data = tips, x= 'tip', binwidth=2)
plt.show()
````

En ese caso, el ancho del contenedor predeterminado puede ser demasiado peque?o, creando brechas inc?modas en la distribuci?n:
