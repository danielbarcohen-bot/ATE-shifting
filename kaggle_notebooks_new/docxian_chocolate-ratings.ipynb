#### Rating Scale:
* 4.0 - 5.0 = Outstanding
* 3.5 - 3.9 = Highly Recommended
* 3.0 - 3.4 = Recommended
* 2.0 - 2.9 = Disappointing
* 1.0 - 1.9 = Unpleasant

```python
# packages

# standard
import numpy as np
import pandas as pd

# plot
import matplotlib.pyplot as plt
import seaborn as sns

# wordcloud
from wordcloud import WordCloud, STOPWORDS, ImageColorGenerator
```

<a id='1'></a>
# Import and Preprocessing

```python
# load data and first glance
df = pd.read_csv('../input/chocolate-ratings/chocolate_ratings.csv')
df.head()
```

```python
# dimensions of data (rows,columns)
df.shape
```

```python
# data frame overview
df.info()
```

```python
# prep percentage field
def clean_perc(x):
    return pd.to_numeric(x.replace('%',''))

df['Cocoa Percent'] = df['Cocoa Percent'].apply(clean_perc)
```

```python
# add categorical version of rating
df['Rating_Cat'] = 'Unpleasant'
df.loc[df.Rating >= 2,   'Rating_Cat'] = 'Disappointing'
df.loc[df.Rating >= 3,   'Rating_Cat'] = 'Recommended'
df.loc[df.Rating >= 3.5, 'Rating_Cat'] = 'Highly Recommended'
df.loc[df.Rating >= 4,   'Rating_Cat'] = 'Outstanding'
```

# EDA

```python
df.Rating_Cat.value_counts().plot(kind='bar')
plt.title('Rating Categories - Frequency')
plt.grid()
plt.show()
```

```python
top20_company_loc = df['Company Location'].value_counts()[0:20]
plt.figure(figsize=(12,4))
top20_company_loc.plot(kind='bar')
plt.title('Company Location - Top 20')
plt.grid()
plt.show()
```

```python
top20_country_orig = df['Country of Bean Origin'].value_counts()[0:20]
plt.figure(figsize=(12,4))
top20_country_orig.plot(kind='bar')
plt.title('Country of Bean Origin - Top 20')
plt.grid()
plt.show()
```

```python
top20_ingredients = df['Ingredients'].value_counts()[0:20]
plt.figure(figsize=(12,4))
top20_ingredients.plot(kind='bar')
plt.title('Ingredients - Top 20')
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(12,4))
df['Review Date'].value_counts().sort_index().plot(kind='bar')
plt.title('Review Date')
plt.grid()
plt.show()
```

```python
df['Cocoa Percent'].plot(kind='hist')
plt.title('Cocoa Percent')
plt.grid()
plt.show()
```

```python
df['Rating'].plot(kind='hist')
plt.title('Rating Distribution')
plt.grid()
plt.show()
```

# Rating vs Features

```python
plt.scatter(df['Cocoa Percent'], df['Rating'], alpha=0.25)
plt.xlabel('Cocoa Percent')
plt.ylabel('Rating')
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(12,4))
sns.violinplot(data = df, x='Review Date', y='Rating')
plt.title('Rating by Year')
plt.grid()
plt.show()
```

```python
temp_list = top20_company_loc.index.tolist()
df_temp = df[df['Company Location'].isin(temp_list)]

plt.figure(figsize=(12,4))
sns.violinplot(data = df_temp, x='Company Location', y='Rating')
plt.title('Rating by Company Location - Top 20 only')
plt.xticks(rotation=90)
plt.grid()
plt.show()
```

```python
temp_list = top20_country_orig.index.tolist()
df_temp = df[df['Country of Bean Origin'].isin(temp_list)]

plt.figure(figsize=(12,4))
sns.violinplot(data = df_temp, x='Country of Bean Origin', y='Rating')
plt.title('Rating by Country of Bean Origin - Top 20 only')
plt.xticks(rotation=90)
plt.grid()
plt.show()
```

# Look at top rated chocolates

```python
# select top rated chocolates
df_top = df[df.Rating_Cat=='Outstanding']
```

```python
# bean origin of top rated chocolates
df_top['Country of Bean Origin'].value_counts()
```

```python
# ingredients of top rated chocolates
df_top['Ingredients'].value_counts()

df_top['Ingredients'].value_counts().plot(kind='bar')
plt.title('Ingredients - Top rated chocolates')
plt.grid()
plt.show()
```

#### Wordcloud of characteristics for top rated chocolates

```python
stopwords = set(STOPWORDS)
```

```python
# prepare text
text = ' '.join(txt for txt in df_top['Most Memorable Characteristics'])

# show wordcloud
wordcloud = WordCloud(stopwords=stopwords, max_font_size=50, max_words=500,
                      width = 500, height = 300,
                      background_color='white').generate(text)
plt.figure(figsize=(14,10))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```
