```python
import pandas
from keras.models import Sequential
from keras.layers import Dense
from keras.wrappers.scikit_learn import KerasClassifier
from keras.utils import np_utils
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import KFold
from sklearn.preprocessing import LabelEncoder
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
import numpy as np
# load dataset
dataframe = pandas.read_csv("https://raw.githubusercontent.com/aswinjose89/Datasets/master/iris.csv", header=None)
dataset = dataframe.values
```

# **X = List of features, Y = Output feature, Label encoding to encode multiple class and one hot encoding**

```python
X = dataset[:,0:4].astype(float)
Y = dataset[:,4]
# encode class values as integers
encoder = LabelEncoder()
encoder.fit(Y)
encoded_Y = encoder.transform(Y)
# convert integers to dummy variables (i.e. one hot encoded)
dummy_y = np_utils.to_categorical(encoded_Y)
```

```python
encoder.classes_
```

# **Data should be seperated into Old(X1, Y1) & New(X2, Y2) Data**

```python
#Considered Old Data
X1= np.concatenate([X[0:45], X[50:55],X[100:125]])
Y1=np.concatenate([dummy_y[0:45], dummy_y[50:55],dummy_y[100:125]])

#Considered New Data
X2= np.concatenate([X[46:49], X[56:100],X[126:]])
Y2=np.concatenate([dummy_y[46:49], dummy_y[56:100],dummy_y[126:]])
```

# **Apply pareto ratio and train Neural Network model using old data and predict the accuracy**

```python
trainx1, testx1, trainy1, testy1 = train_test_split(X1,Y1, test_size=.15, random_state=42)

model = Sequential()
model.add(Dense(8, input_dim=4, activation='relu'))
model.add(Dense(3, activation='softmax'))
# Compile model
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(trainx1,trainy1, epochs=200, batch_size=5, verbose=0)
y_pred1=model.predict(testx1)
y_pred1_indices= np.argmax(y_pred1, axis=1)
testy1_indices= np.argmax(testy1, axis=1)
from sklearn.metrics import accuracy_score, classification_report
print('Testing Accuracy:', accuracy_score(testy1_indices, y_pred1_indices)*100)
```

# **Apply pareto ratio and train Neural Network model using new data and predict the accuracy**

```python
trainx2, testx2, trainy2, testy2 = train_test_split(X2,Y2, test_size=.15, random_state=42)
# Compile model
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(trainx2,trainy2, epochs=200, batch_size=5, verbose=0)
y_pred2=model.predict(testx2)
y_pred2_indices= np.argmax(y_pred2, axis=1)
testy2_indices= np.argmax(testy2, axis=1)
from sklearn.metrics import accuracy_score, classification_report
print('Testing Accuracy:', accuracy_score(testy2_indices, y_pred2_indices)*100)
```

# **Predicting Updated Model using Old Data to verify incremental learning is working fine**

```python
y_pred1=model.predict(testx1)
y_pred1_indices= np.argmax(y_pred1, axis=1)
testy1_indices= np.argmax(testy1, axis=1)
from sklearn.metrics import accuracy_score, classification_report
print('Testing Accuracy:', accuracy_score(testy1_indices, y_pred1_indices)*100)
```
