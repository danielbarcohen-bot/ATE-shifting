```python
data = pandas.read_csv("/kaggle/input/body-signal-of-smoking/smoking.csv")
data["gender"] = data["gender"].map({"M":1, "F":0})
data["oral"] = data["oral"].map({"Y":1, "N":0})
data["tartar"] = data["tartar"].map({"Y":1, "N":0})
data = pandas.concat((data[data["smoking"] == 0].sample(sum(data["smoking"]), random_state = 2200), 
                      data[data["smoking"] == 1]), axis = 0)
```

```python
x = data[data.columns[1:-1]]
ss = StandardScaler()
x = ss.fit_transform(x)
y = data[data.columns[-1]]
train_x, test_x, train_y, test_y = train_test_split(x, y, test_size = 0.3, random_state =2200)
```

```python
cls = XGBClassifier(booster = "gbtree", colsample_bylevel = 0.7, colsample_bytree = 1, 
                    eta = 0.2, gamma = 0.01, tree_method = "exact", max_depth = 18, 
                   max_leaves = 327, n_estimators = 800, reg_alpha = 4.145, reg_lambda = 4.146)
clf = cls.fit(train_x, train_y)
predict_y = clf.predict(test_x)
print("auc: {}".format(metrics.roc_auc_score(predict_y, test_y)))
print("accuracy: {}".format(sum(predict_y == test_y)/len(test_y)))
```
