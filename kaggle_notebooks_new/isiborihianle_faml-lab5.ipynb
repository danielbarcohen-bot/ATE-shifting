```python
# Import the pandas library as 'pd'
import pandas as pd

# Define the URL or file path for the Titanic dataset
url = "/kaggle/input/titanicdataset-traincsv/train.csv"

# Read the dataset into a Pandas DataFrame named 'df'
df = pd.read_csv(url)
```

```python
# Find the number of columns in the DataFrame
num_columns = len(df.columns)

# Get the column names
column_names = df.columns.tolist()

# Print the number of columns and column names
print("Number of columns:", num_columns)
print("Column names:", column_names)
```

**Understand one categorical attribute ** 

1. How many male and female passengers? 
Display the distribution of Attribute ["Sex"] using seaborn.countplot.


```python
# Import the seabone library as 'sns'
import seaborn as sns

# pass the dataframe to data and the column to x or y
sns.countplot(x = "Sex", data = df)  
```

2. How many passengers in each class? 
Display the distribution of Attribute [“PClass”] using Use seaborn.countplot.


```python
# pass the dataframe to data and the column to x or y
sns.countplot(x = "Pclass", data = df)  
```

**Understand the relationship between two categorical attributes (2 variables)**
1. How many female and male passengers survived? 
Display the distribution of Attribute [“Survived”] grouped by Attribute ["Sex] using seaborn.countplot.


```python
#pass the dataframe to data and the column to x or y and hue = group by
sns.countplot(x = "Survived", data =df, hue = "Sex") 
```

2. Which class is likely to survive? 
Display the distribution of Attribute [“PClass”] grouped by Attribute ["Survived”] using seaborn.countplot.


```python
#pass the dataframe to data and the column to x or y and hue = group by
sns.countplot(x = "Pclass", data =df, hue = "Survived") 
```

**Understand the relationship between one categorical attribute and one numerical attribute (2 variable)**
1. What is the age distribution of female and male passengers? 
Display the distribution of Attribute [“Age”] grouped by Attribute ["Sex] using seaborn. histplot


```python
#pass the dataframe to data and the column to x or y and hue = group by
sns.histplot(data = df, x= "Age", hue= "Sex", multiple = "stack")
```
