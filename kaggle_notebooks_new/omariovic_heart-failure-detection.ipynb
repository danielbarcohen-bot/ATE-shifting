```python
dataset['Sex'] = dataset['Sex'].map({'F': 0, 'M': 1})
dataset['ChestPainType'] = dataset['ChestPainType'].map({'ATA': 0, 'NAP': 1, 'TA': 2, 'ASY': 3})
dataset['RestingECG'] = dataset['RestingECG'].map({'Normal': 0, 'LVH': 1, 'ST': 2}) 
dataset['ExerciseAngina'] = dataset['ExerciseAngina'].map({'N': 0, 'Y': 1})
dataset['ST_Slope'] = dataset['ST_Slope'].map({'Up': 0, 'Down': 1, 'Flat': 2})
```

```python
corr_matrix = dataset.corr()
def absolute(x):
    return np.abs(x)
plt.figure(figsize = (10, 6))
sns.heatmap(corr_matrix, annot = True, cmap = 'Blues')
plt.show()
```

```python
data = dataset.drop(['HeartDisease'], axis = 1).copy()
target = dataset['HeartDisease'].copy()
```

```python
data = StandardScaler().fit_transform(data)
```

```python
X_train, X_test, y_train, y_test = train_test_split(data, target, test_size = 0.25, random_state = 289)
```

```python
model = BaggingClassifier(LogisticRegression(solver = 'liblinear'), n_estimators = 100)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
print(classification_report(y_pred, y_test))
```
