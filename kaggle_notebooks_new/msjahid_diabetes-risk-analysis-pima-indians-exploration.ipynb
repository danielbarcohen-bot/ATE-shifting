````python
def visualize_null_values(df):
    # Determine the null values:
    null_counts = df.isnull().sum().to_frame()
    null_counts = null_counts.rename(columns={0: 'Null'})
    # Determine the not null values:
    not_null_counts = df.notna().sum().to_frame()
    not_null_counts = not_null_counts.rename(columns={0: 'Not Null'})
    # Combine the dataframes:
    df_null_counts = pd.concat([null_counts, not_null_counts], ignore_index=False, axis=1).reset_index()
    df_null_counts = df_null_counts.rename(columns={'index': df.columns[0].replace("_", " ").title()})
    
    # Generate Plot
    fig = px.bar(df_null_counts, x=df.columns[0].replace("_", " ").title(), y=['Not Null', 'Null'])
    fig.update_xaxes(categoryorder='total descending')
    fig.update_layout(
        title={'text': "Missing Values Analysis",
               'xanchor': 'center',
               'yanchor': 'top',
               'x': 0.5},
        xaxis_title=df.columns[0].replace("_", " ").title(),
        yaxis_title="Count",
        plot_bgcolor='rgba(0,0,0,0)',  # Setting plot background color to transparent
        paper_bgcolor='#ffffff',  # Setting paper background color to white
        legend_title_font_color="black",  # Setting legend title text color to black
        legend_font_color="#0000ff",  # Setting legend item text color to blue
        legend_bgcolor='rgba(0,0,0,0)',  # Setting legend background color to transparent
        font=dict(color="#000000")  # Setting font color to black
    )
    fig.show()
````

````python
# Example usage:
visualize_null_values(df)
````

````python
def format_column_name(df):
    
    # Define a function to format a single column name
    def format_single_column_name(column_name):
        
        if 'BMI' in column_name:
            return 'bmi'
        
        # Check if the column already contains underscores
        if '_' in column_name:
            return column_name  # Return the original column name if it already contains underscores
        
        # Replace camel case with underscores
        column_name = re.sub(r'(?<!^)(?=[A-Z])', '_', column_name)
        
        return column_name  # Return the modified column name
    
    # Apply format_single_column_name function to each column name
    mapped_names = [format_single_column_name(col) for col in df.columns]
    
    # Create a dictionary mapping original column names to modified column names
    column_mapping = dict(zip(df.columns, mapped_names))
    
    # Rename columns using the mapping
    df = df.rename(columns=column_mapping)
    
    # Convert column names to lowercase
    df.columns = df.columns.str.lower()
    
    return df
````

Let's delve into the summary statistics for our __numerical__ attributes:

````python
def calculate_statistics(df, opacity=0.5, font_size='17px'):
    # Filter numeric columns
    numeric_df = df.select_dtypes(include=['number'])
    
    stat = pd.DataFrame()  # Initialize an empty dataframe

    # Measure of central tendency
    stat['Mean'] = numeric_df.mean()  # for mean of all columns
    stat['Median'] = numeric_df.median()  # for median of all columns
    stat['Mode'] = numeric_df.mode().iloc[0]  # for mode of all columns

    # Measure of dispersion
    stat['Range'] = (numeric_df.max() - numeric_df.min())
    stat['Stdev'] = numeric_df.std()
    stat['Variance'] = numeric_df.var()

    # Interquartile range
    stat['IQR'] = [stats.iqr(numeric_df[i]) for i in numeric_df.columns]

    # Quantiles
    stat['25%'] = numeric_df.quantile(q=0.25, axis=0)
    stat['50%'] = numeric_df.quantile(q=0.50, axis=0)
    stat['75%'] = numeric_df.quantile(q=0.75, axis=0)

    # Minimum and maximum
    stat['Min'] = numeric_df.min()
    stat['Max'] = numeric_df.max()

    # Skewness and Kurtosis
    stat['Skewness'] = numeric_df.skew()
    stat['Kurtosis'] = numeric_df.kurt()

    # Transpose the DataFrame
    stat = stat.T

    # Apply the float format
    stat = stat.applymap('{:,.2f}'.format)

    # Apply the background color with transparency
    styled_df = stat.style.apply(lambda x: [f'background: rgba(173, 216, 230, {opacity})' for val in x], axis=1)
    
    # Apply font size
    styled_df = styled_df.set_properties(**{'font-size': font_size})
    
    return styled_df
````

````python
def variable_summary(df, top=1, opacity=0.5, font_size='20px'):
    # Initialize an empty list to store individual summaries
    summaries = []

    # Iterate through each column in the DataFrame
    for col in df.columns:
        dtype = df[col].dtype  # Data type of the column
        count = df[col].count()  # Count of non-null values
        unique = df[col].nunique()  # Number of unique values
        missing = df[col].isna().sum()  # Number of missing values
        
        # Calculate the top n most frequent values
        value_counts = df[col].value_counts().head(top)
        frequent_value = ', '.join([f"{value} ({count})" for value, count in value_counts.items()])
        
        # Append the summary information to the list
        summaries.append([col, dtype, count, unique, missing, frequent_value])

    # Create a DataFrame from the list of summaries
    summary_df = pd.DataFrame(summaries, columns=['Variable', 'Dtype', 'Count', 'Unique', 'Missing', 'Frequent Value'])

    # Apply background color with opacity
    styled_summary_df = summary_df.style.apply(lambda x: [f'background: rgba(42, 46, 120, {opacity})' for val in x], axis=1)

    # Apply font size to the entire DataFrame
    styled_summary_df.set_table_styles([{'selector': '', 'props': [('font-size', font_size)]}])

    return styled_summary_df
````

````python
def plot_categorical_distribution(df, categorical_features, columns_per_row=2):
    # Set up the subplot layout
    num_features = len(categorical_features)
    num_rows = (num_features - 1) // columns_per_row + 1

    # Choose a color palette with enough colors for all the features
    palette = sns.color_palette("dark", num_features)

    # Set subplot background color
    subplot_color = 'silver'

    # Create the figure and axes
    fig, ax = plt.subplots(nrows=num_rows, ncols=columns_per_row, figsize=(8 * columns_per_row, 5 * num_rows), facecolor='silver')

    # Flatten the axes for easier indexing
    ax = ax.flatten()

    # Loop through each categorical feature and plot its distribution
    for i, col in enumerate(categorical_features):
        # Calculate frequency percentages
        value_counts = df[col].value_counts(normalize=True).mul(100).sort_values()

        # Plot the bar chart
        value_counts.plot(kind='barh', ax=ax[i], width=0.8, color=palette[i])

        # Add frequency percentages to the bars
        for index, value in enumerate(value_counts):
            ax[i].text(value, index, f'{value:.1f}%', fontsize=12, weight='bold', va='center')

        ax[i].set_xlim([0, 100])
        ax[i].set_xlabel('Frequency Percentage', fontsize=12)
        ax[i].set_title(f'{col}', fontsize=14, color=palette[i], fontname='Serif')
        ax[i].set_facecolor(subplot_color)

    # Hide any remaining subplots if there are fewer features than expected
    for j in range(num_features, num_rows * columns_per_row):
        ax[j].axis('off')

    # Adjust the layout and display the plot
    plt.suptitle('Exploring Numeric and Categorical Data', fontsize=18, color='#FF5733', fontname='Calibri', fontweight='bold')
    plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.85, wspace=0.4, hspace=0.4)
    plt.show()
````

````python
def plot_continuous_vs_target(df, continuous_features, target):
    # Set color palette
    palette = sns.color_palette("dark")

    # Create the subplots
    fig, ax = plt.subplots(len(continuous_features), 2, figsize=(15, 23), gridspec_kw={'width_ratios': [1, 2]}, facecolor='silver')

    # Loop through each continuous feature to create barplots and KDE plots
    for i, col in enumerate(continuous_features):
        # Barplot showing the mean value of the feature for each target category
        graph = sns.barplot(data=df, x=target, y=col, ax=ax[i,0], palette=palette)

        # KDE plot showing the distribution of the feature for each target category
        for val, color in zip(df[target].unique(), palette):
            sns.kdeplot(data=df[df[target]==val], x=col, fill=True, linewidth=2, ax=ax[i,1], label=str(val), color=color)

        ax[i,1].set_yticks([])
        ax[i,1].legend(title=target, loc='upper right', facecolor='lightgray')

        # Add mean values to the barplot
        for cont in graph.containers:
            graph.bar_label(cont, fmt='%.3g')

        # Set face color to silver
        ax[i,0].set_facecolor('silver')
        ax[i,1].set_facecolor('silver')

        # Set plot title
        title = col.replace('_', ' ').title()  # Convert column name to title format
        ax[i,0].set_title(title, color=palette[i], fontname='Serif', fontsize=14)
        ax[i,1].set_title(f'KDE - {title}', color=palette[i], fontname='Serif', fontsize=14)

        # Set axis labels
        x_label = target.replace('_', ' ').title()  # Convert target column name to title format for x-axis label
        y_label = col.replace('_', ' ').title()    # Convert continuous feature column name to title format for y-axis label
        ax[i,0].set_xlabel(x_label, color=palette[i], fontname='Serif', fontsize=12)
        ax[i,0].set_ylabel(y_label, color=palette[i], fontname='Serif', fontsize=12)
        ax[i,1].set_xlabel(y_label, color=palette[i], fontname='Serif', fontsize=12)
        ax[i,1].set_ylabel('Density', color=palette[i], fontname='Serif', fontsize=12)

    # Set the title for the entire figure
    plt.suptitle('Continuous Features vs Target Distribution', fontsize=16, color='#FF5733', fontname='Calibri', fontweight='bold')
    plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.95, wspace=0.4, hspace=0.7)
    plt.show()
````

````python
def plot_stripplots(data, data_columns, hue_column):
    """
    Generates strip plots for each feature against the 'outcome' column.

    Parameters:
        data (pd.DataFrame): The dataframe containing the data to plot.
        data_columns (list): A list of column names to be used for plotting.
        hue_column (str): The column name used for coloring the plots.
    """
    fig, axs = plt.subplots(figsize=(15, 15), facecolor='silver')
    
    # Get the number of unique values in the hue_column
    num_unique_values = len(data[hue_column].unique())
    
    # Generate a color palette with the same number of colors as unique values
    palette = sns.color_palette("dark", num_unique_values)
    
    # Create color palettes for x-axis and y-axis
    x_palette = sns.color_palette("magma", len(data_columns))
    y_palette = sns.color_palette("dark", len(data_columns))

    j = 1

    for i, column in enumerate(data_columns[:8]):
        # Determine the index for x_palette and y_palette using modulo
        x_index = i % len(x_palette)
        y_index = i % len(y_palette)
        
        plt.subplot(4, 2, j)
        sns.stripplot(x=hue_column, y=column, data=data, palette=palette)
        plt.xlabel(hue_column.replace('_', ' ').title(), color=x_palette[x_index])
        plt.ylabel(column.replace('_', ' ').title(), color=y_palette[x_index])
        plt.title(f'{column.replace("_", " ").title()} by {hue_column.replace("_", " ").title()}', color='#3c3c3c')
        plt.gca().set_facecolor('silver')  # Set facecolor for current subplot
        j += 1

    plt.tight_layout()
    plt.show()
````

````python
def plot_scatter_matrix(data, data_columns, hue_column):
    """
    Generates scatter plots for all combinations of the specified columns in the dataframe.
    
    Parameters:
        data (pd.DataFrame): The dataframe containing the data to plot.
        data_columns (list): A list of column names to be used for plotting.
        hue_column (str): The column name used for coloring the plots.
    """
    
    # Generate all possible combinations of x and y columns
    column_combinations = list(itertools.combinations(data_columns, 2))
    
    # Calculate the number of rows and columns for subplots
    num_plots = len(column_combinations)
    num_cols = 2
    num_rows = (num_plots + num_cols - 1) // num_cols
    
    # Create a figure and axis objects
    fig, axs = plt.subplots(num_rows, num_cols, figsize=(20, 5*num_rows), facecolor='silver')
    
    # Flatten the axis object if there is only one row
    if num_rows == 1:
        axs = [axs]
    elif num_rows > 1:
        axs = axs.ravel()
    
    # Create a color palette for each unique category in hue_column
    palette = sns.color_palette("dark", len(data[hue_column].unique()))
    
    # Create color palettes for x-axis and y-axis
    x_palette = sns.color_palette("dark", len(data_columns))
    y_palette = sns.color_palette("dark", len(data_columns))
    
    # Create scatter plots for each combination
    for i, (x_column, y_column) in enumerate(column_combinations):
        # Determine the index for x_palette and y_palette using modulo
        x_index = i % len(x_palette)
        y_index = i % len(y_palette)
        
        # Create the scatter plot with the specified hue column and color palette
        sns.scatterplot(data=data, x=x_column, y=y_column, hue=hue_column, ax=axs[i], palette=palette)
        axs[i].set_xlabel(x_column.replace('_', ' ').title(), color=x_palette[x_index])  # Replace underscores and set color
        axs[i].set_ylabel(y_column.replace('_', ' ').title(), color=y_palette[y_index])  # Replace underscores and set color
        axs[i].set_title(f'{x_column.replace("_", " ").title()} vs {y_column.replace("_", " ").title()} {hue_column}', color='#3c3c3c')  # Replace underscores and set color
        axs[i].legend(title=hue_column.replace("_", " ").title(), facecolor='lightgray')
        
        # Set background color of each subplot to silver
        axs[i].set_facecolor('silver')
    
    # Adjust layout
    plt.tight_layout()
    plt.show()
````

````python
def plot_boxplots(df, columns, columns_per_row=2):
    # Define the format function
    def format_func(value, tick_number):
        if int(value) == value:  # Check if the value is an integer
            return f'{int(value)}'
        else:
            return f'{round(value, 2)}'

    # Choose a color palette with enough colors for all the columns
    palette = sns.color_palette("dark", len(columns))

    # Set subplot background color
    subplot_color = 'silver'

    num_cols = len(columns)
    num_rows = (num_cols - 1) // columns_per_row + 1

    plt.figure(figsize=(8 * columns_per_row, 5 * num_rows), facecolor='silver')
    for i, col in enumerate(columns):
        ax = plt.subplot(num_rows, columns_per_row, i + 1)
        sns.boxplot(data=df, y=col, color=palette[i], ax=ax)  # Change x to y for vertical orientation
        ax.set_facecolor(subplot_color)  # Set subplot background color

        # Set y-axis formatter
        ax.yaxis.set_major_formatter(FuncFormatter(format_func))

        ax.set_ylabel('')  # Hide y-axis label
        title = col.replace('_', ' ').title()
        ax.set_title(title, fontsize=13, color=palette[i], fontname='Serif')  # Set title for each subplot

    plt.suptitle('Boxplots of Each Columns', fontsize=18, color='#FF5733', fontname='Calibri', fontweight='bold')  # Set the title at the top
    # Adjust the spacing between subplots
    plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.94, wspace=0.4, hspace=0.4)
    plt.show()
````

<h2 align="left"><font color=red>Outlier Removal Processes</font></h2>

<b>Introduction</b>
    <br><br>
Outliers are data points that significantly deviate from the rest of the dataset, potentially skewing statistical analyses and machine learning models. Therefore, it's crucial to detect and appropriately handle outliers in data preprocessing.

<b>Approaches to Outlier Removal<b>
1. Univariate Method
   This approach focuses on detecting outliers in individual variables without considering their relationships with other variables. Common techniques include:

   - Z-score method: Removing data points that fall beyond a certain number of standard deviations from the mean.
   - Interquartile Range (IQR) method: Identifying outliers based on the spread of the data around the median.
   - Winsorization: Capping or clipping extreme values at a predefined percentile of the data.

    
2. Multivariate Method
   This approach considers the relationships between multiple variables when identifying outliers. Techniques include:

   - Mahalanobis distance: Measuring the distance of each data point from the centroid of the data cloud, considering the covariance structure of the data.
   - Robust methods: Utilizing robust statistical measures less affected by outliers, such as robust regression or robust covariance estimation.

Choosing the Right Method
The choice of outlier removal method depends on various factors such as the distribution of the data, the presence of influential variables, and the objectives of the analysis. It's essential to carefully evaluate the impact of outlier removal on the dataset and the downstream analysis to ensure meaningful information is not lost.

````python
def plot_violin_grid(df, columns, columns_per_row=3, hue='Species'):
    # Set the seaborn style to remove white grid lines
    sns.set_style("dark")
    
    # Choose a color palette with enough colors for all the columns
    palette = sns.color_palette("tab10", len(columns))  # You can choose any suitable color palette

    # Calculate the number of rows required based on the number of columns and columns per row
    num_cols = len(columns)
    num_rows = (num_cols - 1) // columns_per_row + 1

    # Create figure and axes
    fig, axs = plt.subplots(num_rows, columns_per_row, figsize=(14, 4.5 * num_rows), facecolor='silver')

    # Flatten axes for easier iteration
    axs = axs.flatten()

    # Set subplot background color
    subplot_color = 'silver'

    # Loop through each numeric column in the dataset
    for i, col in enumerate(columns):
        if i < num_cols:  # Ensure not to exceed the number of columns
            row_index = i // columns_per_row  # Calculate the row index
            col_index = i % columns_per_row   # Calculate the column index

            sns.violinplot(x=hue, y=col, data=df, ax=axs[i], palette=[palette[i]])
            axs[i].set(facecolor=subplot_color)  # Set subplot background color

            # Set plot title
            title = col.replace('_', ' ').title()  # Convert column name to title format
            axs[i].set_title(f'{title} by {hue.replace("_", " ").title()}', color=palette[i], fontname='Serif', fontsize=16, fontweight='bold')

            # Set x-axis and y-axis labels
            axs[i].set_xlabel(hue.replace('_', ' ').title(), color=palette[i])
            axs[i].set_ylabel(col.replace('_', ' ').title(), color=palette[i])

    # Hide empty axes if there are fewer columns than the allocated number of subplots
    for j in range(num_cols, num_rows * columns_per_row):
        fig.delaxes(axs[j])
    
    # Adjust layout
    plt.tight_layout()

    # Show plot
    plt.show()
````

    <br><br>
<b>Correlation</b>
    <br><br>
A measure of the extent of interdependence between variables.
<br><br>
<b>Pearson Correlation:<b>
    <br><br>
The Pearson Correlation measures the linear dependence between two variables X and Y.
    <br><br>
    The resulting coefficient is a value between -1 and 1 inclusive, where:
    <br><br>

   * __`1`__: Perfect positive linear correlation.
   * __`0`__: No linear correlation, the two variables most likely do not affect each other.
   * __`-1`__: Perfect negative linear correlation.
    
<br>
<b>P-value<b>
    <br><br>
What is this P-value? The P-value is the probability value that the correlation between these two variables is statistically significant. Normally, we choose a significance level of 0.05, which means that we are 95% confident that the correlation between the variables is significant.
    <br><br>
    By convention, when the
    <br>

   * __`p-value is < 0.001`__: we say there is strong evidence that the correlation is significant.
   * __`p-value is < 0.05`__: there is moderate evidence that the correlation is significant.
   * __`p-value is < 0.1`__: there is weak evidence that the correlation is significant.
   * __`p-value is > 0.1`__: there is no evidence that the correlation is significant.

````python
def plot_corr_heatmap_single(dataframe, figsize=(20, 17), cmap="cividis"):
    """
    Plot a correlation heatmap for the given dataframe.

    Args:
    - dataframe (pd.DataFrame): The input dataframe.
    - figsize (tuple): Figure size.
    - cmap (str): Color map for the heatmap.

    Returns:
    - None
    """
    palette = sns.color_palette("dark")

    plt.figure(figsize=figsize)
    matrix = np.triu(dataframe.corr())
    heatmap = sns.heatmap(dataframe.corr(), annot=True, linewidth=.8, mask=matrix, cmap=cmap)

    # Set x-axis and y-axis labels to title case with dark color palette
    x_labels = [label.replace('_', ' ').title() for label in dataframe.columns]
    y_labels = [label.replace('_', ' ').title() for label in dataframe.columns]
    heatmap.set_xticklabels(x_labels, color=palette[0], fontsize=12)  # Set x-axis labels
    heatmap.set_yticklabels(y_labels, color=palette[0], fontsize=12)  # Set y-axis labels
    
    # Modify tick label color using the palette
    for i, label in enumerate(heatmap.get_xticklabels()):
        label.set_color(palette[i % len(palette)])  # Use modulo operator to loop through the color palette
    for i, label in enumerate(heatmap.get_yticklabels()):
        label.set_color(palette[i % len(palette)])  # Use modulo operator to loop through the color palette
    
    # Set the background color of the entire figure and heatmap to silver
    plt.gcf().set_facecolor('silver')
    heatmap.set_facecolor('silver')

    plt.show()
````

````python
def plot_correlation_heatmap(df, columns):
    # Define color palette
    palette = sns.color_palette("dark", len(columns))
    
    # Calculate the correlation matrix
    corrmat = df.corr()

    # Define the colormap
    cmap = sns.diverging_palette(240, 10, s=75, l=40, n=9, center="dark", as_cmap=True)

    # Create a figure and axes
    f, ax = plt.subplots(figsize=(20, 12), facecolor='silver')

    # Plot the heatmap with specified colors
    heatmap = sns.heatmap(corrmat, annot=True, cmap=cmap, fmt=".2f", vmin=-1, vmax=1, annot_kws={"size": 14}, square=True)

    # Set font size for x-axis and y-axis labels
    heatmap.set_xticklabels([col.replace('_', ' ').title() for col in corrmat.columns], rotation=90, fontsize=16)
    heatmap.set_yticklabels([col.replace('_', ' ').title() for col in corrmat.index], rotation=0, fontsize=16)
    
    # Modify tick label color using the palette
    for i, label in enumerate(ax.get_xticklabels()):
        label.set_color(palette[i % len(palette)])  # Use modulo operator to loop through the color palette
    for i, label in enumerate(ax.get_yticklabels()):
        label.set_color(palette[i % len(palette)])  # Use modulo operator to loop through the color palette

    # Add title
    plt.title('Correlation Heat Map of Different Flowers Features', pad=45, fontsize=25, color='black')

    # Show plot
    plt.show()
````

````python
def z_test_comparison(df, columns=None, unique_col=None, name=None):
    try:
        if columns is None:
            raise ValueError("Columns argument is required.")
            
        if unique_col is None:
            raise ValueError("Unique column argument is required.")
            
        if unique_col not in df.columns:
            raise ValueError(f"Unique column '{unique_col}' not found in DataFrame.")
            
    except ValueError as ve:
        print(ve)
        return

    # Get the unique varieties
    specific = df[unique_col].unique()
    
    alpha=0.05

    # Lists to store results
    comparison_results = []
    columns_list = []
    category1_list = []
    category2_list = []

    # Perform Z-test for each pair of varieties and each column
    for column in columns:
        for category1, category2 in combinations(specific, 2):
            data1 = df[df[unique_col] == category1][column]
            data2 = df[df[unique_col] == category2][column]

            z_score, p_value = wstat.ztest(data1, data2)

            if abs(p_value) <= alpha:
                result = color.GREEN + 'Yes' + color.RESET
            else:
                result = color.RED + 'No' + color.RESET

            comparison_results.append(result)
            columns_list.append(column)
            category1_list.append(category1)
            category2_list.append(category2)

    # Create DataFrame
    df_comparison = pd.DataFrame({
        'Group': columns_list,
        'Outcome->1': category1_list,
        'Outcome->0': category2_list,
        'Significant Difference': comparison_results
    })
    
    print(f"{color.BLUE}Z-test for variety of {name}".center(80) + color.END)

    # Loop through each column and print the corresponding table
    for column in columns:
        # Print table for the current column with colored column names
        print(tabulate(df_comparison[df_comparison['Group'] == column], headers='keys', tablefmt="double_grid"))
        # Print an empty line for separation
        print()
````

    <h2 style="color:#0A2342">Feature Importance</h2>
    <p>
        Feature importance is a crucial concept in machine learning that helps us understand the relative contribution of each feature towards predicting the target variable. By identifying important features, we can focus our analysis and model-building efforts on the most relevant aspects of the data.
    </p>
    <h3 style="color:#0A2342">Overview</h3>
    <p>
        Feature importance can be assessed using various techniques, each providing insights into the significance of features in predictive modeling. Some common methods include:
    </p>
    <ul>
        <li><strong>Pearson's Correlation Coefficient:</strong> This statistical measure quantifies the strength and direction of the linear relationship between two continuous variables. A correlation coefficient close to 1 or -1 indicates a strong positive or negative correlation, respectively, while a coefficient near 0 suggests no linear relationship.</li>
        <li><strong>Feature Importance in Decision Trees:</strong> Decision tree-based models, such as Random Forests or Gradient Boosting Machines, assign importance scores to features based on their contribution to reducing impurity or error in the model. Features with higher importance scores are considered more influential in predicting the target variable.</li>
        <li><strong>Permutation Importance:</strong> This technique evaluates the impact of each feature by randomly shuffling its values and measuring the resulting decrease in model performance. Features that lead to a significant drop in performance when permuted are deemed important for the model's predictions.</li>
    </ul>
    <h3 style="color:#0A2342">Visualization</h3>

        <li><strong>Feature Importance Plots:</strong> For tree-based models, feature importance plots visualize the importance of each feature in the model, often represented as a bar chart or heatmap.</li>
    </ul>
    <h3 style="color:#0A2342">Usage</h3>
    <p>

    </p>
    <ul>
        <li><strong>Feature Selection:</strong> Identifying and selecting the most important features can improve model performance, reduce overfitting, and enhance interpretability.</li>
        <li><strong>Insights Generation:</strong> Understanding feature importance provides insights into the underlying relationships between features and the target variable, guiding further data exploration and hypothesis generation.</li>

````python
# Function to plot feature importance
def plot_feature_importance(X, y, palette="dark", columns_to_aggregate=None):
    """
    Plot feature importances using a horizontal bar plot.

    Parameters:
        X (pd.DataFrame): Feature matrix.
        y (pd.Series): Target vector.
        palette (str or list, optional): Color palette name or list of colors. Defaults to "dark".
        columns_to_aggregate (list, optional): List of columns to aggregate feature importances. Defaults to None.
    """
    feat_selection = ExtraTreesClassifier()
    feat_selection.fit(X, y)
    feat_importances = pd.Series(feat_selection.feature_importances_, index=X.columns)

    if columns_to_aggregate:
        for column in columns_to_aggregate:
            column_columns = [col for col in X.columns if col.startswith(column + "_")]
            if column_columns:
                feat_importances[column] = feat_importances[column_columns].sum()
                feat_importances.drop(column_columns, inplace=True)

    custom_palette = sns.color_palette(palette, len(feat_importances))

    plt.figure(figsize=(10, 8), facecolor='silver')
    barplot = sns.barplot(x=feat_importances.nlargest(len(feat_importances)).values,
                          y=feat_importances.nlargest(len(feat_importances)).index,
                          palette=custom_palette)

    for i, label in enumerate(barplot.get_yticklabels()):
        label.set_color(custom_palette[i])

    plt.gca().set_facecolor('silver')
    plt.xlabel('Feature Importance', fontsize=14)
    plt.ylabel('Features', fontsize=14)
    plt.title('Feature Importance Analysis', fontsize=16)
    plt.show()

    return feat_importances
````

    <p>
        Overfitting and underfitting are common challenges in machine learning models, affecting their ability to generalize well to new, unseen data.
    </p>
    <h3 style="color:#0A2342">Overfitting</h3>
    <p>
        Overfitting occurs when a model learns the training data too well, capturing noise or random fluctuations instead of the underlying patterns. It leads to high performance on the training data but poor performance on unseen data.
    </p>
    <h3 style="color:#0A2342">Underfitting</h3>
    <p>
        Underfitting happens when a model is too simple to capture the underlying structure of the data. It results in poor performance on both the training and unseen data.
    </p>
    <h3 style="color:#0A2342">Causes and Solutions</h3>
    <ul>
        <li><strong>Overfitting:</strong> Causes include using a very complex model, too many features, or insufficient regularization. Solutions involve simplifying the model, reducing the number of features, or applying regularization techniques such as L1 or L2 regularization.</li>
        <li><strong>Underfitting:</strong> Causes include using a very simple model or insufficient training data. Solutions include using a more complex model, adding more features, or obtaining more training data.</li>
    </ul>
    <h3 style="color:#0A2342">Prevention and Detection</h3>
    <p>
        Techniques such as cross-validation, regularization, and monitoring performance metrics on both training and validation datasets can help prevent and detect overfitting and underfitting.
    </p>
    <h3 style="color:#0A2342">Example</h3>
    <p>
        Consider a dataset for predicting housing prices where a linear regression model is used. If the model captures every minor variation in the training data, including noise, it may overfit the data. Conversely, if the model is too simplistic and fails to capture the true relationship between features and prices, it may underfit the data.
    </p>
    <h3 style="color:#0A2342">Summary</h3>
    <p>
        Overfitting and underfitting are common challenges in machine learning that can affect model performance and generalization. Understanding their causes, solutions, and prevention techniques is crucial for building robust and reliable machine learning models.

    <center style="font-size: 22px; color: Blue;"><u>Hyperparameters Tuning Technique of Machine Learning</u></center><br><br>
    <p style="text-align: justify;">GridSearchCV (Grid Search Cross-Validation) is a technique used to tune hyperparameters of machine learning models. It systematically searches for the optimal hyperparameters within a predefined grid of values by evaluating the model's performance using cross-validation.</p><br>
    <p style="text-align: justify;"><b>Here's a brief overview of how GridSearchCV works:</b></p>
    <ol style="list-style-type: decimal; padding-left: 20px;">
        <li style="text-align: justify;">Parameter Grid Definition: Define a grid of hyperparameters to be tuned. Each hyperparameter is specified along with a list of values to try.</li><br>
        <li style="text-align: justify;">Cross-Validation: Split the dataset into multiple folds for cross-validation. The data is divided into training and validation sets for each fold.</li><br>
        <li style="text-align: justify;">Model Training: For each combination of hyperparameters in the grid:
            <ul style="list-style-type: circle; padding-left: 20px;">
                <li>Train the model on the training data.</li>
                <li>Evaluate the model's performance on the validation data.</li>
            </ul>
        </li><br>
        <li style="text-align: justify;">Performance Evaluation: Use a scoring metric (such as accuracy, F1-score, etc.) to evaluate the model's performance for each combination of hyperparameters.</li><br>
        <li style="text-align: justify;">Hyperparameter Selection: Identify the combination of hyperparameters that yields the best performance based on the chosen scoring metric.</li><br>
        <li style="text-align: justify;">Model Evaluation: Optionally, evaluate the best model on a separate test set to estimate its performance on unseen data.</li>
    </ol><br>
    <p style="text-align: justify;">GridSearchCV automates this process, making it easy to find the optimal hyperparameters for a given model and dataset. It helps in improving model performance and generalization by fine-tuning the hyperparameters.</p>

<h2><b><span style="color:#DAA03DFF">?? Step 5.4.3   </span><span style="color:#FF5733">| ??  Confusion Matrix for KNN</span></b></h2>

<div style="background-color:#eef1fb; padding: 20px; border-radius: 10px; box-shadow: 0 2px 4px 0 rgba(0, 0, 0, 0.1);border:2px solid #0A2342">
    <center style="font-size: 22px; color: Blue;">_?_?_? <u>Confusion Matrix</u> ?_?_?_</center><br><br>

A confusion matrix is a table that describes the performance of a classification model. It is a useful tool for evaluating the performance of a classifier, especially in binary classification problems.

In a binary classification problem, a confusion matrix is typically represented as follows:


\begin{array}{|c|c|}
\hline
\text{Predicted} & \text{Actual Positive} & \text{Actual Negative} \\
\hline
\text{Positive} & \text{TP (True Positive)} & \text{FP (False Positive)} \\
\text{Negative} & \text{FN (False Negative)} & \text{TN (True Negative)} \\
\hline
\end{array}


- **True Positive (TP)**: Instances correctly predicted as "Positive".
- **False Positive (FP)**: Instances incorrectly predicted as "Positive" (actually "Negative").
- **True Negative (TN)**: Instances correctly predicted as "Negative".
- **False Negative (FN)**: Instances incorrectly predicted as "Negative" (actually "Positive").
<br><br><br>
- **Precision**: $$ \frac{TP}{TP + FP} $$
<br><br><br>
- **Recall**: $$ \frac{TP}{TP + FN} $$
<br><br><br>
- **F1-score**: $$ 2 \times \frac{Precision \times Recall}{Precision + Recall} $$
<br><br><br>
- **Accuracy**: $$ \frac{TP + TN}{TP + FP + FN + TN} $$
<br><br><br>
- **Macro Average**: $$ \frac{Precision_{Positive} + Precision_{Negative}}{2} $$, $$ \frac{Recall_{Positive} + Recall_{Negative}}{2} $$, $$ \frac{F1-score_{Positive} + F1-score_{Negative}}{2} $$
<br><br><br>
- **Weighted Average**: $$ \frac{Precision_{Positive} \times Support_{Positive} + Precision_{Negative} \times Support_{Negative}}{Support_{Positive} + Support_{Negative}} $$, $$ \frac{Recall_{Positive} \times Support_{Positive} + Recall_{Negative} \times Support_{Negative}}{Support_{Positive} + Support_{Negative}} $$, $$ \frac{F1-score_{Positive} \times Support_{Positive} + F1-score_{Negative} \times Support_{Negative}}{Support_{Positive} + Support_{Negative}} $$
<br><br>

### Example

Consider a binary classification problem where we have 100 instances of each class.


\begin{array}{|c|c|c|}
\hline
\text{Predicted} & \text{Actual Positive} & \text{Actual Negative} \\
\hline
\text{Positive} & 85 \text{ (TP)} & 10 \text{ (FP)} \\
\text{Negative} & 15 \text{ (FN)} & 90 \text{ (TN)} \\
\hline
\end{array}


In this example:

- There are 100 instances of the "Actual Positive" class.
- The classifier correctly predicted 85 instances as "Positive" (True Positives).
- The classifier incorrectly predicted 15 instances as "Negative" when they were actually "Positive" (False Negatives).
- There are also 100 instances of the "Actual Negative" class.
- The classifier incorrectly predicted 10 instances as "Positive" when they were actually "Negative" (False Positives).
- The classifier correctly predicted 90 instances as "Negative" (True Negatives).
- **Precision**: $$ \frac{85}{85 + 10} = 0.8947 $$<br><br><br>
- **Recall**: $$ \frac{85}{85 + 15} = 0.8500 $$<br><br><br>
- **F1-score**: $$ 2 \times \frac{0.8947 \times 0.8500}{0.8947 + 0.8500} = 0.8717 $$<br><br><br>
- **Accuracy**: $$ \frac{85 + 90}{85 + 10 + 15 + 90} = 0.8750 $$<br><br><br>
- **Macro Average Precision**: $$ \frac{0.8947 + 0.9000}{2} = 0.8973 $$<br><br><br>
- **Macro Average Recall**: $$ \frac{0.8500 + 0.9000}{2} = 0.8750 $$<br><br><br>
- **Macro Average F1-score**: $$ \frac{0.8717 + 0.9000}{2} = 0.8859 $$<br><br><br>
- **Weighted Average Precision**: $$ \frac{0.8947 \times 100 + 0.9000 \times 100}{100 + 100} = 0.8973 $$<br><br><br>
- **Weighted Average Recall**: $$ \frac{0.8500 \times 100 + 0.9000 \times 100}{100 + 100} = 0.8750 $$<br><br><br>
- **Weighted Average F1-score**: $$ \frac{0.8717 \times 100 + 0.9000 \times 100}{100 + 100} = 0.8859 $$

````python
def clf_plot(y_pred):
    '''
    1) Plotting Confusion Matrix
    2) Plotting Classification Report'''
    # Define the colormap
    cmap = sns.diverging_palette(240, 10, s=75, l=40, n=9, center="dark", as_cmap=True)
    
    cm = metrics.confusion_matrix(y_test, y_pred)
    cr = pd.DataFrame(metrics.classification_report(y_test, y_pred, digits=3, output_dict=True)).T
    cr.drop(columns='support', inplace=True)
    
    fig, ax = plt.subplots(1, 2, figsize=(15, 5), facecolor='silver')
    
    # Left AX : Confusion Matrix
    ax[0] = sns.heatmap(cm, annot=True, fmt='', cmap=cmap, ax=ax[0])
    ax[0].set_xlabel('Predicted labels', fontsize=18)
    ax[0].set_ylabel('True labels', fontsize=18)
    ax[0].set_title('Confusion Matrix', fontsize=25)
    ax[0].xaxis.set_ticklabels(['Yes', 'No'])
    ax[0].yaxis.set_ticklabels(['Yes', 'No'])
    
    plt.gca().set_facecolor('silver')  # Set subplot background color to silver
    
    # Right AX : Classification Report
    ax[1] = sns.heatmap(cr, cmap=cmap, annot=True, linecolor='white', linewidths=0.5, ax=ax[1])
    ax[1].xaxis.tick_top()
    ax[1].set_title('Classification Report', fontsize=25)
    plt.show()
````

Decision Tree is a versatile classification algorithm that constructs a tree-like structure by recursively splitting the dataset based on the feature that provides the most information gain.

- **`criterion`:** Split criterion. Options include 'gini', 'entropy', and 'log_loss'
- **`max_depth`:** Maximum depth of the tree.
- **`min_samples_split`:** Minimum number of samples required to split an internal node.
- **`min_samples_leaf`:** Minimum number of samples required to be at a leaf node.
- **`max_features`:** Number of features to consider when looking for the best split. Options include 'auto', 'sqrt', 'log2', and None.
    
### Parameter Tuning Grid:

```python
param_grid = {

'max_features': ['auto', 'sqrt', 'log2', None] # Number of features to consider when looking for the best split
}
```

This grid defines the range of values for the hyperparameters of the Decision Tree classifier, including 'criterion', 'max_depth', 'min_samples_split', 'min_samples_leaf', and 'max_features'. You can utilize this grid for tuning the Decision Tree classifier using techniques such as grid search.

<div style="border-radius: 10px; padding: 15px; background-color: #D6EAF8; font-size: 115%; text-align: left;">
    
<h2 align="left"><font color="blue">Random Forest Classification</font></h2>
    
Random Forest is an ensemble learning method that constructs multiple decision trees during training and outputs the mode of the classes as the prediction.

### Parameters:
- **`n_estimators`:** Number of trees in the forest.
- **`max_depth`:** Maximum depth of the trees.
- **`min_samples_split`:** Minimum number of samples required to split an internal node.
- **`min_samples_leaf`:** Minimum number of samples required to be at a leaf node.
- **`max_features`:** Number of features to consider when looking for the best split. Options include 'auto', 'sqrt', 'log2'.
- **`bootstrap`:** Whether bootstrap samples are used when building trees.
    
### Parameter Tuning Grid:
```python
param_grid_rf = {
    'n_estimators': [10, 20, 30],                      # Number of trees in the forest
    'max_depth': [10, 20, 30],                         # Maximum depth of the trees
    'min_samples_split': [2, 5, 10, 15, 20],           # Minimum number of samples required to split an internal node
    'min_samples_leaf': [1, 2, 4, 6, 8],               # Minimum number of samples required to be at a leaf node
    'max_features': ['auto', 'sqrt', 'log2'],          # Number of features to consider when looking for the best split
    'bootstrap': [True]                                # Whether bootstrap samples are used when building trees
}
```
___
This grid defines the range of values for the hyperparameters of the Random Forest classifier, including 'n_estimators', 'max_depth', 'min_samples_split', 'min_samples_leaf', 'max_features', and 'bootstrap'. You can utilize this grid for tuning the Random Forest classifier using techniques such as grid search.

````python
# Define the parameter grid for Random Forest
param_grid_rf = {
    'n_estimators': [10, 20, 30],                         # Number of trees in the forest
    'max_depth': [10, 20, 30],                            # Maximum depth of the trees
    'min_samples_split': [2, 5, 10, 15, 20],              # Minimum number of samples required to split an internal node
    'min_samples_leaf': [1, 2, 4, 6, 8]  ,               # Minimum number of samples required to be at a leaf node
    'max_features': ['auto', 'sqrt', 'log2'],            # Number of features to consider when looking for the best split
    'bootstrap': [True]                                 # Whether bootstrap samples are used when building trees
}
# Create an instance of RandomForestClassifier
rf_classifier = RandomForestClassifier(random_state=42)

# Perform grid search using cross-validation
rf_search = GridSearchCV(estimator=rf_classifier, param_grid=param_grid_rf, cv=5, scoring='accuracy')

# Fit the grid search to the data
rf_search.fit(X_train, y_train)

# Get the best parameters and best score
best_params_rf = rf_search.best_params_
best_score_rf = rf_search.best_score_

# Print best parameters and best score with colored text
print(f'{color.BLUE}Best Parameters for Random Forest:{color.END} {best_params_rf}')
print(f'{color.LIGHT_MAGENTA}Best Score for Random Forest:{color.END} {best_score_rf}')
````

````python
# Random Forest
rf = RandomForestClassifier(**rf_search.best_params_)

# Fit Random Forest model
rf.fit(X_train, y_train)

# Predict on the training and testing sets
y_pred_train_rf = rf.predict(X_train)
y_pred_test_rf = rf.predict(X_test)

# Calculate accuracy scores
train_score_rf = round(accuracy_score(y_train, y_pred_train_rf), 4)
test_score_rf = round(accuracy_score(y_test, y_pred_test_rf), 4)

print(f'{color.CYAN}Random Forest Train Score:{color.END} {train_score_rf}')
print(f'{color.LIGHT_MAGENTA}Random Forest Test Score:{color.END} {test_score_rf}')
````

<h2 align="left"><font color="blue">Support Vector Machine (SVM) Classification</font></h2>
    
Support Vector Machine (SVM) is a powerful supervised learning algorithm used for classification tasks.

### Parameters:
- **`C`:** Regularization parameter.
- **`kernel`:** Kernel type. Options include 'linear', 'poly', 'rbf', and 'sigmoid'.
- **`gamma`:** Kernel coefficient. Options include 'scale' and 'auto'.
- **`shrinking`:** Whether to use the shrinking heuristic. Options include True and False.
- **`probability`:** Whether to enable probability estimates. Options include True and False.
- **`class_weight`:** Weights associated with classes. Options include None and 'balanced' (use 'balanced' for imbalanced datasets).
- **`max_iter`:** Maximum number of iterations. Options include -1 and 1000 (-1 for no limit).

### Parameter Tuning Grid:
```python
param_grid_svm = {
    'C': [0.1, 1, 10],                     # Regularization parameter
    'kernel': ['linear', 'poly', 'rbf', 'sigmoid'],           # Kernel type
    'gamma': ['scale', 'auto'],            # Kernel coefficient
    'shrinking': [True, False],            # Whether to use the shrinking heuristic
    'probability': [True, False],          # Whether to enable probability estimates
    'class_weight': [None, 'balanced'],    # Weights associated with classes (use 'balanced' for imbalanced datasets)
    'max_iter': [-1, 1000]                 # Maximum number of iterations (-1 for no limit)
}
```
___
This grid defines the range of values for the hyperparameters of the Support Vector Machine (SVM) classifier. You can utilize this grid for tuning the SVM classifier using techniques such as grid search.

````python
# Define the parameter grid for SVM
param_grid_svm = {
    'C': [0.1, 1, 10],                     # Regularization parameter
    'kernel': ['linear', 'poly', 'rbf', 'sigmoid'],           # Kernel type
    'gamma': ['scale', 'auto'],            # Kernel coefficient
    'shrinking': [True, False],            # Whether to use the shrinking heuristic
    'probability': [True, False],          # Whether to enable probability estimates
    'class_weight': [None, 'balanced'],    # Weights associated with classes (use 'balanced' for imbalanced datasets)
    'max_iter': [-1, 1000]                 # Maximum number of iterations (-1 for no limit)
}

# Create an instance of SVM classifier
svm_classifier = SVC()

# Perform grid search using cross-validation
svm_search = GridSearchCV(estimator=svm_classifier, param_grid=param_grid_svm, cv=5, scoring='accuracy')

# Fit the grid search to the data
svm_search.fit(X_train, y_train)

# Get the best parameters and best score
best_params_svm = svm_search.best_params_
best_score_svm = svm_search.best_score_

# Print best parameters and best score with colored text
print(f'{color.BLUE}Best Parameters for SVM:{color.END} {best_params_svm}')
print(f'{color.LIGHT_MAGENTA}Best Score for SVM:{color.END} {best_score_svm}')
````

````python
# Create an instance of SVC with the best parameters
svm = SVC(**svm_search.best_params_)

# Fit SVM model
svm.fit(X_train, y_train)

# Predict on the training and testing sets
y_pred_train_svm = svm.predict(X_train)
y_pred_test_svm = svm.predict(X_test)

# Calculate accuracy scores
train_score_svm = round(accuracy_score(y_train, y_pred_train_svm), 4)
test_score_svm = round(accuracy_score(y_test, y_pred_test_svm), 4)

print(f'{color.CYAN}SVM Train Score:{color.END} {train_score_svm}')
print(f'{color.LIGHT_MAGENTA}SVM Test Score:{color.END} {test_score_svm}')
````
