## 使用 Numpy 处理数组和矩阵

NumPy 允许您有效地操作**向量**（一维数组）和**矩阵**（二维数组），以及执行向量化数学运算。


### 向量和矩阵

**向量** 是一个一维数组：

```python
x = np.array([1, 2, 3, 4])
```

**矩阵**是一个二维数组：

```python
X = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])
```

### 桌子的形状（`shape`）

```python
x.shape
X.shape
```

* 向量的形式为 `(n,)`
* 矩阵的形式为“(n_rows, n_columns)”

该表格对于检查操作的兼容性至关重要。



### 使用“reshape”改变形状

`reshape` 方法允许您**修改数组的形状**而不更改其值。

```python
x = np.arange(1, 11)
X = x.reshape(5, 2)
```

* `x` 包含 10 个元素
* `X` 是 5 行 2 列的矩阵
* 元素总数必须保持不变

```python
x.size == X.size
```

`reshape` 通常用于传递：

* 从向量到矩阵
* 从线性表示到表格表示



### 总和与平均值

```python
np.sum(x)
np.mean(x)
```

对于矩阵，可以指定轴：

```python
np.sum(X, axis=0)  #每列总和
np.sum(X, axis=1)  #每行总和
```

### 点积

两个向量之间的**点积**是线性回归中的基本运算。

```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

np.dot(u, v)
```

它对应于：

$$
u \cdot v = \sum_i u_i v_i
$$

### 矩阵向量乘法

```python
X = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

w = np.array([0.5, 1.0])

X @ w
```

此操作是线性模型的核心：

$$
\hat{y} = X w
$$

### `*` 和 `@` 之间的区别

```python
X * w    #逐个元素乘法
X @ w    #矩阵向量积
```

* `*`：逐元素运算
* `@`: 矩阵乘积


### 逐个元素的操作

```python
x + 1
x * 2
x ** 2
```

这些操作被矢量化并取代显式循环。

### 使用 `df.describe()` 进行描述性统计

`describe()` 方法允许您快速获取 DataFrame 数字列的**统计摘要**。

```python
df.describe()
```

它自动计算描述**数据分布**的几个指标。


#### `describe()` 提供的统计信息

对于每个数字列，“describe()”返回以下统计信息。



##### 1. `计数`

* **非零**值的数量
* 如果值缺失，则可能小于行数



##### 2.`意思`

* 数值的算术平均值
* 给出变量的总体中心值


##### 3.`std`（标准差）

* 测量数据围绕平均值的**离散度**
* 标准差越大，数值越分散



##### 4.`分钟`

* 列中观察到的最小值



##### 5. 四分位数 (25%, 50%, 75%)

四分位数将**有序**数据分成四个相等的部分。

###### 第一个四分位数 (25%)

* 25% 的值**小于或等于**该值
* 经常注意到 (Q_1)



###### 第二个四分位数 (50%) — 中位数

* 将数据分成两等份的值
* 50% 的值低于，50% 高于
* 对应于**中位数**


###### 第三个四分位数 (75%)

* 75% 的值**小于或等于**该值
* 经常注意到 (Q_3)



###### 四分位数的解释

四分位数允许您：

*了解数据的**分布**
* 确定**中心分散**
* 检测**非典型值**（异常值）

四分位数范围定义为：

$IQR = Q_3 - Q_1$

##### 6.`最大`

* 列中观察到的最大值



### 使用 Seaborn 可视化分布：箱线图

**Seaborn** 是一个构建在 Matplotlib 之上的统计可视化库。
它提供更具表现力的图形，更适合探索性数据分析，并具有面向 *DataFrame* 的语法。

我们将使用 Seaborn 来表示变量“co2”的分布。


#### 1.提醒：箱线图原理

**箱线图** 从五个描述性统计数据合成数值分布：

* 最小值
* 第一个四分位数 (Q_1) (25%)
* 中位数 (Q_2) (50%)
* 第三个四分位数 (Q_3) (75%)
* 最大

该框代表四分位数范围：

[
IQR = Q_3 - Q_1
]

该区间包含中间 50% 的数据。


#### 2.图形结构

* 盒子从(Q_1)延伸到(Q_3)
* 中心线对应**中位数**
*须线延伸至非异常值

这种类型的图表可以让您快速评估：

* 分散性
* 不对称性
* 数据集中度



#### 3.异常值的定义

如果出现以下情况，则将观察值视为异常值：

[
x < Q_1 - 1.5 \times IQR
]

或

[
x > Q_3 + 1.5 \times IQR
]

这些值通常以点的形式单独表示。


#### 4. 使用 Seaborn 实施

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))

sns.boxplot(y=df["co2"])

plt.title("Boxplot of the CO2 variable")
plt.show()
```

#### 5.异常值的解释

异常值可能会导致：

* 测量误差
* 输入错误
* 非典型事件
* 罕见但合法的现象

应避免任何自动删除。
在做出任何清洁决定之前，必须进行背景分析。



## 使用 Seaborn 进行分布可视化

**Seaborn** 是一个构建在 Matplotlib 之上的统计可视化库。
它使得使用适合 DataFrame 的语法有效地表示数值变量的分布成为可能。

用于探索性分析的三种基本可视化是：

* 直方图
* 密度曲线（KDE）
* 小提琴情节



### Seaborn 的直方图

```python
plt.figure(figsize=(6, 4))

sns.histplot(
    data=df,
    x="co2",
    bins=100
)

plt.title("Distribution of the CO2 variable")
plt.show()
```

直方图通过将观察结果分组为**间隔（箱）**来表示分布：

* 每个bin对应一个值范围
*高度表示该范围内的观测值数量
* `bins` 参数控制 bin 的数量

太少的垃圾箱隐藏了分布的精细结构。
数字太高会放大随机波动。

因此，箱数的选择直接影响解释。



### 密度曲线 (KDE)

**核密度估计 (KDE)** 提供连续且平滑的分布估计。

与直方图不同：

* 不存在离散类别的划分
* 我们得到一条连续曲线
* 曲线下面积为1

```python
plt.figure(figsize=(6, 4))

sns.kdeplot(
    data=df,
    x="co2",
    fill=True
)

plt.title("Density estimation (KDE)")
plt.show()
```

曲线的高度反映了观测值的相对集中度。



### 直方图 + KDE

Seaborn 允许您组合两种表示形式：

```python
plt.figure(figsize=(6, 4))

sns.histplot(
    data=df,
    x="co2",
    bins=100,
    kde=True
)

plt.title("Histogram with density estimation")
plt.show()
```

条形显示观察到的频率。
叠加曲线显示了潜在的连续趋势。




### Seaborn 的小提琴情节

```python
plt.figure(figsize=(6, 4))

sns.violinplot(
    y=df["co2"],
    inner="box"
)

plt.title("Violin plot of the CO2 variable")
plt.show()
```

小提琴图也依赖于密度估计。

* 小提琴的宽度对应于密度
* 大面积表示高浓度
* 狭窄区域表示稀有值

`inner="box"` 选项显示在里面：

* 中位数
* 四分位数
* 异常值

因此，小提琴图结合了：

* 统计摘要（箱线图类型）
* 分布的连续表示


### 图形的互补性

* 直方图描述了离散类别的分布
* KDE提供连续近似
* 小提琴图结合了密度和描述性统计

这些可视化是互补的，构成任何建模之前探索性分析的重要步骤。


### 使用 Seaborn 绘制散点图

**点云**允许您研究**两个数值变量**之间的关系。

散点图让我们可以观察到：

* 变量之间存在关系


**分散**

**异常值**

## 使用 `update_traces` 进行自定义

`px.*` 函数生成一个包含一个或多个**痕迹**（图形对象）的图形。

方法：

```python
fig.update_traces(...)
```

允许您在创建轨迹后修改轨迹的图形属性。

例子 ：


这里：

* `opacity` 修改透明度
* `marker_line_color` 设置轮廓颜色
* `marker_line_width` 设置轮廓的粗细

这种方法区分：

1. 创建图表 (`px.*`)
2.美观调整（`update_traces`）

## Python 对象


### 用 Python 构建一个类

Python 中的 **类** 允许您通过分组来定义自定义对象类型：

* **数据**称为*属性*
* **函数**称为*方法*

以下示例说明了 Python 类的基本元素。

```python
#定义一个非常简单的类来说明属性+方法
class Calculatrice:
    def __init__(self):
        #实例属性：每个对象都会有自己的“内存”
        self.memoire = 10

    def add(self, x):
        #修改内部状态（内存）的方法
        self.memoire = self.memoire + x
```

#### 1. 类定义

```python
class Calculatrice:
```

“class”关键字允许您定义一个新类。
`Calculator` 是可以实例化的对象类型的名称。



#### 2. `__init__` 构造函数

```python
def __init__(self):
```

* `__init__` 是类的 **构造函数**
* 创建对象时**自动**调用
* 用于**初始化对象的属性**

在此示例中，构造函数不接受任何其他参数，但它可以接收一些参数。



#### 3. `self` 参数

* `self` 代表类的**当前实例**
* 它始终是实例方法的**第一个参数**
* 它提供对对象的属性和其他方法的访问

如果没有 self，Python 将不知道代码引用的是哪个对象。



#### 4.实例属性

```python
self.memoire = 10
```

* `memory` 是一个 **实例属性**
* 每个“计算器”对象都有**其自己的“内存”值**
* 该属性是在调用构造函数时创建的

因此，两个不同的对象具有独立的记忆。



#### 5. 类方法

```python
def add(self, x):
```

* **方法**是在类内部定义的函数
* 它通过“self”作用于对象的内部状态
* 这里，`add`方法修改了`memory`属性

```python
self.memoire = self.memoire + x
```

该指令更新对象的内部状态。



#### 6. 创建和使用对象

```python
calc = Calculatrice()   #来自制造商的电话
calc.add(5)             #调用一个方法
print(calc.memoire)     #海报 15
```

* `Calculator()` 创建一个新实例
* `calc.add(5)` 修改该对象的内部内存
* 方法通过点符号调用


#### 7. 总结

* 类定义对象的**结构和行为**
* `__init__` 初始化属性
* `self` 指的是当前实例
* 属性存储状态
* 方法修改或利用此状态


### 在Python中实例化一个对象

**实例化**对应于从类中具体创建对象。
它允许您从**抽象定义**（类）移动到**内存中可用的对象**。

```python
#实例化：创建计算器对象
c = Calculatrice()
```

#### 1. 类调用

```python
Calculatrice()
```

* 用括号调用类会触发**新对象的创建**
* Python为该对象预留内存空间
* 然后**自动**调用 `__init__` 构造函数



#### 2.自动调用构造函数`__init__`

教学期间：

```python
c = Calculatrice()
```

Python 隐式执行：

```python
Calculatrice.__init__(c)
```

在我们的示例中，构造函数初始化属性：

```python
self.memoire = 10
```

因此，新创建的对象立即具有内部状态。



#### 3. 创建引用变量

```python
c = ...
```

* `c` 是一个 **参考变量**
* 指向内存中创建的`Calculator`对象
* 对象的存在与引用它的变量的名称无关



#### 4.对象的初始状态

实例化后：

```python
c.memoire
```

返回值：

```text
10
```

从“Calculator”类实例化的每个新对象都有其**自己的独立内存**。



#### 5.类/对象区别

* **类**：行为和结构的描述（`Calculator`）
* **Object**：从类（`c`）创建的具体实例

同一个类可以产生**所需数量的对象**，每个对象都有自己的状态。



#### 6. 总结

* 实例化在内存中创建一个新对象
* 自动调用`__init__`构造函数
* 属性在创建时初始化
* 变量(`c`)引用创建的对象



### 访问属性和调用方法

一旦对象被实例化，就可以：

* 从 **读取其属性**
* **调用其方法**来修改其内部状态

以下说明说明了这两种机制。

```python
#读取内存属性
c.memoire

#调用方法：向内存添加 20
c.add(20)

#检查新的内存值
c.memoire
```

#### 1. 读取属性

```python
c.memoire
```

* “内存”属性通过**点符号**访问
* Python读取`c`引用的对象中存储的值
* 没有进行状态修改

此时，返回值为：

```text
10
```

#### 2. 调用方法

```python
c.add(20)
```

* `add` 是一个 **`Calculator` 类的方法**
* 调用也是通过点符号进行的
* 参数“20”被传递给参数“x”
* `self`参数自动链接到`c`对象

方法内执行的语句是：

```python
self.memoire = self.memoire + x
```

它修改了对象的内部状态。



#### 3.改变对象的状态

调用该方法后：

* `memory` 的值从 `10` 变为 `30`
* 只要该对象存在于内存中，该对象就会保留这个新值



#### 4.属性的新读取

```python
c.memoire
```

该指令允许您验证内部状态是否已更新。
现在返回值是：

```text
30
```

#### 5. 要点

* **属性** 存储对象的状态
* **方法**可以读取和修改此状态
* 通过方法进行的任何修改对于相关对象来说都是**持久的**

### 提醒 — 格式化字符串（f 字符串）

在 Python 中，**f-string** 允许您直接在字符串中插入变量。

字符串前面的字母“f”表示应该计算大括号“{}”中的表达式。



#### 插入变量

```python
year = 3
capital = 1234.567

print(f"Year {year}: capital = {capital}")
```

结果 ：

```
Année 3 : capital = 1234.567
```

#### 格式化数字

可以**控制数字的显示**。

```python
print(f"{capital:.2f}")
```

* `.2f` 表示：

  * `f`：实数（浮点数）
  * `.2`: 小数点后两位

结果：

```
1234.57
```

为了显示，数字被四舍五入。



#### 指令分解

```python
print(f"Year {year}: capital = {self.capital:.2f}")
```

* `f"..."`: 格式化字符串
* `{year}`: 插入`year`变量的值
* `{self.capital:.2f}`：插入大写并保留两位小数
* `print`：打印笔记本输出中的文本
