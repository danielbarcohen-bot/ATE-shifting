## **Landasan Teori**

Algoritma pembelajaran mesin dapat secara umum diklasifikasikan menjadi dua kategori - pembelajaran terawasi (supervised learning) dan pembelajaran tak terawasi (unsupervised learning). Algoritma ini mengikuti cara yang mudah atau sederhana untuk mengklasifikasikan himpunan data yang diberikan melalui sejumlah klaster, yang telah ditentukan sebelumnya. Algoritma K-Means bekerja secara iteratif untuk menetapkan setiap titik data ke salah satu dari K kelompok berdasarkan fitur-fitur yang disediakan. Titik-titik data dikelompokkan berdasarkan kesamaan fitur.

## **Intuisi K-Means Clustering**
K-Means Clustering dapat digunakan untuk menemukan grup intrinsik dengan dataset tanpa label dan gambar yang mereferensikannya. Ini berdasarkan basis klustering centroid.
Centroid adalah poin data yang berada pada pusat dari sebuah cluster. Pada basis kluster centroid, setiap cluster direpresentasikan oleh sebuah centroid. Itu merupakan sebuah algoritma iterativ yang mana gagasan dari persamaannya berasal dari seberapa dekat sebuah titik data dengan centroid dari kluster. K-Means Clustering bekerja sebagai berikut: algoritma klustering K-Means menggunakan prosedur iterativ untuk mengirim hasil akhir. Algoritma tersebut membutuhkan angka dari kluster K dan dataset sebagai input. Dataset merupakan koleksi dari fitur untuk setiap titik data. Algoritma dimulai dengan estimasi inisial untuk K setiap centroid. Kemudian algoritma berulang di antara dua langkah yaitu:
### Langkah Penugasan Data
Setiap Centroid mendefinisikan salah satu kluster. Pada langkah ini, setiap titik data ditetapkan ke centroid terdekat, yang didasarkan pada jarak Euclidean kuadrat. Jadi, jika ci adalah koleksi centroid di dalam set C, maka setiap titik data ditetapkan ke kluster berdasarkan jarak minimun Euclidean.

### Langkah Pembaruan Centroid
Pada langkah ini, setiap centroid dikomputasi ulang dan diperbaharui. Ini dilakukan dengan mengambil rata-rata dari semua titik data yang ditugaskan ke cluster centroid tersebut.
Algoritma kemudian melakukan iterasi antara langkah 1 dan langkah 2 hingga kriteria penghentian terpenuhi. Kriteria penghentian berarti tidak ada titik data yang mengubah cluster, jumlah jarak diminimalkan atau sejumlah iterasi maksimum tercapai. Algoritma ini dijamin konvergen ke hasil. Hasilnya mungkin merupakan arti optimal lokal yang menilai lebih dari satu proses algoritma dengan centroid awal acak dapat memberikan hasil yang lebih baik.

## **Menentukan Nilai K**
Algoritma K-Means bergantung pada penemuan jumlah kluster dan label data untuk nilai K yang telah ditentukan sebelumnya. Untuk menemukan jumlah kluster dalam data, kita perlu menjalankan algoritma pengelompokan K-Means untuk nilai K yang berbeda dan membandingkan hasilnya. Oleh karena itu, performa algoritma K-Means bergantung pada nilai K. Kita sebaiknya memilih nilai K yang optimal yang memberikan performa terbaik. Terdapat berbagai teknik yang tersedia untuk menemukan nilai K yang optimal. Teknik yang paling umum adalah metode Elbow.

## **Elbow Method**
Metode elbow digunakan untuk menentukan jumlah kluster yang optimal dalam pengelompokan K-Means. Metode elbow memplot nilai dari fungsi biaya yang dihasilkan oleh berbagai nilai K. Diagram di bawah ini menunjukkan bagaimana metode elbow bekerja:



```python
Image('/kaggle/input/indonesia-internet-speed-2022/gambar3.png')
```

Ketika nilai K meningkat, distorsi rata-rata akan menurun. Kemudian setiap kluster akan memiliki lebih sedikit jarak anggota, dan jarak-jarak tersebut akan lebih dekat dengan pusat kluster masing-masing. Namun, peningkatan distorsi rata-rata akan semakin melambat seiring dengan peningkatan nilai K. Nilai K di mana peningkatan distorsi mencapai puncak penurunan yang paling tajam disebut sebagai "elbow" atau siku, di mana kita sebaiknya berhenti membagi data menjadi kluster yang lebih lanjut.

Terlihat bahwa ada data yang memiliki nilai yang hilang (*missing value)*, yaitu kolom "mobile" sebanyak 65 nilai.

### Menghapus data yang hilang

```python
df.dropna(inplace=True)
```

### Melihat kembali apakah ada data yang memiliki nilai yang kosong (*missing value*)

```python
df.isnull().sum()
```

Mengdeklarasikan DataFrame baru yang berisikan "mobile" dan "broadband" saja.

```python
X.info()
```

### Melakukan **scaling** angka.

```python
cols = X.columns
```

```python
from sklearn.preprocessing import MinMaxScaler

ms = MinMaxScaler()

X = ms.fit_transform(X)
```

```python
X = pd.DataFrame(X, columns=[cols])
```

```python
X.head()
```

## **Melakukan Perhitungan K-Means Clustering**

### Menerapkan K-Means dengan nilai K = 2.

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=0)

kmeans.fit(X)

label_2 = kmeans.fit_predict(X)
```

### Melihat nilai **inersia**
Semakin rendah nilai inersia, semakin baik. Dan jika inersia sama dengan nol, maka metode ini adalah metode yang optimal untuk diterapkan di dataset ini.

```python
kmeans.inertia_
```

### Melihat nilai **centeroid** dari klaster-klaster.
K-Means Clustering bertujuan untuk memilih centroid yang akan meminimalisir nilai inersia

```python
kmeans.cluster_centers_
```

Melihat model scattering untuk nilai K = 2.

```python
plt.scatter(X['mobile'], X['fixed'], c=kmeans.labels_)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], marker='x', color='red')
plt.xlabel('Mobile')
plt.ylabel('Broadband')
plt.title('K-Means Clustering K = 2')
plt.show()
```

### Melihat Elbow Method untuk mengetahui nilai K ideal
Siku yang terlihat signifikan merupakan nilai K optimal.

```python
inertia = []
for k in range(1, 16):
    kmeans = KMeans(n_clusters=k, random_state=0)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

plt.plot(range(1, 16), inertia, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Inersia')
plt.title('Elbow Method')
plt.show()
```

Dari elbow method tersebut, terlihat bahwa nilai K = 2 merupakan siku yang terlihat signifikan mengalami penurunan nilai inersia.

### Mencoba menggunakan nilai K = 4

```python
kmeans = KMeans(n_clusters=4, random_state=0)

kmeans.fit(X)

label_4 = kmeans.fit_predict(X)

plt.scatter(X['mobile'], X['fixed'], c=kmeans.labels_)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], marker='x', color='red')
plt.xlabel('Mobile')
plt.ylabel('Broadband')
plt.title('K-Means Clustering K = 4')
plt.show()
```

### Export ke CSV
Menambahkan kedua output tersebut di dalam dataframe awal, kemudian meng-*export* nya ke file "k-means-result.csv".

```python
df['k=2'] = label_2
df['k=4'] = label_4
print(df)
df.to_csv('k-means-result.csv', index=False)
```
