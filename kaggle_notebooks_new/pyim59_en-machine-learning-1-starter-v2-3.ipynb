## Arrays and matrices with Numpy

NumPy allows you to efficiently manipulate **vectors** (1D arrays) and **matrices** (2D arrays), as well as perform vectorized mathematical operations.


### Vectors and matrices

A **vector** is a 1D array:

```python
x = np.array([1, 2, 3, 4])
```

A **matrix** is a 2D array:

```python
X = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])
```


### Shape of a table (`shape`)

```python
x.shape
X.shape
```

* a vector has a form `(n,)`
* a matrix has a form `(n_rows, n_columns)`

The form is essential to check the compatibility of operations.



### Changing shape with `reshape`

The `reshape` method allows you to **modify the shape of an array** without changing its values.

```python
x = np.arange(1, 11)
X = x.reshape(5, 2)
```

* `x` contains 10 elements
* `X` is a matrix of 5 rows and 2 columns
* the total number of elements must remain the same

```python
x.size ==
```

`reshape` is often used to pass:

* from a vector to a matrix
* from a linear representation to a tabular representation



### Sums and averages

```python
np.sum(x)
np.mean(x)
```

For a matrix, it is possible to specify the axis:

```python
np.sum(X, axis=0) # sum per column
np.sum(X, axis=1) # sum per line
```


### Dot product

The **dot product** between two vectors is a fundamental operation in linear regression.

```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

np.dot(u, v)
```

It corresponds to:

$$
u \cdot v = \sum_i u_i v_i
$$



### Matrix-vector multiplication

```python
X = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

w = np.array([0.5, 1.0])

x@w
```

This operation is central in linear models:

$$
\hat{y} = X w
$$



### Difference between `*` and `@`

```python
x * w # element-by-element multiplication
X @ w # matrix-vector product
```

* `*`: element-by-element operation
* `@`: matrix product


### Element-by-element operations

```python
x + 1
x * 2
x**2
```

These operations are vectorized and replace explicit loops.

### Descriptive statistics with `df.describe()`

The `describe()` method allows you to quickly obtain a **statistical summary** of the numeric columns of a DataFrame.

```python
df.describe()
```

It automatically calculates several indicators describing the **data distribution**.


#### Statistics provided by `describe()`

For each numeric column, `describe()` returns the following statistics.



##### 1. `count`

* Number of **non-zero** values
* May be less than the number of rows if values are missing



##### 2. `mean`

* Arithmetic average of values
* Gives an overall central value of the variable


##### 3. `std` (standard deviation)

* Measures the **dispersion** of data around the mean
* The larger the standard deviation, the more dispersed the values are



##### 4. `min`

* Minimum value observed in the column



##### 5. Quartiles (25%, 50%, 75%)

Quartiles divide the **ordered** data into four equal parts.

###### First quartile (25%)

* 25% of values are **less than or equal** to this value
* Noted often (Q_1)



###### Second quartile (50%) — median

* Value that divides the data into two equal parts
* 50% of values are below, 50% above
* Corresponds to the **median**


###### Third quartile (75%)

* 75% of values are **less than or equal** to this value
* Noted often (Q_3)



###### Interpretation of quartiles

Quartiles allow you to:

* understand the **distribution** of data
* identify the **central dispersion**
* detect **atypical values** (outliers)

The interquartile range is defined by:

$IQR = Q_3 - Q_1$



##### 6. `max`

* Maximum value observed in the column


### Visualization of the distribution with Seaborn: boxplots

**Seaborn** is a statistical visualization library built on top of Matplotlib.
It offers more expressive graphics that are better suited to exploratory data analysis, with a *DataFrame* oriented syntax.

We will use Seaborn to represent the distribution of the variable `co2`.


#### 1. Reminder: boxplot principle

A **boxplot** synthesizes a numerical distribution from five descriptive statistics:

* minimum
* first quartile (Q_1) (25%)
* median (Q_2) (50%)
* third quartile (Q_3) (75%)
* maximum

The box represents the interquartile range:

[
IQR = Q_3 - Q_1
]

This interval contains the middle 50% of the data.


#### 2. Graphic structure

* The box extends from (Q_1) to (Q_3)
* The central line corresponds to the **median**
*Whiskers extend to non-outlier extreme values

This type of graph allows you to quickly evaluate:

* dispersion
* asymmetry
* data concentration



#### 3. Definition of outliers

An observation is considered an outlier if:

[
x < Q_1 - 1.5 \times IQR
]

or

[
x > Q_3 + 1.5 \times IQR
]

These values are represented individually, usually as points.


#### 4. Implementation with Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))

sns.boxplot(y=df["co2"])

plt.title("Boxplot of the CO2 variable")
plt.show()
```


#### 5. Interpretation of outliers

Outliers can result:

* measurement errors
* entry errors
* atypical events
* rare but legitimate phenomena

Any automatic deletion should be avoided.
A contextual analysis is necessary before any cleaning decision.


## Distribution visualization with Seaborn

**Seaborn** is a statistical visualization library built on top of Matplotlib.
It makes it possible to efficiently represent the distributions of numerical variables with a syntax adapted to DataFrames.

Three fundamental visualizations for exploratory analysis are:

* the histogram
* the density curve (KDE)
* the violin plot



### Histogram with Seaborn

```python
plt.figure(figsize=(6, 4))

sns.histplot(
    data=df,
    x="co2",
    bins=100
)

plt.title("Distribution of the CO2 variable")
plt.show()
```

The histogram represents the distribution by grouping observations into **intervals (bins)**:

* each bin corresponds to a range of values
*height indicates the number of observations in this range
* the `bins` parameter controls the number of bins

Too few bins hide the fine structure of the distribution.
Too high a number amplifies random fluctuations.

The choice of the number of bins therefore directly influences the interpretation.


### Density curve (KDE)

**Kernel Density Estimation (KDE)** provides a continuous and smooth estimation of the distribution.

Unlike the histogram:

* there is no division into discrete classes
* we obtain a continuous curve
* the area under the curve is 1

```python
plt.figure(figsize=(6, 4))

sns.kdeplot(
    data=df,
    x="co2",
    fill=True
)

plt.title("Density estimation (KDE)")
plt.show()
```

The height of the curve reflects the relative concentration of observations.



### Histogram + KDE

Seaborn allows you to combine the two representations:

```python
plt.figure(figsize=(6, 4))

sns.histplot(
    data=df,
    x="co2",
    bins=100,
    kde=True
)

plt.title("Histogram with density estimation")
plt.show()
```

The bars show the observed frequencies.
The superimposed curve shows the underlying continuous trend.


### Violin plot with Seaborn

```python
plt.figure(figsize=(6, 4))

sns.violinplot(
    y=df["co2"],
    inner="box"
)

plt.title("Violin plot of the CO2 variable")
plt.show()
```

The violin plot also relies on a density estimate.

* the width of the violin corresponds to the density
* large areas indicate high concentration
* narrow areas indicate rarer values

The `inner="box"` option displays inside:

* the median
* quartiles
* outliers

The violin plot therefore combines:

* a statistical summary (boxplot type)
* a continuous representation of the distribution


### Complementarity of graphics

* The histogram describes the distribution by discrete classes
* KDE provides continuous approximation
* The violin plot combines density and descriptive statistics

These visualizations are complementary and constitute an essential step in the exploratory analysis before any modeling.


### Scatter plot with Seaborn

The **point cloud** allows you to study the relationship between **two numerical variables**.
It constitutes the reference graphical tool for analyzing a potential dependence and introducing **linear regression**.

Each point corresponds to an observation from the dataset.

* The horizontal axis represents an explanatory variable (x)
* The vertical axis represents a target variable (y)
* Each point corresponds to a couple ((x_i, y_i))

The scatter plot allows us to observe:

* the existence of a relationship between the variables
* the form of the relationship (linear or not)
* the dispersion of observations
* the presence of atypical values


From the point cloud, we can analyze:

**Overall trend**

**Dispersion**
The deviation of points around an average trend reflects variability.

**Outliers**
Points far from the main cloud may correspond to atypical observations.

## Customization with `update_traces`

The `px.*` functions generate a figure containing one or more **traces** (graphic objects).

The method:

```python
fig.update_traces(...)
```

allows you to modify the graphic properties of traces after their creation.

Example:

```python
fig = px.histogram(df, x="co2", nbins=50)

fig.update_traces(
    opacity=0.7,
    marker_line_color="black",
    marker_line_width=1
)

fig.show()
```

Here:

* `opacity` modifies transparency
* `marker_line_color` sets the outline color
* `marker_line_width` sets the thickness of the outline

This approach separates:

1. creating the graph (`px.*`)
2. aesthetic adjustment (`update_traces`)

You can also modify the overall layout via:

```python
fig.update_layout(width=600, height=400)
```

### Building a class in Python

A **class** in Python allows you to define a custom object type by grouping:

* **data** called *attributes*
* **functions** called *methods*

The following example illustrates the fundamental elements of a Python class.

```python
# Definition of a very simple class to illustrate attributes + methods
class Calculator:
    def __init__(self):
        # Instance attribute: each object will have its own 'memory'
        self.memory = 10

    def add(self, x):
        # Method that modifies the internal state (memory)
        self.memory = self.memory + x
```



#### 1. Class definition

```python
class Calculator:
```

The `class` keyword allows you to define a new class.
`Calculator` is the name of the type of object which can then be instantiated.



#### 2. The `__init__` constructor

```python
def __init__(self):
```

* `__init__` is the **constructor** of the class
* It is called **automatically** when creating an object
* It is used to **initialize the attributes** of the object

In this example, the constructor does not take any additional parameters, but it could receive some.



#### 3. The `self` parameter

* `self` represents **the current instance** of the class
* It is always the **first parameter** of instance methods
* It provides access to the attributes and other methods of the object

Without `self`, Python would not know which object the code refers to.



#### 4. Instance attributes

```python
self.memory = 10
```

* `memory` is an **instance attribute**
* Each `Calculator` object has **its own `memory` value**
* The attribute is created when calling the constructor

Thus, two distinct objects have independent memories.



#### 5. Class methods

```python
def add(self, x):
```

* A **method** is a function defined inside a class
* It acts on the internal state of the object via `self`
* Here, the `add` method modifies the `memory` attribute

```python
self.memory = self.memory + x
```

This instruction updates the internal state of the object.



#### 6. Creating and using an object

```python
calc = Calculator() # constructor call
calc.add(5) # call a method
print(calc.memory) # prints 15
```

* `Calculator()` creates a new instance
* `calc.add(5)` modifies the internal memory of this object
* Methods are called via dot notation


#### 7. Summary

* A class defines the **structure and behavior** of objects
* `__init__` initializes attributes
* `self` refers to the current instance
* Attributes store state
* Methods modify or exploit this state


### Instantiating an object in Python

**Instantiation** corresponds to the concrete creation of an object from a class.
It allows you to move from an **abstract definition** (the class) to an **object usable in memory**.

```python
# Instantiation: creation of a Calculator object
c = Calculator()
```

#### 1. Class call

```python
Calculator()
```

* Calling the class with parentheses triggers the **creation of a new object**
* Python reserves memory space for this object
* The `__init__` constructor is then called **automatically**



#### 2. Automatic call of constructor `__init__`

During instruction:

```python
c = Calculator()
```

Python implicitly executes:

```python
Calculator.__init__(c)
```

In our example, the constructor initializes the attribute:

```python
self.memory = 10
```

The newly created object therefore immediately has an internal state.



#### 3. Creating the reference variable

```python
c = ...
```

* `c` is a **reference variable**
* It points to the `Calculator` object created in memory
* The object exists independently of the name of the variable that references it



#### 4. Initial state of the object

After instantiation:

```python
c.memory
```

returns the value:

```text
10
```

Each new object instantiated from the `Calculator` class has its **own independent memory**.



#### 5. Class/object distinction

* **Class**: description of behavior and structure (`Calculator`)
* **Object**: concrete instance created from the class (`c`)

The same class can give rise to **as many objects as necessary**, each with its own state.



#### 6. Summary

* Instantiation creates a new object in memory
* The `__init__` constructor is called automatically
* Attributes are initialized upon creation
* The variable (`c`) references the created object


### Accessing attributes and calling methods

Once the object has been instantiated, it is possible:

* from **read its attributes**
* to **call its methods** to modify its internal state

The following instructions illustrate these two mechanisms.

```python
# Reading the memory attribute
c.memory

# Calling a method: adds 20 to memory
c.add(20)

# Check the new memory value
c.memory
```

#### 1. Reading an attribute

```python
c.memory
```

* The `memory` attribute is accessed via **dot notation**
* Python reads the value stored in the object referenced by `c`
* No state modification is made

At this point, the returned value is:

```text
10
```



#### 2. Calling a method

```python
c.add(20)
```

* `add` is a **method of the `Calculator` class**
* The call is also made via dot notation
* Argument `20` is passed to parameter `x`
* The `self` parameter is automatically linked to the `c` object

The statement executed inside the method is:

```python
self.memory = self.memory + x
```

Which modifies the internal state of the object.



#### 3. Changing the state of the object

After calling the method:

* the value of `memory` changes from `10` to `30`
* the object keeps this new value as long as it exists in memory



#### 4. New reading of the attribute

```python
c.memory
```

This instruction allows you to verify that the internal state has been updated.
The returned value is now:

```text
30
```



#### 5. Key takeaway

* **attributes** store the state of the object
* **methods** can read and modify this state
* Any modification via a method is **persistent** for the object concerned

### Reminder — Formatted character strings (f-strings)

In Python, **f-strings** allow you to directly insert variables inside a character string.

The letter `f` placed in front of the string indicates that expressions enclosed in braces `{}` should be evaluated.



#### Inserting variables

```python
year = 3
capital = 1234,567

print(f"Year {year}: capital = {capital}")
```

Result:

```
Year 3: capital = 1234,567
```



#### Formatting numbers

It is possible to **control the display** of numbers.

```python
print(f"{capital:.2f}")
```

* `.2f` means:

  * `f`: real number (float)
  * `.2`: two digits after the decimal point

Result:

```
1234.57
```

The number is rounded for display.



#### Breakdown of the instruction

```python
print(f"Year {year}: capital = {self.capital:.2f}")
```

* `f"..."`: formatted string
* `{year}`: inserting the value of the `year` variable
* `{self.capital:.2f}`: insertion of capital with two decimal places
* `print`: prints text in notebook output
