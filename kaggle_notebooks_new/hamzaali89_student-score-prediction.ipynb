# ✅ Q1. Import Required Libraries

```python
import numpy as py
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
```

 ✅ Q2. Load the Dataset

* Load the dataset into a DataFrame using `pandas.read_csv()`.
* Display the first few rows using `.head()`.
* Show the basic structure using `.info()`.

```python
dataset=pd.read_csv('/kaggle/input/student-study-hours/score.csv')
dataset.info()
dataset.head()
```

 ✅ Q3. Prepare the Data

* Separate the features (`Hours`) and target (`Scores`).
* Use:
  `X = df[['Hours']]`
  `y = df['Scores']`

---

```python
x=dataset.iloc[:,:-1].values
y=dataset.iloc[:,-1].values
```

```python
print(x)
```

### ✅ Q4. Train the Simple Linear Regression Model

* Import `LinearRegression` from `sklearn.linear_model`.
* Fit the model on **all the data** using `.fit()`.



```python
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
regressor.fit(x,y)
```


### ✅ Q5. Make Predictions

* Use the trained model to predict scores using `.predict(X)`.
* Display both actual and predicted scores side by side.

```python
y_pred=regressor.predict(x)
```

```python
print(y_pred)
```

### ✅ Q6. Evaluate the Model

* Evaluate the model using the **entire dataset**.
* Calculate and display:

  * **Mean Absolute Error (MAE)**
  * **Mean Squared Error (MSE)**
  * **R² Score**
* Use `mean_absolute_error()`, `mean_squared_error()`, and `r2_score()` from `sklearn.metrics`.


```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
# Use the y_pred calculated on the entire dataset
y_pred=regressor.predict(x)
print(mean_absolute_error(y,y_pred))
print(mean_squared_error(y,y_pred))
print(r2_score(y,y_pred))
```

### ✅ Q7. Predict a Custom Value

* Predict the score of a student who studies for **9.25 hours** using the trained model

---


```python
y_pred=regressor.predict([[9.25]])
print(y_pred)
```
