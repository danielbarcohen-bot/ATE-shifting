```python
# ????? ? ??? ????
import pandas as pd
import numpy as np
from sklearn.preprocessing import power_transform

df = pd.read_csv('../input/bigdatacertificationkr/basic1.csv')
df.head(5)
```

```python
# ??? ?? ???
print("?? ?? ?:", df.shape)
df = df[df['age']>=20]
print("?? ?? ?:", df.shape)
```

```python
# ????? 'f1' ?? ??? ??
print("??? ?? ?: \n", df.isnull().sum())
print("???: ",df['f1'].mode()[0])
df['f1'] = df['f1'].fillna(df['f1'].mode()[0])
print("??? ?? ?: \n", df.isnull().sum())   
```

```python
# 'f1'??? ?-?? yeo-johnson ? ???
df['y'] = power_transform(df[['f1']]) # method ??? ?? ?-??’yeo-johnson’
df['y'].head()
```

```python
# 'f1'??? ?-?? yeo-johnson ? ???
df['y'] = power_transform(df[['f1']],standardize=False) # method ??? ?? ?-??’yeo-johnson’
df['y'].head()
```

```python
# 'f1'??? ??-?? box-cox ? ???
df['b'] = power_transform(df[['f1']], method='box-cox')
df['b'].head()
```

```python
# 'f1'??? ??-?? box-cox ? ???
# ?? 2022.6.21 
# standardize=False? python? R? ???? ?? ??? ??? ??????! ?? ??? ?? ???? ?? ??? (?? ??? ??? ?? ??? ?????.)
df['b'] = power_transform(df[['f1']], method='box-cox', standardize=False)
df['b'].head()
```
