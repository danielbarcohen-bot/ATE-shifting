```python
import pandas
import numpy
import sklearn
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import linear_model
from sklearn import tree
from sklearn import metrics
```

```python
data = pandas.read_csv("/kaggle/input/house-prices-advanced-regression-techniques/train.csv")
```

```python
def transform(data):
    
    def map_exterior(x):
        if x in ["MetalSd", "Wd Sdng"]:
            return -1
        elif x in ["CemntBd", "VinylSd"]:
            return 2
        else:
            return 0
        
    def map_mas(x):
        if x == "BrkCmn":
            return 0
        elif x == "None":
            return 1
        elif x == "BrkFace":
            return 2
        elif x == "Stone":
            return 3
        
    def map_ex_qual(x):
        if x == "Fa":
            return 0
        elif x == "TA":
            return 1
        elif x == "Gd":
            return 2
        elif x == "Ex":
            return 3
        
    def map_ex_cond(x):
        if x == "Po":
            return -1
        elif x == "Fa":
            return 0
        elif x == "TA":
            return 1
        elif x == "Gd":
            return 2
        elif x == "Ex":
            return 3
        
    def map_foundation(x):
        if x == "PConc":
            return 1
        elif x in ["CBlock", "BrkTil", "Slab"]:
            return 0
        else:
            return -1
        
    def map_bsmt_qual(x):
        if x == "Fa":
            return 0
        elif x == "TA":
            return 1
        elif x == "Gd":
            return 2
        elif x == "Ex":
            return 3
        
    def map_bsmt_cond(x):
        if x == "Po":
            return -1
        elif x == "Fa":
            return 0
        elif x == "TA":
            return 1
        elif x == "Gd":
            return 2
        elif x == "Ex":
            return 3
        
    def map_bsmt_exp(x):
        if x == "No":
            return 0
        elif x == "Mn":
            return 1
        elif x == "Av":
            return 2
        elif x == "Gd":
            return 3
        
    def map_bsmt_fin(x):
        if x == "Unf":
            return 0
        elif x == "LwQ":
            return 1
        elif x == "Rec":
            return 2
        elif x == "BLQ":
            return 3
        elif x == "ALQ":
            return 4
        elif x == "GLQ":
            return 5
    
    def map_heat_qc(x):
        if x == "Po":
            return -1
        elif x == "Fa":
            return 0
        elif x == "TA":
            return 1
        elif x == "Gd":
            return 2
        elif x == "Ex":
            return 3
        
    def map_kit_qual(x):
        if x == "Fa":
            return 0
        elif x == "TA":
            return 1
        elif x == "Gd":
            return 2
        elif x == "Ex":
            return 3
        
    def map_firep_qu(x):
        if x == "Po":
            return -1
        elif x == "Fa":
            return 0
        elif x == "TA":
            return 1
        elif x == "Gd":
            return 2
        elif x == "Ex":
            return 3
        
    def map_gar_type(x):
        if x in ["CarPort", "Detchd"]:
            return -1
        elif x in ["Basment", "2Types"]:
            return 0
        elif x == "Attchd":
            return 1
        elif x == "BuiltIn":
            return 2
        
    def map_gar_fin(x):
        if x == "Unf":
            return 1
        elif x == "RFn":
            return 2
        elif x == "Fin":
            return 3
        
    def map_gar_qua(x):
        if x == "Po":
            return -1
        elif x == "Fa":
            return 0
        elif x == "TA":
            return 1
        elif x in ["Gd", "Ex"]:
            return 2
        
    def map_gar_cond(x):
        if x in ["Po", "Fa"]:
            return -1
        elif x == "TA":
            return 0
        else:
            return 1
        
    def map_sale_cond(x):
        if x in ["Abnorml", "AdjLand"]:
            return -1
        elif x == "Partial":
            return 1
        else:
            return 0
        
    def map_year_built(x):
        if x <= 1950:
            return 0
        elif x <= 2000:
            return 1
        else:
            return 2
        
    def map_year_remove(x):
        if x <= 1950:
            return 0
        elif x <= 1980:
            return 1
        elif x <= 2000:
            return 2
        else:
            return 3
    
    data["MSZoning"] = data["MSZoning"].map({"FV":0, "RH":1, "RM":2, "RL":3, "C (all)":10})
    data["LotFrontage"] = data["LotFrontage"].fillna(data["LotFrontage"].mean())
    data["Street"] = data["Street"].map({"Grvl":0, "Pave":1})
    data["Alley"] = data["Alley"].map({"Grvl":0, "Pave":1})
    data["Alley"] = data["Alley"].fillna(-1)
    data["LotShape"] = data["LotShape"].map({"Reg":0, "IR1":1, "IR2":2, "IR3":3})
    data["LandContour"] = data["LandContour"].map({"Lvl":0, "Bnk":1, "HLS":2, "Low":3})
    data["LotConfig"] = data["LotConfig"].map({"Inside":0, "Corner":1, "CulDSac":2, "FR2":3, "FR3":4})
    data["LandSlope"] = data["LandSlope"].map({"Gtl":0, "Mod":1, "Sev":2})
    locs = ["OldTown", "Edwards", "NridgHt", "Sawyer", "BrkSide"]
    for i in locs:
        data[i] = data["Neighborhood"] == i
        data[i] = data[i].apply(lambda x: 1 if x == True else 0)
    data["Feedr"] = data["Condition1"] == "Feedr"
    data["Feedr"] = data["Feedr"].apply(lambda x: 1 if x == True else 0)
    data["Artery"] = data["Condition1"] == "Artery"
    data["Artery"] = data["Artery"].apply(lambda x: 1 if x == True else 0)
    data["BldgType"] = data["BldgType"].apply(lambda x: 0 if x in ["1Fam", "TwnhsE"] else 1)
    data["HouseStyle"] = data["HouseStyle"].map({"1Story":0, "2Story":1, "1.5Fin":-1, 
                        "SLvl":0, "SFoyer":0, "1.5Unf":-1, "2.5Unf":0, "2.5Fin":1})
    data["Exterior1st"] = data["Exterior1st"].apply(lambda x: map_exterior(x))
    data["MasVnrType"] = data["MasVnrType"].fillna("None")
    data["MasVnrType"] = data["MasVnrType"].apply(lambda x:map_mas(x))
    data["MasVnrArea"] = data["MasVnrArea"].fillna(0.0)
    data["ExterQual"] = data["ExterQual"].apply(lambda x:map_ex_qual(x))
    data["ExterCond"] = data["ExterCond"].apply(lambda x:map_ex_qual(x))
    data["Foundation"] = data["Foundation"].apply(lambda x:map_foundation(x))
    data["BsmtQual"] = data["BsmtQual"].fillna("TA")
    data["BsmtQual"] = data["BsmtQual"].apply(lambda x:map_bsmt_qual(x))
    data["BsmtCond"] = data["BsmtCond"].fillna("TA")
    data["BsmtCond"] = data["BsmtCond"].apply(lambda x:map_bsmt_cond(x))
    data["BsmtExposure"] = data["BsmtExposure"].fillna("No")
    data["BsmtExposure"] = data["BsmtExposure"].apply(lambda x:map_bsmt_exp(x))
    data["BsmtFinType1"] = data["BsmtFinType1"].apply(lambda x:(map_bsmt_fin(x)))
    data["HeatingQC"] = data["HeatingQC"].apply(lambda x:map_heat_qc(x))
    data["CentralAir"] = data["CentralAir"].apply(lambda x:-1 if x == "N" else 0)
    data["Electrical"] = data["Electrical"].apply(lambda x:-1 if x != "SBrkr" else 0)
    data["KitchenQual"] = data["KitchenQual"].apply(lambda x :map_kit_qual(x))
    data["FireplaceQu"] = data["FireplaceQu"].fillna("Po")
    data["FireplaceQu"] = data["FireplaceQu"].apply(lambda x:map_firep_qu(x))
    data["GarageType"] = data["GarageType"].fillna("CarPort")
    data["GarageType"] = data["GarageType"].apply(lambda x:map_gar_type(x))
    data["GarageYrBlt"] = data["GarageYrBlt"].apply(lambda x:1 if x>= 2000 else 0)
    data["GarageFinish"] = data["GarageFinish"].fillna(0)
    data["GarageFinish"] = data["GarageFinish"].apply(lambda x:map_gar_fin(x))
    data["GarageQual"] = data["GarageQual"].fillna("Po")
    data["GarageQual"] = data["GarageQual"].apply(lambda x:map_gar_qua(x))
    data["GarageCond"] = data["GarageCond"].fillna("Po")
    data["GarageCond"] = data["GarageCond"].apply(lambda x:map_gar_cond(x))
    data["PavedDrive"] = data["PavedDrive"].apply(lambda x:-1 if x == "N" else 0)
    data["SaleType"] = data["SaleType"].apply(lambda x:1 if x == "New" else 0)
    data["SaleCondition"] = data["SaleCondition"].apply(lambda x:map_sale_cond(x))
    data["location"] = data["NridgHt"]
    data["area"] = data["Feedr"] + data["Artery"]
    data["YearBuilt"] = data["YearBuilt"].apply(lambda x:map_year_built(x))
    data["YearRemodAdd"] = data["YearRemodAdd"].apply(lambda x:map_year_remove(x))
    data["ExterCond"] = data["ExterCond"].fillna(1.0)
    data["BsmtFinType1"] = data["BsmtFinType1"].fillna(0.0)
    data["GarageFinish"] = data["GarageFinish"].fillna(1.0)
    data['MSZoning'] = data["MSZoning"].fillna(3.0)
    data['BsmtFinSF1'] = data["BsmtFinSF1"].fillna(0.0)
    data["TotalBsmtSF"] = data["TotalBsmtSF"].fillna(0.0)
    data["BsmtFullBath"] = data["BsmtFullBath"].fillna(0.0)
    data["BsmtHalfBath"] = data["BsmtHalfBath"].fillna(0.0)
    data["KitchenQual"] = data["KitchenQual"].fillna(1.0)
    data["GarageCars"] = data["GarageCars"].fillna(2.0)
    data["GarageArea"] = data["GarageArea"].fillna(0.0)
    
    if "SalePrice" in data.columns:
        data["y"] = data["SalePrice"]
        data = data.drop(columns = "SalePrice")
        
    data = data.drop(columns = ["Feedr", "Artery", "YrSold", "MiscVal", "MiscFeature", 
                            "Fence", "PoolQC", "PoolArea", "Functional", "Heating", "BsmtFinType2", 
                            "BsmtFinSF2", "BsmtUnfSF", "Exterior2nd", "RoofMatl", "RoofStyle", "Condition1",
                           "Condition2", "Neighborhood", "Utilities", 'OldTown', 'Edwards', 'NridgHt', 'Sawyer', 'BrkSide',
                           "3SsnPorch", "ScreenPorch"])
    return data
```

```python
t_data = transform(data)
```

```python
x = t_data[t_data.columns[1:-1]]
y = t_data["y"]
train_x, test_x, train_y, test_y = train_test_split(x, y, test_size = 0.2, 
                                                    random_state = 2200)
cls = linear_model.LinearRegression(n_jobs = -1)
clf = cls.fit(train_x, train_y)
predict_y = clf.predict(test_x)
print("model evaluation\nrmse: {}\nr_squared: {}".format(metrics.mean_squared_error(test_y, predict_y) **0.5, 
                                       metrics.r2_score(test_y, predict_y)))
```
