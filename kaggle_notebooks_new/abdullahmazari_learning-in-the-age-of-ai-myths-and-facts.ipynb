        1. Do AI Users Perform Better Than Non-AI Users?
    </h2>
</div>


```python

# Set plot style
sns.set(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (10, 6)

# 1?? uses_ai vs final_score
plt.figure(figsize=(8,5))
sns.boxplot(x='uses_ai', y='final_score', data=df)
plt.title('Impact of AI Usage on Students\' Final Scores', fontsize=14, fontweight='bold')
plt.xlabel('Uses AI (0=No, 1=Yes)', fontsize=12)
plt.ylabel('Final Score', fontsize=12)
plt.show()

# Optional: mean scores
print("Mean final_score by AI usage:")
print(df.groupby('uses_ai')['final_score'].mean(), "\n")

```

###  *?? Insight:*
The average final score is almost identical for students who use AI and those who do not (?56.8 in both cases\
This suggests that AI usage alone does not significantly improve or harm overall academic performance. Other factors—such as study consistency, sleep, or conceptual understanding—are likely more influential than simply whether a student uses AI.

    <h2 style="margin: 0; font-size: 28px;">
        2. Do AI Users Pass More Often Than Non-AI Users?
    </h2>
</div>


```python

# uses_ai vs passed
plt.figure(figsize=(8,5))
sns.countplot(x='uses_ai', hue='passed', data=df)
plt.title('AI Usage vs Passed')
plt.xlabel('Uses AI (0=No, 1=Yes)')
plt.ylabel('Count of Students')
plt.show()

# Percentage of passing by AI usage
pass_percent = df.groupby('uses_ai')['passed'].mean() * 100
print("Pass percentage by AI usage:\n", pass_percent, "\n")
```

### *?? Insight*

The pass percentage is nearly the same for both groups—88.7% for non-AI users and 89.0% for AI users.\
This indicates that using AI does not meaningfully change the likelihood of passing. While AI users show a slightly higher pass rate, the difference is marginal and suggests that AI usage alone is not a decisive factor in academic success.

    background: linear-gradient(90deg, #4CAF50, #FFEB3B);

    <h2 style="margin: 0; font-size: 28px;">
        3. Does AI Usage Influence Students’ Performance Categories?
    </h2>
</div>


```python
# uses_ai vs performance_category
plt.figure(figsize=(8,5))
sns.countplot(x='uses_ai', hue='performance_category', data=df)
plt.title('Impact of AI Usage on Performance Category', fontsize=14, fontweight='bold')
plt.xlabel('Uses AI (0=No, 1=Yes)', fontsize=12)
plt.ylabel('Count of Students', fontsize=12)
plt.show()

# Percentage of students in each performance category by AI usage
performance_percent = df.groupby('uses_ai')['performance_category'].value_counts(normalize=True).unstack() * 100
print("Percentage of performance categories by AI usage:\n")
print(performance_percent)
```

### *?? Insight* 

The distribution of performance categories is very similar for AI users and non-AI users.
Both groups have nearly identical proportions of High, Medium, and Low performers. AI users show a slightly higher share of Medium performers and a slightly lower share of High performers, but the differences are minimal.

Overall, this suggests that AI usage does not substantially shift students into higher or lower performance categories. Performance level appears to be influenced more by study habits, consistency, and conceptual understanding than by AI usage alone.

    <h2 style="margin: 0; font-size: 28px;">
        4. Does Higher AI Dependency Affect Students’ Concept Understanding?
    </h2>
</div>


```python
# ai_dependency_score vs concept_understanding_score
plt.figure(figsize=(8,5))
sns.scatterplot(x='ai_dependency_score', y='concept_understanding_score', hue='uses_ai', data=df)
plt.title('AI Dependency Score vs Concept Understanding Score')
plt.xlabel('AI Dependency Score')
plt.ylabel('Concept Understanding Score')
plt.show()

# Optional: correlation
corr_dep_concept = df[['ai_dependency_score','concept_understanding_score']].corr().iloc[0,1]
print(f"Correlation between AI dependency and concept understanding: {corr_dep_concept:.2f}")
```

### *?? Insight*

The correlation between AI dependency and conceptual understanding is 0.02, which is effectively zero.
This indicates no meaningful linear relationship between how much students rely on AI and their level of conceptual understanding. In other words, higher AI dependency neither clearly improves nor degrades conceptual learning based on this metric alone.


<div style="
    background: linear-gradient(90deg, #4CAF50, #FFEB3B);

    <h2 style="margin: 0; font-size: 28px;">
       5. Does the Purpose of AI Usage Affect Student Performance?
    </h2>
</div>


```python
# Create performance categories
df['performance_category'] = pd.cut(
    df['final_score'],
    bins=[0, 50, 70, 85, 100],
    labels=['Low', 'Average', 'Good', 'Excellent']
)

performance_percent = (
    df.groupby('ai_usage_purpose')['performance_category']
      .value_counts(normalize=True)
      .mul(100)
      .round(2)
      .unstack()
)

print("Percentage of performance categories by AI usage purpose:\n")
print(performance_percent)

performance_percent.plot(
    kind='bar',
    stacked=True,
    figsize=(10,6)
)

plt.title('Performance Distribution by AI Usage Purpose')
plt.xlabel('AI Usage Purpose')
plt.ylabel('Percentage of Students')
plt.legend(title='Performance Category', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

### ?? Insight:
While AI usage purpose alone does not drastically change performance outcomes, students using AI for homework and note-taking show a slightly higher proportion of good and excellent performance compared to other purposes.

    background: linear-gradient(90deg, #4CAF50, #FFEB3B);

       6. Does the Purpose of AI Usage Influence Concept Understanding?
    </h2>
</div>


```python
# AI Usage Purpose vs Concept Understanding Score 
plt.figure(figsize=(8,5))
sns.boxplot(x='ai_usage_purpose', y='concept_understanding_score', data=df, order=['Exam Prep','Notes','Doubt Solving'])
plt.title('AI Usage Purpose vs Concept Understanding')
plt.xlabel('AI Usage Purpose')
plt.ylabel('Concept Understanding Score')
plt.show()

summary = (
    df.groupby('ai_usage_purpose')['concept_understanding_score']
      .agg(['mean','median','std','count'])
      .round(2)
)

print("Concept Understanding Summary by AI Usage Purpose:\n")
print(summary)

```

### *?? Insight:*
Notes (mean = 5.56) and Homework (mean = 5.55) have the highest average concept understanding scores, with a median of 6, indicating that students using AI for these purposes tend to achieve slightly better conceptual clarity.

Exam Prep (mean = 5.46) and Coding (mean = 5.43) fall in the middle, suggesting moderate conceptual understanding.

Doubt Solving (mean = 5.32) shows the lowest average score, indicating that while AI helps resolve immediate questions, it may not always translate into deeper conceptual learning.

The standard deviations (?2.8–3.0) are similar across all categories, suggesting consistent variability in student understanding regardless of AI usage purpose.

?? Key Takeaway

Using AI for structured learning activities such as notes and homework is associated with slightly better conceptual understanding than using AI primarily for doubt solving or exam preparation.

        7. Does the Choice of AI Tool Influence Student Performance?
    </h2>
</div>


```python
# AI Tools Used vs Performance Category ---
plt.figure(figsize=(10,6))
sns.countplot(x='ai_tools_used', hue='performance_category', data=df)
plt.title('AI Tools Used vs Performance Category')
plt.xlabel('AI Tools Used')
plt.ylabel('Number of Students')
plt.xticks(rotation=45)
plt.legend(title='Performance Category')
plt.show()

# Optional: calculate average scores per purpose/tool
avg_scores = df.groupby('ai_usage_purpose')[['final_score','concept_understanding_score']].mean().reset_index()
print("Average Scores by AI Usage Purpose:\n", avg_scores)

tool_scores = df.groupby('ai_tools_used')[['final_score','concept_understanding_score']].mean().sort_values(by='concept_understanding_score', ascending=False)
print("\nAverage Scores by AI Tool:\n", tool_scores)
```

### *?? Insight:*
?? AI Usage Purpose and Academic Performance

The average final scores across different AI usage purposes are very close, ranging from 56.36 to 57.19, indicating that AI usage purpose alone does not strongly differentiate exam performance.

Homework users achieve the highest final score (57.19) and strong conceptual understanding (5.55), suggesting that AI is most effective when used for guided practice and reinforcement.

Coding also shows relatively high performance (final score = 57.09), indicating that hands-on, problem-solving use of AI supports learning outcomes.

Exam Prep and Notes fall slightly behind in final scores but show good conceptual understanding, particularly Notes, which has the highest understanding score (5.56).

Doubt Solving shows the lowest averages in both final score (56.36) and conceptual understanding (5.32), implying that resolving isolated doubts may not always lead to deeper learning.

?? AI Tools and Learning Outcomes

Performance differences across AI tools are minimal, suggesting that how AI is used matters more than which tool is used.

Gemini users show the highest final score (57.21) and conceptual understanding (5.56), but the margin over other tools is small.

Copilot, ChatGPT+Gemini, and Claude demonstrate nearly identical performance, indicating comparable educational effectiveness.

ChatGPT users show slightly lower conceptual understanding (5.41), though overall performance remains similar.

?? Key Takeaways

Structured use of AI for homework and note-taking is associated with slightly better conceptual understanding and academic performance, while the choice of AI tool has minimal impact compared to usage strategy.

```python
# Study Hours per Day vs Final Score
plt.figure(figsize=(8,5))
sns.scatterplot(x='study_hours_per_day', y='final_score', hue='uses_ai', data=df)
plt.title('Study Hours per Day vs Final Score')
plt.xlabel('Study Hours per Day')
plt.ylabel('Final Score')
plt.legend(title='Uses AI')
plt.show()

# Optional: correlation
corr_hours_score = df[['study_hours_per_day','final_score']].corr().iloc[0,1]
print(f"Correlation between study hours per day and final score: {corr_hours_score:.2f}")
```

### *?? Insight*
The correlation between study hours per day and final score is extremely weak (r = 0.02), indicating that spending more time studying does not automatically lead to better academic performance. This suggests that how students study—including consistency, focus, and learning strategies—is more important than the total number of hours spent studying. Effective study habits and balanced routines likely play a greater role in improving outcomes than simply increasing study time.

```python
# Study Consistency Index vs Improvement Rate ---
plt.figure(figsize=(8,5))
sns.scatterplot(x='study_consistency_index', y='improvement_rate', hue='uses_ai', data=df)
plt.title('Study Consistency vs Improvement Rate')
plt.xlabel('Study Consistency Index')
plt.ylabel('Improvement Rate')
plt.legend(title='Uses AI')
plt.show()

# Optional: correlation
corr_consistency_improve = df[['study_consistency_index','improvement_rate']].corr().iloc[0,1]
print(f"Correlation between study consistency and improvement rate: {corr_consistency_improve:.2f}")
```

### *?? Insight*

The correlation between study consistency and improvement rate is nearly zero (r = ?0.01), indicating no meaningful linear relationship in this dataset. This suggests that consistency alone does not directly translate into short-term improvement; instead, improvement may depend on a combination of factors such as study methods, feedback quality, concept clarity, and support systems (including AI usage). Consistency is still important, but it appears to be effective only when paired with effective learning strategies rather than as an isolated factor.

```python
sns.set(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (10,6)

# Concept Understanding vs Final Score
plt.figure(figsize=(8,5))
sns.scatterplot(x='concept_understanding_score', y='final_score', hue='uses_ai', data=df)
plt.title("Concept Understanding vs Final Score", fontsize=14, fontweight='bold')
plt.xlabel("Concept Understanding Score", fontsize=12)
plt.ylabel("Final Score", fontsize=12)
plt.show()

# Correlation
corr_concept_final = df[['concept_understanding_score','final_score']].corr().iloc[0,1]
print(f"Correlation between Concept Understanding and Final Score: {corr_concept_final:.2f}\n")
```

### *?? Insight*

The correlation between concept understanding and final score is moderate and positive (r = 0.42), indicating that students with stronger conceptual clarity tend to achieve higher academic performance. This suggests that deep understanding of concepts is a significantly stronger predictor of success than factors such as study duration or AI usage alone. Focusing on learning strategies that improve conceptual understanding—such as problem-solving, active recall, and meaningful use of AI—can lead to better academic outcomes.

This suggests that AI dependency alone is not a key driver of learning depth; instead, the way AI is used—such as for guided practice or clarification—matters far more than the level of dependence. Effective learning appears to be shaped primarily by study strategies and engagement rather than AI reliance.
