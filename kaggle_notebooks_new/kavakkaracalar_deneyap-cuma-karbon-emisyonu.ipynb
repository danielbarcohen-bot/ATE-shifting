```python
from sklearn.model_selection import train_test_split
```

```python
import numpy
```

```python
girdi = numpy.array(sadece_avusturalya_sadece_otoban_emisyonu["year"]).reshape(-1, 1)
```

```python
girdi[0:8]
```

```python
etiket = numpy.array(sadece_avusturalya_sadece_otoban_emisyonu["value"]).reshape(-1, 1)
```

```python
etiket[0:8]
```

```python
egitim_girdi,test_girdi,egitim_etiket,test_etiket = train_test_split(girdi,etiket,test_size = 0.10,random_state = 80)
```

```python
len(egitim_girdi)
```

```python
len(test_girdi)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
model = DecisionTreeRegressor()
```

```python
model.fit(egitim_girdi,egitim_etiket)
```

```python
tahmin = model.predict(test_girdi)
```

```python
from sklearn.metrics import mean_squared_error
```
