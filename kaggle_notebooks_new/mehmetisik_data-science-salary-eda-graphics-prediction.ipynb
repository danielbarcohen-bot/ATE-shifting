    <h3 style="color: #0C6A86; margin-top: 0;">📁 Dataset Variables Overview</h3>
    <p>The dataset contains 11 variables defined as follows:</p>
    <ul>
        <li><strong>work_year:</strong> The year the salary was paid.</li>
        <li><strong>experience_level:</strong> Experience level on the job throughout the year, with possible values:
            <ul>
                <li><strong>EN - Entry-level</strong></li>
                <li><strong>MI - Junior Mid-level</strong></li>
                <li><strong>SE - Intermediate Senior-level</strong></li>
                <li><strong>EX - Expert Executive-level / Director</strong></li>
            </ul>
        </li>
        <li><strong>employment_type:</strong> Type of Employment for the Role
            <ul>
                <li><strong>PT - Part-time</strong></li>
                <li><strong>FT - Full-time</strong></li>
                <li><strong>CT - Contract</strong></li>
                <li><strong>FL - Freelance</strong></li>
            </ul>
        </li>
        <li><strong>job_title:</strong> Role worked throughout the year.</li>
        <li><strong>salary:</strong> Total Gross Salary Amount paid.</li>
        <li><strong>salary_currency:</strong> Currency of the salary paid, as ISO 4217 Currency Code.</li>
        <li><strong>salary_in_usd:</strong> Salary in USD (divided by average USD exchange rate for the relevant year, via <a href="https://fxdata.foorilla.com/" target="_blank">fxdata.foorilla.com</a>).</li>
        <li><strong>employee_residence:</strong> Primary country where the employee resided during the working year, as ISO 3166 Country Code.</li>
        <li><strong>remote_ratio:</strong> Total amount of work done remotely, with possible values as follows:
            <ul>
                <li><strong>0</strong> - No Remote Work (less than 20%)</li>
                <li><strong>50</strong> - Partially Remote</li>
                <li><strong>100</strong> - Fully Remote (more than 80%)</li>
            </ul>
        </li>
        <li><strong>company_location:</strong> Country of the employer's main office or contract branch, as ISO 3166 Country Code.</li>
        <li><strong>company_size:</strong> Average number of people employed for the company throughout the year
            <ul>
                <li><strong>S</strong> - Less than 50 Employees (small)</li>
                <li><strong>M</strong> - 50 to 250 Employees (medium)</li>
                <li><strong>L</strong> - More than 250 Employees (large)</li>
            </ul>
        </li>
    </ul>

# <b><div style='padding:15px;background-color:#006600;color:white;border-radius:15px;font-size:100%;text-align: center'>Capturing / Detecting Numeric and Categorical Variables</div></b>

```python
def grab_col_names(dataframe, cat_th=10, car_th=20):
    """

    Returns the names of categorical, numeric and categorical but cardinal variables in the data set.
    Note Categorical variables include categorical variables with numeric appearance.

    Parameters
    ------
        dataframe: dataframe
                Variable names of the dataframe to be taken
        cat_th: int, optional
                class threshold for numeric but categorical variables
        car_th: int, optinal
                class threshold for categorical but cardinal variables

    Returns
    ------
        cat_cols: list
                Categorical variable list
        num_cols: list
                Numeric variable list
        cat_but_car: list
                List of cardinal variables with categorical appearance

    Examples
    ------
        import seaborn as sns
        df = sns.load_dataset("iris")
        print(grab_col_names(df))


    Notes
    ------
        cat_cols + num_cols + cat_but_car = total number of variables
        num_but_cat is inside cat_cols.
        The sum of the 3 return lists equals the total number of variables: cat_cols + num_cols + cat_but_car = number of variables

    """

    # cat_cols, cat_but_car
    cat_cols = [col for col in dataframe.columns if dataframe[col].dtypes == "O"] 

    num_but_cat = [col for col in dataframe.columns if dataframe[col].nunique() < cat_th and
                   dataframe[col].dtypes != "O"]

    cat_but_car = [col for col in dataframe.columns if dataframe[col].nunique() > car_th and
                   dataframe[col].dtypes == "O"]

    cat_cols = cat_cols + num_but_cat

    cat_cols = [col for col in cat_cols if col not in cat_but_car] 

    num_cols = [col for col in dataframe.columns if dataframe[col].dtypes != "O"] 

    num_cols = [col for col in num_cols if col not in num_but_cat] 
    
    print(f"Observations: {dataframe.shape[0]}") 
    print(f"Variables: {dataframe.shape[1]}") 
    print(f'cat_cols: {len(cat_cols)}') 
    print(f'num_cols: {len(num_cols)}') 
    print(f'cat_but_car: {len(cat_but_car)}') 
    print(f'num_but_cat: {len(num_but_cat)}') 


    return cat_cols, num_cols, cat_but_car, num_but_cat
```

```python
def high_correlated_cols(dataframe, plot=False, corr_th=0.70):
    corr = dataframe.corr()
    cor_matrix = corr.abs()
    upper_triangle_matrix = cor_matrix.where(np.triu(np.ones(cor_matrix.shape), k=1).astype(np.bool))
    drop_list = [col for col in upper_triangle_matrix.columns if any(upper_triangle_matrix[col] > corr_th)]
    if plot:
        import seaborn as sns
        import matplotlib.pyplot as plt
        sns.set(rc={'figure.figsize': (12, 8)})
        sns.heatmap(corr, cmap="RdBu", annot=True, fmt=".2f")  # annot=True added here
        plt.show()
    return drop_list
```

```python
# Rare Encoder implementation

def rare_encoder(dataframe, rare_perc):
    temp_df = dataframe.copy()

    rare_columns = [col for col in temp_df.columns if temp_df[col].dtypes == 'O'
                    and (temp_df[col].value_counts() / len(temp_df) < rare_perc).any(axis=None)]

    for var in rare_columns:
        tmp = temp_df[var].value_counts() / len(temp_df)
        rare_labels = tmp[tmp < rare_perc].index
        temp_df[var] = np.where(temp_df[var].isin(rare_labels), 'Rare', temp_df[var])

    return temp_df
```

# <b><div style='padding:15px;background-color:#006600;color:white;border-radius:15px;font-size:100%;text-align: center'>Modelling</div></b> 

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
scaler = MinMaxScaler()
#scaler_s = StandardScaler()
#scaler_r = RobustScaler()

df_fea= df_fea.drop(["salary"], axis=1)
y=df_fea['salary_in_usd']
X=df_fea.drop(['salary_in_usd'],axis=1)
m=X
X = scaler.fit_transform(X)
y= y.values.reshape(-1,1)
y = np.log(y)


def regression_(x, y):
    rf = RandomForestRegressor()
    lg = LGBMRegressor()
    kn = KNeighborsRegressor()
    et = ExtraTreesRegressor() 
    gb = GradientBoostingRegressor()
    dt = DecisionTreeRegressor()
    xgb = XGBRegressor()

    algos = [rf,lg , kn, et, gb, dt,xgb]
    algos_names = ["Random Forest","LGBM", 'KNeighbors', 'ExtraTree', 'GradientBoosting',
                   'DecisionTree','XGB']

    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=.2, random_state=42)

    r_score = []
    mse = []
    mae = []

    result = pd.DataFrame(columns=['R_square', 'MSE', 'MAE'], index=algos_names)

    for algo in algos:
        pred = algo.fit(x_train, y_train).predict(x_test)
        r_score.append(r2_score(y_test, pred))
        mse.append(mean_squared_error(y_test, pred) ** .5)
        mae.append(mean_absolute_error(y_test, pred))

    result.R_square = r_score
    result.MSE = mse
    result.MAE = mae

    return result.sort_values('R_square', ascending=False)
```
