```python
import pandas

df = pandas.read_csv("/kaggle/input/imdb-top-250-tv-shows/IMDB.csv")

```

```python
df['Episodes'] = df['Episodes'].str.replace(' eps', '')

df
```

```python
from sklearn import linear_model
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
import numpy as np

x = df[['Episodes']]
x_p = df['Episodes']

y = df['Rating']
```

```python
import matplotlib.pyplot as plt

plt.scatter(x_p, y)
plt.show()
```

```python
x_1d = [element for sublist in x.values for element in sublist]
x_1d
```

```python
y.values
```

```python
from scipy import stats

slope, intercept, r, p, std_err = stats.linregress(y.values, y.values)

def myfunc(x):
#     print(slope, x, intercept)
    return int(slope) * int(x) + int(intercept)

mymodel = list(map(myfunc, x_1d))

plt.scatter(x_1d, y)
plt.plot(x_1d, mymodel)
plt.show()
```
