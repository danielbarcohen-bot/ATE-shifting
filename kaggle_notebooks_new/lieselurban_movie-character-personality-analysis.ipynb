# Movie Character Personality Analysis

**Welcome to the movie character personality analysis!** Here, we will walk through some common Enneagram/Meyers Briggs (MBTI) personality analysis and visualizations. The dataset used is the Movie Character MBTI Dataset found on Kaggle. First, we import the main libraries including Numpy (math and statistical manipulation), Pandas (data manipulation and analysis), and Matplotlib and Seaborn (data visualization)

```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import matplotlib.pyplot as plt
import seaborn as sns

import os
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))
```

Next, we import the dataset using pandas.read_csv to scan and convert the data into a Pandas Dataframe. I also did some simple cleaning, removing the null and nonexistent values to make the future analysis easier. Finally, I shortened the number of columns to only what was necessary for our next analysis. Take a look at the dataframe snippet below: this is a sneak-peek of the over 11,000 rows of movie personality data!

```python
df = pd.read_csv('../input/movie-character-mbti-dataset/mbti.csv')

test_df = df[(df['mbti'].str.contains("XXXX")) & (df['enneagram'].notnull())]
df = df[(df["mbti"].str.contains("XXXX") == False) & (df['enneagram'].notnull())]
df = df.iloc[:, 0:5]

df.head(8)
```

Next, to get a better idea of each personality type individually, we plot a bar graph of the two personality categories in ascending order. Below is the graph for **Meyers Briggs** Personalities:

```python
df.mbti.value_counts().sort_values().plot(kind = 'bar', figsize = (12, 6), color=['#ABDEE6', '#CBAACB', '#FFFFB5', '#FFCCB6', '#F3B0C3'])
plt.title('Meyers Briggs Personalities')
```

And here we have **Enneagram** personality types in ascending order:

```python
df.enneagram.value_counts().sort_values().plot(kind = 'bar', figsize = (12, 6), color=['#FF968A', '#FFAEA5', '#FFC5BF', '#FFD8BE', '#FFC8A2'])
plt.title('Enneagram Personalities')
```

Using the groupby function in Pandas, we can gather the different personality types together and see the counts for combinations for each Enneagram/MBTI combo (for instance, Enneagram 1w2 and Meyers Briggs ESFJ).

```python
grouped = df.groupby(['enneagram', 'mbti'])['enneagram', 'mbti', 'role'].count()
grouped.head(15)
```

Instead of viewing all 281 grouping in a dataframe, a heatmap allows us to view every pairing frequency in one rectangular frame. A heatmap displays values by color to visualize common frequencies. Here, the darker the blue is, the higher the frequency by count of the corresponding Enneagram and MBTI type:

```python
fig, ax = plt.subplots(figsize=(11,9))

personality_heatmap = pd.pivot_table(df, values='role', index=['mbti'], columns='enneagram', aggfunc= 'count')
percentages = (personality_heatmap / personality_heatmap.sum())
df_heatmap = sns.heatmap(data = percentages, linewidths=0.5, cmap='Blues')
```

Finally, we can use the format from the heatmap to find the maximum values from each row of Enneagram types. 

```python
percentages = (personality_heatmap / personality_heatmap.sum())

print("Highest Personality Combinations by MBTI/Enneagram:\n")
print(personality_heatmap.idxmax())
```
