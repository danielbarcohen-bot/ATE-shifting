↪️How to deal with wrongly recorded values or missing data is one of the important things in data preprocessing. </p> Below, I discuss some common methods for replacing suspicious or missing values:</p> 

✔️Replace with constant value:</p> 
You may choose a constant value (for example, zero or the mean of the entire feature) for all missing values in a feature.</p> 
        
✔️Replace with mean or median:
Use the mean or median of non-missing values in a feature to replace missing values.

✔️Use previous or next value:</p> 
If the data is ordered, you can replace the missing value with its previous or next value.</p> 
        
✔️Interpolation:</p> 
Use value estimation methods such as linear interpolation or spline interpolation to predict the missing value according to its adjacent values.</p> 
        
✔️Using machine learning algorithms:</p> 
You can use machine learning algorithms such as K-Nearest Neighbors (KNN), Decision Trees, or Random Forest to predict the missing value using other features in the data.</p> 
        
✔️random initialization:</p> 
You can replace the missing value with a random value from a given distribution (such as a normal distribution).
Each of these methods has its advantages and disadvantages, and the right choice depends on the data type and its structure. Also, they may require testing and evaluation to find the best method for your particular data.</p>

➡️In the next step, we go to the "sc_w" column, where 180 samples in this column have zero value.</p>
This column is related to the "sc_h" column because both features are about horizontal and vertical screen checking</p>
To find the relationship between these two features, we can use correlation coefficient and scatter plot to measure the strength and direction of this relationship.</p>
✅In the case of "sc_h" and "sc_w", this correlation coefficient indicates that the vertical height (sc_h) and horizontal width (sc_w) of the screen follow each other on average. In other words, if a mobile phone is tall, there is a high probability that its width will also increase, and vice versa.</p>
✅The values in the sc-w column are linearly related to the values in the sc-h column, and there is a 0 value for each sc-h value, you can use a linear method to estimate the appropriate values for the 0 values. One of the common linear methods for estimating missing or zero values is using linear regression. You can build a linear regression model based on the available values of the "sc-h" and "sc-w" columns and then use this model to predict the zero values.

```python
sns.scatterplot(x='sc_h', y='sc_w', data=df)
plt.title('Scatter Plot between sc_h and sc_w')
plt.xlabel('sc_h')
plt.xticks (range (5,20,1))
plt.ylabel('sc_w')
plt.yticks([i for i in range(0, 20)], [i for i in range(0, 20)])
plt.grid()
plt.show()
```

```python
correlation_coefficient = df['sc_h'].corr(df['sc_w'])
print(f'Correlation Coefficient between sc_h and sc_w: {correlation_coefficient}')
```

```python
df1 =df.copy()
```

```python

# The data we have to train the model (available values)
train_data = df1[df1['sc_w'] != 0][['sc_h', 'sc_w']]

# The values we want to predict (zero values)
predict_data = df1[df1['sc_w'] == 0][['sc_h', 'sc_w']]


model = LinearRegression()
model.fit(train_data[['sc_h']], train_data['sc_w'])

# Predict values for zero values
predicted_values = model.predict(predict_data[['sc_h']])

# Actual values for zero values
actual_values = df1[df1['sc_w'] == 0]['sc_w']

# Original data (excluding zero values)
original_data = df1[df1['sc_w'] != 0]

# Replace zero values in the original DataFrame with predicted values
df1.loc[df1['sc_w'] == 0, 'sc_w'] = predicted_values

#  RMSE
rmse = np.sqrt(mean_squared_error(actual_values, predicted_values))
print(f"Root Mean Squared Error (RMSE): {rmse}")

# Create a new DataFrame for the predicted data
predicted_data_df = pd.DataFrame({'sc_h': predict_data['sc_h'], 'sc_w': predicted_values})

# Scatter Plot
plt.figure(figsize=(10, 6))

# Plot the original data
sns.scatterplot(x='sc_h', y='sc_w', data=original_data, label='primary')

# Plot the predicted data
sns.scatterplot(x='sc_h', y='sc_w', data=predicted_data_df, label='predicte', marker='x', color='red')

plt.title('Scatter Plot Among sc_h and sc_w')
plt.xlabel('sc_h')
plt.xticks (range (5,20,1))
plt.ylabel('sc_w')
plt.yticks([i for i in range(0, 20)], [i for i in range(0, 20)])
plt.legend()
plt.show()

```

```python

# Set the style of seaborn
sns.set(style="whitegrid")

# Define the categorical variables
categorical_variables = ['blue','dual_sim', 'four_g', 'three_g', 'touch_screen', 'wifi', 'price_range']

# Color
pastel_colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC99', '#c2c2f0', '#ffb3e6', '#c2f0c2']

# Create subplots
fig, axes = plt.subplots(len(categorical_variables), 2, figsize=(15, 4 * len(categorical_variables)))

# Loop over each row (variable) and plot a countplot and a pie chart
for i, column in enumerate(categorical_variables):
    # Calculate the number of occurrences for each category
    category_counts = df1[column].value_counts()

    # Plot countplot
    ax_countplot = sns.countplot(data=df1, x=column, palette=pastel_colors, ax=axes[i, 0])
    ax_countplot.set_title(f"{column} Number distribution")

    # Add labels to countplot
    for p in ax_countplot.patches:
        ax_countplot.annotate(f'{p.get_height()}', (p.get_x() + p.get_width() / 2., p.get_height()),
                              ha='center', va='baseline', fontsize=12, color='black', xytext=(0, 5),
                              textcoords='offset points')

    # Plot pie chart
    ax_pie = axes[i, 1]
    ax_pie.pie(category_counts, labels=category_counts.index, autopct='%1.1f%%', startangle=90, colors=pastel_colors)
    ax_pie.set_title(f"{column} Percentage distribution")

plt.tight_layout()
plt.show()
```

```python
#Check the ratios
def Decision_Tree_Classifier_model(df_norm, test_ratio):
    # define x and y
    x = df_norm[['battery_power','clock_speed','fc','int_memory','m_dep','mobile_wt','n_cores','pc','px_height','px_width','ram','sc_h','sc_w','talk_time','blue','dual_sim','four_g','three_g','touch_screen','wifi']].values
    y = df_norm['price_range'].values.reshape(-1 ,1)

    # define x and y test & train
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=test_ratio, random_state=1)

    # fit model
    DTC = DecisionTreeClassifier( max_depth = 5,ccp_alpha = 0.007)
    DTC.fit(x_train, y_train)
    y_pred = DTC.predict(x_test)

    # Printing results
    print(colored('*' * 40 + 'confusion_matrix' + '*' * 40, 'red'))
    print('confusion matrix : \n', confusion_matrix(y_test, y_pred))
    print(colored('*' * 40 + 'classification_report' + '*' * 40, 'green')) 
    print('Classification Report :\n', classification_report(y_test, y_pred))
    print(colored('*' * 40 + 'Accuracy' + '*' * 40, 'black'))
    print ("Accuracy :" , metrics.accuracy_score(y_test , y_pred))
    

test_ratios = [0.3, 0.2, 0.1]

for test_ratio in test_ratios:
    print(f"\nTraining Model with {int((1-test_ratio)*100)}% training data and {int(test_ratio*100)}% test data:")
    Decision_Tree_Classifier_model(df_norm, test_ratio)
```
