```python
# Use pandas.read_csv() to create a dataframe named em_df from the dataset at 
# URL: /kaggle/input/global-fossil-co2-emissions-by-country-2002-2022/GCB2022v27_MtCO2_flat.csv

em_df = pd.read_csv('/kaggle/input/global-fossil-co2-emissions-by-country-2002-2022/GCB2022v27_MtCO2_flat.csv')
em_df
```

```python
# Select only four columns of the dataset: Country, Year, Total, and Per Capita

em_df = em_df[['Country', 'Year', 'Total', 'Per Capita']]
em_df
```

## Exercise 1 Use Conditional Index to Filter Data

```python
# Use pandas.DataFrame.loc[] to select countries with positive total emissions

# Replace this line with your code
em_df = em_df.loc[(em_df['Total'] > 0),]

em_df
```

```python
# Use pandas.DataFrame.loc[] to filter out rows of which the value of the Country column is equal to Global
# or International Transport

# Replace this line with your code
country_df = em_df.loc[(em_df['Country'] != 'Global') & (em_df['Country'] != 'International Transport'),]

country_df
```

```python
# Use pandas.DataFrame.loc[] to select rows of which the value of the year column is equal to 2021
# and store the selected rows in country_2021_df

# Replace this line with your code
country_2021_df = country_df.loc[(country_df['Year'] == 2021),]

country_2021_df
```

## Exercise 2 Sort Data By Values

```python
# Use both pandas.DataFrame.loc[] and pandas.DataFrame.sort_values() to find out 
# the Top 5 countries with the highest total Carbon Emission at 2021

# Replace this line with your code
country_2021_df.sort_values(['Total'], ascending=False)
```

## Exercise 3 Create Frequency Distribution

```python
# Use pandas.DataFrame.value_counts() to create a frequency distribution with 5 intervals of 
# Total Carbon Emission across countries at 2021

# The frequency distribution should not be sorted nor be normalized

# Replace this line with your code
country_2021_df['Total'].value_counts(sort=False, normalize=False, bins=5)
```

## [Optional] Exercise 6 Use Groupby Operation to Find Countries with the Fastest Growth in Ecological Footprint of Consumption

```python
# Use pandas.DataFrame.loc[] to create a dataframe named country_period_df 
# for records of Carbon Emission per Capita between 2000 and 2021

# Replace this line with your code
country_period_df = country_df.loc[(country_df['Year'] >= 2000),]

country_period_df
```

```python
# Use pandas.DataFrame.groupby() and pandas.DataFrame.agg() to find the Carbon Emission per Capita
# for each country (group by country) at the start and the end of the time period from 2000 to 2021
# Pass the result to a dataframe named country_group_df

# Hint: use agg({'Per Capita': ['first', 'last']}) to obtain the first and the last rows of the column named total for each group

# Replace this line with your code
country_group_df = country_period_df.groupby(by='Country', as_index=False).agg({'Per Capita': ['first', 'last']})

country_group_df
```

```python
# Calculate the change in the Carbon Emission per Capita from 2000 to 2021 for each country
# Create a new column in country_group_df named change to store the result

# Hint: Use a tuple to represent a multi-level column index. 
# E.g., country_group_df[('Per Capita', 'change')] creates a sub-column named change under the column named Per Capita

# Replace this line with your code
country_group_df[('Per Capita', 'change')] = country_group_df[('Per Capita', 'last')] - country_group_df[('Per Capita', 'first')]

country_group_df
```

```python
# Use pandas.DataFrame.sort_values() to find out the Top 20 countries with the largest growth in 
# the Carbon Emission per Capita from 2000 to 2021

# Hint: Use iloc[:20] to select the first 20 rows in a dataframe

# Replace this line with your code
country_group_df.sort_values([('Per Capita', 'change')], ascending=False).iloc[:20,]
```
