```python
class CompoundInterestCalculator:
    def __init__(self, capital_initial, taux):
        self.capital = capital_initial
        self.taux = taux

    def next_year(self):
        self.capital = self.capital * (1 + self.taux)

    def simulate(self, n):
        for year in range(1, n + 1):
            self.next_year()
            print(f"Year {year}: capital = {self.capital:.2f}")
```

```python
calc = CompoundInterestCalculator(1000, 0.05)
calc.simulate(5)
```
