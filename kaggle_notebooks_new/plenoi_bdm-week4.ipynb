### 15. Linear Regression

```python
sales_by_day = df_clean.groupby('Date').agg(total_qty=('Qty', 'sum'),total_amount=('Amount', 'sum')).reset_index()
sales_by_day['x'] = sales_by_day.index+1
```

```python
import seaborn as sns
sns.lmplot(x='x',y='total_amount',data=sales_by_day,fit_reg=True) 
```

```python
from sklearn import linear_model
regr = linear_model.LinearRegression()
X = sales_by_day['x'].values.reshape(-1,1)
y = sales_by_day['total_amount'].values.reshape(-1,1)
regr.fit(X, y)
print("y = "+str(regr.intercept_)+ " + "+str(regr.coef_[0])+"X")
```

### 16. One-Way Anova 

```python
df_clean['Size'].unique()
```

```python
import scipy.stats as stats
group1 = df_clean[df_clean['Size']=='L']['Amount']
group2 = df_clean[df_clean['Size']=='S']['Amount']
_, lpValue = stats.levene(group1, group2)
fValue, pValue = stats.f_oneway(group1, group2)
print("Levene Test : "+str(lpValue))
print("Anova Test: F = "+ str(fValue) + ", p =" + str(pValue))
```

### 17. ARIMA
ARIMA stands for AutoRegressive Integrated Moving Average. It is a class of models that captures a suite of different standard temporal structures in time series data. Before we apply ARIMA, we need to ensure that the series is stationary, which involves checking if its statistical properties such as mean, variance, and autocorrelation are constant over time.

```python
from statsmodels.tsa.stattools import adfuller

adf_test = adfuller(sales_by_day['total_amount'])
# Output the results
print('ADF Statistic: %f' % adf_test[0])
print('p-value: %f' % adf_test[1])
```

A p-value below 0.05 indicates stationarity, and our data meets this criterion, so we do not need to difference it.

#### Find p, d, q parameters of ARIMA
p from ACF = 1
d from PCF = 2
q from ACF = 2

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
# Lag is the correlation between different time frame
plot_acf(sales_by_day['total_amount'], lags=30)
plot_pacf(sales_by_day['total_amount'], lags=30)
plt.show()
```

#### https://analyticsindiamag.com/ai-mysteries/quick-way-to-find-p-d-and-q-values-for-arima/

```python
from statsmodels.tsa.arima.model import ARIMA

model = ARIMA(sales_by_day['total_amount'], order=(1, 2, 2))
model_fit = model.fit()
```

```python
forecast = model_fit.get_forecast(steps=5)
```

```python
# Create a plot to compare the forecast with the actual test data
plt.figure(figsize=(14,7))
plt.plot(sales_by_day['total_amount'], label='Training Data')
plt.plot(forecast.predicted_mean, label='Forecasted Data', color='green')
plt.title('ARIMA Model Evaluation')
plt.xlabel('Date')
plt.ylabel('Amounts')
plt.legend()
plt.show()
```
