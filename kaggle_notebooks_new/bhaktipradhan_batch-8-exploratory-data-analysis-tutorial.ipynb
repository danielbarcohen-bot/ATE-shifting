# Exploratory Data Analysis

Exploratory Data Analysis (EDA). It is a step in which we need to explore the data set, know the data set before you pass it to next steps like Feature Engineering, Model Building and Model Evaluation using different Machine Learning algorithm.

By defination - Exploratory Data Analysis refers to the critical process of performing initial investigations on data so as to discover patterns, to spot anomalies, to test hypothesis and to check assumptions with the help of summary statistics and graphical representations. 

Exploratory data analysis (EDA) is the process of examining and analyzing data sets in order to summarize their main characteristics. Before any formal modelling or statistical testing is done, it is normally carried out at the start of a data analysis project.

The main aim of exploratory data analysis is to obtain confidence in your data to an extent where you’re ready to engage a machine learning algorithm. By doing this you can get to know whether the selected features are good enough to model, are all the features required, are there any correlations based on which we can either go back to the Data Preprocessing step or move on to modeling.

It is very important for a data scientist to be able to understand the nature of the data without making assumptions.
the goal is to try and answer all questions on the data.

# Basics of Statistics

## Measures of Central Tendency

### Mean
The mean is the arithmetic average for calculating the mean just add up all of the values and divide by the number of observations in your dataset.

```python
print('The Mean of SalePrice is ',data['SalePrice'].mean())
```

The average sale price of a house in our dataset is close to 180,000 with most of the values falling within the 130,000 to 215,000 range.

### Median
The median is the middle value.

It is the value that splits the dataset in half.

To find the median, order your data from smallest to largest, and then find the data point that has an equal amount of values above it and below it.

The method for locating the median varies slightly depending on whether your dataset has an even or odd number of values.

```python
print('The Median of SalePrice is ',data['SalePrice'].median())
```

### Mode
The mode is the value that occurs the most frequently in your data set i.e. has the highest frequency.

On a bar chart, the mode is the highest bar.

If the data have multiple values that are tied for occurring the most frequently, you have a multimodal distribution.

If no value repeats, the data do not have a mode.

```python
print('The Mode of SalePrice is ', data['SalePrice'].mode())
```

### Modality
* A distribution might be unimodal with one prominent peak, bimodal with two prominent peaks, or uniform with no prominent peaks.
* With more than two prominent peaks a distribution is usually said to be multimodal.
* A bimodal distribution might indicate that there are two distinct groups in your data.

## Measures of Spread/Dispersion
Variability or spread is the measure of how close or far the data lie from the center of the distribution. The less variability, the more data are closer to the center. The more variability, the more data are spread further away from it.

Very Rare - Many machine learning algorithms assume that the data follows a normal distribution, or a bell-shaped curve, where most of the values are clustered around the mean. However, in real-world scenarios, data can be non-normal, meaning that it can have different shapes, outliers, skewness, or kurtosis
* bell shaped curve/normal distribution

##### Normal Distribution
* The curve is symmetric around the Mean
* Mean, Median, and Mode are all the same/same

Empirical Rule for Normal Distribution
* 68.27% of data lies within 1 standard deviation of the mean
* 95.45% of data lies within 2 standard deviations of the mean
* 99.73% of data lies within 3 standard deviations of the mean


This rule enables us to check for Outliers

#### Skewness
Skewness is also another measure to check for normality 

Normal Distribution is symmetric, which means its tails on one side are the mirror image of the other side. In the Normal Distribution, Mean, Median and Mode are equal.

But this is not the case with most datasets. We call these types of distributions Skewed Distributions. 

Left Skewed Distribution
    When data points cluster on the right side of the distribution, then the tail would be longer on the left side. This is the property of Left Skewed Distribution. The tail is longer in the negative direction so we also call it Negatively Skewed Distribution. The mean is generally smaller than the median.

Here, Mode > Median > Mean.


Right Skewed Distribution
    When data points cluster on the left side of the distribution, then the tail would be longer on the right side. This is the property of Right Skewed Distribution. Here, the tail is longer in the positive direction so we also call it Positively Skewed Distribution. The mean is generally higher than the median.

Mode < Median < Mean

The shorter the left or right tail, the closer the mean to the median.

Pandas has a function to compute skewness

* If the value is less than -0.5, we consider the distribution to be negatively skewed or left-skewed
* If the value is greater than 0.5, we consider the distribution to be positively skewed or right-skewed
* If the value is between -0.5 and 0.5, we consider the distribution to be approximately symmetric

#### Kurtosis
Kurtosis is a statistical measure used to describe the degree to which scores cluster in the tails or the peak of a frequency distribution.

For the symmetric type of distribution, the Kurtosis value will be close to Zero

Pandas has a function to compute Kurtosis

* If the Kurtosis value is close to Zero, we call such types of distributions as Mesokurtic distribution. Normal Distribution have this.
* If the Kurtosis value is less than zero, we call such types of distributions as Leptokurtic Distribution. Here, Tail will be fatter and will have longer distribution/Peak. This shows there are extreme values present in the data,
* If the Kurtosis value is greater than zero, we call such types of distributions as Platykurtic Distribution. It will have a thinner tail and a shorter distribution.


```python
print('The skewness of SalePrice is ',data['SalePrice'].skew())
print('The Kurtosis of SalePrice is- ',data['SalePrice'].kurt())
```

# Identify and handle Missing Value - ?, NaN, 0, “N/A”, Blank Cell
Why missing values treatment is required? -  
Missing data in the training data set can reduce the power / fit of a model or can lead to a biased model because we have not analysed the behavior and relationship with other variables correctly. It can lead to wrong prediction or classification.



Why my data has missing values? -  
* MCAR (Missing completely at random) - if the probability of a missing data value is independent of any observation/columns in the data set, then the data are said to be missing completely at random (MCAR). There’s no relationship between whether a data point is missing and any values in the data set, missing or observed. For example MCAR is a weighing scale that ran out of batteries.
* MAR (Missing at random) - If the probability of being missing is the same only within groups defined by the observed data, then the data are missing at random (MAR). For example, when a weighing scale placed on a soft surface may produce more missing values than when placed on a hard surface.
* Missing not at random (MNAR). if the probability of a missing data value is unknown. For example, when a weighing scale goes bad and is giving missing data but is not noted.


#### Detetecting missing - 
Standard Missing Values - isNull detects 
Non-Standard Missing Values - isNull doesnt detects example XX, 9999 etc...

 ## Clean Missing Data

### How to Treat Missing Value

#### Connect with Data Source Collection / Extraction source to load it back
At times due to technical issue it could happen a column/ part of data set is not populated as you were not able to connect to that source. 

#### Fix by Dropping the column or row 

Rows and columns could be deleted if the number of missing values are insignificant in number, as that would not impact analysis.

      df.dropna(axis=0, thresh=1, inplace=True)

      df.dropna(axis=1, thresh=1, inplace=True)

But this is not suggested most times as though this is simple, as this method reduces the power of model because it reduces the sample size.

It is always better to keep data than to discard it. 

Sometimes you can drop variables if the data is missing for more than 60% observations but only if that variable is insignificant.

#### Fix by Imputation
Imputation is a method to fill in the missing values with some value.

The value could be
* Mean / Median imputation is one of the most frequently used methods for numerical columns. It consists of replacing the missing data for a given attribute by the mean or median of all known values of that variable.
* Mode/Static imputation is done for categorical columns. It is either word None/Missing/? or most frequent observation in the data set.

Some ways are

* Mark it as missing or with value you feel is apt wrt the column and business domain - For the categorical columns
* Mark it as 0 or with value you feel is apt wrt the column and business domain - For the numerical columns
* Impute with mode value - For the categorical column, you can replace the missing values with mode values i.e. the frequent ones.
* Impute with mean value - For the numerical column, But before replacing with mean value, it is advisable to check that the variable shouldn’t have extreme values .i.e. outliers.
* Impute with median value - For the numerical column, In case you have extreme values such as outliers it is advisable to use the median approach.

Imputation can be Univariate or MultiVariate

```python
# PoolQC has missing % is 99 %. So, there is fill by Missing
data['PoolQC'] = data['PoolQC'].fillna('Missing')

# Others where missing % is above and around 50 % replace with Missing 
data['MiscFeature'] = data['MiscFeature'].fillna('Missing')
data['Alley'] = data['Alley'].fillna('Missing')
data['Fence'] = data['Fence'].fillna('Missing')
data['FireplaceQu'] = data['FireplaceQu'].fillna('Missing')
```

```python
#GarageType, GarageFinish, GarageQual and GarageCond these are replacing with Missing
for col in ['GarageType', 'GarageFinish', 'GarageQual', 'GarageCond']:
    data[col] = data[col].fillna('Missing')
```

```python
#GarageYrBlt, GarageArea and GarageCars these are replacing with zero
for col in ['GarageYrBlt', 'GarageArea', 'GarageCars']:
    data[col] = data[col].fillna(int(0))
```

```python
#BsmtFinType2, BsmtExposure, BsmtFinType1, BsmtCond, BsmtQual these are replacing with Missing
for col in ('BsmtFinType2', 'BsmtExposure', 'BsmtFinType1', 'BsmtCond', 'BsmtQual'):
    data[col] = data[col].fillna('Missing')
```

```python
# Filling 'LotFrontage' according to Neighborhood. At times you fill data using alternative stategies like 
# -- imputing with median but with median of certain sub group
data['LotFrontage'] = data.groupby("Neighborhood")["LotFrontage"].transform(
    lambda x: x.fillna(x.median()))
```

```python
data['MasVnrArea'] = data['MasVnrArea'].fillna(int(0))
data['MasVnrType'] = data['MasVnrType'].fillna('Missing')
```

```python
data['Electrical'] = data['Electrical'].fillna(data['Electrical']).mode()[0]
```

# Outlier 
An outlier is an observation that lies "far away" from other values. 

Let’s take an example, we do customer profiling and find out that the average annual income of customers is  0.8 million dollar. But, there are two customers having annual income of 4 million dollar and 4.2 million dollar. These two customers annual income is much higher than rest of the population. These two observations will be seen as Outliers.


They may be an indicator of data errors or a rare events, and should be investigated carefully


What causes Outliers?
* Data Entry Errors - Human errors such as errors caused during data collection, recording, or entry can cause outliers in data
* Measurement Error - It is the most common source of outliers. This is caused when the measurement instrument used turns out to be faulty
* Experimental Error - Another cause of outliers is experimental error. Time to take recording was incorrect.
* Intentional Outlier - This is commonly found in self-reported measures that involves sensitive data.
* Data Processing Error - Whenever we perform data mining, we extract data from multiple sources. It is possible that some manipulation or extraction errors may lead to outliers in the dataset.
* Natural Outlier - When an outlier is not artificial (due to error), it is a natural outlier


What is the Impact of Outliers on a dataset?


Outliers can drastically change the results of the data analysis and statistical modeling. There are numerous unfavourable impacts of outliers in the data set:

* It increases the error variance and reduces the power of statistical tests
* If the outliers are non-randomly distributed, they can decrease normality
* They can introduce bias

Hence they should be identified and fixed.

Formula to get outlier
* The potential outliers lie outside the range of: [Q1 - (1.5 × IQR), Q3 + (1.5 × IQR)]
* The problematic outliers lie outside of: [Q1 - (3 × IQR), Q3 + (3 × IQR)].

```python
potenQ1 = Q1 - (1.5 * IQR)
potenQ3 = Q3 + (1.5 * IQR)
outlierQ1 = Q1 - (3 * IQR)
outlierQ3 = Q3 + (3 * IQR)
print('Q1 - ', Q1)
print('Q3 - ', Q3)
print('IQR - ', IQR)
print('The potential outliers lie outside the range of: [Q1 - (1.5 × IQR), Q3 + (1.5 × IQR)] are (', potenQ1, potenQ3,')');
print('The problematic outliers lie outside of: [Q1 - (3 × IQR), Q3 + (3 × IQR)] are (', outlierQ1, outlierQ3,')')
```

##### How to identify Outliers
Most commonly used method to detect outliers is visualization. We use various visualization methods, like Box-plot, Histogram, Scatter Plot . Some analysts also various thumb rules to detect outliers. Some of them are:

* Any value, which is beyond the range of -1.5 x IQR to 1.5 x IQR
* Use capping methods. Any value which out of range of 5th and 95th percentile can be considered as outlier
* Data points, three or more standard deviation away from mean are considered outlier
* Outlier detection is merely a special case of the examination of data for influential data points and it also depends on the business understanding

##### Outlier  Treatment
* Drop the outlier value
* Transforming and binning values
* Imputing - Replace the outlier value


# Univariate Analysis and Bivariate/Multivariate Analysis

To bring important aspects of that data we shall explore the data in two ways
* Univariate Analysis - Univariate analysis explores variables (attributes) one by one. Univariate analysis is done on both categorical or numerical variables. 
* Bivariate/Multivariate Analyssis - Bivariate analysis is the simultaneous analysis of two or more variables (attributes). It explores the concept of relationship between two variables, whether there exists an association and the strength of this association, or whether there are differences between two variables and the significance of these differences.

Both these analysis can be done non-graphical or graphical way.

Non graphical exploratory data analysis is the first step when beginning to analyze your data as part of the general data analysis approach.

This preliminary data analysis step focuses on four points
* measures of central tendency, i.e. the mean, the median and mode,
* measures of spread, i.e. variability, variants and standard deviation,
* the shape of the distribution, and
* the existence of outliers.

Non graphical exploratory data analysis may be followed by, or be engaged in concurrently with graphical exploratory data analysis. Graphical EDA involves visual exploratory analysis of the data.

There are different statistical and visualization techniques of investigation for univariate and bivariate analysis

# Final Thoughts
EDA is undoubtedly/unarguably the most crucial step before we go Feature Selection / PCA / Encoding etc and Machine Learning

What are the key steps to perform exploratory data analysis on a new dataset

* Understand the data
The first step of EDA is to understand the data you are working with. This involves checking the data source, the data format, the data size, the data types, and the data quality. You should also look for any missing values, outliers, duplicates, or errors in the data. You can use descriptive statistics, such as mean, median, mode, standard deviation, and range, to get a sense of the distribution and variability of the data. You can also use metadata, such as column names, descriptions, and units, to understand the meaning and context of the data.

* Explore the data
The next step of EDA is to explore the data and look for patterns, trends, relationships, and anomalies. This involves using visualizations, such as charts, graphs, plots, and maps, to display the data in different ways and reveal its characteristics. You can use different types of visualizations depending on the data type and the question you want to answer. For example, you can use histograms, box plots, or density plots to show the distribution of a single variable, or scatter plots, line plots, or heat maps to show the correlation or interaction between two or more variables.



* Transform the data
The third step of EDA is to transform the data and make it ready for further analysis or modeling. This involves applying various operations, such as filtering, sorting, grouping, aggregating, merging, or reshaping, to manipulate the data and extract relevant information. You can also use feature engineering, which is the process of creating new variables or modifying existing ones, to enhance the data and capture more insights. For example, you can create new features based on existing ones, such as extracting the day of the week from a date column, or based on domain knowledge, such as calculating the body mass index from the height and weight columns.


* Test the data
The final step of EDA is to test the data and validate your assumptions and hypotheses. This involves using statistical tests, such as t-tests, ANOVA, chi-square, or regression, to measure the significance and effect of the variables and their relationships. You can also use machine learning techniques, such as clustering, classification, or regression, to build models and evaluate their performance and accuracy. You should also compare the results of your tests and models with your expectations and intuition, and check for any inconsistencies or surprises.
