# Regresi?n lineal m?ltiple

Esto tambi?n se aplica de forma similar a la Regresi?n Lineal M?ltiple. El objetivo sigue siendo ajustar una l?nea que muestre mejor la relaci?n entre una variable independiente y el objetivo. La diferencia es que en la Regresi?n Lineal M?ltiple, tenemos que tratar con al menos 2 caracter?sticas o variables independientes.

Se usar? el conjunto de datos "advertising.csv", que describe para productos el impacto entre su venta y la inversi?n en medios publicitarios.  El conjunto de datos tiene un total de doscientas piezas de datos.

* 3,17.2,45.9,69.3,9.3

```python
# Pr?ctica: Regresi?n lineal m?ltiple. Caso publicidad en medios
# Fernando Bordignon - http://saberesdigitales.unipe.edu.ar

# Carga de librer?as

import pandas
from sklearn.model_selection import train_test_split # Entrenamiento de validaci?n cruzada y divisi?n del conjunto de pruebas
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
 
# Los datos almacenados se cargan en un datafframe
datos = pandas.read_csv("../input/advertisingcsv/Advertising.csv");
datos.columns = ['caso','TV','Radio','Prensa','Ventas']  # se cambian los nombres de las columnas
datos
```

```python
# Se construyen los vectores de caracter?sticas X e Y

X = datos[['TV', 'Radio', 'Prensa']]; # Volver al marco de datos
print (type (X), "", X.shape)         # Devuelve la dimensi?n de X type X
    
Y = datos['Ventas']                   # Volver al tipo y lista de serie
print (type (Y), "", Y.shape)         # Devuelve la dimensi?n de X type X
```

```python
# Se divide el conjunto de entrenamiento y el conjunto de prueba
#  75% para entrenamiento y 25% para prueba

# random_state Asigne un valor al lugar donde necesita establecer random_state. Cuando ejecuta este c?digo varias veces, puede obtener exactamente los mismos resultados y otros pueden ejecutar este c?digo para replicar su proceso. Si no se establece este par?metro, se seleccionar? una semilla al azar y el resultado de la ejecuci?n variar? en consecuencia. Aunque se puede ajustar random_state, el modelo que se desempe?a bien en el conjunto de entrenamiento despu?s del ajuste puede no funcionar bien en el conjunto de entrenamiento desconocido, por lo que generalmente se selecciona aleatoriamente un valor de random_state como par?metro.
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,random_state=1);
print(X_train.shape,"  ",X_test.shape,"  ",Y_train.shape,"  ",Y_test.shape);
```

```python
# Ejecuci?n de la regresi?n lineal

lrg = LinearRegression();

model = lrg.fit (X_train, Y_train); # entrenamiento
print(model)

print (lrg.intercept_); #intercepci?n de salida
coef = zip (['TV', 'Radio', 'Peri?dico'], lrg.coef_) # Caracter?sticas y coeficientes correspondientes 

print("\n")
for T in coef :
         print (T); # coeficiente de salida
```

```python
# Pron?stico

y_pred = lrg.predict(X_test);

print (y_pred); # valor de prueba de salida
```

```python
# Medidas de evaluaci?n 

# Las medidas de evaluaci?n son de precisi?n
# Aqu? hay tres medidas de evaluaci?n de uso com?n para la regresi?n lineal. ·
 #   Error absoluto medio (MAE);
 # · Error cuadr?tico medio (MSE);
 # · Error cuadr?tico medio (RMSE). Aqu? RMES se utiliza para evaluaci?n y medici?n.
    
print("Pron?stico");

print(type(y_pred),type(Y_test));
print(len (y_pred), len (Y_test)); 

sum_mean = 0;
for i in range (len(y_pred)): 
    sum_mean+=(y_pred[i]-Y_test.values[i])**2;
    sum_erro = np.sqrt (sum_mean / len (y_pred)); 

print ("Error cuadr?tico medio", sum_erro);
```

```python
# Se dibuja una curva ROC

plt.figure()
plt.plot (range (len (y_pred)), y_pred, 'b', label = "predicci?n")
plt.plot(range(len(Y_test)),Y_test,'r',label="test")
plt.xlabel("n?mero de ventas")
plt.ylabel("ventas (miles de d?lares)")
plt.legend ()

plt.show()
```
