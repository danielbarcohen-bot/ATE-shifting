# Regresión lineal múltiple

Esto también se aplica de forma similar a la Regresión Lineal Múltiple. El objetivo sigue siendo ajustar una línea que muestre mejor la relación entre una variable independiente y el objetivo. La diferencia es que en la Regresión Lineal Múltiple, tenemos que tratar con al menos 2 características o variables independientes.

Se usará el conjunto de datos "advertising.csv", que describe para productos el impacto entre su venta y la inversión en medios publicitarios.  El conjunto de datos tiene un total de doscientas piezas de datos.

* #,TV,Radio,Newspaper,Sales
* 1,230.1,37.8,69.2,22.1
* 2,44.5,39.3,45.1,10.4
* 3,17.2,45.9,69.3,9.3
* .....

```python
# Práctica: Regresión lineal múltiple. Caso publicidad en medios
# Fernando Bordignon - http://saberesdigitales.unipe.edu.ar

# Carga de librerías

import pandas
from sklearn.model_selection import train_test_split # Entrenamiento de validación cruzada y división del conjunto de pruebas
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
 
# Los datos almacenados se cargan en un datafframe
datos = pandas.read_csv("../input/advertisingcsv/Advertising.csv");
datos.columns = ['caso','TV','Radio','Prensa','Ventas']  # se cambian los nombres de las columnas
datos
```

```python
# Se construyen los vectores de características X e Y

X = datos[['TV', 'Radio', 'Prensa']]; # Volver al marco de datos
print (type (X), "", X.shape)         # Devuelve la dimensión de X type X
    
Y = datos['Ventas']                   # Volver al tipo y lista de serie
print (type (Y), "", Y.shape)         # Devuelve la dimensión de X type X
```

```python
# Se divide el conjunto de entrenamiento y el conjunto de prueba
#  75% para entrenamiento y 25% para prueba

# random_state Asigne un valor al lugar donde necesita establecer random_state. Cuando ejecuta este código varias veces, puede obtener exactamente los mismos resultados y otros pueden ejecutar este código para replicar su proceso. Si no se establece este parámetro, se seleccionará una semilla al azar y el resultado de la ejecución variará en consecuencia. Aunque se puede ajustar random_state, el modelo que se desempeña bien en el conjunto de entrenamiento después del ajuste puede no funcionar bien en el conjunto de entrenamiento desconocido, por lo que generalmente se selecciona aleatoriamente un valor de random_state como parámetro.
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,random_state=1);
print(X_train.shape,"  ",X_test.shape,"  ",Y_train.shape,"  ",Y_test.shape);
```

```python
# Ejecución de la regresión lineal

lrg = LinearRegression();

model = lrg.fit (X_train, Y_train); # entrenamiento
print(model)

print (lrg.intercept_); #intercepción de salida
coef = zip (['TV', 'Radio', 'Periódico'], lrg.coef_) # Características y coeficientes correspondientes 

print("\n")
for T in coef :
         print (T); # coeficiente de salida
```

```python
# Pronóstico

y_pred = lrg.predict(X_test);

print (y_pred); # valor de prueba de salida
```

```python
# Medidas de evaluación 

# Las medidas de evaluación son de precisión
# Aquí hay tres medidas de evaluación de uso común para la regresión lineal. ·
 #   Error absoluto medio (MAE);
 # · Error cuadrático medio (MSE);
 # · Error cuadrático medio (RMSE). Aquí RMES se utiliza para evaluación y medición.
    
print("Pronóstico");

print(type(y_pred),type(Y_test));
print(len (y_pred), len (Y_test)); 

sum_mean = 0;
for i in range (len(y_pred)): 
    sum_mean+=(y_pred[i]-Y_test.values[i])**2;
    sum_erro = np.sqrt (sum_mean / len (y_pred)); 

print ("Error cuadrático medio", sum_erro);
```

```python
# Se dibuja una curva ROC

plt.figure()
plt.plot (range (len (y_pred)), y_pred, 'b', label = "predicción")
plt.plot(range(len(Y_test)),Y_test,'r',label="test")
plt.xlabel("número de ventas")
plt.ylabel("ventas (miles de dólares)")
plt.legend ()

plt.show()
```
