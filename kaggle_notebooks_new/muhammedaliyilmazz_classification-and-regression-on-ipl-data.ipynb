##  **Dataset Overview – IPL Match Data (Ball-by-Ball Level)**

###  General Description:

This dataset provides a comprehensive and granular look into Indian Premier League (IPL) cricket matches. Each row represents a **single ball delivery**, allowing deep-level analysis of match dynamics, player performance, and tactical decisions.

* **Granularity**: Ball-by-ball data
* **Scope**: All seasons, all venues, all teams
* **Use Cases**: Outcome prediction, player performance analytics, strategy modeling, and more

# Classification Modeling (Who Will Win the Match?)

## 1. Import required libraries

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

## 2. Use cleaned data

```python
df_model = df.copy()
```

## 3. Drop rows with missing match winner

```python
df_model = df_model[df_model['match_won_by'].notna()]
```

## 4. Select basic features for modeling

```python
features = ['toss_winner', 'toss_decision', 'venue', 'season', 'batting_team', 'bowling_team']
df_model = df_model[features + ['match_won_by']]
```

## 5. Encode categorical variables

```python
le = LabelEncoder()
for col in df_model.columns:
    if df_model[col].dtype == 'object':
        df_model[col] = le.fit_transform(df_model[col])
```

## 6. Define X and y

```python
X = df_model.drop('match_won_by', axis=1)
y = df_model['match_won_by']
```

## 7. Train-test split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=537, stratify=y)
```

## 8. Feature scaling

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## 9. Initialize models

```python
models = {
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=537),
    'XGBoost': XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=537)
}
```

## 10. Train, predict, and evaluate

```python
results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    print(f"Model: {name}")
    print("Accuracy:", acc)
    print("Classification Report:\n", classification_report(y_test, y_pred))

    results.append({'Model': name, 'Accuracy': acc})
```

## 11. Summary table

```python
results_df = pd.DataFrame(results)
results_df.sort_values(by='Accuracy', ascending=False)
```

## 12. Accuracy bar plot

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Accuracy', y='Model', data=results_df.sort_values(by='Accuracy', ascending=True))
plt.title('Model Accuracy Comparison')
plt.xlabel('Accuracy')
plt.ylabel('Model')
plt.tight_layout()
plt.show()
```

# Regression Modeling

## 1. Import libraries

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

## 2. Use cleaned data

```python
df_reg = df.copy()
```

## 3. Remove rows with missing or zero runs

```python
df_reg = df_reg[df_reg['team_runs'].notna()]
df_reg = df_reg[df_reg['team_runs'] > 0]
```

## 4. Select features and target

```python
features = ['venue', 'season', 'batting_team', 'bowling_team', 'toss_decision', 'overs']
df_reg = df_reg[features + ['team_runs']]
```

## 5. Encode categorical features

```python
le = LabelEncoder()
for col in df_reg.columns:
    if df_reg[col].dtype == 'object':
        df_reg[col] = le.fit_transform(df_reg[col])
```

## 6. Define X and y

```python
X = df_reg.drop('team_runs', axis=1)
y = df_reg['team_runs']
```

## 7. Train-test split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=537)
```

## 8. Feature scaling

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## 9. Initialize models

```python
models = {
    'Random Forest': RandomForestRegressor(n_estimators=100, random_state=537),
    'XGBoost': XGBRegressor(n_estimators=100, random_state=537)
}
```

## 10. Train, predict, and evaluate

```python
results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    print(f"Model: {name}")
    print("MSE:", round(mse, 2))
    print("RMSE:", round(np.sqrt(mse), 2))
    print("R² Score:", round(r2, 4))
    print('-'*40)

    results.append({'Model': name, 'R2_Score': r2, 'RMSE': np.sqrt(mse)})
```

## 11. Results summary table

```python
results_df = pd.DataFrame(results).sort_values(by='R2_Score', ascending=False)
results_df
```

## 12. Visualization

```python
plt.figure(figsize=(8,5))
sns.barplot(x='R2_Score', y='Model', data=results_df)
plt.title('Model R² Score Comparison')
plt.xlabel('R² Score')
plt.ylabel('Model')
plt.tight_layout()
plt.show()
```
