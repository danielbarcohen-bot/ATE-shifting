# 5 Common Business Statistical Measures

Business leaders often use a variety of statistical measures to gain insights from their data and inform their decisions. Here are five common ones:

1. Mean (Average): The mean is the sum of all values divided by the number of values. It gives a general idea of the magnitude of the data.

2. Median: The median is the middle value in a set of values. If the number of observations is even, it is the average of the two middle numbers. Unlike the mean, it is not skewed by outliers, so it can sometimes give a more accurate picture of the data.

3. Standard Deviation: The standard deviation measures the amount of variation or dispersion in a set of values. A low standard deviation means that the values tend to be close to the mean, while a high standard deviation means that the values are spread out over a wider range.

4. Correlation: Correlation measures the relationship between two variables. The correlation coefficient ranges from -1 to 1. A value of 1 means a strong positive correlation (as one variable increases, so does the other), while a value of -1 means a strong negative correlation (as one variable increases, the other decreases). A value of 0 means there is no correlation.

5. Percentage Change/Growth Rate: Often used to track the growth of a metric over time, this measure indicates the rate at which a quantity increases or decreases for a given set of returns. This is often used in sales, revenue, and user growth analyses.

````python
import pandas as pd

# Retrieve and print the highest 'SalePrice' column
max_value = df['SalePrice'].max()
print(f'Max Sale Price: ${max_value:,.2f}')

# Retrieve and printe the lowest 'SalePrice' column
min_value = df['SalePrice'].min()
print(f'Min Sale Price: ${min_value:,.2f}')
# Calculate and print the mean of the 'SalePrice' column
mean = df['SalePrice'].mean()
print(f"Mean Sale Price: ${mean:,.2f}")

# Calculate and print the median of the 'SalePrice' column
median = df['SalePrice'].median()
print(f"Median Sale Price: ${median:,.2f}")

# Calculate and print the standard deviation of the 'SalePrice' column
std_dev = df['SalePrice'].std()
print(f"Standard Deviation of Sale Price: ${std_dev:,.2f}")

# Calculate and print the correlation between 'SalePrice' and 'Overall Qual'
correlation = df['SalePrice'].corr(df['Overall Qual'])
print(f"Correlation between Sale Price and Overall Quality: {correlation:,.2f}")

# Calculate and print the percentage change in 'SalePrice' over the dataset
df['SalePriceChange'] = df['SalePrice'].pct_change()
percent_change = df['SalePriceChange'].mean() * 100
print(f"Average Percent Change in Sale Price: {percent_change:,.2f}%")
````

# Common types of Data Distributions

Normal Distribution (Gaussian Distribution): This is perhaps the most important and most commonly observed distribution. A normal distribution has a bell-shaped curve, with the mean, median, and mode all located at the center of the distribution. The data is symmetric about the mean. Examples can include heights or IQ scores in a sufficiently large population.

Uniform Distribution: In a uniform distribution, all outcomes are equally likely, and the distribution is symmetric. An example can be rolling a fair six-sided dice, where each side (1-6) has an equal chance of turning up.

Binomial Distribution: This distribution is used when there are exactly two mutually exclusive outcomes of a trial, often termed as "success" and "failure". The binomial distribution gives the probabilities of observing various numbers of the "success" outcome in a set number of trials. An example could be the number of heads when flipping a coin a certain number of times.

Poisson Distribution: This distribution applies when an event can occur a number of times within a given length of time or space. The key characteristic is that the events are independent, meaning the occurrence of one event does not affect the occurrence of other events. For instance, the number of calls received by a call center per hour can follow a Poisson distribution.

Exponential Distribution: The exponential distribution describes the time between events in a Poisson point process, i.e., a process in which events occur continuously and independently at a constant average rate. It has a parameter λ called rate parameter, and its equation is derived from the Poisson distribution.

Chi-Square Distribution: This distribution is used in hypothesis testing and is related to standard normal distribution. It is used when dealing with variance and standard deviation, and particularly when comparing observed data with theoretical models.

## Binomial Distribution 
A coin flip can be represented by a binomial distribution with parameters $$n=1$$ (since each trial consists of flipping the coin once) and $$p=0.5$$ (since the probability of getting a head or a tail is 0.5 if the coin is fair). Here we simulate 10000 coin flips and plot the distribution.

````python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Number of trials and probability of success
n = 1
p = 0.5

# Generate 10000 random numbers from a binomial distribution
data = np.random.binomial(n, p, size=10000)

# Map the data to the outcomes of a coin flip
outcomes = ['heads' if i else 'tails' for i in data]

# Create a histogram of the outcomes with seaborn
sns.histplot(outcomes, discrete=True)

# Add title and labels
plt.title('Binomial Distribution of Coin Flips')
plt.xlabel('Outcome')
plt.ylabel('Frequency')

# Show the plot
plt.show()
````

## Poisson Distribution
Scenario: Use a Poisson distribution to model the number of calls received by a call center per hour, assuming the calls come in at a constant average rate and each call is independent of the others.

Let's say the call center receives an average of 10 calls per hour. Here's how you can generate a Poisson distribution for this situation:

````python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Average number of calls per hour
lambda_val = 10

# Generate 10000 random numbers from a Poisson distribution
data = np.random.poisson(lambda_val, size=10000)

# Create a histogram of the data with seaborn
sns.histplot(data, bins=30, color='skyblue', kde=True)

# Add title and labels
plt.title('Poisson Distribution of Call Center Calls')
plt.xlabel('Number of Calls')
plt.ylabel('Frequency')

# Show the plot
plt.show()
````

## Exponential Distribution
If the events in a Poisson process occur with a known constant mean $$λ = 10$$, the time between events is exponentially distributed. In this context, the rate (λ) represents the average number of events per interval.

If we consider the same call center scenario, with an average of 10 calls per hour (i.e., λ=10), the time between calls would be exponentially distributed as $$  \text{mean} = 1/λ$$.

Here's how you can generate an exponential distribution for this situation:

````python
# Average number of calls per hour
lambda_val = 10

# Generate 10000 random numbers from an exponential distribution
data = np.random.exponential(1/lambda_val, size=10000)

# Create a histogram of the data with seaborn
sns.histplot(data, bins=30, color='skyblue', kde=True)

# Add title and labels
plt.title('Exponential Distribution of Time Between Calls')
plt.xlabel('Time Between Calls (Hours)')
plt.ylabel('Frequency')

# Show the plot
plt.show()
````

## Chi-square distribution
The Chi-square distribution is often used in hypothesis testing and is related to the standard normal distribution. It is used mainly in the case of estimations and to test the goodness of fit, independence of two attributes, and homogeneity. The distribution is denoted with a parameter called degrees of freedom.

Let's generate a Chi-square distribution with 4 degrees of freedom:

````python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Degrees of freedom
df = 4

# Generate 10000 random numbers from a chi-square distribution
data = np.random.chisquare(df, size=10000)

# Create a histogram of the data with seaborn
sns.histplot(data, bins=30, color='skyblue', kde=True)

# Add title and labels
plt.title('Chi-square Distribution')
plt.xlabel('Values')
plt.ylabel('Frequency')

# Show the plot
plt.show()
````

## Does this data have skew?

Skewness is a measure of the asymmetry of the probability distribution of a real-valued random variable about its mean. It quantifies the extent and direction of skew (departure from horizontal symmetry).

1. **Positive Skew:** When the tail on the right side (larger end) of the distribution is longer or fatter, the function is said to have positive skewness. In this scenario, the mean and median will be greater than the mode.

2. **Negative Skew:** When the tail on the left side (smaller end) of the distribution is longer or fatter, the function is said to have negative skewness. In this case, the mean and median will be less than the mode.

3. **No Skew:** If the distribution is symmetrical, meaning the left and right sides of the distribution are mirror images of each other, then the distribution has no skew. For a perfectly symmetrical distribution, the mean, median, and mode are equal.

Understanding the skewness of data is important in data analysis as it can influence the choice of statistical methods used to analyze the data. For example, many statistical techniques assume that the data follows a symmetric (often normal) distribution. If the data is skewed, it might need to be transformed to make it more symmetric, or a non-parametric method might be more appropriate.
