# **Read Dataset**

```python
sdf = spark.read.csv('/kaggle/input/adult-income', header=True,inferSchema=True)
```

```python
sdf.show()
```

```python
sdf.dtypes
```

# **One Hot Encoding**

```python
from pyspark.ml.feature import OneHotEncoder
```

```python
from pyspark.ml.feature import StringIndexer
```

```python
sdf.columns
```

```python
si = StringIndexer(inputCols=['Job',
 'Qualification',
 'MaritalStatus',
 'Gender'], outputCols = ['IJob',
 'IQualification',
 'IMaritalStatus',
 'IGender'])
```

```python
si_fitted = si.fit(sdf)
```

```python
sdf = si_fitted.transform(sdf)
```

```python
sdf.show()
```

```python
ohe = OneHotEncoder(inputCols= ['IsAdult',
 'IJob',
 'IQualification',
 'IMaritalStatus',
 'IGender'], outputCols=['DIsAdult',
 'DJob',
 'DQualification',
 'DMaritalStatus',
 'DGender'], dropLast=False)
```

```python
ohe_fitted = ohe.fit(sdf)
```

```python
sdf = ohe_fitted.transform(sdf)
```

```python
sdf.show()
```
