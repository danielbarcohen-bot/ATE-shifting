```python
import seaborn
import statsmodels.api as sm
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
```

```python
data = seaborn.load_dataset("iris")

print(data)
```

```python
scat1 = seaborn.regplot(x="sepal_length", y="petal_length", data=data)
plt.xlabel('Chi?u d?i ??i hoa')
plt.ylabel('Chi?u d?i c?nh hoa')
plt.title('Scatterplot cho m?i li?n h? gi?a ??i hoa v? c?nh hoa')
```

```python
print ("m? h?nh h?i quy OLS cho m?i li?n h? gi?a ??i hoa v? c?nh hoa Iris")
reg1 = smf.ols('petal_length ~ sepal_length', data=data).fit()
print (reg1.summary())
```

# Bi?n gi?i th?ch l? ph?n lo?i

```python
import numpy
import pandas
import statsmodels.api as sm
import seaborn
import statsmodels.formula.api as smf 
import matplotlib.pyplot as plt
```

```python
pandas.set_option('display.float_format', lambda x:'%.2f'%x)
```

```python
data = pandas.read_csv('/kaggle/input/nesarc-huit/nesarc.csv', low_memory=False)
```

```python
data['IDNUM'] =pandas.to_numeric(data['IDNUM'], errors='coerce')
data['TAB12MDX'] = pandas.to_numeric(data['TAB12MDX'], errors='coerce')
data['MAJORDEPLIFE'] = pandas.to_numeric(data['MAJORDEPLIFE'], errors='coerce')
data['NDSymptoms'] = pandas.to_numeric(data['NDSymptoms'], errors='coerce')
data['SOCPDLIFE'] = pandas.to_numeric(data['SOCPDLIFE'], errors='coerce')
data['S3AQ3C1'] = pandas.to_numeric(data['S3AQ3C1'], errors='coerce')
data['AGE'] =pandas.to_numeric(data['AGE'], errors='coerce')
data['SEX'] = pandas.to_numeric(data['SEX'], errors='coerce')
data['S3AQ3B1'] = pandas.to_numeric(data['S3AQ3B1'], errors='coerce')
```

```python
sub1=data[(data['AGE']<=25) & (data['CHECK321']==1) & (data['S3AQ3B1']==1)]
```

```python
reg1 = smf.ols('NDSymptoms ~ MAJORDEPLIFE', data=sub1).fit()
print (reg1.summary())
```

```python
# NDSymptoms = 2.18 + 1.36 * MAJORDEPLIFE
```

```python
sub1 = sub1[['NDSymptoms', 'MAJORDEPLIFE']].dropna()
```

```python
# group means & sd
print ("Trung b?nh")
ds1 = sub1.groupby('MAJORDEPLIFE').mean()
print (ds1)
print ("?? l?ch chu?n")
ds2 = sub1.groupby('MAJORDEPLIFE').std()
print (ds2)
```

```python
seaborn.catplot(x="MAJORDEPLIFE", y="NDSymptoms", data=sub1, kind="bar", errorbar=None)
plt.xlabel('Tr?m c?m')
plt.ylabel('Trung b?nh s? l??ng tri?u ch?ng nghi?n Nicotine')
```
