```python
import os

import numpy
import pandas
import seaborn

from scipy.stats import zscore

from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score
```

```python
train_data = pandas.read_csv("/kaggle/input/playground-series-s3e16/train.csv", index_col="id")
test_data = pandas.read_csv("/kaggle/input/playground-series-s3e16/test.csv", index_col="id")

train_labels = train_data.pop("Age")
```

```python
sex_mapping = {sex: index for index, sex in enumerate(train_data["Sex"].unique())}
```

```python
train_data["Sex"] = train_data["Sex"].replace(sex_mapping)
test_data["Sex"] = test_data["Sex"].replace(sex_mapping)
```

```python
seaborn.boxplot(data=train_data[train_data.columns[-4:]])
```

```python
def remove_outliners(data_frame):
    for column in data_frame.columns:
        if column == "Sex":
            continue
        data_frame[column] = data_frame[column].clip(data_frame[column].mean() - 3 * data_frame[column].std(), data_frame[column].mean() + 3 * data_frame[column].std())
    return data_frame   
```

```python
train_data = remove_outliners(train_data)
test_data = remove_outliners(test_data)
```

```python
seaborn.boxplot(data=train_data[train_data.columns[-4:]])
```

```python
scaler = StandardScaler()

train_data = scaler.fit_transform(train_data)
test_data = scaler.transform(test_data)
```

```python
# estimator_range = [50, 100, 150, 200, 250]
# scores = []

# for number in estimator_range:
#     random_forest = RandomForestRegressor(n_estimators=number, random_state=42)
    
#     score = cross_val_score(random_forest, train_data, train_labels, cv=5, scoring='neg_mean_squared_error')
    
#     scores.append(numpy.mean(-score))

# best_estimators = estimator_range[numpy.argmin(scores)]
```

```python
random_forest = RandomForestRegressor(n_estimators=100, random_state=42)
random_forest.fit(train_data, train_labels)

prediction = random_forest.predict(test_data)
```

```python
result = pandas.read_csv("/kaggle/input/playground-series-s3e16/sample_submission.csv")

result["Age"] = prediction
```

```python
result.to_csv(os.path.join("/kaggle/working/", "result_with_CV.csv"), index=False)
```
