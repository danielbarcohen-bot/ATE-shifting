Let's check if our male and female height variables are normally distributed. To do this, we use the Shapiro-Wilk test.

```python
train.columns
```

```python
numeric_features=['Male Height in Cm', 'Female Height in Cm',
       'mean_height']
```

```python
# Shapiro-Wilk Test
from scipy.stats import shapiro
# normality test
for feature in numeric_features:
	stat, p = shapiro(train[feature])
	print('Statistics=%.3f, p=%.3f' % (stat, p))
	# interpret
	alpha = 0.05
	if p > alpha:
		print('Sample looks Gaussian (fail to reject H0)')
	else:
		print('Sample does not look Gaussian (reject H0)')
```

Excellent! Because variables are normally distributed we can use parametric tests. To determine the degree of correlation between variables, we use the construction of the Pearson correlation matrix.

```python
correlation = train.corr()
print(correlation['Female Height in Cm'].sort_values(ascending = False),'\n')
```

```python
k= 3
cols = correlation.nlargest(k,'Female Height in Cm')['Female Height in Cm'].index
print(cols)
cm = np.corrcoef(train[cols].values.T)
f , ax = plt.subplots(figsize = (14,12))
sns.heatmap(cm, vmax=.8, linewidths=0.01,square=True,annot=True,cmap='viridis',
            linecolor="white",xticklabels = cols.values ,annot_kws = {'size':12},yticklabels = cols.values)
```

With a high degree of probability, according to the average height of men, it is possible to predict the height of women in the country.

Let's transform the dataset to use machine learning algorithms.

```python
train.drop(['Rank','Country Name','mean_height'],axis=1,inplace=True)
```

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, KFold, GridSearchCV
from sklearn.linear_model import LinearRegression, ElasticNet,Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error, mean_squared_log_error
import xgboost as xgb
from sklearn.ensemble import RandomForestRegressor
```

```python
y=train['Female Height in Cm']
X=train.drop(['Female Height in Cm'],axis=1)
```

```python
X_train, X_test, Y_train, Y_test=train_test_split(X,y,test_size=0.1,random_state=42)
```

```python
models = [RandomForestRegressor(), LinearRegression(),ElasticNet(), KNeighborsRegressor(),xgb.XGBRegressor(),Ridge()]
scores = dict()

for m in models:
    m.fit(X_train, Y_train)
    y_pred = m.predict(X_test)

    print(f'model: {str(m)}')
    print(f'RMSE: {mean_squared_error(Y_test,y_pred)}')
    print(f'MAE: {mean_absolute_error(Y_test, y_pred)}')
    print('-'*30, '\n')
```

Ridge regression is ideal for solving our problem. Let's try to find the optimal hyperparameters of the model using the Optuna library.

```python
!pip install optuna
```

```python
import optuna 
from optuna import Trial, visualization
from optuna.samplers import TPESampler
from xgboost import XGBClassifier
from sklearn.model_selection import cross_val_score
```

```python
def objective(trial):
    alpha = trial.suggest_float("alpha", 0, 10)
    intercept = trial.suggest_categorical("fit_intercept", [True, False])
    tol = trial.suggest_float("tol", 0.001, 0.01, log=True)
    solver = trial.suggest_categorical("solver", ["auto", "svd","cholesky", "lsqr", "saga", "sag"])

    ## Create Model
    regressor = Ridge(alpha=alpha,fit_intercept=intercept,tol=tol,solver=solver)
    ## Fit Model
    regressor.fit(X_train, Y_train)

    return mean_squared_error(Y_test, regressor.predict(X_test))
```

```python
study2 = optuna.create_study(study_name="RidgeRegression")
study2.optimize(objective, n_trials=50)
```

```python
from optuna import visualization
print('Best trial: score {},\nparams {}'.format(study2.best_trial.value,study2.best_trial.params))
hist = study2.trials_dataframe()
hist.head()
```
