```python
# Diabetes utilizando a base Play Tennis

import pandas as pd
import numpy as np

df = pd.read_csv('/kaggle/input/review/diabetes.csv')

df

def entropia(feature):
    
    probs = feature.value_counts(normalize=True)
    impureza = -1 * np.sum(np.log2(probs) * probs)
        
    return(round(impureza, 3))

print('Entropia atributo classe:', entropia(df['Glucose']))

df['Glucose'].value_counts()

def ganho_informacao(df, target, descriptive_feature):
    
    print('Atributo classe:', target)
    print('Atributo avaliado:', descriptive_feature)
           
    target_entropy = entropia(df[target])

    entropy_list = list()
    weight_list = list()
    
    for nivel in df[descriptive_feature].unique():
        df_feature_level = df[df[descriptive_feature] == nivel]
        entropy_level = entropia(df_feature_level[target])
        entropy_list.append(round(entropy_level, 3))
        weight_level = len(df_feature_level) / len(df)
        weight_list.append(round(weight_level, 3))

    print('Entropia das ramifica??es:', entropy_list)    

    entropia_atual = np.sum(np.array(entropy_list) * np.array(weight_list))
    print('Impureza atual:', entropia_atual)
    
    information_gain = target_entropy - entropia_atual
    print('Ganho de informa??o:', information_gain)
    
    print('====================')

    return(information_gain)

for feature in df.drop(columns='Glucose').columns:
    feature_info_gain = ganho_informacao(df, 'Glucose', feature)
```
