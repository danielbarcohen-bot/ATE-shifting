    CV 1.057 V6 LB 0.666 quantile(0.7)

Overall situation is strange CV of median is better than for Ridge+TE, but LB is much worse. 


```python
%%time
Y = df_de_train.iloc[:,5:].values
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
verbose = 100
for predict_method in ['mean', 'median', 'percentile60', 'percentile70']:
    print( predict_method )
    mrrmse_list = [] ; r2_list = []
    for fold, val_cell_type in enumerate(validation_cell_types):
        mask_va = (df_de_train.cell_type == val_cell_type) & ~df_de_train.sm_name.isin(train_sm_names)
        mask_tr = ~mask_va # 485 or 487 training rows
        print('fold:', fold, val_cell_type, mask_va.sum() , mask_tr.sum(), mask_va.sum()+ mask_tr.sum() ,  )

        X_train_categorical = df_de_train[['cell_type','sm_name']][mask_tr]
        X_valid_categorical = df_de_train[['cell_type','sm_name']][mask_va]    
        Y_train = Y[mask_tr,:]
        Y_valid = Y[mask_va,:]    
        print( 'X_valid_categorical.shape', X_valid_categorical.shape , 'X_train_categorical.shape',X_train_categorical.shape, 'Y_train.shape', Y_train.shape, 'Y_valid.shape', Y_valid.shape  )

        if predict_method == 'mean':
            vec1 = np.mean( Y_train, axis = 0)
        elif predict_method == 'median':
            vec1 = np.median( Y_train, axis = 0)
        elif predict_method == 'percentile60':
            vec1 = np.percentile( Y_train,60, axis = 0)
        elif predict_method == 'percentile70':
            vec1 = np.percentile( Y_train,70, axis = 0)

        Y_pred_valid = np.zeros_like( Y_valid) + vec1[ np.newaxis,:] # broadcasting vec1 to Y_pred sizes 
        mrrmse = np.sqrt(np.square(Y_pred_valid - Y_valid).mean(axis=1)).mean()
        r2 = r2_score( Y_valid , Y_pred_valid ) 
        print(f"# Fold {fold}: mse: {mrrmse:5.3f} r2: {r2:5.3f} val='{val_cell_type}'")    
        mrrmse_list.append( mrrmse )
        r2_list.append( r2 )
        print()

    mrrmse = np.array(mrrmse_list).mean()    
    r2 = np.array(r2_list).mean()    
    if verbose >= 10:
        print(f"# {predict_method} Overall mrrmse: {mrrmse:5.3f}, r2: {r2:5.3f}" )
    print()
    print('--------------------------------------------------------')
    
```

# Optuna param optimization for Catboost

First optimization (1000+ iterations) lead to very bad conclusion - we get CV 1.106 LB 0.664 - exactly as for prediction by means. 
So the model have not learned anything but just predicted by mean. 

May be we need to optimize other params also but no much optimism 

    First optimization we tried: 
    'n_estimators':  trial.suggest_categorical('n_estimators', [1,2,5,10,20, 50, 100, 200, 500]) , # 
    'depth' : trial.suggest_int('depth', 1,16) ,
    'learning_rate': trial.suggest_float('learning_rate',0.01, 0.1) ,  


```python
%%time
import optuna

import catboost
from catboost import CatBoostRegressor, Pool
from sklearn.multioutput import MultiOutputRegressor
categorical_features = ['cell_type','sm_name']

import time

dict_trial_count = {0:0}
def objective(trial):
    verbose = 1
    dict_trial_count[0] += 1
    t0 = time.time()

    params = {
        'random_state' :  0,
        'verbose' :  0, 
        'cat_features': categorical_features, 
        'n_estimators':  trial.suggest_categorical('n_estimators', [1,2,5,10,20, 50, 100, 200, 500]) , # 
        'depth' : trial.suggest_int('depth', 1,16) ,
        'learning_rate': trial.suggest_float('learning_rate',0.01, 0.1) ,  
        'loss_function': 'RMSE'
    }
    
    model = MultiOutputRegressor( CatBoostRegressor(**params)) 
    
    mrrmse_list = []
    for fold, val_cell_type in enumerate(validation_cell_types):
        mask_va = (df_de_train.cell_type == val_cell_type) & ~df_de_train.sm_name.isin(train_sm_names)
        mask_tr = ~mask_va # 485 or 487 training rows
        if verbose >= 10:
            print(  'fold:', fold, val_cell_type, mask_va.sum() , mask_tr.sum(), mask_va.sum()+ mask_tr.sum() )


        Yr_train = reducer.fit_transform(Y[mask_tr,:])
        X_train_categorical = df_de_train[['cell_type','sm_name']][mask_tr]
        X_valid_categorical = df_de_train[['cell_type','sm_name']][mask_va]    
        #Yr_valid = reducer.transform(Y[mask_va,:])
        if verbose >= 1000:        print('Yr_train.shape',Yr_train.shape, 'X_valid_categorical.shape', X_valid_categorical.shape , 'X_train_categorical.shape',X_train_categorical.shape )

        model.fit(X_train_categorical, Yr_train )
        Yr_valid_pred = model.predict(X_valid_categorical) # 
        Y_valid = reducer.inverse_transform( Yr_valid_pred )

        mrrmse = np.sqrt(np.square(Y[mask_va,:] - Y_valid).mean(axis=1)).mean()
        r2 = r2_score( Y[mask_va,:] ,Y_valid  )
        if verbose >= 10:
            print(f"# Fold {fold}, {val_cell_type}: mse: {mrrmse:5.3f} r2: {r2:5.3f} val='{val_cell_type}'")
        mrrmse_list.append(mrrmse)    

        if verbose >= 10: print()

    
    mrrmse = np.array(mrrmse_list).mean()    
    if verbose >= 1:
        print(params)
        print(f"# Overall {mrrmse:5.3f}", 'trial_count = ', dict_trial_count[0] ,'timing: %.1f'%(time.time()-t0) )
    
    return mrrmse

```

```python
%%time

n_trials = 0
if n_trials > 0:
    print('Optimization starts.  n_trials = ', n_trials)
    dict_trial_count[0] = 0


    optuna.logging.set_verbosity(optuna.logging.WARNING)
    import warnings
    warnings.filterwarnings("ignore", category=FutureWarning)

    study = optuna.create_study()
    study.optimize(objective, n_trials=n_trials)

    # Output for best found params: 
    print(); print('Best params:')
    print(study.best_params)  # E.g. {'x': 2.002108042}
```
