```python
import pandas as pandas
import matplotlib.pyplot as plt
import numpy as np

sales_frame = pandas.read_csv("../input/stores-area-and-sales-data/Stores.csv")
sales_frame
```

```python
## sorting by Store_Sales
sales_frame.sort_values('Store_Sales', ascending=False)
```

```python
#selecting top 10 store sales and displaying data
top_10_sales = sales_frame.sort_values('Store_Sales', ascending=False).head(10)
top_10_sales
labels = top_10_sales["Store ID "]

Store_Sales = top_10_sales["Store_Sales"]

x = np.arange(len(labels)) 

width = 0.45

fig, axs = plt.subplots()


rects1 = axs.bar(x, Store_Sales, width, label='Sales of Stores')



axs.set_ylabel('Store Sales')
axs.set_title('Sales of top 10 Stores')
axs.set_xticks(x, labels)
axs.legend()
axs.bar_label(rects1, padding=3)


fig.set_size_inches(10, 6)
fig.tight_layout()

plt.show()
```

## Daily Customer Count Visualization against Sales

```python

Yfac = Store_Sales
Xfac = top_10_sales["Daily_Customer_Count"]
#print(type(Xfac.iloc[0]))
plt.figure(figsize=(15,10))
plt.scatter(Xfac,Yfac,s=15*5)
plt.xlabel('Count of Customers')
plt.ylabel('Sales in $')
plt.title('Customers & Sales of top 10 Stores')
plt.xlim(200, 2000)
plt.annotate
for xItem, yItem in zip(Xfac,Yfac):
    
    plt.text(float(xItem)-90.0, float(yItem)+450.0, '({}, {})'.format((xItem), (yItem)),fontsize=12,
             verticalalignment='center')

plt.grid(True)


```

```python
Yfac = sales_frame["Store_Sales"]
Xfac = sales_frame["Daily_Customer_Count"]
#print(type(Xfac.iloc[0]))
plt.figure(figsize=(15,10))
plt.scatter(Xfac,Yfac,s=15)
plt.xlabel('Count of Customers')
plt.ylabel('Sales in $')
plt.title('Customers & Sales of All Stores')
plt.xlim(200, 2000)
plt.yscale("linear")
plt.grid(True)
```

## Store Sales and Items Available Graph

```python
Yfac = sales_frame["Store_Sales"]
Xfac = sales_frame["Items_Available"]
#print(type(Xfac.iloc[0]))
plt.figure(figsize=(15,10))
plt.scatter(Xfac,Yfac,s=15,c='r')
plt.xlabel('Items Available')
plt.ylabel('Sales in $')
plt.title('Items Available & Sales of All Stores')

plt.grid(True)
```

## Store Sales and Store Area Graph

```python
Yfac = sales_frame["Store_Sales"]
Xfac = sales_frame["Store_Area"]
#print(type(Xfac.iloc[0]))
plt.figure(figsize=(15,10))
plt.scatter(Xfac,Yfac,s=15,c='g')
plt.xlabel('Store Area')
plt.ylabel('Sales in $')
plt.title('Store Area & Sales of All Stores')

plt.grid(True)
```

```python
## adding spend per customer by store
sales_frame["spend_per_customer"]=sales_frame["Store_Sales"]/sales_frame["Daily_Customer_Count"]
```

## spend per customer graph

```python
sales_frame_2 = sales_frame.sort_values('spend_per_customer', ascending=False).head(10)
Yfac = sales_frame_2["spend_per_customer"]
Xfac = sales_frame_2["Store ID "]
#print(type(Xfac.iloc[0]))
plt.figure(figsize=(15,7))
plt.scatter(Xfac,Yfac,s=35,c='#f57105')
plt.xlabel('Store Id')
plt.ylabel('Sales per customer in $')
plt.title('Sales per customer by store')
for xItem, yItem in zip(Xfac,Yfac):
    
    plt.text(float(xItem)-9.0, float(yItem)+100.0, '({}, {})'.format((xItem), round(yItem,2)),fontsize=12,
             verticalalignment='center')

plt.grid(True)
```
