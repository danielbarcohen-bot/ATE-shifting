```python
X = data.drop('price', axis=1)
y = data['price'] 
```

```python
#X = pd.get_dummies(X, drop_first=True)
from sklearn.preprocessing import LabelEncoder
lb=LabelEncoder()
X['mainroad']=lb.fit_transform(X['mainroad'])
X['guestroom']=lb.fit_transform(X['guestroom'])
X['basement']=lb.fit_transform(X['basement'])
X['hotwaterheating']=lb.fit_transform(X['hotwaterheating'])
X['airconditioning']=lb.fit_transform(X['airconditioning'])
X['prefarea']=lb.fit_transform(X['prefarea'])
X['furnishingstatus']=lb.fit_transform(X['furnishingstatus'])
```

```python
X.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
y.head()
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
sc=model.score(X_test,y_test)
```

```python
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
acc=accuracy_score(y_test, y_pred)
print("Mean Squared Error:", mse)
print("R-squared:", r2)
print("Accuracy:", acc)
```
