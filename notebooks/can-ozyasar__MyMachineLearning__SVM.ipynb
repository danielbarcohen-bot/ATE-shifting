Bu dersimizde Machine Learning modellerinden SVM - Support Vector Machines Modelini Python'da ?eker hastal??? veri setini ?rne?iyle uygulamal? olarak ??renece?iz.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split


# Outcome = 1 Diabet/?eker Hastas?
# Outcome = 0 Sa?l?kl?
data = pd.read_csv("svm_dataset.csv") # diyabet dataseti ile ?ok benzer bir dataset ( de?ii?tirilmi? hali )
data.head()

```

```python


seker_hastalari = data[data.Outcome == 1] # outcome hastal?k durumunu beirtiyor 1 ise hasta,0 ise sa?l?kl? anlam?na geliyor
saglikli_insanlar = data[data.Outcome == 0]


#  gloucose'a bakarak ?rnek bir ?izim yapal?m:
# Program?m?z?n sonunda makine ??renme modelimiz sadece glikoza de?il t?m di?er verilere bakarak bir tahmin yapacakt?r..
plt.scatter(saglikli_insanlar.Age, saglikli_insanlar.Glucose, color="green", label="sa?l?kl?", alpha = 0.4)
plt.scatter(seker_hastalari.Age, seker_hastalari.Glucose, color="red", label="diabet hastas?", alpha = 0.4)

plt.xlabel("Age")
plt.ylabel("Glucose")
plt.legend()
plt.show()


# kan bas?nc?-ya?
plt.scatter(seker_hastalari.Age, seker_hastalari.BloodPressure, color="red", label="diabet hastas?", alpha = 0.4)
plt.scatter(saglikli_insanlar.Age, saglikli_insanlar.BloodPressure, color="green", label="sa?l?kl?", alpha = 0.4)

plt.xlabel("Age")
plt.ylabel("BloodPressure")
plt.legend()
plt.show()


# ins?lin-ya?
plt.scatter(seker_hastalari.Age, seker_hastalari.Insulin, color="red", label="diabet hastas?", alpha = 0.4)
plt.scatter(saglikli_insanlar.Age, saglikli_insanlar.Insulin, color="green", label="sa?l?kl?", alpha = 0.4)

plt.xlabel("Age")
plt.ylabel("Insulin")
plt.legend()
plt.show()
```

```python


# x ve y eksenlerini belirleyelim
y = data.Outcome.values
x_orjinal = data.drop(["Outcome"],axis=1)   
# Outcome s?tununu(dependent variable) ??kar?p sadece independent variables b?rak?yoruz
# ??kn? algoritmam?z x de?erleri i?erisinde grupland?rma yapacak..


# normalization yap?yoruz - x_orjinal i?erisindeki de?erleri sadece 0 ve 1 aras?nda olacak ?ekilde hepsini g?ncelliyoruz
# E?er bu ?ekilde normalization yapmazsak y?ksek rakamlar k???k rakamlar? ezer ve algoritmay? yan?ltabilir!
x = (x_orjinal - np.min(x_orjinal))/(np.max(x_orjinal)-np.min(x_orjinal))

# ?nce
print("Normalization ?ncesi ham veriler:\n")
print(x_orjinal.head())


# sonra 
print("\n\n\nNormalization sonras? yapay zekaya e?itim i?in verece?imiz veriler:\n")
print(x.head())
    

```

```python

# train datam?z ile test datam?z? ay?r?yoruz
# train datam?z sistemin sa?l?kl? insan ile hasta insan? ay?rt etmesini ??renmek i?in kullan?lacak
# test datam?z ise bakal?m makine ??renme modelimiz do?ru bir ?ekilde hasta ve sa?l?kl? insanlar? ay?rt edebiliyor mu diye 
# test etmek i?in kullan?lacak...% 20 lik k?sm?n? kullanacak
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.2, random_state=3)



```

```python
# SVM modelimizi olu?turuyoruz.
```

```python
svm_model = SVC(random_state = 3)
svm_model.fit(x_train,y_train)
 
```
