```python
import os
import pandas
pandas.set_option('display.show_dimensions', False) 
# ?????????? ???????????? ??????? ?????????, ?.?. ???????? ???? ???????? ???? ???????

def dir_read(path): # ?????? ??????? ?????????? ? ????? ?? ????? ??????
    num = 1
    for file in os.listdir(path):
        full_path = os.path.join(path, file)  # ?????? ???? ? ?????/?????
        if os.path.isfile(full_path):         # ????????? ?????? ????
            print(num, ": ", file)
            num += 1
    print('_'*80)

def file_read(path): # ?????? ? ????? ??? ???????
    pandas.set_option('display.max_columns', None)
    pandas.set_option('display.max_rows', None)
    data = pandas.read_csv(os.path.join(path,'Lr3.csv'), sep='\t')
    # ?????? ????? ? ????????? ??????? ? ?????????
    return data

def file_read_dict(path): # ?????? ? ????? ??? ???????
    data = file_read(path)
    data_dict = dict()

    for i in data:
        data_dict[i] = list(data[i])

    for k, v in data_dict.items():
        print('{}: {}'.format(k,v))

    print('_'*80)

def file_sort(path, obj, asc): # ??????????
    print('?????????? ?? {}'.format(obj))
    data = file_read(path)
    data = data.sort_values(by=obj, ascending=asc)

    print(data)
    print('_' * 80)


def file_filter(path, obj, val): # ??????????
    print('?????????? ?? {}'.format(obj))
    data = file_read(path)
    if type(obj) == int:
        data = data[data[obj] == int(val)]
    else:
        data = data[data[obj].str.contains(val)]

    print(data)
    return data

path = os.getcwd() # ??????? ??????????


dir_read(path) # ????? ?????? ??????
print('????????? ?????')
print(file_read(path))# ?????? ? ????? ??? ???????
print('????????? ?????')
file_read_dict(path) # ?????? ? ????? ??? ???????

file_sort(path, 'car_pass', True) # ????? ??????????? ?? ???????? car_pass
                                                
file_sort(path, '?', False) # ????? ??????????? ?? ???????? ?

data_filtered = file_filter(path, 'enable_time', '12:10') # ?????????? enable_time ?? 12:10

data_filtered.to_csv(os.path.join(path,'Lr3_fl.csv'), index = False, sep='\t', encoding='utf-8', header=True)
```
