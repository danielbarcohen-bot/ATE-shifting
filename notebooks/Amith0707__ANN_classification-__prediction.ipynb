```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
## LOad the ANN trained model,scaler model,ohe

model=load_model('model.keras')

##Load the encoder and scaler

with open('ohe.pkl','rb') as file:
    label_encoder_geo=pickle.load(file)
with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender=pickle.load(file)
with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
input_data = {
    'CreditScore': 850,
    'Geography': 'Germany',  # Choose a geography likely associated with churn
    'Gender': 'Female',
    'Age': 50,
    'Tenure': 7,
    'Balance': 200000,  # High balance can be linked to customer retention
    'NumOfProducts': 4,
    'HasCrCard': 1,
    'IsActiveMember': 0,  # If customer is not active, they may be more likely to churn
    'EstimatedSalary': 150000
}
```

```python
## One-Hot Encode 'Geography'

geo_encoded=label_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df=pd.DataFrame(geo_encoded,columns=label_encoder_geo.get_feature_names_out(['Geography']))
```

```python
geo_encoded_df
input_df=pd.DataFrame([input_data])
```

```python
## Combine one-hot encoded columns with input data
input_data=pd.concat([input_df.reset_index(drop=True),geo_encoded_df],axis=1)
input_data
```

```python
## Encode categorical Variables
input_df['Gender']=label_encoder_gender.transform(input_df['Gender'])
input_df
```

```python
## Concatinating one hot encoded and gender one

input_df=pd.concat([input_df.drop("Geography",axis=1),geo_encoded_df],axis=1)
```

```python
## Scaling the data
input_scaled=scaler.transform(input_df)
input_scaled
```

```python
## Prediction time
prediction=model.predict(input_scaled)
prediction
```

```python
prediction_proba=prediction[0][0]
if prediction_proba>0.5:
    print("Customer is likely to churn")

else:
    print("THe customer is not likely to churn")
```
