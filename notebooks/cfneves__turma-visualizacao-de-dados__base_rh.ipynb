## ?? C?LULA 10 ? M?dulo datetime: o b?sico

**Por que isso importa?**  
A coluna `Data_Admissao` veio como **texto** (`object`).  
O Python n?o sabe que `'13/08/2024'` ? uma data ? pra ele ? s? uma string.  
Enquanto for texto, voc? **n?o consegue** calcular diferen?as, filtrar por per?odo, extrair m?s ou ano.

**Dois comandos essenciais:**

| Comando | Significado | Exemplo |
|---|---|---|
| `strptime` | string ? parse ? time (l? um texto e vira data) | `'13/08/2024'` ? `datetime(2024, 8, 13)` |
| `strftime` | string ? format ? time (pega uma data e formata como texto) | `datetime(2024, 8, 13)` ? `'Agosto de 2024'` |

**C?digos de formato de data:**
| C?digo | Significado | Exemplo |
|---|---|---|
| `%d` | Dia com zero ? esquerda | 08 |
| `%m` | M?s com zero ? esquerda | 08 |
| `%Y` | Ano com 4 d?gitos | 2024 |
| `%B` | Nome do m?s por extenso | August |

````python
from datetime import datetime, timedelta

# ?? strptime: texto ? objeto data ?????????????????????????????????????????
# Pegamos o primeiro valor da coluna Data_Admissao como exemplo
data_texto = '13/08/2024'

# datetime.strptime(texto, formato)
# O formato descreve COMO a string est? escrita:
# '%d/%m/%Y' = dia/m?s/ano
data_obj = datetime.strptime(data_texto, '%d/%m/%Y')

print(f'String original  : "{data_texto}"')
print(f'Objeto datetime  : {data_obj}')
print(f'Tipo             : {type(data_obj)}')
print()

# ?? strftime: objeto data ? texto formatado ???????????????????????????????
# Agora podemos formatar a data do jeito que quisermos
formatado = data_obj.strftime('%d de %B de %Y')  # 13 de August de 2024
print(f'Formatado        : {formatado}')

# Extraindo partes da data
print(f'Ano              : {data_obj.year}')
print(f'M?s              : {data_obj.month}')
print(f'Dia              : {data_obj.day}')
````

**O que ? `timedelta`?**  
? o resultado da subtra??o entre duas datas ? a diferen?a de tempo.  
Se voc? subtrair uma data de outra, recebe um `timedelta` com a diferen?a em dias.

````python
# ?? timedelta: diferen?a entre duas datas ????????????????????????????????
# datetime.today() ? data e hora de agora
hoje = datetime.today()

# Subtrair duas datas retorna um objeto timedelta
diferenca = hoje - data_obj

print(f'Data de admiss?o : {data_texto}')
print(f'Hoje             : {hoje.strftime("%d/%m/%Y")}')
print(f'Diferen?a        : {diferenca.days} dias')
print(f'Em anos          : {diferenca.days / 365.25:.1f} anos')

# Por que 365.25 e n?o 365?
# Todo 4 anos existe um ano bissexto com 366 dias.
# 365.25 ? a m?dia correta: (365 ? 3 + 366) / 4 = 365.25
````

## ?? C?LULA 11 ? Convertendo a coluna inteira com pd.to_datetime

A c?lula anterior mostrou como converter **uma** string em data.  
Mas nosso DataFrame tem 1.000 linhas.  

`pd.to_datetime()` converte a coluna **inteira de uma vez** ? sem loop, muito mais eficiente.

````python
# Antes da convers?o:
print(f'Tipo ANTES : {df["Data_Admissao"].dtype}')
print(f'Valor antes: "{df["Data_Admissao"].iloc[0]}"')
# .iloc[0] ? acessa a primeira linha pelo ?ndice (iloc = integer location)

print()

# Converter a coluna inteira
# Trabalhamos em uma c?pia para n?o alterar o df original
df_limpo = df.copy()
# .copy() ? cria uma c?pia independente
# Sem isso, mudan?as em df_limpo afetariam o df original

df_limpo['Data_Admissao'] = pd.to_datetime(
    df_limpo['Data_Admissao'],
    format='%d/%m/%Y',  # formato da string de entrada
    errors='coerce'     # datas inv?lidas viram NaT (Not a Time) em vez de erro
)

print(f'Tipo DEPOIS: {df_limpo["Data_Admissao"].dtype}')
print(f'Valor depois: {df_limpo["Data_Admissao"].iloc[0]}')
print(f'NaT gerados : {df_limpo["Data_Admissao"].isna().sum()}')
````

## ?? C?LULA 12 ? Extraindo informa??es da data com o acessor `.dt`

Depois de converter para datetime, o pandas libera o **acessor `.dt`**  
que d? acesso a propriedades da data:

| C?digo | O que retorna | Exemplo |
|---|---|---|
| `.dt.year` | Ano | 2024 |
| `.dt.month` | M?s (n?mero) | 8 |
| `.dt.quarter` | Trimestre | 3 |
| `.dt.day_name()` | Nome do dia da semana | "Tuesday" |

````python
# Criar colunas derivadas de data
df_limpo['Ano_Admissao'] = df_limpo['Data_Admissao'].dt.year
df_limpo['Mes_Admissao'] = df_limpo['Data_Admissao'].dt.month
df_limpo['Trimestre']    = df_limpo['Data_Admissao'].dt.quarter

# Calcular tempo de casa em anos
# pd.Timestamp.today() ? data atual no formato do pandas
hoje_ts = pd.Timestamp.today()

df_limpo['Tempo_Casa_Anos'] = (
    (hoje_ts - df_limpo['Data_Admissao']).dt.days / 365.25
).round(1)
# Subtra??o de datas ? timedelta
# .dt.days ? extrai a diferen?a em dias
# / 365.25 ? converte dias em anos
# .round(1) ? arredonda para 1 casa decimal

# Ver o resultado
cols = ['Nome', 'Data_Admissao', 'Ano_Admissao', 'Mes_Admissao', 'Trimestre', 'Tempo_Casa_Anos']
df_limpo[cols].head(6)
````

## ?? C?LULA 13 ? Criando sua primeira fun??o

**O que ? uma fun??o?**  
? um bloco de c?digo com nome pr?prio que voc? pode reutilizar.  
Em vez de copiar o mesmo c?digo 10 vezes, voc? define uma vez e chama quando precisar.

```python
def nome_da_funcao(parametro):
    # o que a fun??o faz
    return resultado
```

- `def` ? palavra-chave que define uma fun??o
- `parametro` ? valor de entrada que a fun??o recebe
- `return` ? valor de sa?da que a fun??o entrega
- A indenta??o (4 espa?os) define o que est? dentro da fun??o

````python
# Fun??o que classifica o tempo de casa em uma categoria de texto
# Por que criar isso? Categorias s?o mais f?ceis de usar em gr?ficos
# e filtros no Power BI do que n?meros decimais.

def classificar_tempo(anos):
    """
    Recebe: anos (n?mero) ? o tempo de casa do funcion?rio
    Retorna: uma categoria em texto
    """
    # pd.isna() verifica se o valor ? nulo (NaN ou NaT)
    if pd.isna(anos):
        return 'Sem data'
    elif anos < 1:
        return 'Menos de 1 ano'
    elif anos < 3:
        return '1 a 3 anos'
    elif anos < 5:
        return '3 a 5 anos'
    else:
        return 'Mais de 5 anos'


# Testar a fun??o com valores individuais antes de aplicar no DataFrame
print('Testando a fun??o:')
print(f'  0.5 anos ? "{classificar_tempo(0.5)}"')
print(f'  2.1 anos ? "{classificar_tempo(2.1)}"')
print(f'  4.8 anos ? "{classificar_tempo(4.8)}"')
print(f'  9.0 anos ? "{classificar_tempo(9.0)}"')
````

````python
# .apply(fun??o) ? aplica a fun??o em cada linha da coluna
# ? como fazer um loop em cada c?lula, mas de forma mais eficiente

df_limpo['Faixa_Tempo_Casa'] = df_limpo['Tempo_Casa_Anos'].apply(classificar_tempo)

print('Distribui??o por faixa de tempo de casa:')
print(df_limpo['Faixa_Tempo_Casa'].value_counts())
````

## ?? C?LULA 14 ? Filtrando por per?odo de datas

Agora que a data ? um tipo real, podemos fazer filtros por per?odo.  
`pd.DateOffset(years=1)` ? um deslocamento de exatamente 1 ano no calend?rio.

````python
hoje = pd.Timestamp.today()

# Definir os pontos de corte
um_ano_atras    = hoje - pd.DateOffset(years=1)
cinco_anos_atras = hoje - pd.DateOffset(years=5)

# Filtrar por per?odo
# >= significa 'maior ou igual a' ? admitidos depois de 1 ano atr?s
novos     = df_limpo[df_limpo['Data_Admissao'] >= um_ano_atras]
veteranos = df_limpo[df_limpo['Data_Admissao'] <  cinco_anos_atras]

print(f'Admitidos nos ?ltimos 12 meses : {len(novos)} funcion?rios')
print(f'Veteranos (5+ anos de casa)    : {len(veteranos)} funcion?rios')

print()

# Comparativo de sal?rio: novos vs veteranos
print('Sal?rio m?dio:')
print(f'  Novos     : R$ {novos["Salario"].mean():,.2f}')
print(f'  Veteranos : R$ {veteranos["Salario"].mean():,.2f}')
````

````python
# Sal?rio m?dio por departamento
# .mean()             ? calcula a m?dia de cada grupo
# .round(2)           ? arredonda para 2 casas
# .sort_values(...)   ? ordena do maior para o menor
# ascending=False     ? decrescente (False = do maior para o menor)

print('Sal?rio m?dio por departamento:')
print(
    df_limpo.groupby('Departamento')['Salario']
    .mean()
    .round(2)
    .sort_values(ascending=False)
)
````

````python
# Tempo m?dio de casa por departamento
media_depto = (
    df_limpo
    .groupby('Departamento')['Tempo_Casa_Anos']
    .mean()
    .round(1)
    .sort_values(ascending=False)
    .rename('Media_Anos')   # .rename() ? renomeia a coluna resultante
    .reset_index()          # .reset_index() ? converte o ?ndice em coluna comum
)

print('Tempo m?dio de casa por departamento:')
print(media_depto.to_string(index=False))

# .idxmax() ? retorna o NOME do ?ndice com maior valor
maior = media_depto.set_index('Departamento')['Media_Anos'].idxmax()
menor = media_depto.set_index('Departamento')['Media_Anos'].idxmin()
print(f'\n? Maior reten??o: {maior}')
print(f'? Menor reten??o: {menor}')
````
