### Apply - ????????? ??????? ? ??????? ??? ??????

```python
data = pd.read_csv( 'train.csv', delimiter = ',' )
data.head(20)
```

```python
def ageGroup( row ):
    """
    ??????? ??????? ????????? ???????? ? ??????
    """
    
    # ?????????, ??? ???????? ???????? ?? ????? NaN
    if not pd.isnull( row['Age'] ):
        if row['Age'] <= 18:
            return 'Child'

        if row['Age'] >= 65:
            return 'Retiree'

        return 'Young'
    
    # ???? ???????? ???????? NaN, ?? ?????????? Undef
    return 'Undef'
```

```python
# ???????? ??????? ageGroup ? DataFrame ? ??????? ????????? ? ????????? ??????? ageGroup

data['ageGroup'] = data.apply( ageGroup, axis = 1 )
data.head(10)
```

### ??????????? DataFrame

Merge - ?????? JOIN ? SQL

```python
# ??????? ??? DataFrame

df1 = pd.DataFrame( {
    'key1': ['one', 'two', 'three', 'only1'],
    'value': [1, 2, 3, 4]
} )
```

```python
df1
```

```python
df2 = pd.DataFrame( {
    'key2': ['one', 'two', 'three', 'only2'],
    'value': [11, 12, 13, 14]
} )
```

```python
df2
```

```python
# ????????? ??? ???????? ??????, ??????? ???? ? df1
# ???? ????? ????????? ????????, ?? ????? left_on = ['key1', ...] ? right_on = ['key2', ...]

df1.merge( df2, how = 'left', left_on = 'key1', right_on = 'key2' )
```

```python
# ????????? ??? ???????? ??????, ??????? ???? ? df2

df1.merge( df2, how = 'right', left_on = 'key1', right_on = 'key2' )
```

```python
# ????????? ??? ???????? ?????? (???????????)

df1.merge( df2, how = 'outer', left_on = 'key1', right_on = 'key2' )
```

```python
# ????????? ?????? ????? ???????? ??????

df1.merge( df2, how = 'inner', left_on = 'key1', right_on = 'key2' )
```
