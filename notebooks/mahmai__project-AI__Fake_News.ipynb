```python
import re
```

```python
def wordopt(text):
    # بنحول النص لاحرف صغيره
    text = text.lower()

    # بنمسح اللينكات
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    #html بنمسح تاجات ال
    text = re.sub(r'<.*?>', '', text)

    # بنمسح علامات الترقيم
    text = re.sub(r'[^\w\s]', '', text)

    # بنمسح الارقام
    text = re.sub(r'\d', '', text)

    # بنمسح اي سطر جديد
    text = re.sub(r'\n', '', text)

    return text
```

```python
news['text'] = news['text'].apply(wordopt) # بنطبق الداله علشان يتم تعديل العمود تيكست ف نيوز مباشرة ليحتوي علي النصوص النظيفة بعد تطبيق الداله
```

```python
def output_label(n):                                                       # داله بتحول النتيجه الرقيمه لنصوص
    if n == 0:                                                             # 0 يعني الخبر فيك
        return "It's a Fake News"                                          # 1 يعني الخبر صحيح
    elif n == 1:
        return "It's a Real News"
```

```python
def manual_testing(news):
    testing_news = {"text": [news]}
    new_def_test = pd.DataFrame(testing_news)                                       # داله بتعمل اختبار يدوي علي خبر معين صحيح او فيك
    new_def_test["text"] = new_def_test["text"].apply(wordopt)                      # اول حاجه الداله بتاخد النص اللي عاوزه تختبره كـ انبوت ف البرامتر نيو
    new_x_test = new_def_test["text"]                                               # وبعدين بنعمل عمليه كلير للنص نمسح مثلا علامات ترقيم ونخلي النص كله حروف صغيره مثلا يعني
    new_xv_test = vectorization.transform(new_x_test)                               # بعد البروسيسنج بياخد النص النضيف وبيحوله ل فيكتورز تمشي مع المودلز
    pred_lr = LR.predict(new_xv_test)                                               # بعدين بنتنبأ ان الخبر ده صحيح ولا فيك باستخدام المودلز
    pred_gbc = gbc.predict(new_xv_test)                                             # واخيرا بيظهرلنا نتائج كل مودلز
    pred_rfc = rfc.predict(new_xv_test)
    return """LR Prediction: {}
    GBC Prediction: {}
    RFC Prediction: {}""".format(
        output_label(pred_lr[0]), output_label(pred_gbc[0]), output_label(pred_rfc[0])
    )
```

```python
news_article = str(input())     # ببطلب مننا ندخل النص يدوي ك انبوت وبعدين بيحول لسلسه نصيه
```

```python
manual_testing(news_article)  #نتيجه كل مودلز
```

# ***Naive Bayes***

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
from sklearn.metrics import accuracy_score
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
# بنقسم البيانات الي مجموعات لتيست وترين
# الاكس هنا الفيتشرز زي مثلا نص او رقم
# الواي هنا الليبل زي مثلا خبر فيك ولا حقيقي
#20% من الداتا هنستخدمها للتيست
#80% للترين
# random_state بنستخدمه لتحديد عشوائيه التقسيم
```

```python
tfidf = TfidfVectorizer(stop_words='english', max_df=0.7)
# تجاهل الكلمات التي تظهر في أكثر من 70% من المستندات
```

```python
x_train_tfidf = tfidf.fit_transform(x_train)
```

```python
x_test_tfidf = tfidf.transform(x_test)
```

```python
nb_model = MultinomialNB()
```

```python
nb_model.fit(x_train_tfidf, y_train)
```

```python
y_pred = nb_model.predict(x_test_tfidf) #بنعمل تنبؤ للنصوص الجديده ف التيست باستخدام مودل nb
```

```python
accuracy = accuracy_score(y_test, y_pred)
print(f"Naive Bayes Model Accuracy: {accuracy * 100:.2f}%")
# هنا بنحسب الاكيورسي وبنحدد مدي كفاءه المودل
```
