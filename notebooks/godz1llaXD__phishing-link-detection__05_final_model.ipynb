```python
import pandas as pd

X_PATH = r"E:\file_main\Major_2\data\processed\X_features.csv"
y_PATH = r"E:\file_main\Major_2\data\processed\y_labels.csv"

X = pd.read_csv(X_PATH)
y = pd.read_csv(y_PATH).squeeze()

print(X.shape, y.shape)
```

```python
#DO NOT RETRAIN
from sklearn.ensemble import RandomForestClassifier

rf_final = RandomForestClassifier(
    n_estimators=200,
    max_depth=20,
    random_state=42,
    n_jobs=-1
)

rf_final.fit(X, y)
```

```python
rf_final.predict(X[:5])
rf_final.predict_proba(X[:5])
```

```python
import os

MODEL_DIR = r"E:\file_main\Major_2\models"
os.makedirs(MODEL_DIR, exist_ok=True)
```

```python
import joblib

MODEL_PATH = f"{MODEL_DIR}\\phishing_model.pkl"

joblib.dump(rf_final, MODEL_PATH)

print("Model saved at:", MODEL_PATH)
```

```python
FEATURE_PATH = f"{MODEL_DIR}\\feature_columns.pkl"

joblib.dump(list(X.columns), FEATURE_PATH)

print("Feature schema saved")
```

```python
import json

metadata = {
    "model": "RandomForestClassifier",
    "n_estimators": 200,
    "max_depth": 20,
    "features": list(X.columns)
}

with open(f"{MODEL_DIR}\\model_metadata.json", "w") as f:
    json.dump(metadata, f, indent=4)
```
