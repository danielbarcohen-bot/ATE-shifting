### I.1. Normalisation

Ici, nous allons r?aliser les deux fonctions de nomalisation : standard et min-max.
Nous allons prendre une matrice $X[N, M]$ de $N$ ?chantillons et $M$ colonnes.
La normalisation standard d'une colonne $j$ peut ?tre d?crite comme : 
$$standard(X_j) = \frac{X_j - \mu(X_j)}{\sigma(X_j)}$$
La nomalisation min-max d'une colonne $j$ peut ?tre d?crite comme : 
$$minmax(X_j) = \frac{X_j - min(X_j)}{max(X_j) - min(X_j)}$$


```python
# TODO: Standardisation d'une matrice
# Entr?e : la matrice des donn?es (N ?chantillons X  M caract?ristiques)
# Sortie : vecteur de M moyennes, vecteur de M ?cart-types, une matrice normalis?e
def norm_std(X: np.ndarray) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    
    moy   = X.mean(axis=0)
    std   = X.std(axis=0)
    X_new = (X-moy)/std
    return moy, std, X_new


#=====================================================================
# TEST UNITAIRE
#=====================================================================
# (array([4. , 3. , 0.5]),
#  array([1.87082869, 2.        , 0.5       ]),
#  array([[ 1.60356745,  1.        , -1.        ],
#         [-1.06904497, -1.        ,  1.        ],
#         [-0.53452248,  1.        , -1.        ],
#         [ 0.        , -1.        ,  1.        ]]))
#---------------------------------------------------------------------

X = np.array([
    [7, 5, 0],
    [2, 1, 1],
    [3, 5, 0],
    [4, 1, 1],
])
norm_std(X)
```

```python
# TODO: Normalisation min-max d'une matrice
# Entr?e : la matrice des donn?es (N ?chantillons X  M caract?ristiques)
# Sortie : vecteur de M max, vecteur de M min, une matrice normalis?e
def norm_minmax(X: np.ndarray) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    mx    = X.max(axis=0)
    mn    = X.min(axis=0)
    X_new = (X - mn)/(mx-mn)
    return mx, mn, X_new

#=====================================================================
# TEST UNITAIRE
#=====================================================================
# (array([7, 5, 1]),
#  array([2, 1, 0]),
#  array([[1. , 1. , 0. ],
#         [0. , 0. , 1. ],
#         [0.2, 1. , 0. ],
#         [0.4, 0. , 1. ]]))
#---------------------------------------------------------------------

X = np.array([
    [7, 5, 0],
    [2, 1, 1],
    [3, 5, 0],
    [4, 1, 1],
])

norm_minmax(X)
```

### I.2. Encodage One-Hot

Etant donn? un vecteur $A[N]$ repr?sentant une caract?ristique nominale donn?e, on veut encoder les valeurs en utilisant One-Hot. Pour faciliter la t?che, on vous donne l'algorithme d?taill? : 
1. Trouver les valeurs uniques dans le vecteur $A$ ; on appele ?a : un vocabulaire $V$
1. Cr?er une matrice $X[N, |V|]$ en recopiant le vecteur $V$ $N$ fois. Dans python, on peut recopier un vecteur en utilisant l'instruction : [V] * N
1. Comparer l'?galit? entre chaque ligne de $A$ et chaque ligne (qui est un vecteur) de $X$.
1. Transformer les bool?ens vers des entiers

```python
# TODO: Encodage One-Hot
# Entr?e : un vecteur d'une caract?ristique (N ?chantillons)
# Sortie : vecteur du vocabulaire V, matrice N X |V|
def one_hot(A: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    V = np.unique(A)
    X = [V] * A.size
    X = (X == A.reshape(-1, 1)).astype(int)
    
    return V, X

#=====================================================================
# TEST UNITAIRE
#=====================================================================
# (array(['COLD', 'HOT', 'MILD'], dtype='<U4'),
#  array([[0, 1, 0],
#         [0, 0, 1],
#         [1, 0, 0],
#         [0, 1, 0],
#         [0, 0, 1]]))
#---------------------------------------------------------------------

A = np.array(['HOT', 'MILD', 'COLD', 'HOT', 'MILD'])
one_hot(A)
```

### I.3. Binarisation

Etant donn? un vecteur $A[N]$ repr?sentant une caract?ristique num?rique donn?e, on veut encoder les valeurs en 0 ou 1 selon un seuil $s$.
La binarization d'un ?l?ment $A_i$ est donn?e par :
$$A_i' = \begin{cases}
1 & \text{si } A_i \ge s\\
0 & \text{sinon}\\
\end{cases}$$

```python
# TODO: Binarisation
# Entr?e : un vecteur d'une caract?ristique (N ?chantillons), un seuil de binarisation
# Sortie : un vecteur binaris? (N ?chantillons)
def bin(A: np.ndarray, seuil: float) -> np.ndarray:
    return (A >= seuil).astype(int)

#=====================================================================
# TEST UNITAIRE
#=====================================================================
# array([1, 0, 0, 0, 1, 1])
#---------------------------------------------------------------------

A = np.array([5, 2, 1, -1, 6, 4])

bin(A, 4)
```

### I.4. Analyse en composantes principales (ACP)

Algorithme:
1. Centrer et r?duire les donn?es (Standardisation) (X) pour avoir (X_std).
1. Calculer la matrice de covariance (X_cov) en se basant sur (X_std).
1. Calculer les valeurs propres (val_prop) et les vecteurs propres (vec_prop) de (X_cov). 
1. Ordonner les valeurs propres d'une maniere descendante et selon cette ordre, ordonner les vecteurs propres
1. Garder seulement les nbr_comp vecteurs propres premiers pour former le vecteur caracteristique (vec_car) 
1. Transformer les donnees $X\_reduit = (vec\_car^T \cdot X\_std^T)^T$

```python
# TODO: ACP
# Entr?e : la matrice des donn?es (N ?chantillons X  M caract?ristiques), nombre de composantes: nbr_comp
# Sortie : une matrice reduite (N ?chantillons X nbr_comp caract?ristiques)
def ACP(X: 'np.array[M, N]', nbr_comp: int) -> 'np.array[M, nbr_comp]':
    
    # 1. Centrer et reduire les donnees
    X_std = norm_std(X)[2]
     
    # 2. matrice de covariance (X_cov)
    X_cov = (1/X_std.shape[0]) * ( np.transpose(X_std) @ X_std )
     
    # 3. valeurs propres et vecteurs propres
    val_prop, vec_prop = np.linalg.eigh(X_cov)

    # 4. Ordonner vec_prop selon l'ordre descendant de val_prop
    vec_ord = np.fliplr(vec_prop)
     
    # 5. creer le vecteur caracteristique
    vec_car = vec_ord[:,:2]
     
    # 6. Transformer les donnees
    X_reduit = X_std @ vec_car
     
    return X_reduit

#=====================================================================
# TEST UNITAIRE
#=====================================================================
# array([[-2.00488333, -0.74288047],
#        [ 1.75181002,  0.27206396],
#        [-0.99259008,  1.14038556],
#        [ 1.24566339, -0.66956905]])
#---------------------------------------------------------------------

X = np.array([
    [7, 5, 0],
    [2, 1, 1],
    [3, 5, 0],
    [4, 1, 1],
])

ACP(X, 2)
```
