## Importa m?dulo com fun??es para medir o desempenho do classificador

```python
from sklearn import metrics
from sklearn.metrics import ConfusionMatrixDisplay
```

### Importa??o dos dados - Iris dataset

```python
from sklearn.datasets import load_iris
iris = load_iris()
class_names = iris.target_names
X = iris.data
y = iris.target
```

### Separa em dois conjuntos de dados (treinamento e teste)

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25) 
```

## Aplica??o do Grid_Search

```python
from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

clf = KNeighborsClassifier()

param_grid = {'n_neighbors': np.arange (1,50)}
grid = GridSearchCV(clf, param_grid, cv=5)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size =.2, random_state=42)
grid_results = grid.fit(X_train ,y_train )
# Melhor resultado (melhores par?metros)
print("Melhor resultado: %f utilizando os par?metros %s" % (grid_results.best_score_, grid_results.best_params_))
# An?lise dos resultados dos testes
medias = grid_results.cv_results_['mean_test_score']
desvios = grid_results.cv_results_['std_test_score']
params = grid_results.cv_results_['params']
for media, desvio, param in zip(medias, desvios, params):
    print("%f (%f) utilizando: %r" % (media, desvio, param))
```
