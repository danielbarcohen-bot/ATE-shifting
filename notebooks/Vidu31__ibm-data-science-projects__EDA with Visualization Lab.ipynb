### TASK  4: Visualize the relationship between FlightNumber and Orbit type


For each orbit, we want to see if there is any relationship between FlightNumber and Orbit type.


```python
# Plot a scatter point chart with x axis to be FlightNumber and y axis to be the Orbit, and hue to be the class value
# Create the scatter plot
sns.scatterplot(
    x='FlightNumber',        # Setting x-axis to FlightNumber
    y='Orbit',               # Setting y-axis to Orbit
    hue='Class',             # Color-coding by the 'class' column (success/failure)
    data=df                  # DataFrame containing the data
)

# Show the plot
plt.show()

# Calculate the success rate for each orbit
success_rate = df.groupby('Orbit')['Class'].mean() * 100  # Mean of 'class' gives success rate (1 = success, 0 = failure)

# Reset the index to have a proper DataFrame for plotting
success_rate_df = success_rate.reset_index()

# Rename the 'class' column to 'Success Rate'
success_rate_df.rename(columns={'Class': 'Success Rate'}, inplace=True)

# Display the success rate DataFrame
print(success_rate_df)
```

You can observe that in the LEO orbit, success seems to be related to the number of flights. Conversely, in the GTO orbit, there appears to be no relationship between flight number and success.

### TASK  6: Visualize the launch success yearly trend


You can plot a line chart with x axis to be <code>Year</code> and y axis to be average success rate, to get the average launch success trend.


The function will help you get the year from the date:


```python
# Define the year list globally - better function
year = []

# Function to extract the year from the date column
def Extract_year(df):
    global year
    year = df['Date'].dt.year  # Use pandas to extract the year from the Date column
    return year
```

```python
# Plot a line chart with x axis to be the extracted year and y axis to be the success rate
# Plot a line chart with x axis to be the extracted year and y axis to be the success rate
# Calculate the success rate for each year
# Assume 'Date' is in datetime format, if not, convert it
df['Date'] = pd.to_datetime(df['Date'])

# Extract years using the function
df['Year'] = Extract_year(df)

# Calculate the success rate for each year
success_rate_by_year = df.groupby('Year')['Class'].mean() * 100  # Success rate as percentage

# Reset the index for plotting
success_rate_df = success_rate_by_year.reset_index()

# Rename 'class' column to 'Success Rate'
success_rate_df.rename(columns={'Class': 'Success Rate'}, inplace=True)

# Create the line plot
sns.lineplot(
    x='Year',                  # Set x-axis to Year
    y='Success Rate',           # Set y-axis to Success Rate
    data=success_rate_df        # Data containing success rate per year
)

# Add labels and title
plt.xlabel('Year')
plt.ylabel('Success Rate (%)')
plt.title('Success Rate by Year')

# Show the plot
plt.show()
```

you can observe that the sucess rate since 2013 kept increasing till 2020

Use the function <code>get_dummies</code> and <code>features</code> dataframe to apply OneHotEncoder to the column <code>Orbits</code>, <code>LaunchSite</code>, <code>LandingPad</code>, and <code>Serial</code>.
