#**(D) Exploratory Data Analysis (Selected Dataset Only)**

```python
import pandas as pd
import kagglehub
import os

# Download the latest version of the dataset
path = kagglehub.dataset_download("aliiihussain/amazon-sales-dataset")

# kagglehub returns a directory path, so we append the filename
csv_path = os.path.join(path, "amazon_sales_dataset.csv")
df = pd.read_csv(csv_path)
print("Path to dataset files:", path)

# Display basic information and the first few rows
print("Data Info:")
print(df.info())
print("\nFirst 5 rows:")
print(df.head())
print("\nDescriptive Statistics:")
print(df.describe())
```

```python
import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub
import os
from sklearn.ensemble import IsolationForest
from scipy.stats import chi2_contingency, f_oneway
from collections import Counter
from itertools import combinations


# --- COURSE TOPIC ALIGNMENT TECHNIQUES ---

# 1. ANOMALY DETECTION (Isolation Forest)
# We detect outliers based on 'total_revenue' and 'quantity_sold'
clf = IsolationForest(contamination=0.01, random_state=42)
df['anomaly'] = clf.fit_predict(df[['total_revenue', 'quantity_sold']])


# 2. FREQUENT ITEMSETS (Category Co-occurrence)
# We find which product categories are often bought in the same region on the same day
sessions = df.groupby(['order_date', 'customer_region'])['product_category'].apply(list).reset_index()
category_pairs = []
for items in sessions['product_category']:
    if len(items) > 1:
        category_pairs.extend(combinations(sorted(set(items)), 2))


# --- BEYOND-COURSE TECHNIQUES ---

# 3. CHI-SQUARED ANALYSIS
# Testing independence between Product Category and Customer Region
contingency_table = pd.crosstab(df['product_category'], df['customer_region'])
chi2, p_chi2, dof, ex = chi2_contingency(contingency_table)



# 4. ANOVA ANALYSIS
# Testing if average Total Revenue differs significantly across Product Categories
categories = df['product_category'].unique()
revenue_groups = [df[df['product_category'] == cat]['total_revenue'] for cat in categories]
f_stat, p_anova = f_oneway(*revenue_groups)




#------------------------------------------------
#------------------------------------------------
#Charts
#------------------------------------------------
#------------------------------------------------

# 1. ANOMALY DETECTION (Isolation Forest)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='quantity_sold', y='total_revenue', hue='anomaly', palette={1: 'tab:blue', -1: 'tab:red'})
plt.title('Anomaly Detection: Revenue vs Quantity Outliers')
plt.show()


# 2. FREQUENT ITEMSETS (Category Co-occurrence)
common_pairs = Counter(category_pairs).most_common(5)
print("Top 5 Frequent Category Pairs:")
for pair, count in common_pairs:
    print(f"{pair}: {count} occurrences")
print("\n")


# 3. CHI-SQUARED ANALYSIS
plt.figure(figsize=(10, 6))
sns.heatmap(contingency_table, annot=True, fmt='d', cmap='YlGnBu')
plt.title(f'Chi-squared Heatmap (p-value: {p_chi2:.3f})')
plt.show()



# 4. ANOVA ANALYSIS
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='product_category', y='total_revenue')
plt.title(f'ANOVA: Revenue by Category (p-value: {p_anova:.3f})')
plt.xticks(rotation=45)
plt.show()



# --- SUMMARY OF STATISTICAL RESULTS ---
print("--- Statistical Summary ---")
print(f"Anomalies detected: {len(df[df['anomaly'] == -1])}")
print(f"Chi-squared P-value: {p_chi2:.4f} (Independent if > 0.05)")
print(f"ANOVA P-value: {p_anova:.4f} (No significant difference if > 0.05)")
```

#**(E) Initial Insights and Direction**

### (E) Initial Insights and Direction

#### 1. Observations
- **Statistical Uniformity**: The ANOVA test yielded a high p-value (0.9085), showing that average revenue is consistent across categories. This suggests that "product category" alone is not a strong predictor of transaction value in this dataset.
- **Independence of Variables**: The Chi-squared test (p-value 0.3807) indicates that customer region and payment method are independent. This implies that Amazon’s payment infrastructure and regional preferences are highly standardized.
- **Anomaly Distribution**: Outliers detected via Isolation Forest are concentrated in the top 1% of revenue earners, rather than being distributed across different quantity levels.

#### 2. Hypotheses
- **Hypothesis 1**: While global averages are uniform, high-support thresholds in frequent itemset mining may be masking "niche" regional co-occurrences that are more profitable than the common "Books + Electronics" pairings.
- **Hypothesis 2**: Transactional anomalies are driven by "Bulk Purchase" behaviors rather than "High Unit Price" luxury items, meaning the model is flagging volume-based outliers rather than value-based ones.

#### 3. Potential Research Questions (RQs)
- **RQ1: Sensitivity Analysis of Support**: How do varying minimum support thresholds affect the "lift" and "conviction" of association rules between disparate categories like Fashion and Home Decor?
- **RQ2: Feature Engineering for Anomalies**: Does the inclusion of 'Customer Rating' as a feature in the Isolation Forest significantly improve the separation of fraudulent transactions from legitimate high-value sales?
- **RQ3: Granular Variance**: If category-level revenue is uniform (per ANOVA), does a sub-categorical analysis (e.g., separating 'Laptops' from 'Cables' within Electronics) reveal the variance that the high-level analysis missed?
