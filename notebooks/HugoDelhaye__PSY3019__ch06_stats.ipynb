<div class="alert alert-success">
    <h4>Test-T, 1 ?chantillon</h4>
<li> scipy.stats.ttest_1samp()
<li> tester la valeur d'une moyenne de population
<li> (techniquement si les observations sont tir?es d'une distribution gaussienne

<div class="alert alert-success">
    <h4>Test-T, 2 ?chantillon</h4>
<li> scipy.stats.ttest_ind()
<li> ttest_ind va calculer le test T pour les moyennes de deux ?chantillons IND?PENDANTS
<li> tester la diff?rence entre les populations
<li> ttest_1samp a montr? que le VIQ moyen dans les populations femmes et hommes ?tait diff?rent
<li> ttest_ind va montrer si cela est significatif
<li>Renvoie:<ol>
    <li>la statistique T 
    <li>la valeur de p</ol></div>

```python
femme_viq = cerveau_df[cerveau_df['Gender'] == 'Female']['VIQ']
homme_viq = cerveau_df[cerveau_df['Gender'] == 'Male']['VIQ']
```

```python
stats.ttest_ind(femme_viq, homme_viq)
```

```python
res = stats.ttest_ind(femme_viq, homme_viq)
print(res.statistic, res.pvalue)
```

<div class="alert alert-success">
    <h4>Tests appari?s: mesures r?p?t?es sur les m?mes individus</h4>
<li> scipy.stats.ttest_ind()
<li> PIQ, VIQ et FSIQ donnent 3 mesures de QI
<li> pour tester si FISQ et PIQ sont significativement diff?rents, il faudrait faire un test ? 2 ?chantillons:
<li> nous testerons d'abord en faisant l'hypoth?se que les valeurs FSIQ et PIQ sont ind?pendantes

<div class="alert alert-success">
    <h4>Tests appari?s: mesures r?p?t?es sur les m?mes individus</h4>

<ul><li>les observations: FSIQ et PIQ sont mesur?s sur les m?mes individus

<div class="alert alert-success">

```python
model_anova.pvalues.loc['education:gender']
```
