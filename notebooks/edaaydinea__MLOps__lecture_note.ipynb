# Data Journey and Data Storage

## Data Journey
**1. Data Journey and Provenance:**
   - **Definition:** Data provenance refers to tracking the lifecycle of data and its transformations throughout the production pipeline, including feature engineering and model training.
   - **Importance:** Understanding data provenance is crucial for debugging, reproducibility, and compliance (e.g., GDPR).
   - **Components:**
     - **Raw Data:** Initial features and labels from sources.
     - **Transformations:** Changes in data format, feature engineering, etc.
     - **Artifacts:** Results from pipeline components, including transformed data, models, schemas, and metrics.

**2. Metadata and Tracking:**
   - **ML Metadata Library:** Helps track data and model changes during training.
   - **Purpose:** Assists with debugging, reproducibility, and understanding model results.

**3. Artifacts:**
   - **Definition:** Outputs generated by pipeline components, including all stages of transformed data and model outputs.
   - **Tracking:** Essential for understanding pipeline operations and comparing training runs.

**4. Data Lineage and Provenance:**
   - **Synonyms:** Data lineage and data provenance are often used interchangeably.
   - **Purpose:** Helps trace the sequence of artifacts and understand how data transforms and evolves through the pipeline.
   - **Application:** Debugging, comparing training runs, and ensuring compliance with regulations.

**5. Version Control:**
   - **Code Version Control:** Managed with tools like GitHub.
   - **Environment Versioning:** Achieved using Docker and Terraform for consistent environments.
   - **Data Versioning:**
     - **Importance:** Tracks changes in data files, allowing restoration of previous versions.
     - **Tools:** DVC (Data Version Control), Git LFS (Large File Storage).

**6. GDPR Compliance:**
   - **Requirement:** Organizations must track the origin, changes, and location of personal data.
   - **Data Lineage:** Provides a means to determine how data was used and transformed, which is key for compliance.

**7. Model Understanding:**
   - **Relation to Data:** The model represents a transformation of the data.
   - **Evolution Tracking:** Helps understand how the model changes during training and optimization.

**8. Practical Considerations:**
   - **Data Versioning Tools:** Essential for managing large files and ensuring reproducibility.
   - **Debugging and Comparisons:** Provenance helps in debugging and understanding discrepancies in results from different training runs.

### Key Notes:
- **Data Provenance:** Essential for tracking and understanding data and model transformations.
- **Artifacts:** Include all outputs from the pipeline; tracking these is crucial for debugging.
- **Versioning:** Includes code, environment, and data versioning, each with specialized tools.
- **GDPR Compliance:** Tracking data provenance is necessary for personal data regulations.
- **Model Evolution:** Provenance helps understand model changes and performance.


## Introduction to ML Metadata

**1. Importance of ML Metadata:**
   - **Legal and Regulatory Considerations:** Essential for interpreting models and tracing data lineage to limit legal liability in business, healthcare, and financial decisions.
   - **Tracking and Debugging:** Metadata helps track artifacts and pipeline changes, similar to logging in software development.

**2. ML Metadata (MLMD):**
   - **Function:** Tracks and retrieves metadata associated with ML workflows, useful for analyzing pipeline components and debugging.
   - **Integration:** Can be integrated with ML pipelines or used independently. When integrated, direct interaction with MLMD may not be required.

**3. Key Components:**
   - **Artifacts:** Elementary units of data, either input to or output from ML pipeline components. Stored in a relational database for properties and in a file system or block store for large objects.
   - **Executions:** Records of component runs, including runtime parameters. Each execution is associated with one type of component.
   - **Contexts:** Groupings of artifacts and executions, representing metadata for projects, experiments, and pipelines.
   - **Types:** Define properties of artifacts, executions, and contexts.
   - **Relationships:** Track interactions between artifacts and executions, e.g., input and output events.

**4. Architecture of ML Metadata:**
   - **Components:** Connected to a centralized metadata store, allowing independent access to metadata throughout the pipeline.
   - **GUI Console:** Optional interface for tracking progress and accessing metadata.
   - **Artifact and Execution Flow:**
     - **Artifacts:** Inputs and outputs to components.
     - **Executions:** Recorded interactions between artifacts and components.
     - **Events:** Record relationships between artifacts and executions (input and output events).

**5. Data Storage:**
   - **Backend Storage:** Metadata stored in databases like SQLite or MySQL, with large objects managed in file systems or block stores.

**6. Summary:**
   - MLMD facilitates tracking metadata and understanding pipeline results, essential for debugging and analyzing model performance. It provides insight into the training process, both current and historical.

### Key Notes:
- **MLMD Purpose:** Track metadata, artifacts, executions, and data lineage for ML pipelines.
- **Artifacts and Executions:** Core units of MLMD, with artifacts as data units and executions as component run records.
- **Contexts and Relationships:** Group and relate metadata for comprehensive tracking.
- **Architecture:** Centralized store with optional GUI for pipeline tracking; supports various backend storage solutions.

# Evolving Data

## Schema Development

Understanding and developing schemas is crucial for managing evolving data in machine learning pipelines. Here’s a breakdown of key concepts and considerations for schema development:

### **What is a Schema?**
- **Definition:** A schema is a structured representation of the features in a dataset or project. It outlines the data types, constraints, and attributes of features.
- **Components:**
  - **Feature Name:** Identifies the data field.
  - **Feature Type:** Specifies data types like integer, float, string, or categorical.
  - **Required Status:** Indicates if the feature is mandatory.
  - **Valency:** For features with multiple values (e.g., lists or arrays), it defines the minimum and maximum number of values.
  - **Range and Categories:** Defines permissible values or ranges for features.
  - **Default Values:** Provides defaults for missing data.

### **Importance of Schemas in ML Pipelines**
- **Data Evolution:** As datasets change over time, schemas help manage and track these changes. They are crucial for adapting to new data distributions and feature sets.
- **Data Validation:** Schemas help identify issues such as missing values or incorrect data types, ensuring data quality and consistency.

### **Handling Evolving Data**
- **Initial Schema Creation:** When first receiving data, your schema should reflect the expected range of values and categories, even if the initial dataset only covers a small portion.
- **Anomaly Detection:** Schemas can help spot anomalies and data issues, which may indicate the need for schema updates or further investigation.

### **Requirements for ML Pipeline Platforms**
1. **Reliability:**
   - **Data Consistency:** The platform should handle inconsistent data gracefully.
   - **Error Management:** It should manage runtime errors and misconfigurations efficiently.
   - **Orchestration:** Ensure smooth execution across pipeline components.

2. **Scalability:**
   - **Training Data:** The platform must efficiently handle large volumes of training data and utilize accelerators (e.g., GPUs, TPUs).
   - **Serving:** In online scenarios, it should scale to handle varying request traffic, balancing latency and cost.

3. **Anomaly Detection:**
   - **Schema Updates:** Your system should detect anomalies and alert you to update the schema if necessary.
   - **Error Handling:** Treat data anomalies with the same importance as code bugs, ensuring they are addressed promptly.

### **Leveraging Schemas for Automation**
- **Automatic Feature Engineering:** Schemas can be key inputs for automated processes that generate features from raw data, improving efficiency and consistency in model development.

### **Summary**
Schemas are vital for managing the complexities of evolving data in ML pipelines. They provide a structured way to define and validate data, handle inconsistencies, and adapt to changes over time. Implementing robust schema management practices helps ensure that your ML pipelines remain reliable, scalable, and effective in handling real-world data challenges.

## Schema Environments

### **Schema Environments in Evolving Data**

Managing schema environments is crucial for handling the dynamic nature of data in production ML pipelines. Here’s a comprehensive look at how schema environments work and their importance:

### **Understanding Schema Environments**

1. **Schema Evolution:**
   - **Definition:** As your data and business needs evolve, your schemas may need to change to reflect new data patterns, features, or requirements.
   - **Version Control:** Just like code, schemas should be versioned to manage changes over time. This allows you to maintain different versions of schemas for development, testing, and production environments.

2. **Multiple Schema Versions:**
   - **Development, Testing, and Production:** It's common to have separate schemas for different stages of the pipeline. For example, a development schema might be less strict than a production schema.
   - **Training and Serving:** Different schemas might be required for training and serving environments. For instance, a schema used for model training might require certain features that are not present during serving (e.g., labels).

### **Handling Schema Environments**

1. **Creating Schema Environments:**
   - **Purpose:** Define different schemas to handle variations in data for different scenarios, such as development, testing, training, and serving.
   - **Example:** You might need one schema for training data where all features (including labels) are required, and another schema for serving data where some features (like labels) may be absent.

2. **Schema Customization:**
   - **Customizing for Use Cases:** Customize schemas based on the specific requirements of each environment. For instance, if a feature is only available during training, adjust the schema used for serving to exclude that feature.

### **Example: Managing Multiple Schemas**

1. **Inferring Serving Schema:**
   - **Tool:** Use TensorFlow Data Validation (TFTV) to infer and validate schemas against your serving data.
   - **Process:** Generate statistics for the serving dataset and check for anomalies using TFTV.
  
  ```python

   stats_options = tfdv.StatsOptions(schema=schema, infer_type_from_schema=True)

   eval_stats = tfdv.generate_statistics_from_csv(data_location = SERVING_DATASET,
                                                 stats_options=stats_options)

   serving_anomalies = tfdv.validate_statistics(statistics=eval_stats, schema=schema)
   tfdv.display_anomalies(serving_anomalies)
   ```

2. **Handling Anomalies:**
   - **Scenario:** If a required feature (like a label) is missing in the serving data, TFTV will flag it as an anomaly.
   - **Solution:** Adjust the schema to reflect the reality of the serving environment. For example, if the label is not available during serving, update the schema to make it optional for the serving environment.

3. **Implementing Schema Environments:**
   - **Setup:** Create two schema environments (e.g., training and serving) with their respective requirements.
   - **Modification:** For the serving schema, remove features that won’t be present in serving data.
   - **Validation:** Use the updated schema to validate the serving data and ensure there are no anomalies.

```python

schema.default_environment.append('TRAINING')
schema.default_environment.append('SERVING')

tfdv.get_feature(schema, "Cover_Type").not_in_environment.append("SERVING")

serving_anoamlies = tfdv.validate_statistics(statistics=eval_stats, schema=schema, environment='SERVING')

tfdv.display_anomalies(serving_anomalies)
# No anomalies found

```

### **Code Implementation Overview**

1. **Define Schema Environments:**
   - Create schemas for different environments (training and serving).
   - Modify schemas as needed to reflect differences in data availability.

2. **Validate Data:**
   - Use the appropriate schema to validate data for each environment.
   - Address any anomalies flagged by the schema validation process.

### **Summary**

- **Schema Evolution:** Regularly update and fine-tune schemas to adapt to changes in data.
- **Reliability and Scalability:** Ensure your schema management supports reliable and scalable data processing.
- **Schema Environments:** Implement different schemas for development, testing, training, and serving to handle variations in data effectively.

By managing schema environments properly, you can ensure that your ML pipelines handle evolving data gracefully, detect and address anomalies, and maintain robust and scalable data processing workflows.

# Enterprise Data Storage

## Feature Stores

#### **Feature Stores Overview**
- **Definition:** Central repositories for storing documented, curated, and access-controlled features used in machine learning.
- **Purpose:** Facilitate sharing, discovery, and usage of curated features; interface between feature engineering and model development.
- **Benefits:**
  - **Reduces Redundancy:** Avoids duplicating feature engineering work across different pipelines.
  - **Enables Data Sharing:** Allows teams to access and leverage common data across various modeling scenarios.
  - **Centralized Management:** Simplifies data management and consistency across projects.

#### **Key Features**
- **Online and Offline Processing:** Supports both real-time (online) and batch (offline) feature processing.
- **Access Control:** Role-based permissions ensure secure access to features.
- **Feature Transformation:** Enables aggregation, anonymization, and other transformations.
- **Compliance:** Supports GDPR compliance through data anonymization and purging.
- **Metadata Management:** Facilitates feature discovery and documentation through metadata.

#### **Data Handling**
- **Online Usage:** Strict latency requirements; pre-computed features often used to ensure low latency.
- **Batch Processing:** Less stringent latency requirements; suitable for processing historical features.
- **Time Travel Avoidance:** Ensures that features used in predictions are available at the time of serving; prevents using future data in predictions.

#### **Operational Aspects**
- **Pre-Computing:** Regularly scheduled pre-computation (e.g., cron jobs) to handle feature engineering and updates.
- **Integration:** Can integrate with monitoring tools for feature processing and data adjustments.
- **Training and Serving Consistency:** Ensures the same features are used in both training and serving to avoid skew.

#### **Goals of Feature Stores**
- **Unified Management:** Provide a centralized system for managing feature data.
- **Scalability:** Designed to scale from individual users to large enterprises.
- **Performance:** Maintain performance and efficiency in feature access and processing.
- **Consistency:** Ensure consistent data usage and point-in-time correctness for model predictions.

## Data Warehouse

#### **Overview**
- **Definition:** A data warehouse is a centralized repository that aggregates data from multiple sources for processing and analysis, optimized for read operations and batch processing.
- **Purpose:** Originally designed for big data and business intelligence applications; also valuable for production machine learning.

#### **Key Features**
- **Subject-Oriented:** Focuses on specific topics or areas, such as customers or vendors.
- **Data Aggregation:** Collects data from various sources, including relational databases and files.
- **Timestamped Data:** Maintains context by recording the time when data was generated.
- **Nonvolatile:** Old data versions are preserved; allows historical data access and analysis.

#### **Advantages**
- **Enhanced Analysis:** Time-stamped data helps maintain context and improves analysis capabilities.
- **Improved Data Quality:** Consistent schema ensures better data quality and consistency.
- **High ROI:** Studies indicate that data warehouses offer significant returns on investment for various use cases.
- **Efficient Read and Query:** Typically provides fast access to data and high read/query efficiency.

#### **Comparison with Databases**
- **Purpose:**
  - **Data Warehouse:** Optimized for complex analysis and historical data.
  - **Database:** Primarily for transaction processing and real-time data availability.
- **Data Availability:**
  - **Data Warehouse:** May have a delay between data entry and availability.
  - **Database:** Data is available immediately after storage.
- **Historical Data:**
  - **Data Warehouse:** Stores data over time; supports historical data analysis.
  - **Database:** Generally does not focus on historical data.
- **Data Volume:**
  - **Data Warehouse:** Handles larger volumes of data.
  - **Database:** Typically manages smaller datasets.
- **Query Complexity:**
  - **Data Warehouse:** Supports complex, long-running queries.
  - **Database:** Queries are simpler and execute in real-time.
- **Normalization:**
  - **Data Warehouse:** Normalization is not required.
  - **Database:** Normalization is necessary to ensure data integrity.

## Data Lakes

#### **Overview**
- **Definition:** A data lake is a repository that stores data in its raw, unprocessed format, typically as blobs or files. It aggregates data from various sources, accommodating structured, semi-structured, and unstructured data.
- **Purpose:** Provides a flexible and scalable storage solution for large volumes of diverse data types, often used by data professionals.

#### **Key Features**
- **Raw Data Storage:** Data is stored in its original format without preprocessing or schema enforcement.
- **Data Variety:** Can include structured data (e.g., relational databases), semi-structured data (e.g., CSV files), and unstructured data (e.g., images, documents).
- **Flexibility:** Changes to data are easier to implement due to lack of schema constraints.

#### **Comparison with Data Warehouses**
- **Data Format:**
  - **Data Lake:** Stores data in raw format without a consistent schema.
  - **Data Warehouse:** Stores data in a consistent, schema-defined format.
- **Purpose Determination:**
  - **Data Lake:** The purpose of the data is often not determined at the time of storage.
  - **Data Warehouse:** Data is stored for specific, pre-defined purposes.
- **User Base:**
  - **Data Lake:** Typically used by data professionals such as data scientists.
  - **Data Warehouse:** Often used by business professionals for analysis and reporting.
- **Data Changes:**
  - **Data Lake:** Easier to modify and adapt due to schema-less nature.
  - **Data Warehouse:** Changes to data can be complex and costly due to schema requirements.

#### **Summary of Enterprise Data Storage Solutions**
- **Feature Stores:** Centralized repositories for curated features used in machine learning, aimed at reducing redundancy and enhancing feature discovery.
- **Data Warehouses:** Subject-oriented, read-optimized repositories for structured data, designed for complex queries and high efficiency in analysis.
- **Data Lakes:** Flexible, schema-less repositories for raw and diverse data types, facilitating broad data collection and storage with easy modification.
