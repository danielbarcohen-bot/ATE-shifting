## `pandas` `pd.cut()` e `pd.qcut()`

* `pd.cut()`: ordena valores em *bins* (faixas de valores)
    - `bins=x`: quantidade de faixas de valores desejada 
* `pd.qcut()`: mesma coisa que `pd.cut()` mas usa *quantis*
    - `q=x`: quantidade de quantis
        * 3: tercis
        * 4: quantis
        * 5: quintis
        * 10: decis

```python
# 4 faixas (Quantis)
pd.cut(corona['Confirmed'],bins=4)
```

```python
# 3 (Tercis)
pd.cut(corona['Confirmed'], 3)
```

## `pandas` - `join()`

| Op??o    | Comportamento                                                                       |
| -------- | ----------------------------------------------------------------------------------- |
| `inner`  | Usa somente as combina??es de chaves (*keys*) observadas em ambos *DataFrames*      |
| `left`   | Usa todas as combina??es de chaves (*keys*) encontradas no *DataFrames* da esquerda |
| `right`  | Usa todas as combina??es de chaves (*keys*) encontradas no *DataFrames* da direita  |
| `output` | Usa todas as combina??es de chaves (*keys*) observadas em ambos *DataFrames* juntos |

![pandas-join](http://www.datasciencemadesimple.com/wp-content/uploads/2017/09/join-or-merge-in-python-pandas-1.png)
