# UFOS
El famoso dataset de avistamientos de OVNIs (Objetos Voladores No Identificados) contiene informaci?n recopilada sobre avistamientos de fen?menos inexplicables en el cielo, que han sido reportados por testigos en todo el mundo a lo largo de los a?os. Este dataset contiene varias columnas que describen diferentes aspectos de cada avistamiento, como la fecha y hora del avistamiento, la ubicaci?n geogr?fica, la forma del objeto visto, la duraci?n del avistamiento, y a veces incluye detalles adicionales proporcionados por los testigos.

Haremos algo muy simple. A partir de este dataset, tomaremos la longitud y latitud para entrenar un modelo que nos diga en que pa?s nos encontramos en base a estos dos par?metros.

## Cleaning Data

```python
import pandas as pd
import numpy as np

ufox = pd.read_csv('./data/ufos.csv')
ufox.head(3)
```

```python
ufos = pd.DataFrame({'Country': ufox['country'],'Latitude': ufox['latitude'],'Longitude': ufox['longitude']})

ufos.Country.unique()
```

```python
# Reducir la cantidad de datos eliminando los valores nulos
ufos.dropna(inplace=True)

ufos.info()
```

```python
# Import Scikit-learn's `LabelEncoder` library to convert the text values for countries to a number:
from sklearn.preprocessing import LabelEncoder

ufos['Country'] = LabelEncoder().fit_transform(ufos['Country'])

ufos.head()
```

## Build the model

```python
from sklearn.model_selection import train_test_split

Selected_features = ['Latitude','Longitude']

X = ufos[Selected_features]
y = ufos['Country']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
# Train your model using logistic regression:
from sklearn.metrics import accuracy_score, classification_report 
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)

print(classification_report(y_test, predictions))
print('Predicted labels: ', predictions)
print('Accuracy: ', accuracy_score(y_test, predictions))
```

The accuracy isn't bad (around 95%), unsurprisingly, as `Country` and `Latitude/Longitude` correlate.

```python
# Importar el LabelEncoder
from sklearn.preprocessing import LabelEncoder

# Definir las nuevas coordenadas
new_latitude = 53.200000
new_longitude = -2.916667

# Realizar una predicci?n para las nuevas coordenadas
predicted_country_encoded = model.predict([[new_latitude, new_longitude]])

# Decodificar el pa?s predicho
predicted_country_decoded = LabelEncoder().fit(ufox['country']).inverse_transform(predicted_country_encoded)

# Imprimir el pa?s predicho
print("Pa?s predicho para las coordenadas ({}, {}): {}".format(new_latitude, new_longitude, predicted_country_decoded))

```
