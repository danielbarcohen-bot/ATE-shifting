```python
import janitor as jn
Xbad = pd.DataFrame(
  {
    "A": [1, None, 3],
    " sales numbers ": [20.0, 30.0, None],
  }
)

jn.clean_names(Xbad)
```

```python
def clean_col(name):
  return (
    name.strip().lower().replace(" ", "_")
  )

Xbad.rename(columns=clean_col)
```

```python
Xbad.fillna(10)
```

```python
jn.fill_empty(
  Xbad,
  columns=["A", " sales numbers "],
  value=10,
)
```
