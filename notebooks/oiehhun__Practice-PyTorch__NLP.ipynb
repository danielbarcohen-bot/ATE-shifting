```python
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    'This is last chance.',
    'and if you do not have this chance.',
    'you will never get any chance.',
    'will you do get this one?',
    'please, get this chance.'
]
vect = CountVectorizer()
vect.fit(corpus)
vect.vocabulary_
```

### TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

doc = ['I like machin learning', 'I love deep learning', 'I run everyday']
tfidf_vectorizer = TfidfVectorizer(min_df=1)
tfidf_matrix = tfidf_vectorizer.fit_transform(doc)
doc_distance = (tfidf_matrix * tfidf_matrix.T)
print('???? ??', str(doc_distance.get_shape()[0]) + 'x' + str(doc_distance.get_shape()[1]), '??? ??????.')
print(doc_distance.toarray())
```

### Word2Vec

```python
from nltk.tokenize import sent_tokenize, word_tokenize
import warnings
warnings.filterwarnings(action='ignore')
import gensim
from gensim.models import Word2Vec

sample = open('/home/xogns5037/Basic_PyTorch/data/peter.txt', 'r', encoding='UTF-8')
s = sample.read()

f = s.replace('\n', ' ')
data = []

for i in sent_tokenize(f):
    temp = []
    
    for j in word_tokenize(i):
        temp.append(j.lower())
    data.append(temp)

data
```

```python
model1 = gensim.models.Word2Vec(data, min_count=1, vector_size=100, window=5, sg=0)

print("Cosine similarity between 'peter'" + "'wendy' - CBOW : ", model1.wv.similarity('peter', 'wendy'))
```

```python
from gensim.test.utils import common_texts
from gensim.models import FastText

model = FastText('home/xogns5037/Basic_PyTorch/data/peter.txt', vector_size=4, window=3, min_count=1, epochs=10)
sim_score = model.wv.similarity('peter', 'wendy')
print(sim_score)
```

```python
sim_score = model.wv.similarity('peter', 'hook')
print(sim_score)
```
