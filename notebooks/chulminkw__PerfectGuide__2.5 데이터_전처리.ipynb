* StandardScaler

```python
from sklearn.datasets import load_iris
import pandas as pd
# ?? ??? ?? ???? DataFrame?? ?????. 
iris = load_iris()
iris_data = iris.data
iris_df = pd.DataFrame(data=iris_data, columns=iris.feature_names)

print('feature ?? ?? ?')
print(iris_df.mean())
print('\nfeature ?? ?? ?')
print(iris_df.var())
```

```python
from sklearn.preprocessing import StandardScaler

# StandardScaler?? ??
scaler = StandardScaler()
# StandardScaler ? ??? ? ??. fit( ) ? transform( ) ??.  
scaler.fit(iris_df)
iris_scaled = scaler.transform(iris_df)

#transform( )? scale ??? ??? ?? numpy ndarry? ???? ?? DataFrame?? ??
iris_df_scaled = pd.DataFrame(data=iris_scaled, columns=iris.feature_names)
print('feature ?? ?? ?')
print(iris_df_scaled.mean())
print('\nfeature ?? ?? ?')
print(iris_df_scaled.var())
```

* MinMaxScaler

```python
from sklearn.preprocessing import MinMaxScaler

# MinMaxScaler?? ??
scaler = MinMaxScaler()
# MinMaxScaler ? ??? ? ??. fit() ? transform() ??.  
scaler.fit(iris_df)
iris_scaled = scaler.transform(iris_df)

# transform()? scale ??? ??? ?? numpy ndarry? ???? ?? DataFrame?? ??
iris_df_scaled = pd.DataFrame(data=iris_scaled, columns=iris.feature_names)
print('feature?? ?? ?')
print(iris_df_scaled.min())
print('\nfeature?? ?? ?')
print(iris_df_scaled.max())
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

# ?? ???? 0 ?? 10??, ??? ???? 0 ?? 5?? ?? ??? ??? ??? ??
# Scaler???? fit(), transform()? 2?? ?? ???? ????? reshape(-1, 1)? ?? ??
train_array = np.arange(0, 11).reshape(-1, 1)
test_array =  np.arange(0, 6).reshape(-1, 1)
```

```python
# ??? 0, ??? 1? ???? MinMaxScaler?? ??
scaler = MinMaxScaler()
# fit()?? ?? train_array ???? ???? 0, ???? 10?? ??.  
scaler.fit(train_array)
# 1/10 scale? train_array ??? ???. ?? 10-> 1? ???.
train_scaled = scaler.transform(train_array)
 
print('?? train_array ???:', np.round(train_array.reshape(-1), 2))
print('Scale? train_array ???:', np.round(train_scaled.reshape(-1), 2))
```

```python
# ??? ??? MinMaxScaler? test_array? fit()?? ?? ?? ???? ???? 0, ???? 5?? ??? 
scaler.fit(test_array)
# 1/5 scale? test_array ??? ???. ?? 5->1? ??.  
test_scaled = scaler.transform(test_array)
# train_array ?? ??
print('?? test_array ???:', np.round(test_array.reshape(-1), 2))
print('Scale? test_array ???:', np.round(test_scaled.reshape(-1), 2))
```
