### Label Encoding

```python
from sklearn.preprocessing import LabelEncoder
LE=LabelEncoder()
```

```python
df['model']=LE.fit_transform(df['model'])
```

```python
df['fuelType']=LE.fit_transform(df['fuelType'])
```

```python
df['transmission']=LE.fit_transform(df['transmission'])
```

```python
df.info()
```

```python
features=['model','year','transmission','mileage','fuelType','tax','mpg','engineSize']
```

```python
X=df[features]
Y=df['price']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.4,random_state=1)
```

```python
Y_train.shape
```

```python
Y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
LR=LinearRegression()
```

```python
lr=LR.fit(X_train,Y_train)
```

```python
pred=lr.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
```

```python
print(r2_score(Y_test,pred))
```

```python
print(mean_squared_error(Y_test,pred))
RMSE=np.sqrt(mean_squared_error(Y_test,pred))
print('RMSE : ',RMSE)
```

### Gradient Boosting


```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
regressor=GradientBoostingRegressor(max_depth=2,n_estimators=10,learning_rate=0.1)
```

```python
regressor.fit(X_train,Y_train)
```

```python
errors=[mean_squared_error(Y_test,pred) for pred in regressor.staged_predict(X_test)]
```

```python
best_n_estimators = np.argmin(errors)
```

```python
best_n_estimators
```

```python
best_reg=GradientBoostingRegressor(max_depth=2,n_estimators=best_n_estimators,learning_rate=0.5)
```

```python
best_reg.fit(X_test,pred)
```

```python
GB_pred=best_reg.predict(X_test)
```

```python
mean_absolute_error(Y_test,GB_pred)
```

```python
mean_absolute_error(Y_test,pred)
```

```python
r2_score(Y_test,GB_pred)
```
