```python
import cv2
import numpy as np
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# ??CIFAR-10??
from keras.datasets import cifar10
(x_train, y_train), (x_test, y_test) = cifar10.load_data()

# ???SIFT???
sift = cv2.SIFT_create()

def extract_features(images):
    features = []
    for img in images:
        gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        kp, des = sift.detectAndCompute(gray, None)
        if des is not None:
            des_mean = np.mean(des, axis=0)  # ??????
        else:
            des_mean = np.zeros(128)  # ??????
        features.append(des_mean)
    return np.array(features)

# ???????????(?????,??????)
x_train_sift = extract_features(x_train[:1000])
x_test_sift = extract_features(x_test[:200])

# ??SVM???
clf = make_pipeline(StandardScaler(), SVC(kernel='rbf', C=1.0))
clf.fit(x_train_sift, y_train[:1000].ravel())

# ??
y_pred = clf.predict(x_test_sift)
print("???????:", accuracy_score(y_test[:200], y_pred))
```
