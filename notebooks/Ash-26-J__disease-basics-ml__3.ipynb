```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn import svm
from sklearn.metrics import accuracy_score
```

```python
diabitic =pd.read_csv('/content/diabetes.csv')
diabitic.head()
```

data spliting

```python
diabitic['Outcome'].value_counts()
```

```python
x=diabitic.drop(columns='Outcome',axis=1)
y=diabitic['Outcome']
```

```python
print(x)
```

```python
print(y)
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(x)
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=0.2,stratify=y,random_state=2)
```

```python
print(xtrain)
```

```python
classifier = svm.SVC(kernel='linear')
```

```python
classifier.fit(xtrain,ytrain)
```

```python
xtrainp = classifier.predict(xtrain)
train= accuracy_score(xtrainp,ytrain)
print(train)
```

```python
xtestp = classifier.predict(xtest)
test= accuracy_score(xtestp,ytest)
print(test)
```

Building prediction system

```python
inputdata = (6, 148, 72, 35, 0, 33.6, 0.627, 50) # Example 8-feature input tuple
inputdata_as_numpy_array = np.asarray(inputdata) # Use the correct inputdata tuple
inputdata_reshaped = inputdata_as_numpy_array.reshape(1,-1)
std_data = scaler.transform(inputdata_reshaped)
prediction = classifier.predict(std_data)
print(prediction)
if (prediction[0] == 0):
  print('The person is not diabetic')
else:
  print('The person is diabetic')
```
