## Qu? es el sesgo

 El sesgo mide la distancia que hay entre el valor esperado de nuestro modelo y el verdadero valor,

$$\text{sesgo}(\hat\beta)=\mathbb{E}(\hat\beta)-\beta$$



## El mejor estimador insesgado..

Pues bien, veamos c?mo se relaciona todo esto con la regresi?n lineal. La ecuaci?n del modelo de regresi?n es,

$$y = X\beta + \varepsilon$$

Donde $y$ es el vector de la variable respuesta (la que queremos predecir), $X$ es la matrz con las variables predictoras (que usaremos para hacer las predicciones) y $\beta$ es el vector con los par?metros del modelo de regresi?n (los valores de la ecuaci?n de la recta). Recordemos que para encontrar la soluci?n ?ptima del modelo de regresi?n resolv?amos la ecuaci?n,

$$
J(\beta)=\text{min}\left\{\frac{1}{n}\sum_{i=1}^n(y_i-\hat{y}_i)^2\right\}
$$

Si no recuerdas de donde viene esta f?rmula no te preocupes, puedes leerlo [en el post anterior](https://blog.noaxacademy.com/el-abc-del-machine-learning-regresion-lineal-con-python/). No quiero entrar en muchos detalles t?cnicos, pero es importante saber que matem?ticamente, la soluci?n del modelo de regresi?n lineal es el **mejor estimador lineal insesgado**, es decir, es el que tiene la menor varianza de entre todos los posibles tipos de modelos lineales insesgados. Esto en principio es bueno... ?o no?

Aqu? entra en juego el equilibrio entre el sesgo y la varianza, lo que en ingl?s se llama el **variance bias tradeoff**.


## El equilibrio entre sesgo y varianza

Cuando se construye un modelo de machine learning, existe un equilibrio entre su sesgo y su varianza. 
* Si el sesgo disminuye, la varianza aumenta;
* Si la varianza disminuye, el sesgo aumenta.

Y adem?s, tambi?n se sabe que el aumentar la complejidad del modelo, es decir, **aumentar el n?mero de variables** aumenta la varianza (y disminuye el sesgo). 



## La penalizaci?n ridge

Y aqu? llegamos a **la clave para entender la regresi?n ridge**: Hemos dicho que el modelo de regresi?n normal es el mejor estimador insesgado, pero ?puede ser que consigamos mejores predicciones con un modelo que tenga un poco de sesgo, pero que a cambio reduzca mucho la varianza?

La respuesta, como te puedes imaginar, es si, y eso es precisamente lo que hace la penalizaci?n ridge. Matem?ticamente, los modelos ridge resuelven la ecuaci?n,

$$
J(\beta)=\text{min}\left\{\frac{1}{n}\sum_{i=1}^n(y_i-\hat{y}_i)^2+\alpha\sum_{j=1}^p\beta_j^2\right\}
$$

Esta ecuaci?n tiene dos partes, la primera parte es la misma que en un modelo de regresi?n no penalizado, y la segunda parte es la que se conoce como **penalizaci?n ridge**. La soluci?n del modelo ridge ser?n aquellos valores de $\beta$ que minimicen la funci?n $J(\beta)$.

**?Pero c?mo funciona esta penalizaci?n?** Si los valores de $\beta$ fuesen muy grandes, el valor de la penalizaci?n ridge ser?a muy grande, pero lo que queremos es buscar el valor m?nimo de $J(\beta)$. Es decir, que con esta penalizaci?n estamos "forzando" a que los coeficientes $\beta$ no sean excesivamente grandes. De forma intuitiva, valores peque?os de los coeficientes har?n que la varianza sea menor. 

## Optimizaci?n del modelo

F?jate por ?ltimo en ese $\alpha$ que aparece en la f?rmula de la penalizaci?n. Eso es lo que se conoce como un hiper par?metro del modelo de regresi?n, y es un coeficiente que nosotros, como analistas de datos, debemos ajustar. Valores muy grandes de $\alpha$ hacen que la importancia de la penalizaci?n sea mayor, y que los coeficientes $\beta$ se hagan m?s peque?os para conseguir minimizar $J(\beta)$, y valores muy peque?os de $\alpha$ hacen que la penalizaci?n no tenga tanta importancia. De hecho, si $\alpha=0$ (este es el valor m?s peque?o que puede tomar) estaremos resolviendo un modelo no penalizado.

En la pr?ctica, para encontrar el valor ?ptimo de $\alpha$ suele considerarse una lista de valores posibles, y se resuelve el modelo con todos ellos. Por ?ltimo se selecciona el $\alpha$ que minimice el error.


Veamos c?mo resolver un modelo de regresi?n ridge en Python. Para ello usaremos un conjunto de datos de $442$ pacientes con diabetes. En este conjunto de datos se midieron $10$ variables, la edad, sexo, ?ndice de masa corporal, presi?n sangu?nea, y otras seis mediciones de serum en sangre, as? como una variable adicional que mide el progreso de la enfermedad en los pacientes. Nuestro objetivo es usar las $10$ variables predictoras para construir un modelo que prediga el progreso de la enfermedad.

En primer lugar cargamos las librer?as de python que utilizaremos en esta secci?n:

```python
import numpy as np 
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

Ahora cargamos los datos usando la funci?n `load_diabetes()`, y los escalamos para que tengan media $0$ y desviaci?n t?pica $1$.

```python
diabetes = load_diabetes()
x = diabetes.data
y = diabetes.target

# Split into train / test
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=100, random_state=42)

# Scale
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

Construimos en primer lugar un modelo de regresi?n no penalizado usando la parte de entrenamiento (train) de nuestros datos, y despu?s evaluamos el modelo calculando su error cuadr?tico medio en la parte de test.

```python
linear_model = LinearRegression()
linear_model.fit(x_train, y_train)

lm_prediction = linear_model.predict(x_test)
lm_error = mean_squared_error(y_test, lm_prediction)
```

Ahora vamos a construir los modelos de regresi?n ridge. Para eso definimos un vector de posibles valores para $\alpha$. Lo habitual con estos par?metros es definirlos usando escala logar?tmica. En este ejemplo consideramos 500 posibles valores entre $10^{-10}$ y $10^2$. Para cada posible valor de $\alpha$, entrenamos el modelo en la parte de train, y lo testeamos en la parte de test.

```python
alpha = 10**np.linspace(-10, 2, 500)
error_array = np.zeros(len(alpha))

for idx, a in enumerate(alpha):
  ridge_regression = Ridge(alpha=a)
  ridge_regression.fit(x_train, y_train)
  prediction = ridge_regression.predict(x_test)
  error_array[idx] = mean_squared_error(y_test, prediction)
```

```python
# Optimal alpha
smallest_error = np.min(error_array)
optimal_alpha = alpha[np.argmin(error_array)]

print(f'El valor ?ptimo de alpha es: {np.round(optimal_alpha, 2)}, con el que se obtiene un error de {np.round(smallest_error, 2)}')
```

Por ?ltimo vamos a representar los errores que hemos obtenido con los distintos modelos. En el eje vertical representamos el error, y en el eje horizontal representamos el hiper par?metro $\alpha$. La l?nea verde es el error que tenemos usando el modelo de regresi?n no penalizado, y la l?nea azul es el error de los modelos ridge en funci?n del valor de $\alpha$. Como podemos ver, usando un modelo de regresi?n ridge conseguimos un error de predicci?n menor que con un modelo no penalizado para valores de $\alpha$ menores de 80 (aproximadamente), y el valor ?ptimo de $\alpha$ con el que se obtiene el menor error es 26.48.



```python
import matplotlib.pyplot as plt

plt.plot(alpha, error_array)
plt.axhline(y=lm_error, color='g', linestyle='--')
```
