````python
import pandas as pd
import numpy as np

# Cargar la lista de estudiantes desde el archivo CSV
path = r"C:\Users\marco\Documentos\docencia\groups_list\numerico.xlsx"
estudiantes = pd.read_excel(path) 
estudiantes 
````

Definitivamente, el archivo de lista de estudiantes de la universidad de antioquia debe ser le?do por pandas.read_csv(path, sep = '\t') pues es un archivo donde la informaci?n est? separada por tabulaciones. 

````python
# Convertir la columna 'Nombre' a una lista de Python
lista_estudiantes = estudiantes['Nombre'].tolist()

# Funci?n para llamar a un estudiante aleatoriamente y registrar su nota de seguimiento
def asistencia():
    global lista_estudiantes
    if lista_estudiantes:
        estudiante_aleatorio = np.random.choice(lista_estudiantes)
        print(f"Llamando a {estudiante_aleatorio}...")
        asistencia = input(f"?Asisti? {estudiante_aleatorio}? (s/n): ").strip().lower()
        
        # Validaci?n de entrada
        if asistencia == 'n':
            # Obtener el ?ndice del estudiante en el DataFrame
            indice = estudiantes[estudiantes['Nombre'] == estudiante_aleatorio].index[0]
            # Incrementar su valor en la columna 'Parcial'
            estudiantes.at[indice, 'Asistencia'] += 4
            print(f"Actualizado: {estudiante_aleatorio} ahora tiene {estudiantes.at[indice, 'Asistencia']} en la columna 'Asistencia'.")
        
        elif asistencia != 's':
            print("Entrada inv?lida. Por favor, introduce 's' para s? o 'n' para no.")
            return  # Volver a preguntar para el mismo estudiante si la entrada es inv?lida
        
        # Remover al estudiante de la lista para no llamarlo de nuevo
        lista_estudiantes.remove(estudiante_aleatorio)
    else:
        print("Se han llamado a todos los estudiantes del grupo completo.")

# Funci?n para guardar cambios al final de la sesi?n
def guardar_cambios():
    estudiantes.to_csv(path, index=False)
    print("Cambios guardados en el archivo CSV.")

# Ejemplo de uso
while lista_estudiantes:
    asistencia()

# Guardar los cambios una vez que todos los estudiantes hayan sido llamados
guardar_cambios()

````

### 1. Concepto de integral definida y su interpretaci?n geom?trica


En lenguaje Python La integral definida de una funci?n continua $f(x)$ en un intervalo $[a, b]$ se denota como:  
$$
\int_{a}^{b} f(x)\,dx = \lim_{n \to \infty} \sum_{i = 0}^{n-1} f(x_{i})\Delta x 
$$


En la literatura de los textos de C?lculo La integral definida de una funci?n continua $f(x)$ en un intervalo $[a, b]$ se denota como:  

$$
\int_{a}^{b} f(x)\, dx = \lim_{n \to \infty} \sum_{i = 1}^{n} f(x_{i-1}) \Delta x 
$$


# Geom?tricamente  

* $x_{i}$ son puntos del dominio $[a,b]$ de la funci?n.
* $f(x_{i})$ son alturas o longitudes verticales de los puntos de la funci?n en $x_{i}$.

````python
# Visualizaci?n de partici?n regular izquierda de n?mero de subintervalos variable
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

# Definimos la funci?n a graficar (puedes modificarla)
def f(x):
    return np.sin(x) + 2

# Par?metros del intervalo [a, b]
a, b = 0, 2*np.pi

# Funci?n principal de visualizaci?n
def plot_particion(n=5):
    # Configuraci?n inicial
    plt.figure(figsize=(10, 5))
    x = np.linspace(a, b, 1000)
    y = f(x)
    
    # Graficar la funci?n
    plt.plot(x, y, 'b-', lw=2, label='f(x)')
    
    # Calcular puntos de la partici?n
    delta = (b - a)/n
    particion = [a + i*delta for i in range(n+1)]
    
    # Dibujar segmentos verticales rojos en x_{i-1}
    for i in range(n):
        x_prev = particion[i]
        y_prev = f(x_prev)
        plt.plot([x_prev, x_prev], [0, y_prev], 
                'r-', lw=2, alpha=0.7)
    
    # Marcar puntos de la partici?n
    plt.plot(particion, [0]*len(particion), 'ko', 
            markersize=5, label='Puntos de partici?n')
    
    # A?adir etiquetas y estilo
    plt.title(f'Partici?n regular de [{a}, {b:.2f}] con {n} subintervalos')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.ylim(0, None)
    plt.show()

# Crear widget interactivo
interact(plot_particion,
         n=IntSlider(value=5, min=1, max=50, 
                    description='Subintervalos',
                    style={'description_width': 'initial'}))

````

### 2. M?todo de rect?ngulos con extremos izquierdos



#### Definici?n y justificaci?n del m?todo


El m?todo de los rect?ngulos con extremos izquierdos es un procedimiento num?rico para aproximar integrales definidas mediante la suma de ?reas de rect?ngulos. 

Se basa en dividir el intervalo $[a, b]$ en $n$ subintervalos de igual longitud $\Delta x$:
$$


$$
\Delta x = \frac{b - a}{n}
$$


Cada subintervalo $[x_{i}, x_{i+1}]$ define un rect?ngulo cuya altura se eval?a en el extremo izquierdo $x_{i}$, de manera que la suma de las ?reas de estos rect?ngulos aproxima la integral:  

$$
\int_{a}^{b} f(x)\ dx = I \approx \sum_{i=0}^{n-1} f(x_i) \Delta x
$$



## Interpretaci?n geom?trica


Cada rect?ngulo tiene su base en $\Delta x$ y su altura en $f(x_i)$. 

* Cuando $f(x)$ es creciente en $[a, b]$, este m?todo proporciona una subestimaci?n del ?rea;  
* cuando $f(x)$ es decreciente, sobreestima la integral.



#### Aproximaci?n con diferentes cantidades de subintervalos


Al aumentar $n$, la precisi?n de la aproximaci?n mejora, ya que los rect?ngulos se ajustan mejor a la curva. 

En el l?mite, cuando $n \to \infty$, la aproximaci?n converge al valor exacto de la integral.
