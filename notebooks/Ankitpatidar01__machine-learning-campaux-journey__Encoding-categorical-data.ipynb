```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('customer.csv')
```

```python
df = df.iloc[:,2:]
```

```python
df.head()
```

```python
# now we split our test from our train data
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(df.iloc[:,0:2] , df.iloc[:,-1] , test_size=0.2)
```

```python
x_train
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG' , 'PG']])
# categories is a list which contain each column categorical data in order from ascedning to descending value assigned
#i.e poor should have least value and rich get greater number
# same as in pg school
```

```python
oe.fit(x_train)
```

```python
x_train_transformed = oe.transform(x_train)
x_test_transformed = oe.transform(x_test)
```
