```python
def predict():
    vf = pandas.read_csv("../vfl3-data.csv")
    vfdata = vf.set_index("Team", drop=False)
    match = pandas.read_csv("../vfl3-matches.csv")
    pattern = '[0-9,:,(,)]'
    match["Away"] = [re.sub(pattern, '', i) for i in match["Away"]]
    match["Home"] = [re.sub(pattern, '', i) for i in match["Home"]]
    Homematch = match.set_index("Home", drop=False)
    Awaymatch = match.set_index("Away", drop=False)
    Awaymatch = Awaymatch["Away"].str.strip()

    l = []
    for team in vfdata["Team"]:
        for home in Homematch["Home"]:
            if home == team:
                d = {}
                d["Team"] = home
                d["HTP"] = vfdata.loc[team,"HTP"]
                d["HGS"] = vfdata.loc[team,"HGS"]
                d["HGS Avg"] = vfdata.loc[team,"HGS"] / vfdata.loc[team,"HTP"]
                d["HGA"] = vfdata.loc[team,"HGA"]
                d["HGA Avg"] = vfdata.loc[team,"HGA"] / vfdata.loc[team,"HTP"]
                l.append(d)
    df = pandas.DataFrame(l)

    l = []
    for team in vfdata["Team"]:
        for away in Awaymatch:
            if away == team:
                d = {}
                d["Team"] = away
                d["ATP"] = vfdata.loc[team,"ATP"]
                d["AGS"] = vfdata.loc[team,"AGS"]
                d["AGS Avg"] = vfdata.loc[team,"AGS"] / vfdata.loc[team,"ATP"]
                d["AGA"] = vfdata.loc[team,"AGA"]
                d["AGA Avg"] = vfdata.loc[team,"AGA"] / vfdata.loc[team,"ATP"]
                l.append(d)
    adf = pandas.DataFrame(l)
    df = df.set_index("Team", drop=False)
    adf = adf.set_index("Team", drop=False)


    HTotP = df["HTP"].sum()
    HAVP = HTotP/len(df["Team"])
    HTotGS = df["HGS"].sum()
    HAVGS = HTotGS/len(df["Team"])
    HTotGSA = df["HGS Avg"].sum()
    HAVGSA = HTotGSA/len(df["Team"])
    HTotGA = df["HGA"].sum()
    HAVGA = HTotGA/len(df["Team"])
    HTotGAA = df["HGA Avg"].sum()
    HAVGAA = HTotGAA/len(df["Team"])

    ATotP = adf["ATP"].sum()
    AAVP = ATotP/len(adf["Team"])
    ATotGS = adf["AGS"].sum()
    AAVGS = ATotGS/len(adf["Team"])
    ATotGSA = adf["AGS Avg"].sum()
    AAVGSA = ATotGSA/len(adf["Team"])
    ATotGA = adf["AGA"].sum()
    AAVGA = ATotGA/len(adf["Team"])
    ATotGAA = adf["AGA Avg"].sum()
    AAVGAA = ATotGAA/len(adf["Team"])

    l = []
    for team in df["Team"]:
        d = {}
        d["Home"] = team
        d["Att"] = df.loc[team,"HGS Avg"] / HAVGSA
        d["Def"] = df.loc[team,"HGA Avg"] / HAVGAA
        l.append(d)
    home_final = pandas.DataFrame(l)

    l = []
    for team in adf["Team"]:
        d = {}
        d["Away"] = team
        d["Att"] = adf.loc[team,"AGS Avg"] / AAVGSA
        d["Def"] = adf.loc[team,"AGA Avg"] / AAVGAA
        l.append(d)
    away_final = pandas.DataFrame(l)
    home_final = home_final.set_index("Home", drop=False)
    away_final = away_final.set_index("Away", drop=False)

    l = []
    for home, away in zip(match["Home"],Awaymatch):
        d = {}
        d["Home"] = home
        d["HGX"] = home_final.loc[home,"Att"] * away_final.loc[away,"Def"] * df.loc[home,"HGS Avg"]
        d["Away"] = away
        d["AGX"] = away_final.loc[away,"Att"] * home_final.loc[home,"Def"] * adf.loc[away,"AGS Avg"]
        l.append(d)
    xG = pandas.DataFrame(l)
    predict.xG = xG.set_index("Home", drop=False)
    



def sinceawin():
    win_count = "0"
    loss_count = "0"
    day = 0
    for items in range(0,30):
        day = day + 1
        match = pandas.read_csv("../vfl3-matches.csv")
        pattern = '[0-9,:,(,)]'
        match["Away"] = [re.sub(pattern, '', i) for i in match["Away"]]
        match["Home"] = [re.sub(pattern, '', i) for i in match["Home"]]

        driver.switch_to.default_content()
        driver.switch_to.frame(driver.find_element_by_tag_name("iframe"))
        driver.switch_to.frame("vflframe")
        matchday = "matchday%s" % str(items+1)
        select_match = driver.find_elements_by_id(matchday)[0]
        driver.execute_script("return arguments[0].scrollIntoView();", select_match)
        select_match.click()
        time.sleep(2)
        driver.switch_to.default_content()
        driver.switch_to.frame(driver.find_element_by_tag_name("iframe"))
        print("Hold on...", end="\r")
        wait = WebDriverWait(driver, 60)
        time.sleep(1)
        wait.until(lambda x: x.find_element_by_class_name("matchlist-item-content"))
        soup = BeautifulSoup(driver.page_source, "html.parser")
        if len(soup.find("span",{"class":"state"}).text) == 0:
            teams = soup.find_all("div",{"class":"matchlist-item-teams"})
            select_teams = ["VFL Madrid","VFL Paris","VFL Amsterdam","VFL Berlin","VFL London","VFL Rome","VFL Oslo","VFL Stockholm"]
            idx = 0
            if day == int(matchday.replace("matchday","")):
                for newteams in teams:
                    unteam = newteams.text
                    team = re.split(' - ',unteam)
                    home = team[0].strip()
                    away = team[1].replace("+","").replace("9","").strip()
                    unteamer = teams[0].text
                    teamer = re.split(' - ',unteamer)
                    homer = teamer[0].strip()
                    awayer = teamer[1].replace("+","").replace("9","").strip()
                    while homer != match["Home"][0] and awayer != match["Away"][0]:
                        teams = soup.find_all("div",{"class":"matchlist-item-teams"})
                        unteamer = teams[0].text
                        teamer = re.split(' - ',unteamer)
                        homer = teamer[0].strip()
                        awayer = teamer[1].replace("+","").replace("9","").strip()
                        match = pandas.read_csv("../vfl3-matches.csv")
                        pattern = '[0-9,:,(,)]'
                        match["Away"] = [re.sub(pattern, '', i) for i in match["Away"]]
                        match["Home"] = [re.sub(pattern, '', i) for i in match["Home"]]
                        print("Haven't gotten match yet", end="\r")
                        time.sleep(10)
                    else:
                        try:
                            wait.until(lambda x: x.find_element_by_class_name("matchlist-item-market"))
                            a = 0
                            print(matchday)
                            soup = BeautifulSoup(driver.page_source,"html.parser")
                            if day == int(matchday.replace("matchday","")):
                                content = soup.find_all("div",{"class":"matchlist-item-content"})[idx].find("div",{"class":"matchlist-item-markets"}).find_all("div",{"class":"matchlist-item-market"})
                                for items in content:
                                    if items.find("div",{"class":"matchlist-market-name"}).text == "  Over/under (2.5)  ":
                                        loc = a
                                        break
                                    a = a + 1
                                for step in select_teams:
                                    ##if idx == 8:
                                        ##idx = idx - 1

                                    for newstep in select_teams:
                                        if day == int(matchday.replace("matchday","")):
                                            if step == home and newstep == away or step == away and newstep == home:
                                                predict()
                                                if predict.xG.loc[home,"HGX"] >= 1.7 and predict.xG.loc[home,"AGX"] >=1.7 or predict.xG.loc[home,"HGX"] > 2.4 or predict.xG.loc[home,"AGX"] > 2.4:
                                                    select = driver.find_elements_by_class_name("matchlist-item-content")[idx].find_elements_by_class_name("matchlist-item-markets")[0].find_elements_by_class_name("matchlist-item-market")[loc].find_elements_by_class_name("matchlist-betting-options")[0].find_elements_by_class_name("btnOdd")[1]
                                                    driver.execute_script("return arguments[0].scrollIntoView();", select)
                                                    select.click()
                                                    time.sleep(1)
                                                    driver.switch_to.default_content()
                                                    driver.switch_to.frame(driver.find_element_by_tag_name("iframe"))
                                                    driver.execute_script("window.scroll(0, 0);")
                                                    time.sleep(1)
                                                    stake_amt = driver.find_elements_by_class_name("bs-stake-input")
                                                    si = 100
                                                    stake_amt[0].send_keys(Keys.CONTROL + "a")
                                                    stake_amt[0].send_keys(Keys.DELETE)
                                                    stake_amt[0].send_keys(si)
                                                    time.sleep(2)
                                                    stake = driver.find_elements_by_class_name("bs-stake")
                                                    stake[0].click()
                                                    time.sleep(3)
                                                    print("Sleeping...")
                                                    driver.switch_to.default_content()
                                                    driver.switch_to.frame(driver.find_element_by_tag_name("iframe"))
                                                    time.sleep(5)
                                                    break

                                            else:
                                                print("Not Found", end="\r")


                        except Exception as e:
                            print(e)
                            time.sleep(2)
                            break
                        idx = idx + 1
        else:
            print("Not yet...", end="\r")
            continue
            
            
    driver.switch_to.default_content()
    driver.switch_to.frame(driver.find_element_by_tag_name("iframe"))    
```
