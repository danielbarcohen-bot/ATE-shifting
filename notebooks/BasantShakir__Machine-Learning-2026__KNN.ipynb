```python
#KNN
import warnings
warnings.filterwarnings('ignore')

# 1?? ??????? ????????
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
```

```python

# 2?? ????? ????????
data = {
    'Hours_Study': [2, 3, 4, 6, 8],
    'Sleep_Hours': [9, 6, 8, 5, 3],
    'Result': [0, 0, 1, 1, 1]
}
df = pd.DataFrame(data)
```

```python
# 3?? ????? ?????? ???????
X = df[['Hours_Study', 'Sleep_Hours']]
y = df['Result']
```

```python
# 4?? ????? ???????
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X, y)
```

```python
# 5?? ?????? ???? ?????
test = [[5, 7]]
prediction = knn.predict(test)
print("Predicted Class (1=Pass, 0=Fail):", prediction)
```

```python
# 1?? ??????? ????????

from sklearn.naive_bayes import CategoricalNB
from sklearn.preprocessing import LabelEncoder
```

```python
# 2?? ????????
df = pd.DataFrame({
    'Weather': ['Sunny', 'Sunny', 'Rainy', 'Rainy', 'Sunny'],
    'Play': ['No', 'No', 'Yes', 'Yes', 'Yes']
})
```

```python

# 3?? ????? ????? ?????? ??? ?????
le_weather = LabelEncoder()
le_play = LabelEncoder()

df['Weather_enc'] = le_weather.fit_transform(df['Weather'])
df['Play_enc'] = le_play.fit_transform(df['Play'])
```

```python

# 4?? ????? ???????
X = df[['Weather_enc']]
y = df['Play_enc']
model = CategoricalNB()
model.fit(X, y)
```

```python
# 5?? ?????? ???? ?????
test = [[le_weather.transform(['Sunny'])[0]]]
pred = model.predict(test)
print("Prediction:", le_play.inverse_transform(pred))
```
