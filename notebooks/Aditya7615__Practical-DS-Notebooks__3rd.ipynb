# Machine Learning :
is a subset of artificial intelligence that focuses on the development of algorithms and statistical models that enable computers to perform tasks without explicit instructions. It involves the use of data to train models, allowing them to make predictions or decisions based on new, unseen data. Machine learning has various applications, including image recognition, natural language processing, and recommendation systems.

Common Techniques in Machine Learning:
1. Supervised Learning: Involves training a model on labeled data, where the correct output is known. The model learns to map inputs to outputs and can make predictions on unseen data.
2. Unsupervised Learning: Involves training a model on unlabeled data, where the correct output is not known. The model learns to identify patterns and structures in the data.
3. Reinforcement Learning: Involves training a model to make decisions by rewarding desired behaviors and penalizing undesired ones. The model learns to maximize cumulative rewards over time.

## Data Cleaning :
Data cleaning is the process of identifying and correcting or removing errors, inconsistencies, and inaccuracies in a dataset to improve its quality and reliability. It is an essential step in data preprocessing before analysis or modeling.
### Common Steps in Data Cleaning:
1. Handling Missing Values: This involves identifying and addressing missing data points in the dataset. Common techniques include removing rows or columns with missing values, imputing missing values using statistical methods (e.g., mean, median, mode), or using machine learning algorithms to predict missing values.
2. Removing Duplicates: This involves identifying and removing duplicate records in the dataset to ensure that each observation is unique.
3. Correcting Inconsistencies: This involves identifying and correcting inconsistencies in the dataset, such as inconsistent formatting, spelling errors, or incorrect data types.
4. Handling Outliers: This involves identifying and addressing outliers in the dataset, which are data points that are significantly different from the rest of the data. Common techniques include removing outliers, transforming data, or using robust statistical methods.
5. Standardizing Data: This involves transforming data to a common format or scale to ensure consistency and comparability across different variables.
6. Validating Data: This involves checking the accuracy and completeness of the dataset to ensure that it meets the requirements for analysis or modeling.

## missing value handeling :
Missing value handling is an important step in data preprocessing, as missing data can lead to biased or inaccurate results in analysis or modeling. Here are some common techniques for handling missing values:
1. Removing Missing Values: This involves removing rows or columns with missing values from the dataset. This technique is simple but can lead to loss of valuable information if a significant portion of the data is missing.
2. Imputing Missing Values: This involves replacing missing values with estimated values based on other data points in the dataset. Common techniques include:
   - Mean/Median/Mode Imputation: Replacing missing values with the mean, median, or mode of the non-missing values in the same column.
   - Regression Imputation: Using regression models to predict missing values based on other variables in the dataset.
   - K-Nearest Neighbors (KNN) Imputation: Using the values of the k-nearest neighbors to estimate missing values.
3. Using Indicator Variables: This involves creating a new binary variable that indicates whether a value is missing or not. This technique can help preserve information about the missingness of the data.
4. Using Advanced Techniques: This involves using more advanced techniques such as multiple imputation, which involves creating multiple datasets with different imputed values and combining the results, or using machine learning algorithms to predict missing values based on other variables in the dataset.

# Encoding :
Encoding is the process of converting categorical data into a numerical format that can be used by machine learning algorithms. Categorical data is data that can be divided into distinct categories or groups. Encoding is essential because most machine learning algorithms work with numerical data and cannot directly process categorical data.
### Common Encoding Techniques:
1. One-Hot Encoding: This technique creates a new binary variable for each category in the categorical variable. For example, if a categorical variable has three categories (A, B, C), one-hot encoding would create three new binary variables (is_A, is_B, is_C) that indicate whether each observation belongs to that category.
2. Label Encoding: This technique assigns a unique integer to each category in the categorical variable. For example, if a categorical variable has three categories (A, B, C), label encoding would assign the values 0, 1, and 2 to each category, respectively.
3. Ordinal Encoding: This technique is used for categorical variables that have a natural order or ranking. It assigns a unique integer to each category based on its order. For example, if a categorical variable has three categories (Low, Medium, High), ordinal encoding would assign the values 0, 1, and 2 to each category, respectively.
4. Binary Encoding: This technique converts each category into a binary representation. It is useful for categorical variables with a large number of categories, as it reduces the dimensionality of the data.
5. Target Encoding: This technique replaces each category with the mean of the target variable for that category. It is useful for categorical variables with a large number of categories and can help capture the relationship between the categorical variable and the target variable.

### Outlier :
An outlier is a data point that is significantly different from the other data points in a dataset. Outliers can be caused by
measurement errors, data entry errors, or natural variability in the data. Outliers can have a significant impact on
the results of statistical analyses and machine learning models, so it is important to identify and handle them appropriately.
### Methods to handle outliers:
1. Z-Score Method: The Z-score method is a statistical technique used to identify outliers in a dataset. It measures how many standard deviations a data point is from the mean of the dataset. A Z-score greater than 3 or less than -3 is typically considered an outlier.
2. IQR Method: The IQR method is a statistical technique used to identify outliers in a dataset. It measures the spread of the data by calculating the interquartile range (IQR), which is the difference between the first quartile (Q1) and the third quartile (Q3). Data points that fall below Q1 - 1.5 * IQR or above Q3 + 1.5 * IQR are typically considered outliers.

```python
data_a = pd.read_csv(r"/Users/adityagoyal/Downloads/Dataset/loan_data_set.csv")
data_a.head(3)
```

```python
q1 = data_a["CoapplicantIncome"].quantile(0.25)
q3 = data_a["CoapplicantIncome"].quantile(0.75)
```

IQR method to remove outliers

```python
IQR = q3 - q1
lower_bound = q1 - 1.5 * IQR
upper_bound = q3 + 1.5 * IQR
lower_bound, upper_bound
```

# Feature Scaling :
is a technique used to standardize the range of independent variables or features of data. In machine learning, it is important to scale features so that they have a similar range, as this can improve the performance of algorithms that are sensitive to the scale of the data.
### 1. Standardization :
Standardization is a technique used to scale features so that they have a mean of 0 and a standard deviation of 1. This is done by subtracting the mean of the feature from each data point and then dividing by the standard deviation.
### 2. Normalization :
Normalization is a technique used to scale features so that they have a range of 0 to 1. This is done by subtracting the minimum value of the feature from each data point and then dividing by the range (maximum value - minimum value).
### 3. Min-Max Scaling :
Min-Max scaling is a technique used to scale features so that they have a range of 0 to 1. This is done by subtracting the minimum value of the feature from each data point and then dividing by the range (maximum value - minimum value).
### 4. Robust Scaling :
Robust scaling is a technique used to scale features so that they are less sensitive to outliers. This is done by subtracting the median of the feature from each data point and then dividing by the interquartile range (IQR).
### 5. MaxAbs Scaling :
MaxAbs scaling is a technique used to scale features so that they have a range of -1 to 1. This is done by dividing each data point by the maximum absolute value of the feature.
### 6. Log Transformation :
Log transformation is a technique used to scale features that have a skewed distribution. This is done by taking the logarithm of each data point.
### 7. Power Transformation :
Power transformation is a technique used to scale features that have a skewed distribution. This is done by raising each data point to a power.
### 8. Quantile Transformation :
Quantile transformation is a technique used to scale features so that they have a uniform distribution. This is done by mapping each data point to its corresponding quantile in a uniform distribution.
