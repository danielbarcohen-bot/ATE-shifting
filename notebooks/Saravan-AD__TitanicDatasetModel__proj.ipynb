```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data=pd.read_csv('train.csv')
data_x=data.drop(columns='Survived')
data_y=data['Survived']
```

```python
print(data_x)
print(data_y)
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(data_x,data_y,test_size=0.24,random_state=1)
```

```python
X_train['Age'] = X_train.groupby(['Pclass', 'Sex'])['Age'].transform(
    lambda x: x.fillna(x.median())
)
X_train['Deck'] = X_train['Cabin'].apply(lambda x: str(x)[0] if pd.notnull(x) else np.nan)
print(X_train['Deck'].value_counts(dropna=False))
X_train['Deck'] = X_train['Deck'].fillna('U')
X_train['Embarked']=X_train['Embarked'].fillna(X_train['Embarked'].mode()[0])
print(X_train['Embarked'].value_counts(dropna=False))

y_train=y_train.values
print(y_train)
```

```python
X_train = X_train[['Pclass','Sex','Age','SibSp','Parch','Deck','Embarked']]
print(X_train)
```

```python
# print(x[27])
# print(y)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

ct=ColumnTransformer(transformers=[('encoder',OneHotEncoder(handle_unknown='ignore'),['Sex', 'Deck','Embarked'])],remainder='passthrough')
X_train=np.array(ct.fit_transform(X_train))
```

```python
np.set_printoptions(threshold=np.inf)  # inf = no truncation


# Now print your DataFrame
print(X_train)
```

```python
from sklearn.preprocessing import StandardScaler

sc=StandardScaler()
X_train[:,14:16]=sc.fit_transform(X_train[:,14:16])
```

```python
from sklearn.linear_model import LogisticRegression

model1=LogisticRegression(random_state=0)
model1.fit(X_train,y_train)
```

```python
X_test['Age'] = X_test.groupby(['Pclass', 'Sex'])['Age'].transform(
    lambda x: x.fillna(x.median())
)
X_test['Deck'] = X_test['Cabin'].apply(lambda x: str(x)[0] if pd.notnull(x) else np.nan)
print(X_test['Deck'].value_counts(dropna=False))
X_test['Deck'] = X_test['Deck'].fillna('U')
X_test['Embarked']=X_test['Embarked'].fillna(X_test['Embarked'].mode()[0])
print(X_test['Embarked'].value_counts(dropna=False))


```

```python
X_test=np.array(ct.transform(X_test))
X_test[:,14:16]=sc.transform(X_test[:,14:16])
```

```python
y_test=y_test.values
print(y_test)
```

```python
print(model1.predict(X_test))
```

```python
predictions = model1.predict(X_test)

# Ensure y_test is a numpy array (it might already be from train_test_split)
y_test_array = y_test if isinstance(y_test, np.ndarray) else np.array(y_test)

# Method 1: Using numpy column_stack
comparison_array = np.column_stack((predictions, y_test_array))
print(comparison_array)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score
cm = confusion_matrix(y_test, predictions)
print("Confusion Matrix:")
print(cm)

print("\n" + "="*50)

# Method 1: Basic confusion matrix with labels
print("\nConfusion Matrix with Labels:")
print("                Predicted")
print("                0    1")
print(f"Actual    0    {cm[0,0]:3d}  {cm[0,1]:3d}")
print(f"          1    {cm[1,0]:3d}  {cm[1,1]:3d}")

print("\n" + "="*50)
```

```python
accuracy = accuracy_score(y_test, predictions)
print(accuracy)
```
