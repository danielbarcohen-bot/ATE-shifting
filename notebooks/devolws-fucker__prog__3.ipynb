```python
# ??????????? ??????????? ??????????

# Pandas ???????????? ??? ?????? ? ??????? ? ????????? ???????
import pandas as pd

# ??????????? ??????? ??? ?????????? ?????? ?? ????????? ? ???????? ???????, ? ????? ??? ?????? ??????????????? ??????
from sklearn.model_selection import train_test_split, GridSearchCV

# ??????????? ?????? ??? ??????????? ?????????????? ?????? ? ?????????????? ?????????
from sklearn.preprocessing import LabelEncoder, StandardScaler

# ??????????? ??????? ??? ?????? ???????? ??????
from sklearn.metrics import classification_report, accuracy_score

# ??????????? ?????? ??? ????????????? ? ???????? ?????????
from sklearn.linear_model import LogisticRegression, LinearRegression

# ??????????? ??????? ??? ?????????? ?????????????????? ?????? (MSE)
from sklearn.metrics import mean_squared_error
```

```python
# ?????? ?????? ?? CSV-????? 'weather.csv' ? ??????? DataFrame ? ??????? pandas
df = pd.read_csv('weather.csv', encoding='utf-8')

# ??????????? ???????? ??????? 'RainToday' ?? ????? 'Yes' ? 'No' ? ?????????? ???????? True ? False ??????????????
df['RainToday'] = df['RainToday'].map({'Yes': True, 'No': False}).astype(bool)

# ??????????? ???????? ??????? 'RainTomorrow' ?? ????? 'Yes' ? 'No' ? ?????????? ???????? True ? False ??????????????
df['RainTomorrow'] = df['RainTomorrow'].map({'Yes': True, 'No': False}).astype(bool)

# ??????? ?????? 5 ????? DataFrame ??? ???????????? ? ???????
print(df.head(5))

# ??????? ???? ?????? ??????? ??????? DataFrame
print(df.dtypes)
```

```python
# ?????????? ?????? ????????, ??????? ????? ??????? ?? DataFrame
columns_to_drop = ['Humidity3pm', 'Pressure3pm', 'Cloud3pm', 'Temp3pm', 'WindDir9am']

# ??????? ????????? ??????? ?? DataFrame
df = df.drop(columns=columns_to_drop, axis=1)

# ??????? ?????? 5 ????? ???????????? DataFrame ??? ???????? ?????????
print(df.head(5))
```

```python
# ??????? ?????? ? ?????????????? ?????????? ?? DataFrame
df.dropna(inplace=True)
```

```python
# ??????? ??????? ??? ???????? ???????? LabelEncoder ??? ??????? ??????????????? ????????
label_encoders = {}

# ???????? ?? ???? ???????? DataFrame, ??????? ????? ??? ?????? 'object' (?????????????? ????????)
for column in df.select_dtypes(include=['object']).columns:
    # ??????? ????? LabelEncoder ??? ???????? ???????
    label_encoders[column] = LabelEncoder()
    
    # ??????????? ?????????????? ???????? ? ???????? ? ????????? ?? ??????? ? DataFrame
    df[column] = label_encoders[column].fit_transform(df[column])

# ??????? ?????? 5 ????? DataFrame ??? ???????? ?????????
print(df.head(5))
```

```python
# ??????? ?????? StandardScaler ??? ??????????????? ???????? ?????????
scaler = StandardScaler()

# ???????? ??? ???????, ??????? ????? ???????? ???? ?????? (int32, int64, float32, float64)
numeric_features = df.select_dtypes(include=['int32', 'int64', 'float32', 'float64']).columns

# ????????? ??????????????? ?? ???? ???????? ????????? ? ????????? ?? ??????? ? DataFrame
df[numeric_features] = scaler.fit_transform(df[numeric_features])

# ??????? ?????? 5 ????? DataFrame ??? ???????? ?????????
print(df.head(5))
```

```python
# ????????? DataFrame ?? ??????? ????????? X ? ?????? ??????? ?????????? Y
X = df.drop('RainTomorrow', axis=1)  # ??????? ??????? 'RainTomorrow' ? ????????? ??? ????????? ??????? ? X
Y = df['RainTomorrow']               # ????????? ??????? 'RainTomorrow' ? ???????? ??????? ?????????? Y

# ????????? ?????? ?? ????????? ? ???????? ??????
# test_size=0.2 ????????, ??? 20% ?????? ????? ???????????? ??? ????????????, ? 80% ??? ????????
# random_state=42 ???????????? ??? ????????????????? ???????????
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
# ???????? ?????? ???????? ?????????
linear_model = LinearRegression()
```

```python
# ?????????? ????? ?????????? ??? ?????? ??????????? ?????????? ?????? ???????? ?????????
param_grid_linear = {
    'fit_intercept': [True, False],  # ????????, ???????????, ?????? ?? ?????? ????????? ???????? (????????)
    'copy_X': [True, False],          # ????????, ???????????, ??????? ?? ?????????? ??????? X (??? ????????????? inplace-????????)
    'positive': [True, False],        # ????????, ???????????, ?????? ?? ???????????? ???? ????????????????
}
```

```python
# ????? ??????????? ?????????? ? ?????????????? ?????-?????????
grid_search_linear = GridSearchCV(linear_model, param_grid_linear, cv=5)
grid_search_linear.fit(X_train, Y_train)
```

```python
# ????????? ?????? ??????????
best_params_linear = grid_search_linear.best_params_
print("?????? ?????????:", best_params_linear)
```

```python
# ???????? ?????? ? ??????? ???????????
best_svm_linear = grid_search_linear.best_estimator_
best_svm_linear.fit(X_train, Y_train)
```

```python
# ???????????? ?? ???????? ?????? ??????
Y_pred = best_svm_linear.predict(X_test)
```

```python
# ?????? ?????? ?? ???????? ?????? ??????
mse_linear = mean_squared_error(Y_test, Y_pred)
print("?????????????????? ?????? ??????:", mse_linear)
```
