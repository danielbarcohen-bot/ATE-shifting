## <strong>`OneHotEncoder`</strong>

- O conceito por tr?s do OneHotEncoding ? produzir um ?nico atributo de sa?da com uma matriz densa a partir de atributos num?ricos. O OneHotEncoding espera atributos num?ricos e pode ser utilizado o StringIndexer para transform?-los.
- OneHotEncoding ? uma t?cnica de processamento de dados que produz um ?nico atributo de sa?da com uma matriz densa a partir de n atributos num?ricos. Por exemplo, se tivermos um atributo de entrada que pode ter tr?s valores diferentes 'A', 'B', 'C', o OneHotEncoder criar? tr?s atributos bin?rios correspondentes. No c?digo a seguir, est? sendo utilizado o **`pyspark`**, uma biblioteca para processamento de dados distribu?dos em Python, para criar uma sess?o do Spark. Primeiramente, a biblioteca **`findspark`** ? importada para permitir o uso do Spark em um ambiente Python sem a necessidade de configurar manualmente as vari?veis de ambiente. Em seguida, ? importada a biblioteca **`pyspark`.sql** para acessar as funcionalidades relacionadas ao SQL no Spark. A fun??o **`init()`** do **findspark** ? chamada para inicializar o Spark, e ent?o ? criada uma inst?ncia da classe **`SparkSession`** utilizando a fun??o **`builder`**. O nome da aplica??o ? definido como "onehot" atrav?s da fun??o **`appName`()**, e a sess?o do Spark ? obtida ou criada atrav?s da fun??o **`getOrCreate`()**. Essa sess?o do Spark pode ent?o ser utilizada para realizar opera??es de processamento de dados distribu?dos.

```python
import findspark, pyspark
from pyspark.sql import SparkSession

findspark.init()
spark = SparkSession.builder.appName("onehot").getOrCreate()
```

No c?digo a seguir, est? sendo importada a biblioteca **pyspark.ml.feature** e a fun??o **`OneHotEncoder`** dela. Essa fun??o ? utilizada para transformar vari?veis categ?ricas em vetores de caracter?sticas bin?rias, um processo comum em an?lise de dados e machine learning. O **OneHotEncoder** cria uma coluna para cada categoria ?nica na vari?vel categ?rica, atribuindo o valor 1 ? categoria presente e 0 ?s demais, permitindo que algoritmos de aprendizado de m?quina entendam e processem esses dados categ?ricos de forma eficiente.

```python
from pyspark.ml.feature import OneHotEncoder
```

No c?digo a seguir, est? sendo carregado um conjunto de dados chamado "Churn.csv" utilizando a **biblioteca** **`Spark`** em **`Python`**. O m?todo ``load`` ? utilizado para carregar o arquivo CSV, especificando o formato como CSV, o separador como ponto e v?rgula, e inferindo o esquema dos dados. A fun??o ``show`` ? ent?o chamada para exibir as primeiras cinco linhas do conjunto de dados carregado.

```python
churn = spark.read.load("../Churn.csv", format="csv", sep=";", inferSchema=True, header=True)
churn.show(5)
```

No c?digo a seguir, est? sendo utilizado o **`pyspark`**, uma biblioteca para processamento de big data em Python, especificamente o m?dulo de aprendizado de m?quina. Primeiro, ? importado o **`StringIndexer`** da biblioteca **pyspark.ml.feature**. Em seguida, duas vari?veis s?o criadas: **indice** e **indiceonehot**. O **StringIndexer** ? aplicado para transformar duas colunas do conjunto de dados, **Geography** e **Gender**, em ?ndices num?ricos, com as sa?das armazenadas nas colunas **Indexer_c1** e **Indexer_c2** respectivamente. Isso ? ?til para algoritmos de aprendizado de m?quina que exigem entradas num?ricas. O m?todo **`fit`** ? usado para ajustar o **StringIndexer** ao conjunto de dados, e **`transform`** ? usado para aplicar a transforma??o. O resultado ? uma nova estrutura de dados com as colunas originais substitu?das pelos ?ndices num?ricos correspondentes.

```python
from pyspark.ml.feature import StringIndexer

indice = StringIndexer(inputCol="Geography", outputCol="Indexer_c1")
indiceonehot = indice.fit(churn).transform(churn)

indice = StringIndexer(inputCol="Gender", outputCol="Indexer_c2")
indiceonehot = indice.fit(indiceonehot).transform(indiceonehot)
```

No c?digo a seguir, est? sendo utilizado o m?todo **`select`** do objeto **indiceonehot** para selecionar as colunas **Indexer_c1** duas vezes e ent?o exibir o resultado usando o m?todo **`show`()**. Este c?digo provavelmente faz parte de um processo de transforma??o de dados em um contexto de aprendizado de m?quina, onde o objetivo pode ser visualizar o resultado da codifica??o one-hot de uma ou mais colunas do conjunto de dados.

```python
indiceonehot.select("Indexer_c1","Indexer_c1").show()
```

No c?digo a seguir, est? sendo utilizado o **`OneHotEncoder`** para transformar vari?veis categ?ricas em vetores bin?rios. Primeiro, ? criado um objeto **onehot** que especifica as colunas de entrada e sa?da para a codifica??o one-hot. Em seguida, ? criado um modelo utilizando a fun??o **`fit`** com os dados de entrada **indiceonehot**. Ap?s o ajuste do modelo, ? aplicada a transforma??o utilizando a fun??o **`transform`**, e os resultados s?o armazenados na vari?vel **onehotout**. Por fim, ? selecionado um subconjunto das colunas codificadas utilizando o m?todo **`select`** e exibido na tela utilizando o m?todo **`show`**. Este c?digo ? ?til para converter vari?veis categ?ricas em um formato adequado para serem utilizadas em algoritmos de aprendizado de m?quina que requerem entradas num?ricas, como regress?o log?stica ou redes neurais.

```python
onehot = OneHotEncoder(inputCols=["Indexer_c1","Indexer_c1"], outputCols=["onehot_c1","onehot_c2"])
modelo = onehot.fit(indiceonehot)
onehotout = modelo.transform(indiceonehot)

onehotout.select("onehot_c1","onehot_c2").show(truncate=False)
```
