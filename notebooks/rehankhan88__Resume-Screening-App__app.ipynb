Clean Data

```python
import re 
def cleanResume(txt):
    cleanTxt = re.sub('http\s+\s',' ',txt)
    cleanTxt = re.sub('RT|cc',' ',cleanTxt)
    cleanTxt = re.sub('#\s+\s',' ',cleanTxt)
    cleanTxt = re.sub('@\s+',' ',cleanTxt)
    cleanTxt = re.sub('[%s]' % re.escape('''!"#$%&"()*+,-./:;<=>?@[\]^_`{|}~'''),' ',cleanTxt)
    cleanTxt = re.sub(r'[^\x00-\x7f]',' ',cleanTxt)
    cleanTxt = re.sub('\s+',' ',cleanTxt)
    
    return cleanTxt
```

```python
cleanResume('my ###hkhj website like @jdklsjds is this http://helowords and access it')
```

```python
df['Resume'] = df['Resume'].apply(lambda x: cleanResume(x))
```

Vectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(stop_words='english')

tfidf.fit(df['Resume'])
requredText = tfidf.transform(df['Resume'])
```

Splitting

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(requredText, df['Category'], test_size=0.2, random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

Train the model and print the classification

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.multiclass import OneVsOneClassifier
from sklearn.metrics import accuracy_score

clf = OneVsOneClassifier(KNeighborsClassifier())
clf.fit(x_train,y_train)
ypred = clf.predict(x_test)
print(accuracy_score(y_test,ypred))  
```
