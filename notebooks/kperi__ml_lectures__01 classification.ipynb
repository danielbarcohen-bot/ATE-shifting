```python
from sklearn import datasets
from sklearn.datasets import fetch_openml


import matplotlib.pyplot as plt
%matplotlib inline

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler 

from sklearn.pipeline import Pipeline, FeatureUnion
from sklearn.model_selection import GridSearchCV
```

```python
X, y = fetch_openml('mnist_784', version=1, return_X_y=True)
```

```python
example = X[0].reshape(28,28)
plt.imshow( example , cmap='gray', vmin=0) 
y[0]
```

## Steps 

- Feature engineering
- Train/Validation/Test splits
- Fit the model 
- Evaluate the model 

```python
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.50, random_state=42)
X_train.shape, y_train.shape
```

```python
clf = LogisticRegression(max_iter=50, solver='sag')
pipeline = Pipeline([("scale", StandardScaler()), ("clf", clf)])
pipeline
```

```python
pipeline.fit(X_train, y_train)
```

```python
ret = pipeline.predict(X_test)
accuracy_score( y_test, ret )
```
