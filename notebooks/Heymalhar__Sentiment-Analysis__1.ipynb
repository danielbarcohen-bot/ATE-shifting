```python
# Importing the dependencies

import numpy as np
import pandas as pd
import re
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
# Downloading the stopwords

import nltk
nltk.download('stopwords')
```

```python
# Printing the stopwords in English

print(stopwords.words('english'))
```

Data Processing

```python
# Making the code read the columns, or naming the columns

column_names = ['target', 'id', 'date', 'flag', 'user', 'text']

data = pd.read_csv('Data.csv', names=column_names, encoding='ISO-8859-1')

data.head()
```

```python
# Counting the missing values in the dataset

data.isnull().sum()
```

```python
# Checking the distribution of the target column
# Basically checks the number or occurence of distinct values in a particular columns

data['target'].value_counts()
```

```python
# Converting the label '4' to '1'
# Therefore, 0 is negative and 1 is positive

data.replace({'target':{4:1}}, inplace=True)

data['target'].value_counts()

data.tail()
```

0 ----> Negative Tweet

1 ----> Positive Tweet

Stemming = the process of reducing a word to its rootword.

Eg: Actors, Actress, Acting = Act

```python
port_stem = PorterStemmer()
```

```python
# Function for stemming

def stemming(content):

    # Removing non-alphabetic characters
    stemmed_content = re.sub('[^a-zA-Z]', ' ', content)

    # Converting to lower case
    stemmed_content = stemmed_content.lower()

    # Creating a list
    stemmed_content = stemmed_content.split()

    # Removing the stopwords
    stemmed_content = [port_stem.stem(word) for word in stemmed_content if not word in stopwords.words('english')]

    # Joining
    stemmed_content = ' '.join(stemmed_content)

    return stemmed_content
```

```python
data['stemmed_content'] = data['text'].apply(stemming)
```

```python
data.head()
```

```python
print(data['stemmed_content'])
```

```python
# Separating the data and the label

X = data['stemmed_content'].values
Y = data['target'].values

print(X)

print(Y)
```

Splitting the data into training and testing data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=2)
```

```python
print(X.shape)
print(X_train.shape)
print(X_test.shape)
```

```python
print(Y.shape)
print(Y_train.shape)
print(Y_test.shape)
```

```python
# Converting the textual data to numerical data

vectorizer = TfidfVectorizer()

X_train = vectorizer.fit_transform(X_train)
X_test = vectorizer.transform(X_test)

print(X_train)
print(X_test)
```

```python
# Training the Machine Learning model, which is Logistic Regression

model = LogisticRegression(max_iter=1000)
```

```python
model.fit(X_train, Y_train)
```

Model Evaluation

```python
# Accuracy score on the training data

X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)

print("The accuracy score on the training data is: " ,training_data_accuracy)
```

```python
# Accuracy score on the test data

X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(Y_test, X_test_prediction)

print("Accuracy on test data: ", test_data_accuracy)
```

The final model accuracy is 77.8%
