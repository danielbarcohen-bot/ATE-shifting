## Agregar Encabezados

Hecha un vistazo a nuestro dataset. Pandas autom?ticamente ajusta el encabezado con un integer que comienza en 0.

Para describir de mejor manera nuestra informaci?n, podemos introducir encabezados. Esta informaci?n esta disponible en: https://archive.ics.uci.edu/ml/datasets/Automobile.
Por lo tanto, debemos agregarla de forma manual.

Primero, creamos una lista "headers" que incluye el nombre de todas las columnas en orden. Luego, usamos `dataframe.columns = headers` para reemplazar los por la lista que creamos.

```python
# En el lab original ya estaban separadas y ordenadas como correspond?an, pero yo no quer?a copiar y pegar y me dio pereza abrir y cerrar comillas asi que lo hice de esta manera.

headers = "symboling, normalized-losses, make, fuel-type, aspiration, num-of-doors, body-style, drive-wheels, engine-location, wheel-base, length, width, height, curb-weight, engine-type, num-of-cylinders, engine-size, fuel-system, bore, stroke, compression-ratio, horsepower, peak-rpm, city-mpg, highway-mpg, price"

headers=[header.strip() for header in headers.split(",")]

print(headers)
```

Reemplazamos los headers y revisamos nuestro dataframe.

```python
df.columns = headers
df.head(10)
```

Necesitamos reemplazar el s?mbolo *`"?"`* por NaN as? que dropna() remover? los valores faltantes:

```python
df1 = df.replace('?',np.nan)
```

Podemos remover los valores faltantes de la columna "price" de la siguiente manera:

```python
df = df1.dropna(subset=['price'], axis=0)
df.head(20)
```

Ahora, hemos le?do exitosamente nuestro dataset en bruto y agregamos correctamente los encabezados en el dataframe.
