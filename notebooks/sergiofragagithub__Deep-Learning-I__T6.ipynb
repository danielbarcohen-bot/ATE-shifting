# Classifica??o

Vamos come?ar fazendo um modelo simples de classifica??o e entender a forma geral de desenvolvimento de modelos do tipo sequencial e as classes empregadas.

Para esse exemplo vamos criar um conjunto de dados simples com o `make_moons` do `Scikit-learn`.

## Dados

````
from sklearn.datasets import make_moons

X, y = make_moons(noise=0.352, random_state=1, n_samples=100)

sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=y, marker='o', s=50, edgecolor='k', legend=False).set_title("Data")
plt.show()
````

### Train e Test Split

````
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=123)
````

## Import `torch`

Esses s?o os imports essenciais para iniciarmos o nosso modelo.

````
import torch
import torch.nn as nn
from torch.autograd import Variable
````

## Definindo o Modelo

Um modelo ? definido como uma extens?o da classe `Module`. Os m?todos principais s?o o m?todo construtor da classe `__init__` que inicializa os elementos da rede e o m?todo `forward`,  

````
class Model(nn.Module):
    def __init__(self, input_dim):
        super(Model, self).__init__()
        self.layer1 = nn.Linear(input_dim, 50)
        self.layer2 = nn.Linear(50, 50)
        self.layer3 = nn.Linear(50, 2)
        
    def forward(self, x):
        x = torch.relu(self.layer1(x))
        x = torch.relu(self.layer2(x))
        # x = torch.sigmoid(self.layer3(x))        # voc? pode querer empregar a sigmoid ou a softmax
        x = torch.softmax(self.layer3(x), dim=1)
        return x
````

`Linear` implementa uma camada densa e tem como par?metros b?sicos o n?mero de elementos de entrada e sa?da,

```
torch.nn.Linear(in_features, out_features, bias=True)
```
E ir? aplicar a transforma??o:

$$ y = x A^T + b $$

sobre os dados.

$\rightarrow$ Importante: esse m?dulo suporta tipos `TensorFloat32`.

Embora tenhamos apenas duas classes dados vamos empregar aqui uma classifica??o multiclasse. Desse modo voc? pode notar a sa?da do modelo com dimens?o 2, correspondendo ?s duas classes de dados.


## Inicializa??o

N?o temos aqui uma fun??o `compile` como no `TensorFlow` mas precisamos inicializar nosso modelo, e definir a fun??o de perda e o otimizador que desejamos empregar.

Voc? encontrar? no `PyTorch` otimizadores e fun??es de perda semelhantes a que vimos no `TensorFlow` como o Adam, SGD para otimizadores e CrossEntropyLoss, BCELoss (Binary Cross Entropy Loss) e MSELoss para fun??es de perda.

````
model     = Model(X_train.shape[1])
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
loss_fn   = nn.CrossEntropyLoss()
model
````

No `PyTorch` precisamos transformar explicitamente nossas entradas em tensores, tendo ainda aten??o nos tipos de dados suportados por cada m?dulo,

````
X_train = Variable(torch.from_numpy(X_train)).float()
y_train = Variable(torch.from_numpy(y_train)).long()
X_test  = Variable(torch.from_numpy(X_test)).float()
y_test  = Variable(torch.from_numpy(y_test)).long()
````

Neste ponto voc? pode aplicar a rede n?o treinada a um conjunto de dados e observar as sa?das produzidas pelo modelo n?o treinado,

````
model(X_train)[0:10] # exibindo somente as 10 primeiras ocorr?ncias
````

## Treinamento

O treinamento ? bastante semelhante ao que vimos no modelo de subclasses do `TensorFlow` e segue essencialmente:

```
for epoch in EPOCHS:

    # Forward da rede
    y_pred = model(X_train)

    # C?lculo do erro
    loss = loss_fn(y_pred, y_train)
    
    # Zera os gradientes anteriores do otimizador
    optimizer.zero_grad()

    # Recupera os gradientes 
    loss.backward()

    # Atualiza os par?metros da rede com base nos gradientes
    optimizer.step()
```
Sendo os demais comandos opcionais para exibi??o do hist?rico do treinamento.


````
import tqdm # somente para display da evolu??o do loop

EPOCHS  = 100

loss_list     = np.zeros((EPOCHS,))
accuracy_list = np.zeros((EPOCHS,))

for epoch in tqdm.trange(EPOCHS):
    y_pred = model(X_train)
    loss = loss_fn(y_pred, y_train)
    loss_list[epoch] = loss.item()
    
    # Zero gradients
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    with torch.no_grad():
        y_pred = model(X_test)
        correct = (torch.argmax(y_pred, dim=1) == y_test).type(torch.FloatTensor)
        accuracy_list[epoch] = correct.mean()
````

A instru??o `with torch.no_grad()` indica ao `PyTorch` que n?o precisar? obter os gradientes das opera??es aninhadas.

### Curva de Aprendizado

Obtemos ent?o a Curva de Aprendizado. Embora n?o fa?a parte do treinamento da rede ? algo comumente necess?rio para o ajuste do modelo e que frequentemente implementamos.

````
def plot_history(loss_list,  accuracy_list):
  fig, (ax1, ax2) = plt.subplots(2, figsize=(6, 8), sharex=True)

  ax1.plot(accuracy_list)
  ax1.set_ylabel("validation accuracy")
  ax2.plot(loss_list)
  ax2.set_ylabel("validation loss")
  ax2.set_xlabel("epochs")
  plt.show()

  return

plot_history(loss_list,  accuracy_list) 
````

# Regress?o

Vamos criar agora um modelo simples de regress?o. Come?aremos empregando o mesmo conjunto de dados gerado a partir da fun??o $cos(\frac{3}{2}  \pi x)$  da trilha anterior. 

## Dados

````
def true_fun(X):
    return np.cos(1.5 * np.pi * X)

np.random.seed(0)

n_samples = 30

x = np.sort(np.random.rand(n_samples))
y = true_fun(x) + np.random.randn(n_samples) * 0.1

plt.scatter(x,y)
xr = np.linspace(0,1,100)
sns.lineplot(x=xr,y=true_fun(xr),color='orange')

plt.show()
````

### Normaliza??o

````
X = (x - x.mean()) / x.std()
y = (y - y.mean()) / y.std()
````

### Train e Test Split

````
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=123)
````

## Import `torch`

````
import torch
import torch.nn as nn
from torch.autograd import Variable
````

## Definindo o Modelo

Tamb?m constru?mos aqui um modelo de 3 camadas, mas agora nossa sa?da ? um ?nico neur?nio e deve retornar o valor da transforma??o linear sem qualquer fun??o de ativa??o adicional.

````
class Model(nn.Module):
    def __init__(self, input_dim):
        super(Model, self).__init__()
        self.layer1 = nn.Linear(input_dim, 10)
        self.layer2 = nn.Linear(10, 8)
        self.layer3 = nn.Linear(8, 1)
        
    def forward(self, x):
        x = torch.tanh(self.layer1(x))
        x = torch.tanh(self.layer2(x))
        x = self.layer3(x)
        return x
````

## Inicializa??o

Do mesmo modo como fizemos para nosso modelo de classifica??o precisamos inicializar a rede e transformar nossas entradas em tensores.

````
model     = Model(1)
optimizer = torch.optim.SGD(model.parameters(), lr=0.001, momentum=0.9)
loss_fn   = nn.MSELoss()
model
````

Vamos empregar aqui o otimizador SGD e, sendo um modelo de regress?o, ? mais apropriado empregar uma fun??o custo de m?dia do erro quadr?tico.

````
X_train = Variable(torch.from_numpy(X_train)).float()
print( 'Type: ', type(X_train), '\nShape: ',  X_train.shape, '\nContent: ',  X_train[0:5] )
y_train = Variable(torch.from_numpy(y_train)).float()
X_test  = Variable(torch.from_numpy(X_test)).float()
y_test  = Variable(torch.from_numpy(y_test)).float()

X_train = X_train.unsqueeze(-1) # o mesmo que X_train.reshape(len(X_train),1)
y_train = y_train.unsqueeze(-1)
X_test = X_test.unsqueeze(-1)
y_test = y_test.unsqueeze(-1)

print( 'Type: ', type(X_train), '\nShape: ',  X_train.shape, '\nContent: ',  X_train[0:5] )
````

Note que agora todos os dados s?o `float` e fazemos o reshape dos dados para um formato an?logo ao que empregamos antes.

Novamente nosso modelo est? pronto para aplicarmos a rede n?o treinada a um conjunto de dados e observarmos as sa?das do modelo antes do treinamento,

````
model(X_train)[0:5] # Exibindo apenas as 5 primeiras ocorr?ncias
````

## Treinamento

O treinamento n?o tem diferen?a com rela??o ao modelo de classifica??o, mas precisamos ajustar a coleta do hist?rico de treinamento para os dados referentes a regress?o.

````
def plot_history(loss_train, loss_val):

  plt.plot(loss_train, label = 'Train Error')
  plt.plot(loss_val, label = 'Val Error')
  plt.xlabel('Epoch')
  plt.ylabel('Mean Square Error')
  plt.legend()

  plt.show()
````

````
import tqdm

EPOCHS  = 1000

loss_train    = np.zeros((EPOCHS,))
loss_val      = np.zeros((EPOCHS,))

for epoch in tqdm.trange(EPOCHS):
    y_pred = model(X_train)
    loss = loss_fn(y_pred, y_train)
    loss_train[epoch] = loss.item()
    
    # Zero gradients
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    with torch.no_grad():
        y_pred = model(X_test)
        loss = loss_fn(y_pred, y_test)
        loss_val[epoch] = loss.item()

plot_history(loss_train,  loss_val) 

````

## Predi??o

Finalmente, podemos empregar o modelo para a predi??o de um range de valores e observar o resultado obtido.

## Dataset

No `PyTorch`, um conjunto de dados ? representado por uma classe `Python` que herda da classe `Dataset`. Esse `Dataset` pode ter v?rias formas, mas voc? pode pensar, no caso de um problema de classifica??o, como uma lista de tuplas `Python` em que cada tupla ? da forma (*features*, *classe*).

Os m?todos essencias dessa classe s?o:

* __init__(self) : Construir uma lista de tuplas. Um arquivo CSV que ser? carregado e processado; dois tensores , um para *features*  e outro para *classes*; e muitas outras coisas a depender do problema.

* __get_item__(self, index): permite que o conjunto de dados seja indexado  para que possa funcionar como uma lista (dataset[i]) e deve retornar uma tupla (*features*, *classe*) correspondente ao ?ndice solicitado. 

* __len__(self): retorna o tamanho de todo o conjunto de dados.


Assim para nosso exemplo `biopsy` podemos, por exemplo, construir:

````
import pandas as pd
from torch.utils.data import Dataset
from sklearn.preprocessing import LabelEncoder

class CSVDataset(Dataset):
    # load the dataset
    def __init__(self, path):
        # load the csv file as a dataframe
        df = pd.read_csv(path, header=0, index_col=0)
        
        # some preparation 
        df = df.drop(columns=['ID'])
        df.V6 = df.V6.fillna( df.V6.mean() )

        # store the inputs and outputs
        self.X = df.values[:, :-1]
        self.y = df.values[:, -1]

        # ensure input data is floats
        self.X = self.X.astype('float32')

        # label encode target and ensure the values are floats
        self.y = LabelEncoder().fit_transform(self.y)
        self.y = self.y.astype('float32')
        self.y = self.y.reshape((len(self.y), 1))


    # number of rows in the dataset
    def __len__(self):
        return len(self.X)

    # get a row at an index
    def __getitem__(self, idx):
        return [self.X[idx], self.y[idx]]

path = 'https://vincentarelbundock.github.io/Rdatasets/csv/MASS/biopsy.csv'
dataset = CSVDataset(path)

````

Cada uma das __len__() tuplas ? ent?o da forma (*features*, *classe*):

## DataLoader 

At? agora, usamos todos os dados de treinamento em cada etapa do treinamento, isto ?, cada itera??o para ajuste dos pesos faz o forward da rede para todos os dados, calcula o erro e com os gradientes ajusta os pesos. Nosso *batch* (ou lote) de treinamento ? todo o conjunto de dados. Isso pode ser feito quando a quantidade de dados ? bastante pequena mas, em casos mais pr?ticos, devemos empregar *mini-batchs* para o ajuste dos par?metros da rede a cada itera??o. 

A classe da DataLoader do `PyTorch` faz este trabalho para n?s. Ele recebe um `Dataset` e o tamanho do lote desejado, e cria um objeto que se comportar? como um iterador para podermos fazer um loop sobre ele e processar cada mini-batch pode vez.


````
from torch.utils.data import DataLoader

train = DataLoader(train_dataset, batch_size=16, shuffle=True)
````

## Definindo o modelo

Poder?amos empregar aqui o mesmo modelo anterior de classifica??o multiclasse e empreg?-lo pode ser um bom exerc?cio para voc?. Mas temos apenas duas classes de sa?da nos dados e podemos ent?o empregar uma classifica??o bin?ria. A sa?da de rede ? ent?o um ?nico neur?nio que retornar? a *chance* (~ probabilidade) de uma classe de dados tendo como sa?da a fun??o log?stica. 



````
import torch.nn as nn

class MLP(nn.Module):
    # define model elements
    def __init__(self, n_inputs):
        super(MLP, self).__init__()

        self.hidden1 = nn.Linear(n_inputs, 10)
        self.act1 = nn.ReLU()

        self.hidden2 = nn.Linear(10, 8)
        self.act2 = nn.ReLU()

        self.hidden3 = nn.Linear(8, 1)
        self.act3 = nn.Sigmoid()

    # forward  
    def forward(self, X):
        X = self.hidden1(X)
        X = self.act1(X)
        X = self.hidden2(X)
        X = self.act2(X)
        X = self.hidden3(X)
        X = self.act3(X)
        return X

        
````

## Inicializa??o

Agora empregaremos a fun??o de custo *Binary Cross Entropy Loss* e o otimizador Adam, e inicializamos nosso modelo.

````
from torch.optim import SGD, Adam

model = MLP(9)
loss_fn = nn.BCELoss()
# optimizer = SGD(model.parameters(), lr=0.01, momentum=0.9) # Voc? pode optar por esse se quiser
optimizer = Adam(model.parameters(), lr=0.001)

print( model )
````

As transforma??es para tensores s?o feitas pelas classes `Dataset` e `DataLoader` e voc? pode aplicar o modelo ainda n?o treinado a um *lote* da pr?xima itera??o diretamente:

````
X_train, y_train = next(iter(train))

model(X_train)
````

## Treinamento

Apenas para tornar o c?digo mais simples n?o iremos gerar o hist?rico de acuracidade como antes e iremos apenas exibir a evolu??o da fun??o de custo.

````
EPOCHS = 100
loss_list = np.zeros((EPOCHS,))

for epoch in range(EPOCHS):
    loss_batch = 0

    for x_batch, y_batch in train:
        # compute the model output
        y_pred = model(x_batch)
        loss = loss_fn(y_pred, y_batch)
        # Zero gradients
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        loss_batch = loss_batch + loss.item()
    
    loss_list[epoch] = loss_batch / len(x_batch)
````

````
def plot_history(loss_list):

  plt.plot(loss_list)
  plt.ylabel("Train Loss")
  plt.xlabel("Epoch")
  plt.show()

plot_history(loss_list) 
````
