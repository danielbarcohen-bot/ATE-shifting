## Parámetros que se pueden cambiar

### LogisticRegression

- `max_iter`: número máximo de iteraciones.
- `C`: controla regularización. Valores pequeños aplican más regularización.
- `solver`: algoritmo de optimización.

### RandomForest

- `n_estimators`: cantidad de árboles.
- `max_depth`: profundidad máxima.
- `min_samples_split`: mínimo de muestras para dividir un nodo.
- `random_state`: reproducibilidad.

No se debe modificar el conjunto de prueba para mejorar artificialmente los resultados.
