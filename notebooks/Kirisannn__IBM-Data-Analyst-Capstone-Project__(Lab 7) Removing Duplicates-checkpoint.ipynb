### Step 3: Identifying Duplicate Rows


**Task 1: Identify Duplicate Rows**
  1. Count the number of duplicate rows in the dataset.
  2. Display the first few duplicate rows to understand their structure.


```python
dups=df[df.duplicated(keep=False)]
print(dups.shape[0])
dups.head()
```

### Step 4: Removing Duplicate Rows


**Task 2: Remove Duplicates**
   1. Remove duplicate rows from the dataset using the drop_duplicates() function.
2. Verify the removal by counting the number of duplicate rows after removal .


```python
df.duplicated(keep=False).value_counts()
```

```python
df_clean = df.drop_duplicates(keep='first')
if (df.shape[0] - df_clean.shape[0] == dups.shape[0] / 2):
    df = df_clean
    print("Successfully removed duplicates")
```

### Step 5: Handling Missing Values


**Task 3: Identify and Handle Missing Values**
   1. Identify missing values for all columns in the dataset.
   2. Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value.


```python
# 1. Identifying missing values
missing=df.isna()
```

```python
missing_stats=missing.sum().sort_values(ascending=False).to_frame(name='count')
missing_stats['Missing (%)'] = (missing_stats['count'] / len(df) * 100).round(3)
missing_stats
```

```python
print(f"EdLevel missing percentage: {missing_stats.loc['EdLevel', 'Missing (%)']}%")
```

```python
print(f"ConvertedCompYearly missing percentage: {missing_stats.loc['ConvertedCompYearly', 'Missing (%)']}%")
```

```python
# Define function to impute columns according to data type
def impute_column(col):
    if col.dtype == 'O': # Object data type
        return col.fillna(col.mode()[0]) # Fill missing with first mode
    else: # Numeric data types
        return col.fillna(col.median())        
```

```python
# Impute only EdLevel (categorical)
df['EdLevel'] = impute_column(df['EdLevel'])
```

```python
# Checking successful imputation of columns via lack of missing values
print(f"EdLevel number of missing values: {df['EdLevel'].isna().sum()}")
```

### Step 6: Normalizing Compensation Data


**Task 4: Normalize Compensation Data Using ConvertedCompYearly**
   1. Use the ConvertedCompYearly column for compensation analysis as the normalized annual compensation is already provided.
   2. Check for missing values in ConvertedCompYearly and handle them if necessary.


```python
scaled = df.copy()
```

```python
# Min-Max Scaling
scaled['ConvertedCompYearly'] = (scaled['ConvertedCompYearly'] - scaled['ConvertedCompYearly'].min()) / (scaled['ConvertedCompYearly'].max() - scaled['ConvertedCompYearly'].min())
```

```python
# Impute only ConvertedCompYearly (numeric)
scaled['ConvertedCompYearly'] = impute_column(scaled['ConvertedCompYearly'])
print(f"ConvertedCompYearly number of missing values: {scaled['ConvertedCompYearly'].isna().sum()}")
```
