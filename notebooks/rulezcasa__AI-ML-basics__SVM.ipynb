```python
#Splitting dependent and independent variables
#Independent variables
x=df.iloc[:,1:4]
x.head()
```

```python
#Dependent variable
y=df.Purchased
y.head()
```

```python
#Label encoding
from sklearn.preprocessing import LabelEncoder
```

```python
#Creating an object
le=LabelEncoder()
```

```python
#Encoding
x.Gender=le.fit_transform(x.Gender)
x.head()
```

```python
#Feature scaling
from sklearn.preprocessing import MinMaxScaler
```

```python
#Creating object
ms=MinMaxScaler()
```

```python
#Scaling
x_scaled=pd.DataFrame(ms.fit_transform(x),columns=x.columns)
x_scaled.head()
```

```python
#Train-test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x_scaled,y,test_size=0.2,random_state=0)
```

### SVC model building 

```python
#Model building
from sklearn.svm import SVC
model=SVC(probability=True)
```

```python
rand_list={"C":[2,3,4,5,6,7,8,10],
            "gamma":[0.1,0.2,0.5,0.4,0.8]}
```

```python
#Hyperparameter tuning
from sklearn.model_selection import RandomizedSearchCV
rand_search= RandomizedSearchCV(model,param_distributions=rand_list,n_iter=20,cv=5)
```

```python
#Fitting model
rand_search.fit(x_train,y_train)
```

```python
#Predicting using test data and comparing it with actual values
pred=rand_search.predict(x_test)
```

```python
new=pd.DataFrame({"Actual":y_test,"Predicted":pred})
new
```

```python
#Evaluation of model
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,roc_auc_score,roc_curve
```

```python
#Accuracy score
accuracy_score(y_test,pred)
```

```python
#Confusion matrix
confusion_matrix(y_test,pred)
```

```python
pd.crosstab(y_test,pred)
```

```python
print(classification_report(y_test,pred))
```

```python
#Declaring variables
TN=58
FP=0
FN=6
TP=16
```

```python
#Calculating precision
precision=TP/(TP+FP)
precision
```

```python
#Calculating recall
recall=TP/(FN+TP)
recall
```

```python
#Calculating F1 score
F1=2*precision*recall/(precision+recall)
F1
```

```python
#ROC-AUC curve
probability=rand_search.predict_proba(x_test)[:,1]
probability
```

```python
#ROC curve
fpr,tpr,thresholds=roc_curve(y_test,probability)
```

```python
plt.plot(fpr,tpr)
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.show()
```
