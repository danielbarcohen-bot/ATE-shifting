```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
X_train
```

#### ?? ?? - ??????(Decision Tree)
<img align="left" src='http://drive.google.com/uc?export=view&id=1OZmnohHM8_932fiNnokTRdC6ZQumrPeL'/>

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model_dt = DecisionTreeClassifier(random_state=0)
```

```python
model_dt.fit(X_train, y_train)
```

#### ?? ?? - ???? ????(Logistic Regression)
<img src="https://www.saedsayad.com/images/LogReg_1.png">

```python
from sklearn.linear_model import LogisticRegression

model_lr = LogisticRegression(random_state=0)
model_lr.fit(X_train, y_train)
```

#### ?? ?? - ?????(Neural Network)
<img src="https://www.astroml.org/_images/fig_neural_network_1.png">

```python
from sklearn.neural_network import MLPClassifier

model_nn = MLPClassifier(random_state=0)
model_nn.fit(X_train, y_train)
```

#### ?? ??

```python
# Decision Tree
model_dt.score(X_test, y_test)
```

```python
# Logistic Regression
model_lr.score(X_test, y_test)
```

```python
# Neural Network
model_nn.score(X_test, y_test)
```

```python
# Final Model
model_best = model_dt
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

#### ?? ?? - ??????

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

#### ?? ??

```python
print(model.coef_)
print(model.intercept_)
```

#### ?? ??

```python
model.score(X_test, y_test) # R-square(r2_score)
```

#### ???? ???

```python
y_pred = model.predict(X_test)
y_pred
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
plt.scatter(y_test, y_pred, edgecolors=(0, 0, 0))
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=4)
plt.xlabel('Measured')
plt.ylabel('Predicted')
```

#### ??? ?? ?? - ??????

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(random_state=0)
model.fit(X_train, y_train)
model.score(X_test, y_test)
```
