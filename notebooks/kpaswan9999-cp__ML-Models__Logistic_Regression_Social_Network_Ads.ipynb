### Encoding Categorical Data
-This step prepares the categorical feature for model training.

```python
le = LabelEncoder()
df["Gender"] = le.fit_transform(df["Gender"])
```

## Step 6: Feature?Target Separation

```python
x=df[['Gender','Age','EstimatedSalary']]
y=df[['Purchased']]
```

### Traing and Testing the Features (Train-Test Splitting)
- Prevents overfitting
- Allows evaluation on unseen data

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

### Feature Scaling (Standardization).

```python
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

## Step 7: Model Initialization & Model Training

```python
model=LogisticRegression()
model.fit(x_train,y_train)
```

### Making Predictions
- model is going to predict the target values for the test dataset.
- These predictions will be used to evaluate the model?s performance.

```python
y_pred=model.predict(x_test)
```

### Confusion Matrix
- The confusion matrix is created to evaluate the performance of the classification model.

```python
cm=confusion_matrix(y_test,y_pred)
print(cm)
```

```python
print("Accuracy:",accuracy_score(y_test,y_pred))
print(classification_report(y_test, y_pred))
```

### Observation:
- The Logistic Regression model achieved an accuracy of 83.75%, indicating good overall performance in predicting whether users will purchase the product.
- The model performs better at predicting non-purchasing users (class 0) than purchasing users (class 1), as shown by the higher recall for class 0.

## Conclusion
- After preprocessing, encoding, and training the model, it achieved an accuracy of about 83.75%, indicating good predictive performance.
- The model predicts non-purchasing users more accurately than purchasing users, but overall it effectively captures the relationship between user demographics and purchasing behavior.
