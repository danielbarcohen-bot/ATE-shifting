```python
import pandas
Data = pandas.read_csv("Newdataset/family_suitability.csv")
Data
```

```python
Data.describe()
```

```python
array = Data.values
X = array[:,0:6]
Y = array[:,6]
```

```python
from sklearn import model_selection
X_train,X_test,Y_train,Y_test = model_selection.train_test_split(X,Y, test_size=0.3,random_state=100)
```

```python
# load a classification model we are using svc- support vectror machine
from sklearn.svm import SVC
model = SVC()
model.fit(X_train,Y_train)
```

```python
predictions = model.predict(X_test)
print('Model predicted',predictions)
print('Expected', Y_test)
```

```python
from sklearn.metrics import accuracy_score
print('Accuracy_score',accuracy_score(Y_test,predictions))
```

```python
results = [[6,9,10,00,2,00,]]
outcome = model.predict(results)
print('results:',outcome)
```
