```python
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.decomposition import PCA
import warnings

warnings.filterwarnings('ignore')

df_heart = pd.read_csv('heart_disease_uci.csv')
df_heart.head()
```

```python
df_heart.drop(['id', 'dataset'], axis=1, inplace=True, errors='ignore')
```

```python
df_heart.fillna(df_heart.median(numeric_only=True), inplace=True)
```

```python
for col in df_heart.columns:
    if df_heart[col].dtype == 'object':
        df_heart[col].fillna(df_heart[col].mode()[0], inplace=True)
target = (df_heart['num'] > 0).astype(int)
features = df_heart.drop('num', axis=1)
```

```python
le = LabelEncoder()
categorical_cols = ['sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'thal']
for col in categorical_cols:
    if col in features.columns:
        features[col] = le.fit_transform(features[col].astype(str))
```

```python
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)
```

```python
pca = PCA(0.95)
features_pca = pca.fit_transform(features_scaled)
```

```python
X_train, X_test, y_train, y_test = train_test_split(features_pca, target, test_size=0.2, random_state=40)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
```

```python
lr.score(X_test, y_test)
```

```python
cross_val_score(lr, features_pca, target, cv=3).mean()
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier()
dt.fit(X_train, y_train)
```

```python
dt.score(X_test, y_test)
```

```python
cross_val_score(dt, features_pca, target, cv=3).mean()
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(X_train, y_train)
```

```python
rf.score(X_test, y_test)
```

```python
cross_val_score(rf, features_pca, target, cv=3).mean()
```

```python
from sklearn.svm import SVC
svm = SVC()
svm.fit(X_train, y_train)
```

```python
svm.score(X_test, y_test)
```

```python
cross_val_score(svm, features_pca, target, cv=3).mean()
```

```python
from sklearn.naive_bayes import GaussianNB
nb = GaussianNB()
nb.fit(X_train, y_train)
```

```python
nb.score(X_test, y_test)
```

```python
cross_val_score(nb, features_pca, target, cv=3).mean()
```
