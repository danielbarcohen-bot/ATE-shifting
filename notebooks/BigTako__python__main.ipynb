```python
#14
import pandas as pd

df = pd.read_csv("csv/NationalNames.csv")
unique_names_by_year = df.groupby(['Year', 'Name'])['Gender'].nunique()

gender_neutral_names_by_year = unique_names_by_year[unique_names_by_year == 2]

year_with_most_gender_neutral_names = gender_neutral_names_by_year.reset_index().groupby('Year').count().idxmax()['Name']

count_of_gender_neutral_names = len(gender_neutral_names_by_year[gender_neutral_names_by_year.index.get_level_values('Year') == year_with_most_gender_neutral_names])

print("Year Gender_neutral_names\n" + str(year_with_most_gender_neutral_names) + "                 " + str(count_of_gender_neutral_names))
```

```python
#26
import pandas
import matplotlib.pyplot as pyplot

csv_path = 'csv/NationalNames.csv'
df = pandas.read_csv(csv_path)
pyplot.xlabel('Year')
pyplot.ylabel('Count')
john_df = df.loc[
    (df['Name'] == 'John') &
    (df['Gender'] == 'F')]
mary_df = df.loc[
    (df['Name'] == 'Mary') &
    (df['Gender'] == 'M')]
pyplot.plot(john_df['Year'], john_df['Count'], label="John")
pyplot.plot(mary_df['Year'], mary_df['Count'], label="Mary")
pyplot.legend(title = "Name")
```
