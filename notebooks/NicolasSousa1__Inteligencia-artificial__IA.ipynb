### Case: Score de Cr?dito dos Clientes

Voc? foi contratado por um banco para conseguir definir o score de cr?dito dos clientes. Voc? precisa analisar todos os clientes do banco e, com base nessa an?lise, criar um modelo que consiga ler as informa??es do cliente e dizer automaticamente o score de cr?dito dele: Ruim, Ok, Bom



```python
# IA para previs?o de score 

#passo 1 importar a base de dados e verificar se ? necessario tratar a base dados
#passo 2 transformar os textos da base de dados em numeros para a IA usando o LabelEncoder
#passo 3 escolher as colunas que ser?o usadas para treinar a IA
#passo 4 criar a IA
#passo 5 Treinar a IA
#passo 6 testar a IA e escolher o melhor modelo
#passo 7 importar a tabela com os novos clientes
#passo 8 usar o melhor modelo de IA para prever o score de novos clientes
```

```python
#passo 1 importar a base de dados e verificar se ? necessario tratar a base dados

import pandas as pd

tabela = pd.read_csv("clientes.csv") # importa a base de dados
display(tabela)
```

```python

# verificar se temos valores vazios ou valores reconhecidos em formato errado
print(tabela.info())
print(tabela.columns)
```

```python
#passo 2 transformar os textos da base de dados em numeros para a IA usando o LabelEncoder

from sklearn.preprocessing import LabelEncoder
# vai transformar as colunas de texto em n?meros, ex: profissoes vai sair de cientista, professor, mecanico, etc para 0, 1, 2, etc

codificador_profissao = LabelEncoder()
tabela["profissao"] = codificador_profissao.fit_transform(tabela["profissao"])

codificador_mix_credito = LabelEncoder()
tabela["mix_credito"] = codificador_mix_credito.fit_transform(tabela["mix_credito"])

codificador_comportamento_pagamento = LabelEncoder()
tabela["comportamento_pagamento"] = codificador_comportamento_pagamento.fit_transform(tabela["comportamento_pagamento"])

print(tabela.info())
```

```python
#passo 3 escolher as colunas que ser?o usadas para treinar a IA

# y ? a coluna que queremos que o modelo calcule
# x vai todas as colunas que vamos usar para prever o score de credito, n?o vamos usar a coluna id_cliente porque ela ? um numero qualquer que nao ajuda a previsao

x = tabela.drop(["score_credito", "id_cliente"], axis=1)
y = tabela["score_credito"]

from sklearn.model_selection import train_test_split

# separamos os dados em treino e teste. Treino vamos dar para os modelos aprenderem e teste vamos usar para ver se o modelo aprendeu corretamente
x_treino, x_teste, y_treino, y_teste = train_test_split(x, y, test_size=0.3, random_state=1)
```

```python
#passo 4 criar a IA

from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

modelo_arvore = RandomForestClassifier() # modelo arvore de decisao
modelo_knn = KNeighborsClassifier() # modelo do KNN (nearest neighbors - vizinhos mais proximos)


```

```python
#passo 5 Treinar a IA

# treinando os modelos
modelo_arvore.fit(x_treino, y_treino)
modelo_knn.fit(x_treino, y_treino)
```

```python
#passo 6 testar a IA e escolher o melhor modelo

from sklearn.metrics import accuracy_score

# calculamos as previsoes
previsao_arvore = modelo_arvore.predict(x_teste)
previsao_knn = modelo_knn.predict(x_teste.to_numpy())

# comparamos as previsoes com o y_teste
# nesse score queremos a maior acuracia
print(accuracy_score(y_teste, previsao_arvore))
print(accuracy_score(y_teste, previsao_knn))
```

```python
#passo 7 importar a tabela com os novos clientes


#importando a tabela com os novos clientes
tabela_novos_clientes = pd.read_csv("novos_clientes.csv")

#usando o LabelEncoder na nova tabela

nova_codificador_profissao = LabelEncoder()
tabela_novos_clientes["profissao"] = nova_codificador_profissao.fit_transform(tabela_novos_clientes["profissao"])

nova_codificador_mix_credito = LabelEncoder()
tabela_novos_clientes["mix_credito"] = nova_codificador_mix_credito.fit_transform(tabela_novos_clientes["mix_credito"])

nova_codificador_comportamento_pagamento = LabelEncoder()
tabela_novos_clientes["comportamento_pagamento"] = nova_codificador_comportamento_pagamento.fit_transform(tabela_novos_clientes["comportamento_pagamento"])

display(tabela_novos_clientes.info())


```

```python
#passo 8 usar o melhor modelo de IA para prever o score de novos clientes

#fazendo a previsao do score dos novos clientes com a IA

nova_previsao = modelo_arvore.predict(tabela_novos_clientes)
display(nova_previsao)
```
