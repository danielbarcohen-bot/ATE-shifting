```python
import numpy
import matplotlib.pyplot as plt
import seaborn
import pandas
```

## Unemployment rate by region

```python
unemployment_eurostat = pandas.read_csv(".data/unemployment/europe_unemployment_rate.csv",low_memory=False)
```

```python
unemployment_eurostat.head()
```

```python
unemployment_eurostat.tail()
```

```python
seaborn.heatmap(unemployment_eurostat.isnull(),cbar=False)
```

```python
color = seaborn.color_palette()
cnt_srs = unemployment_eurostat.State.value_counts()

plt.figure(figsize=(12,8))
seaborn.barplot(cnt_srs.index, cnt_srs.values, alpha=0.8, color=color[1])
plt.ylabel('Number of Occurrences', fontsize=12)
plt.xlabel('States', fontsize=12)
plt.title('Count the states', fontsize=15)
plt.xticks(rotation='vertical')
plt.show()
```

```python
grouped_df = unemployment_eurostat.groupby(["Country"])["Rate"].aggregate("mean").reset_index()

plt.figure(figsize=(12,8))
seaborn.pointplot(grouped_df['Country'].values, grouped_df['Rate'].values, alpha=0.8, color=color[2])
plt.ylabel('Mean rate', fontsize=12)
plt.xlabel('States', fontsize=12)
plt.title("Average of mean", fontsize=15)
plt.xticks(rotation='vertical')
plt.show()
```
