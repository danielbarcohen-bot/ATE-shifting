```python
subdf = df[["satisfaction_level", "promotion_last_5years", "salary"]]
subdf.sample(10)
```

Deal with salary as a dummy variable

```python
salary_dummy = pd.get_dummies(subdf.salary, prefix= "salary")
```

```python
# concat with main df
df_dummies = pd.concat([subdf, salary_dummy], axis= "columns")
```

```python
df_dummies.head(20)
```

```python
# remove salary column with text data
df_dummies.drop("salary", axis= "columns", inplace= True)
```

```python
df_dummies.head()
```

```python
# let df_dummies be X
X = df_dummies
X.head()
```

```python
y = df.left
```

```python
df.left.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size= 0.3)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model
```

```python
model.fit(X_train, y_train)
```

```python
model.predict(X_test)
```

accuracy of model

```python
model.score(X_test, y_test)
```
