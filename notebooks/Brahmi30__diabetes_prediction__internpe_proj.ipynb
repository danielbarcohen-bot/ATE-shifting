```python
dataset=pd.read_csv("diabetes.csv")
```

separatig data and labels

```python
d1=dataset.drop(columns="Outcome",axis=1)#dataset with no outcome column
d2=dataset["Outcome"]#dataset with only outcome column
```

```python
d1.head()
```

```python
d2.head()
```

Data Standardization

```python
scaler=StandardScaler()
```

```python
scaler.fit(d1)
```

```python
s_data=scaler.transform(d1)
```

```python
print(s_data)
```

```python
d1=s_data
d2=dataset["Outcome"]
```

```python
print(d1)
print(d2)
```

Train Text Split

```python
d1_train,d1_test,d2_train,d2_test=train_test_split(d1,d2,test_size=0.2,stratify=d2,random_state=2)
```

```python
print(d1.shape,d1_train.shape,d1_test.shape)
```

Training

```python
classifier=svm.SVC(kernel="linear")
```

```python
classifier.fit(d1_train,d2_train)
```

Model Evaluation

Accuracy score???

```python
d1_train_acc=classifier.predict(d1_train)
data_train_accuracy=accuracy_score(d1_train_acc,d2_train)
```

```python
print("Accuracy score: ",data_train_accuracy)
```

Test data

```python
d1_test_acc=classifier.predict(d1_test)
data_test_accuracy=accuracy_score(d1_test_acc,d2_test)
```

```python
print(data_test_accuracy)
```

A predictive System

```python
input = (5,116,74,0,0,25.6,0.201,30)
columns = dataset.drop(columns="Outcome").columns
input_df = pd.DataFrame([input], columns=columns)
std_data = scaler.transform(input_df)
predict = classifier.predict(std_data)
print(predict)
if predict[0] == 0:
    print("Person is non-diabetic")
else:
    print("Person is diabetic")
```
