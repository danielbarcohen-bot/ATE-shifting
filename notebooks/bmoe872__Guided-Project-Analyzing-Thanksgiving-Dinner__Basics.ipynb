```python
def parse_age(age):
    if pandas.isnull(age):
        return None
    
    result = age.split(' ')[0]
    result = result.replace('+', '')
    return int(result)

data["int_age"] = data["Age"].apply(parse_age)
```

```python
data["int_age"].describe()
```

```python
# These numbers were found using one side of an age range.
# i.e. if 18 - 25, 18 would be used

# This does not seem to be an accurate estimate, as the
# estimate is coming from one side of the age range.
# To get a more accurate estimate, outside of getting more specific data,
# it might be better to assume the average of the age range, and do a
# calculation on that. The data does look to be distributed evenly though
# The above approach might be more work than its worth
```

```python
import pandas

data = pandas.read_csv("thanksgiving.csv", encoding="Latin-1")

def parse_household_income(income_range):
    if pandas.isnull(income_range):
        return None
    
    result = income_range.split(' ')[0]
    if result == "Prefer":
        return None
    
    result = result.replace("$", "")
    result = result.replace(",", "")
    
    return int(result)

data["int_income"] = data["How much total combined money did all members of your HOUSEHOLD earn last year?"].apply(parse_household_income)
print(data["int_income"].describe())
```

```python
# We took the lower bound of the range to find the above results
# This is a rather good representation, and I think its actually
# a bit conservative given that we're looking at the lower bound.
```

```python
less_150000 = data[data["int_income"] < 150000]

less_150000["How far will you travel for Thanksgiving?"].value_counts()
```

```python
over_150000 = data[data["int_income"] > 150000]

over_150000["How far will you travel for Thanksgiving?"].value_counts()
```

```python
# There are definitely fewer people making over 150000
# It seems like more people in the lower than 150000 will be staying home
# overall. Which is not what this was supposed to find according to the
# exercise definition. This would seem to suggest a good amount of 
# people making lower incomes live at home. However there is a smaller
# sample of people making over 150000, and of those they all seem to stay
# home. SO the original definition is correct, but I was surprised to
# see the numbers as high as they were for the lower income.
```
