# Load Data

```python
import pandas as pd
import plotly.express as px
import dash
from dash import dcc, html, Input, Output,  dash_table

# get head of data
df = pd.read_csv('./data/netflix_titles.csv')
print(df.columns)
print(df.dtypes)
print(df.isnull().sum())
print(df.shape)
```

# Feature engineering

```python
df['Top 3 cast'] = df['cast'].apply(lambda x: ', '.join(x.split(', ')[:3]) if pd.notnull(x) else x)
df.head()
```

# Ideas
- Data table to view Data Structure
- Top ten movies in a specific country
- Top ten director in a specific country
- Recomender system (possibly)
- Sort by top rating

# Creating dashboard

```python

columns_for_table = ['title', 'Top 3 cast', 'director', 'country']

app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1("Netflix in Numbers", style={'color': '#D3DAD9'}),
    html.Hr(),
    html.Img(src='https://images.ctfassets.net/y2ske730sjqp/1aONibCke6niZhgPxuiilC/2c401b05a07288746ddf3bd3943fbc76/BrandAssets_Logos_01-Wordmark.jpg?w=940', 
            style={'height': '300px', 'width': '400px', 'margin-bottom': '20px', 'border-radius': '10px'}),
    dash_table.DataTable(
        id='data-table',
        columns=[{"name": i, "id": i} for i in columns_for_table],
        data=df[columns_for_table].to_dict('records'),
        page_size=10,
            style_table={'overflowX': 'auto'},
            style_cell={'textAlign': 'left', 'backgroundColor': '#715A5A', 'color': '#D3DAD9'},
            style_header={
                'backgroundColor': '#44444E',
                'fontWeight': 'bold'
            },
            filter_action='native',
        )
    ], style={'textAlign': 'center', 'backgroundColor':  '#44444E', 'padding': '10px'})

if __name__ == '__main__':
    app.run(debug=True, port=8050)
```
